## 1. NLP와 언어 모델


### 1-1 NLP (Natural Language Processing)

자연어란, 인간이 소통을 위해 자연스럽게 발생시켜 사용하는 언어를 말한다.
컴퓨터가 이해하는 프로그램 언어와 대비대는 개념

- NLP의 역할: 이 모호한 자연어를 컴퓨터가 이해할 수 있는 숫자로 변환하여 유의미한 가치를 창출하는 기술

- NLP란, 인간의 언어의 의미를 분석하고, 컴퓨터가 처리할 수 있도록 하는 AI 기술 분야

- NLP는 비정형 데이터인 글자를 컴퓨터가 계산할 수 있는 숫자로 바꾸고, 그 안에서 의미를 추출하는 모든 과정을 포함


#### 자연어의 특징

- 이산적 데이터: 언어는 단어라는 독립적이고 딱 끊어지는 단위로 구성된다 -> 컴퓨터가 이해할 수 있도록 임베딩 변환 과정 필요

- 순차적 데이터: 단어가 나열되는 순서에 따라 문장의 의미가 완전히 결정된다 -> 데이터의 앞뒤 관계와 위치정보를 처리할 수 있는 모델 구조가 필요

- 중의성과 모호성: 동일한 형태의 단어라도 문맥에 따라 뜻이 천차만별 -> 단어 하나가 아닌 문맥을 전체적으로 파악하는 기술이 핵심

- 희소성: 단어의 조합은 무한에 가깝지만, 우리가 가진 데이터에 나타나는 조합은 극히 일부임 -> 한 번도 본 적 없는 문장이 들어와도 단어 간 유사도 추론가능한 일반화 능력 필요

- 가변성과 진화: 언어는 고정되어 있지 않음 -> 데이터로부터 유연하게 패턴을 학습하는 머신러닝 방식이 필수적



### 1-2 언어모델

- 가장 넓은 의미에서 언어 모델은 인간이 말하고 쓰는 방식을 모방하는 시스템을 의미

- 언어 모델은 단어 시퀀스에 확률을 할당하여, 가장 자연스러운 단어 시퀀스를 찾아내는 모델

- 언어모델은 크게 통계 기반 vs 인공 신경망 기반으로 구분됨

## 2. 통계적 언어 모델(SLM)

### 2-1. Statistic Language Model이란?

- 조건부 확률: 앞에 나오는 단어 뒤에 어떤 단어가 나올지 확률을 활용하는 방식

- 카운트 기반: 
    - 이전 단어 시퀀스의 등장 빈도를 통해 다음 단어의 확률을 계산
    - 카운트 기반의 통계적 언어 모델은 희소 문제가 존재
    - 모델은 이전에 경험한 적 있는 조합에 대해서만 잘 예측할 수 있음 -> 이에 새로운 문장 조합에도 적용할 수 있는 신경망 모델 필요


### 2-2. N-gram

- N-gram이 나오게 된 배경:
    - SLM은 희소하게 등장하는 것에 대한 예측을 못하며, 문장이 길어질수록 다음 단어 계산하기 힘듬
    - 전체 문장 대신에 문장의 일부만 보는것 -> N-gram

- N-gram의 한계:
    - 데이터 부족: 훈련 데이터에서 나타나지 않은 단어 조합은 확률이 0이 됨
    - N의 크기가 커질 수록 훈련데이터에서 해당 조합을 찾기 어려워져 희소 문제가 심해짐


### 2-3. Perplexity

- PPL: 언어 모델이 문장을 얼마나 혼란스러워하는지를 수치로 나타낸 것

- 문장 W가 희소하게 등장하면 모델이 훈련 데이터에서 더 적게 봄 -> W가 희소하게 등장하여 PPL 커짐

- 단어 개수 N이 커지면 함께 보는 단어들이 더 많아져, 희소한 부분이 있더라도 그 안에 예측이 쉬운 단어들도 같이 있어 PPL을 낮춤

- 신경망 기반 언어모델은 일반화 능력이 뛰어남 -> 이에 신경망 언어 모델로 관심이 쏠림

## 3. 신경망 언어 모델 (NLM)

- 임베딩: 단어를 컴퓨터의 언어로 번역하는 다리 역할

- SLM: 유사한 분야의 언어도 완전히 다른 이름으로 해석

- 신경망 모델: 유사한 분야의 언어는 비슷한 공간에 임베딩 됨

## 4. 딥러닝 기반 언어 모델

### 4-1. Transformer의 등장

- Transformer: RNN 셀을 반복 사용하는 것 대신, 인코더와 디코더를 여러 층으로 깊게 쌓은것

- 이를 통해 언어 모델의 표현력이 높아져, 현재의 딥러닝 기반 언어모델이 생성됨


### 4-2. BERT

#### BERT란?

BERT = Bidiretional Encoder Representations from Transformers

즉 양방향으로 작동하는 Transformer -> Bert는 encoder 기반, GPT는 decoder 기반

- BERT의 특징
    - 양방향: 문장을 읽을 때 앞뒤를 동시에 고려 -> 한 단어를 이해할 때 그 단어의 좌우 문맥을 동시에 보고 이해
    - 트랜스포머 기반: 트랜스포머 구조의 encoder 부분만 사용
    - 사전학습 + 파인튜닝: 큰 데이터로 먼저 일반적인 언어 능력을 배우고, 나중에 특정 과제에 맞게 조절

#### BERT의 학습 방법

BERT의 기본 구조는 transformer의 encoder을 여러 개 쌓아 올린 구조.

- 처리 과정: Input -> Embedding -> 트랜스포머 인코더 12개 or 24개 -> Output

- 입력 문장을 embedding 한 뒤, 여러 개의 encoder을 통과하면서 각 단어의 의미를 문맥적으로 풍부하게 만듬

- 사전 학습
    - Masked Language Model
        - 빈 칸 채우기 방식으로 학습: 무작위로 빈칸을 뚫어 이 빈칸에 어떤 단어가 들어갈 지 모델이 예측하며 학습
        - input text의 15% 정도의 단어를 랜덤으로 마스킹

    - Next Sentence Prediction(NSP)
        - 두 문장에 대해, 이 문장들의 논리 흐름을 학습

    - BERT 이후의 모델 -> RoBERTa, ALBERT



### 4-3. GPT

#### GPT란?

BERT는 transformer encoder 기반, GPT는 decoder 기반

GPT도 지금까지의 단어를 바탕으로 다음 단어를 예측 -> 왼쪽의 단어들 만 보고, 오른쪽에 올 단어 예측

few-shot learning도입 -> 몇개의 예시를 보고 task 수행


### 4-4. LLM

LLM  = Large Language Model의 줄임말

- LLM의 작동 방식: 학습과 예측이라는 두가지 핵심 단계를 통해 작동함

    - 대규모 학습: 수십억개의 문장으로 이루어진 방대한 데이터를 학습 -> 이때 단어간 관계, 문맥적 의미, 문법구조등 언어 패턴을 스스로 터득

    - 예측 및 생성: 학습이 완료된 LLM은 사용자의 질문이나 지시를 받으면, 다음에 올 가장 확률 높은 단어를 순서대로 예측하며 응답을 생성(GPT방식)

- LLM의 한계

    - hallucination -> GPT에서 발생했던 문제 - 현재 해결됨

    - 업데이트 비용 -> LLM은 파라미터 수가 많기 때문에 새로운 정보를 반영할 때 계산량이 많음


### 4-5. RAG

- RAG의 필요성: 학습된 이후에 새로운 지식에 대해서 답변 불가 -> RAG사용

- RAG의 구조:

    - 질의 인코더: 사용자의 질문을 벡터화

    - 지식 검색기: 인코딩된 질문을 바탕으로 외부지식에서 관련 내용 검색

    - 지식 증강 생성기: 검색한 내용을 참고해서 답변 생성

- RAG의 장점: 최신 정보를 검색해서 가져올 수 있는 최신 정보 반영 기능이 가장 큰 장점 + hallucination 감소

- CAG: Credibility - Aware Generation

    - 기존 RAG: 외부 문서를 검색해 모델에 제공 -> 지식 부족, hallucination 완화 but 검색된 문서의 신뢰도를 판단 못함

    - 이에 CAG -> 문서의 신뢰도를 스스로 판단하고 활용할 수 있도록 훈련시키는 기법


- CAG의 구조: 질문 입력 -> 정보 검색 -> 각 정보의 신뢰도 평가 -> 높은 신뢰도 정보 우선 사용 -> 최종 답변 생성


### 4-6. LangChain

- Langchain: LLM + RAG를 구현하는 것이 복잡하므로, pipeline을 쉽게 구성할 수 있도록 LangChain 프레임워크 사용

    - Langchain의 특징:

        - 추상화: LLM 서비스를 만들 때 필요한 작업 간소화

        - 표준화: 비슷한 기능을 갖추고 있는 요소들을 똑같은 형식을 갖춘 컴포넌트로 표준화

     - 체이닝: 자주 활용하는 주요 컴포넌트를 쉽게 연결해 LLM 서비스의 로직 쉽게 파악

- LangGraph:

    - Langchain의 한계: 중간에 결과가 마음에 들지 않을때 이전 단계로 돌아가는 반복 로직 구현 어려움 -> LangGraph 등장

    - 비선형적인 작업형태로 cycle, loop를 만듬
        - node: 특정 작업을 수행하는 단위로 랭체인의 컴포넌트 사용
        - edge: 노드와 노드를 연결하는 선 -> 조건부 엣지를 통해 다음에 어디로 갈 것인지 결정


- LangSmith:

    - LLM 기반 어플리케이션을 위한 통합 개발 및 운영 플랫폼

    - 핵심 기능:

        - 디버깅 및 추적: LLM 어플리케이션의 모든 실행 과정을 단계별로 시각화

        - 테스트 및 평가: 미리 준비된 질문과 정답 데이터셋을 만들어, 개발 중인 LLM 어플리케이션의 성능 자동 평가

        - 모니터링: 실제 사용자들이 애플리케이션을 어떻게 사용하는지 모니터링

        - 프롬프트 허브: 다양한 프롬프트를 저장, 버전 관리, 쉽게 테스트할 수 있는 공간 제공


- 외부 API 활용:

    - 내부 지식만으로 해결할 수 없는 질문에 대해서 외부 API와 연동해서 답변 가능

## 5. sLM(small Language Model)

- sLM이란?
    - 소규모 언어 모델은 자연어 콘텐츠를 처리, 이해 및 생성할 수 있는 AI 모델 -> LLM보다 규모와 범위가 작음

    - LLM보다 작고 효율적이어서 메모리와 연산 자원을 적게 사용

- 모델 압축

    - 가지치기: 신경망에서 중요도가 더 낮거나 중복되거나 불필요한 매개변수를 제거 -> 가지치기되는 매개 변수에는 가중치 또는 뉴런자체도 포함

    - 양자화: 고정밀 데이터를 저정밀 데이터로 변환 -> 계산 부하를 줄이고 추론 속도 높임

    - 저차원 분해: 큰 가중치 행렬을 더 작은 저차원 행렬로 분해 -> 복잡한 행렬 연산 단순화 가능

    - 지식 증류: 사전학습된 교사 모델의 학습 내용을 학생 모델로 이전하는 것

- sLM의 장점

    - 성능적 측면: 효과적 성능 / 지연 시간 감소 / 보안에 용이
    - 비용적 측면: 접근성 / 효율성 / 에너지 효율적 / 비용 절감

- sLM의 한계

    - 편향: 더 큰 모델에 존재하는 편향까지 함께 전달 가능 -> 성능 저하로 이어질 수 있음

    - 제한된 일반화 & 복잡한 작업에서의 성능 저하

    - 할루시네이션 발생

- LLM과 sLM의 결합

    - 복잡하고 어려운 작업에는 LLM, 가볍고 간단한 작업에는 SLM을 적절히 배분해 사용
