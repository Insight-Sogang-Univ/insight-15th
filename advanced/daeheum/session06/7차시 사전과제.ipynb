## PART 1. 추천 시스템 기초


### 1. 추천 시스템이란?

- 추천 시스템: 사용자의 정보 데이터를 분석, 개인의 취향에 맞는 아이템을 추천하는 알고리즘

#### 1-1. 사용되는 데이터의 종류와 특징

- 명시적 데이터

    - 고객들에 의해 입력된 데이터, 사용자가 선호도를 직접적으로 표현한 데이터

    - 명백한 데이터를 얻기 쉽지 않음

    - 사용자의 취향이 정확하게 반영 되지만, 데이터 자체를 얻기 힘들다

- 암시적 데이터

    - 고객들의 행동을 기반으로 수집된 데이터 -> 사용자의 선호도가 간접적으로 표현 됨

    - 사용자가 남기는 발자취와 같다.

    - 데이터를 얻기 쉽지만, 사용자의 취향이나 선호도를 직접적으로 알 수 없음


#### 1-2. 추천시스템의 분류

- 내용 기반 필터링

    - 콘텐츠의 유사도를 기반으로 사용자에게 추천

    - 사용자가 이용하거나 선택한 아이템과 유사한 아이템을 추천하는 방식

- 협업 필터링

    - 다수의 사용자로부터 얻은 선호도 정보를 기반으로 사용자에게 추천

        - 두 명의 사용자가 비슷한 관심사를 가지고 있다면, 한 사용자의 데이터를 바탕으로 다른 사용자에게 추천해주는 방식

        - 사용자 기반, 아이템 기반 존재

- 하이브리드 필터링

    - 2가지 이상의 다양한 종류의 추천시스템 알고리즘을 조합하여 만들어진 알고리즘

    - 콘텐츠 기반 필터링과 협업 필터링 각각의 단점 보완 위해 제시됨

### 2. 내용 기반 필터링

#### 2-1. 사용자 프로필

- 추천시스템에 사용되는 데이터

    - 직접 지정형: 사용자에게 자신이 선호하는 아이템의 특징을 명시적으로 지정

    - 간접 지정형: 사용자의 과거 행동 이력에 기반하여 사용자의 프로필을 작성하는 법

- 사용자의 프로필을 획득하는 방식

    - 특성 가중치는 사용자가 가지고 있는 아이템  특성 가중치의 평균 값을 활용

    - 사용자의 기호를 따로 파악하거나, 설문조사 또는 자발적인 키워드 입력: Like/Dislike, Rate

    - 아이템에 대한 유저의 기존 평가를 바탕으로 분류 문제를 해결


#### 2-2 유사도

- 유사도란? -> 추천 시스템을 이해하는데 중요한 수학적 개념

    - 영상, 이미지, 텍스트 등의 형태로 존재하는 아이템 정보 데이터를 벡터화한 이후, 아이템 벡터간의 유사도 측정

    - 측정된 유사도를 바탕으로 고객이 선호하는 아이템과 비교하여, 유사도가 높은 아이템 선별


- TF-IDF(Term Frequency - Inverse Document Frequency)

    - TF: 특정 문서내에 특정 단어가 얼마나 자주 등장하는지를 의미하는 단어 빈도

    - IDF: 전체 문서에서 특정 단어가 얼마나 자주 등장하는지를 의미하는 역문서 빈도

    - 이를 통해 다른 문서에서는 등장하지 않지만, 특정 문서에서만 자주 등장하는 단어를 찾아 문서 내 단어의 가중치 계산

    - TF - IDF 계산 = TF * IDF


#### 2-3 내용 기반 필터링의 장단점

| **장점** | **단점** |
| --- | --- |
| 아이템이 가지고 있는 **feature 자체**에만 집중->다른 유저의 평점 등에 영향을 받지 않음→ 콜드스타트 문제에서 자유로움 | **정확한 feature를 추출하지 못하면 정확도가 떨어짐** |
| item의 feature를 통해서 추천을 하기 때문에 추천이 된 이유를 설명하기 용이 | 기존의 item과 유사한 item 위주로만 추천하기에 **새로운 장르의 item을 추천하기 어려움** |

### 3. 협업 필터링

#### 3-1. 메모리 기반 협업 필터링

- 이웃 기반 협업 필터링 / 사용자 또는 아이템 간 유사도 기반 / 사용자 A와 비슷한 취향을 가진 사용자가 선호하는 아이템 추천

- 사용자 기반: 특정 사용자와 타 사용자 간의 유사도를 측정한 뒤, 가장 유사도가 높은 Top-N 사용자 추출

- 아이템 기반: 사용자들이 그 아이템을 평가한 평가 척도가 유사한 아이템들을 추천하는 기준

- User-based VS Item-based

    - User-based CF

        - 추천 정확성 & 효율성: 아이템 수가 사용자 수보다 많은 환경에서 유리

        - 안정성: 아이템이 수시로 바뀌는 환경에 적합

        - 다양성 -> 특정 사용자가 독특한 아이템 평가시, 뜻밖의 아이템 추천 가능

        - 실제 활용 -> 아이템이 자주 바뀌고 사용자 수가 적은 환경

    - Item-based CF

        - 추천 정확성 & 효율성: 사용자 수가 아이템수보다 많은 환경에서 유리

        - 안정성: 아이템 리스트 변화가 적고 고정적인 환경에 적합

        - 다양성: 유사한 아이테 위주로 추천 -> 상대적으로 다양성, 참신성 부족

        - 실제 활용 -> 사용자 유입이 많은 대형 플랫폼


#### 3-2. 메모리 기반 협업 필터링의 한계

- 희소 행렬 문제

    - 주어진 평점/선호도 데이터 적을시 성능 저하

        - 실제로 유저-아이템 행렬 대부분은 희소 행렬임

        - NBCF를 적용하려면 적어도 sparsity ratio가 99.5%를 넘으면 안됨

- 확장성 문제

    - 아이템/유저의 증가에 따른 확장성 감소

    - 아이템에 대한 평점을 예측하기 위해, i에 대해 평가를 한 모든 유저와의 유사도를 구해야하지만, 유저 증가 -> 연산량 증가 -> 성능 감소

    - 이에 따라 KNN CF 사용

- 콜드 스타트 문제

    - 협업 필터링은 기존의 평점/행동 데이터를 기반으로 유사도 계산하지만, 데이터가 없는 경우 추천 불가능

        - User Cold Start: 신규 유저 유임 - 평가 이력이 없어 유사 유저 탐색 불가

        - Item Cold Start: 신규 아이템 등록 - 아. 묻 평가하지 않아 유사도 계산 불가

    - 이에 직접 지정형 프로필 활용 / Content-Based Filtering / 인구통계학적 정보 활용


#### 3-3. 모델 기반 협업 필터링

- 모델 기반 협업 필터링이란?

    - 단순 유사도를 계산하는 것이 아닌, 데이터로부터 패턴을 학습한 모델을 만들어 평점 예측

    - 관측된 평점 데이터를 학습하여, 관측되지 않은 평점을 예측 하는 것


- 잠재 요인 : 사용자와 아이템 사이의 관계를 설명해주는, 눈에 보이지 않는 숨겨진 특성

- Latent Factor Model: 주어진 데이터로는 알 수 없는 사용자 - 아이템 간의 숨겨진 특성을 찾기 위해 주어진 Matrix

    - 사용자-잠재요인 / 아이템-잠재요인으로 각각 분해하여 학습

    - Factor Model과의 차이점은 축(차원)이 무엇을 의미하는지, 몇개의 차원이 존재하는 지 알 수 없음


- 행렬 분해: 행렬을 분해해 평점을 유저의 잠재 요인과 아이템의 잠재요인으로 설명 가능

    - 입력 행렬: R - 사용자가 모든 노래를 듣거나 평점 주지 않았을 것 -> 희소 행렬

    - 잠재 요인 행렬로 분해:

        - X^(T): 사용자의 잠재 요인 행렬 -> users * factors

        - Y: 아이템의 잠재 요인 행렬 -> factors * songs

## Part 2. 딥러닝 기반 추천 시스템

### 1. VAE 기반 협업 필터링

사용자의 아이템 상호작용 벡터를 입력으로 받아, 잠재 공간에서 재구성하도록 VAE를 학습하는 방식



- 인코더: 사용자 이력을 잠재변수 z 분포로 매핑

- 디코더: z로부터 아이템 분포를 복원

- 장점: 기존 MF처럼 단일 포인트 벡터가 아닌, 확률적 잠재 분포로 사용자 선호를 모델링하여 불확실성 반영 / 희소 데이터에서도 강건


#### 1-1. MF의 한계와 딥러닝의 등장

- MF의 근본적 한계

    - 선형서으이 한계: 평점 예측이 단순 잠재 벡터의 내적으로만 이루어짐 -> 비선형 활성화 함수를 가진 신경망 사용

    - 불확실성 무시: 사용자 취향을 단 하나의 벡터값으로만 표현 -> 확률 분포로 취향을 모델링하여 불확실성 반영

#### 1-2. 왜 AE가 아닌 VAE?

- 딥러닝의 비선형성을 추천 시스템에 도입한 초기 시도 -> 일반 AE

- VAE는 불확실성을 모델링함으로써 희소성에 대한 해답을 제시함

| 구분 | AutoEncoder (AE) | Variational AutoEncoder (VAE) |
| --- | --- | --- |
| **잠재 벡터 표현** | 단일 값 | **확률 분포** (평균 $\mu$, 분산 $\sigma$) |
| **데이터 불확실성** | 반영 불가 | **데이터의 불확실성을 모델링**하여 안정성 확보 |
| **적용 시 장점** | 비선형성 학습 | 희소한 데이터에서 더욱 안정적이며, 일반화 성능 우수 |

#### 1-3. VAE 기반 CF의 작동 구조

- VAE 기반 CF는 사용자의 아이템 상호작용 벡터를 입력으로 받아, 잠재 공간에서 재구성하도록 학습

- 입력: 사용자의 아이템 상호작용 벡터

- 인코더: 사용자의 이력을 잠재변수 z분포로 매핑

- 잠재 벡터: 샘플링된 최종 취향 벡터

- 디코더: z로부터 아이템 분포를 복원


#### 1-4. VAE: 목적 함수와 KL Divergence

- VAE의 핵심은 목적함수인 ELBO에 있으며, 아래의 목표를 동시에 최적화

    - 재구성 오차 최소화(첫째 항)

        - 디코더가 잠재 벡터 z를 사용하여 원래 입력 x를 최대한 비슷하게 복원

    - KL Divergence 항(둘째 항)

        - 인코더가 출력하는 잠재 분포가 사전 분포와 유사하도록 강제

        - AE의 문제점: 데이터가 희소한 유저 A,B의 잠재 벡터가 불규칙한 공간에 놓여 있어, 이들 사이를 연결하기 어려움

        - VAE의 강점: KL Divergence가 잠재 공간을 정규 분포 근처로 압축하여 부드럽게 만듬


    - VAE - CF의 주요 장점

        - 확률적 잠재 분포 모델링: 확률적 잠재분포로 사용자 선호 모델링 -> 불확실성 반영 및 노이즈에서 자유로움

        - 희소한 상호작용 데이터에서도 살아남음: KL Divergence를 통해 잠재 공간을 정규화해, 데이터 연결 능력이 뛰어남

        - 비선형성 학습: 신경망 구조를 통해 복잡하고 비선형적인 패턴 학습

### 2. GNN 기반 사용자 선호 예측

#### 2-1. GNN

- GNN은 사용자와 아이템의 상호작용을 누가 무엇을 좋아했다의 집합으로 보는 것을 넘어, 서로 연결된 거대한 관계망을 형성

- Graph = Node와 Edge를 모아 구성한 자료구조

    - 관계, 상호작용과 같은 추상적인 개념을 다루기에 적합

    - Non-Euclidean space도 표현하고 학습 가능


#### 2-2. GNN의 구조

- GNN 작동 구조

    - 노드는 개별 객체, 엣지는 객체 간 관계나 연결성 의미

    - 메시지 패싱: 각 노드가 이웃 노드로부터 정보를 받아 전달

        - 메시지 생성: 이웃 노드로에 전달할 메시지

        - 집계: 받은 여러 이웃의 메시지를 하나로 합침

        - 갱신: 자신의 노드 상태를 업데이트


#### 2-3. GNN의 프로세스

- GNN은 그래프를 입력 받아 메시지 패싱 수행, 각 노드는 자신의 이웃 노드들의 정보를 수집해 자신의 임베딩 업데이트


- a. 그래프 구조 생성

    - 노드와 엣지 정의 후, 각 피처를 담아 표현하는 것

- b. 메시지 패싱

    - 그래프에서 노드간 정보를 전달하고 상호작용하는 방법

    - 메시지 생성 - 집계 - 갱신의 세단계로 이루어짐

    - 각 노드는 이웃 노드에만 메시지 전달

    - 메시지 생성

        - 각 노드가 이웃에게 보낼 자신의 정보를 생성

        - 이우셍게 보내는 메시지는 임베딩 되어 있음

        - 노드는 연결된 이웃마다 전달할 메시지를 생성하기 위해 메시지 함수 사용

    - 집계

        - 메시지 생성 이후, 여러 이웃 노드에서 메시지를 하나의 벡터로 통합: Aggregation

    - 갱신

        - 집계된 이웃 정보와 자신의 이전 상태를 결합해 새로운 임베딩 계산

        - 이 과정을 반복하여, 더 멀리 있는 이웃의 정보까지 반영


#### 2-4. LightGCN

- Light GCN

    - 추천 시스템에 특화된 GNN

    - 특징 변환 제거: 레이어마다 복잡했던 학습 파라미터 삭제

    - 비선형 활성화 함수: ReLU와 같은 활성화 함수를 사용하지 않고, 이웃 정보를 합침

    - Normalized 이웃 임베딩의 합으로만 구성


#### 2-5. GNN 활용 사례

- Pinterest, 핀테크(송금 네트워크 자체를 분석), B2C 비즈니스


#### 2-6. GNN의 장단점

- 장점

    - 고차원 협업 시그널: 단순히 나의 잠재요인 분석을 넘어, 나와 취향이 비슷한 사용자들과 내가 본 아이템과 연관된 다른 아이템들의 정보를 직접 반복

    - 정형화되지 않은 데이터 학습 기능: GNN은 관계를 학습하기 때문에, 각 노드가 정형화된 데이터로 정의되지 않아도 관계망 속에서 정보를 효과적으로 전파

    - Cold Start 문제 완화에 효과적

- 단점

    - 과평탄화 문제: 모든 노드의 임베딩이 같은 값으로 수렴하는 현상

    - 과압축 문제: GNN의 층이 깊어질수록 집계하는 이웃의 범위가 기하급수적으로 커짐 -> 하나의 벡터에 압축 -> 정보 손실

    - 계산 복잡도 증가: 실제 세계에서 막대한 GNN을 적용해야 하므로, 막대한 자원과 메모리 요구

    - 해석의 어려움: 순서 반영 불가

### 3.Transformer 기반 추천 시스템

#### 3-1. 기존 추천시스템의 순차적 한계

- CF의 순차적 한계: 순서 무시, 문맥 부족, 예측의 한계

- GNN의 순차적 한계: RNN과 같은 장기 의존성 문제 지님

    - 지역적 정보 전파의 한계

    - 장기 의존성 문제

#### 3-2. Transformer: 사용자 행동을 순서로 접근

- 행동 순서가 중요한 이유: 의도의 파악 -> Tansformer은 시퀀스 내의 문맥과 핵심 의도를 파악하는데 강력

    - 문맥에 따라 달라진는 아이템의 의미

    - 시간 순서를 뛰어넘는 핵심 의도의 포착

#### 3-3. Transformer 기반 주요 추천 시스템 모델


- SASRec

    

    - 단방향 어텐션으로 과거 정보만 활용하여 미래 행동을 추측하는 방식으로 훈련을 받은 추천 시스템 모델

    - 현재의 아이템은 오직 과거의 아이템들만 참조하여 어텐션 연산 가능 -> 시간 흐름을 명확하게 반영

    - 문제점: 계산량 많음 / 단방향성 존재

- BERT4Rec

    

    - 시퀀스의 일부를 MASK 토큰으로 가린후, 이 아이템이 과거와 미래의 모든 아이템을 참조하는 양방향 어텐션 연산

    - 학습 과정에서 전체 선호도 파악하는데 초점이 맞춰져 있음

### 4. 제언: 오늘나르이 추천 시스템

#### 4-1. LLM의 시대

- 관점의 전환: 추천을 랭킹 문제가 아닌 언어 문제로 푼다

    - 과거: 조합별로 미리 학습된 공식에 따라 점수 출력

    - LLM: 추천을 점수 계산이 아닌 문장으로 출력

        - ID가 아닌 의미 기반 추천 / 멀티모달 데이터 활용

- 새로운 가능성: 대화형 추천 / Zero-shot 추천(새로운 아이템도 아이템의 텍스트 설명 만으로 추천 가능)

- P5 / TALLRec

    - LLM이 풀 수 있도록 하나의 거대한 텍스트 프롬프트 문제로 통일시킨 모델

        - 기존 추시 -> 모든 정보를 벡터로 임베딩 한 뒤, 유사도 측정 -> 엄청난 컴퓨팅 자원과 시간 소모

        - TALLRec -> 이미 만들어진 범용 LLM을 추천 작업에 맞게 효율적으로 튜닝하는 프레임 워크


#### 4-2. 인기 편향과 XAI 문제

    - 인기 편향: 추천 시스템이 소수의 인기 아이템을 과도하게 노출 but 다수의 비인기 아이템은 관심을 못받음

    - XAI: 추천한 이유에 대해서 설명할 수 없음


#### 4-3. 컴퓨팅 자본, 콜드 스타트 문제를 고려한 추시 도입

    - 비용 문제: 딥러닝을 전면 도입하기엔 너무 비쌈 -> 선형 기반 추천 시스템과 병행

        - 1단계: 후보군 생성 -> 사용자가 조금이라도 좋아할 만한 아이템 골라냄(선형 모델 사용)

        - 2단계: 순위 매기기 -> 후보군 중 사용자가 가장 좋아할 순서대로 정교하게 순위를 매김(딥러닝 모델 사용)
