# 1. NLP와 언어 모델
## (1) NLP (Natural Langauge Processing)
- NLP의 역할: 이 모호한 '자연어'를 컴퓨터가 이해할 수 있는 '숫자(데이터)'로 변환하여 유의미한 가치를 창출하는 기술.

### NLP 과정
1. 데이터 수집 (Data Collection)
2. 텍스트 전처리 (Text Preprocessing)
3. 텍스트(를 수치로) 표현 (Text Representation)
4. (목적에 맞게) 모델 학습 및 예측 (Modeling)
5. 후처리 및 응용 (Postprocessing & Application)

### 자연어의 특징
1. 이산적(연속적의 반대말) 데이터 (Discreteness) : 그래서 NLP(이산적 단어 -> 연속적 벡터)로 변환해야 함
2. 순차적 데이터 (Sequentiality) : 순서의 영향을 받음
3. 중의성과 모호성 (Ambiguity) : 동일 단어도 문맥에 따라 뜻 달라짐
4. 희소성 (Sparsity) : 우리가 가진 데이터에 나타나지 않는 문장이 들어와도 추론할 수 있는 일반화 능력 중요
5. 5. 가변성과 진화 (Variability & Evolution) : 언어는 고정되어 있지 않으므로 유연하게 패턴을 학습하는 머신러닝 방식 필수

## (2) 언어 모델
- 언어 모델은 이전 단어들이 주어졌을 때 다음 단어를 예측하도록 훈련되어, 단어 시퀀스(문장)에 확률을 할당하여 가장 자연스러운 단어 시퀀스를 찾아내는 방식으로 인간이 말하고 쓰는 방식을 모방하는 시스템이다.
- 통계 기반 vs 인공 신경망 기반 으로 나뉨.

# 2. 통계적 언어 모델 (SLM)
## (1) Statistic Language Model이란?
- 통계 기반 언어 모델(SLM)은 분포 가설(비슷한 문맥에서 함께 나타나는 단어들은 비슷한 의미를 가진다는 가설)이라는 아이디어에 근거함
- 조건부 확률 : 다음으로 더 높은 확률로 올 수 있는 단어를 선택함
- 카운트 기반 : 이전 단어 시퀀스의 등장 빈도를 통해 다음 단어의 확률을 계산 → 희소문제(모델이 이전에 경험한 적 있는 조합에 대해서만 잘 예측할 수 있는 문제)가 있음
- SLM은 새로운 문장 조합에 약하다 → 그래서 새로운 문장 조합에도 적용할 수 있는 신경망 모델이 필요합니다

## (2) N-gram
- SLM과 비슷하지만, 전체 문장 대신에 문장의 일부(N-1개 단어)만 보고 예측
- 한계 : 희소 문제(SLM과 동일), 문맥 제한(N-1개 단어만 고려해 앞부분 반영X)

## (3) Perplexity(PPL)
- PPL: 언어 모델이 문장을 얼마나 "혼란스러워하는지(perplexed)"를 수치로 나타낸 것
- PPL이 낮다 = 모델이 예측에 덜 혼란스러움 = 모델이 문맥을 잘 이해함 = 모델의 성능이 좋음
- 신경망 기반 언어 모델은 단어와 문맥 간의 관계를 학습하여 일반화하는 능력이 뛰어남 → 이에 따라 신경망 언어 모델 쪽에 관심이 쏠림!

# 3. 신경망 언어 모델 (NLM)
- 통계적 언어 모델 한계 극복
- 신경망 언어 모델의 키워드는 임베딩 : 토큰화된 각 단위는 임베딩 단계에서 숫자 벡터로 변환됨
- 통계적 모델(SLM)이 단어를 단순히 '이름'으로만 기억했다면, 신경망 모델(NLM)은 모든 단어를 거대한 다차원 지도 위의 한 점(좌표)으로 보냅니다. 이것을 워드 임베딩(Word Embedding)이라고 합니다.
- 핵심 포인트: NLM은 단어를 '기호'에서 '의미를 가진 숫자(벡터)'로 바꿈으로써, 데이터에 없는 문맥도 어느 정도 예측할 수 있는 일반화 능력을 얻게 되었습니다.

# 4. 딥러닝 기반 언어모델
## (1) Transformer의 등장
- RNN과 LSTM의 한계(병렬화 불가, 기울기 소실 및 폭주, 장기 의존성 문제) 극복
- 인코더와 디코더가 여러 Layer로 구성됨
- 이러한 방식으로 신경망을 깊게 쌓으면서, 언어 모델의 표현력(capacity)이 높아져 현재의 딥러닝 기반 언어모델을 만들 수 있었습니다. 
- 전이학습(NLP의 표준 학습 방법론) : ① 사전 학습 (Pre-training) -> ② 미세 조정 (Fine-tuning)

## (2) BERT
###  1️⃣ BERT가 뭐예요?
- BERT = **Bidirectional** **Encoder** Representations from **Transformers**
- 텍스트의 앞뒤 문맥을 동시에 고려해서 / 입력값을 벡터로 바꿔주는 / 딥러닝 모델이다.
- 양방향이니까 단어의 좌우 문맥을 동시에 보고 이해, 이때 BERT는 encoder 기반이고 뒤에서 나오는 GPT는 decoder 기반임
- 트랜스포머 기반: 트랜스포머 구조의 encoder 부분만 사용
- 사전 학습 + 파인 튜닝

###  2️⃣ 그래서 BERT는 어떻게 학습했나요?
- BERT의 기본 구조는 transformer의 encoder를 여러 개 쌓아 올린 구조
- [처리 과정] : input → Embedding → 트랜스포머 인코더 12개(or 24개) → Output
- BERT는 입력 문장을 embedding한 뒤, 여러 개의 encoder를 통과하면서 각 단어의 의미를 문맥적으로 풍부하게 만들어요.
#### [사전학습]
##### (1) Masked Language Model(MLM)

- 빈칸 채우기 방식 학습 → 자연스럽게 문맥을 이해하고 추론하는 능력 향상
##### (2) Next Sentence Prediction(NSP)
- 두 문장에 대해, 이 문장들이 서로 앞뒤 관계인지, 아니면 그냥 무작위 위치에 있는 두 개의 문장인지를 학습 → 두 문장이 이어지는 논리 흐름을 학습

### 3️⃣ BERT 이후의 모델은 무엇이 있나요?
(1) RoBERTa: BERT의 사전 훈련 프로세스를 개선하여 성능을 향상시킨 실전형 모델  
(2) ALBERT (A Lite BERT): BERT의 경량화 버전으로, 가볍지만 성능은 유지

## (3) GPT
### 1️⃣ GPT는 뭐예요?
- BERT는 transformer의 encoder 기반이었고, GPT는 decoder 기반임
- 마스킹을 이용하여, GPT도 지금까지의 단어를 바탕으로 다음 단어를 예측
- GPT3부터 few-shot(몇 개의 예시를 보고 task 수행) learning 도입
- zero-shot: 예시 없이 문제 지시만 주고 task 수행, one-shot: 1개의 예시만 보고 task 수행

## (4) LLM(Large Language Model, 거대 언어 모델)
- ex) ChatGPT,LLaMA, PALM, Bard, Gemini 등
- 위 BERT, GPT에 이어서 LLM이 등장했는데요, LLM은 아래와 같은 특징을 가지고 있습니다.
1. Scaling Law: "크기가 곧 실력이다"
2. 창발적 능력(Emergent Abilities): "배운 적 없는 일을 하다"
3. RLHF: "똑똑한 AI를 '착한 AI'로 길들이기"
4. 패러다임의 전환: "튜닝(Tuning)에서 프롬프트(Prompt)로"

### [LLM의 작동 방식]
- LLM은 크게 '학습'과 '예측'이라는 두 가지 핵심 단계를 통해 작동함
   
(1) 대규모 학습 (Training) : 방대한 데이터를 학습, 언어의 패턴을 스스로 터득  
(2) 예측 및 생성 (Prediction & Generation) : 다음에 올 가장 확률 높은 단어를 순서대로 예측

-  LLM은 BERT가 아닌 GPT 방식을 사용합니다. 즉, decoder를 기반으로 해요. 그 이유는 GPT 구조가 문장 생성에 훨씬 유리하기 때문입니다.

### [LLM의 한계]
hallucination, 업데이트 비용

## (5) RAG
### 1️⃣ RAG의 필요성
- 외부에서 최신 정보를 가져와서 답변하게 하기 위한 것이 RAG (Retrieval Augmented Generation)입니다! -> 질문에 대해 모델이 모르면 검색해서 답하는 구조

### 2️⃣ RAG의 구조
  
- 질의 인코더(Query encoder): 사용자의 질문을 벡터로 바꿉니다.
- 지식 검색기(Knowledge Retriever): 인코딩된 질문을 바탕으로 외부 지식에서 관련 내용을 검색합니다.
- 지식 증강 생성기(Knowledge Augmented Generator): 검색한 내용을 참고해서 답변을 생성합니다.

### 3️⃣ RAG의 장점
최신 정보 반영 가능, hallucination 문제도 감소, 정보 정확성도 증가

### 4️⃣ CAG
- Credibility-Aware Generation
- 기존 RAG : 검색된 문서가 신뢰도가 낮거나 잘못된 정보일 경우 생성된 답변 품질과 정확도 하락
- CAG는 모델이 문서를 단순히 활용하는 수준을 넘어 문서의 신뢰도를 스스로 판단하고 활용할 수 있도록 훈련시키는 기법
- [CAG의 구조]
    - 질문 입력 → 정보 검색 → 각 정보의 신뢰도 평가 → 높은 신뢰도 정보 우선 사용 → 최종 답변 생성

## (6) LangChain
### 1️⃣ LangChain
LLM+RAG 모델 pipeline을 쉽게 구성할 수 있도록 도와주는 역할을 하는 것이 LangChain 프레임워크
- LLM을 기반으로 앱 제작 시 필요한 구성 요소를 쉽게 연결
- 파이썬과 JavaScript 라이브러리 제공 → 다양하게 활용 가능

[LangChain의 특징]
- 추상화: LLM 서비스를 만들 때 필요한 각종 작업을 간결하게 표현(간소화)
- 표준화: 비슷한 기능을 갖추고 있는 요소들을 똑같은 형식을 갖춘 컴포넌트로 표준화
- 체이닝: 자주 활용하는 주요 컴포넌트를 쉽게 연결해서 LLM 서비스의 로직을 쉽게 파악 (Input → 컴포넌트 → Output)

[LangChain의 한계]
LangChain은 선형적인 체인 구조를 주로 사용하여, 중간에 조건에 따라 다른 작업을 수행하거나, 결과가 마음에 들지 않을 때 이전 단계로 돌아가 다시 실행하는 등의 복잡한 로직을 구현하기 어렵습니다.

### 2️⃣ LangGraph
- LangChain의 한계 극복
- LangGraph는 비선형적인 작업 형태로, 사이클 (Cycle), 루프 (Loop)를 만들 수 있습니다.
- 노드 (Node): 특정 작업을 수행하는 단위로, 랭체인의 컴포넌트를 그대로 가져와 노드로 사용 할 수 있습니다.  →  ‘무엇을 할 것인가’
- 엣지 (Edge): 노드와 노드를 연결하는 선으로, 어떤 노드에서 다음 노드로 이동할지를 결정합니다. (조건부 엣지 - "만약 A 조건이면 B 노드로, C 조건이면 D 노드로 가라" 와 같은 분기 로직(if-else문)을 쉽게 구현할 수 있습니다.) → ‘다음에 어디로 갈 것인가’

### 3️⃣ LangSmith
- LangChain 프레임워크를 만든 팀에서 개발한, 대규모 언어 모델(LLM) 기반 애플리케이션을 위한 통합 개발 및 운영 플랫폼
- [핵심 기능] : 디버깅 및 추척, 테스트 및 평가, 모니터링, 프롬프트 허브

### 4️⃣ 외부 API 활용하기
LLM 내부 지식만을 이용해서 해결할 수 없는 질문에 대해서는 외부 API와 연동해서 답변할 수 있습니다.



# 5. sLM (small Language Model)
### 1️⃣ sLM이란?
- 소규모 언어 모델(sLM)은 자연어 콘텐츠를 처리, 이해 및 생성할 수 있는 AI 모델입니다. 이름에서 알 수 있듯이 sLM은 대규모 언어 모델(LLM)보다 규모와 범위가 작습니다.
- sLM은 LLM보다 더 작고 효율적이어서 메모리와 연산 자원을 적게 사용합니다. 따라서 모바일 앱처럼 리소스가 제한된 환경, 또는 데이터 네트워크 없이 사용자 쿼리에 대한 오프라인 추론을 통해 응답을 생성해야하는 상황에 적합합니다.

### 2️⃣ 모델 압축
(1) 가지치기 : 가지치기는 신경망에서 중요도가 더 낮거나 중복되거나 불필요한 매개변수를 제거  
(2) 양자화 : 양자화는 고정밀 데이터를 저정밀 데이터로 변환, 계산 부하를 줄이고 추론 속도를 높일 수 있음  
(3) 저차원 분해 : 큰 가중치 행렬을 더 작은 저차원 행렬로 분해하는 것, 복잡한 행렬 연산을 단순화  
(4) 지식 증류 : 사전 학습된 "교사 모델"의 학습 내용을 "학생 모델"로 이전하는 것, 학생 모델은 교사 모델의 예측과 일치시키고 기본 추론 과정을 모방하도록 학습

### 3️⃣ sLM의 예
DistilBERT, Gemma, GPT-4o mini, Llama, Ministral, Phi 등이 있습니다!

### 4️⃣ sLM의 장점
- 성능적 측면 : 효과적인 성능, 지연 시간 감소, 보안
- 비용적 측면 : 접근성, 효율성, 에너지 효율, 비용 절감

### 5️⃣ sLM의 한계
편향까지 큰 모델로부터 학습 가능, 제한된 일반화 & 복잡한 작업에서의 성능 저하, 할루시네이션(환각)

### 6️⃣ LLM과 sLM의 결합 (결합 추론)
  
복잡하고 어려운 작업에는 LLM, 가볍고 간단한 작업에는 SLM을 적절히 배분해서 사용합니다. 이를 통해 LLM 수준의 높은 정확도를 달성하면서도 LLM 토큰 사용량을 최대 90%까지 절약할 수 있습니다.
