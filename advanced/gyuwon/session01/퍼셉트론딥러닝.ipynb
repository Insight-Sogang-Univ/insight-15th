## 1. 인공신경망의 시작 : 퍼셉트론

### 1-1 인공신경망이란?(Artificial Neural Network)

#### ① 인공신경망

> 우리 뇌의 신경망(뇌를 구성하는 신경 세포끼리 서로 연결되어 정보를 처리하는 구조)을 모방한 모델

생물학적 뉴런이 정보를 처리하는 구조를 본 뜸

1. 입력($x_1, x_2, …, x_n$)들을 받아 각각 **가중치**($w_1, w_2, … ,w_n$)와 **편향**($b$)을 적용함
2. 종합한 값을 활성화 함수에 넘깁니다. 이때 활성화 함수는 임계값을 기준으로 함
    - 기준 미달 → **0**에 가깝게 죽여서 내보냄
    - 기준 통과 → **유의미한 값**으로 살려서 내보냄
3. output이 출력됨

####  ② 퍼셉트론

> 초기 형태의 인공신경망. 신경망(딥러닝)의 기원이 되는 선형 이진 분류기(Minear Binary Classifier)

##### <퍼셉트론의 출력값을 구하는 절차>

> 입력 -> 가중합 -> 활성화 함수 -> 출력

1. 입력값과 가중치를 곱함

2. 곱한 값들의 총합을 구함

3. 총합이 임계값(𝜃)을 넘으면 1, 넘지 않으면 0을 출력

* 활성화 함수 : 임계값(𝜃) 초과 여부에 따라 출력값을 결정(사용하는 활성화 함수에 따라 출력값이 달라질 수 있음)

* Training(학습) : 원하는 출력값을 내보내도록 가중치를 조정해가는 작업

#### ③ 용어 정리

1. 가중치(weight) 

    - 입력값이 출력값에 주는 영향력(중요도)을 결정

    - 각 입력값에는 각각의 가중치가 존재

    - 가중치의 값이 클수록 해당 입력 값이 중요함을 의미

2. 활성화 함수(activiation function)

    - 신호가 일정 이상의 값(임계값 𝜃)을 넘으면 1, 넘지 못하면 0을 출력하는 퍼셉트론의 성질은 활성화 함수로 **계단 함수** 를 이용

    - 이후 등장한 여러가지 발전된 신경망들은 계단 함수 외에도 여러 다양한 활성화 함수를 이용(Sigmoid, Tanhm, ReLU, leaky ReLU)

3. 임계값(Threshold, 𝜃)

    - 활성화 함수에 사용되는 값

    - $$
        \begin{aligned}&\text{if } \sum_{i}^{n} w_i x_i \ge \theta \rightarrow y=1 \\&\text{if } \sum_{i}^{n} w_i x_i < \theta \rightarrow y=0\end{aligned}
    $$

    - 계단 함수에서는 가중합이 임계값을 넘으면 1, 넘지 못하면 0을 출력

4. 편향(bias)

    - 계단함수의 식에서 임계값을 좌변으로 넘기면 편향(b)으로 표현 가능

    - $$
        \begin{aligned}&\text{if } \sum_{i}^{n} W_i x_i + b \ge 0\rightarrow y=1 \\&\text{if } \sum_{i}^{n} W_i x_i + b < 0\rightarrow y=0\end{aligned}
        $$

    - 그림으로 표현하면 입력값 1에 편향 b 가 곱해진 형태로 표현됨

        

    - 임계값과 편향은 표현의 차이!

### 1-2 단층 퍼셉트론과 그 한계(Single-Layer Pereptron)

#### ① 단층 퍼셉트론

> 단순히 입력층과 출력층으로만 이루어진 퍼셉트론 -> 값을 보내는 단계 + 값을 받아서 출력하는 단계로만 이루어짐

#### ② 논리 게이트(Logic Gate) 구현

> 디지털 회로에서 기본적인 논리 연산을 수행하는 데 사용되는 구성 요소

* 게이트 연산에 쓰이는 것은 n 개의 입력값과 하나의 출력값

* AND, OR, NOT, NAND, NOR, XOR 등 다양한 종류가 있으며 이러한 논리 게이트를 조합하여 복잡한 논리 회로 구성 가능

#### ③ 대표적인 논리 게이트

*  AND 게이트 : 입력값이 모두 1인 경우에만 출력값이 1이 나오는 구조

*  OR 게이트 : 입력값이 모두 0인 경우에 출력값이 0이고, 나머지 경우에는 모두 출력값이 1인 구조

* NAND 게이트 : 입력값이 모두 1인 경우에만 출력값이 0, 나머지 경우에는 모두 출력값이 1인 구조

* XOR 게이트 : 입력값 두 개가 서로 다른 값을 갖고 있을 때에만 출력값이 1인 구조

#### ④ 단층 퍼셉트론의 한계

> 퍼셉트론은 입력값과 가중치의 선형 결합으로 이루어진 구조이기에 단층 퍼셉트론은 직선 하나로 두 영역을 나눌 수 있는 선형 분류 문제에 대해서만 구현 가능 -> **XOR 게이트를 구현할 수 없다** 

### 1-3 다층 퍼셉트론(Multi-Layer Perceptron, MLP)

기존의 AND, NAND, OR 게이트를 조합하면 XOR 게이트를 만들 수 있음

* 중간에 층을 추가하여(입력층 + 하나 이상의 은닉층 + 출력층)으로 구성함

* XOR 문제보다 더욱 복잡한 문제를 해결할 땐 중간에 수많은 은닉층을 추가할 수 있음

* 두 개 이상의 은닉층을 갖는 MLP 가 딥러닝의 시초 모델로 발전함

    - 심층 신경망(Deep Neural Network, DNN) : 은닉층이 2개 이상인 신경망

    - 딥러닝 : 심층 신경망을 학습(가중치를 스스로 찾도록) 시킴

## 2. 딥러닝 모델은 어떻게 학습하는가

### 2-1 순전파와 역전파(Forward & Backward Propagation)

#### ① 학습 알고리즘의 기본 구조


1. 순전파를 통해 입력값으로부터 예측 수행(시험 문제를 열심히 푼다)

2. 오차 계산을 통해 예측 정확도 측정(내가 쓴 답과 실제 정답을 비교해서 예측 정확도 측정)

3. 역전파를 통해 출력층에서 다시 입력층 방향으로 오차 전달(정답을 보고 다시 공부를 수행)

4. 가중치 업데이트(똑똑해진 머리)

5. 1~4 반복

####  ②  연쇄 법칙 (Chain Rule)

> 합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다는 법칙

* 입력값이 최종 출력에 얼마나 영향을 미쳤는지 계산하려면 각 층이 다음 층에 미친 영향을 순서대로 곱해 나가야함 -> 각 단계가 연쇄적으로 결과에 얼만큼 영향을 줬는지 알 수 있음

* 연쇄 법칙이 필요한 이유 : 딥러닝 모델을 학습시키려면 오차를 줄여야 하는데 이를 위해 역전파 과정에서 각 가중치가 최종 손실에 얼마나 기여했는지를 계산해야 하는데 이 때 연쇄법칙이 사용됨

* 딥러닝 모델에는 은닉층이 여러 개 존재하는데 각 은닉층마다 손실에 주는 영향을 곱해서 최종 손실에 대한 값을 나타냄

    - A -> B -> C 의 3층 구조에서 (A→C의 영향) = (B→C의 영향) × (A→B의 영향)

* 연쇄법칙을 통해 구하는 기울기 = '손실 함수를 줄이기 위해 가중치를 어떻게 바꿀까?'를 알려주는 신호

#### ③ 기울기 소실 문제 (Vanishing Gradient Problem)

> 은닉층이 매우 많은 경우에 각 단계마다 미분값이 1보다 작은 값이면 한없이 0에 가까워지게 됨 -> 앞쪽 층의 가중치가 거의 업데이트되지 않아서 학습이 제대로 이뤄지지 않는 문제 생김

#### ④ 전체적인 흐름 정리

1. 순전파 = 예측하기 

    - 입력 x → [가중치 w1] → [가중치 w2] → 예측값 ŷ

2. 손실계산 = 얼마나 틀렸는지 측정하기 

    - 손실 L = (예측값 ŷ - 실제값 y)²

3. 역전파 = 책임 추적하기

    - 손실 L ← [가중치 w2의 책임은?] ← [가중치 w1의 책임은?]

    - w1 책임 = w1이 w2에 준 영향 × w2가 L에 준 영향 -> 연쇄법칙

4. 가중치 업데이트 = 수정하기 

    - 4번 단계의 반복으로 모델의 정확도가 올라감

### 2-2 손실 함수 (Loss Function)

####  ① 손실 함수란?

> 모델의 예측이 얼마나 틀렸는지를 수치로 표현하는 방법 -> 실제값(정답)과 예측값(출력값)의 차이를 계산하는 데에 사용됨

####  ②  손실 함수의 종류

##### 1. 평균 제곱 오차(Mean Squarred Error, MSE)

$$
\frac{1}{N}\sum(y-\hat{y})^2 \ \ \ \ \ \ (y: 실제값,\ \ \hat{y} : 예측값)
$$

* 예측 값이 연속적이고, 예측값과 실제값 간의 차이를 직접적으로 측정하는 회귀 문제에 적합함

##### 2. 이진 크로스 엔트로피(Binary Cross-Entropy Loss, BCE)

$$
L= -(y \log \hat{y} + (1-y) \log(1-\hat{y}))
$$

$$
y : 실제값(0,1)이진값,\ \ \hat{y} : 예측값(확률)
$$

* 이진 분류 문제(로지스틱 회귀)에서 사용하는 손실함수

    - 실제 정답이 1일 때

        - 예측 확률이 1에 가까울수록 → loss값은 0에 가까워짐 = 잘 맞힘
        
        - 예측 확률이 0에 가까울수록 → loss가 급격히 증가 (확신했지만 틀림)

    - 실제 정답이 0일 때

        - 예측 확률이 0에 가까울수록 → loss값은 0에 가까워짐 = 잘 맞힘
        
        - 예측 확률이 1에 가까울수록 →  loss가 급격히 증가 (확신했지만 틀림) 

* MSE 와의 차이 : MSE 는 단순히 loss 의 크기만 보는 반면, BCE 는 얼마나 확신을 가지고 예측했는지도 loss 에 반영함


##### 3. 크로스 엔트로피 오차(Cross Entropy Error, CEE)

$$
-\sum y \log \hat{y}
$$

$$
y : 원-핫 인코딩\ 값 (1,0),\ \ \hat{y} : 예측값(확률)
$$

* 다중 클래스 분류 문제에서 예측 확률이 실제 클래스와 얼마나 일치하는 지를 평가하는 데 적합함

* 두 확률 분포(진짜 분포/예측 분포)의 유사도를 측정하는 데에 유용

* BCE 와 거의 유사하지만 다중 분류에서 사용

| **손실 함수** | **사용 상황** | **표현식** | **예시** |
| --- | --- | --- | --- |
| **MSE** | 회귀 문제 | $L = \frac{1}{n} \sum_{i=1}^{n}(y_i - \hat{y}_i)^2$ | 주택 가격 예측, 온도 예측, 주식 가격 예측 |
| **BCE** | 이진 분류 문제 | $L = -(y \log(\hat{y}) + (1 - y) \log(1 - \hat{y}))$ | 이메일 스팸 필터링(스팸/정상), 질병 진단(양성/음성) |
| **CEE** | 다중 클래스 
분류 문제 | $L = -\sum_{i=1}^{C} y_i \log(\hat{y}_i)$ | 손글씨 숫자 인식(MNIST), 이미지 분류(CIFAR-10) |

### 2-3 활성화 함수 (Activation Function)

####  ① 활성화 함수란?

신경망의 각 층은 

> 입력값 × 가중치 + 편향  →  활성화 함수  →  출력값

의 계산을 하는데 이때 활성화 함수가 없으면 층을 아무리 많이 쌓아도 결국 선형 계산의 반복에 불과하기에 앞서 봤던 XOR 같은 복잡한 패턴을 학습할 수 없음

=> 활성화 함수는 각 층의 출력값에 비선형성을 추가해서 신경망이 복잡한 패턴도 학습할 수 있게 해주는 함수

|  | **단층 퍼셉프론** | **다층 퍼셉트론** |
| --- | --- | --- |
| **활성화 함수** | 계단 함수 | Sigmoid, ReLU 등 |
| **출력값** | 0 또는 1 (이진 출력) | 연속적인 값 (확률 또는 실수) |
| **특징** | 선형 문제만 풀 수 있음 | 복잡한 비선형 문제도 학습 가능 |

####  ② 계단함수 (Heaviside Step Function)

$$
h(x) = \begin{cases} 1 & (x \geq 0) \\ 0 & (x < 0) \end{cases}
$$

> 입력값과 가중치의 곱을 모두 더한 뒤, 그 합이 특정 기준(임계값)을 넘으면 1, 아니면 0을 출력하는 함수 -> 주로 단층 퍼셉트론에서 사용됨

* 입력 신호 : 입력값과 가중치의 곱

* 출력 신호 : 0 또는 1

####  ③ 시그모이드 함수

$$
f(x)=\frac{1}{1+e^{−x}}
$$

* 계단 함수와의 차이점 : 계단 함수는 0 또는 1만 출력하는 반면, 시그모이드 함수는 x 값에 따라 출력값이 연속적으로 변화함 -> 확실히 맞다/틀리다 가 아닌 얼마나 확실한 지 표현 가능

#### ④ ReLU 함수

$$
f(x)=max(0,x)
$$

* 입력이 0보다 크면 그대로, 0 이하면 0을 출력하는 비선형 함수

* 딥러닝에서 가장 많이 사용됨

    - 계산이 간단하고 효율적호(단순한 비교 연산)

    - 비선형성 제공(은닉층에 비선형성 추가)

    - 기울기 소실 문제 완화(양수 구간에서 미분값이 항상 1임)

#### ⑤ Tanh(쌍곡탄젠트)함수

$$
f(x)=tanh(x)=\frac{e^x-e^{−x}}{e^x+e^{−x}}
$$

* 시그모이드 함수의 변형으로, 출력값의 범위가 -1 ~ 1

* 시그모이드(0~1)와 달리 출력값의 중심이 0이기 때문에, 입력값의 중심을 0으로 맞추고 싶을 때 사용 -> 중심이 0이면 학습 과정에서 가중치 업데이트가 더 안정적으로 이루어질 수 있음

* 시그모이드와 마찬가지로, 입력값이 매우 크거나 작으면 미분값이 0에 가까워져 기울기 소실 문제가 발생할 수 있음

#### ⑥ Softmax 함수

$$
f(x_i)=\frac{e^{x_i}}{∑_je^{x_j}}
$$

* 각 클래스에 대한 출력값을 확률로 변환하는 함수

* 다중 클래스 분류 문제의 출력층에서 사용 가능(출력값이 모두 0~1 사이이고, 전체 합이 1이 되기 때문에 각 클래스에 속할 확률을 직접 읽을 수 있음)

| 활성화함수 | 출력 범위 | 주요 사용처 | 장점 | 단점 |
| --- | --- | --- | --- | --- |
| 계단 함수 | 0 또는 1 | 단층 퍼셉트론 | 구조가 단순 | 미분 불가 |
| Sigmoid | 0 ~ 1 | 이진 분류 출력층 | **확률**로 해석 가능 | **기울기 소실 문제** |

| Softmax | 0 ~ 1 (합=1) | 다중 분류 출력층 | 클래스별 확률 직접 출력 | 출력층 전용, 은닉층엔 부적합 |

### 2-4 경사하강법 (Gradient Descent)


> 손실 함수를 최소화하여 모델을 학습시키는 방법 => 요차를 가장 작게 만드는 가중치와 편향을 찾는 방법

📌 딥러닝에서의 경사하강법은 데이터가 너무 방대하기에 학습 과정에서 모든 데이터를 한 번에 집어 넣으면 메모리가 부족해 시간이 오래걸림 -> 배치(batch) 필요

##### 1. 배치

* 학습 과정에서 가중치와 같은 매개변수를 조정하기 위해 사용하는 데이터의 묶음 의미. 사용자가 임의로 정할 수 있는 하이퍼파라미터

* 배치 크기가 너무 큰 경우, 학습이 느려지고 메모리가 부족해질 수 있음

* 배치 크기가 너무 작으면, 가중치가 자주 업데이트 되다 보니 학습이 불안정해질 수 있음

##### 2. 에포크(Epoch)

* 모든 데이터셋을 학습하는 횟수

##### 3. 이터레이션(Iteration) 또는 스텝(Step)

* 한 번의 에포크를 끝내기 위해서 필요한 배치의 수

* 1 Epoch에서 다루는 데이터의 수 = 배치의 크기 * 이터레이션(step)

####  ① 배치 경사하강법

> 오차를 구할 때 전체 데이터를 한 묶음으로 묶음 (Batch size = Data size)

* 전체 데이터를 통해 학습시키기 때문에, 업데이트 횟수가 적음 (1 Epoch 당 1 업데이트)

* 전체 데이터를 모두 한 번에 처리하므로 메모리가 많이 필요

* 항상 같은 데이터 (전체)에 대해서 경사를 구하기 때문에 수렴이 안정적

####  ② 배치 크기가 1인 확률적 경사하강법(Stochastic Gradient Descent, SGD)

기존 경사하강법은 전체 데이터에 대해서 게산을 하기에 시간이 너무 오래 걸림
> 하지만 배치 크기가 1인 확률적 경사하강법은 매개변수 값을 조정할 때 전체 데이터가 아니라, 랜덤으로 선택된 하나의 데이터에 대해서만 계산함

* 하나의 데이터에 대해서만 메모리에 저장하면 되므로 자원이 적은 컴퓨터에서도 쉽게 사용이 가능

* 매개변수의 변경폭이 불안정하고, 정확도가 낮을 수 있음


####  ③ 미니 배치 경사하강법(Mini-Batch Gradient Descent)

* 배치 크기가 1인 경사하강법은 정확도가 낮을 수 있기에 나온 방법

* 배치 크기를 적절히 지정하여 해당 데이터 개수만큼에 대해 매개 변수의 값을 조정

* 전체 데이터를 계산하는 것보다는 빠르고, 배치 크기가 1인 경사 하강법보다 안정적
⇒ 가장 많이 사용되는 경사 하강법

### 2-5 옵티마이저(Optimizer)

####  ① 모멘텀(Momentum)

> 변수가 가던 방향으로 계속 가도록 속도를 추가

* 경사하강법 + 관성

* 경사하강법에서 계산된 접선의 기울기에 바로 한 시점 전의 접선의 기울기값을 일정 비율만큼 반영

####  ② 알엠에스프롭(RMSProp)

기존 경사하강법은 기울기가 큰 방향에서는 너무 많이, 작은 방향에서는 너무 작게 이동
> 각 학습률을 조정해서, 안정적이고 빠르게 학습하도록 함

📌 급한 경사에서는 속도를 줄이고
완만한 경사에서는 속도를 유지하거나 증가

####  ③ 아담(Adam)

> RMSProp과 Momentum을 합침 + 방향과 스텝 사이즈를 적절하게 합친 방법

각 파라미터의 1차 모멘트(기울기의 평균)와 2차 모멘트(기울기의 제곱 평균)을 사용하여 학습률을 조정
