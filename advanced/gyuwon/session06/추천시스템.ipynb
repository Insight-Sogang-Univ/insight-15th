### 1. 추천 시스템이란?

> 사용자의 정보 데이터를 분석, 개인의 취향에 맞는 아이템을 추천하는 알고리즘

#### 1-1. 사용되는 데이터의 종류와 특징

* 명시적 데이터(Explicit Data)

    - **고객들에 의해 입력된 데이터. 사용자가 선호도를 직접적으로 표현한 데이터.**
    
        - Ex) 평점, 좋아요, 투표, 스케일 점수
    
    - 명백한 데이터를 얻기 쉽지 않음. 모든 사용자가 리뷰나 평점, 별점 등을 남기지 않기 때문
    
    - 사람의 취향은 변함 -> 과거의 선호도가 현재와도 동일한지 알 수 없음
    → 당시의 사용자의 **취향이 정확하게 반영**된다는 장점이 있으나, **데이터 자체를 얻기 
    힘들다**는 단점

* 암시적 데이터 (Implicit Data)

    - **고객들의 행동을 기반으로 수집된 데이터. 사용자의 선호도가 간접적으로 표현된 데이터**

        - Ex) 조회수, 영상시청 기간, 구매 기록, 페이지 체류 기간

    - 사용자가 남기는 발자취 -> 사용자가 서비스를 이용하면 로그데이터나 검색/클릭 기록은 무조건 남음
    
    - 높은 수치가 높은 선호도를 의미하진 않음

    - 데이터를 얻기 굉장히 쉽지만, 사용자의 취향이나 선호도를 직접적으로 알 수 없다는 단점

---

#### 1-2. 추천시스템의 분류

* 내용 기반 필터링(Content-Based Filtering)

    - 콘텐츠의 유사도를 기반으로 사용자에게 추천하는 방법

    - 사용자가 이용하거나 선택한 아이템과 유사한 아이템을 추천하는 방식

* 협업 필터링(Collaborative Filtering)

    - 서비스 내에 있는 **다수의 사용자로부터 얻은 선호도 정보**를 기반으로 사용자에게 추천하는 방법

        - 두 명의 사용자가 비슷한 관ㅅ미사를 가지고 있다면 한 사용자의 데이터를 바탕으로 다른 사용자에게 추천해주는 방식, 사용자 간의 선호도를 서로 고려하여 많은 선택 사항들로부터 아이템을 걸러내거나 선택

        - 사용자 기반(User-Based Filtering): 특정 사용자와 유사한 취향의 사용자가 선호한 아이템 추천

        - 아이템 기반 (Item-Based Filtering): 사용자가 선호한 아이템과 유사한 아이템을 추천

* 하이브리드 필터링(Hybrid Filtering)

    - 2가지 이상의 다양한 종류의 추천시스템 알고리즘을 조합하여 만들어진 알고리즘

    - 콘텐츠 기반 필터링과 협업 필터링 각각의 단점들을 보완하기 위해 제시된 방식

### 2. 내용 기반 필터링(Content-Based Filtering)

#### 2-1. 사용자 프로필

* 추천 시스템에 사용되는 데이터    
    
    * 직접 지정형 : 사용자에게 자신이 선호하는 아이템의 특징을 명시적으로 지정하도록 하는 방법

        - OTT 서비스 가입시, 제시하는 몇 개 영화들에 대한 선호도를 체크해야 하는 경우

    * 간접 지정형 : 사용자의 과거 행동 이력에 기반하여 사용자의 프로필을 작성하는 방법

        - SF 영화를 즐겨본 이력이 존재하는 유저에게 인터스텔라를 추천해주는 경우

* 사용자의 프로필을 획득하는 방식

    - 특성 가중치는 사용자가 가지고 있는 아이템 특성 가중치의 평균 값을 활용

    - 사용자의 기호를 따로 파악하거나 설문조사 또는 자발적인 키워드 입력 등으로 프로필 획득

        - Like/Dislike, Rating from 0 to 5, 유저가 직접 작성한 댓글

        - 검색 로그, 아이템 선택 후 구매 혹은 선택, 특정 아이템을 위해 비슷한 아이템을 tag한 이력

    - 아이템에 대한 유저의 기존 평가를 바탕으로 분류 문제를 해결하는 것과 같음

        - 주어진 User profile 정보를 바탕으로 새로운 아이템을 유저가 평가한 metric 중 하나를 선택
        
        - 아이템에 대한 유저의 기존 평가 $= \{r_1,r_2, …r_n\}$

* 좋은 사용자 프로필을 구성할수록, 추천 시스템의 정확도와 만족도가 높아짐 -> 다양한 방식(명시적+암묵적)을 혼합하여 사용하는 것이 일반적

#### 2-2. 유사도(Similarity)

> 추천시스템을 이해하는데 중요한 수학적 개념

* 영상, 이미지, 텍스트 등의 형태로 존재하는 아이템 정보 데이터를 벡터화한 이후 아이템 벡터간의 유사도를 측정 -> 측정된 유사도를 바탕으로 구객이 선호하는 아이템과 비교하여 유사도가 높은 아이템 선별

* step

    1. 인풋 데이터 만들기

        - 아이템 정보 활용

        - 테이블 형태로 데이터 정리

        - 숫자로 변환하기(원-핫 인코딩)

    2. 유저 정보 활용

        - 유저의 취향 파악

        - 유저를 n개의 feature로 표현

    3. 벡터의 내적과 유사성

        - 벡터 : n개의 feature로 만들어진 데이터 -> n차원의 벡터로 표현 가능

        - 벡터 내적 : 두 벡터의 동일 위치 값들끼리 곱한 후 모두 더한 값

            - 유사성 : 내적 값이 클수록 두 벡터의 유사성이 높음

* TF-IDF(Term Frequency - Inverse Document Frequency) : 특정 문서 내에 특정 단어가 얼마나 자주 등장하는 지를 의미하는 단어 빈도(TF) 와 전체 문서에서 특정 단어가 얼마나 자주 등장하는지를 의미하는 역문서 빈도(IDF)를 통해서 "다른 문서에서는 등장하지 않지만 특정 문서에서만 자주 등장하는 단어"를 찾아서 문서 내 단어의 가중치를 계산하는 방법

    - TF(d,t) : 특정 문서 d에서의 특정 단어 t의 등장 횟수

    - DF(t) : 특정 단어 t가 등장한 문서의 수

#### 2-3. 내용 기반 필터링의 장단점

| **장점** | **단점** |
| --- | --- |
| 아이템이 가지고 있는 **feature 자체**에만 집중하기 때문에 다른 유저의 평점 등에 영향을 받지 않음 | item의 feature를 추출해야 하고 이를 통해서 추천하기 때문에 정확한 feature를 추출하지 못하면 정확도가 떨어짐 |
| item의 feature를 통해서 추천을 하기 때문에 추천이 된 이유를 설명하기 용이함 | 기존의 item과 유사한 item 위주로만 추천하기에 **새로운 장르의 item을 추천하기 어려움** |

> Cold start problem : 추천 시스템이 새로운 유저들에 대한 충분한 정보가 수집된 상태가 아니라서, 해당 유저들에게 적절한 제품을 추천해주지 못하는 문제

### 3. 협업 필터링 (Collaborative Filtering)

> 협업 필터링의 가정 : 과거에 비슷한 선호도를 가진 사용자들은 앞으로도 비슷한 선호도를 가질 가능성이 높음

#### 3-1. 메모리 기반 협업 필터링 (Memory-Based Collaborative Filtering)

* 사용자 또는 아이템 간 유사도를 기반으로 함

* 비교적 초기에 개발된 알고리즘

* 사용자 A와 비슷한 취향을 가진 사용자가 선호하는 아이템을 추천

* 접근법 2가지

    - 사용자 기반 : 특정 사용자와 유사한 다른 사용자를 TOP-N으로 선정해 이 TOP-N 사용자가 좋아하는 아이템을 추천 

        - 특정 사용자와 타 사용자 간의 유사도를 측정한 뒤 가장 유사도가 높은 TOP-N 사용자를 추출해 그들이 선호하는 아이템을 추천

        - 유저 간의 유사도가 높을수록 가중치를 부여하는 방식, 같은 그룹의 다른 유저가 선호하는 아이템을 추천함

        - 추천 정확성 & 효율성: 아이템 수가 사용자 수보다 많은 환경에서 유리
        
        - 안정성: 아이템이 수시로 바뀌는 환경에 적합 (사용자 유입이 적은 플랫폼)
        
        - 다양성: 특정 사용자가 독특한 아이템을 평가한 경우, 뜻밖의 독창적인 아이템 추천 가능 → 다양성과 참신성이 강점
        - 실제 활용: 아이템이 자주 바뀌고 사용자 수가 적은 환경 → 뉴스, 소셜 미디어, 음악 초기 모델에서 주로 활용됨(알 수도 있는 친구 추천, 음악 추천 서비스 등)

    - 아이템 기반 : 사용자들이 그 아이템을 좋아하는지/싫어하는지의 평가 척도가 유사한 아이템을 추천하는 기준이 되는 알고리즘(”아이템간의 속성”이 얼마나 유사한가 X)

        - 추천하고자 하는 target item과 유사한 아이템의 평점을 활용

        - 추천 정확성 & 효율성: 사용자 수가 아이템 수보다 **많은 환경**에서 유리
        
        - 안정성: 아이템 리스트 변화가 적고 **고정적인** 환경에 적합
        
        - 다양성: 유사한 아이템 위주로 추천되어 다양성과 참신성은 상대적으로 부족
        
        - 계산 속도: 사용자별 개별 계산이 불필요해 **빠르고 안정적(Stable)**
        
        - 실제 활용: 사용자 유입이 많은 대형 플랫폼에 유리 → **아마존, 넷플릭스** 등이 IBCF 채택






---

#### 3-2. 메모리 기반 협업 필터링의 한계

* 희소 행렬 문제

    - 실제로 유저-아이템 행렬 **대부분의 원소는 비어 있음 (sparse matrix, 희소 행렬)**

    - NBCF를 적용하려면 **적어도 sparsity ratio(행렬 전체 원소 중 비어있는 원소의 비율)가 99.5%를 넘지 않는** 것이 좋음
    
    **⇒ 그렇지 않을 경우: model-based CF 이용!**

* 확장성 문제

    - 아이템/유저가 계속 늘어날 경우, **확장성**이 떨어짐 **(Scalability)**
    
    - 아이템 i에 대한 평점을 예측하기 위해서는, **아이템 i에 대해 평가를 한 모든 유저와의** 유사도를 구해야 함. **BUT! 유저가 많아지면?** 연산은 계속 늘어나고, 오히려 성능이 떨어지기도 함
    
    **⇒ 따라서 KNN CF 이용!**
    
* 콜드 스타트 문제 

    - 협업 필터링은 **기존의 평점/행동 데이터를 기반**으로 유사도를 계산하는데, 데이터가 아예 없는 경우 추천이 불가능함

| 유형 | 상황 | 설명 |
| --- | --- | --- |
| **User Cold Start** | 신규 유저 유입 | 평가 이력이 없어 유사 유저 탐색 불가 |
| **Item Cold Start** | 신규 아이템 등록 | 아무도 평가하지 않아 유사도 계산 불가 |

* 해결 방법

    - **직접 지정형 프로필** 활용 (가입 시 선호 장르/아이템 직접 선택)
    
    - **Content-Based Filtering** 과 혼합하여 아이템 특성 기반으로 보완
    
    - **인구통계학적 정보** (나이, 성별 등) 활용

#### 3-3. 모델 기반 협업 필터링 (Model-Based Collaborative Filtering)

> 메모리 기반 CF(NBCF)의 한계(Sparsity, Scalability, Cold Start)를 보완하기 위해 등장한 방식으로, 단순히 유사도를 계산하는 것이 아니라, 데이터로부터 패턴을 학습한 모델을 만들어 평점을 예측함 -> 관측된 평점 데이터를 학습(Training) 하여, 관측되지 않은 평점을 예측(Inference) 하는 방식

* 메모리 기반 협업 필터링은 관측된 평점 정보만으로 유사도를 계산하고 추천을 수행하지만, 이는 숨겨진 선호 요인이나 아이템의 특성을 반영하지 못하고, 단순히 유사도에 의존하기 때문에 결측이 많은 환경이나 복잡한 상황에서 정확도가 떨어짐

* 잠재 요인(Latent Factor) : 사용자와 아이템 사이의 관계를 설명해주는, 눈에 보이지 않는 숨겨진 특성(요인)

    - Latent Factor Model : 주어진 데이터로는 알 수 없는 사용자-아이템 간의 숨겨진 특성(feature)을 찾기 위해 주어진 Matrix를 사용자-잠재요인 아이템-잠재요인으로 각각 분해하여 학습하는 모델(유저와 아이템을 구분 없이, 동일한 차원의 벡터 공간에 좌표로 표현)

        - 유저와 아이템이 같은 공간에 놓이기 때문에, **서로의 거리나 각도로 유사도 파악 가능**

        - 사용자와 아이템을 보이지 않는 축(=잠재적인 차원)들을 사용해 나타낼 수 있고, 이를 통해 평점을 예측할 수 있다고 가정하는 모델

* 행렬 분해 (Matrix Factorization)

    - User-Item Matrix (유저-아이템 행렬)에는 유저의 취향+아이템의 특성이 녹아 있지만, 그 결과값만 볼 수 있음
    
    - 이 행렬을 분해하여 평점을 유저의 잠재 요인과 아이템의 잠재요인의 내적으로 설명할 수 있다면, 이 방법을 통해 평점을 예측할 수 있음

### 4. VAE 기반 협업 필터링 (Variational Autoencoder CF)

> 사용자의 아이템 상호작용 벡터(예: 구매/시청 이력)를 입력으로 받아, 잠재 공간에서 재구성하도록 VAE를 학습하는 방식

* 구조 

    - 인코더: 사용자 이력을 잠재변수 z 분포로 매핑
    
    - 디코더: z로부터 아이템 분포를 복원

* 장점

    - 기존 MF처럼 단일 포인트 벡터가 아니라 확률적 잠재 분포로 사용자 선호를 모델링 하여 불확실성을 반영
    
    - 희소한 상호작용 데이터에서도 강건

#### 4-1. MF 의 한계와 딥러닝의 등장

| MF의 근본적 한계 | 내용 | 딥러닝의 해법 |  |
| --- | --- | --- | --- |
| **선형성의 한계** | 평점 예측($\hat{r}_{u, i}$)이 단순한 잠재 벡터의 내적(Dot Product)으로만 이루어집니다. | 비선형 활성화 함수(ReLU 등)를 가진 신경망을 사용하여 복잡한 패턴을 학습합니다. |  |
| **불확실성 무시** | 사용자 취향을 단 하나의 벡터값(Point Estimate)으로만 표현합니다. | **확률 분포**로 취향을 모델링하여 데이터의 불확실성을 반영합니다. |  |

=> 이 한계를 극복하기 위해 VAE 등장

---

#### 4-2. 왜 AutoEncoder(AE)가 아닌 VAE(Variational AE)인가?

> 딥러닝의 비선형성을 추천 시스템에 도입한 초기 시도: 일반적인 AutoEncoder(AE)
→ VAE(Variational AutoEncoder)는 불확실성을 모델링함으로써 추천 시스템의 고질적인 문제인 희소성(Sparsity)에 대한 해답을 제시

| 구분 | AutoEncoder (AE) | Variational AutoEncoder (VAE) |
| --- | --- | --- |
| **잠재 벡터 표현** | 단일 값 (Point Estimate) | **확률 분포** (평균 $\mu$, 분산 $\sigma$) |
| **데이터 불확실성** | 반영 불가 (단순 복원에 집중) | **데이터의 불확실성을 모델링**하여 안정성 확보 |
| **적용 시 장점** | 비선형성 학습 | 희소한 데이터에서 더욱 안정적이며, 일반화 성능 우수 |

---

#### 4-3. VAE 기반 CF (Mult-VAE)의 작동 구조

> 사용자의 아이템 상호작용 벡터(
𝑥
x﻿, 예: 구매/시청 이력)를 입력으로 받아, 잠재 공간에서 재구성하도록 학습하는 방식

| 구성 요소 | 역할 |
| --- | --- |
| **입력** ($x$) | 사용자의 아이템 상호작용 벡터 |
| **인코더** | 사용자 이력을 잠재변수 $z$ 분포로 매핑 |
| **잠재 벡터** ($z$) | 샘플링된 최종 취향 벡터 |
| **디코더** | $z$로부터 아이템 분포를 복원 |

---

#### 4-4. VAE 의 비밀 병기 : 목적 함수와 KL Divergence

> 목적 함수인 ELBO(Evidence Lower Bound)에 있으며, 이를 통해 두 가지 목표를 동시에 최적화

$\mathcal{L}(x; \theta, \phi) = \underbrace{E_{q_\phi(z|x)} [ \log p_\theta(x|z) ]}_{\text{재구성 오차 최소화 (Reconstruction Loss)}} - \underbrace{D_{\text{KL}}(q_\phi(z|x) || p(z))}_{\text{잠재 공간 정규화 (KL Divergence)}}$

- 재구성 오차 최소화 (첫 번째 항):

    - 목표: 디코더가 잠재 벡터 $z$를 사용하여 원래 입력 $x$를 최대한 비슷하게 복원하도록 만듦 (AE의 역할과 동일)

- KL Divergence 항 (두 번째 항):

    - 목표: 인코더가 출력하는 잠재 분포 $q(z|x)$가 사전 분포 $p(z)$ (표준 정규 분포)와 유사하도록 강제
    - 역할: 이 항은 모델이 특정 데이터에 과적합하여 잠재 공간을 불규칙하게 만드는 것을 방지하고, 잠재 공간을 넓고 부드럽게(Smooth) 유지하도록 함

        - AE : 퍼즐 조각들 : 유저 벡터들이 의미 없는 공간에 제멋대로 흩어져 있음

        - VAE : 맞춰진 퍼즐: KL Divergence가 유저 벡터들을 표준 정규 분포 주변에 부드럽고 연결된 상태로 유지하도록 강제

- VAE-CF의 주요 장점

    -  확률적 잠재 분포 모델링 (불확실성 반영)

        - 기존 MF처럼 단일 포인트 벡터가 아니라 확률적 잠재 분포로 사용자 선호를 모델링하여, 예측 시의 불확실성을 반영하고 노이즈에 자유로워짐

    - 희소한 상호작용 데이터에서도 살아남음

        - KL Divergence를 통해 잠재 공간을 정규화하여 데이터 연결 능력이 뛰어나며, 희소성 문제에 효과적으로 대응

    - 비선형성 학습

        - 신경망 구조를 통해 MF의 한계였던 복잡하고 비선형적인 패턴을 학습하여 추천 성능을 극대화

### 5. GNN 기반 사용자 선호 예측

#### 5-1. GNN(Graphic Neural Network, 그래프 신경망)

> 사용자와 아이템의 상호작용을 ‘누가(User) 무엇을(Item) 좋아했다’의 집합으로 보는 것을 넘어, 이들이 서로 연결된 거대한 관계망(Graph)을 형성

* 그래프(Graph) : 노드(Node)와 그 노드를 잇는 선(Edge)을 모아 구성한 자료 구조

    - 노드 : 현실 세계의 개체

    - 엣지 : 개체 간 관계나 상호작용

    - 그래프 사용하는 이유

        - 관계, 상호작용과 같이 다소 추상적인 개념을 다루기에 적합

        - Non-Euclidean space도 표현하고 학습할 수 있음

---

#### 5-2. GNN 의 구조

* 그래프 구조 생성 : 노드는 개별 객체, 엣지는 객체 간 관계나 연결성을 의미

* 매시지 패싱 : 각 노드가 이웃 노드로부터 정보를 받아 전달

    - 메시지 생성 : 이웃 노드로에 전달할 메시지 생성

    - 집계 : 받은 여러 이웃의 메시지를 하나로 합침

    - 갱신 : 자신의 노드 상태를 업데이트 함

* 최종 출력

---

#### 5-3. GNN 의 프로세스

> GNN은 그래프를 입력 받아 메시지 패싱(Message Passing)을 수행. 이때, 각 노드는 자신의 이웃 노드들의 정보(잠재 요인)를 수집하여 자신의 은닉 상태(임베딩)를 업데이트

* 이 과정을 K번 반복하여 더 멀리 있는 이웃의 정보까지 반영 -> 이웃들의 취향까지 반영한 최종 임베딩을 얻고, 두 노드 간 선호도를 예측할 수 있음

* 그래프 구조 생성 : 노드와 엣지를 정의한 후, 각각의 피처를 담아 표현

* 메시지 패싱 : 그래프에서 노드 간 정보를 전달하고 상호작용하는 방법

    - 보통 하나의 GNN 레이어에서 ‘메시지 생성 - 집계 - 갱신’ 의 세 단계로 이루어짐(k개의 레이어로 구성되어 있다면 k번의 과정 반복됨)

    - 메시지 생성 : 각 노드가 이웃에게 보낼 자신의 정보를 생성

    - 집계 : 메시지 생성 이후, 여러 이웃 노드에서 받은 메시지를 하나의 벡터로 통합

    - 갱신 : 집계된 이웃 정보와 자신의 이전 상태를 결합해 새로운 임베딩을 계산

---

#### 5-4. LightGCN (Light Graph Convolutional Networks)

* LightGCN : GNN 의 한종류로 추천 시스템에 특화되어 복잡성을 제거하고 최적화한 버전

    - 특징 변환(Feature Transformation) 제거
        
        - 레이어마다 복잡했던 학습 파라미터 (
        𝑊) 삭제

    - 비선형 활성화 함수 (Non-linear Activation) 제거

        - ReLU와 같은 활성화 함수를 사용하지 않고, 이웃 정보를 합침 → 주로 정규화된 합 혹은 평균 사용

    - 정규화된(Normalized) 이웃 임베딩의 합(Aggregation)만으로 구성 -> CF의 본질에 충실하게 ‘그래프 구조’와 ’전파’라는 뼈대만 남긴, 매우 가볍고 효율이 좋은 모델

---

#### 5-5. GNN 활용 사례

* SNS(Pinterest)

* 핀테크(사기 및 자금 세탁 방지)

* B2c 비즈니스(우버 이츠 메뉴 추천)

---

#### 5-6. GNN 의 장단점

* 장점

    - 고차원 협업 시그널 (High-order Collaborative Signals) 포착 : 단순히 '나'의 잠재 요인(MF, VAE) 분석을 넘어, 나와 취향이 비슷한 사용자들(이웃)과 내가 본 아이템과 연관된 다른 아이템들의 정보를 직접 반복해서 반영할 수 있음

    - 정형화되지 않은 데이터 학습 가능 : GNN은 ‘관계’를 학습하기 때문에, 각 노드가 정형화된 데이터로 정의되지 않아도 관계망 속에서 정보를 효과적으로 전파할 수 있음

    - Cold Start 문제 완화에 효과적 : 신규 사용자나 아이템에 대한 데이터가 부족해 추천을 생성하기 어려웠던 기존 방식과 달리, 노드의 고유 특징과 주변 이웃과의 관계를 활용해 즉시 임베딩을 생성할 수 있음

* 단점

    - 과평탄화 (Over-smoothing) 문제 : 모든 노드의 임베딩이 같은 값으로 수렴하는 현상 -> 메시지 전달을 많이 반복하여 GNN 모델의 레이어가 많아질수록, 연결된 그래프 내의 모든 노드 간의 임베딩이 비슷해지는 문제가 발생

    - 과압축 (Oversquashing) 문제 : GNN의 층이 깊어질수록 집계하는 이웃의 범위가 기하급수적으로 커지는데, 이를 고정된 크기의 벡터 하나에 모두 압축하려고 하면 정보가 손실될 수도 있음

    - 계산 복잡도 ↑ : 실제 세계에서는 수십억 개의 노드와 엣지가 포함된 그래프에 GNN을 적용해야 하므로 막대한 자원과 메모리가 요구

    - 해석의 어려움 : GNN은 모든 상호작용을 집합으로 보지만, 순서를 반영하지는 못함

### 6. Transformer 기반 추천 시스템

#### 6-1. 기존 추천 시스템의 순차적 한계

* CF의 순차적 한계 

    - **순서 무시:** [신발 → 바지 → 셔츠]를 구매한 사용자와 [셔츠 → 바지 → 신발]을 구매한 사용자를 동일하게 간주

    - **문맥 부족:** '방금' 구매한 아이템과 '1년 전'에 구매한 아이템의 중요도를 구분하기 어려움

    - **예측의 한계:** 사용자의 '다음' 행동을 예측하기보다는 '일반적인' 선호도를 찾음

* GNN 의 순차적 한계

    - **지역적 정보 전파의 한계**

        - GNN은 기본적으로 메시지 패싱 통해 관계를 학습
        
        - 따라서, 세션 기반 GNN이 [A → B → C] 순서를 방향성 있는 그래프로 표현하더라도, 'C'가 'A'의 정보를 받으려면 'B'를 거쳐야 함

        - 이 과정에서 정보가 여러 단계를 거치며 희석될 수 있음

- **장기 의존성 문제**

    - 이 방식은 Transformer의 Self-Attention처럼 '3일 전 클릭(A)'과 '방금 전 클릭(E)'의 관계를 동적으로 직접 계산하는 데 한계가 있음
    
    - GNN은 정의된 그래프 구조(이웃)에 의존하는 반면, 어텐션은 시퀀스 내 모든 아이템 간의 관계를 거리에 상관없이 한 번에 유연하게 계산

---

#### 6-2. Transformer: 사용자 행동을 순서(Sequence)로 접근

> 행동 순서(문맥)가 중요한 이유: 의도의 파악 -> Transformer는 시퀀스 내의 문맥과 핵심 의도를 파악하는 데 강력

---

#### 6-3. Transformer 기반 주요 추천 시스템 모델

* SASRec : 단방향 어텐션으로 과거 정보만 활용하여 미래 행동을 추측하는 방식으로 훈련을 받은 추천 시스템 모델

    - **방식:** 단방향(Uni-directional) 어텐션 (GPT와 유사)
    
    - **어텐션 계산:** 현재 아이템은 '오직 과거의 아이템'들만 참조할 수 있음 (미래 정보 차단)
    
    - **목표:** 다음 아이템 예측 (Next-item prediction)

        - `[신발, 바지, 셔츠]` 시퀀스가 주어지면, `[??]` 다음에 올 아이템을 예측

    - **특징:**
        
        - 과거 정보만 활용하여 시간의 흐름을 명확하게 반영
        
        - 학습 과정에서 사용자의 '다음 행동'을 예측하는 데 초점이 맞춰져 있음
    
    - **문제점 :**
        
        - **계산량 많음 :** 각 타임 step 에 따른 loss 모두 계산 (하나의 시퀀스에서 t-1 번 loss 계산)
        
        - **단방향성 :** 이전에 등장한 값만 참고할 수 있고 미래의 값은 참고할 수 없음

**→ 이 두가지 문제점을 보완해서 등장한 알고리즘이 - BERT4rec**

* BERT4Rec : 양방향 어텐션으로 과거 정보와 미래 정보를 활용하여 예측하는 방법을 훈련한 모델

    - **방식:** 양방향(Bi-directional) 어텐션 (BERT와 유사)
    
    - **어텐션 계산:** 시퀀스의 일부를 `[MASK]` 토큰으로 일부러 가리고, 가려진 아이템이 '과거와 미래의 모든 아이템'을 참조할 수 있음 
        
    - **목표:** 시퀀스 내 가려진 아이템 예측 (Masked-item prediction)

    - **특징:**

        - 아이템의 양방향 문맥(과거와 미래)을 모두 고려
        
        - 학습 과정에서 시퀀스 전체의 선호도를 파악하는 데 더 초점이 맞춰져 있음

* SASRec vs. BERT4Rec

    - SASRec: 단방향 어텐션 → 최대한 현실과 유사한 방식으로 과거 데이터만 보고 학습

    - BERT4Rec: 양방향 어텐션 → 아이템 간의 숨겨진 복잡한 맥락을 훨씬 더 깊고 풍부하게 학습
