- **단층 퍼셉트론**  
  
인공신경망이란 세포 간 연결을 통한 정보처리 구조 모델, 입력과 가중치, 편향(임계값을 구하는 함수에서, 임계값을 좌변으로 넘김)을 적용한다. 기준 미달일 시 0에 수렴하게 만듦. 퍼셉트론의 출력값을 구하기 위해 입력, 가중합(입력이 출력에 주는 영향도를 고려. 가중치 크기에 따라 중요도 결정.), 활성화, 출력의 흐름을 따른다.   
    - 단층 퍼셉트론의 한계  
    값을 보내고, 받아서 출력하는 단계로만 이루어진다. AND(입력이 모두 1인 경우에만 출력이 1이 나옴), NAND(입력이 1인 경우에만 출력이 0), OR(입력이 모두 0인 경우에만 출력이 0) 논리 게이트를 구현할 수 있다.   
    그러나 단충 퍼셉트론은 xor게이트를 구현할 수 없는 이유는 직선 하나로만 두 영역을 나누는 선형분류에 특화되어 잇기 때문이다. 


- **다층 퍼셉트론**  
퍼셉트론의 층을 더 쌓아서 조합한다. 중간층을 추가하여 세단계로 나눈다. 더욱 복잡한 문제 해결을 위해 중간에 은닉충을 추가로 더할 수 있다.   
- **딥러닝 모델의 학습 방법**  
    - 순전파와 역전파, 연쇄 법칙  
    학습 알고리즘의 기본 구조는 순전파를 통해 입력값으로 부터 예측을 수행하는 것. 오차 계산을 통해 정확도를 측정후 역전파로 다시 오차 전달한다. 가중치 업데이트 후 같은 과정 반복. 
    연쇄법칙이란 합성함수의 미분을 각함수의 미분 곱으로 나타낼 수 있는 법칙이다. 각 층마다 손실에 주는 영향 곱해 최종 손실 값 나타냄. 모델의 성능을 높이려면 오차를 계산해 순전파와 역전파 활용. 
    - 손실 함수의 정의와 종류
    손실계산은 예측과 실제 정답을 비교하여 오차를 계한 하는것. 모델의 예측이 얼마나 어긋났는지 수치로 표현하는 방법이 손실 함수이다. 
    평균 제곱 오차, 이진 크로스 엔트로피, 크로스 엔트로피 오차 등이 있다. 
    - 활성화 함수의 정의와 종류
    신경망의 각 층은 입력*가중치+편향  으로 활성화 함수 계산, 출력값 산출한다. 계단 함수와 시그모이드 함수가 있다. relu함수는 입력 값이 0보다 크고 작은지에 따라 출력이 달라지는 비선형 함수이다. 
    - 옵티마이저의 정의와 종류
    경사하강법을 더 효율적이고 안정적으로 수행하기 위한 방법, 효과적인 최솟값을 찾아갈 수 있다. 모멘텀은 변수가 가던 방향으로 계속 사도록 속도를 추가하는, 물리학 법칙을 응용한 방법. 알엠에스프롭은 경사를 조정하기 위해 학습률을 바꿔 빠르고 안정적인 학습을 찾음. 아담은 방향과 학습률을 모두 잡기 위한 방법.
