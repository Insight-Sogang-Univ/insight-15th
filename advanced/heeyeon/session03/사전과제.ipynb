## RNN의 한계
1. 장기 의존성 2. vanishing gradient
> 극복하기 위한 lstm 모델

RNN은 순차 데이터, 즉 앞디 연관 데이터를 처리하기 위해 등장한 신경망 구조이다. 입력에 활용하는 신경망. x는 인풋 레이어, w는 가중치를 의미.

이전 시점의 정보를 현재 시점으로 전달하면서 문맥을 학습할 수 있다는 장점이 있지만, 문장이 길어질수록 이전 정보를 제대로 유지하지 못하는 문제가 발생한다.

특히 긴 문장에서 초반 정보가 뒤로 갈수록 점점 희미해지는 장기 의존성(Long-Term Dependency) 문제가 대표적이다. 즉 데이터 후반부에 초반 정보가 반영이 안 되는 것이다.

또한 역전파 과정에서 기울기가 지나치게 작아지는 Gradient Vanishing 문제도 발생하기 때문에 긴 시퀀스 학습이 어렵다. 이것은 또 문장이 길어지면 가중치 값이 늘어나 과도하게 후반부 데이터에 반영된다는 거다. RNN은 데이터를 순차적으로 처리해야 하므로 병렬 연산이 어렵고 학습 속도가 느리다는 단점도 존재한다.

### Attention의 등장 배경 및 장점

Attention 메커니즘은 RNN의 한계를 해결하기 위해 등장하였다.그전에 seq2spq란 입력출력 개수 안 같아도 다양한 변환 작업에 적용이 가능하단 거다. 기존 RNN은 문장 전체 정보를 하나의 고정된 벡터에 압축하려 했기 때문에 긴 문장에서 정보 손실이 심했다.

Attention은 현재 단어를 처리할 때 입력 문장 전체를 참고하여 중요한 단어에 더 집중할 수 있도록 만든 방식이다. 즉 기좀 모델은 *고정된 벡터에 정보를 압축시키기에 중요 정보가 손실*되므로 이를 해결하기 위해 등장한 모델이다. 디코더가 단어 생성할 때마다 입력 문장 전체를 다시 본다. 이를 통해 멀리 떨어진 단어 간 관계도 효과적으로 학습할 수 있게 되었으며, 병렬 처리가 가능해져 학습 속도도 크게 향상되었다. 또한 문맥상 중요한 단어에 높은 가중치를 부여할 수 있어 번역이나 문장 이해 성능이 크게 개선되었다.

### Attention is All You Need

“Attention is All You Need”는 2017년 Google이 발표한 논문으로, Transformer 구조를 처음 제안한 연구이다. 이로 인해 딥러닝 모델이 대부분 self attention 방식을 택하게 되었다. 순차적 게산 구조를 제거. 이 논문에서는 기존 자연어처리 모델에서 사용되던 RNN과 CNN을 제거하고 오직 Attention만을 이용하여 문장을 처리하는 새로운 구조를 제시하였다. Transformer는 병렬 처리가 가능하고 긴 문장에서도 높은 성능을 보였기 때문에 자연어처리 분야의 패러다임을 크게 바꾸었다. 이후 BERT, GPT와 같은 대규모 언어 모델들이 모두 Transformer 구조를 기반으로 발전하게 되었다.
> llm 탄생, nlp 평정, 컴퓨터 비전 등으로 영역 확장.

## Self-Attention

>query, key, value 이해하기: 가장 관련 정보 효과적으로 찾아내기

Self-Attention은 하나의 문장 내부에서 단어들 사이의 관계를 계산하는 메커니즘이다. 문장 속 각 단어가 다른 단어들과 얼마나 관련 있는지를 계산하여 문맥 정보를 반영한다. 예를 들어 문장 속 대명사가 어떤 단어를 가리키는지 파악하거나, 멀리 떨어진 단어 간 의미 관계를 이해할 수 있도록 도와준다. 기존 RNN과 달리 모든 단어를 동시에 계산할 수 있기 때문에 병렬 처리에 유리하며 긴 문장에서도 효과적으로 문맥 정보를 유지할 수 있다.

### Query, Key, Value 개념

Self-Attention에서는 각 단어를 Query(Q), Key(K), Value(V)라는 세 가지 벡터로 변환하여 사용한다. Query는 현재 어떤 정보를 찾고 싶은지를 나타내는 벡터이고, Key는 각 단어가 가진 특징 정보를 의미한다. Value는 실제로 전달되는 정보 역할을 한다. Attention 과정에서는 Query와 Key의 유사도를 계산하여 어떤 단어에 집중해야 하는지를 결정하고, 그 결과를 바탕으로 Value 정보를 가중합하여 최종 문맥 표현을 만든다.

### Cross Attention vs Self-Attention 개념 및 과정

핵심 아이디어, 작동 방식, 참조 대상 등에서 달라진다. 기존은 인코더와 디코더간 연결과 정보간 병목 현상 해결 위함이지만 셀프 어텐션은 시퀀스 내부의 문백 관계 파악 위함이다. 이는 목적의 차이를 서술한 것이다.

Self-Attention은 Query, Key, Value가 모두 동일한 입력 문장에서 생성되는 방식이다. 즉 문장 내부 단어들끼리 관계를 계산한다. 반면 Cross-Attention은 Query는 Decoder에서 생성되고 Key와 Value는 Encoder의 출력에서 생성된다. 따라서 Decoder가 문장을 생성할 때 Encoder의 입력 정보를 참고할 수 있게 된다.

### Scaled Dot-Product Attention

> Dot-Product Attention은 Attention Score를 계산하는 가장 기본적인 방법 중 하나. 쿼리(Query)와 키(Key) 벡터를 내적(Dot Product)하여 유사도를 구함.

즉 어텐션 계산 공식 자체를 의미한다. 현재 트랜스포머 방식의 표준.

dot 과정에서 내적 진행 수 값의 크기에 맞게 스케일링을 추가한다. 나눗셈으로 내적값 과도 증가 방지. 문맥적 표현을 계산의 결과로 얻어질 수 있다.

Scaled Dot-Product Attention은 Transformer에서 사용하는 핵심 Attention 계산 방식이다. 먼저 Query와 Key의 내적값을 계산하여 단어 간 유사도를 구한다. 이후 값이 지나치게 커지는 것을 방지하기 위해 Key 차원의 제곱근으로 나누는 scaling 과정을 수행한다. 그 다음 softmax 함수를 적용하여 Attention Score를 확률 형태로 변환하고, 이를 Value와 곱하여 최종 결과를 얻는다. 이 과정은 어떤 단어가 현재 문맥에서 중요한지를 수치적으로 계산하는 역할을 한다.

## Multi-Head Attention

Multi-Head Attention은 Attention을 여러 번 동시에 수행하는 구조이다. 하나의 Attention만 사용하면 특정 관계만 집중해서 학습할 가능성이 있지만, 여러 개의 Head를 사용하면 서로 다른 관계를 동시에 학습할 수 있다. 예를 들어 어떤 Head는 문법적 관계를, 다른 Head는 의미적 관계를 학습할 수 있다. 각 Head에서 계산된 결과는 하나로 합쳐져 최종 출력으로 사용된다. 이를 통해 모델은 문장을 더욱 풍부하고 다양한 관점에서 이해할 수 있게 된다.

### Multi-Head Attention 개념

Multi-Head Attention은 입력 데이터를 여러 개의 서로 다른 공간으로 투영한 뒤 각각 독립적으로 Attention을 수행하는 방식이다. 각 Head는 서로 다른 가중치를 사용하기 때문에 다양한 특징을 학습할 수 있다. 이후 모든 Head의 결과를 연결하여 최종 출력 벡터를 생성한다. 이 구조는 Transformer의 성능 향상에 매우 중요한 역할을 한다.

### Multi-Head Attention 작동 방식

8명의 헤드에 단어 벡터를 할당하고, 각 헤드들이 각자 분석을 수행하며, 결합 및 최종 투영을 진행한다.

입력 벡터는 먼저 여러 개의 Query, Key, Value로 나뉜다. 이후 각 Head에서 독립적으로 Scaled Dot-Product Attention을 수행한다. 각 Head의 결과는 concatenate 과정을 통해 하나로 합쳐지며, 마지막 선형 변환을 거쳐 최종 출력이 생성된다. 이 과정 덕분에 모델은 다양한 문맥 정보를 동시에 학습할 수 있다.

## Transformer 전체 아키텍처

Transformer는 Encoder와 Decoder 구조로 이루어진 모델이다.

### 전처리 단계 (토큰화 - 임베딩 - Positional Encoding)

토큰화는 문장을 단어 또는 subword 단위로 분리하는 과정이다: 입력 텍스트를 모델이 처리할 수 있는 단위 즉 토큰으로 바꾸는 첫번째 단계..

이후 각 토큰은 임베딩 과정을 거쳐 고차원 벡터로 변환된다. 즉 임베딩은 많은 숫자로 이루어진 긴 벡터로 단어를 나타내는 것이다.

마지막으로 Positional Encoding을 추가하여 각 단어의 위치 정보를 모델에 전달한다. 이를 통해 Transformer는 순서를 고려하면서 문장을 처리할 수 있게 된다.이는 단어의 순서 정보를 임베딩 벡터에 추가하는 단계이다.

### Encoder와 Decoder

Encoder는 입력 문장의 문맥 정보를 추출하는 역할을 한다.-문장 이해, 요약, 의미벡터로 변환하는 역할. 인코더레이어로 구성되어있음.  여러 층의 Self-Attention과 Feed Forward Network로 구성되어 있으며 입력 문장을 풍부한 표현으로 변환한다.

Decoder는 Encoder의 출력을 참고하여 다음 단어를 예측하며 문장을 생성한다. 이 과정에서 Decoder는 Self-Attention과 Cross-Attention을 함께 사용하여 이전 출력과 입력 문맥을 동시에 고려한다. 이러한 Encoder-Decoder 구조 덕분에 Transformer는 번역, 요약, 질의응답 등 다양한 자연어처리 작업에서 높은 성능을 보인다.
