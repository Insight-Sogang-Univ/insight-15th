1. 생성 모델
    - 분류모델과 생성모델의 차이

    생성모델이란 학습 데이터를 학습해 분포를 따르는 유사 데이터를 생성하는 모델. 분류모델은 각 데이터에 정답 라벨이 있었으나, 데이터가 어떤 레이블에 속할지 결정하는 모델이었다. 그러나 생성모델은 샘플에 레이블을 지정하지 않고 학습 데이터와 유사한 새로운 데이터를 만들어내는 모델이다.
    - 생성모델의 분류
        - AE: 오토 인코더는 입력과 동일한 출력을 만드는 것으로, 데이터 복원이나 특성 학습에 자주 쓰인다. 입력데이터를 아주 작은 잠재 벡터로 통과시키기 위한 핵심만 남기고 깎아냄. 이후 병목 통과 후 디코더로 복원한다. 
        - VAE: 데이터의 잠재 구조를 학습해 원본과 닮았지만 새로운 데이터를 생성해내는 모델. 가우시안 분포로 학습함. 어떤 점을 뽑아도 그럴듯한 이미지를 생성함. 텍스트 생성에 많이 쓰임 
        - GAN
            - 구조와 장단점: 생성적 적대 신경망. 생성자와 판별자 신경망이 서로 적대적으로 경쟁하며 훈련을 통해 정교하게 수행함. 이미지 생성, 영상 합성에 사용됨 
        - 확산모델
            - 순확산과 역확산: 입력 이미지에 노이즈를 조금씩 추가하여 완전한 노이즈로 만든뒤 이를 거꾸로 복원하는과정을 통해 입력 이미지와 유사한 확률 분포를 만들어냄/ 원본>노이즈>원본
            순확산은 데이터에 점진적으로 노이즈를 추가하는 과정이고, 역확산은 노이즈 데이터에서 원본데이터를 재구성 하는 과정이다. 
2. AI 최적화
    - 전이학습: 여러가지 딥려닝 모델을 이용하는 과정에서 우리는 데이터 부족의 문제를 직면하게 되므로, 이미 학습된 내용을 가져다 쓸 방법을 찾은 것이 전이 학습. 
        - 전이학습의 개념과 분류: 도메인과 태스크, 소스와 타겟의 개념을 통한 전이학습의 공식적 정의는 다음과 같다. 
        > 
        - Pre-Training과 Fine-tuning: 트랜스포머이전의 전이 학습에는 사전 학습된 좋은 모델이라는 개념 자체가 지금처럼 당연하지 않았다. 트랜스포머 이후의 전이학습에는 사전학습된 거대모델들이 등장하기 시작했고, 모델을 어떻게 내용도에 맞게 조정하는지 따지는 것이 더 중요. 
    - AI Alignment: 인공지능 정렬은 단순히 모델 성능을 높이는 것이 아니라, 인간의 의도와 목표, 윤리적 가치에 부합하도록 만드는 것. *Helpful(도움이 되는가) / Honest(정직한가) / Harmless(무해한가)*
        - RLHF: 모델이 사람이 원하는 답을 이해하게 만들기 위해 인간의 피드백을 기반으로 모델을 강화학습으로 조정함. 정답 여부보다 인간이 선호하는 답변을 학습한다. >인간이 좋아할 대답만 골라하는 보상해킹, 기만적 행동을 보일 위험도 존재함. 
    - 프롬프트 엔지니어링: 언제나 만능인 것은 아니다. 악의적인 사용자가 프롬프트를 조정하거나, 인공지능이 그럴 듯한 거짓말을 하거나, 모델 의존성에 따라 결과가 다르게 나타나는 문제, 성능한계 등이 있을 수 있다.
