- NLP  
> 소통을 위해 자연스럽게 발생. 컴퓨터 언어인 자바와 대비됨.
자연어와 프로그래밍 언어 차이점? : c언어는 일대일로 대응되고, 에러 나면 안 돌아감, 그러나 자연어는 일대다 대응으로 자연발생이므로 유연하고 모호함.  

-nlp란 이 모호한 자연어로 숫자를 변환해 가치 창출하는 기술, 자연어 의미 분석, 컴퓨터 처리 인공지능 기술 분야. 비정형 데이터인 글자를 컴푸터가 처리하도록 챗봇 등에 사용됨

>자연어의 특징? 이산적, 순차적, 중의성, 희소성, 가변성 등이 있다.  언어모델이란 단어시퀀스에 확률 할당하여 가장 자연스러운것을 찾아내는 모델.

- 통계기반 모델  
    - SLM: 분포 가설에 근거(비슷한 문맥 함께 발생 단어는 비슷한 의미가짐. 즉 자주 같이 등장하는 언어로 그 단어 의미 짐작. '조건부 확률=다음에 올 자연스러운 단어. 확률 활용' "카운트 기반=직관적이지만 희소문제를 가짐. 등장 빈도로 다음 단어 활률 계산함")  
        - SLM에 사용되는 분포 가설: 함께 등장하는 단어 들로 의미 추측한다는 가설이다.   
        ---
    - N-gram: 이전 모델은 희소 문제를 해결 못함. 이번 모델은 전체 대신 일부만 본다. N-1인 이유는 예측 단어 합하여 n개의 단어를 보기 위함.    
        - N-gram이 나온 배경과 한계: 그러나 이 모델 또한 한계가 있다. 데이터가 부족하면 확률이 0이 되어 시퀀스를 예측 불거, n의 크기가 커질 수록 해당 조합 찾기 어려워 희소 문제가 더욱 심각해짐. 또한 문맥이 제한된다.    
        ---
    - PPL: 언어 모델이 문장 얼마나 혼란스러워 하는지 나타내는 수치. 문장이 희소 등장 할 수록 ppl이 커지고 n이 커질수록 ppl 낮아짐.  
        -이 수치가 낮다는 것은 모델이 예측에 덜 혼란스럽다는 의미이므로, 모델의 성능이 좋다는 의미까지 내포한다.   

    ---
    - NLM: 신경망 언어모델. 임베딩: 토큰화된 단어는 임베딩에서 숫자의 벡터로 표현된다...    
        - SLM과의 차이: slm은 통계적 모델로 단순히 단어를 이름으로 기억. 신경망 모델인 해당 모델은 거대한다차원 지도 활용. 지도 위에서 아주 가까운의미의 단어를 찾아낼 수 있다 ex: 농구와 축구가 단순 다른 단어가 아니라 의미상 유사도가 높다는 것까지 찾아냄.   
            > 신경망 모델은 미학습 단어도 유사 단어로 추론 가능. 가중치만 저장하면 되므로 공간 활용 좋음. 의미 파악을 벡터로 가능케 함. ..

- sLM  
    - 모델 압축: 소규모 언어 모델은 자연어 콘텐츠를 처리, 이해 및 생성할 수 있는 AI 모델. sLM은 대규모 언어 모델(LLM)보다 규모와 범위가 작다. 수십억의 매개변수, llm은 수천억. 그러나 더 작고 효율적이므로 메모리 연산 자원을 적게 사용한다는 장점이 있다!!  
    > 모델 압축(가지치기, 양자화, 저차원 분해, 지식 증류)
    - sLM의 장단점: 성능이 효과적이지만 큰 모델이 존재하는 편향까지 함께 전달 가능하다. 이는 성능저하로 이어질 수 있으며 환각 문제또한 당연히 발생 가능    
    - sLM과 LLM의 결합: 결합 추론. 복잡 작업과 간단 작업을 나누어 각각 다르게 잘 배분하여 토큰 사용량 절약.
