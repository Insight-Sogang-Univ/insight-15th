- ## 추천 시스템의 분류  

> 추천 시스템이란: 정보 데이터를 사용자의 취향에 맞게 조정하는 알고리즘.   
- 명시적 데이터: 사용자가 평점처럼 직접 선호도를 표현한 것.정확도는 높지만 데이터 자체를 구하기 어렵다.  
- 암시적: 행동 기반 간접적 데이터. 조회수나 구매 기록 등의 발자취와 같다. 데이터를 얻기 쉽지만 해석하는 인사이트가 필요하다.

# 내용 기반 필터링:  
콘텐츠 유사도 기반. 이전에 선택한 아이템과 유사한 아이템을 추천한다. 즉 한 사용자의 행동이력을 데이터롸 하는 것이다.
- 사용자 프로필: 자신이 선호하는 아이템을 직접 명시적으로 지정하는 방법이다. 직접 지정하거나, 과거 행동 이력 기반으로 간접적으로 지정하게끔 할 수 있다. 특성 가중치는 아이템 가중치의 평균값을 활용하고, 키워드 임력 등으로 프로필을 만든다. 이는 아이템에 대한 유저의 기존 평가를 바탕으로 분류 문제를 해결하는 것과 같다.    

- 유사도: 추천 시스템을 이해하는 데 중요한 수학적인 개념이다. 아이템 정보 데이터를 벡터화한 후, 벡터 간 유사도를 측정한다. 그것을 바탕으로 고객 선호 아이템과 유사도가 높은 아이템을 선별.
- TF-IDF: 단어가 얼마나 자주 등장하는지를 의미하는 단어빈도와(특정 문서에서}, 전체 문서에서 얼마나 자주 등장하는지를 의미하는 역문서 빈도를 비고하여, 문서 내 단어의 가중치를 계산하는 방법이다.  

> idf(d,t) = log(\frac{n}{1+df(t)}

- 행렬 분해(Matrix Factorization)
    - 행렬 분해의 대표적 4가지 방식
유저 아이템 매트릭스가 그 결과값만 볼수 있는 것과는 다르게, 이 행렬을 분해하면 평점을 유저의 잠재 요인과 아이템의 잠재요인을 내적으로 설명해서 평점을 예측할 수 있다.   

> 입력 행렬:  (노래/유저) 사용자가 모든 노래를 듣거나 평점 주지 않았을 것이기 때문에, 이 R은 희소 행렬일 가능성이 높다 노래(아이템)의 잠재 요인 행렬: factors(행) x songs(열) 입력 행렬의 빈칸은 두 잠재 요인 벡터의 내적으로 예측된다!  

(* 실제로는 예측 평점이 정해진 척도(예: 1~5) 안에 들도록 복잡한 학습 과정과 편향 처리가 추가된다.)   

    svd: 수학적 분해기법은 행렬 대수학에 기반한다  
    sgd: 최적화알고리즘은 경사하강법 기반으로 희소성 문제 해결    
    nmf: 분해 행렬에 비음수 적용. 제약 조건 기반 모델/ 직관적인 의미 부여가 가능하다.     
    pmf: 평점 예측에 확률 분포를 사용하는 확률적 모델.     

    > 메모리 기반 협업 필터링 한계 극복방법이다. 행렬 분해를 통해 유저와 아이템의 잠재 요인 벡터를 만들어 내적 예산으로 평점을 예측한다.
