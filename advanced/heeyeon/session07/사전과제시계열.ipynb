# 시계열 데이터란?  
## 시계열 데이터의 정의:   
        -> 시간의 흐름은 순차적이다. 연속된 데이터 간에는 현재 데이터와 과거 간의 상관관계가 존재하며, 이는 순서가 중요 변수인 **시퀀스 데이터**로 간주됨.   
> 미래 데이터를 예측하고, 시간 흐름이 있는 데이터를 분석 모델링하는 통계적인 방법과 기법이다. 즉 데이터셋의 전체적인 흐름도와 구조 파악, 예측, 이상치 탐지를 하는 전 과정을 포함함.  
    
- 웹 트래픽 예측을 통한 서버 최적화
- 주가, 환율, 금리 등의 변동 예측
- 사용자의 앱 사용 로그를 이용한 사용자 이탈 예측


    - 시계열 데이터의 특성:   
        1. 단변량 시계역: 단일 변수가 시간이 흐름에 따라 점차 변화하는 데이터. 트렌드와 패턴 이해에 유용함.   
        ex) 주식 가격, 온도 기록, 매출 데이터  
        2. 다변량 시계열: 한 기점에 측정된 여러 변수가 동시에 시간의 흐름에 따라 변하는 데이터. 즉 서로 영향을 주고 받는 것이 가능한 다양한 변수 들을 상관관계를 보는 것이 중요하다.   
        ex) 경제지표나 기후모델  

        ---

        1. **시간 의존성**> 데이터는 시간에 따라 변화한다.  
        즉 시간 순서가 중요한 특성이므로 이를 임의적으로 바꾸면 데이터의 의미 자체가 뒤바뀌게 된다. 과거 값이 현재에 영향을 미치므로, 각 시점간 상관관계도 중요 변수가 된다.   
        2. **계절성**> 일, 주, 년, 월 등 일정 주기에 따라 반복적으로 나따나는 어떠한 패턴 흐름. (여름에 보이는 소비자 패턴-다음 년도에 반복됨)  
        3. 추세: 시간의 흐름에 따라 크게 보이는 어떠한 장기적인 증가 감소 경향. (올라간다 or 내려간다.)   
        4. **자기 상관성**: 시간의 흐름 데이터이니 자연스럼게 이전 시전 값이 이후 시점 값에 영향을 미치게 된다. 어제가 덥다면 보통은 오늘도 덥다.  

            > 만약 시계열 데이터에 회귀분석을 돌리려면 허위 상관관계(두 변수간 인과관계가 없음에도 높은 상관관계가 나타나는 현상)가 나타나므로, 회귀분석 이전에 반드시 정상성 및 추세확인이 필요하게된다.  
            *자기상관성은 회귀의 적이지만, 시계열 모델에선 근거가 된다.(AR은 자기상관성 자체를 활용해서 모델 예측하므로)*
        ---

    - 시계열 데이터의 분석요인:   
      
    1. 추세 요인: 시간 흐름도 전반에 걸쳐 나타나는 큰 그림. 연도별로 꾸준히 상승하는 매출 추세 같은 것.
    2. 계절 요인: 일정한 주기 별로 뚜렷하게 나타나는 패턴. 반복된다. 특정 시즌에 크게 증가하는 매출이라던가
    3. 순환요인은 계절성과 조금 다르다. 반복되긴하지만 더 길고 불규칙하게 변동이 있는 것이다. 경기 호황과 불황은 기간이 꽤 길고 변동이 불규칙하다. 특히 경제 요인처럼 외부 요인에 의해 발생하는 경우. *데이터가 순환요인 주기보다 짧을 경우 아예 무시하기도 함.  
    4. 불규칙 요인: 위의 것들을 모두 제외하고 남은 노이즈. 자연재해처럼 예측이 불가능한 이상치이다.

## 시계열 데이터 전처리:  
특성이 왜곡되지 않기 위해 필수적이다.

1.  결측치/이상치 처리:   
  
    **a**. 값이 특정 시점에서 측정 x 혹은 누락된 상태. 이것이 존재하면 추세나 계절성이 왜곡되므로 대치와 보간이 필요하다. (삭제시 흐름이 깨지거나 단순 대치 시 패턴 왜곡의 가능성이 있음.)  

     > 그래서 보간법? 결측값을 합리적으로 메워주는 과정. 이전 이후 데이터의 변화 흐름과 비슷한 비율로 변했을 것이라 추정한다. ex) 선형, 다항, 스플라인, 조각별 상수 보간법

    **b** 측정된 값이 일반적인 범위에서 벗어났을 경우, 특정 신호의 중요한 의미일 수 도 있으므로 신중한 접근이 필요.

    > 이상치   

    이전 교육 세션에서 배웠던 IQR 3배 범위 안의 정상 값으로 기준을 두거나, Z-score을 통해 임계값을 벗어나면 이상치로 판단하거나, 시계열을 추세, 계절성, 잔차로 분해해 이 마지막 잔차가 특정 임곗값을 초과하면 이상으로 참지한다.

    > 모델
     분류 및 회귀트리를 이용하는 CART, 클러스터에서 멀리 떨어진 아웃라이어 소수 클러스터를 K-MEANS로 찾아내는 클러스터링 기반 이상 탐지, 인공신경망 기반 비지도 학습 오토 인코더 등이 있다.

- 2. EDA (SMOOTHING)
    - 노이즈를 줄이고 전체 흐름을 더 명확히 파악하는, 부드럽게 스무싱하는 기술.  
    - n개의 기간을 평균 합하여, 시간이 지날수록 그 평균합하는 기간도 함께 움직이는 방식이다. 이 n기간이 길수록 더 긴 기간을 평균에 포함하므로, 노이즈가 더 제거되어 매끄러운 곡선이 왼성된다!   

        -단순 이동: 기간 내 모든 값 평균
        -가중 이동 최근 값에 더 큰 가중치 부여
        -지수 이동 최근 데이터에 지수적으로 많은 가중치 부여  

    - 3. 요소분해란? 시계열 데이터가 주어졌을 때 추세성, 계절성, 순환성, 불규칙성으로 분해하는 과정.
    각 요소를 분리함으로서 각각의 패턴을 더 명확히 이해 가능하다. 이후 예측이나 이상 탐지도 정교하게 수행가능.  

    “시간 t에서의 값은 여러 요인(추세, 계절성 등)의 영향을 받아 결정된다”  
        > Yt=f(Tt,St,Ct,Rt)  
        -가법 모형일 경우? Yt=Tt+St+Ct+Rt (관측값은 각 변동요인의 더하기로 설명됨. 독립적으로, 모든 요소 영향이 일정. 계절성 진폭 일정할 때 적합)  
        -승법 모형일 경우? Yt=Tt×St×Ct×Rt (관측값은 각 변동요인의 곱하기로 설명됨. 독립적으로, 모든 요소 영향이 일정하지 않음. 계절성 진폭 변동할 때 적합, 추세 커질수록 계절성이나 불규칙성도 비례해서 커짐.)

## 정상성 판단
    - 정상성의 정의 / 강한 정상성과 약한 정상성  
평균이나 분산 같은 통계 특성이 어제나 오늘이나 일정한 것. 예측이 가능해야 통계 모델링에 적용키 좋음
강: 시점 상관 없이 확률 분포 똑같음. 왜도 첨도도 참고 가능  
> 약:평균도 분산도 일정수중 유지, 자기 공분산성이 시차에만 의존   


    - 정상성의 판단 - 시각 / 통계  
그래프를 봤을 때 정상적인 데이터는 평균 중심 일정하게 진동함. 계절성이 있거나 추세(차분하면 정상 시계열로 변환 가능)가 있으면 비정상 데이터.  
과거 충격이 영구적으로 누적되는 단위근이 있는지 확인한다. 이는 비정상성의 원인. 시계열이 평균회귀 못하도록 막음.  (ADF VS KPSS: 가정이 데이터의 비정상성을 가정하는지에 따라 달라짐.)

- 정상성 확보 방법  
비정상 데이터는 대부분 추세, 게절성, 분산에 변동이 존재한다. 이는 우리주변의 일상적인 데이터들로, EDA를 통해 해당 요인을 확인한후 요소 분해를 통해 그 부분만 제거한다.   

    차분이란: 연이은 관측값의 차이를 계산해서, 변화량을 계산하고, 비정상적인 변동을 제거할 수 있다.  
    주기적인 패턴이 있을 경우 시계열 수준에서 나타나는 차분을 통해 평균 변화를 일정하게 만든다.  > 차분을 통해 랜덤 데이터 생성 가능.  
    또한, 로그 변환을 통해 분산의 정상성을 확보하고 안정화 시킬 수 있따. (변동 폭이 점점 변화가 생길 경우)

## 분석  
- 단변량 시계열 분석  
    - 안정적 시계열 분석  
        - AR 자기 회귀 모델, 작년 데이터가 현재 데이터를 설명해주리라 가정. 시계열 데이터의 자기상관성을 활용.  AR(P)는 과거 P개 시점 데이터로 선형 조합을 만드는 모델,  
        - MA 이동 평균 모델, 어제의 충격이 오늘 결과에 영햐을 미친다는 과정, 과거 시점의 잔차를 독립 변수로 현재 값을 예측한다. 작차는 서로 독립적, MA(q)는 과거 Q개의 예측 오차를 선형 결합으로 예측하는 모델  
        - AR,MA의 전제조건  
        정상성(1−ϕ1z−ϕ2z2−⋯−ϕpzp=0의 모든 해의 값의 절댓값이>1인 것이 정상성을 확보한다.) 가역성(MA 모형을 AR(∞) 형태로 재표현할 수 있는지 여부, ar 모델은 체크할 필요 없음. 오차항도 관측 가능한 과거 값으로 표현할 수 있어야함) 으로 데이터의 특성과 모델 구조의 안정성을 고려한다.
        - ARMA  
        “과거 관측값 + 과거 오차”로 현재 값을 설명하는 모델, Yt=ϕ1Yt−1+⋯+ϵt+θ1ϵt−1+⋯, 패턴과 예측 오차까지 모두 보정   
        - 차수 결정 방법 : 자기 상관 함수 (ACF, PACF)  
            arma에서 사용할 q개의 오차 결정.
            부분상관 자기 계수는 arma에서 p개의 시점을 결정하는 방법이다.  
    - 불안정 시계열 분석  
        - ARIMA / SARIMA  
        안정적 시게열 모델에 정상성 확보 과정이 추가된 상태.  
        ARIMA 차분 변환을 통해 AR, MA, ARMA로 정상화. 백색잡음, 자기회귀, 이동평균 등의 특수사례가 있다.
        SARIMA 계절성을 처리할 수 있고, ARIMA에 계절성 변동을 반영한 모델. 하이퍼파라미터는 자료 읽기만으로 이해가 잘 되지 않아 세션때 더욱 공부해 볼 예정이다.  
- 다변량 시계열 분석
    - VAR: 시계열 분석에서 여러 변수 간의 상호 의존성과 상호 작용을 모델링하는 데 사용하는 다변량 시계열 모델  

    시계열의 벡터를 예측하는 단변량(univariate)은 자기회귀 모델의 일반적인 형태이다. 이와 다르게 다변향은 XY가 서로의 과거를 함께 사용한다.  

    y1,t=c1+ϕ11,1y1,t−1+ϕ12,1y2,t−1+e1,t   
    y2,t=c2+ϕ21,1y1,t−1+ϕ22,1y2,t−1+e2,t   

-> 두 개의 시계열이 서로의 과거 값에 영향을 주면서 동시에 변화하는 모델(VAR(1))!!

- 딥러닝을 활용한 시계열 분석  
    - RNN / Tranformer / Foundation Model  

    - RNN 순환신경망. 순차적 패턴이 중요한 시계열 데이터. 그러나 병렬 처리 효과는 낮음,
    x1 → x2 → x3 → x4
     ↓    ↓    ↓    ↓
    h1 → h2 → h3 → h4
이전 기억(h)이 다음 입력에 계속 전달됨,   
    - TRANSFORMER 인포머(긴 시계열 처리 최적화→ 중요하지 않은 attention 줄임), 오토포머(trend / season 분해해서 예측), TFT(과거 데이터 + 외부 변수 같이 사용) 등이 있음."이 시점은 과거의 어떤 시점과 관련 있는가?" 장기 예측에 특화되거나, 주기성 포착하거나, 과거 데이터뿐 아니라 다양항 외부변수를 사용하기도. 긴 거리 관계 잘 기억하고, 언어/시계열 모두 강함  
    - 거대 데이터로 사전 학습하여 문제 해결하는 범용 모델 FOUNDATION MODEL. 사전 학습( 큰 데이터로 일반 패턴 학습) 후 파인 튜님(특정 문제에 맞게 조정). GPT 계열 LLM 기반이거나 시간 패턴 일반화 모델인 시계열 전용으로 유형이 나뉜다.
