## 1. 인공신경망의 시작 : 퍼셉트론

### 1-1. 인공신경망이란?
- 인공신경망 : 우리 뇌의 신경망 (뇌를 구성하는 신경 세포끼리 서로 연결되어 정보를 처리하는 구조)를 모방한 모델
    - 입력 -> 가중치, 편향 적용 -> 활성화 함수 (임계값 기준) -> 결과 출력
- 퍼셉트론 : 초기 형태의 인공신경망, 신경망(딥러닝 기원), 선형 이진 분류 모델
- 용어 정리
    1. 가중치 : 입력값이 출력값에 주는 영향력(중요도)을 결정
    2. 활성화 함수 : 신호가 일정 이상의 값(임계값)을 넘으면 1, 넘지 못하면 0을 출력 
    3. 임계값 : 활성화 함수에 사용되는 값
    4. 편향 : 임계값을 좌변으로 넘겨 표현

### 1-2. 단층 퍼셉트론과 그 한계
- 단층 퍼셉트론 : 단순히 입력층과 출력층으로만 이루어진 퍼셉트론
- 논리 게이트 구현 
    - 디지털 회로에서 기본적인 논리 연산을 수행하는 데 사용되는 구성 요소
    - n 개의 입력값과 하나의 출력값 사용
    - AND, OR, NOT, NAND, NOR, XOR 등 다양한 종류
    - 논리 게이트 종류
        - AND 게이트 : 입력값이 모두 1인 경우에만 출력값이 1이 나오는 구조
        - OR 게이트 : 입력값이 모두 0인 경우에 출력값이 0이고, 나머지 경우에는 모두 출력값이 1인 구조
        - NAND 게이트 : 입력값이 모두 1인 경우에만 출력값이 0, 나머지 경우에는 모두 출력값이 1인 구조
        - XOR 게이트 : 입력값 두 개가 서로 다른 값을 갖고 있을 때에만 출력값이 1인 구조
- 단층 퍼셉트론의 한계
    -  XOR 게이트 구현 불가능 : 입력값과 가중치의 선형 결합으로 이루어진 구조 -> 선형 분류 문제에 대해서만 구현 가능

### 1-3. 다층 퍼셉트론
- 중간에 층을 추가하여 (입력층 + 하나 이상의 은닉층 + 출력층)으로 구성
- 두 개 이상의 은닉층을 갖는 MLP가 딥러닝의 시초 모델로 발전





## 2. 딥러닝 모델의 학습

### 2-1.순전파와 역전파
- 학습 알고리즘 구조
    1. 순전파를 통해 입력값으로부터 예측 수행
    2. 오차 계산을 통해 예측 정확도 측정
    3. 역전파를 통해 출력층에서 다시 입력층 방향으로 오차 전달
    4. 가중치 업데이트
    5. 1~4 반복
- 연쇄 법칙
    - 순전파/역전파의 과정 : 연쇄법칙을 통해 일어남
    - 합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다는 법칙
    - 딥러닝 모델 학습 중 오차 줄이기 위해, 역전파 과정에서 각 가중치가 최종 손실에 얼마나 기여했는지를 계산
    - 각 은닉층마다 손실에 주는 영향을 곱해서 최종 손실에 대한 값
- 기울기 소실 문제
    - 은닉층 다수 & 각 단계마다 미분값이 1보다 작은 값 -> 기울기가 0이 되는 문제 발생
    - 학습이 제대로 이뤄지지 X
- 전체적인 흐름
    1. 순전파 = 예측하기
    > 입력 x → [가중치 w1] → [가중치 w2] → 예측값 ŷ
    2. 손실 계산 = 얼마나 틀렸는지 측정하기
    > 손실 L = (예측값 ŷ - 실제값 y)²
    3. 역전파 = 책임 추적하기
    > 손실 L ← [가중치 w2의 책임은?] ← [가중치 w1의 책임은?]
    4. 가중치 업데이트 = 수정하기

### 2-2. 손실 함수
- 실제값(정답)과 예측값(출력값)의 차이를 계산하는 데에 사용되는 함수
- 종류
    1. 평균 제곱 오차 (Mean Squarred Error, MSE)
    > $$\frac{1}{N} \sum (y - \hat{y})^2
\quad (y : 실제값,\ \hat{y} : 예측값) $$
    2. 이진 크로스 엔트로피 (Binary Cross-Entropy Loss, BCE)
    > $$
L = -\left( y \log \hat{y} + (1 - y)\log(1 - \hat{y}) \right)
\quad (y : 실제값(0,1),\ \hat{y} : 예측값(확률))
$$
    3. 크로스 엔트로피 오차 (Cross Entropy Error, CEE)
    > $$
-\sum y \log \hat{y}
\quad (y : 원\text{-}핫 인코딩 값(1,0),\ \hat{y} : 예측값(확률))
$$

### 2-3. 활성화 함수
- 각 층의 출력값에 비선형성을 추가해서 신경망이 복잡한 패턴도 학습할 수 있게 해주는 함수
1. 계단 함수
>$$
h(x) =
\begin{cases}
1 & (x \ge 0) \\
0 & (x < 0)
\end{cases}
$$
2. 시그모이드 함수
> $$
f(x) = \frac{1}{1 + e^{-x}}
$$
3. ReLU 함수
> $$
f(x) = \max(0, x)
$$
4. Tanh(쌍곡탄젠트)함수
> $$
f(x) = \tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}
$$
5. Softmax 함수
> $$
f(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}
$$

### 2-4. 경사하강법
- 오차를 가장 작게 만드는 가중치와 편향을 찾는 방법
> $$
\frac{\partial J}{\partial w} > 0 \;\Rightarrow\; \text{decrease } w
$$

> $$
\frac{\partial J}{\partial w} < 0 \;\Rightarrow\; \text{increase } w
$$

> $$
w = w - \alpha \frac{\partial J}{\partial w}
$$

> $$
\alpha : \text{learning rate}
$$

- 배치 : 학습 과정에서 가중치와 같은 매개변수를 조정하기 위해 사용하는 데이터의 묶음
- 에포크 : 모든 데이터셋을 학습하는 횟수

1. 배치 경사하강법
    > $$
w_n = w_{n-1} - \alpha \nabla f(w_{n-1})
$$
    - 특징
        - 전체 데이터 학습 -> 업데이트 횟수가 적음 (1 Epoch 당 1 업데이트)
        - 전체 데이터를 모두 한 번에 처리 -> 메모리가 많이 필요
        - 항상 같은 데이터에 대해서 경사 계산 -> 수렴이 안정적
2. 배치 크기가 1인 확률적 경사하강법
    - 전체 데이터 X, 랜덤으로 선택된 하나의 데이터 계산
    - 특징
        - 자원이 적은 컴퓨터에서도 쉽게 사용이 가능
        - 매개변수의 변경폭이 불안정하고, 정확도가 낮을 수 있음
3. 미니 배치 경사하강법
    - 배치 크기를 적절히 지정 -> 해당 데이터 개수만큼에 대해 매개 변수의 값을 조정
