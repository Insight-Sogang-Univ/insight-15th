## 합성곱 신경망 (CNN)
### 픽셀과 채널
- 픽셀
  - 입력 이미지의 가장 기본적인 데이터 단위
  - 이미지 데이터를 저장할 때 픽셀의 색을 표현하는 값을 2차원 배열로 표현
- 채널
  - 픽셀의 색을 숫자로 표현하는 방식
  - 그레이 스케일(흑백 이미지), RGB (컬러 이미지)

### CNN
- 컴퓨터는 이미지를 인식할 수 없으므로, 원본 이미지를 컴퓨터가 이해할 수 있도록 **픽셀 단위로 구분**
- 이때 이미지 공간 정보를 유지한 상태로 학습
- 부분연결 구조 & Convolution 연산으로 특정 피쳐에 집중 -> 효율적인 파라미터 사용
- 특징 어느 정도 추출 후 flatten -> 공간 정보 손실 정도 낮음

### 합성곱과 풀링
- CNN 모델의 구조
  - 특징 추출 영역
    - Convolution Layer와 Pooling Layer를 여러 겹 쌓는 형태로 구성
      - Convolution Layer: 입력 데이터에 필터를 적용 후 활성화 함수를 반영하는 필수 요소
      - Pooling Layer: Convolution Layer 다음에 위치하며, 이미지의 공간 크기를 줄여 중요한 특징을 강조하는 역할
  - 이미지 분류 영역
    - CNN 모델의 마지막에 위치
    - Fully Connected Layer로 구성
  - Feature Learning(Conv + Pool)과 Classification(FC Layer) 사이에 3차원을 1차원 배열 형태로 만드는 Flatten 위치
  1. Convolution Layer
  : 전체 이미지를 일정 크기의 필터로 훑으면서 이미지의 특징을 학습
  2. Padding : filter를 거치며 정보를 잃어버리는 것을 막기 위해 이미지 바깥을 0으로 둘러싸는 기법
  3. Stride : 이미지를 몇 칸씩 점프하면서 훑을 것인지 결정하는 값
  4. Pooling : 피처맵에서 특징을 추출하기 위해 이미지를 격자 형태로 분할한 뒤, 해당 영역 안에서 가장 큰 값 또는 평균값을 취하는 것

### 대표적인 CNN 아키텍처
- AlexNet : 머신러닝에서 벗어나 최초로 GPU를 이용해 학습한 대규모 CNN 모델
- VGGNet : 이미지 인식에 강한 깊은 층의 CNN 모델
- ResNet : 잔차 연결 개념을 도입해 깊은 신경망의 소실 문제를 해결한 모델

## 이미지 딥러닝 응용
### 이미지 분류
- 이미지를 픽셀 값으로 인식 -> 사전 정의된 클래스 중 하나로 분류
### 객체 탐지
- 한 이미지에서 객체와 이 객체를 둘러싸는 가장 작은 직사각형으로 정의되는 bounding box를 찾는 작업
### 이미지 캡셔닝
- image captioning은 이미지 내에 있는 객체에 대한 판단뿐만 아니라 객체들 간의 관계를 파악하고 자연어의 형태로 알맞게 표현
- CNN 이용한 특징 추출 -> Attention Network를 이용하여 단어 생성 -> RNN을 이용하여 문장 완성

# RNN
## 순차 데이터
- 순서에 의미가 있으며, 순서가 달라질 경우 의미가 손상되는 데이터
- 데이터를 구성하는 요소들의 순서 자체가 핵심 정보
- 이전 시점의 데이터가 다음 시점의 데이터에 영향을 미침
## RNN 구조 및 원리
- 순환하는 구조를 가진, 순차 데이터 처리에 특화된 인공 신경망
- 구조
  1. 시퀀스 데이터를 입력받아 순서 정보를 유지하며 처리
  2. 순환하는 은닉층이 매 시점의 은닉 상태 업데이트
  3. 이전 시점의 값을 현재 시점으로 넘겨 줌
- 가중치 공유
  - 동일한 가중치를 시퀀스의 다양한 시점에서 반복적으로 사용
  - 시퀀스 데이터의 길이와 시점 위치에 상관없이 효과적으로 작동
  - 학습 파라미터의 수 감소
  일반화 능력 향상
## RNN의 한계 및 장기 의존성 문제
- 역전파를 구하는 과정에서 연쇄법칙에 의해 미분값이 반복적으로 곱해짐 (등비수열 형태)
  1. 기울기 소실 : 기울기가 0으로 수렴
  2. 기울기 폭주 : 기울기가 무한대로 발산
  3. 느린 훈련 시간
- 장기 의존성 문제
  - 시퀀스 앞부분의 중요한 정보를 잊어버려 맥락 파악 능력이 급격히 저하되는 현상

# LSTM & GRU
## LSTM
- 장기기억과 단기기억 및 출력을 따로 분리해서 관리
  1. Forget Gate : 정보를 얼마나 잊어버릴지 결정
  2. Input Gate : 현재 정보를 얼마나 사용할지 결정
  3. Output Gate : 다음 층으로 어떤 정보를 전달할지 결정
- RNN의 기울기 소실 문제를 해결하기 위한 구조
- RNN보다 학습 파라미터가 많아짐
- 문맥 이해가 중요한 작업, 데이터 양이 충분하고 복잡한 패턴을 학습해야 하는 경우, 계산 효율이 중요한 경우 사용

## GRU
- 모든 기억과 행동이 하나의 상태로 통합
  1. Update Gate : 과거 정보와 새 정보의 비율 조정
  2. Reset Gate : 과거 기억 무시
- 자원이 제한적이거나 학습 속도가 중요한 경우, 데이터 양이 적고 과적합 위험이 있는 경우, 실시간 예측이 필요한 경우 사용

# Seq2Seq
- 한 시퀀스를 다른 시퀀스로 변환하는 작업 수행하는 딥러닝 모델
- 구조
  - 인코더 : 의미를 받아서 정리
    - 입력 -> 컨텍스트 벡터로 압축
  - 디코더 : 결과를 출력
- 입력과 출력의 길이가 달라도 가능
- 한계
  - 병목현상 : 고정된 길이에 입력 시퀀스의 정보를 모두 담지 못해 손실 발생할 수 있음
  - 기울기 소실 문제
  - 병렬화 불가능
