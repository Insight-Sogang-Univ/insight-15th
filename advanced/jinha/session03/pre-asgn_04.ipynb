## Attention의 등장
- 기존 Seq2Seq 모델은 입력 문장의 모든 정보를 하나의 고정된 Context Vector에 압축
- 병목 현상 -> 중요 정보 손실 문제 발생
- 중요한 정보에 더 집중하는 매커니즘
  - 현재 시점에서 가장 중요한 부분에 더 높은 가중치를 부여
  - 출력 시점마다 새로은 context vertor 계산 (병렬화)
- 장점
  1. 시간 경과에 대한 유연성
    - 시간적, 순서적 거리의 제약 X
  2. 공간에 대한 유연성
    - 픽셀 간의 거리에 상관없이 한 번에 전체 이미지 관계 학습 가능
  3. 병렬화
    - 모든 단어의 관계를 한 번에 독립적으로 계산 -> 병렬 처리 가능

## Attention is All YOU NEED
- 오직 Attention만을 사용하는 모델 -> self attention에 기반한 **Transformer** 모델
- 모델링에 Recurrence 및 Convolution이 필수가 아님을 증명

## Transformer
- Attention만으로 인코더와 디코더 구현
- Self-Attention
  - 문장 안에서 어떤 단어가 다른 단어들과 얼마나 중요한 관계를 맺고 있는지 한 번에 파악하는 기술
  - 문장 자체를 뜯어보며 내부 관계를 파악
  - 시퀀스 내부의 문맥 및 의존 관계 파악
  - Query, Key, Value
    - Query : 지금 당장 내가 알고 싶거나, 초점을 맞추고 있는 대상
    - Key : 검색 대상이 되는 모든 정보들이 달고 있는 '이름표' 또는 '색인'
    - Value : 키와 한 쌍으로 묶여 있는 '실제 내용물'
  - 핵심 과정
    1. 단어의 프로필 만들기 (벡터 임베딩)
    2. 단어 간 관계 점수 계산 (내정 및 정렬 점수)
    3. 중요도 배분 (소프트맥스 & 어텐션 가중치)
    4. 새로운 벡터 생성
  - Cross-Attention vs. Self-Attention
    - RNN vs Attention Only
    - 병렬 처리 불가능 vs 가능
    - 서로 다른 시퀀스 vs 동일한 시퀀스
- Scaled Dot-Product Attention
  - Dot-Product Attention
    - Attention Score를 계산하는 가장 기본적인 방법 중 하나
    - 쿼리(Query)와 키(Key) 벡터를 내적(Dot Product)하여 유사도 측정
    - 문제 : Attention Score의 극단화 -> Vanishing Gradient 문제 발생
  - Dot-Product Attention 과정에서 내적을 진행 후, 값의 크기를 맞추는 스케일링 과정 추가
- Multi-Head Attention
  - 셀프 어텐션을 여러 개의 머리(Head)로 동시에, 서로 다른 관점에서 실행
  - 한 단어와 다른 단어 간의 관계를 여러 차원으로 나누어 병렬로 학습
  - 작동 방식
    1. 분할 - 각 헤드에 단어 벡터 할당
    2. 병렬 어텐션 계산 - 각 헤드들이 각자 분석 수행
    3. 결합 및 최종 투영
- 전처리 단계
  1. 토큰화
  - 입력 테스트를 모델이 처리할 수 있는 단위(토큰)으로 나눔
  2. 임베딩
  - 토큰화된 각 단위(토큰)이 숫자 벡터로 변환
  - 벡터의 길이 => 벡터의 차원
  3. Positional Encoding
  - 단어들의 순서 정보를 임베딩 벡터에 추가
  - 등장한 순서, 위치 정보를 제공하기 위해 사용
- 인코더와 디코더
  1. 인코더
  - Multi-Head Attention
  - Feed-Forward
    - 입력 벡터의 차원을 확장하고 비선형 변환을 적용하여 새로운 표현을 생성하는 신경망 구조
  2. 디코더
  - Masked Multi-Head Attention
    - 미래 시점의 단어 정보를 참고하지 못하도록 마스크 적용
  - Encoder-Decoder Attention (Cross-Attention)
    - 디코더가 인코더의 출력(Context)을 참고하면서 현재 생성 중인 단어를 입력 문장의 의미와 연결
