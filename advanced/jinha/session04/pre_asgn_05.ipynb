## 1. NLP와 언어 모델
### (1) NLP
- 자연어 : 인간이 소통을 위해 사용하는 언어 <-> 프로그래밍 언어
  - 특징
    1. 이산적 데이터 : 단어라는 독립적인 단위로 구성
    2. 순차적 데이터 : 순서에 따라 의미 결정
    3. 중의성과 모호성 : 문맥에 따라 뜻 천차만별
    4. 희소성 : 단어의 조합은 무한, but 데이터에는 극히 일부
    5. 가변성과 진화 : 언어는 계속해서 변함
- NLP
  - 자연어를 컴퓨터가 이해할 수 있는 숫자(데이터)로 변환하여 유의미한 가치를 창출하는 기술
  - 비정형 데이터인 글자를 벡터로 바꾸고 그 안에서 의미를 추출하는 모든 과정
  - 비정형 데이터인 글자를 컴퓨터가 계산할 수 있는 '숫자(Vector)'로 바꾸고, 그 안에서 의미를 추출하는 모든 과정을 포함
### 통계적 언어모델 (SML)
- 분포 가설(Distributional Hypothesis)에 근거
  - "비슷한 문맥에서 함께 나타나는 단어들은 비슷한 의미를 가진다"는 가설
- 이전 단어 시퀀스의 등장 빈도를 통해 다음 단어의 확률을 계산
- 희소 문제(Sparsity Problem) : 새로운 조합에 약함

### N-gram
- SLM은 희소하게 등장한 것에 대해서는 잘 예측 X, 문장이 길어질수록 다음 단어를 계산하기 어려움 -> 문장의 일부만 보자 : N-gram 등장
- 한계
  - 희소 문제
  1. 데이터 부족
  2. N의 크기 클수록 심해짐
  - 문맥 제한
    - 앞부분 정보를 뒷부분 결정에 반영하지 못함

### Perplexity(PPL)
- 언어 모델이 문장을 얼마나 "혼란스러워하는지(perplexed)"를 수치로 나타낸 것
- 문장 W가 희소하게 등장할수록 커지고, 단어 개수 N이 커질수록 낮아짐

### 신경망 언어 모델 (NLM)
- 통계적 언어 모델의 한계를 극복하기 위해 등장
- 모든 단어를 거대한 다차원 지도 위의 한 점(좌표)으로 보냄 : SLM과의 차이

## 딥러닝 기반 언어모델
### BERT
-  Bidirectional Encoder Representations from Transformers
- 한 단어를 이해할 때 그 단어의 좌우 문맥을 동시에 보고 이해
- input → Embedding → 트랜스포머 인코더 12개(or 24개) → Output

###  GPT
- decoder 기반
- 지금까지의 단어를 바탕으로 다음 단어를 예측
- few-shot learning 도입

### LLM
- Large Language Model
- '학습'과 '예측'이라는 두 가지 핵심 단계를 통해 작동
  1. 대규모 학습 (Training)
  2. 예측 및 생성 (Prediction & Generation)
### RAG
- 외부에서 최신 정보를 가져와서 답변하게 하기 위한 것
- 최신 정보 반영 가능

### LangChain
- LLM+RAG를 직접 구현하는 것은 복잡하니까, 이러한 모델 pipeline을 쉽게 구성할 수 있도록 도와주는 것
- 특징
  1. 추상화
  2. 표준화
  3. 체이닝

### sLM
    - 모델 압축
    1. 가지치기
    2. 양자화
    3. 저차원 분해
    4. 지식 증류
    - sLM의 장점
      - 효과적인 성능, 지연 시간 감소, 보안
      - 접근성, 효율성, 에너지 효율, 비용 절감
    - 단점
      - 편향, 제한된 일반화, 복잡한 작업에서의 성능 저하, 할루시네이션
    - sLM과 LLM의 결합
      - 복잡하고 어려운 작업에는 LLM, 가볍고 간단한 작업에는 SLM을 적절히 배분해서 사용
