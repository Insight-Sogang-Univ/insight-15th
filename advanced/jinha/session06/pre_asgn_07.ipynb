## 1. 추천 시스템이란?
- 사용자의 정보 데이터를 분석, 개인의 취향에 맞는 아이템을 추천하는 알고리즘
- 데이터 종류
  1. 명시적 데이터 : 직접 입력된 데인터
  2. 암시적 데이터 : 행동을 기반으로 수집된 데이터
- 분류
  1. 내용 기반 필터링 : 콘텐츠의 유사도 기반
  2. 협업 필터링 : 다수의 사용자로부터 얻은 선호도 정보 기반
  3. 하이브리드 필터링 : 추천시스템 알고리즘을 결합
## 2. 내용 기반 필터링
  1. 사용자 프로필 획득
    - 데이터
      1. 직접 지정형 : 사용자가 직접 체크
      2. 간접 지정형 : 사용자 과거 행동 이력에 기반하여 작성
    - 사용자의 특성 가중치 평균값 활용 -> 데이터를 통한 프로필 획득 -> 분류 문제 해결
  2. 유사도
    - 아이템 벡터 간 유사도 측정 -> 유사도 높은 아이템 선별
    - TF-IDF
      - 단어 빈도(TF)와 역문서 빈도(IDF)를 통해 문서 내 단어의 가중치 계산
      - 값이 높을수록 그 단어가 해당 문서에서 더 중요한 역할
  3. 내용 기반 필터링 장단점
    - 장점
      - 다른 유저의 평점 등에 영향 X
      - 추천 이유 설명하는 것 용이
    - 단점
      - 정확한 피처 추출하지 못한다면 정확도 떨어짐
      - 새로운 장르 추천 어려움
## 3. 협업 필터링
  1. 메모리 기반 협업 필터링 (이웃 기반 협업 필터링)
    - 사용자 혹은 아이템 간 유사도 기반
    - 비슷한 취향을 가진 사용자가 선호하는 아이템 추천
    - 한계
      - 희소 행렬 문제 : 데이터가 적을수록 성능 저하
      - 확장성 문제 : 데이터가 늘어날수록 연산량 늘어나 확장성 떨어짐
      - 콜드 스타트 문제 : 데이터가 아예 없는 경우 불가능
  2. 모델 기반 협업 필터링
    - 데이터로부터 패턴을 학습한 모델을 만들어 평점 예측
    - Latent Model
      - 잠재 요인 : 관측된 데이터 분석하여 수치로 '추정'한 값
      - 잠재 요인 찾기 위해 사용자-잠재요인/아이템-잠재요인 으로 분해하여 학습 -> 평점을 예측하는 모델
      1. factor model
        - 유저와 아이템 구분 X, 동일한 차원의 벡터 공간에 좌표로 표현
      2. latent factor model
        - 축과 차원의 개수를 모델이 알아서 결정, 알 수 없음
## 4. 행렬 분해
1. SVD
  - 수학적 분해 기법
  - 행렬 대수학에 기반하여 행렬 분해
  - 결측치 처리 필요
2. SGD
  - 경사 하강법 기반  최적화 알고리즘
  - 관측된 평점에 대해서만 학습 진행
3. NMF
  - 제약 조건 기반 모델
  - 분해된 행렬에 비음수 제약 조건 추가
4. PMF
  - 확률적 모델
  - 평점 예측 + 확률 분호 사용 -> 불확실성까지 모델링

#딥러닝 기반 추천 시스템
## 1. VAE 기반 협업 필터링

- 사용자의 아이템 상호작용 벡터 입력 -> 잡재 공간에서 재구성하도록 VAE 학습
- 불확실성 모델링 -> 희소성 해답 제시
- 작동 구조 : 입력 -> 인코더 -> 잠재 벡터 -> 디코더
- 목적함수 : ELBO + 잠재 공간 정규화

## 2. GNN 기반 사용자 선호 예측
- '관계' 자체를 모델링
- 그래프를 사용하여 예측
- 작동 구조 : 그래프 구조 생성 -> 메시지 패싱 -> 최종 출력
- 활용 사례 : 핀터레스트, 금융 사기와 자금 세탁 탐지, 우버 이츠 메뉴 추천
- 장점
  - 고차원 협업 시그널 포착
  - 정형화되지 않은 데이터 학습 가능
  - 콜드 스타트 문제 완화에 효과적
- 단점
  - 과평탄화 문제
  - 과압축 문제
  - 계산 복잡도 증가
  - 해석의 어려움

## Transformer 기반 추천 시스템
- 순서와 문맥 이해 도입 -> 의도 파악
- 주요 모델
  - SASRec
    - 단방향 어텐션 (과거 아이템 참조, 미래 아이템 차단)
    - 다음 아이템 예측
    - 한계 : 계산량 많음, 단방향성
  - BERT4Rec
    - 양방향 어텐션
    - 마스크로 가리고 ㄱㅏ려진 아이템 예측
    - 아이템 간 숨겨진 맥락 훨씬 더 풍부하게 학습
