## 시계열 데이터
- 순차적인 시간 흐름에 따라 수집된 데이터
- 상관관계 존재, 순서가 중요한 시퀀스 데이터
- 과거 데이터가 미래에 어떤 영향을 미치느냐가 중요
1. 단변량 시계열 : 하나의 숫자가 시간에 따라 변하는 것
2. 다변량 시계열 : 여러 숫자가 동시에 시간에 따라 변하는 것
- 주 목적 : 미래 데이터 예측

## 시계열 데이터의 특징
1. 시간의존성 : 시간에 따라 변화
2. 계절성 & 추세 : 반복적으로 나타나는 변동, 흐름에 따라 나타나는 장기적인 증가/감소 경향
3. 자기상관성 : 이전 시점의 값이 이후 시점의 값에 영향

## 시계열 데이터의 분석요인
1. 추세 요인 : 데이터 전반에 걸쳐 시간에 따라 지속적으로 증가/감소하는 경향
2. 계절 요인 : 일정한 주기로 반복되는 패턴
3. 순환 요인 : 계절성보다 더 길고 불규칙한 주기로 반복되는 변동
4. 불규칙 요인 : 위 세 개 모두 제거한 후 남은 예측 불가능한 변동

## 시계열 데이터 전처리
1. 데이터 확인 / 시각화
2. 결측치 / 이상치 확인 및 보간
3. 평활화 진행 : 이동평균 이용
4. 요소 분해 (개별 성분으로 분해)
- 가법모형 : 시계열 = 추세 + 계절성 + 불규칙성
- 승법모형 : 시계열 = 추세 x 계절성 x 불규칙성
1. seasonal decompose : 고정된 계절성 반영
2. STL : 시간에 따라 변하는 계설성 반영

## 정상성 판단
- 정상성 : 시계열 데이터의 통계적 특성이 시간에 따라 일정한 수준으로 유지되는 것
- 강한 정상성 : 완벽하지만 비현실적인 기준
- 약한 정상성 : 실무에서 쓰는 현실적인 기준
1. 시각적으로 정상성 파악
2. 통계적 검정 : 단위근 확인
- ADF (단위근 검정)
- KPSS (정상성 검정)

## 정상성 확보 방법
1. 추세 제거 : 차분
2. 계절성 제거 : 계절 차분
3. 분산 안정화 : 로그 변환

## 단변량 시계열 분석

- 안정적 시계열 분석
  - 자기 회귀 모델(AR) : 과거의 데이터가 미래 값을 설명
  - 이동 평균 모델(MA) : 어제의 충격이 오늘의 결과에 영향
- AR,MA의 전제조건
  1. 정상성
  2. 가역성
- ARMA
  - AR(p) 모형의 정상성 + MA(q) 모형의 가역성 모두 만족 -> 예측
- 차수 결정 방법
  - 자기 상관 함수 (ACF, PACF)
- 불안정 시계열 분석
  - ARIMA : 차분, 변환을 통해 AR, MA, ARMA로 정상화
  - SARIMA : ARIMA 모델에 계절성 변동을 반영

## 다변량 시계열 분석
- VAR : 자신의 과거 값과 다른 변수들의 과거 값을 함께 사용하여 미래를 예측하는 모델

## 딥러닝을 활용한 시계열 분석
### RNN
- 적합한 경우
  - 순차적 패턴이 중요한 시계열 데이터
  - 과거 정보가 미래 예측에 영향을 주는 문제
  - 시계열 길이와 목적에 따라 모델 선택 가능

### Tranformer
1. 장기 예측 특화
2. 주기성 포착 특화
3. TFT

### Foundation Model
- 거대한 데이터로 사전 학습하여 다양한 문제를 해결할 수 있는 범용 모델
