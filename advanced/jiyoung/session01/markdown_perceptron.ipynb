#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> 논리 게이트 </span>

##### AND, NAND, OR 논리 게이트를 쉽게 구현 가능!

논리 게이트란?
- 디지털 회로에서 기본적인 논리 연산을 수행하는 데 사용되는 구성 요소
- 게이트 연산에 쓰이는 것은 n 개의 입력값과 하나의 출력값
- AND, OR, NOT, NAND, NOR, XOR 등 다양한 종류가 있으며, 이러한 논리 게이트를 조합하여 복잡한 논리 회로를 구성할 수 있음

1. AND 게이트
: 입력값이 모두 1인 경우에만 출력값이 1이 나오는 구조



2. OR 게이트
: 입력값이 모두 0인 경우에 출력값이 0이고, 나머지 경우에는 모두 출력값이 1인 구조



3. NAND 게이트: 
입력값이 모두 1인 경우에만 출력값이 0, 나머지 경우에는 모두 출력값이 1인 구조



4. XOR 게이트: 
입력값 두 개가 서로 다른 값을 갖고 있을 때에만 출력값이 1인 구조



단층 퍼셉트론의 한계
→ XOR 게이트를 구현할 수 없다!!

## <span style="background-color:#F993C0; color:#311d26; padding:4px; border-radius:0px;"> 순전파와 역전파 </span>

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> 기본 구조 </span>

학습 알고리즘의 기본 구조
1. 순전파를 통해 입력값으로부터 예측 수행
2. 오차 계산을 통해 예측 정확도 측정
3. 역전파를 통해 출력층에서 다시 입력층 방향으로 오차 전달
4. 중치 업데이트
5. 1~4를 반복

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> 연쇄 법칙 </span>

합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다는 법칙

: 딥러닝 모델을 학습시키려면 오차를 줄여야 함. 이를 위해 역전파 과정에서 각 가중치가 최종 손실에 얼마나 기여했는지를 계산해야 하는데, 이때 연쇄법칙을 사용!

- 딥러닝 모델의 성능을 높이려면, 모델의 오차를 계산해 순전파와 역전파를 활용해서 오차를 줄이는 것이 필요!
- 손실 함수 = 오차값과 실제값을 수치로 표현
- 연쇄법칙을 통해 구하는 기울기 = ‘손실 함수를 줄이기 위해 가중치를 어떻게 바꿀까?’를 알려주는 신호

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> 기울기 소실 문제 </span>

기울기들을 계속해서 곱했는데, 결국 기울기가 0이 되는 문제
- 기울기 소실 문제가 있으면 앞쪽 층의 가중치가 거의 업데이트되지 않아서 학습이 제대로 이뤄지지 않는 문제가 생김

## <span style="background-color:#F993C0; color:#311d26; padding:4px; border-radius:0px;"> 손실 함수와 활성화 함수 </span>

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> 손실 함수의 종류 </span>

손실 함수는 모델이 정확한지 오차를 계산하는 방법 
⇒ 실제값(정답)과 예측값(출력값)의 차이를 계산하는 데에 사용되는 함수

---

1. 평균 제곱 오차 (Mean Squarred Error, MSE)   

    $\frac{1}{N}\sum(y-\hat{y})^2 \ \ \ \ \ \ (y: 실제값,\ \ \hat{y} : 예측값)$

    예측 값이 연속적이고, 예측 값과 실제 값 간의 차이를 직접적으로 측정하는 회귀 문제에 적합

2. 이진 크로스 엔트로피 (Binary Cross-Entropy Loss, BCE) 

    $L= -(y \log \hat{y} + (1-y) \log(1-\hat{y}))$   

    $y : 실제값(0,1)이진값,\ \ \hat{y} : 예측값(확률)$

    이진 분류 문제(로지스틱 회귀)에서 사용하는 손실 함수

3. 크로스 엔트로피 오차 (Cross Entropy Error, CEE)   

    $-\sum y \log \hat{y}$   

    $y : 원-핫 인코딩\ 값 (1,0),\ \ \hat{y} : 예측값(확률)$   

    다중 클래스 분류 문제에서 예측 확률이 실제 클래스와 얼마나 일치하는 지를 평가하는 데 적합


| **손실 함수** | **사용 상황** | **표현식** | **예시** |
| --- | --- | --- | --- |
| **MSE** | 회귀 문제 | $L = \frac{1}{n} \sum_{i=1}^{n}(y_i - \hat{y}_i)^2$ | 주택 가격 예측, 온도 예측, 주식 가격 예측 |
| **BCE** | 이진 분류 문제 | $L = -(y \log(\hat{y}) + (1 - y) \log(1 - \hat{y}))$ | 이메일 스팸 필터링(스팸/정상), 질병 진단(양성/음성) |
| **CEE** | 다중 클래스 
분류 문제 | $L = -\sum_{i=1}^{C} y_i \log(\hat{y}_i)$ | 손글씨 숫자 인식(MNIST), 이미지 분류(CIFAR-10) |

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> 활성화 함수의 종류 </span>

1. 계단함수 (Heaviside Step Function)

- 입력값과 가중치의 곱을 모두 더한 뒤, 그 합이 특정 기준(임계값)을 넘으면 1, 아니면 0을 출력하는 함수
- 주로 단층 퍼셉트론에서 사용

2. 시그모이드 함수
- 입력 신호를 0과 1 사이의 연속적인 값으로 변환하는 비선형 함수
- 계단 함수는 0 또는 1만 출력하는 반면, 시그모이드 함수는 x값에 따라 출력값이 연속적으로 변화

3. ReLU 함수
- 입력이 0보다 크면 그대로, 0 이하면 0을 출력하는 비선형 함수
- 단순하지만 딥러닝에서 가장 많이 사용되는 활성화 함수

4. Tanh(쌍곡탄젠트)함수
- 시그모이드 함수의 변형으로, 출력값의 범위가 -1 ~ 1
- 시그모이드(0~1)와 달리 출력값의 중심이 0이기 때문에, 입력값의 중심을 0으로 맞추고 싶을 때 사용

5. Softmax 함수
- 각 클래스에 대한 출력값을 확률로 변환하는 함수
- 다중 클래스 분류 문제의 출력층에서 사용

## <span style="background-color:#F993C0; color:#311d26; padding:4px; border-radius:0px;"> 경사 하강법 </span>

⇒ 오차를 가장 작게 만드는 가중치와 편향을 찾는 방법!

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> 배치 경사하강법 </span>

가장 기본적인 경사하강법으로, 오차를 구할 때 전체 데이터를 한 묶음으로 묶음

- 전체 데이터를 통해 학습시키기 때문에, 업데이트 횟수가 적음 (1 Epoch 당 1 업데이트)
- 전체 데이터를 모두 한 번에 처리하므로 메모리가 많이 필요
- 항상 같은 데이터 (전체)에 대해서 경사를 구하기 때문에 수렴이 안정적

## <span style="background-color:#F993C0; color:#311d26; padding:4px; border-radius:0px;"> 옵티마이저 </span>

경사하강법을 더 효율적이고 안정적으로 수행하기 위해서는 옵티마이저가 필요

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> 모멘텀 </span>

- 경사하강법 + 관성
- 경사하강법에서 계산된 접선의 기울기에 바로 한 시점 전의 접선의 기울기값을 일정 비율만큼 반영

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> RMSProp </span>

기존의 경사하강법 문제: 기울기가 큰 방향에서는 너무 많이, 작은 방향에서는 너무 작게 이동    
⇒ 따라서 각 학습률을 조정해서, 안정적이고 빠르게 학습하도록!

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> Adam </span>

RMSProp과 Momentum을 합친, 방향과 스텝 사이즈를 적절하게 합친 방법
- 각 파라미터의 1차 모멘트(기울기의 평균)와 2차 모멘트(기울기의 제곱 평균)을 사용하여 학습률을 조정
