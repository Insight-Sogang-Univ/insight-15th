## 05. 언어 모델

## <span style="background-color:#F993C0; color:#311d26; padding:4px; border-radius:0px;"> NLP </span>

자연어: 인간이 소통을 위해 자연스럽게 발생시켜 사용하는 언어

<특징>
1. 이산적 데이터
2. 순차적 데이터
3. 중의성과 모호성
4. 희소성
5. 가변성과 진화

NLP의 역할: 이 모호한 '자연어'를 컴퓨터가 이해할 수 있는 '숫자(데이터)'로 변환하여 유의미한 가치를 창출하는 기술

#### 언어 모델

가장 넓은 의미에서 언어 모델은 인간이 말하고 쓰는 방식을 모방하는 시스템을 의미
- 언어 모델은 단어 시퀀스(문장)에 확률을 할당하여, 가장 자연스러운 단어 시퀀스를 찾아내는 모델
- 언어 모델이 문장의 자연스러움을 판단하게 하는 가장 보편적인 방법은 '이전 단어들'이 주어졌을 때 '다음 단어'를 예측하도록 훈련시키는 것
- 언어 모델은 크게 통계 기반 vs. 인공 신경망 기반으로 구분


## <span style="background-color:#F993C0; color:#311d26; padding:4px; border-radius:0px;"> 통계기반 모델 </span>

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> SLM </span>

#### SLM 분포 가설

"비슷한 문맥에서 함께 나타나는 단어들은 비슷한 의미를 가진다"는 가설    
→ 즉, 어떤 단어의 주변에 자주 등장하는 단어들을 보면 그 단어의 의미를 짐작할 수 있다는 뜻!

- 조건부 확률
    - 더 높은 확률로 뒤에 올 수 있는 것을 선택

- 카운트 기반
    - 이전 단어 시퀀스의 등장 빈도를 통해 다음 단어의 확률을 계산
    - 희소 문제를 가짐(이전에 경험한 적 있는 조합에 대해서만 잘 예측할 수 있음)

SLM은 새로운 문장 조합에 약하다 → 그래서 새로운 문장 조합에도 적용할 수 있는 신경망 모델이 필요

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> N-gram </span>

#### N-gram 배경 및 한계

SLM은 희소하게 등장한 것에 대해서는 잘 예측을 못하고, 문장이 길어질수록 다음 단어를 계산하기 어려움

전체 문장 대신에 문장의 일부만 보는게 어떨까? 해서 나온 것이 N-gram



N-gram의 한계: 
1. 희소 문제
    - 데이터 부족: 훈련 데이터에서 나타나지 않은 단어 조합(N-gram)은 확률이 0이 되어, 해당 단어 시퀀스를 예측하거나 평가할 수 없음
    - N의 크기(n-gram)가 커질수록 훈련 데이터에서 해당 조합을 찾기 어려워져 희소 문제가 심해짐
2. 문맥 제한
    - 앞선 N-1개의 단어만 고려하므로, 문장 앞부분에 나온 정보를 뒷부분 결정에 반영하지 못함

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> PPL </span>

언어 모델이 문장을 얼마나 "혼란스러워하는지(perplexed)"를 수치로 나타낸 것

- 문장 W가 희소하게 등장할수록 PPL이 커지고, 단어 개수 N이 커질수록 PPL이 낮아짐
- PPL이 낮다 = 모델이 예측에 덜 혼란스러움 = 모델이 문맥을 잘 이해함 = 모델의 성능이 좋음

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> NLM </span>

#### SLM과의 차이



→ 통계적 모델(SLM)이 단어를 단순히 '이름'으로만 기억했다면, 신경망 모델(NLM)은 모든 단어를 거대한 다차원 지도 위의 한 점(좌표)으로 보냄 = 워드 임베딩!

| **특징** | **통계적 모델 (N-gram)** | **신경망 모델 (NNLM)** |
| --- | --- | --- |
| **미학습 단어 조합** | 확률 0 (예측 불가) | 유사 단어를 통해 확률 추론 가능 |
| **저장 공간** | 데이터가 늘수록 기하급수적 증가 | 가중치(Weight)만 저장하면 됨 |
| **의미 파악** | 불가능 (단순 빈도수) | 가능 (임베딩 벡터 간 거리 계산) |

## <span style="background-color:#F993C0; color:#311d26; padding:4px; border-radius:0px;"> 딥러닝기반 모델 </span>

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> BERT </span>

BERT = Bidirectional Encoder Representations from Transformers    
- Bidirectional : 텍스트의 앞뒤 문맥을 동시에 고려해서
- Encoder : 입력값을 벡터로 바꿔주는
- Transformers : 딥러닝 모델

[BERT의 특징]
- 양방향: 문장을 읽을 때 앞뒤를 동시에 고려
    → 한 단어를 이해할 때 그 단어의 좌우 문맥을 동시에 보고 이해
- 트랜스포머 기반: 트랜스포머 구조의 encoder 부분만 사용
- 사전 학습 + 파인 튜닝: 큰 데이터로 먼저 일반적인 언어 능력을 배우고, 나중에 특정 과제에 맞게 살짝 조정

[BERT의 기본 구조]   
transformer의 encoder를 여러 개 쌓아 올린 구조



[처리 과정].   
input → Embedding → 트랜스포머 인코더 12개(or 24개) → Output

[사전 학습].   
(1) Masked Language Model(MLM)
- 빈 칸 채우기 방식으로 학습



(2) Next Sentence Prediction(NSP)
- 두 문장에 대해, 이 문장들이 서로 앞뒤 관계인지, 아니면 그냥 무작위 위치에 있는 두 개의 문장인지를 학습

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> GPT </span>

BERT는 transformer의 encoder 기반, GPT는 decoder 기반

-  GPT도 지금까지의 단어를 바탕으로 다음 단어를 예측
- GPT 3부터 few-shot learning이 도입

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> LLM </span>

'Large Language Model'의 줄임말로, 우리말로는 '거대 언어 모델'
- 방대한 양의 데이터를 학습하여 자연어 및 기타 유형의 콘텐츠를 이해하고 생성하여 광범위한 작업을 수행할 수 있는 기초 모델의 범주
- ex) ChatGPT,LLaMA, PALM, Bard, Gemini 등

[LLM의 작동 방식]



(1) 대규모 학습   
(2) 예측 및 생성

[LLM의 한계]

- hallucination → GPT에서 발생했던 유명한 문제였는데, 지금은 이 문제가 해결됨
- 업데이트 비용 → LLM은 파라미터 수가 엄청나게 많기 때문에 새로운 정보를 반영할 때 계산량이 많음

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> RAG </span>

외부에서 최신 정보를 가져와서 답변하게 하기 위한 것이 RAG!

#### RAG의 구조



- 질의 인코더(Query encoder): 사용자의 질문을 벡터로 바꿉니다.
- 지식 검색기(Knowledge Retriever): 인코딩된 질문을 바탕으로 외부 지식에서 관련 내용을 검색합니다.
- 지식 증강 생성기(Knowledge Augmented Generator): 검색한 내용을 참고해서 답변을 생성합니다.

[RAG의 장점]

- 최신 정보 반영 가능
- hallucination 문제 감소
- 정보 정확성 증가

#### CAG
- RAG의 한계를 보완하기 위해 제안된 프레임워크   
+) 외부 문서를 검색해 모델에 제공 → 지식 부족 문제와 hallucination 어느 정도 완화
- 검색된 문서 중 일부가 신뢰성이 낮거나 잘못된 정보일 경우 생성된 답변의 품질과 정확도가 크게 하락
- CAG는 이러한 잘못된 컨텍스트가 모델 성능에 미치는 부정적 영향을 최소화하고, 모델이 문서를 단순히 활용하는 수준을 넘어 문서의 신뢰도를 스스로 판단하고 활용할 수 있도록 훈련시키는 기법

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> Langchain </span>

#### 아이디어와 특징

- LLM을 기반으로 앱 제작 시 필요한 구성 요소를 쉽게 연결
- 파이썬과 JavaScript 라이브러리 제공 → 다양하게 활용 가능

[LangChain의 특징]
- 추상화: LLM 서비스를 만들 때 필요한 각종 작업을 간결하게 표현(간소화)
- 표준화: 비슷한 기능을 갖추고 있는 요소들을 똑같은 형식을 갖춘 컴포넌트로 표준화
- 체이닝: 자주 활용하는 주요 컴포넌트를 쉽게 연결해서 LLM 서비스의 로직을 쉽게 파악 (Input → 컴포넌트 → Output)

## <span style="background-color:#F993C0; color:#311d26; padding:4px; border-radius:0px;"> sLM </span>

소규모 언어 모델(sLM)은 자연어 콘텐츠를 처리, 이해 및 생성할 수 있는 AI 모델

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> 모델 압축 </span>

(1) 가지치기    
신경망에서 중요도가 더 낮거나 중복되거나 불필요한 매개변수를 제거



(2) 양자화   
고정밀 데이터를 저정밀 데이터로 변환



(3) 저차원 분해   
큰 가중치 행렬을 더 작은 저차원 행렬로 분해



(4) 지식 증류   
사전 학습된 "교사 모델"의 학습 내용을 "학생 모델"로 이전



#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> 장단점 </span>

[장점]   
성능적 측면
- 효과적인 성능: 작은 모델을 해당 업무에 맞게 fine-tuning하면 필요한 기능만 빠르고 안정적으로 수행 가능
- 지연 시간 감소: 모델 크기가 작기 때문에 계산량 ↓
- 보안: 작은 모델이라서 기업 내부 서버나 개인 기기에서 직접 실행하기가 상대적으로 쉬움 → 민감한 데이터를 외부 클라우드 API로 보내지 않아도 됨

비용적 측면
- 접근성: sLM은 고성능 GPU 서버가 없어도 사용 가능→ 진입 장벽 ↓
- 효율성: 저장 공간, 메모리, 연산 자원을 적게 사용 → 같은 작업을 처리할 때 더 적은 자원으로 결과 도출
- 에너지 효율: 계산량이 적으므로 전력 소비 ↓
- 비용 절감: 모델이 작기 때문에 추론 비용, 서버 비용, GPU 비용, API 호출 비용 ↓

[한계]
- 편향: 더 큰 모델에 존재하는 편향까지 함께 전달 가능 → 성능 저하로 이어질 수 있음
- 제한된 일반화 & 복잡한 작업에서의 성능 저하: 특정 작업에 대해 미세 조정하기 때문에 포괄적인 지식이 필요한 작업에 취약
- 할루시네이션(환각): 정보가 적고 추론 능력이 낮으므로 당연히 발생 가능

#### <span style="background-color:#633a4c; color:#fde9f2; padding:4px; border-radius:0px;"> sLM+LLM </span>

복잡하고 어려운 작업에는 LLM, 가볍고 간단한 작업에는 SLM을 적절히 배분해서 사용
