## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 1. 인공신경망 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.1 인공신경망의 개념 </span>
#### 0. 인공지능 > 머신러닝 > 딥러닝
- **인공지능** : 사고나 학습 등 인간이 가진 지적 능력을 컴퓨터를 통해 구현하는 기술
- **머신러닝** : 컴퓨터가 스스로 학습해 인공지능의 성능을 향상 시키는 기술
- **딥러닝** : 인간의 뉴런과 비슷한 인공신경망 방식으로 정보를 처리 (가지돌기 -> 세포체 -> 축살돌기 -> 시냅스)

#### 1. **인공신경망** (Artificial Neural Network)
1. **개념** : <span style="background-color:yellow">우리 뇌의 신경망을 모방한 모델</span>
   - <span style="color:blue">신경망</span> : 뇌를 구성하는 신경 세포끼리 서로 연결되어 정보를 처리하는 구조
2. **인공신경망의 연결 방식**
   - **입력**들$x_1, x_2, …, x_n$을 받아 각각 **가중치**$w_1, w_1, …, w_n$와 **편향**$b$ 적용
   - 종합한 값을 **활성화 함수**에 넘김. 활성화 함수는 **임계값**을 기준으로!
      - 기준 미달 -> **0에 가깝게** 죽여서 내보냄
      - 기준 통과 -> **유의미한 값**으로 살려서 내보냄
   - **output** 출력

#### 2. **퍼셉트론** 
1. **개념** : <span style="background-color:yellow">초기 형태의 인공신경망, 신경망(딥러닝)의 기원이 되는 선형 이진 분류기</span><br>
2. **퍼셉트론의 출력값을 구하는 절차**
   1. <span style="color:blue">입력</span> : 입력값과 가중치를 곱함
   2. <span style="color:blue">가중합</span> : 곱한 값들의 총합을 구함 (편향 포함은 옵셥)
   3. <span style="color:blue">활성화 함수 & 출력</span> : 총합이 임계값을 넘으면 1, 안 넘으면 0 출력
      1. 이 경우, 활성화함수는 step 함수 (어떤 활성화함수를 쓰느냐에 따라 출력이 달라짐)
      2. training : 원하는 출력값을 내보내도록 가중치를 조정하는 작업

#### 3. **용어 정리** 
1. **가중치**($w_1, w_1, …, w_n$)
   1. 역할 : 입력값이 출력값에 주는 영향력 결정
   2. 각 입력값 당 각각의 가중치 O
2. **활성화함수**
   1. 퍼셉트론은 1 or 0을 출력하므로 **계단 함수**(Step)를 사용
      1. 다른 신경망들은 계단 함수 외에도 다양한 활성화함수 사용
          * <span style="color:green">ex. Sigmoid, Tanh, ReLU, leaky ReLU ...</span>
  
![활성화함수 모양](image.png)

3. **임계값**(${\theta}$)
   1. 활성화함수에 사용되는 기준 값
4. **편향**
   1. <span style="color:blue">임계값을 좌변으로 넘기면 편향으로 표현 가능</span>
   2. 그림으로 표현 시 입력값 1에 편향 $b$가 곱해진 형태
   3. 즉, 임계값과 편향은 표현의 차이!
$$\begin{aligned}&\text{if } \sum_{i}^{n} W_i x_i + b \ge  0\rightarrow y=1 \\&\text{if } \sum_{i}^{n} W_i x_i + b < 0\rightarrow y=0\end{aligned}$$

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.2 단층 퍼셉트론 </span>

#### 1. **개념**
- **입력층**(값을 보내는 단계)) -> **출력층**(값을 받아서 출력하는 단계)으로만 이루어진 단순한 퍼셉트론
- Keras를 활용해 단층 퍼셉트론 구현한 코드
    ```python
    from tensorflow.keras.models import Sequential
    from tensorflow.keras.layers import Dense

    # 층을 한 층씩 순서대로 쌓는 함수
    model = Sequential()

    # 4개의 출력: Dense(4,...), 2개의 입력: input_dim=2
    model.add(Dense(4, input_dim=2, activation='softmax'))
    ```

#### 2. **논리 게이트 구현 (Logic Gate)**
- <span style="background-color:yellow">논리게이트란</span>
  - 디지털 회로에서 기본적인 논리 연산을 수행하는 데 사용하는 구성 요소
  - 게이트 연산에 쓰이는 것은 n개의 입력값과 하나의 출력값
  - 종류 : `AND`, `OR`, `NOT`, `NAND`, `NOR`, `XOR` 등
- 단층 퍼셉트론을 이용하면 AND, NAND, OR의 논리게이트를 쉽게 구현 가능

#### 3. **대표적인 논리게이트**
1. **AND 게이트** : 입력값이 모두 1인 경우에 출력값이 1이 나오는 구조
2. **OR 게이트** : 입력값이 모두 0인 경우 출력값이 0, 그 외에는 모두 출력값이 1인 구조
3. **NAND 게이트** : 입력값이 모두 1인 경우 출력값이 0, 그 외에는 모두 출력값이 1인 구조
4. **XOR 게이트** : 입력값 두개가 서로 다른 값을 갖고 있을 때에만 출력값이 1인 구조

#### 4. **단층 퍼셉트론의 한계**
> XOR 게이트 구현 X <br>
- <span style="color:blue">**퍼셉트론** = 입력값과 가중치의 선형 결합
- **단층 퍼셉트론**은 직선 하나로 두 영역을 나눌 수 있는 **선형 분류 문제**에 대해서만 구현 가능한데, <br> XOR 게이트는 직선이 아닌, **곡선/비선형 영역**으로 분리해야만 구현할 수 있음 <br>-> 따라서 다중 퍼셉트론이 필요!

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.3 다층 퍼셉트론 (MLP) </span>
#### 1. **개념**
- 입력층와 출력층 사이에 **하나 이상의 은닉층**이 추가된 퍼셉트론
  - **심층 신경망**(DNN) : 두 개 이상의 은닉층을 갖는 MLP
  - **딥러닝** : 심층 신경망을 학습시킴
- Keras를 활용해 단층 퍼셉트론 구현한 코드
    ```python
    from tensorflow.keras.models import Sequential
    from tensorflow.keras.layers import Dense

    model2 = Sequential()

    # 5개의 입력과 10개의 출력, 활성화함수는 'relu' 
    model2.add(Dense(10, input_dim=5, activation='relu'))

    # 이어서 7개의 출력, 활성화함수는 'relu' 
    model2.add(Dense(7, activation='relu'))

    # 이어서 4개의 출력, 활성화함수는 'softmax' 
    model2.add(Dense(4, activation='softmax'))
    ```
- 즉, **XOR 게이트**도, 기존의 AND, NAND, OR 게이트를 조합해, 즉 **퍼셉트론의 층을 쌓아**(다층 퍼셉트론으로) 구현 가능

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 2. 딥러닝 모델 학습 방법 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.1 순전파와 역전파 (Forward&Backward Propagation)</span>
#### 1. **학습 알고리즘의 기본 구조**
1. **순전파**를 통해 입력값으로부터 예측 수행
   1. 이동 과정 : 입력층 -> 정답
2. **오차(손실) 계산**을 통해 예측 정확도 측정
3. **역전파**를 통헤 출력층에서 다시 입력층 방향으로 오차 전달
   1. 이동 과정 : 정답 -> 입력층
4. **가중치** 업데이트
5. 1-4 반복 -> 반복을 통해, 오차값이 0에 가까워지게 됨

#### 2. **연쇄법칙** 
- 순전파/역전파의 과정이 일어나는 방식
- **개념** : 합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다
  - <span style="color:blue">즉, 입력값이 최종 출력값에 얼마나 영향을 미쳤는지 계산하기 위해 <span style="background-color:yellow">각 층이 다음 층에 미친 영향</span>을 순서대로 곱해나가야하기 때문</span>
- **목적** : 딥러닝 모델의 성능 향상을 위한 모델의 **오차 감소**를 위해 역전파 과정에서 **각 가중치가 최종 손실에 얼마나 기여**했는지 계산할 때 사용
- **계산 방법** : 각 은닉층마다 손실에 주는 영향을 곱해 최종 손실에 대한 값을 나타냄
  - 즉, **A→C의 영향** = B→C의 영향 × A→B의 영향
$$\frac{∂C}{∂A}=\frac{∂C}{∂B}×\frac{∂B}{∂A}$$
  - 은닉층이 늘어나도, <span style="color:blue">모든 층에 걸친 기울기를 연쇄적으로 곱해나가는 방식</span>으로 계산
- <span style="background-color:yellow">**손실함수**</span> : 오차값과 실제값을 수치로 표현한 것
- **연쇄법칙을 통해 구하는 기울기** : 손실함수를 줄이기 위해 가중치를 어떻게 바꿀까?를 알려주는 신호로서 기능

#### 3. **기울기 소실 문제** 
- **개념** : <span style="background-color:yellow">각 층의 기울기(미분값)이 1보다 작아 연쇄작용을 통해 곱해진 기울기가 0에 가까워지는 문제</span>
- **영향** : 앞쪽 층의 가중치가 거의 업데이트되지 않아 학습이 제대로 이뤄지지 않음
  - 은닉층의 수가 많아질 수록, **앞쪽 층의 영향력이 거의 사라지게 된다**는 것

#### 4. **딥러닝 모델의 학습 흐름 정리** 
1. <span style="background-color:yellow">**순전파**</span> = 예측 하기
   - 입력값이 각 층을 통과하면 예측값이 산출
   - 입력 x → [가중치 w1] → [가중치 w2] → 예측값 ŷ
2. <span style="background-color:yellow">**손실 계산**</span> = 얼마나 틀렸는지 측정
   - 예측값과 실제 정답을 비교해 오차(손실) 계산
   - 손실 L = (예측값 ŷ - 실제값 y)²
3. <span style="background-color:yellow">**역전파**</span> = 책임 추적 (연쇄법칙 사용)
   - 계산된 손실을 바탕으로, 출력층부터 거슬러 올라가며 각 가중치가 손실에 얼마나 기여했는지 계산
   - 손실 L ← [가중치 w2의 책임은?] ← [가중치 w1의 책임은?]
     - w1의 책임을 구하는 방법 : ${w_1이 w_2에 준 영향} \times {w_2가 L에 준 영향}$
4. <span style="background-color:yellow">**가중치** 업데이트</span>
   - 계산값을 바탕으로 가중치 수정 -> 반복을 통해 정확도 향상

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.2 손실 함수 </span>

#### 1. **개념**
- 모델의 예측이 얼마나 틀렸는지, 즉 **오차를 수치로 표현**하는 방법
  - **실제값(정답)과 예측값(출력값) 간의 차이**를 계산하는 데 사용되는 함수
- <span style="color:green">종류 : 평균 제곱오차(MSE, 회귀), 크로스 엔트로피 오차(분류)</span>

#### 2. **종류**
1. <span style="background-color:yellow">**평균 제곱오차**(MSE)</span>
   - `회귀 문제`에 적합 : 예측값이 연속적, 예측값과 실제값 간의 차이를 직접적으로 측정
$$\frac{1}{N}\sum(y-\hat{y})^2 \ \ \ \ \ \ (y: 실제값,\ \ \hat{y} : 예측값)$$
   - <span style="color:grey">주택 가격 예측, 온도 예측, 주식 가격 예측</span>

2. <span style="background-color:yellow">**이진 크로스 엔트로피**(BCE)</span>
   - `이진 분류 문제(로지스틱 회귀)`에 적합
     - 실제 정답이 1일 때
       - 예측 확률이 1에 가까울 수록 loss값은 0에 가까워짐 -> 잘 맞힘
       - 예측 확률이 0에 가까울수록 → loss가 급격히 증가 (확신했으나 틀리게됨)
     - 실제 정답이 0일 때
       - 예측 확률이 0에 가까울수록 loss값은 0에 가까워짐 = 잘 맞힘
       - 예측 확률이 1에 가까울수록 → loss가 급격히 증가 (확신했지만 틀리게 됨)
   - MSE는 loss의 크기만 보지만, BCE는 **얼마나 확신을 갖고 예측**했는지도 loss에 반영
     - 따라서 낮은 확률로 예측해서 맞춘 경우 + 높은 확률로 예측해 틀린 경우 = loss가 더 큼
$$L= -(y \log \hat{y} + (1-y) \log(1-\hat{y}))$$
   - <span style="color:grey">예시 : 이메일 스팸 필터링(정상/비정상), 질병 진단(양/음)</span>

3. <span style="background-color:yellow">**크로스 엔트로피 오차**(CEE)</span>
   - `다중 클래스 분류 문제`에 적합 : 예측확률이 실제 클래스와 얼마나 일치하는지 평가
     - 두 확률 분포(진짜 분포/예측 분포)의 유사도 측정에 유용
     - 이진 크로스 엔트로피와 거의 유사하나, CEE는 **다중분류**에서 사용
   - 공식에 -가 붙는 이유 : log에 0-1사이의 확률값을 넣기 때문
$$-\sum y \log \hat{y}$$
   - 예시 : 손글씨 숫자 인식, 이미지 분류 등

4. 코드
```python
import torch
import torch.nn as nn

riterion = nn.BCEWithLogitsLoss() # 이진 분류
criterion = nn.CrossEntropyLoss() # 다중 분류
```

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.3 활성화 함수 </span>

#### 1. **개념**
- 각 층의 출력값에 **비선형성**을 추가해 신경망이 복잡한 패턴을 학습할 수 있게하는 함수
- **목적** : 선형 반복의 계산망으로는 복잡한 패턴(XOR 등)을 학습할 수 없기 때문

- 단층/다층 퍼셉트론에서의 구분

    |             | 단층 퍼셉프론          | 다층 퍼셉트론                  |
    |-------------|------------------------|--------------------------------|
    | 활성화 함수 | 계단 함수              | Sigmoid, ReLU 등               |
    | 출력값      | 0 또는 1 (이진 출력)   | 연속적인 값 (확률 또는 실수)   |
    | 특징        | 선형 문제만 풀 수 있음 | 복잡한 비선형 문제도 학습 가능 |

#### 2. 활성화 함수의 종류
1. <span style="background-color:yellow">**계단함수 (Heaviside Step Function)**</span>
   - **개념** : 입력치와 가중치의 곱을 모두 더한 뒤, 그 합이 특정 기준(임계값)을 넘으면 1 or 0을 출력하는 함수
     - 입력 신호 : 입력값과 가중치의 곱
     - 출력 신호 : 0 or 1
   - **입력값** : 뉴런이 얼마나 쉽게 활성화되는지 제어
   - **가중치** : 각 입력 신호가 결과에 미치는 영향력 제어
$$h(x) = \begin{cases} 1 & (x \geq 0) \\ 0 & (x < 0) \end{cases}$$

2. <span style="background-color:yellow">**시그모이드 함수**</span>
   - **개념** : 입력 신호를 0-1 사이 연속적인 값으로 변환하는 비선형 함수
   - 계단함수와 비교
     - 계단함수는 0 or 1만 출력 <-> 시그모이드 함수는 x값에 따라 출력값이 연속적으로 변화
       - -> 맞다 아니다가 아닌, **얼마나 확실한지 표현** 가능
   - 기울기 소실 문제 발생 가능성 O (입력값이 매우 크거나 작을 경우)
$$f(x)=\frac{1}{1+e^{−x}}$$

3. <span style="background-color:yellow">**ReLU 함수**</span>
   - **개념** : 입력이 0보다 크면 그대로, 0이하면 0을을 출력하는 비선형 함수
   - 딥러닝에서 가장 많이 사용되는 활성화 함수
     - 계산이 간단&효율 (0보다 크냐 아니냐만 수행)
     - 은닉층에 **비선형성** 제공
     - 기울기 소실 문제 완화 (ReLU는 양수 구간에서 미분값이 항상 1이기 때문)

4. <span style="background-color:yellow">**Tanh(쌍곡탄젠트) 함수**</span>
   - **개념** : 시그모이드 함수의 변형, 출력값의 범위 : -1<x<1
   - 출력값의 중심 = 0 -> **입력값의 중심을 0으로** 맞추고 싶을 때 사용
     - 중심이 0이면 학습 과정에서 가중치 업데이트가 더 안정적으로 이루어짐
   - 기울기 소실 문제 발생 가능성 O (입력값이 매우 크거나 작을 경우)
$$f(x)=tanh(x)=\frac{e^x-e^{−x}}{e^x+e^{−x}}$$

5. <span style="background-color:yellow">**Softmax 함수**</span>
   - **개념** : 각 클래스에 대한 출력값을 확률로 변환
   - 다층 클래스 분류 문제의 출력층에서 사용
     - 출력값이 모두 0-1 사이, 전체 합이 1 -> 각 클래스에 속할 확률을 직접 읽을 수 O
     - **`크로스 엔트로피 손실 함수`와의 조합**이 학습 효율이 높음

6. **총정리**

    | 활성화함수 |   출력 범위  |    주요 사용처   |              장점             |             단점             |
    |:----------:|:------------:|:----------------:|:-----------------------------:|:----------------------------|
    | 계단 함수  | 0 또는 1     | 단층 퍼셉트론    | 구조가 단순                   | 미분 불가                    |
    | Sigmoid    | 0 ~ 1        | 이진 분류 출력층 | 확률로 해석 가능              | 기울기 소실 문제             |
    | Tanh       | -1 ~ 1       | 은닉층           | 출력 중심이 0으로 학습 안정적 | 기울기 소실 문제             |
    | ReLU       | 0 ~ ∞        | 은닉층 전반      | 계산 빠름, 기울기 소실 완화   | 음수 입력 시 출력이 항상 0   |
    | Softmax    | 0 ~ 1 (합=1) | 다중 분류 출력층 | 클래스별 확률 직접 출력       | 출력층 전용, 은닉층엔 부적합 |

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.4 경사하강법 </span>

#### 1. **기초 지식**
- **경사하강법(Gradient Descent)** : 손실함수를 최소화, 즉 오차를 가장 작게 만드는 가중치와 편향을 찾아 모델을 학습
- 딥러닝에서도 경사하강법을 사용하나, 데이터가 너무 방대까지 때문에 **배치**를 사용!
1. <span style="background-color:yellow">**배치**</span>
   - **개념** : 학습 과정에서 가중치와 같은 매개변수를 조정하기 위해 사용하는 **데이터의 묶음**
     - 사용자가 임의로 정할 수 있는 하이퍼파라미터
   - **특징**
     - 배치 크기가 너무 크면, 학습이 느려지고 메모리가 부족해질 수 O
     - 배치 크기가 너무 작으면, 가중치가 자주 업데이트되다보니 학습이 불안정해질 수 O
2. <span style="background-color:yellow">**에포크(Epoch)**</span>
   - 개념 : 모든 데이터셋을 학습하는 횟수
3. <span style="background-color:yellow">**이터레이션(Iteration)/스텝(Step)**</span>
   - 개념 : 한 번의 에포크를 끝내기 위해 필요한 배치의 수
   - 1 에포크에서 다루는 데이터의 수 = 배치의 크기 * 이터레이션(Step)

#### 2. **배치 경사하강법(GD)**
- **개념** : 기본 경사하강법, 오차를 구할 때 전체 데이터를 한 묶음으로 묶음
  - 즉, 배치 사이즈 = 데이터 사이즈
- **특징**
  - 업데이트 횟수가 적음 (1에포크 당 1업데이트)
  - 많은 메모리가 필요 (전체 데이터를 한 번에 처리하기 때문)
  - 안정적인 수렴 (항상 같은 데이터에 대해 경사를 구하기 때문)
- 코드
    ```python
    model.fit(X_train, y_train, batch_size=len(X_train))
    ```

#### 3. **배치 크기가 1인 확률적 경사하강법(Stochastic Gradient Descent, SGD)**
- **개념** : 랜덤으로 선택된 하나의 데이터에 대해 계산
  - 기존 배치 경사하강법보다 더 적은 데이터를 사용하므로 빠르게 계산 가능
- **특징**
  - 하나의 데이터에 대해서만 메모리를 저장하면 되므로 자원이 적은 컴퓨터에서도 쉽게 사용 가능
  - 매개변수의 변경폭이 불안정, 낮은 정확도 가능성
- 코드
    ```python
    model.fit(X_train, y_train, batch_size=1)
    ```

#### 4. **미니 배치 경사하강법**
- **개념** : 가장 많이 사용, 배치 크기를 적절히 지정해 해당 데이터 개수만큼에 대해 매개 변수의 값을 조정
  - 배치 경사하강법보다는 빠르고, SGD보다는 안정적
- **배치 사이즈 결정법**
  - 일반적으로 $2^n$에 해당하는 숫자로 선택
  - 별도로 크기 지정 X 시 기본값은 32
- 코드
    ```python
    model.fit(X_train, y_train, batch_size=128)
    ```

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.5 옵티마이저 </span>
> 경사하강법을 더 효율적, 안정적으로 수행하기 위해 필요!<br>
> 옵티마이저를 통해 더 **효과적인 최솟값**을 찾아갈 수 있음!
<br>

#### 1. **모멘텀**
- **관성**을 응용한 방법
  - <span style="color:blue">경사하강법에서 계산된 접선의 기울기에 바로 한 시점 전의 접선의 기울기값을 일정 비율만큼 반영</span>
- 따라서 **local minima**에 빠져도 관성의 힘으로 넘어갈 수 있음
- 코드 (보통 알파 0.9값을 적용해 업데이트를 조정함)
    ```python
    tf.keras.optimizers.SGD(lr=0.01, momentum=0.9)
    ```
$$변수가\,가던\,방향으로\,계속\,가도록\,속도를\,추가!$$

#### 2. **알엠에스프롭(RMSProp)**
- 기존의 경사하강법이 기울기가 큰 방향에서는 너무 많이, 작은 방향에서는 너무 작게 이동하는 점 보완
  - 각 **학습률을 조정** -> 안정적이고 빠르게 학습 가능!
- 방법
  - **최근 기울기를 참고**해 
    - **급한 경사**에서는 속도를 줄이고
    - **완만한 경사**에서는 속도 유지 or 증가
  - 기존 경사하강법은 무조건 현재 기울기만 참고한다는 점과 대비!

#### 3. **아담(Adam)**
- RMSProp + Momentom -> 방향과 학습률을 모두 고려
- 각 파라미터의 1차 모멘트(기울기의 평균), 2차 모멘트(기울기의 제곱평균)을 사용해 학습률 조정

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.6 데이터 증강 </span>

1. **필요성**
- 현실의 데이터는 변형, 변경되는 경우가 많음
- 따라서 기존 데이터에서 다양한 변형을 만들어 학습 데이터를 늘리면 더욱 효과적인 학습이 가능!

2. **개념**
- 학습 데이터셋에 일부러 다양한 변화를 가하는 것
- **목적** : 데이터 부족 문제 해결, 모델의 견고함 향상, 과적합 감소
- **적용 분야** : 비전(뒤집기, 회전, 부분 크롭, 밝기 조정 등...), 자연어 처리(동의어 대체, 문장 섞기, 번역 후 역번역 등...)
- **주의점** : 변형 데이터가 원본의 본질적 특징을 잃지 X, 태스크에 적절한 증강 방식 선택

3. **데이터 증강의 장점**
- 모델 성능 향상 : 일반화된 패턴을 익힐 수 있음
- 데이터 의존성 감소 : 특정 패턴에 과적합 X, 다양한 패턴 학습
- 과적합 완화 : 작은 데이터셋의 한계 극복
- 데이터 프라이버시 보호 : 원본 데이터의 노출 최소화하며 성능을 향상
