## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 1. 합성곱 신경망(CNN) </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.1 CNN(Convolutional NN)의 개념 </span>
1. **배경지식**
   1. <span style="color:blue">픽셀</span> : 입력 이미지의 가장 기본적인 데이터 단위
   2. <span style="color:blue">채널</span> : 픽셀의 색을 숫자로 표현하는 방식 (*grey scale, RGB, HSV 등*)
2. **CNN이 필요한 이유**
   1. 이미지는 픽셀값들이 모여서 만들어지기 때문에, 픽셀 정보들을 그냥 학습하면 컴퓨터는 원본이미지에서 물체의 특징을 학습하지 못함
   2. 사람은 이미지를 <span style="background-color:yellow">영역 위주로, 물체의 특징 위주로</span> 살핌 -> CNN은 컴퓨터가 **이미지의 공간정보**를 유지한 상태로 학습할 수 있게 해줌
3. **MLP vs CNN**
   1. <span style="background-color:yellow">**MLP**</span> : FC Layer를 활용해 분류/식별 진행
   2. <span style="background-color:yellow">**CNN**</span> : **Conv + Pool로 특징 유지** -> FC Layer를 활용해 분류/식별 진행

#### 1. **이미지 처리에 MLP가 아닌 CNN이 필요한 이유**
1. **이미지 데이터의 구조 유지 가능**
   - 이미지는 <span style="background-color:yellow">**픽셀 간 공간적 관계**</span>가 중요
     - **MLP** : 픽셀을 일렬로 펼쳐 정리
     - **CNN** : 이미지의 2차원 구조를 유지 + 주변 픽셀 관계와 시각적 특징을 학습
2. **<span style="background-color:yellow">변형</span>(위치, 크기, 회전 등)에 강함**
   - **MNP** : 이미지 변형 시 학습한 가중치가 무력화 -> 인식 성능 저하
   - **CNN**
     - <span style="color:blue">Convolution(합성곱)</span>으로 **특징**을 추출 -> 모델이 이미지의 특징 학습
     - <span style="color:blue">Pooling(풀링)</span>으로 차원 축소 -> 위치/크기 변화에 덜 민감
     - -> CNN은 **특정 영역 내 어떤 특징**이 있는지에 집중 -> 위치 변화에 강함
3. **파라미터 효율성**
   - **MNP** : 완전 연결 구조 -> 파라미터 급증, 가중치 수가 이미지 픽셀 개수만큼 생성
     - 과적합 가능성 증가, 이미지가 커질수록 학습이 느려짐
   - **CNN** : <span style="color:blue">부분 연결 구조 + Convolution 연산</span> -> 이미지 특정 피처에만 집중
     - 효율적인 파라미터 사용 가능
4. **공간 정보 보존 가능**
   - **MNP** : 이미지를 1차원 벡터로 바로 변환(flatten)
     - 이미지 공간 정보가 많이 손실됨
   - **CNN** : 다차원 이미지를 그대로 입력받음 + 공간 정보 보존하며 학습
     - <span style="color:blue">특징을 추출한 후 flatten</span> -> 손실 정도가 낮음

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2. **CNN의 구성**  </span>
#### 1. **구조**
1. **이미지의 위치별 특징 추출** : `Convolution`, `Pooling`
   1. **Convolution Layer** : 입력 데이터에 필터를 적용 -> 활성화 함수를 반영
   2. **Pooling Layer** : Conventional Layer 다음에 위치, 이미지의 공간 크기를 줄여 중요한 특징을 강조
      -> 계산 효율성 향상, 과적합 방지
   3. **하이퍼 파라미터** (`padding`, `stride`) :연산 과정에서 출력의 크기, 정보 보존 정도 조절을 위해 활용
   2. **클래스 분류** : `Fully Connected Layer`
      1. Fully Connected Layer(FC Layer) : CNN 모델의 마지막에 위치
         1. Conv+Pool - FC Layer 사이에 `Flatten` 위치

2. **CNN의 다양한 구조** 
   1. <span style="background-color:yellow">**Convolution Layer**</span>
      1. **개념** : 전체 이미지를 일정 크기 필터로 훑으며 이미지의 특징 학습
      2. **방법**
         1. 필터 or 커널이 이미지를 일정 크기의 필터로 훑으며 특징 학습
            1. **Feature Map** : input data에 filter의 수호에 따라 결정 (activation map)

   2. **Padding** : filter를 거치며 정보를 잃어버리는 것을 막기 위해 **이미지 바깥을 0으로** 둘러싸는 기법
      - **Stride = 1**: 한 칸씩 이동 (세밀한 특징 추출)
      - **Stride = 2**: 두 칸씩 이동 (빠른 처리, 크기 축소)
      → 계산량 감소, 겹치는 영역을 줄여 효율적으로 요약

   3. **Stride**(보폭) : 이미지를 몇 칸씩 점프시키며 훑을 것인지 결정
      1. **Convenlution Layer = 1**: 한 칸씩 이동 (세밀한 특징 추출)
      2. **Stride = 2**: 두 칸씩 이동 (빠른 처리, 크기 축소)
      3. → 계산량 감소, 겹치는 영역을 줄여 효율적으로 요약

   4. <span style="background-color:yellow">**Pooling**</span> : 특징 추출을 위해 이미지를 격자 형태로 분할 -> 해당 영역 안에서 가장 큰 값 or 평균값을 구하는 것
      1. **Max Pooling** : 최댓값 선택 (대부분의 선택)
      2. **Average Pooling** : 평균값 선택 

#### 2. **대표적인 CNN 아키텍처** 
1. **AlexNet** : 머신러닝이 아닌, 최초로 GPU를 이용해 학습한 대규모 CNN 모델
   1. 8개 레이어 구조 (5개 Concolutional Layer + 3개 FC layer)
   2. ReLU 함수 이용 -> 기울기 소실 문제 해소
   3. Dropout을 통해 과적합 방지
2. **VGGNet** : 이미지 인식에 강한 깊은 층의 CNN 모델
   1. Conv + Pool 여러번 반복
   2. 필터 크기 : 3 고정
   3. 전이 학습에 많이 활용됨
   4. 초반 연산량이 매우 많음
3. **ResNet** : 잔차 연결 개념을 도입해 깊은 신경망의 소실 문제를 해결한 모델
   1. 잔차 연결 : 이전 레이어의 출력을 몇개 건너뛴 다음의 레이어의 입력으로 직접 추가하는 방식
   2. 층이 깊어져도 기울기 소실 문제 X
   3. 현재까지도 많은 모델의 기반이 됨

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 2. 이미지 딥러닝에 CNN 응용 </span>
- CNN은 `Conv + Pool`의 반복으로 **특징을 학습**
- `FC Layer`를 적용하여 이미지를 **분류**
- 이때 `Conv + Pool`에 여러 방법을 적용 가능 -> 그에 따라 모델의 성능이 달라짐

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.1 이미지 분류 </span>
> 이미지를 **수치 데이터(픽셀 값)로 인식**해 사전 정의된 클래스 중 하나로 분류

- **주요 모델 특징**
  - **AlexNet**: 비교적 단순한 구조, 최초 GPU 활용 대규모 CNN, ReLU 함수로 기울기 소실 해소
    - 주의 : 작은 데이터셋에 적절, 학습 개념 설명용
  - **VGGNet**: 더 깊은 구조, 복잡한 블록 구조 적용
    - 주의 : 계산량 증대, 메모리 요구 높음
  - <span style="background-color:yellow">**ResNet**</span>: 잔차 연결(Skip Connections) 도입으로 매우 깊은 네트워크 구현 가능
    - 주의 : 과적합 주의, 배치 크기/규제 조절 필요
  - <span style="background-color:yellow">**Vision Transformer(ViT)**</span>: 트랜스포머 구조 응용 및 패치 기반 입력 처리
    - 주의 : 이미지 처리 모델 + 순차 처리 모델의 결합, 대규모 데이터 필요
- *분류 외에도 이미지에서 **특정 객체의 위치를 찾아내**는 등 다양한 작업 가능*

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.2 객체 탐지 </span>
> 이미지 내 객체 판별, 객체를 둘러싸는 **Bounding Box**의 위치 탐색

- **모델 적용 방식**
  - 구조: CNN을 통해 **Feature Map** 얻고 그 위에 **두 가지 헤드** 구성
    - <span style="color:blue">**Classification Head**</span>: 객체의 **종류**(클래스) 예측
    - <span style="color:blue">**Regression Head**</span> : 객체의 **좌표**(x, y, 가로, 세로) 예측
- **대표 모델**: *YOLO, Faster R-CNN, SSD 등*

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.3 이미지 캡셔닝 </span>
> 이미지 내 객체 & 객체들 간 관계를 파악 -> **자연어 문장**으로 표현

- **모델 적용 방식**
  - 모델 결합: 특징 추출을 위한 <span style="background-color:yellow">**CNN**</span> + 문장 생성을 위한 <span style="background-color:yellow">**RNN**</span>의 결합
- **이미지 캡셔닝 단계**
  1. **특징 추출**: CNN(ResNet-101)을 이용한 이미지 데이터 처리
  2. **단서 생성**: 어텐션 네트워크를 통한 중요 영역 가중치 부여 및 단어 추출
  3. **문장 완성**: 가중치를 바탕으로 추출된 주요 단어들을 RNN으로 조합하여 최종 문장 완성

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 3. RNN (Recurrent Neural Network) </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.1 순차 데이터의 개념 </span>
#### 1. **정의**
- <span style="background-color:yellow">**순서에 의미**가 있고, 순서가 달라질 경우 **의미가 손상**되는 데이터</span>
- **분류** : <span style="color:purple">**순차데이터** > 시간적 순차 데이터 > 시계열 데이터 (일정한 시간 간격으로 기록되는 경우)
- **순차 데이터 예시**
  1. 음성 & 오디오 데이터 : 소리의 파형은 시간에 따라 연속적으로 변화하는 순차 데이터
  2. 자연어 : 문장, 단락, 문서 등 단어의 순서가 매우 중요한 대표적인 순차 데이터
  3. 생물학적 서열 데이터 : DNA, 단백질 서열처럼 특정 순서로 배열된 순차 데이터
  4. 비디오 데이터 : 여러 이미지 프레임이 순서대로 나열된 형태의 순차 데이터

#### 2. **기존 정형 데이터와 비교**
| 구분        | 일반적인 정형 데이터                                         | 순차 데이터                                                                         |
|-------------|--------------------------------------------------------------|-------------------------------------------------------------------------------------|
| 순서 의존성 | 행이나 열의 순서가 바뀌어도 데이터의 의미가 변하지 X | 요소를 구성하는 순서 자체가 핵심 정보, 순서가 바뀌면 데이터의 의미가 왜곡됨 |
| 샘플 간관계 | 각 관측치는 서로 연관이 없다고 가정                  | 이전 시점의 데이터가 다음 시점의 데이터에 직접적인 영향을 미침.         |
| 데이터 처리 | 각 특성(나이, 키 등)의 개별적인 정보 함유량에 집중           | 데이터 간 맥락과 시간적 흐름을 파악하는 것이 중요                                   |

-> 순차 데이터의 특징을 고려하기 위해서는 기존의 머신러닝 기법이 충분하지 X<br>
-> **특별한 신경망 필요 : <span style="background-color:yellow">과거의 정보를 기억 + 순서의 의미를 학습 가능한 신경망!!</span>**


### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.2 RNN의 개념 </span>
#### 1. **RNN(Recurrent Neural Network) 정의**
> **순환 구조**를 가진, 순차 데이터(Sequence Data) 처리에 특화된 인공 신경망 -> **순환 신경망**
- **특징**
  - **시퀀스 데이터**를 입력받아 **순서 정보 유지**하며 처리
  - <span style="background-color:yellow">순환하는 **은닉층**이 매 시점의 **은닉상태**를 업데이트</span>
    - **은닉층**(신경망 구성 요소, 함수 자체)) vs **은닉상태**(은닉층을 통과한 결과값으로, 매 시점 변하는 숫자로 이우어진 벡터)
  - **이전 시점의 값을 현재 시점으로** 넘김

#### 2. **RNN 구조**
1. **구성 요소**: 
   1. <span style="color:blue">**입력층**($x_t$ : 시간 t에서의 입력)
   2. <span style="color:blue">**은닉층**($h_t$ : 시간 t에서의 은닉상태 -> 이 시점의 메모리)
   3. <span style="color:blue">**출력층**($o_t$ : 시간 t에서의 출력)<br>
2. **가중치 체계**
   - 입력 가중치($U$): 현재 입력이 은닉 상태에 주는 영향 조절
   - <span style="background-color:yellow">**순환 가중치**($V$): 이전 시점의 은닉 상태 정보를 현재로 전달</span>
     - 기존 신경망과의 차이점임!
   - 출력 가중치($W$): 현재 시점의 은닉 상태가 최종 출력에 영향을 미치는 가중치, 결과를 만드는 역할<br>
3. 정보 흐름: **이전 시점의 은닉 상태($h_{t-1}$)와 현재 입력($x_t$)을 결합**하여 새로운 상태 생성 및 다음 시점 전송
   -> 은닉 상태의 순환을 통해 과거 정보를 다음 시점으로 계속해 전달!!

#### 3. **RNN 원리 : 가중치 공유**
1. **핵심 원리**
   1. 시퀀스의 **모든 시점**에서 <span style="background-color:yellow">**동일한 파라미터($U, V, W, b$)를 반복적으로 적용**</span>
2. **가중치 공유의 이점**
   - 데이터 길이에 상관없이 **학습해야 할 가중치 수가 일정하게 유지**됨 -> 모델이 가볍고 빠름
   - 과거 정보와 현재 입력을 결합하는 **일반적 규칙**을 학습 -> 새로운 패턴에도 대응 가능

#### 4. **RNN 한계 및 장기 의존성 문제**
1. **한계점**
   - 전체 시퀀스를 **전부 읽은 후 역전파**가 이루어짐
     - 역전파 구하는 과정에서 연쇄 법칙에 의해 다음 미분값이 반복적으로 곱해짐
   - 이로 인해 생기는 문제들...
     - **기울기 소실** : 역전파 과정에서 미분값이 반복 곱해지며 기울기가 0으로 수렴함 -> 시퀀스 뒤쪽의 오차를 학습 X
     - **기울기 폭주** : 미분값이 비정상적으로 커짐 -> 가중치가 무한대로 발산
       - Gradient Clipping으로 해결 가능 *(Gradient의 상한선을 정해서 발산을 피하는 것)*
     - **느린 훈련 시간** : **순차적 계산 구조**로 인해 **병렬 처리가 불가능**한 구조적 한계
       - CNN은 이미지 각 영역에 필터 연산을 동시에 적용해 빠르게 처리가능하나, <br>RNN은 시간이라는 축 때문에 병렬 처리 X

2. **장기 의존성 문제**
   - 시퀀스가 길어질수록 **앞부분의 중요 정보를 상실** -> 긴 문장이나 시계열 분석의 정확도가 급격히 저하
     - 단기 기억력은 준수했으나, **장기 기억력에 문제**가 발생
     - *예: 문장 앞머리의 단어(예: "모스크바")를 기억해야 뒤쪽 단어를 예측할 수 있으나, 거리가 멀어질수록 정보 소실이 발생함*
   - 따라서, 아래와 같은 새로운 모델이 필요함!
     - 장기 기억력 한계 극복
     - 더 길고 복잡한 시퀀스 데이터를 안정적으로 다룰 수 있게
     - 중요한 정보는 오래 기억, 불필요한 정보는 잊는 메모리 구조

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 4. LSTM & GRU </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.1 LSTM (Long Short Term Memory) </span>
#### 1. **개념**
- RNN 한계 극복
  - 정보가 뒤로 충분히 전달되지 못하는 RNN의 기울기 소실 문제 해결을 위한 모델 구조
- <span style="background-color:yellow">**Gate**를 통해 **기억할 내용**과 **잊을 내용**을 선택 <br>-> 중요한 정보를 장기 보존

#### 2. **특징**
- **장기 기억**을 담당하는 `셀 상태`(Cell State)와 **단기 기억**을 담당하는 `은닉 상태`(Hidden State)의 두 가지 층 사용
- 3대 게이트 구성 : 게이트를 통해 필요한 정보들만 통과함
  - **Forget Gate** : 정보 얼마나 잊어버릴지 결정
  - **Input Gate** : 현재 정보를 얼마나 사용할지 결정
  - **Output Gate** : 다음 층으로 어떤 정보를 전달할지 결정

#### 3. **장단점**
- **장점**: 기울기 소실 문제를 효과적으로 완화(해결은 X) -> 이전의 중요 정보를 보존하며 은닉 상태 갱신 가능
- **단점**: 구조가 복잡 -> RNN 대비 학습 파라미터 수가 엄청 많아짐

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.2 GRU (Gated Recurrent Unit) </span>
#### 1. **개념**
- 별도의 **메모리 셀 없이 게이트 수를 줄여 LSTM을 발전**시킨 형태
- **과거 정보와 현재 정보의 반영 비율을 결정**하여 은닉 상태 업데이트

#### 2. **LSTM과의 차이점**
- 게이트 통합: 삭제 게이트와 입력 게이트를 하나의 **Update Gate로 합침**
- 이전 상태를 얼마나 무시할지 결정하는 **Reset Gate** 사용
- **게이트 수 감소(3개→2개)** -> 학습시간 감소 (성능은 유지&개선!)

#### 3. **장단점**
- **장점** : 성능과 효율성 사이의 균형이 좋음 -> LSTM 대비 학습 시간 단축
- **단점** : 긴 시퀀스 처리 시 발생하는 정보 소실 한계는 여전히 존재

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.3 LSTM - GRU 비교 </span>
| 구분 | LSTM | GRU |
| --- | --- | --- |
| **구조** | 3개 게이트 (입력, 삭제, 출력) | 2개 게이트 (업데이트, 리셋) |
| **복잡도** | 더 복잡, 파라미터 수 많음 | 단순, 파라미터 수 적음 |
| **학습 속도** | 느림 (계산량 많음) | 빠름 (계산량 적음) |
| **메모리 사용량** | 큼 | 작음 |
| **장기 의존성 처리** | 더 강력 | LSTM과 비교해 다소 약함 |
| **연구/사례 축적** | 매우 많음 | 비교적 적음 (최근 등장) |
| **성능** | 데이터가 충분하고 문제 복잡성이 높을 때 강력 | 데이터가 적거나 효율성이 중요할 때 유리 |

- 적절한 것을 고르기
   1. **LSTM**
      1. 긴 시퀀스에서 문맥 이해가 중요한 경우, 데이터 양이 충분, 복잡한 패턴 학습이 필요, 모델 성능이 계산 효율보다 중요할 때, 기계번역/언어모델링/장기 시계열 예측 등
   2. **GRU**
      1. 자원이 제한적, 학습 속도가 중요할 때, 데이터 양이적을 때, 실시간 예측이 필요할 때, 음성인식/스트리밍데이터/실시간비디오분석/짧은 시퀀스 기반 텍스트 분류

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 5. Seq2Seq </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 5.1 Seq2Seq의 개념 </span>
- **정의**: 한 시퀀스를 다른 시퀀스로 **변환**하는 작업을 수행하는 딥러닝 모델로, 대표적으로 **기계 번역**에서 사용됨
- **특징**
    - 입력 아이템(단어)의 개수와 출력 개수가 같지 않아도 됨
    - <span style="color:blue">의미를 이해하는 과정(인코더)과 결과를 생성하는 과정(디코더)로 분리돼있음
    - 입력문의 핵심 의도를 압축한 **컨텍스트 벡터**를 통해 정보 전달함

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 5.2 Seq2Seq의 구조 </span>
1. <span style="background-color:yellow">**인코더**</span>
  - 입력된 시퀀스를 읽고 **압축**하여 원문의 특징이 담긴 **요약 정보(컨텍스트 벡터) 준비**
  - 입력 요소: $t$ 시점의 임베딩된 단어 & $t-1$ 시점의 은닉 상태(Hidden State)를 입력으로 받음
  - 출력 결과: 각 시점의 연산 결과로 **t 시점의 은닉 상태를 생성**하며, 마지막 시점의 상태가 디코더로 전달됨<br>
2. <span style="background-color:yellow">**디코더**</span>
  - 인코더로부터 전달받은 **압축된 정보**를 바탕으로 번역문 등 **새로운 시퀀스 생성**
  - 인코더가 보낸 컨텍스트 벡터를 **디코더의 첫 번째 은닉 상태**로 사용함
  - 특징: t 시점에 나온 출력값이 **다음 시점의 입력값으로 다시 사용**되어 순차적으로 문장을 완성함<br>
3. <span style="background-color:yellow">**컨텍스트 벡터**</span>
  - 인코더의 마지막 시점 은닉 상태(Hidden State), **입력 시퀀스의 모든 정보를 요약**한 float형 벡터
  - 입력 시퀀스의 요약정보
  - 차원은 직접 설정 가능 (실제 모델에서는 주로 256, 512, 1024 차원 등을 사용)<br>
4. **기본 작동**
  - 인코더와 디코더 모두 **RNN(or LSTM/GRU)을 여러 개 조합**한 형태
  - 모든 단어는 워드 임베딩 과정을 거침

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 5.3 Seq2Seq의 한계 </span>
1. **병목 현상**
  - 고정된 크기의 컨텍스트 벡터에 모든 정보를 압축해야 함 -> 정보 손실 문제 발생 -> 모델 성능 하락
2. **기울기 소실 문제 발생** : 시퀀스가 길어질수록 앞쪽 단어의 정보가 마지막까지 전달되지 못하는 현상
3. **병렬화 불가능 문제** : 데이터를 순차적으로 처리해야 하므로 시퀀스가 길어질수록 학습 시간이 늘어나는 구조적 제약 </br>-> 이 한계점들 해결을 위해 **Attention 메커니즘**이 도입! (다음주~!)
