## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">1. NLP와 언어 모델</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1. NLP (Natural Language Processing)</span>
1. **개념 : 자연어 vs 프로그래밍 언어**
   - **자연어** : 인간이 소통을 위해 자연스럽게 발생시켜 사용하는 언어 -> 일대다 대응
   - **프로그래밍 언어** : 엄격한 규칙, 논리적, 에러가 있으면 돌아가지 않음 -> 일대일 대응
2. **NLP**란? : 자연어의 의미를 분석해 컴퓨터가 처리할 수 있도록 하는 인공지능 기술 분야
   1. **NLP의 역할** : 모호한 자연어를 컴퓨터가 이해할 수 있는 <span style="background-color:yellow">숫자(Vector)로 변환</span>해 유의미한 가치를 창출
   2. NLP 활용 분야 : *음성 인식, 내용 요약, 번역, 감정분석, 텍스트 분류, 챗봇 등*
3. **분석과정**
   - <span style="color:blue">데이터수집 > 텍스트 전처리 > 텍스트 표현(수치로) > 모델 학습 및 예측 > 후처리 및 응용</span>
   - 한국어 분석 과정 : 형태소 분석 > 구문 분석 > 의미 분석 > 담화 분석
4. **자연어의 특징**
   - **이산적** -> NLP : 컴퓨터가 이해할 수 있도록 연속적인 벡터로 변환하는 과정이 필요
   - **순차적** : 단어가 나열되는 순서가 중요 -> 순차 데이터를 처리할 수 있는 모델 구조가 필요 (RNN, Transformer 등)
   - **중의성과 모호성** -> NLP는 문맥을 전체적으로 파악하는 기술이 필요
   - **희소성** -> NLP는 처음 접하는 문장이 들어와도 단어간 유사도를 이용해 확률적으로 추론하는 일반화 능력 필요
   - **가변성**과 진화 -> NLP : 고정된 규칙이 아닌, 데이터로부터 유연하게 학습 방식 필요

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">2. 언어 모델</span>
1. **개념** : 인간이 말하고 쓰는 모든 방식을 모방하는 시스템
2. **핵심 원리** : <span style="background-color:yellow">단어 시퀀스에 확률 할당 -> 가장 자연스러운 단어 시퀀스 탐색</span>
   1. 문장의 자연스러움을 판단하는 가장 보편적인 방법 : 이전 단어들을 기반으로 다음 단어를 예측하도록 훈련시키기
3. **분류** : **통계** 기반 언어 모델 vs. **인공신경망** 기반 언어 모델

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">2. 통계적 언어 모델 (SLM)</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1. Statistic Language Model(SLM)이란?</span>
> 통계 기반 언어 모델은 <span style="background-color:yellow">분포 가설</span>에 근거함

1. **분포 가설**
   - **개념** : <span style="background-color:yellow">**비슷한 문맥에서 함께 나타나는 단어들은 비슷한 의미를 가진다**는 가설</span>
     - -> 어떤 단어의 **주변에 자주 등장하는 단어들**을 보면 그 단어의 **의미를 짐작**할 수 있다!
   - `조건부 확률`, `카운트 기반 확률`이 분포가설을 설명함

2. <span style="background-color:yellow">**조건부 확률**</span>
   - **개념** : 현재까지 나타난 단어들 **다음에 더 높은 확률로 올 수 있는 것**을 선택하는 것
   - **예시** : "나는 버스를" 다음에 올 단어는 "탔다"가 자연스럽게 이어지는 경우가 더 많으므로 "탔다"를 선택

3. <span style="background-color:yellow">**카운트 기반 확률**</span>
   - **개념** : 이전 단어 시퀀스의 **등장 빈도**를 통해 다음 단어의 확률을 계산하는 것
   - **장점** : 직관적
   - **단점** : <span style="background-color:yellow">**희소성 문제**</span>
     - 이전에 경험해본 적 있는 조합에 대해서만 잘 예측할 수 있고, 새로운 문장 조합에 약함
     - 따라서 **새로운 문장 조합**에도 적용할 수 있는 신경망 모델이 필요

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">2. N-gram</span>
1. **등장 배경**
   - SLM은 희소한 조합에 대해 예측 성능이 낮음 + 문장이 길어질 수록 단어 계산을 잘 못함
   - "*전체 문장 대신 문장의 일부만 보면 어떨까?*" => <span style="background-color:yellow">**N-gram**!</span>

2. <span style="background-color:yellow">**N-gram**</span> : 문장 개수 중 `N-1`개를 참고해 다음에 올 단어를 예측하는 것
   - **확률적 선택**에 따라 더 많이 등장한 것으로 예측하는 것은 SLM과 동일
   - `N-1`인 이유 : 예측하고자하는 단어까지 합해서 N개이기 때문

3. **N-gram의 한계**
   1. **희소 문제**
      1. **데이터 부족** : 훈련 데이터에서 나타나지 않은 단어 조합은 확률이 0<br> -> 해당 단어 시퀀스를 예측/평가할 수 X
      2. **N의 크기** : N의 크기가 커질수록 훈련 데이터에서 해당 조합을 찾기 어려워짐 -> 희소 문제 심화
   2. **문맥 제한** : 앞선 `N-1`개의 단어만 고려 -> 문장 앞부분에 나온 정보를 뒷부분 결정에 반영하지 X

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">3. Perplexity(PPL)</span>
1. **개념** : <span style="background-color:yellow">언어 모델이 얼마나 **혼란**스러워하는지를 수치로 나타낸 것</span>
   - 문장 W가 희소하게 등장할 수록 PPL **증가** (데이터에서 적게 봤으므로 예측할 때 더 혼란스러워짐)
   - 단어개수 N이 커질 수록 PPL이 **감소** (희소한 부분이 있어도, 예측이 쉬운 문제들도 있으므로 PPL을 낮춰줌)
   - 즉! <span style="color:blue">PPL이 낮다 = 모델이 문맥을 잘 이해, 덜 혼란 -> 모델의 성능이 좋다!</span>
2. **신경망 기반 언어 모델에서의 PPL** : 단어, 문맥 간 관계를 학습해 **일반화**하는 능력이 뛰어남 <br> -> 희소하거나 처음 등장한 문장이 등장해도 훨씬 유연하게 대처 가능 (낮은 PPL)

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">3. 신경망 언어 모델 (NLM)</span>
1. **등장 배경** : 통게적 언어 모델의 한계를 극복하기 위해 등장
2. **키워드** : <span style="background-color:yellow">**임베딩**</span>



1. **개념** : <span style="background-color:yellow">모든 단어를 거대한 다차원 지도 위에 한 점(좌표)로 표시 = **워드 임베딩**</span>
2. **SLM v.s NLM 예시와 함께 비교** : "*재밌는 축구/농구를 한다*"
- **<span style="color:blue">통게 모델</span>** : `축구`, `농구` : 완전 다른 이름 모음 
  - 다만, 훈련 데이터에 없는 문장은 **확률이 0**으로 처리됨
- **<span style="color:blue">신경망 모델</span>** : 축구와 농구는 지도 위에서 상당히 가까움 (특징 공유 : 운동, 구기종목, 단체 스포츠)
  - NLM에서는, 단어를 기호에서 의미를 갖는 **숫자(벡터)**로 변환<br> -> 데이터에 없는 분야도 어느정도 예측할 수 O -> **일반화** 능력을 얻게 됨!

1. **한 눈에 비교하기**

    | **특징** | **통계적 모델 (N-gram)** | **신경망 모델 (NLM)** |
    | --- | --- | --- |
    | **미학습 단어 조합** | 확률 0 (예측 불가) | 유사 단어를 통해 확률 추론 가능 |
    | **저장 공간** | 데이터가 늘수록 기하급수적 증가 | 가중치(Weight)만 저장하면 됨 |
    | **의미 파악** | 불가능 (단순 빈도수) | 가능 (임베딩 벡터 간 거리 계산) |

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">4. 딥러닝 기반 언어모델</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1. Transformer의 등장</span>
1. **등장 배경** : RNN과 LSTM 한계
   1. 병렬화 불가, 기울기 소실 및 폭주, 장기 의존성 문제를 해결하기 위해
   2. 이런 문제들 때문에 신경망의 깊이를 깊게 만들지 못했음

2. **Transformer의 개념 복습**
- **트랜스포머의 구조**
  - RNN과의 차이 : RNN 셀의 반복 사용
    - 반면 <span style="background-color:yellow">트랜스포머는 **인코더와 디코더를 여러 층으로 깊게** 쌓음</span><br> => 깊어진 신경망을 통해 언어 모델의 표현력이 높아짐
  - **처리 방식 정리**
    - 인코더에서 문장 전체를 한 번에 입력받아 단어 간 관계를 계산 + 각 층을 거치며 문장 이해 고도화
    - 인코더의 최종 출력을 디코더의 각층으로 전달 -> 각 층은 이전까지 생성한 단어들과 인코더의 출력을 참고해 다음 단어 생성

3. **전이학습**
- **구성** : **사전 학습**(Pre-training) 단계, **미세 조정**(Fine-tuning) 단계
- NLP의 표준 학습 방법론

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">2. BERT</span>
1. **개념** : 트랜스포머의 **인코더** 구조를 기반으로 한 **양방향**(Bidirectional) 문맥 이해 모델
   1. 직역 : *Bidirectional Encoder Representations from Transformers*<br> : 텍스트의 앞뒤 문맥을 동시에 고려해서 입력값을 벡터로 바꿔주는 딥러닝 모델

2. **특징**
   - **양방향** : 특정 단어의 의미를 파악할 때 **좌우 문맥을 동시에 참조**하여 풍부한 의미 정보 추출
   - <span style="background-color:yellow">**트랜스포머 기반** (**인코더** 부분만 사용)</span>
   - **사전학습 + 파인튜닝** : 큰 데이터로 일단 일반적 언어능력 배우기 -> 후에 특정 과제에 맞게 조정 
     - 사전학습 : **MLM**(마스크 언어 모델)과 **NSP**(다음 문장 예측) 기법을 활용한 사전 학습 수행

3. **처리 과정**
   - input -> embedding -> 트랜스포머 인코더 12개/24개 -> output
   - BERP는 입력 문장을 임베딩한뒤, **여러 개의 인코더**를 통과하면서 단어의 **의미를 문맥적으로 풍부**하게 만듦
   - <span style="background-color:yellow">인코더만 있기 때문에, **문장 이해**를 목적으로 함!! (*감정분석, 스팸메일분류, 검색엔진 등*)</span>

4. **사전 학습**
   - <span style="background-color:yellow">**Masked Language Model(MLM)**</span>
     - 인풋 텍스트의 15%정도의 단어를 랜덤으로 마스킹해, 마스킹된 곳에 들어갈 단어를 예측 -> **문맥 이해, 추론하는 능력이 향상됨**
   - <span style="background-color:yellow">**Next Sentence Prediction(NSP)**</span>
     - 두 문장에 대해, 이 문장들이 서로 앞뒤 관계인지, 무작위 위치에 있는지 학습 -> **두 문장이 이어지는 논리 흐름 학습**

5. **이후의 모델**
   - `RoBERTa` : BERT 사전 훈련 프로세스를 개선해 성능 향상
   - `ALBERT`(A Lite BERT) : BERT 경량화 버전 (성능은 유지)

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">3. GPT</span>
1. **개념** : <span style="background-color:yellow">트랜스포머의 **디코더** 구조를 기반으로 한 **단방향**(Left-to-Right) 문장 생성 모델</span>
   - 이전 시점까지 등장한 단어들만을 바탕으로 다음에 올 가장 확률 높은 단어를 예측 (mask)
   - [GPT3~] **Few-shot learning** 도입 : 몇개의 예시를 보고 task를 수행하느냐에 따라 이름 붙임
     - 몇개 예시만으로도 새로운 태스크를 수행하는 능력 확보
     - zero-shot, one-shot, few-shot
   - *텍스트 생성, 요약, 번역, 대화 등 문장의 생성 및 생산 태스크*에 최적화.

2. **BERT vs GPT**

    | 항목                  | **BERT**                                      | **GPT**                                  |
    |-----------------------|-------------------------------------------|--------------------------------------|
    | 방향성                | 양방향 (Bidirectional)                    | 단방향 (Left-to-Right)               |
    | 어떻게?               | 한 단어를 예측할 때 앞뒤 문맥을 모두 참고 | 다음 단어를 예측할 때 앞 단어만 참고 |
    | 사용 목적             | 문장의 이해 (인식)                        | 문장의 생성 (생산)                   |
    | 대표 태스크           | 문장 분류, 질의응답, NER, NSP 등          | 텍스트 생성, 요약, 번역, 대화 등     |
    | Self-Attention 마스크 | 제한 없음 (모든 단어 참조 가능)           | Masked Attention (미래 단어 가림)    |

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">4. LLM</span>
1. **개념** : <span style="background-color:yellow">방대한 데이터를 학습하여 자연어 이해 & 생성 능력이 극대화된 기초 모델의 범주</span>
   - 주로 **GPT 방식의 디코더 구조**를 기반으로 **문장 생성**에 특화됨 
     - 사용 사례 : *텍스트 생성, 기계 번역, 질의응답, 문서요약, 감정분석 등*
   - 예 : *ChatGPT,LLaMA, PALM, Bard, Gemini* 등

2. **주요 특징**
   - **Scaling Law** : 모델 크기, 데이터량, 계산량이 일정 임계치를 넘으면 성능이 폭발적으로 향상됨 
   - **창발적 능력** : 명시적으로 배우지 않은 추론(Reasoning) 및 Zero-shot 수행 능력 발현
   - **RLHF** : 인간 피드백 기반 강화학습을 통해 AI를 인간의 의도와 가치관에 정렬

3. **작동 방식**
   1. <span style="color:blue">**대규모 학습** : 방대한 대규모 데이터 학습 -> 단어 간 관계, 문맥, 세상의 지식까지 언어의 패턴을 스스로 터득</span>
   2. **예측 및 생성** : 학습 완료 -> 질문/지시 받으면, 학습한 패턴 바탕으로 다음에 올 가장 확률 높은 단어를 순서대로 에측해 응답 생성
  
4. **한계점**
   - **Hallucination** : 그럴듯하지만 틀린 정보를 제공하는 환각 현상 (지금은 해결됨)
   - **업데이트 비용** : 방대한 파라미터 수로 인해 새로운 정보 반영 시 엄청 많은 계산량 소요 

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">5. RAG</span>
1. **개념** : <span style="background-color:yellow">모델 외부에서 최신 정보나 특정 도메인 지식을 **검색**하여 답변 생성에 반영하는 기술</span>
   1. 즉, 질문에 대해 모델이 모르면, 검색해서 답하는 구조
   - 직역 : *Retrieval-Augmented Generation* (검색 증강 생성) 

2. **구조**
   - **질의 인코더** : 사용자 질문을 수치화된 벡터로 변환 
   - **지식 검색기** : **관련성 높은 외부 지식** 검색 
   - **지식 증강 생성기** : 검색된 내용을 참고하여 최종 답변 생성 

3. **장점**
   - 최신 정보 반영 가능
   - 정보 정확성 -> 할루시네이션 문제 감소

4. **<span style="background-color:yellow">CAG</span> (Credibility-Aware Generation)**
   - RAG의 한계 보완을 위해 제안된 프레임워크
     - 검색된 문서 중 일부가 신뢰성이 낮거나 잘못된 정보일 경우 생성된 답변 품질, 정확도가 크게 하락
   - 검색된 문서의 **신뢰도를 모델이 스스로 판단**하도록 훈션시키는 기법 -> 답변 품질 향상
   - 구조 : *질문 입력 -> 정보 검색 -> 각 정보의 신뢰도 평가 -> 높은 신뢰도 정보 우선 사용 -> 최종 답변 생성*

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">6. LangChain</span>
#### 1. LangChain
1. **개념** : <span style="background-color:yellow">LLM 기반 애플리케이션 및 RAG 시스템을 쉽게 구축하기 위한 프레임워크</span>
   - LLM 앱 제작에 필요한 구성 요소들을 쉽게 **연결**하는 레고 블록 역할 수행 
   - 파이썬, 자바스크립트 라이브러리 제공 -> 다양하게 활용가능

2. **특징**
   - **추상화** : LLM 서비스 개발 시 필요한 작업들을 간결하게 표현
     - PDF 로드, 임베딩, DB 저장 등 
   - **표준화** : 비슷한 기능을 갖추고 있는 요소들을 똑같은 형식을 갖춘 컴포넌트로 표준화
   - **체이닝** : 자주 활용하는 주요 컴포넌트를 쉽게 연결 -> LLM 서비스의 로직을 쉽게 파악 (인풋->컴포넌트->아웃풋)

#### 2. LangGraph
1. **LangChain의 한계**
   - 이전 단계로 돌아가 다시 실행하는 등의 복잡한 로직 구현이 어려움 => **LangGraph 등장!**

2. **<span style="background-color:yellow">LangGraph</span>의 개념**
   - 비선형적인 작업 형태 -> 사이클, 루프 생성 가능
   - 한 번에 완벽한 답 탐색 X -> 사이클을 통해 검색
   - 구성
     - **노드** : 특정 작업을 수행하는 단위, 랭체인의 컴포넌트를 그대로 가져와 노드로 사용 가능 (무엇을 할 것인가)
     - **엣지** : 노드-노트 연결하는 선, 어떤 노드에서 다음 노드로 이동할지 결정 (다음에 어디로 갈 것인가)
       - **조건부 엣지** : 분기로직 쉽게 구현 가능

3. **LangChain vs LangGraph**

	|                **LangChain**               |                                   **LangGraph**                                  |
	|:--------------------------------------:|:----------------------------------------------------------------------------:|
	| 작업 순서가 명확하고 고정 (순차적)     | 복잡한 자율 에이전트 (순환)                                                  |
	| 복잡한 분기가 순환 로직이 필요 없을 때 | "결과가 만족스러울 때까지 계속 다시 시도" 와 같은 순환/반복 로직이 필요할 때 |
	| 빠르고 간단하게 프로토타입을 만들 때   | 사용자와 여러 번 대화를 주고받으며 상태를 계속 유지하고 관리해야 할 때       |

#### 3. LangSmith
1. **개념** : LangChain 측에서 개발한, LLM 기반 앱을 휘한 통합 개발 및 운영 플랫폼
2. **기능**
   - 디버깅&추적, 테스트&평가, 모니터링, 프롬프트 허브

#### 4. 외부 API 활용
1. LLM 내부 지식만을 이용해 해결할 수 없는 질문에 대해 외부 API와 연동해 답변할 수 O
2. **발급 방법** (OpenAI API 기준)
    - OpenAI 웹사이트 접속 및 회원가입
    - 결제 정보 등록
    - API 키 생성
    - API 키 복사 및 보관
    
- **주의 사항**
    - 생성된 API 키는 **최초 한 번만 전체를 볼 수 있음.** 창을 닫으면 다시는 전체 키를 확인 X
    - 키가 유출되면 다른 사람이 내 API KEY 를 사용하여 GPT 를 사용할 수 있으며, 결제는 제 지갑에서 결제됨<br> → 절대 키를 타인에게 공유 X

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">5. sLM (small Language Model)</span>

1. **sLM의 개념**
   - <span style="background-color:yellow">자연어 콘텐츠를 처리, 이해 및 생성할 수 있는 AI 모델</span>
   - LLM보다 규모, 범위가 **작음** (수백만~수십억 개 수준의 매개변수)
   - 모바일 앱 등 리소스가 제한된 환경이나 데이터 네트워크 없는 오프라인 추론 상황에 적합

2. **모델 압축**
   - **가지치기** : 중요도가 낮거나 중복된 매개변수(가중치, 뉴런)를 제거하여 크기를 줄이고 연산 속도를 향상
   - **양자화** : 32비트 부동 소수점 등 고정밀 데이터를 8비트 정수 등 저정밀 데이터로 변환 -> 계산 부하 및 메모리 절감
   - **저차원 분해** : 큰 가중치 행렬을 더 작은 저차원 행렬로 분해(예: SVD) -> 매개변수 수와 계산 횟수 단축
   - **지식 증류** : 사전 학습된 거대 '교사 모델'의 지식을 작고 가벼운 '학생 모델'로 이전하여 모방 학습

3. **sLM의 예**
   - *DistilBERT, Gemma, GPT-4o mini, Llama, Ministral, Phi 등*

4. **sLM의 장단점**
   - **장점**
     - **성능 및 지연 시간**: 특정 업무 맞춤형 미세 조정을 통한 빠르고 안정적인 수행 및 계산량 감소로 지연 시간 단축
     - **보안**: 외부 클라우드 API 전송 없이 기업 내부 서버나 개인 기기에서 직접 실행 가능하여 민감 데이터 보호 용이
     - **비용 및 효율성**: 고성능 GPU 없이도 사용 가능한 낮은 진입 장벽과 저장 공간, 메모리, 전력 소비 등 전반적인 유지 비용 절감
   - **한계**
     - **편향 및 일반화**: 더 큰 모델의 편향 전달 가능성 및 특정 작업 특화로 인해 포괄적 지식이 필요한 복잡한 작업에서 성능 저하
     - **할루시네이션**: LLM 대비 적은 정보량과 낮은 추론 능력으로 인해 환각 현상 발생 가능.

5. **LLM과 sLM의 결합 (결합 추론)**
   - <span style="color:blue">어렵고 복잡한 작업은 LLM이, 가볍고 간단한 작업은 sLM이 처리하도록 적절히 배분</span>
   - LLM 수준의 높은 정확도를 유지하면서도 LLM 토큰 사용량을 **최대 90%까지 절약** 가능
