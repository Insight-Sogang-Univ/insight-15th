# 01 추천 시스템 기초

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">1 추천 시스템 기초</span>

**추천 시스템 정의**
> 사용자의 정보 데이터를 분석해 개인의 취향에 맞는 아이템을 추천하는 알고리즘!!

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1 데이터의 종류와 특징</span>

#### 1) 명시적 데이터
* **정의**: <span style="background-color:yellow">사용자가 자신의 **선호도를 직접적으로 표현**해 입력한 데이터</span>
* **Ex**: 평점, 좋아요, 투표, 스케일 점수 등
* **장점**: 당시 사용자의 취향이 정확하게 반영됨
* **단점**: 모든 사용자가 리뷰나 평점을 남기지 않으므로 명백한 데이터를 얻기 어려우며, 과거의 선호도가 현재와 동일한지 알 수 없을 정도로 사람의 취향은 변함

#### 2) 암시적 데이터
* **정의**: <span style="background-color:yellow">고객들의 행동을 기반으로 **간접적으로 선호도를 수집**한 데이터</span>
* **Ex**: 조회수, 영상 시청 기간, 구매 기록, 페이지 체류 기간 등
* **장점**: 사용자가 서비스를 이용할 때 로그 데이터나 검색/클릭 기록이 무조건 남으므로 데이터를 얻기 매우 쉬움
* **단점**: 높은 수치가 무조건 높은 선호도를 의미하지는 않음 -> 사용자의 취향이나 선호도를 직접적으로 파악하기 어려움

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">2 추천시스템 분류</span>

#### 1) 내용 기반 필터링
* **개념**: <span style="background-color:yellow">**콘텐츠의 유사도**를 기반으로 사용자가 기존에 이용하거나 선택한 아이템과 유사한 아이템을 추천하는 방식</span>
* **특징**: 한 사용자의 행동 이력을 기준으로 하며, 사용자가 선호한 아이템들의 **속성(특징) 유사도**를 계산해 비슷한 아이템을 추천함
* **Ex**: "*당신이 이 제품을 좋아했다면, 이와 유사한 다른 제품도 좋아할 것!*"

#### 2) 협업 필터링
* **개념**: <span style="background-color:yellow">서비스 내 다수의 사용자로부터 얻은 선호도 정보를 기반으로 **사용자 간의 선호도를 상호 고려**해 아이템을 선택하거나 걸러내는 방식</span>
* **분류**
    * **사용자 기반 필터링**: 특정 사용자와 유사한 취향의 사용자가 선호한 아이템을 추천하는 방식<br> “*비슷한 사용자가 좋아한 상품을 추천*”하는 구조
        * ex.: 사용자 A와 B가 비슷한 상품(라면, 콜라)을 구매했을 때, 사용자 A가 구매한 생수를 사용자 B에게 추천함
    * **아이템 기반 필터링**: 사용자가 선호한 아이템과 유사한 아이템을 추천하는 방식<br> “*선호한 아이템과 유사한 다른 아이템을 추천*”하는 구조
        * ex.: 사용자 B가 콜라를 좋아할 때, 콜라를 좋아한 다른 사용자들이 생수도 자주 구매했다면 두 상품을 유사한 아이템으로 판단해 사용자 B에게 생수를 추천함

#### 3) 하이브리드 필터링
* **개념**: 내용 기반 필터링과 협업 필터링 각각의 단점을 보완하기 위해 2가지 이상의 다양한 추천 알고리즘을 조합한 방식
* **하이브리드 방식 종류**:
    * **<span style="background-color:yellow">Weighted Ensemble</span>**: 여러 모델의 추천 결과를 `Soft Ensemble`(점수) 또는 `Hard Vote` 등의 방법으로 하나로 합쳐 최종 추천 아이템을 결정하는 방식
    * **<span style="background-color:yellow">Mixed</span>**: 플랫폼 내에 특징이 다른 여러 추천 알고리즘을 활용해 그 결과를 동시에 보여주는 방식
    * **<span style="background-color:yellow">Switch<span>**: 사용자 또는 서비스 상태 등 특정 상황을 고려해 여러 추천 알고리즘의 추천 결과를 선택적으로 제공하는 방식
    * **<span style="background-color:yellow">Feature Combination</span>**: 보유 데이터에서 얻을 수 있는 다양한 Feature를 모두 조합해 알고리즘을 학습하고 추천하는 방식
    * **<span style="background-color:yellow">Meta-Level</span>**: 첫 번째 모델의 출력 결과가 다음 모델의 Input이 되는 형태로 여러 추천 알고리즘이 서로의 정보를 학습해 추천하는 방식

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">2 내용 기반 필터링</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1 사용자 프로필</span>

1. **추천 시스템 사용 데이터 지정 방식**
    * **<span style="background-color:yellow">직접 지정형</span>**: 사용자에게 선호하는 아이템의 특징을 명시적으로 지정하도록 하는 방법<br>(ex.: OTT 서비스 가입 시 제시되는 몇 개의 영화에 대한 선호도를 체크하는 경우)
    * **<span style="background-color:yellow">간접 지정형</span>**: 사용자의 과거 행동 이력에 기반해 프로필을 작성하는 방법<br>(ex.: SF 영화를 즐겨본 이력이 존재하는 유저에게 인터스텔라를 추천하는 경우)
2. **사용자 프로필 획득 및 특징**
    * 사용자의 기호를 따로 파악하거나 설문조사, 자발적인 키워드 입력, Like/Dislike, 평점, 유저가 직접 작성한 댓글 등을 통해 프로필을 획득
    * 검색 로그, 아이템 선택 후 구매 혹은 선택, 특정 아이템 유지를 위해 비슷한 아이템을 태그한 이력 등의 암묵적 데이터도 활용
    * 아이템에 대한 유저의 기존 평가를 바탕으로 분류(*Classification*) 문제를 해결하는 것과 같음
    * 좋은 사용자 프로필을 구성할수록 추천 시스템의 정확도와 만족도가 높아짐-> 명시적 방식과 암묵적 방식을 혼합해 사용하는 것이 일반적
3. **특성 가중치 계산 예시**
    * 사용자가 가지고 있는 **아이템 특성 가중치의 평균값**을 사용자 프로필 벡터로 활용함
    * 유저가 아래 영화 3편을 모두 시청한 경우를 계산해보면
        * 어벤져스: $[0.9,\ 0.1,\ 0.5,\ 0.0]$
        * 인터스텔라: $[0.3,\ 0.2,\ 0.9,\ 0.0]$
        * 노트북: $[0.0,\ 0.9,\ 0.0,\ 0.1]$
        * 사용자 프로필 = $\frac{1}{3}\bigl([0.9,0.1,0.5,0.0] + [0.3,0.2,0.9,0.0] + [0.0,0.9,0.0,0.1]\bigr) = [0.4,\ 0.4,\ 0.47,\ 0.03]$
    * -> 분석 결과 해당 유저는 액션 40%, 로맨스 40%, SF 47% 정도를 선호하는 것으로 파악
    * 신규 영화 듄의 특성 벡터가 $[0.4,\ 0.0,\ 0.9,\ 0.0]$ 일 때, 유저 프로필 벡터와의 유사도(코사인 유사도 등)가 높음 -> 최종 추천을 결정!

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">2 유사도</span>

#### 1) 유사도란?
* **정의**: <span style="background-color:yellow">영상, 이미지, 텍스트 등의 아이템 정보 데이터를 벡터화한 후 벡터 간의 거리가 얼마나 가까운지 측정하는 수학적 개념</span>
* **추천 단계**:
    1. **인풋 데이터 생성**: 영화의 장르, 제작 국가, 감독, 출연 배우 등 아이템의 특성(*Feature*)을 테이블 형태로 정리한 뒤, 기계가 이해할 수 있도록 원-핫 인코딩(속하면 1, 아니면 0)을 통해 숫자로 변환함
    2. **유저 정보 활용**: 직접 입력한 프로필이나 시청 기록 기반의 간접 정보를 활용해 유저를 $n$개의 특성(*Feature*) 벡터로 표현함
    3. **벡터 내적과 유사성 계산**: 유저와 아이템을 동일한 차원의 벡터로 표현한 후 벡터 내적(*Dot Product*)을 계산하며, 내적 값이 클수록 유사성이 높다고 판단함
* **영화 추천 사례로 보기**: 사용자(벡터 A)와 스타일이 다른 두 영화(벡터 B, 벡터 C)가 있을 때, '벡터 A - 벡터 B'의 내적이 더 크다면 사용자가 B 영화와 더 유사한 성향을 보인다고 판단해 B 영화를 추천함
* **<span style="background-color:yellow">유사도 방식의 종류</span>**:
    * **자카드 유사도**: 집합 기반 지표로 아이템 집합 간의 교집합 크기를 이용함
    * **코사인 유사도**: 두 벡터 사이의 각도를 측정한 지표로 각도가 작을수록 두 데이터가 유사함을 의미함
    * **유클리디안 거리**: 두 점 사이의 거리를 측정한 지표로 거리가 가까울수록 두 데이터가 유사함을 의미함
    * **피어슨 상관계수**: 아이템 벡터 간의 선형 상관관계를 이용해 측정함

#### 2) TF-IDF란?
* **정의**: 특정 문서 내에서 특정 단어가 얼마나 자주 등장하는지 나타내는 단어 빈도(**TF**)와 전체 문서에서 특정 단어가 얼마나 자주 등장하는지 나타내는 역문서 빈도(**IDF**)를 곱해 문서 내 단어의 가중치를 계산하는 방법
* **핵심 기능**: 다른 문서에서는 등장하지 않지만 특정 문서에서만 자주 등장하는 단어를 찾아내어 각 단어가 해당 문서에서 얼마나 중요한지 구체적인 값으로 표현함
* **수식 관계**:
    * $TF(d,t)$: 특정 문서 $d$에서의 특정 단어 $t$의 등장 횟수
    * $DF(t)$: 특정 단어 $t$가 등장한 전체 문서의 수
    * $IDF(d,t) = \log\left(\frac{n}{1+DF(t)}\right)$ ($n$은 전체 문서 수)
    * $TF\text{-}IDF = TF(d,t) \times IDF(d,t)$
* **계산 예시 (전체 문서 수 $n=3$)**:
    * 문서 A(고양이는 귀엽다), 문서 B(강아지는 귀엽다), 문서 C(고양이는 사냥한다)
    * 단어 '귀엽다'는 문서 A, B에 등장하므로 $DF=2$가 되며, 문서 A에서의 $IDF = \log\left(\frac{3}{1+2}\right) = \log(1) = 0$이 되어 여러 문서에 자주 나오는 중요도 낮은 단어로 인식됨
    * 단어 '사냥한다'는 문서 C에만 등장하므로 $DF=1$이 되며, 문서 C에서의 $IDF = \log\left(\frac{3}{1+1}\right) = \log(1.5) \approx 0.176$이 되어 다른 문서에서 덜 나타나는 중요한 단어로 인식됨
* **Scikit-learn 활용 분석 코드**:
    ```python
    from sklearn.feature_extraction.text import TfidfVectorizer
    from sklearn.metrics.pairwise import cosine_similarity

    # TF-IDF 벡터화
    vectorizer = TfidfVectorizer()
    tfidf_matrix = vectorizer.fit_transform(docs)

    # 코사인 유사도 계산
    similarity = cosine_similarity(tfidf_matrix, tfidf_matrix)

    print(similarity)
    ```
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">3 내용 기반 필터링 장단점</span>
| 구분 | 내용 |
| :--- | :--- |
| **장점** | - 아이템이 가진 특성(*Feature*) 자체에만 집중하므로 다른 유저의 평점 데이터에 영향을 받지 않음<br><br>- 새로운 유저에 대한 정보 부족으로 적절한 추천을 하지 못하는 콜드 스타트 문제(*Cold Start Problem*)에서 자유로움<br><br>- 아이템의 특성을 기반으로 추천하므로 사용자에게 추천 이유를 설명하기 용이함 |
| **단점** | - 아이템의 특성을 직접 추출해야 하므로 정확한 특성을 추출하지 못하면 추천 정확도가 크게 떨어짐<br><br>- 유저가 기존에 소비한 아이템과 유사한 아이템 위주로만 반복 추천되어 완전히 새로운 장르의 아이템을 추천하기 어려움 |


## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">3 협업 필터링</span>

* **기본 가정**
    * 과거에 비슷한 선호도를 가진 사용자들은 앞으로도 비슷한 선호도를 가질 가능성이 높음

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1 메모리 기반 협업 필터링</span>

* **정의**
    * 사용자 또는 아이템 간의 유사도를 기반으로 추천을 수행하는 알고리즘으로, *이웃 기반 협업 필터링(Neighborhood-based Collaborative Filtering)*이라고도 함
* **접근 방식 분류**
    * **사용자 기반 접근**: "당신과 비슷한 고객들이 다음 상품도 구매했습니다"
    * **아이템 기반 접근**: "이 상품을 선택한 다른 고객들은 다음 상품도 구매했습니다"

#### 1) 사용자 기반 접근
* **개념**: <span style="background-color:yellow">유저 간의 **유사도가 높을수록 가중치를 부여**하는 방식</span><br> target user와 유사한 성향을 가진 타 사용자들의 평점을 활용해 같은 그룹의 다른 유저가 선호하는 아이템을 추천함
* **예시**:
    * 사용자 A: 어바웃 타 5점, 노팅힐 4점, 아이언맨 3점
    * 사용자 B: 어바웃 타 5점, 노팅힐 5점, 반지의 제왕 2점
    * 사용자 C: 아이언맨 5점, 어벤져스 5점, 어바웃 타 2점
    * 분석 결과 영화 어바웃 타임과 노팅힐에 높은 점수를 준 사용자 A와 사용자 B는 비슷한 취향을 가짐으로써, 사용자 B가 노팅힐을 아직 보지 않았다면 추천 대상으로 선정함

#### 2) 아이템 기반 접근
* **개념**
  * <span style="background-color:yellow">사용자들이 특정 아이템을 좋아하는지 혹은 싫어하는지의 **평가 척도가 유사한 아이템을 추천 기준**으로 삼는 알고리즘</span>
  * 아이템 자체의 속성 유사도를 계산하는 내용 기반 필터링과는 다름!
* **예시**:
    * 어바웃 타임과 노팅힐은 로맨틱 코미디, 아이언맨과 어벤져스는 액션 장르의 특징을 지님
    * 사용자 C가 어바웃 타임을 낮게 평가했으므로 아직 보지 않은 노팅힐을 추천하는 대신 다른 액션 영화를 추천하는 것이 적합함
    * 반대로 사용자 A는 어바웃 타임을 좋아하므로 평가 패턴이 유사한 영화인 노팅힐을 추천하는 것이 적합함

#### 3) 두 접근 방식 비교

| 비교 항목 | 사용자 기반 협업 필터링 (UBCF) | 아이템 기반 협업 필터링 (IBCF) |
| :--- | :--- | :--- |
| **적합한 환경** | - 아이템 수가 사용자 수보다 많은 환경<br>- 사용자 유입이 적고 아이템이 수시로 변경되는 플랫폼 | - 사용자 수가 아이템 수보다 많은 환경<br>- 아이템 리스트의 변화가 적고 고정적인 환경 |
| **추천 다양성** | - 특정 사용자가 독특한 아이템을 평가한 경우 뜻밖의 독창적인 아이템 추천이 가능해 다양성과 참신성이 강함 | - 유사한 아이템 위주로 추천이 이루어지므로 다양성과 참신성이 상대적으로 부족함 |
| **계산 속도** | - 대규모 사용자 환경에서 실시간 유사도 계산 부담이 큼 | - 사용자별 개별 계산이 불필요하며 이웃 선정을 위한 계산 빈도가 낮아 빠르고 안정적 |
| **실제 활용** | - 알 수도 있는 친구 추천, 초기 음악 추천, 뉴스, 소셜 미디어 플랫폼 등 | - 사용자 유입이 많은 대형 플랫폼인 아마존, 넷플릭스 등 |

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">2 메모리 기반 협업 필터링 한계</span>

* **희소 행렬 문제 (Sparsity)**
    * 실제 유저-아이템 행렬의 대부분 원소는 비어 있는 희소 행렬(*Sparse Matrix*) 형태를 띰
    * 주어진 평점 및 선호도 데이터가 부족해 행렬 전체 원소 중 비어 있는 비율이 너무 높을 경우 추천 성능이 크게 저하됨
* **확장성 문제 (Scalability)**
    * 아이템과 유저의 수가 지속해서 증가할 경우 연산량이 급격히 늘어나 확장성이 떨어짐
    * 특정 아이템의 평점을 예측하기 위해 해당 아이템을 평가한 모든 유저와의 유사도를 매번 구해야 하므로 성능 저하를 유발함
* **콜드 스타트 문제 (Cold Start)**
    * 기존의 평점 및 행동 데이터를 기반으로 유사도를 계산하므로 데이터가 아예 없는 상황에서는 추천이 불가능함
    * **User Cold Start**: 신규 유저 유입 시 평가 이력이 없어 유사 유저 탐색 불가
    * **Item Cold Start**: 신규 아이템 등록 시 아무도 평가하지 않아 유사도 계산 불가
    * **해결 방법**: 가입 시 선호 장르를 직접 선택하는 직접 지정형 프로필 활용, 내용 기반 필터링과 혼합해 보완, 나이 및 성별 등의 인구통계학적 정보 활용

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">3 모델 기반 협업 필터링</span>

* **정의**
    * 메모리 기반 CF의 한계를 보완하기 위해 등장한 방식 관측된 평점 데이터를 학습-> 데이터 뒤에 숨겨진 패턴인 잠재 요인을 도출 + 관측되지 않은 평점을 예측(*Inference*)하는 방식

#### 1) 잠재 요인 모델 (Latent Factor Model)
* **개념**: 사용자와 아이템 사이의 관계를 설명해주는 눈에 보이지 않는 숨겨진 특성(*Feature*)을 찾기 위해 사용자-아이템 상호작용 행렬을 저차원 행렬로 분해하는 모델
* **특징**: 유저와 아이템을 동일한 차원의 잠재 공간 벡터 좌표로 표현해 거리와 각도로 취향 유사도를 파악하나, 축(차원)이 의미하는 바를 사람이 직접 해석할 수는 없음
* **수식 구조**:
    $$ R \approx Q P^T $$
    * $R$: 유저-아이템 평점 행렬 ($m \times n$)
    * $Q$: $k$차원의 유저 잠재 행렬 ($m \times k$)
    * $P$: $k$차원의 아이템 잠재 행렬 ($n \times k$)
    * 유저 $u$의 아이템 $i$에 대한 선호도 추정값 수식:
    $$ r_{ui} = q_i p_u^T = \sum_{s=1}^{k} u_{uk} \times v_{ik} $$

#### 2) 행렬 분해 (Matrix Factorization) 수행 방식

| 구분 | 접근 방식 | 특징 |
| :--- | :--- | :--- |
| **SVD** | 수학적 분해 기법 | - 행렬 대수학에 기반해 원래 행렬을 $A = U \Sigma V^T$ 구조로 분해함<br>- 희소 행렬에는 직접 적용할 수 없어 결측치를 유저 평균 평점 등으로 채우는 전처리가 필요<br>- 추천 시스템에서는 상위 $k$개의 특잇값만 사용하는 **Truncated SVD**를 통해 노이즈를 제거하고 핵심 패턴을 압축 |
| **SGD** | 최적화 알고리즘 | - 관측된 평점과 예측 평점 사이의 오차 제곱합을 최소화하고 과적합 방지를 위한 정규화(*Regularization*) 항을 추가한 목적 함수를 사용함<br>- 관측된 평점에 대해서만 학습을 진행함으로써 SVD의 데이터 희소성 및 계산 복잡성 문제를 우회적으로 해결 |
| **NMF** | 제약 조건 기반 모델 | - 분해된 두 개의 행렬 $P$와 $Q^T$의 모든 원소가 0 이상이어야 한다는 비음수(*Non-Negative*) 제약 조건을 추가함<br>- 잠재 요인 값이 음수가 될 수 없으므로 특정 속성이나 선호도의 강도를 직관적으로 해석할 수 O |
| **PMF** | 확률적 모델 | - 평점 예측 시 고정된 수치가 아닌 불확실성을 포함해 확률 분포 형태로 모델링하는 통계적 접근법<br>- 베이지안 추론 기법과 결합해 정교한 예측을 수행하며 VAE 알고리즘과 연결됨 |

# 02 딥러닝 기반 추천 시스템

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">1 VAE 기반 협업 필터링</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1 MF의 한계 & 딥러닝 등장 배경</span>

* **행렬 분해(MF)의 근본적 한계와 딥러닝의 해법**
    * **선형성의 한계**: 평점 예측($\hat{r}_{u, i}$)이 단순한 잠재 벡터의 내적(*Dot Product*)으로만 이루어져 복잡한 관계 표현이 불가능함 -> 비선형 활성화 함수(*ReLU* 등)를 가진 신경망을 사용해 복잡한 패턴을 학습함
    * **불확실성 무시**: 사용자 취향을 단 하나의 고정된 벡터값(*Point Estimate*)으로만 표현해 데이터의 오차를 반영하지 못함 -> 확률 분포로 취향을 모델링해 데이터의 불확실성을 반영
* **딥러닝 도입 의의**
    * 비선형성과 확률 분포 모델링을 통해 MF의 근본적 한계를 극복하고 추천 성능을 극대화하기 위해 **VAE 기반 알고리즘**이 등장함

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">2 AE vs VAE</span>

* **데이터 희소성과 불확실성 처리 방식의 비유**
    * *사용자 A가 본 영화가 5편에 불과한 희소 행렬 문제 상황을 가정*
    * **AE 접근**: "*A의 취향은 (액션 0.9, 로맨스 0.1) 벡터 하나*" (단일 값 추정) -> 데이터가 부족하면 단일 추정치가 불안정하고 틀리기 쉬워 노이즈에 취약해짐
    * **VAE 접근**: "*A의 취향은 (평균 액션 0.8, 표준편차 0.2)를 따르는 분포임*" (분포 추정) -> 취향의 불확실성 즉 <span style="background-color:yellow">데이터 부족으로 인한 오차 범위까지 모델링에 포함</span>시켜 안정성을 높임

* **구조 및 특성 비교**

    | 구분 | AutoEncoder (AE) | Variational AutoEncoder (VAE) |
    | :--- | :--- | :--- |
    | **잠재 벡터 표현** | - 단일 값 (*Point Estimate*) | - 확률 분포 (평균 $\mu$, 분산 $\sigma$) |
    | **데이터 불확실성** | - 반영 불가 (단순 복원에만 집중) | - 데이터의 불확실성을 직접 모델링해 안정성 확보 |
    | **추천 시스템 적용 장점** | - 비선형성 학습 가능 | - 희소한 데이터에서 더욱 안정적이며 일반화 성능이 우수함 |

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">3 VAE 기반 CF</span>

* **개념**
    * 사용자의 아이템 상호작용 벡터($x$, 예: 구매/시청 이력)를 입력으로 받아 잠재 공간에서 원래의 데이터를 재구성하도록 VAE를 학습하는 방식<br>(*Mult-VAE* 등이 대표적)
* **작동 구조 및 구성 요소**
    * **입력 ($x$)**: 사용자의 아이템 상호작용 벡터를 신경망에 전달함
    * **인코더**: 사용자 이력을 기반으로 인공신경망을 통과 -> 잠재 변수 $z$의 확률 분포로 매핑
    * **잠재 벡터 ($z$)**: 인코더가 생성한 확률 분포로부터 샘플링된 사용자의 최종 취향 벡터
    * **디코더**: 샘플링된 잠재 벡터 $z$를 입력받음 -> 원래의 아이템 분포를 복원하고 선호도를 예측

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">4 목적 함수 & KL Divergence</span>

* **<span style="background-color:yellow">목적 함수: ELBO(Evidence Lower Bound)</span>**
    * VAE의 핵심 최적화 지표로 두 가지 목표인 재구성 오차 최소화와 잠재 공간 정규화를 동시에 달성함
    $$ \mathcal{L}(x; \theta, \phi) = E_{q_\phi(z|x)} [ \log p_\theta(x|z) ] - D_{\text{KL}}(q_\phi(z|x) || p(z)) $$
    * **재구성 오차 최소화 (첫 번째 항)**: 디코더가 잠재 벡터 $z$를 사용해 원래 입력 $x$를 최대한 비슷하게 복원하도록 유도함
    * **KL Divergence 항 (두 번째 항)**: 인코더가 출력하는 잠재 분포 $q(z|x)$가 사전 분포 $p(z)$인 표준 정규 분포와 유사하도록 강제함 <br> -> 모델이 특정 데이터에 과적합 -> 잠재 공간이 불규칙해지는 것을 방지하고 연속적이며 부드러운(*Smooth*) 상태를 유지시킴

* **KL Divergence를 통한 희소성 극복 원리**

    | 모델 | 잠재 공간 상태 | 희소한 데이터 처리 방식 |
    | :--- | :--- | :--- |
    | **AE** | - 유저 벡터들이 의미 없는 공간에 제멋대로 흩어진 퍼즐 조각 상태를 형성함 | - 데이터가 듬성듬성 놓여 있어 빈 공간을 예측할 때 길을 잃고 노이즈에 매우 취약해짐 |
    | **VAE** | - 유저 벡터들을 표준 정규 분포 주변에 부드럽고 연결된 상태의 맞춤 퍼즐로 강제함 | - 데이터가 부족한 신규 유저가 등장해도 주변의 풍부한 데이터를 잇는 연속된 패턴 위에 위치하게 되어 안정적인 예측이 가능함 |
<br>

* **VAE-CF의 주요 장점 요약**
    * **확률적 잠재 분포 모델링**: 단일 포인트 벡터가 아닌 분포 형태로 사용자 선호를 정의 -> 예측 불확실성을 반영 + 노이즈 문제 해결
    * **희소 상호작용 데이터 방어**: *KL Divergence* 정규화를 통해 잠재 공간의 데이터 연결 능력을 극대화 -> 희소성 문제 대응
    * **비선형성 학습**: 심층 신경망 구조를 활용 -> 기존 선형 알고리즘인 MF의 한계를 극복 + 복잡한 패턴 정보를 안정적 학습

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">2 GNN 기반 사용자 선호 예측</span>

* **기존 잠재 요인 모델(VAE 등)의 한계**
    * 사용자-아이템 상호작용을 개별 점의 집합으로 간주해 데이터 내에 존재하는 복잡한 *관계*나 *순서* 정보를 직접적으로 활용하지 못함

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1 GNN이란?</span>

* **정의**
    * 그래프 신경망(*Graph Neural Network*)의 약자
    * 사용자-아이템 상호작용을 단순 집합이 아닌 <span style="background-color:yellow">서로 연결된 거대한 관계망인 그래프(*Graph*) 형태로 모델링해 학습하는 방법론</span>
* **그래프(Graph) 자료 구조**
    * 현실 세계의 개체를 의미하는 **노드(Node)**와 개체 간의 관계 및 상호작용을 나타내는 **엣지(Edge)**로 구성됨
* **GNN 도입 목적 및 특징**
    * 친구 관계, 소비 관계 등 추상적인 개념을 컴퓨터가 이해할 수 있는 **[노드-엣지-노드]** 형태의 명시적 데이터로 구조화함
    * 데이터가 규칙적인 격자 구조 위에 정렬된 유클리드 공간(*Euclidean space*)과 달리<br>, 관계와 구조가 불규칙하고 복잡한 비유클리드 공간(*Non-Euclidean space*)의 데이터를 손실 없이 표현, 학습

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">2 GNN 구조</span>

* **핵심 아이디어**
    * 타겟 노드 주변의 **이웃 노드**(나와 비슷한 사용자, 내가 소비한 아이템) 정보를 나에게 전달 및 요약하는 **메시지 패싱(Message Passing)** 방식을 기반으로 작동
* **기본 구성 레이어 구조**
    * <span style="background-color:yellow> 그래프 구조 생성 -> 메시지 패싱(메시지 생성 -> 집계 -> 갱신) -> 최종 출력 단계로 이어짐</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">3 GNN 프로세스</span>

#### 1) 그래프 구조 생성
* **개념**: 노드와 엣지를 정의하고 각 객체의 특성(*Feature*)을 반영해 수학적 행렬 형태로 변환하는 단계
* **데이터 표현 방식 예시 (넷플릭스 데이터 기준)**:
    * 사용자(U1, U2, U3) 정보와 영화(A, B, C) 정보를 각각의 노드로 생성해 총 6개의 노드를 정의함
    * 사용자가 영화를 시청하고 평점을 남긴 상호작용 행동을 엣지로 연결함
* **그래프 표현 행렬 종류**:
    * **인접 행렬 (Adjacency matrix)**: 노드 간의 연결 관계 유무를 0과 1로 표현한 크기 $6 \times 6$의 정방행렬 (연결 시 1, 미연결 시 0 부여)
    * **피어 행렬 (Feature matrix)**: 관계 정보를 제외하고 노드 고유의 특성(예: 20대, 30대, 액션, SF 등) 정보만을 담아낸 행렬

#### 2) <span style="background-color:yellow">메시지 패싱</span>
* **정의**: 그래프 내 노드들이 정보를 주고받으며 상호작용하는 과정으로, 하나의 레이어 내에서 3단계 순환 과정을 거치며 레이어 수($k$)만큼 반복 수행함
* **상세 연산 단계**:
    * **메시지 생성 (Message Generation)**: 타겟 노드가 이웃 노드에게 보낼 임베딩된 정보를 생성하며, 이전 레이어의 임베딩 결과값($h_u^{(l-1)}$)을 메시지 함수의 입력값으로 활용함
    * **집계 (Aggregation)**: 여러 이웃 노드로부터 전달받은 메시지들을 Sum, Mean, Max, Attention-weighted sum 등의 함수를 사용해 하나의 대표 벡터로 통합함
    * **갱신 (Update)**: 집계 완료된 이웃 정보 요약 벡터와 타겟 노드 자신의 이전 임베딩 상태를 결합(*Combine*)한 후, 신경망 레이어($W^{(l)}$)와 비선형 활성화 함수($\sigma$)를 통과시켜 최종 노드 상태를 업데이트함
* **반복 효과**: 레이어 전파 과정을 $l$번 반복함으로써 $l$단계만큼 멀리 떨어진 이웃 노드의 취향 정보까지 점진적으로 누적 반영할 수 있게 됨

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">4 LightGCN이란?</span>

* **개념**
    * 기존 GCN(*Graph Convolutional Network*) 구조에서 추천 시스템에 불필요한 복잡한 연산 기능을 제거하고 최적화해 설계한 가볍고 효율적인 알고리즘
* **전통적 GNN 아키텍처 종류**:
    * **<span style="background-color:yellow">GCN</span>**: 그래프 위에서 합성곱 연산을 수행하며 **이웃과 타겟 노드에 동일 파라미터**를 적용함
    * **<span style="background-color:yellow">GRN</span>**: GNN과 RNN을 결합해 **시간**에 따른 노드 특성의 동적 변화를 포착함
    * **<span style="background-color:yellow">GAT</span>**: 이웃 노드 간의 중요도를 학습해 가중 평균을 적용하며, **취향이 비슷한 이웃의 의견을 더 가중**해 반영함
    * **<span style="background-color:yellow">GraphSAGE</span>**: 대규모 그래프 학습을 위해 **샘플링 방식을 도입**하고 가중치를 **Concat** 구조로 결합함
* **LightGCN 핵심 변경 사항**
    * **특징 변환 (Feature Transformation) 제거**: 레이어마다 연산 효율을 저해하던 학습 파라미터($W^{(l)}$)를 전면 삭제함
    * **비선형 활성화 함수 (Non-linear Activation) 제거**: 임베딩 전파 과정에서 이미 비선형성이 반영되므로 과적합 방지를 위해 ReLU 등의 활성화 함수를 제거하고 정규화된 합 또는 평균 연산만을 사용함
* **의의**: CF의 본질에 집중해 그래프 구조와 전파라는 뼈대만 남김으로써 학습 연산 효율성과 속도를 극대화함

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">5 GNN 활용 사례</span>

* **소셜 미디어 (Pinterest)**: 대규모 그래프 데이터 내에서 사용자가 상호작용하는 핀, 보드, 카테고리 간의 연결 고리를 분석해 메인 화면 및 관련 이미지 추천 시스템에 활용함
* **핀테크 (금융 사기 및 자금 세탁 방지)**: 단일 고객 정보 분석을 넘어 돈의 흐름을 나타내는 송금 네트워크 그래프 자체를 추적해 자금 세탁 행위 및 금융 사기 패턴을 탐지함
* **B2C 비즈니스 (Uber Eats)**: *GraphSAGE* 알고리즘을 도입해 사용자, 레스토랑, 메뉴 아이템을 노드로 엮고, 사용자의 지리적 위치 정보와 유사 취향 사용자의 소비 패턴을 결합해 메뉴 추천을 고도화함

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">6 GNN 장단점</span>

1. **장점** 
   1. **고차원 협업 시그널 포착**: 단순 개인 성향 분석을 넘어 취향이 비슷한 이웃 및 관련 아이템 정보를 직접 반복 전파해 반영 가능
   2. **비정형 데이터 학습**: 데이터가 정형화된 형태로 고정되지 않아도 복잡한 관계망 구조 속에서 정보를 효과적으로 전파함
   3. **콜드 스타트 문제 완화**: 신규 데이터 유입 시 주변 이웃과의 연결 관계 및 노드 고유 특성을 활용해 즉각적인 임베딩 생성이 가능함
2. **단점**
   1. **과평탄화 (Over-smoothing) 문제**: 레이어가 깊어지고 메시지 전달이 과도하게 반복될수록 복잡한 그래프 내 모든 노드의 임베딩 값이 동일하게 수렴해 변별력이 사라짐
   2. **과압축 (Oversquashing) 문제**: 층이 깊어지면서 집계 대상인 이웃의 범위가 기하급수적으로 확장되나, 이를 고정된 크기의 단일 벡터에 강제로 압축하는 과정에서 주요 정보 손실이 발생함
   3. **높은 계산 복잡도**: 수십억 개의 대규모 노드와 엣지를 처리해야 하므로 막대한 연산 자원과 메모리 요구
   4. **인과적 구매 순서 반영 불가**: 모든 상호작용을 단순 집합 관계로 취급하므로 '신발 구매 후 바지 구매'와 같은 행동의 시간적 선후 관계 및 인과적 의도를 정밀하게 학습하지 X

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">3 Transformer 기반 추천 시스템</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1 기존 추천시스템의 한계</span>

#### 1) CF
* **순서 무시**: 구매나 클릭의 순서를 고려하지 않아 순서가 상이한 구매 사용자를 동일하게 간주
* **문맥 부족**: 방금 구매한 아이템과 1년 전에 구매한 아이템의 중요도 차이를 구분하기 어려움
* **예측의 한계**: 사용자의 동적인 다음 행동을 예측하기보다 고정된 일반 선호도를 파악하는 데 그침

#### 2) GNN
* **지역적 정보 전파의 한계**: 메시지 패싱을 통해 관계를 학습하므로 [A  -> B -> C] 구조에서 C가 A의 정보를 받으려면 반드시 B를 거쳐야 하며 이 과정에서 정보가 희석됨
* **장기 의존성 문제**: 정의된 그래프 구조와 이웃 관계에 의존하므로 3일 전 클릭과 방금 전 클릭의 유연한 직접 관계를 계산하는 데 한계가 존재함

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">2 Transformer의 순차적 특성</span>

* **의도 파악의 강력함**
    * 아이템 임베딩에 시퀀스 임베딩이 추가되어 전체 문맥과 핵심 의도 파악이 가능함
    * *Self-Attention* 구조를 활용해 아이템 간 관계를 거리에 상관없이 동시에 계산하며 장기 의존성 문제를 해결함
* **문맥에 따른 의미 변화 사례 (*Self-Attention*)**
    * 사용자가 **등산화**를 클릭했을 때 직전 행동의 문맥에 따라 예측 결과가 완전히 달라짐
    * **문맥 A (캠핑 준비)**: [텐트 -> 침낭 -> 등산화] 순서일 때 등산화를 캠핑 장비로 해석해 *캠핑용 코펠*, *백패킹 배낭* 등을 추천함
    * **문맥 B (겨울 출퇴근 준비)**: [롱패딩 -> 방한 장갑 -> 등산화] 순서일 때 등산화를 방한 신발로 해석해 *발열 내의*, *울 양말* 등을 추천함
* **시간 순서를 뛰어넘는 핵심 의도 포착 사례 (*Long-Range Dependency*)**
    * [프린터 구매 -> 가습기 검색 -> 강아지 사료 클릭 -> 토너 검색] 시퀀스가 존재할 때 중간에 수많은 다른 행동이 발생해도 상관없음
    * 토너(*Query*)가 유입되었을 때 관련 없는 가습기나 사료를 건너뛰고 과거의 프린터(*Key*) 항목과 가장 강한 연관성을 맺어 *프린터 전용 토너*, *A4 용지*를 정확히 추천

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">3 Transformer 기반 추천 시스템 모델</span>

#### 1) SASRec
* **정의**: 단방향 어텐션을 사용해 과거 정보만을 활용해 미래 행동을 추측하도록 훈련된 추천 시스템 모델
* **구조 및 작동 방식**:
    * **임베딩 레이어**: 각 아이템을 고유한 숫자 벡터로 변환함
    * **셀프 어텐션 레이어**: 미래 정보를 보지 못하도록 차단하는 *Masked Self-Attention* 기법을 적용해 현재 아이템 기준 오직 과거의 아이템들만 참조
    * **포인트와이즈 FFN**: 셀프 어텐션을 통과한 정보를 개별적으로 비선형 변환해 예측 능력을 고도화
    * **예측 레이어**: 학습된 맥락을 바탕으로 다음 단계에 등장할 단일 아이템을 예측
* **한계**: 각 타임스텝에 따른 오차(*Loss*)를 매번 계산하므로 계산량이 많고 미래 시점 맥락을 전혀 참고할 수 X

#### 2) BERT4Rec
* **정의**: 양방향 어텐션을 활용해 과거와 미래의 문맥 정보를 동시에 고려하도록 훈련된 추천 시스템 모델
* **구조 및 작동 방식**:
    * **입력 및 임베딩**: 사용자 행동 시퀀스 중 일부를 **[MASK]** 토큰으로 의도적 치환하며 아이템 임베딩과 순서 정보를 담은 포지셔널 임베딩을 더해 입력값으로 사용
    * **트랜스포머 블록**: 양방향(*Bi-directional*) 셀프 어텐션을 적용해 가려진 아이템을 맞추기 위해 해당 위치의 과거와 미래 문맥을 모두 참조
    * **예측**: 최종 트랜스포머 층을 통과한 **[MASK]** 토큰의 벡터만을 활용해 원래 아이템이 무엇이었는지 예측
* **특징**: 타겟 아이템 정보를 직접 보게 되는 정보 누수(*Information Leakage*) 문제를 **[MASK]** 토큰 치환 방식으로 해결

#### 3) SASRec vs BERT4Rec

| 비교 항목 | SASRec | BERT4Rec |
| :--- | :--- | :--- |
| **어텐션 방향** | - 단방향 (*Uni-directional*) 어텐션 | - 양방향 (*Bi-directional*) 어텐션 |
| **학습 방식** | - 과거 데이터만 보고 다음 아이템 예측 | - 시퀀스 내 가려진 **[MASK]** 아이템 예측 |
| **현실 추론 환경** | - 평소 학습 환경과 동일해 안정적 | - 마지막 위치를 **[MASK]**로 두고 추론 진행함 |
| **성능적 특징** | - 현실과 유사한 시계열 흐름 명확히 반영 | - 숨겨진 복잡한 맥락을 깊게 이해해 성능이 우수함 |

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">4 최근 추천 시스템 현황</span>

1. **LLM 중심 패러다임 전환**
    * 추천 시스템을 점수 계산 기반의 랭킹 문제가 아닌 문장을 출력하는 언어 문제로 접근
    * 아이템 식별자(*ID*)가 아닌 텍스트, 이미지 등 멀티모달 데이터와 의미(*Semantic*) 기반의 서술형 추천을 수행
    * 영화 리뷰나 분위기 텍스트 자체를 이해하므로 평점 데이터가 없는 신작 추천과 대화형 추천 및 제로샷(*Zero-shot*) 추천이 가능해져 강력한 콜드 스타트 해결책으로 부상<br>
2. **주요 LLM 추천 프레임워크**
    * **P5**: 모든 추천 과제를 하나의 거대한 텍스트 프롬프트 문제로 통일해 순수 텍스트로만 추천을 수행하는 대형 모델이나 컴퓨팅 자원 소모가 매우 극심
    * **TALLRec**: 범용적으로 구축된 대형 언어 모델(예: *LLaMA*)을 추천 작업에 맞게 효율적으로 파튜닝(*Fine-Tuning*)해 자원 한계를 극복하는 효율적 프레임워크<br>
3. **해결해야 할 주요 핵심 과제**
    * **인기 편향 (Popularity Bias)**: 시스템이 소수의 인기 아이템만 과도하게 노출하고 비인기 아이템을 소외시키는 현상으로 쏠림 현상을 강화해 사용자 발견의 즐거움을 저해
    * **설명 가능성 (XAI)**: 딥러닝 기반 알고리즘의 고질적인 블랙박스 문제를 해결하고 사용자 및 비즈니스 담당자에게 정당한 추천 근거를 제시해야 하는 요구가 증대됨<br>
3. **현업의 효율적 하이브리드 운영 전략**
    * 대규모 트래픽 환경에서 딥러닝의 높은 서버 비용 문제를 해결하기 위해 선형 모델과 딥러닝을 단계별로 융합
    * **1단계 (후보군 생성)**: 계산이 단순하고 속도가 압도적인 선형 모델(CF, MF 등)을 사용해 수백만 개의 아이템 중 후보군 수백 개를 빠르게 선별
    * **2단계 (순위 매기기)**: 추려진 소수의 후보군만을 대상으로 복잡한 딥러닝 모델(GNN, Transformer, VAE)을 집중 적용해 비선형성과 문맥 정보를 총동원해 정교하게 최종 순위를 확정
