# 01 추천 시스템 기초

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">1 시계열 데이터의 개념</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1 시계열 데이터란?</span>
1. **개념**
- **순차적인 시간의 흐름**에 따라 수집된 데이터
- 연속한 관측치 : 서로 **상관관계**가 존재, **시퀀스 데이터**로 간주
- 시간-수치 데이터, **단어의 순서가 중요한 NLP 데이터**에도 동일 개념 적용 가능
- 활용 분야 : 모니터링 데이터, 주가 데이터, 날씨 데이터, 천문 데이터 등....

2. **단변량 시계열 - 다변량 시계열**
- **단변량 시계열** : 시간에 따라 연속적으로 측정된 **단일 변수**로 이뤄진 시계열 데이터
  - 즉, 하나의 숫자가 시간에 따라 변하는 것
  - 시간 흐름에 따른 패턴, 트렌드 이해에 활용 -> 주식 가격, 온도 기록, 매출 데이터 등
<br>
- **다변량 시계열** : 같은 시간에 기록된 **둘 이상의 변수**로 이루어진 시계열 데이터
  - 즉, 여러 숫자가 동시에 시간에 따라 변하는 것
  - 변수 간 상호작용 및 영향력 파악에 활용 -> 경제 지표, 기후 모델 등

3. **시계열 분석과 목적**
- **주 목적 : 미레 데이터 예측**
- **시계열 분석** = 시계열 데이터를 분석, 모델링하는 통계적 방법과 기법
  - 데이터 구조 파악, 예측, 이상치 탐지를 포괄
  - 다양한 기법들을 데이터 특성, 목적에 따라 선택, 조합해 분석함
    - 데이터가 일변량인 경우 : ARIMA, 지수평활법
    - 데이터가 다변량인 경우 : 전이함수모형

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">2 특징</span>

#### 1) 시간의존성
- **개념**: 시계열 데이터는 시간에 따라 변함 -> 데이터의 순서를 바꾸면 의미가 달라짐
  - 과거의 값이 현재의 값에 영향을 미침
  - 시점 간 상관관계 고려 필요

#### 2) 계절성 & 추세
- **계절성** : 일정 주기에 따라 반복적으로 나타나는 데이터의 변동 
  - -> 주기적 패턴을 고려한 예측 모델링 가능
- **추세** : 시간에 흐름에 따라 나타나는 장기적인 증가/감소 경향 
  - -> 전반적 방향성 파악 가능

#### 3) 자기상관성
- **자기상관성** : 자기 자신끼리 상관관계를 갖는 특성
  - 즉, 한 시점의 값이 과거의 값과 통계적으로 관련 있는 성질 <br> -> 시간이 지남에 따라 값들 간 일정한 패턴&반복이 존재
  - 관측값들은 서로 독립적이지 X
- 자기상관성을 활용해 **과거 데이터를 바탕으로 미래 예측** 가능
- **회귀분석이 불가능함**
  - 오차항 독립성 가정을 위반하기 때문
  - 허위 상관관계 문제 (실제로는 변수간 인과관계가 없는데 공통된 시간 추세로 인해 나타나는 문제)
    - 회귀 시, **사전 처리(각 시계열 변수에서 추세 제거)한 후 분석해 해결**
    - 시계 열 변수 간 회귀 분석시 반드시 정상성 및 추세 확인이 필요
  - 예측의 어려움 & 모델 복잡성 증가
  - **결론** : <span style="background-color:yellow">일반회귀에서는 자기상관성이 가정을 깨뜨리나, 시계열 전용 모델(AR, MR, ARIMA)에서는 자기상관성 자체를 활용해 예측할 수 있음</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">3 시계열 데이터의 분석요인</span>

#### 1) 추세 요인(Trend Factor)
- **개념 : 데이터 전반에 걸쳐 시간에 따라 지속적으로 증가/감소하는 경향**
- 시각적 표현 : 꾸준히 증가/감소하는 선형 그래프
- 예 : 연도별 매출 상승, 출산률 감소 등

#### 2) 계절 요인(Seasonal Factor)
- **일정한(고정된) 주기로 반복되는 패턴** (+외부 요인에 의해 변동이 발생하는 경우)
- 시각적 표현 : 주기적 변동을 나타내는 그래프
- 예 : 크리스마스 시즌만 되면 관련 용품 매출이 증가함

#### 3) 순환 요인(Cycle Factor)
- 계절성보다 더 **길고 불규칙한 주기로 반복되는 변동**
- 시각적 표현 : 길게 상승과 하강을 반복하는 파형 그래프
- 예 : 경기 호황과 불황기

#### 4) 불규칙 요인
- 추세, 계절성, 순환성 모두 제거한 뒤 남은 예측 불가능한 변동
- 노이즈, 우연적 사건, 일시적 변동, 이상치 모두 포함!
- 시각적 표현 : 불규칙한 변동을 나타내는 그래프
- 예 : 자연재해, 팬데믹

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">4 시계열 분석 워크플로우</span>
1. 데이터 전처리 : 결측치 보간, 이상치 제거, 요소 분해
2. 정상성 판단
3. 모델링 및 예측

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">2 시계열 데이터 전처리</span>

- **시계열 데이터 전처리의 필요성**
  - 시계열 데이터 특성 (**시간 의존성, 추세 및 계절성, 자기상관성**)이 왜곡되지 않도록!

- **시계열 전처리 프로세스**
  1. 데이터 확인 및 시각화
  2. 결측치 확인 및 처리 (삭제, 단순 대치, 보간, 이동평균 대치 등)
  3. 이상치 탐지 및 처리 (삭제, 단순 대치, 보간, 이동평균 대치 등)
  4. 평활화 진행 (이동평균 활용)
  5. 요소 분해 (가법 모형, 승법 모형)

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1 결측치/이상치 탐지 및 처리</span>

#### 1) 시계열 데이터 결측치
- **개념**
  - 특정 시점에서 **값이 측정되지 않았거나 누락된 상태**
  - 주로 센서 오류, 수집 누락, 전송 지연 등으로 발생
  - 결측치가 존재하면 추세 및 계절성이 왜곡 -> 합리적인 대치와 보간이 필요

- **결측치 처리 방법**
  - **삭제 및 단순 대치**
    - **결측치 삭제**: 비어 있는 시점의 데이터를 제거하는 가장 단순한 방식 
      - but 시간 흐름의 연속성이 깨지고 분석 구간이 짧아질 수 있음
    - **단순 대치**: 평균, 중앙값, 최빈값, 영(0) 등으로 결측치를 채우기
      - but 패턴 왜곡 가능성이 큼
  - **보간법 (*Interpolation*)**
    - 주어진 관측값들을 사용해 데이터 포인트 사이의 **누락된 값을 수학적 기법으로 추정해 메워주는 과정**
    - 결측치 및 이상치 구간이 **짧거나** 신호의 패턴이 **단순한** 경우 보간 정확도가 높음

- **보간 방법의 종류**
  1. **선형 보간법 (*Linear Interpolation*)**
    - 두 점 사이를 직선으로 연결해 그 사이의 값을 추정
    - 데이터 간의 관계가 **선형적일 때 가장 적합**하며, 계산이 빠르고 직관적임
    - 단점: 비선형 관계나 급격한 변화가 있는 데이터에서는 정확도와 신뢰성이 떨어짐
  2. **다항식 보간법 (*Polynomial Interpolation*)**
    - 주어진 데이터 포인트를 모두 통과하는 다항식 함수로 근사해 값 추정
    - 데이터가 비선형적일 때 유용 
    - but 고차 다항식은 **과적합 문제**를 일으키거나 변동성이 큰 구간에서 급격하게 변할 수 있어 적절한 차수 선택이 필요
  3. **스플라인 보간법 (*Spline Interpolation*)**
    - 구간별 다항식을 사용해 각 구간에서의 보간을 수행
    - 각 구간마다 3차 다항식을 사용하는 **큐빅 스플라인 (*cubic spline*)**이 주로 사용됨
    - 과적합이 적고 안정적이며 부드럽고 연속적인 곡선형 데이터 제공
      - but 선형 보간에 비해 계산량이 많음
  4. **조각별 상수 보간법 (*Piecewise Constant Interpolation*)**
    - 데이터를 여러 구간으로 나누고 + 각 구간에서 동일한 상수 값을 유지 -> 결측치를 채우는 방식
      - *Forward fill*: 직전 관측치 값으로 결측치를 대체함
      - *Backward fill*: 직후 관측치 값으로 결측치를 대체함
      - *최근접 이웃 보간 (*Nearest Neighbor*)*: 앞뒤 기존 관측값 중 거리상 가장 가까운 값을 선택
    - 장점: 단순하고 계산량이 적음 + 잘못된 추정 없이 원본 값만 사용
    - 단점: 부드러움이 없는 계단형 구조 -> 세밀한 추정이 불가능 + 데이터 급변 영역에서 정확도가 낮음

- **회귀분석 vs 보간법**
  - **공통**: 관측값을 바탕으로 관측되지 않은 값에 대한 함수값 유추
  - **차이**
    - **회귀 (*Regression*)**: 관측된 점들의 오차를 가장 적게 만드는 선을 찾는 기법으로 오차를 허용
    - **보간법 (*Interpolation*)**: 관측값을 완전히 신뢰해 관측값 자체를 직접 연결

#### 2) 시계열 데이터 이상치
- **개념**
  - 값이 측정되었으나 일반적이지 않고 극단적인 값을 가진 상태
  - 측정 오류 or 실제로 특이한 이벤트일 가능성
  - 신호의 중요한 의미를 담고 있을 수 O -> 신호에 대한 **사전 지식** 바탕으로 신중하게 접근 필요

1. **이상치 전처리 방법**
  - **IQR (*Interquartile Range*)**
    - 주기성이 없거나 안정적인 구간에서 효과적
    - 하루 단위 등 구간별로 나누어 계산 시 추세의 영향을 줄일 수 있음
    - *장점*: 단순하고 통계 기반이라 계산이 빠르며 해석이 명확
    - *단점*: 추세나 계절성이 강한 시계열, 혹은 비정상 시계열에서는 오탐률이 높음
  - **Z-score**
    - 특정 값이 평균으로부터 몇 표준편차만큼 떨어져 있는지 나타냄
    - 보통 Z-score의 절대값이 3 이상(<span class="math">|Z| ≥ 3</span>)일 때 이상치로 판단함
    - 값의 변동폭이 일정한 시계열이나 구간별 분할 계산 시 효과적
    - *장점*: 계산이 간단하고 직관적임
    - *단점*: 정규분포 가정에 민감, 평균과 표준편차가 시간에 따라 변하는 비정상 시계열에서는 안정적인 임계값 설정이 어려움
  - **STL 분해**
    - 시계열을 추세($T_t$), 계절성($S_t$), 잔차($R_t$)의 세 성분으로 분해<br> -> 잔차가 특정 임계값을 초과하면 이상치로 탐지
    - 잔차에 Z-score나 IQR 기준을 추가 적용할 수 O 
    - 계절성과 추세가 뚜렷한 시계열에 적합 (일별 매출이나 센서 데이터 등)
    - *장점*: 추세와 계절성을 제거하고 진짜 이상한 값만 정교하게 포착할 수 O
    - *단점*: 각 지점 주변 데이터로 작은 회귀모델을 반복 구축하는 **LOESS 회귀** 기반 -> 계산 비용이 크고 파라미터 튜닝이 필요

2. **이상치 탐지 모델**
    - **분류 및 회귀 트리 (*CART*)**
      - 데이터를 분할해 예측
      - 예측 오차가 일정 임계값 이상인 지점을 이상치로 판단
      - 비선형 관계 및 다변량 시계열 포착에 유리 but 노이즈 데이터에서 과적합 위험 O
    - **클러스터링 기반 탐지**
      - 정상 데이터가 특정 군집에 모이는 특성을 이용해 군집 중심에서 멀리 떨어진 개체를 이상치로 판단
      - *K-means, DBSCAN, Isolation Forest* 등이 활용
      - 복잡한 패턴에 유용 but 하이퍼파라미터에 민감, 시간 의존성이 직접 반영되지 X
    - **오토 인코더 (*Autoencoder*)**
      - 인공신경망 기반 비지도 학습 모델
      - 입력을 잠재 공간으로 압축 후 복원할 때 발생하는 **복원 오차**가 큰 데이터를 이상치로 판정 
      - *LSTM Autoencoder* 활용 시 시간적 의존성 학습이 가능 -> 다변량 시계열에 강력
      - but 많은 학습 데이터와 까다로운 튜닝 필요

3. **이상치 처리 방법**
  - 탐지된 이상치를 결측치로 간주 -> 결측치 처리 프로세스(삭제, 단순 대치, 보간 등) 동일하게 적용

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">2 EDA (평활화)</span>

- **평활화 (*Smoothing*)의 개념**
  - 시계열 데이터에서 **노이즈**를 줄여 전체적인 흐름이나 추세를 명확하게 파악하기 위해 데이터를 부드럽게 만드는 기술
    - 노이즈 : 데이터 안에 섞인 설명 불가능한 랜덤 변동
    - 우리가 관측하는 데이터는 신호(Signal) + 노이즈(Noise)의 결합체

- **이동평균 (*Moving Average, MA*)**
  - **시계열의 부분집합 구간(윈도우)을 설정 + 평균을 구해 평활화 효과를 얻는 기법**
  - *이동 윈도우의 길이($N$)가 클수록 긴 기간의 데이터가 포함* -> 노이즈가 많이 제거되고 더 매끄러운 곡선이 생성됨!
  - 사용 목적: 일시적 급등락을 평탄화하는 노이즈 제거, 상승 및 하강 흐름 식별을 통한 추세 파악, ARIMA 등 시계열 예측의 기초자료 활용

- **이동평균의 유형**
  - **단순 이동평균 (*SMA, Simple Moving Average*)**
    - 특정 기간 내 모든 데이터에 **동일한 가중치**를 부여해 단순 평균을 계산
    - 변동이 심하지 않고 비교적 **안정적인 시계열**에 주로 사용됨
    - 코드: `data.rolling(window=n).mean()`
  - **가중 이동평균 (*WMA, Weighted Moving Average*)**
    - 최근 데이터에 더 높은 가중치를 부여해 **최근의 변화**를 더 민감하게 반영
    - 최근 데이터가 상대적으로 더 중요한 분석 상황에 적합함
    - 코드: `data.rolling(window=n).apply(lambda x: (x * weights).sum() / weights.sum(), raw=True)`
  - **지수 이동평균 (*EMA, Exponential Moving Average*)**
    - 최근 데이터에 **지수적으로 감소하는 가중치**를 적용하며 과거의 모든 데이터를 계산에 고려
    - **최근 가격 변동에 매우 빠르게 반응**해야 하는 시계열에 활용 (ex.주식)
    - 코드: `data.ewm(alpha=0.5, adjust=False).mean()`

- **이동평균을 이용한 대치**
  - 이동평균은 원래는 평활화 기법이나, 결측치나 이상치가 발생했을 때 그 주변 $n$개 값의 이동평균을 계산해 해당 위치를 메워주는 처리 기법으로도 활용될 수 O

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">3 요소분해 (Decomposition, 시계열 분해)</span>
- **요소분해란?**
  - 주어진 시계열 데이터를 개별 성분인 **추세성, 계절성, 순환성, 불규칙성**으로 나눠 분석하는 과정
  - $Y_t = f(T_t, S_t, C_t, R_t)$ (순환성은 보통 장기 변동이므로 $C_t$ = 0으로 설정)
  - 필요한 이유 : 다양한 변동 요인이 결합된 복잡한 시계열 그래프를 요소별로 분리해 
    - 각 패턴을 명확히 파악하고
    - 모델링 예측 정확도 향상 및 이상 탐지의 원인 분석을 가능하게 함

#### 1) 분해 모형 (=수학적 형태로 분해)
- **가법 모형 (*Additive Model*)**
  - 공식: $y_t = T_t + S_t + R_t$
  - 시계열 관측값이 각 변동 요인들의 **독립적인 합**으로 설명된다고 가정
  - 시계열 전체에서 **계절성의 진폭이 일정하게 유지될 때** 적합
  - *예*: 여름철 한국에 방문하는 관광객 수가 매년 평균 10만 명씩 일정하게 증가하는 경우

- **승법 모형 (*Multiplicative Model*)**
  - 공식: $y_t = T_t \times S_t \times R_t$
  - 로그 변환을 통해 가법 형태로 변환 가능
  - 추세가 커질수록 계절성이나 불규칙성의 효과도 비례해서 커짐 -> **시간에 따라 진폭이 변할 때** 적합
  - *예*: 여름철 한국에 방문하는 관광객 수가 매년 평균 10%씩 비율로 증가하는 경우

#### 2) 분해 방법 (구현 알고리즘)
- **핵심 특징 비교**
  - **seasonal_decompose**
    - 이동평균 방식 사용
    - 고정된 계절성만 반영 + 이상치의 영향을 많이 받음 -> 데이터 손실이 발생
  - **STL**
    - LOESS 방식 사용
    - 시간에 따라 변하는 계절성을 반영 + 이상치 영향이 적음 -> 데이터 손실이 발생하지 X

1. **Seasonal_Decompose**
   - **원리**
     - 전통적인 시계열 분해 기법<br> -> **중심 이동평균을 통해 추세를 추출한 뒤 잔여 성분에서 계절성을 분리하는 방식**
     - **작동 방식**
       1. 중심 이동평균을 적용해 추세 성분($T_t$) 추정
       2. 원 시계열에서 추세를 제거하는 Detrending 수행 ($D_t = Y_t - T_t$)
       3. 디트렌드 시계열에서 동일 계절 인덱스의 평균을 취해 계절 패턴($S_t$) 도출 (연간 평균이 0이 되도록 정규화)
       4. 잔차 산출 ($R_t = Y_t - T_t - S_t$)

2. **STL -LOESS 기반 분해**
   - **원리**
     - **LOESS** (Locally Estimated Scatterplot Smoothing) 기법 사용
       - 이동평균 대신 **로컬 회귀 모델**을 이어 붙여 부드러운 곡선을 만드는 기법
       - 추세와 계절성이 시간에 따라 변화하는 경우에도 유연하게 포착 + 이상치에 강함
   - **작동 방식**
     1. Inner Loop (반복 수렴): 추세 제거
     2. 각 Cycle-subseries에 LOESS 적용
     3. 저주파 필터링
     4. 계절 성분 확정 $\rightarrow$ 
     5. 계절성 제거 $\rightarrow$ 
     6. LOESS 추세 갱신 단계를 수렴할 때까지 반복
   - **Outer Loop**: 잔차 크기를 기준으로 강건성 가중치를 부여<br> -> 이상치 영향력을 제어한 후 Inner Loop를 재실행

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">3 정상성 판단</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1 정상성이란?</span>
> 시계열 데이터의 평균, 분산 등 **통계적 특성이 시간에 흐름에 상관없이 일정하게 유지**되는 성질<br>
  - **정상 시계열 (*Stationary Time Series*)**: 정상성을 띠는 시계열 데이터<br> -> 과거의 패턴과 성질이 미래에도 동일하게 유지된다고 가정할 수 있어 통계적 모델링에 바로 적용 가능
  - **비정상 시계열 (*Non-stationary Time Series*)**: 시간에 따라 평균이나 분산이 지속적으로 변하는 데이터<br> -> 바로 예측하기 어렵기 때문에 정상성을 띠도록 수학적 처리를 거쳐야 함
<br>
> 데이터가 정상성을 가져야만 미래 예측이 가능해지며, 통계적 모델의 신뢰성을 확보할 수 있음

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">2 강한 정상성 - 약한 정상성</span>

1. **강한 정상성 (*Strict Stationarity*)**
  - 시간을 언제로 잡든 데이터의 전체 **확률 분포 자체가 완전히 동일한 상태**
  - 평균과 분산뿐만 아니라 왜도, 첨도 등 모든 통계적 적률이 일정
  - 현실 세계의 데이터는 잡음과 외부 요인이 항상 존재 -> 강한 정상성을 만족하는 데이터는 거의 존재 X

2. **약한 정상성 (*Weak Stationarity*)**
  - 실무와 통계 모델링에서 사용하는 현실적 기준
    - **평균이 일정**: 시간이 지나도 시계열의 평균값이 일정한 수준을 유지함
    - **분산이 일정**: 데이터의 오르내림 변동 폭이 시간에 관계없이 일정하게 유지됨
    - **자기공분산이 시차에만 의존**: 두 시점 간의 상관관계가 실제 측정 시점이 아닌, 오직 두 시점 사이의 **시간 간격 (*Lag, h*)**에 의해서만 결정됨
  - 전체 확률 분포는 변할 수 있으나 평균, 분산, 자기공분산인 **2차 적률까지만 일정**하면 약한 정상성으로 간주

3. **공분산 vs 자기공분산**
  - **공분산 (*Covariance*)**: 서로 다른 두 변수 $X$와 $Y$가 함께 움직이는 정도
  - **자기공분산 (*Autocovariance*)**: 동일한 시계열 내에서 서로 다른 두 시점 $t$와 $t-k$ 사이의 관계 ($\gamma(k) = \operatorname{Cov}(X_t, X_{t-k})$)
    - *양의 자기공분산*: 과거와 현재가 같은 방향으로 움직이는 관성이 있는 데이터 (예: 온도)
    - *음의 자기공분산*: 과거 값과 현재 값이 반대로 움직이며 되돌아오는 경향이 있는 데이터 (예: 냉장고 온도 유지 시스템)
    - *0의 자기공분산*: 과거 값이 현재 값에 아무런 힌트를 주지 않는 기억이 없는 랜덤 시계열 (예: 백색잡음, 동전 던지기)

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">3 정상성 판단</span>

1. **시각적 정상성 파악**
  - **정상 시계열은 일정한 기준선(평균)을 중심으로 일정한 폭(분산)으로 진동하는 형태**를 보임
  - 비정상 시계열 특징
    - 시간에 따라 지속적으로 상승/하락하는 **추세**가 있거나 
    - 특정 주기로 패턴이 반복되는 **계절성**이 보이는 경우
  - 추세가 뚜렷한 시계열은 -> 차분한 결과로 해석 가능 (추세 제거한 결과로 변환)

2. **통계적 검정과 단위근 (*Unit Root*)**
  - 그래프만으로는 정상성 여부를 확신할 수 X -> 수치적인 가설 검정 필요
  - 대부분의 시계열 검정은 비정상성의 대표적 원인인 **단위근**이 존재하는지 확인하는 과정을 거침
  - **단위근이란**
    - 과거의 충격과 변동이 희석되지 않고 영구적으로 누적되는 특성
    - 시계열 모델의 특성방정식에서 근이 1인 경우 = 비정상 시계열
    - 단위근이 존재하면 평균으로 회귀X -> 분산이 시간에 따라 계속 증가하는 **랜덤워크** 현상이 발생해 미래 예측이 불가능해짐

3. **AR(1) 모델로 보는 단위근**
  - 모델식: $y_t = \alpha y_{t-1} + e_t$ ($\alpha$: 자기회귀 계수, $e_t$: 백색잡음)
  - $\alpha = 0$: 과거의 영향을 받지 않는 완전한 정상성 상태
  - $|\alpha| < 1$: 현재가 과거의 영향을 받지만 시간이 흐를수록 과거 충격이 희석되어 평균으로 회귀하는 **정상 시계열**
  - $\alpha = 1$: 과거의 모든 오차가 누적되어 원상태로 돌아가지 못하고 수렴하지 않는 단위근 상태의 **비정상 시계열**
  - $|\alpha| > 1$: 값이 기하급수적으로 커지는 폭발적 시계열이자 비정상 시계열

#### 1) ADF vs KPSS

  | 구분 | **ADF** 검정 (*Unit Root Test*) | **KPSS** 검정 (*Stationarity Test*) |
  | :--- | :--- | :--- |
  | **관점** | 데이터의 **비정상성**을 가정함 | 데이터의 **정상성**을 반증하려 함 |
  | **귀무가설 ($H_0$)** | 단위근 존재 $\rightarrow$ **비정상 시계열** | 단위근 없음 $\rightarrow$ **정상 시계열** |
  | **대립가설 ($H_1$)** | 단위근 없음 $\rightarrow$ **정상 시계열** | 단위근 존재 $\rightarrow$ **비정상 시계열** |
  | **판정 기준** | p-value < 유의수준일 때 $H_0$ 기각 (정상) | p-value < 유의수준일 때 $H_0$ 기각 (비정상) |
  | **결과 상충 시** | 시계열 데이터에 **확정적 추세**가 존재함을 의심함 | ADF와 반대 방향의 가설 체계를 공유함 |
  | **해결 방법** | 모델에 추세항을 추가 $\rightarrow$ *ADF(Trend)* | 모델에 추세항을 추가 $\rightarrow$ *KPSS(Trend)* |

- **검정 결과 조합 해석**
  - *ADF 정상 / KPSS 정상*: 정상 시계열이므로 그대로 모델링 가능함
  - *ADF 비정상 / KPSS 비정상*: 단위근이 존재하는 확률적 추세이므로 **차분**을 적용해야 함
  - *ADF 비정상 / KPSS 정상*: 확정적 추세가 의심되므로 **추세 제거** 또는 **추세항 모델링**이 필요함

#### 2) ADF 검정 (단위근 검정)
- **정의 및 특징**
  - 시계열이 과거 값에 얼마나 의존하는지 측정해 단위근 존재 여부를 검정하는 대표적 방법
  - p-value가 유의수준보다 작으면 귀무가설을 기각 -> 정상 시계열로 판단
  - 데이터에 확정적 추세가 존재할 경우, 단순 ADF는 이를 단위근으로 착각해 비정상으로 오판하기 쉬우므로 **추세항($\beta t$)을 추가한 ADF(Trend) 모델**을 사용해야 함

#### 3) KPSS 검정 (정상성 검정)
- **정의 및 특징**
  - **ADF의 낮은 검정력을 보완**하기 위해 교차 검정용으로 함께 사용하는 정상성 검정 방법
  - 시계열이 일정한 평균 수준을 유지하는지 측정하며, p-value가 유의수준보다 작으면 귀무가설을 기각 -> **비정상 시계열**로 판단함
  - 확정적 추세가 존재할 때 기본 KPSS는 평균이 변한다고 보아 비정상으로 오판하므로, 구조적으로 추세를 반영해주는 ***KPSS(Trend) 모델***을 적용해야 정상 결과를 얻을 수 O

- **확정적 추세 vs 확률적 추세**
  - **확정적 추세 (*Deterministic Trend*)**
    - 시간에 따라 일정한 방향으로 움직여 완벽히 예측 가능한 추세임
    - 잡음이 섞여도 추세 자체는 일정하게 유지됨 (예: $X_t = 0.05t + \epsilon_t$)
  - **확률적 추세 (*Stochastic Trend*)**
    - 추세선 자체가 시간에 따라 무작위로 이동하는 **랜덤워크** 성격을 띰
    - 과거의 충격이 누적되어 미래 예측이 불가능함 (예: $X_t = X_{t-1} + \epsilon_t$)

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">4 정상성 확보 방법</span>
#### 1. **추세 제거 및 차분**
  - **차분이란?**: 연속된 두 시점의 관측값들의 차이를 계산해 변화량을 구하는 기법
  - **목적**: **평균의 정상성**을 확보하기 위함이며, 대부분의 비정상 시계열은 값이 누적되는 과정이므로 이전 값을 빼줌으로써 정상 시계열을 얻을 수 있음
  - **1차 차분**: $y'_t = y_t - y_{t-1}$ 과 같이 직전 값을 빼는 방식으로, 일반적인 **선형 추세를 제거**할 때 사용함
  - **2차 차분**: 1차 차분 데이터에 한 번 더 차분을 적용하는 방식($y''_t = y'_t - y'_{t-1}$)으로, 데이터 소실이 커져 설명력이 낮아질 수 있으므로 주의해야 함
  - **역차분 (*Inverse Difference*)**: 분석 및 모델링이 완료된 후, 차분된 데이터에 이전 시점의 원래 값을 다시 더해 원래 단위의 데이터로 복원하는 전처리 역산 과정임

#### 2. **계절성 제거 (계절성 차분)**
  - **개념**: 시계열이 일정한 주기($m$)를 가지고 반복될 때, 현재 시점의 값에서 한 주기 전의 값을 빼는 기법임
  - **공식**: $y_t^* = y_t - y_{t-m}$ (예: 12달 주기의 계절성이면 $m=12$ 적용)
  - 데이터에 계절성 패턴이 매우 강하게 나타난다면 1차 차분보다 **계절 차분을 먼저 계산**하는 것이 좋으며, 계절 차분만으로 정상성이 확보되면 추가적인 일반 차분은 진행하지 않음

#### 3. **분산 안정화 & 로그 변환**
  - **개념**: 시간에 따라 시계열의 오르내림 변동 폭(분산)이 점점 커지거나 작아질 때 적용하는 기법임
  - **공식**: $y_t \rightarrow \log(y_t)$
  - 로그를 취하면 데이터의 변동 폭과 비율적인 변화가 줄어들어 **분산의 정상성**을 확보하는 데 도움을 줌
  - **한계점**: 로그 변환은 **음수나 0의 데이터를 처리할 수 없으며**, 분산 변화 패턴이 비선형적이거나 극도로 복잡할 경우에는 효과가 제한적일 수 O

#### 4. **정상성 확보 후 최종 프로세스**
  - 시계열 분석은 **`정상성 판단 → 비정상 유형별 정상성 확보 → 모델링`** 순으로 진행
  - 정상성이 완전히 확보된 후 데이터의 구조적 특성에 맞게 *AR, MA, ARMA, SARIMA* 등의 시계열 모델을 최종 적용

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">4 단변량 시계열 분석</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1 안정적 시계열 분석</span>

#### 1) 자기회귀모델(AR)
- **개념**
  - 과거 자신의 값들을 독립변수로 삼아 현재 값을 예측하는 회귀 모델임
  - 시계열 데이터의 **자기상관성**인 이전 시점의 값이 이후 시점에 영향을 미치는 특성을 활용함

- **수식**
  - **$AR(p)$ 모형**: 과거 $p$개 시점의 데이터 선형 조합을 이용해 예측하는 $p$차 자기 회귀 모형
    - $y_t = c + \phi_1 y_{t-1} + \phi_2 y_{t-2} + \cdots + \phi_p y_{t-p} + \varepsilon_t$
  - **$AR(1)$ 모형**: 현재 값($y_t$)이 바로 직전 1시점 이전의 값($y_{t-1}$)에만 영향을 받는 모형
    - $y_t = c + \phi_1 y_{t-1} + \varepsilon_t$

- **OLS 회귀와 $AR(p)$ 모형의 비교**
  - **OLS 기반 시계열 회귀의 한계**
    - 예측할 미래 시점($Y_{t+1}$)의 예측값을 다음 시점($Y_{t+2}$) 예측의 독립변수로 반복 사용함
    - 이 과정에서 예측 오차가 계속 누적되어 **먼 미래의 예측값이 매우 불안정**해지는 구조적 취약점이 존재함
  - **$AR(p)$ 모형의 우수성**
    - 시계열 예측을 목적으로 설계되어 **예측값을 안정적으로 연쇄 활용하는 구조**가 모델 자체에 포함됨
    - 파라미터 추정 시 OLS(최소자승법) 대신 오차 누적을 방지하는 **MLE(최우도추정법)**를 기반으로 파라미터($\phi$)를 추정하므로 시계열 분석에 적합함!!

#### 2) 이동평균모델(MA)
- **개념**
  - 과거 시점의 **잔차($\varepsilon$)**를 독립변수로 삼아 현재 값을 예측하는 모델
  - 과거에 잘못 예측한 오차, 외부의 갑작스러운 충격을 현재 시점에 반영하는 구조
  - 잔차($\varepsilon$)는 실제값과 예측값의 차이를 의미하며 모두 독립적이고 백색잡음인 상태를 전제함
  - 주의: **평활화 기법인 이동평균과는 이름만 같을 뿐 완전히 다른 개념**임!!

- **백색잡음**
  - $y_t = \varepsilon_t$
  - 평균은 0, 분산은 일정, 과거 값의 영향을 전혀 받지 않고, 오직 무작위 외부 충격에 의해서만 결정되는 정상 시계열

- **수식**
  - **$MA(q)$ 모형**: 과거 $q$개 시점의 예측 오차들의 선형 결합으로 현재 값을 예측하는 $q$차 이동 평균 모형
    - $y_t = \mu + \varepsilon_t + \theta_1 \varepsilon_{t-1} + \theta_2 \varepsilon_{t-2} + \cdots + \theta_q \varepsilon_{t-q}$
  - **$MA(1)$ 모형**: 현재 값($y_t$)이 평소 평균 수준($\mu$)에 당해 시점 충격($\varepsilon_t$)과 바로 직전 1시점 이전의 오차($\varepsilon_{t-1}$)의 영향만을 결합해 결정되는 모형
    - $y_t = \mu + \varepsilon_t + \theta_1 \varepsilon_{t-1}$

#### 3) AR, MA 모델의 전제 조건
- **정상성**
  - AR과 MA 모형은 기본적으로 확률적 안정성을 전제로 추정함 -> 모델 자체의 구조가 정상성을 만족해야 함
  - **AR 모형의 정상성 확보**
    - 특성 방정식인 $1 - \phi_1 z - \phi_2 z^2 - \cdots - \phi_p z^p = 0$에서 도출되는 모든 해($z$)의 절대값이 1보다 커야 함 (지연 연산자의 역수 근 기준으로는 절대값이 1보다 작아야 함)
    - *$AR(1)$의 정상성 조건*: $|\phi_1| < 1$
    - *$AR(2)$의 정상성 조건*: 특성방정식 해의 절대값이 모두 1보다 작아야 안정 상태를 유지하며 평균으로 회귀함
  - **MA 모형의 정상성 확보**: 모형을 구성하는 기본 단위인 잔차(백색잡음)들이 이미 정상성 조건을 완벽히 만족하므로 **MA 모형은 계수 값과 상관없이 항상 정상성을 만족**함
  - 주의 : **데이터 자체의 정상성** -> **ADF, KPSS** 검정으로 판단 / **모델 구조의 정상성** -> **특성방정식**으로 확보

- **가역성**
  - **개념**: MA 모형을 무한 차수의 AR 모형인 $AR(\infty)$ 형태로 상호 변환해 재표현할 수 있는지의 여부
  - **필요성**: MA 모형의 독립변수인 오차항($\varepsilon$)은 실제 환경에서 직접 관찰이 불가능하므로 이를 관측 가능한 과거 관측값($y_{t-n}$)들의 식으로 변환할 수 있어야 모델이 안정적으로 작동함
  - **AR 모형**: 태생적으로 관측값 기반으로 표현되어 있으므로 가역성을 체크할 필요 X
  - **MA 모형**: 특성 방정식인 $1 + \theta_1 z + \cdots + \theta_q z^q = 0$의 모든 해($z$)의 절대값이 1보다 커야 가역성이 만족

#### 4) ARMA 모형
- **개념**
  - 과거 $p$개의 관측값과 과거 $q$개의 예측 오차를 결합해 미래를 예측하는 모형
  - $Y_t = \phi_1 Y_{t-1} + \cdots + \phi_p Y_{t-p} + \varepsilon_t + \theta_1 \varepsilon_{t-1} + \cdots + \theta_q \varepsilon_{t-q}$
  - **성립 조건**: $AR(p)$ 성분의 정상성 조건과 $MA(q)$ 성분의 가역성 조건을 동시에 모두 만족해야 함

- **특수한 경우**
  - **$ARMA(p, 0)$**: 이동평균 항이 존재하지 않아 $AR(p)$ 모형과 완벽히 동일
  - **$ARMA(0, q)$**: 자기회귀 항이 존재하지 않아 $MA(q)$ 모형과 완벽히 동일
  - **$ARMA(0, 0)$**: 과거의 누적 영향이 전혀 없는 백색잡음 모델($y_t = \varepsilon_t$)이 됨

#### 5) 차수 결정 방법
- **자기상관함수 (*ACF*)**
  - 시차 $k$에 따른 두 관측치 간의 전체적인 상관관계를 나타냄
  - 중간 시점들이 미치는 간접적인 파급 효과와 영향력까지 모두 포함해 측정
  - $MA(q)$ 모형의 적절한 차수 $q$를 결정하는 데 사용
- **부분자기상관함수 (*PACF*)**
  - 두 시점 사이의 다른 중간 시점들의 영향을 모두 통제한 상태에서 오직 두 시점 간 직접적 영향만 분석
  - $AR(p)$ 모형의 적절한 차수 $p$를 결정하는 데 사용

- **시각적 차수 판정 예시 분석**
  1. ACF 천천히 감소 / PACF 시점 1 이후 파란 점선 신뢰 구간 안 진입
     1. $p=1, q=0$인 $AR(1)$ 모형이 적합
  2. *ACF 시점 1 이후 신뢰 구간 안 진입 / PACF 천천히 진동 감소*
     1. $p=0, q=1$인 $MA(1)$ 모형이 적합
  3. *ACF 및 PACF 모두 시점 1 이후 신뢰 구간 영역 안으로 수렴*
     1. $p=1, q=1$인 $ARMA(1, 1)$ 모형이 적합

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">2 불안정 시계열 분석</span>

- **불안정 데이터 분석 프로세스**
  - 추세나 계절성이 존재하는 비정상 시계열 데이터 접수
  - 일반 차분 또는 계절 차분을 적용해 정상성 확보
  - 변환된 정상 시계열에 안정적 시계열 모델인 $ARMA$를 적용해 분석함
- **불안정 시계열 모델** : 기존 $ARMA$ 모델에 데이터의 **정상성 확보 과정(차분, 계절성 차분)이 추가**된 형태

#### 1) ARIMA
- **개념**
  - $ARMA(p, q)$ 모델에 정상성 확보를 위한 일반 차분 $d$회 과정을 추가해 결합한 모형임
  - 공식 명칭은 자기회귀 누적이동평균 모형이며 **$ARIMA(p, d, q)$**로 표기함
    - $p$: 자기회귀($AR$) 항의 차수
    - $d$: 정상성 변환을 위한 일반 차분($Integration$) 횟수
    - $q$: 이동평균($MA$) 항의 차수

- **특수한 경우**
  - **$ARIMA(0, 0, 0)$**: 차분과 과거 성분의 영향이 없는 백색잡음 모델임
  - **$ARIMA(p, 0, 0)$**: 차분이 필요 없는 구조로 $AR(p)$ 모형과 동일함
  - **$ARIMA(0, 0, q)$**: 차분이 필요 없는 구조로 $MA(q)$ 모형과 동일함
  - **상수가 없는 $ARIMA(0, 1, 0)$**: 일차 차분 시 백색잡음이 되는 순수 **랜덤워크 (*Random Walk*)** 모델임 ($Y_t = Y_{t-1} + \varepsilon_t$)
    - 현재 값을 예측하는 최선의 지표는 직전 시점의 값이며, 별도의 거시적 방향성 없이 위아래로 무작위하게 휘청이며 과거의 충격을 영구적으로 누적함
  - **상수가 있는 $ARIMA(0, 1, 0)$**: **표류를 포함하는 랜덤워크 (*Random Walk with Drift*)** 모델임 ($Y_t = Y_{t-1} + \varepsilon_t + c$)
    - 무작위 충격과 더불어 매 시점 상수항($c$)이 지속적으로 누적되기 때문에 시간이 흐를수록 상향 또는 하향하는 뚜렷한 거시적 **방향성**을 띠며 증가하거나 감소함

#### 2) SARIMA
- **개념**
  - 일반 $ARIMA$ 모형에 주기적인 패턴을 제어하는 **계절성 차분 및 계절성 파라미터**를 추가해 확장한 모형
  - **하이퍼파라미터**: **$SARIMA(p, d, q)(P, D, Q)_m$**
    - $(p, d, q)$: 비계절성 ARIMA 파라미터 (차수, 차분, 오차항 차수)
    - $(P, D, Q)$: 계절성 ARIMA 파라미터 (계절성 AR 차수, 계절성 차분 횟수, 계절성 MA 차수)
    - $m$: 1년 혹은 1주 단위 내에서 반복되는 **계절성 주기의 길이**임 (예: 월별 데이터에서 연간 주기 시 $m=12$, 일별 데이터에서 주간 주기 시 $m=7$)

- **계절성 데이터의 ACF와 PACF**
  - 강력한 계절성 주기를 가지는 데이터는 ACF 그래프 상에서 계절 주기 $m$의 배수가 되는 시차 지점($lag = 12, 24, \cdots$)마다 주기적으로 값이 급등하는 독특한 잔류 패턴을 보임
  - 이러한 강한 계절성 추세는 비계절성 추세 파악을 방해하므로, 분석 시 계절 차분을 우선 수행해 변동을 깎아낸 후 비계절성 차분을 연속 적용하는 것이 올바름

- **통계적 하이퍼파라미터 ($p, q, d, D$) 결정 방법**
  1. **차분 차수 ($d, D$) 결정**
     - 시각적 분포 확인과 통계적 가설 검정인 **ADF(단위근 검정)**를 순차 실행해 정상성 확보 여부를 엄격히 판단함
     - 데이터가 정상성을 만족하는 최소한의 차분 횟수를 지정하며 대부분 0 또는 1의 범위 내에서 최종 결정됨
  2. **AR, MA 차수 ($p, q$) 범위 추정**
     - 정상성 변환 처리가 완료된 시계열 데이터의 ACF와 PACF 그래프를 시각적으로 검토해 대략적인 차수의 후보군 범위를 1차 추정함
  3. **정보 기준 수치 기반의 그리드 서치 및 최종 확정**
     - 그래프 판독의 주관적 한계를 극복하기 위해 패널티 기반 모델 성능 평가 지표인 **AIC** 또는 **BIC(SC)**를 활용함
     - 통상적으로 $p$와 $q$의 탐색 범위를 0에서 3 사이로 설정한 후 시스템적인 **그리드 서치**를 실행함
     - 산출된 지표 중 **AIC 및 BIC 값이 가장 낮게 측정되는 최적의 조합**을 정교한 최종 모델 파라미터로 선택함
     - 안정적 시계열 모형인 ARMA의 차수 결정 시에도 그래프만으로 판별이 모호할 경우 동일한 그리드 서치 기법을 수행해 확정함

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">5 다변량 시계열 분석</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1 VAR</span>
- **개념**
  - **자신의 과거 값뿐만 아니라 다른 변수들의 과거 값을 함께 사용해** 상호 연관된 다변량 미래 값을 동시에 예측
  - 단변량 AR 모형이 자기 자신의 과거 정보만 사용하는 단방향 구조라면, VAR 모형은 여러 변수가 **서로 영향을 주고받는 양방향 구조**!!

- **수식으로 이해하기**
  - *단변량 $AR(1)$*: $X_t = a_1 X_{t-1} + \varepsilon_t$ (자기자신 과거만 사용)
  - *다변량 $VAR(1)$* (서로의 과거를 함께 사용)
    - $X_t = a_{11} X_{t-1} + a_{12} Y_{t-1} + \varepsilon_{x,t}$
    - $Y_t = a_{21} X_{t-1} + a_{22} Y_{t-1} + \varepsilon_{y,t}$

- **추천 vs 비추천 기준**
  - **적합한 경우**
    - 경제·금융 지표처럼 변수들이 일방향이 아닌 상호 의존적으로 밀접하게 영향을 주고받는 경우
  - **부적합한 경우**
    - 변수의 개수가 너무 많아 파라미터가 폭발적으로 증가하는 경우
    - 변수 간 관계가 모호한 경우, 장기 예측이 필요한 경우

- **변수 선택 방법: 그레인저 인과관계 검정**
  - 통계적으로 **"X의 과거 정보가 Y의 미래 예측에 실질적인 도움이 되는가?"**를 검정해 VAR 모형에 포함할 변수를 선별하는 기법

- **VAR 모델을 활용한 주요 분석 방법**
  - **충격 반응 함수 (*IRF, Impulse-Response Function*)**
    - 특정 변수에 돌발적인 외부 충격이 발생했을 때, 그 충격이 시간 흐름에 따라 다른 변수에 어떤 방향과 크기로 파악되는지 분석
  - **예측오차 분산 분해 (*Variance Decomposition*)**
    - 미래 값을 예측할 때 발생하는 오차의 변동성이 각 변수의 충격에 의해 얼마나 기여되고 설명되는지 그 비율을 분석

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;">6 딥러닝 활용 시계열 분석</span>
- **등장 배경**
  - 데이터의 규모가 거대해질수록 연산 처리 속도가 급격히 저하됨
  - 고정된 크기의 입력 데이터만 처리 가능함
  - 비선형 패턴 포착이 어려움

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">1 순환 신경망(RNN 계열)</span>

- **주요 모델들**
  - **RNN**: 이전 시점의 은닉 상태 정보를 현재 시점으로 순차 전달해 연산하는 가장 기본적인 순환 구조 신경망
  - **LSTM**: 정교한 게이트 메커니즘을 내장해 RNN의 고질적인 전력 손실 및 기울기 소실 문제를 해결하고 **장기 의존성(*Long-Term Dependency*)** 기억을 안정적으로 유지
  - **GRU**: LSTM의 셀 구조를 축소 및 단순화해 연산 속도를 대폭 끌어올리면서도 유사한 수준의 장기 기억 성능을 내도록 설계

- **활용성 및 한계**
  - **적합한 경우**: 주가 데이터, 실시간 센서, 재무제표, 사용자 로그 등 시퀀스의 순차적 흐름과 과거 정보의 전이가 핵심인 도메인에 유용
  - **부적합한 경우**: 수천에서 수만 스텝에 달하는 극단적으로 긴 시계열에서는 학습 효율이 급감하며, 구조적 특성상 **병렬 처리가 불가능**해 트랜스포머 계열 대비 학습 속도가 느림

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">2 Transformer 계열</span>

- **주요 특성**
  - **Informer (장기 예측 특화)**
    - 자기주의집중 연산의 불필요한 부분을 솎아내는 방식을 통해 연산 복잡도를 획기적으로 낮춘 모델
    - 아주 긴 시점의 시퀀스를 한 번에 처리해야 하는 **장기 시계열 예측** 환경에서 연산 효율성이 극대화됨
  - **Autoformer (주기성 포착 특화)**
    - 전통 시계열 기법의 핵심인 **요소 분해** 개념을 딥러닝 네트워크 구조 내부에 직접 통합
    - 연간 주기나 계절적 변동이 뚜렷하고 반복적으로 나타나는 매출 및 수요 예측 도메인에서 우수한 성능
  - **TFT (*Temporal Fusion Transformer*)**
    - 과거 시계열뿐만 아니라 날씨, 공휴일 변수, 고정된 정적 마스터 정보 등 **다양한 종류의 외생 변수를 복합적으로 융합**할 수 있는 아키텍처임
    - 변수별 중요도를 어텐션 스코어로 직접 시각화할 수 있음 
    - **설명력** 확보가 필수적인 유통 수요 예측 실무에서 가장 많이 쓰임

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">3 Foundation Model</span>
- **정의 및 핵심 특징**
  - 대규모의 대량 데이터셋으로 사전 학습되어 별도의 아키텍처 전면 수정 없이도 다양한 시계열 다운스트림 태스크를 범용적으로 해결하는 거대 모델임
  - 특정 도메인 문제에만 국한되지 않는 **뛰어난 범용성**을 가지며, 사전 학습 완료 후 미세조정을 통해 특정 비즈니스에 커스텀 적용

- **기반 유형 분류**
  - **LLM 기반 모델**: 기존 언어 모델이 축적한 시퀀스 이해 능력을 시계열 분석에 그대로 전용해 프롬프팅이나 파인튜닝을 적용하는 형태임
  - **시계열 전용 파운데이션 모델**: 대규모 시계열 데이터만을 타깃으로 삼아 처음부터 대형 사전 학습을 거친 전용 범용 모델임 (예: *TimeGPT, TimesFM, Moirai, Chronos*)

- **LLM 기반 모델 활용 방식**
  - **Without Adaptation**
    - 거대 언어 모델이 내부적으로 시계열적 지식 패턴을 내재하고 있다고 가정
    - 오직 정교한 **프롬프트 엔지니어링**만으로 예측 유도
  - **Adapt LLM**
    - 언어 모델의 가중치 레이어를 시계열 토큰 데이터에 맞추어 레이어 전반 혹은 일부를 직접 **Fine-tuning**하는 방식
  - **Adapt to LLM**
    - 수치형 시계열 소스 자체를 언어 모델이 인지하기 좋은 형태의 텍스트 자연어로 전처리 및 변환해 입력하는 방식임
    - 예: 숫자 배열을 "현재 우상향 추세이며 일정한 주기가 발견됨" 형태로 텍스트화

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">4 딥러닝 모델 선택 가이드</span>
- **핵심 트레이드오프**
  - 순수 `RNN < LSTM/GRU < Transformer` 순으로 갈수록 아키텍처가 무거워지나 고도화된 패턴의 표현 능력이 상승함
  - 전통 통계 모델이 직관적인 설명력은 우수하나 정확도는 딥러닝이 압도적임 (단, TFT 모델은 어텐션 스코어를 활용해 정확도와 설명력을 동시에 취득 가능함)
  - 데이터 표본 수집량이 적은 구조에서는 딥러닝 적용 시 과적합을 유발하므로 오히려 **통계 모델**(ARIMA 등)이 안전하며, 데이터가 무수히 많은 빅데이터 환경에서 트랜스포머의 병렬 효율이 극대화됨
