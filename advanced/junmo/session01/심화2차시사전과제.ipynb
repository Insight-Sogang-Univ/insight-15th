# 1. 인공신경망의 시작: 퍼셉트론

## 1-1. 인공신경망이란?

딥러닝을 이해하기 위해서는 우선 인공신경망과 퍼셉트론에 대한 이해가 필요함.

<aside>

### ① 인공신경망

우리 뇌의 신경망을 모방한 모델

- 생물학적 뉴런이 정보를 처리하는 구조
    1. 가지돌기를 통해 외부 전기신호를 받아들임
    2. 모인 신호가 일정 이상의 크기를 가지면
    3. 축삭돌기를 통해 신호를 전달

- (참고) 뉴런과 퍼셉트론의 연결 방식
    A, C는 각각 뉴런과 뉴런들의 연결을 보여줌. B는 단층 퍼셉트론, D는 여러 퍼셉트론을 연결시킨 다층 퍼셉트론을 나타낸 것임.

1. 입력(x_1, x_2, ..., x_n)들을 받아 각각 가중치(w_1, w_2, ..., w_n)와 편향(b)을 적용함.
2. 종합한 값을 활성화 함수에 넘김. 이때 활성화 함수는 임계값을 기준 삼음.
    - 기준 미달이면? → 0에 가깝게 죽여서 내보냄.
    - 기준 통과면? → 유의미한 값으로 살려서 내보냄.
3. output이 출력됨.
</aside>

<aside>

### ② 퍼셉트론

퍼셉트론은 초기 형태의 인공신경망으로, 신경망의 기원이 되는 선형 이진 분류기임.

퍼셉트론의 출력값을 구하는 절차

⇒ `입력 → 가중합 → 활성화 함수 → 출력` 흐름을 따라감.

1. 입력값과 가중치를 곱함 → ${x_1 \times w_1}$
2. 곱한 값들의 총합을 구함 → ${x_1 \times w_2 + x_2 \times w_2 + x_3 \times w_3 + ... + x_n \times w_n}$
(편향을 더해줄지는 옵션임)
3. 총합이 임계값을 넘으면 1, 넘지 않으면 0을 출력함.
(이 경우, 활성화 함수는 Step 함수임. 밑에 나옴.)

※ 활성화 함수: 임계값 초과 여부에 따라 출력값을 결정함.
     사용하는 활성화 함수에 따라 출력값이 달라질 수 있음.

→ Training(학습): 원하는 출력값을 내보내도록 가중치를 조정해가는 작업

</aside>

<aside>

### ③ 용어 정리

1. 가중치
- 입력값이 출력값에 주는 영향력을 결정함.
- 각 입력값에는 각각의 가중치가 존재함.
- 가중치의 값이 클수록 해당 입력 값이 중요함을 의미함.

---

1. 활성화 함수
- 신호가 일정 이상의 값(임계값)을 넘으면 1, 넘지 못하면 0을 출력하는 퍼셉트론의 성질은 활성화 함수로 계단 함수를 사용함.
- 이후 등장한 여러가지 발전된 신경망들은 계단 함수 외에도 여러 다양한 활성화 함수를 사용함.
    - Step 외의 다양한 활성화 함수
        시그모이드 (로지스틱 회귀 시간에 본 적 있을 것임.)
        Tanh (-1 ~ 1의 출력을 가짐.)
        ReLU (음수 부분을 0으로 만들어, 양수값을 주로 다룰 때는 계산 속도상의 이점을 가지는 것이 특징임.)
        leaky ReLU (ReLU에서 음수 부분에 미세한 기울기를 준 함수임.)
        ReLU가 좋음.

---

1. 임계값
- 활성화 함수에 사용되는 값으로, 보통 세타(${\theta}$)로 표현함.

$$
\begin{aligned}&\text{if } \sum_{i}^{n} w_i x_i \ge \theta \rightarrow y=1 \\&\text{if } \sum_{i}^{n} w_i x_i < \theta \rightarrow y=0\end{aligned}
$$

- 계단 함수에서는 가중합이 임계값을 넘으면 1, 넘지 못하면 0을 출력함.

---

1. 편향
- 위의 계단 함수를 구현한 식에서, 임계값을 좌변으로 넘기면 편향($b$)으로 표현할 수 있음.

$$
\begin{aligned}&\text{if } \sum_{i}^{n} W_i x_i + b \ge 0\rightarrow y=1 \\&\text{if } \sum_{i}^{n} W_i x_i + b < 0\rightarrow y=0\end{aligned}
$$

- 그림으로 표현하면 입력값 1에 편향 b가 곱해진 형태로 표현됨.

임계값과 편향은 표현의 차이라고 보면 됨.
오늘날 딥러닝에서는 임계값 대신 편향 개념을 주로 사용함.

</aside>

## 1-2. 단층 퍼셉트론과 그 한계

<aside>

### ① 단층 퍼셉트론

단순히 입력층과 출력층으로만 이루어진 퍼셉트론을 단층 퍼셉트론이라 함.

위의 인공신경망 예시에서 살펴보았던 과정은 단층 퍼셉트론임.
단층 퍼셉트론은 값을 보내는 단계와, 값을 받아서 출력하는 단계로만 이루어짐.

</aside>

<aside>

### ② 논리 게이트 구현

단층 퍼셉트론을 이용하면 AND, NAND, OR 논리 게이트를 쉽게 구현할 수 있음.

논리 게이트란?

- 디지털 회로에서 기본적인 논리 연산을 수행하는 데 사용되는 구성 요소임.
- 게이트 연산에 쓰이는 것은 n 개의 입력값과 하나의 출력값임.
- AND, OR, NOT, NAND, NOR, XOR 등 다양한 종류가 있으며,
이러한 논리 게이트를 조합하여 복잡한 논리 회로를 구성할 수 있음.

</aside>

<aside>

### ③ 대표적인 논리 게이트

1. AND 게이트
입력값이 모두 1인 경우에만 출력값이 1이 나오는 구조임.

1. OR 게이트
입력값이 모두 0인 경우에 출력값이 0이고, 나머지 경우에는 모두 출력값이 1인 구조임.

1. NAND 게이트
입력값이 모두 1인 경우에만 출력값이 0, 나머지 경우에는 모두 출력값이 1인 구조임.

1. XOR 게이트
입력값 두 개가 서로 다른 값을 갖고 있을 때에만 출력값이 1인 구조임.

</aside>

<aside>

### ④ 단층 퍼셉트론의 한계

→ XOR 게이트를 구현할 수 없음!!

퍼셉트론은 입력값과 가중치의 선형 결합으로 이루어진 구조이기 때문에, 단층 퍼셉트론은 직선 하나로 두 영역을 나눌 수 있는 선형 분류 문제에 대해서만 구현할 수 있음.

XOR 게이트는 직선이 아닌 곡선/비선형 영역으로 분리해야만 구현할 수 있음!!

따라서 퍼셉트론 하나로는 XOR을 구현할 수 없음.

→ 퍼셉트론을 여러 계층으로 쓰면 됨. 그러면 가능함.

</aside>

## 1-3. 다층 퍼셉트론

<aside>

XOR 게이트는 기존의 AND, NAND, OR 게이트를 조합하면 만들 수 있음.
즉, 퍼셉트론의 층을 더 쌓으면 가능함!

- 중간에 층을 추가하여 (입력층 + 하나 이상의 은닉층 + 출력층)으로 구성하면 됨.
- 물론 XOR 문제보다 더욱 복잡한 문제를 해결할 땐 중간에 수많은 은닉층을 추가할 수 있음!

- 두 개 이상의 은닉층을 갖는 다층 퍼셉트론이 딥러닝의 시초 모델로 발전함.
    - 심층 신경망 = 은닉층이 2개 이상인 신경망
    - 딥러닝 = 심층 신경망을 학습(가중치를 스스로 찾도록) 시킴
</aside>

# 2. 딥러닝 모델은 어떻게 학습할까?

## 2-1. 순전파와 역전파

<aside>

### ① 학습 알고리즘의 기본 구조

1. 순전파를 통해 입력값으로부터 예측 수행
2. 오차 계산을 통해 예측 정확도 측정
3. 역전파를 통해 출력층에서 다시 입력층 방향으로 오차 전달
4. 가중치 업데이트
5. 1~4를 반복.

</aside>

<aside>

### ② 연쇄 법칙

연쇄 법칙은 합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다는 법칙임.

딥러닝에서는 입력값이 여러 은닉층을 거친 후 최종 출력값이 나오는데,
이때 입력값이 최종 출력에 얼마나 영향을 미쳤는지 계산하려면 각 층이 다음 층에 미친 영향을 순서대로 곱해나가야 함.

이렇게 하면 각 단계가 연쇄적으로 결과에 얼만큼 영향을 줬는지 알 수 있을 것임. → 이게 바로 연쇄법칙임!!

$$
\frac{∂C}{∂A}=\frac{∂C}{∂B}×\frac{∂B}{∂A}
$$

은닉층이 더 많아져도 원리는 동일하게 모든 층에 걸친 기울기를 연쇄적으로 곱해 나가는 것임.

</aside>

<aside>

### ③ 기울기 소실 문제

그런데... 만약 은닉층도 엄청 많고, 각 단계마다 미분값이 1보다 작은 값이라면 어떻게 될까?

한없이 0에 가까워짐.
⇒ 기울기들을 계속해서 곱했는데, 결국 기울기가 0이 되는 문제가 발생함.

이 문제를 기울기 소실 문제라고 함. 기울기 소실 문제가 있으면 앞쪽 층의 가중치가 거의 업데이트되지 않아서 학습이 제대로 이뤄지지 않는 문제가 생김.

</aside>

<aside>

### ④ 전체적인 흐름 정리

1. 순전파 = 예측하기
2. 손실 계산 = 얼마나 틀렸는지 측정하기
3. 역전파 = 책임 추적하기
4. 가중치 업데이트 = 수정하기

</aside>

## 2-2. 손실 함수

<aside>

### ① 손실 함수란?

손실 함수는 모델의 예측이 얼마나 틀렸는지를 수치로 표현하는 방법임.
⇒ 실제값과 예측값의 차이를 계산하는 데에 사용되는 함수

</aside>

<aside>

### ② 손실 함수의 종류

1. 평균 제곱 오차
2. 이진 크로스 엔트로피
3. 크로스 엔트로피 오차

| 손실 함수 | 사용 상황 | 표현식 |
| --- | --- | --- |
| 평균 제곱 오차 | 회귀 문제 | $L = \frac{1}{n} \sum_{i=1}^{n}(y_i - \hat{y}_i)^2$ |
| 이진 크로스 엔트로피 | 이진 분류 문제 | $L = -(y \log(\hat{y}) + (1 - y) \log(1 - \hat{y}))$ |
| 크로스 엔트로피 오차 | 다중 클래스 분류 | $L = -\sum_{i=1}^{C} y_i \log(\hat{y}_i)$ |

</aside>

## 2-3. 활성화 함수

<aside>

### ① 활성화 함수란?

활성화 함수는 각 층의 출력값에 비선형성을 추가해서 신경망이 복잡한 패턴도 학습할 수 있게 해주는 함수임.

| | 단층 퍼셉트론 | 다층 퍼셉트론 |
| --- | --- | --- |
| 활성화 함수 | 계단 함수 | 시그모이드, ReLU 등 |
| 출력값 | 0 또는 1 (이진 출력) | 연속적인 값 (확률 또는 실수) |
| 특징 | 선형 문제만 풀 수 있음 | 복잡한 비선형 문제도 학습 가능 |

</aside>

<aside>

### ② 계단함수, 시그모이드, ReLU, Tanh, Softmax 등

| 활성화함수 | 출력 범위 | 주요 사용처 | 장점 | 단점 |
| --- | --- | --- | --- | --- |
| 계단 함수 | 0 또는 1 | 단층 퍼셉트론 | 구조가 단순 | 미분 불가 |
| 시그모이드 | 0 ~ 1 | 이진 분류 출력층 | 확률로 해석 가능 | 기울기 소실 문제 |
| Tanh | -1 ~ 1 | 은닉층 | 출력 중심이 0으로 학습 안정적 | 기울기 소실 문제 |
| ReLU | 0 ~ ∞ | 은닉층 전반 | 계산 빠름, 기울기 소실 완화 | 음수 입력 시 출력이 항상 0 |
| Softmax | 0 ~ 1 (합=1) | 다중 분류 출력층 | 클래스별 확률 직접 출력 | 출력층 전용, 은닉층엔 부적합 |

</aside>
