# CNN & RNN 딥러닝 정리

---

## 1. 합성곱 신경망

### 왜 이미지 처리에 CNN을 쓰냐?

MLP로 이미지를 처리하면 세 가지 문제가 생김.

1. **위치 변화에 취약** — 같은 숫자 1이라도 위치가 조금만 바뀌면 완전히 다른 입력으로 인식함
2. **파라미터 폭발** — 1000×1000 이미지를 FC Layer에 연결하면 파라미터가 수십억 개 생겨버림
3. **공간 정보 손실** — 이미지를 1차원으로 펼치는 순간 픽셀 간 공간적 관계가 사라짐

CNN은 이미지의 2차원 구조를 유지한 채로 **지역적 특징**을 학습하기 때문에 이 문제들을 해결함.

---

### CNN 전체 구조

입력 이미지 → Conv + Pool × N → Flatten → FC Layer → 출력

크게 두 파트로 나뉨:
- **특징 추출**: Convolution + Pooling 반복
- **분류**: Flatten 후 Fully Connected Layer

---

### 핵심 연산

**Convolution**
- 필터가 이미지 위를 슬라이딩하면서 특징을 뽑아냄
- 필터의 값이 MLP의 가중치 역할을 함
- 출력: **Feature Map**

**Padding**
- 이미지 바깥을 0으로 채워서 가장자리 정보 손실을 방지함
- 출력 크기를 입력 크기와 동일하게 유지할 때 씀

**Stride**
- 필터가 몇 칸씩 이동할지 결정
- stride가 클수록 feature map 크기가 작아짐

**Pooling**
- Feature map을 격자로 나눠 대표값을 뽑는 연산
- Max Pooling, Average Pooling —> 주로 Max Pooling 씀
- 차원을 줄이고, 위치 변화에 덜 민감하게 만들어줌

---

### 대표 CNN 아키텍처

모델: **AlexNet**, 연도: 2012, 핵심 특징: GPU 학습 최초, ReLU 사용으로 기울기 소실 완화, Dropout으로 과적합 방지

모델: **VGGNet**, 연도: 2014, 핵심 특징: 3×3 필터 고정, 깊이를 늘려 성능 향상. 전이 학습에 많이 쓰임

모델: **ResNet**, 연도: 2015, 핵심 특징: Skip Connection 도입. 152층까지 쌓아도 기울기 소실 없이 학습 가능

ResNet의 핵심 아이디어: 입력 x를 몇 레이어 건너뛰어 출력에 더해줌 → 차이만 학습하면 되니까 학습이 안정적

## 2. 이미지 딥러닝 응용

### 이미지 분류
- 이미지를 픽셀 값으로 변환 → 사전 정의된 클래스 중 하나로 분류
- AlexNet, VGG, ResNet, ViT 등이 사용됨

### 객체 탐지
- 이미지 내 **어디에** 무엇이 있는지 찾는 작업
- CNN으로 Feature Map 추출 후 두 가지 head를 붙임:
  - Classification Head: 클래스 예측
  - Regression Head: 위치 좌표 예측
- 대표 모델: YOLO, Faster R-CNN, SSD

### 이미지 캡셔닝
- 이미지를 보고 자연어 문장으로 설명하는 작업
- **CNN으로 특징 추출** → **Attention Network로 중요 영역 가중치 부여** → **RNN으로 문장 생성**
- CNN과 RNN이 함께 쓰이는 대표적인 사례

## 3. 순환 신경망

### 순차 데이터란?

순서 자체가 의미를 가지는 데이터. 순서가 바뀌면 의미도 바뀜.
- 자연어, 음성, 시계열, 비디오 등

일반적인 정형 데이터는 행의 순서가 바뀌어도 의미가 안 바뀌는데, 순차 데이터는 바뀜.

---

### RNN 구조와 원리

RNN은 **이전 시점의 은닉 상태**를 현재 시점으로 넘겨줘서 과거 정보를 기억함.

h_t = f x_t, h_t-1

- x_t: t 시점 입력
- h_t: t 시점 은닉 상태
- o_t: t 시점 출력

**가중치 공유**가 핵심 — U, V, W 세 가중치를 모든 시점에서 공유함. 덕분에 시퀀스 길이가 달라도 파라미터 수가 일정함.

**은닉층 vs 은닉 상태** 구분:
- 은닉층: 가중치와 활성화 함수를 포함한 구조 자체
- 은닉 상태: 매 시점 갱신되는 출력 벡터

---

### RNN의 한계

**기울기 소실**
- 역전파 시 기울기가 반복적으로 곱해지면서 0으로 수렴
- 시퀀스 앞쪽 정보가 뒤쪽까지 전달이 안 됨

**기울기 폭주**
- 기울기가 1보다 클 때 무한대로 발산
- 가중치가 폭발적으로 커져 학습 불가
- 해결: Gradient Clipping

**느린 훈련 속도**
- 순차적으로 계산해야 해서 병렬 처리 불가능

이 세 가지가 합쳐져 **장기 의존성 문제** 발생 — 시퀀스가 길어질수록 앞쪽 정보를 점점 잊어버림.

## 4. LSTM & GRU

### LSTM

RNN의 기울기 소실 문제를 해결하기 위해 설계됨.

**RNN과의 차이**: 순환 상태가 h_t 하나에서 h_t + c_t 두 개로 늘어남.

**세 가지 Gate로 정보 흐름을 제어함:**

Forget Gate: 과거 cell state에서 얼마나 잊을지 결정

Input Gate: 현재 입력을 얼마나 반영할지 결정

Output Gate: cell state에서 무엇을 뽑아 hidden state로 내보낼지 결정

- **장점**: Vanishing gradient 문제 완화, 긴 시퀀스에서 중요 정보 보존
- **단점**: 구조가 복잡하고 파라미터가 많아 학습이 느림

---

### GRU

LSTM을 단순화한 구조. cell state 없이 h_t 하나로 통합.

**두 가지 Gate만 사용:**
- **Update Gate**: 이전 상태와 현재 입력을 어떤 비율로 반영할지 결정
- **Reset Gate**: 이전 상태를 얼마나 무시할지 결정

- **장점**: LSTM보다 파라미터 수 적고 학습 빠름. 성능은 비슷하거나 더 좋은 경우도 있음
- **단점**: 매우 긴 시퀀스에서는 LSTM보다 다소 약함

**언제 뭘 쓰냐?**
- 데이터가 많고 복잡한 패턴, 긴 시퀀스 → **LSTM**
- 자원이 제한적이거나 빠른 학습이 필요할 때 → **GRU**
- 결국 실험해보고 비교하는 게 정답

## 5. Seq2Seq

### 개념

**한 시퀀스를 다른 시퀀스로 변환**하는 모델. 입력과 출력의 길이가 달라도 됨.

대표 사용 사례: 기계 번역, 텍스트 요약, 챗봇

---

### Encoder-Decoder 구조

입력 시퀀스 → Encoder → Context Vector → Decoder → 출력 시퀀스

**Encoder**
- 입력 시퀀스를 순서대로 읽으면서 정보를 압축
- 마지막 시점의 hidden state = **Context Vector**

**Context Vector**
- 입력 문장 전체를 요약한 고정 크기 벡터
- Decoder의 첫 번째 hidden state로 넘겨짐

**Decoder**
- Context Vector를 받아서 출력 시퀀스를 하나씩 생성
- t 시점 출력이 t+1 시점의 입력으로 들어감

Encoder와 Decoder는 모두 RNN/LSTM/GRU로 구성됨.

---

### Seq2Seq의 한계: 병목 현상

아무리 긴 입력이어도 Context Vector는 **고정된 크기**의 벡터 하나로 압축됨.

→ 긴 문장일수록 정보 손실이 커지고 성능이 떨어짐

이를 해결하기 위해 **Attention 메커니즘**이 도입됨.
