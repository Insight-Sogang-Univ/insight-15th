### <span style="color: #000000;background-color: #ffdba4">**1. 추천 시스템이란?**</span>

<U>**사용자의 정보 데이터를 분석, 개인의 취향에 맞는 아이템을 추천하는 알고리즘**</U>

#### <span style="color: #000000;background-color: #ffc2e5">   **1-(1) 사용되는 데이터의 종류와 특징**</span> 
**1.명시적 데이터**
- 고객들에 의해 입력된 데이터, 사용자가 선호도를 **직접적**으로 표현한 데이터
- 당시 사용자의 취향이 정확하게 반영된다는 장점
- 데이터 자체를 얻기 힘들고 과거 선호도가 현재와 동일한지 모른다는 단점

**2. 암시적 데이터**
- 고객들의 행동을 기반으로 수집된 데이터, 사용자의 선호도가 **간접적**으로 표현된 데이터
- 데이터를 얻기 쉽다는 장점
- 사용자의 취향이나 선호도를 직접적으로 알 수 없다는 단점

#### <span style="color: #000000;background-color: #ffc2e5">   **1-(2) 추천 시스템의 분류**</span> 



**1. 내용 기반 필터링**
- 콘텐츠의 유사도를 기반으로 사용자에게 추천
- 사용자가 이용하거나 선택한 아이템과 유사한 아이템 추천

**2. 협업 필터링**
- 서비스 내에 있는 다수의 사용자로부터 얻은 선호도 정보를 기반으로 사용자에게 추천  
- 두 명의 사용자가 비슷한 관심사를 가지고 있다면, 한 사용자의 데이터를 바탕으로 다른 사용자에게 추천해주는 방식  
    - 사용자 기반 : 특정 사용자와 유사한 취향의 사용자가 선호한 아이템 추천
    - 아이템 기반 : 사용자가 선호한 아이템과 유사한 아이템 추천

**3. 하이브리드 필터링**
- 2가지 이상의 다양한 종류의 추천시스템 알고리즘을 조합하여 만들어진 알고리즘
- 콘텐츠 기반 필터링과 협업 필터링 각각의 단점들을 보완하기 위해 제시

### <span style="color: #000000;background-color: #ffdba4">**2. 내용 기반 필터링**</span>

#### <span style="color: #000000;background-color: #ffc2e5">   **2-(1) 사용자 프로필**</span> 
**[추천 시스템에 사용되는 데이터]**
1. 직접 지정형
    - 사용자에게 자신이 선호하는 아이템의 특징을 명시적으로 지정하도록 하는 방법
2. 간접 지정형
    - 사용자의 과거 행동 이력에 기반하여 사용자의 프로필을 작성하는 방법

<br>

**[사용자의 프로필을 획득하는 방식]**
1. 특성 가중치 : 사용자가 가지고 있는 아이템 특성 가중치의 평균값 활용
2. 사용자의 기호를 따로 파악하거나 설문조사 또는 자발적인 키워드 입력 등으로 프로필 획득
3. 아이템에 대한 유저의 기존 평가를 바탕으로 분류 문제를 해결하는 것과 같음  
-> 좋은 사용자 프로필을 구성할수록, 추천 시스템의 정확도와 만족도가 높아짐  
-> 명시적+암묵적을 혼합하여 사용하는 것이 일반적

#### <span style="color: #000000;background-color: #ffc2e5">   **2-(2) 유사도**</span> 
- 유사도 : <U>**추천시스템을 이해하는데 중요한 수학적 개념**</U>
- 방식
    1. 인풋 데이터 만들기
        - 영상, 이미지, 텍스트 등의 형태로 존재하는 아이템 정보 획득
        - 벡터로 변환(원-핫 인코딩)
    2. 아이템 벡터 간의 유사도 측정
        - 벡터 내적 값이 클수록 두 벡터의 유사성이 높다
    3. 측정된 유사도를 바탕으로 고객이 선호하는 아이템과 유사도가 높은 아이템 선별

<br>

**[TF-IDF(Term Frequency - Inverse Document Frequency)]**
- TF-IDF
    - 단어 빈도(특정 문서에서의 단어 빈도, TF)와 역문서 빈도(전체 문서에서의 단어 빈도, IDF)를 통해서 **다른 문서에서는 등장하지 않지만 특정 문서에서만 자주 등장하는 단어**를 찾아서 문서 내 단어의 가중치를 계산하는 방법
    - TF-IDF는 각 단어가 그 문서에서 얼마나 중요한지를 나타내는 값으로, 높을수록 그 단어가 다른 문서들보다 해당 문서에서 더 중요한 역할을 한다는 뜻

#### <span style="color: #000000;background-color: #ffc2e5">   **2-(3) 내용 기반 필터링의 장단점**</span> 
- 장점
    - 아이템이 가지고 있는 특징 자체에만 집중 -> 다른 유저의 평점에 영향 x
    - 추천된 이유가 명확
- 단점
    - 아이템의 피처를 추출해야하고, 이를 통해서 추천이 이루어짐 -> 정확한 피처를 추출하지 못하면 정확도가 떨어짐
    - 새로운 장르의 아이템을 추천하기 어려움

### <span style="color: #000000;background-color: #ffdba4">**3. 협업 필터링**</span>
- 협업 필터링의 가정 : 과거에 비슷한 선호도를 가진 사용자들은 앞을로도 비슷한 선호도를 가질 가능성이 높다

#### <span style="color: #000000;background-color: #ffc2e5">   **3-(1) 메모리 기반 협업 필터링**</span> 
- = 이웃 기반 협업 필터링
- **유사도 기반**
- 사용자 A와 비슷한 취향을 가진 사용자가 선호하는 아이템을 추천
- 접근법
    - 사용자 기반 : 추천하고자 하는 타겟 유저와 유사한 **사용자**들의 평점 활용
    - 아이템 기반 : 추천하고자 하는 타겟 아이템과 유사한 **아이템**의 평점 활용
    - 새로운 추천이 필요한 플랫폼은 사용자 기반, 사용자에게 아이템에 치중된 플랫폼은 아이템 기반

#### <span style="color: #000000;background-color: #ffc2e5">   **3-(2) 메모리 기반 협업 필터링의 한계**</span> 
**1. 희소 행렬 문제**
- 주어진 평점/선호도 데이터가 적을 경우, 성능 저하
    - 실제로 유저-아이템 행렬 대부분의 원소는 비어있음(희소 행렬)
    - 메모리 기반을 사용하려면 sparsity ratio가 99.5%를 넘지 않는 것이 좋고, 그렇지 않을 경우엔 모델 기반 cf 이용

**2. 확장성 문제**
- 아이템/유저가 계속 늘어날 경우, 확장성이 떨어짐
- 유저가 많아지면 유사도 연산은 계속 늘어나고 오히려 성능이 떨어짐 -> KNN CF 이용


**3. 콜드 스타트 문제**
- 데이터가 아예 없는 경우 추천 불가능
- 크게 두가지 상황에서 발생

| 유형 | 상황 | 설명 |
| --- | --- | --- |
| **User Cold Start** | 신규 유저 유입 | 평가 이력이 없어 유사 유저 탐색 불가 |
| **Item Cold Start** | 신규 아이템 등록 | 아무도 평가하지 않아 유사도 계산 불가 |

- 해결 방법
    - 직접 지정형 프로필 활용
    - content-based filtering과 혼합하여 아이템 특성 기반으로 보완
    - 인구통계학적 정보 활용

#### <span style="color: #000000;background-color: #ffc2e5">   **3-(3) 모델 기반 협업 필터링**</span>
<U>**데이터 뒤에 숨겨진 패턴(잠재요인)을 학습하여 추천의 정확도와 확장성을 동시에 높이는 방식**</U>  
<br>

**[잠재요인]**
- 사용자와 아이템 사이의 관계를 설명해주는, 눈에 보이지 않는 숨겨진 특성(요인)
- 실제 데이터에는 없지만, 관측된 데이터를 분석하여 수치로 측정할 수 있는 값

**[Latent Factor Model]**
- 사용자-아이템 간의 숨겨진 특성을 찾기 위해 주어진 행렬을 '사용자-잠재요인','아이템-잠재요인'으로 각각 분해하여 학습하는 모델


- 차원의 수와 축의 의미를 해석 X
- 유저와 아이템을 동일한 잠재 공간의 벡터로 표현 -> 둘 사이의 거리,각도로 숨겨진 취향 유사도를 파악

**[행렬 분해]**
- 사용자-아이템 행렬을 분해해서 평점을 유저의 잠재요인과 아이템의 잠재요인의 내적으로 설명할 수 있다면, 평점 예측 가능
- 행렬 분해 방식

| **구분** | **접근 방식** | **특징** |
| --- | --- | --- |
| **SVD** | **수학적 분해 기법** | 행렬 대수학에 기반하여 행렬을 분해합니다. 희소 행렬에는 직접 적용 불가하여 결측치 처리가 필요합니다. |
| **SGD** | **최적화 알고리즘** | 경사 하강법 기반의 알고리즘으로, 관측된 평점에 대해서만 학습을 진행하여 희소성 문제를 우회적으로 해결합니다. |
| **NMF** | **제약 조건 기반 모델** | 분해된 행렬에 비음수(Non-Negative)라는 제약 조건을 추가하여 잠재 요인의 해석력을 높입니다. |
| **PMF** | **확률적 모델** | 평점 예측에 확률 분포를 사용하여 불확실성까지 모델링하는 통계적 접근법입니다. → VAE와 연결 |

- SVD
    - 어떤 행렬이든 다음의 세 행렬로 분해
        $$A=UΣV^T$$
        - $A$ : 원래의 $m×n$ 행렬 (사용자-아이템 평점 행렬)
        - $U$ :  $m×m$ 직교 행렬 (사용자 관련 정보)
        - $Σ$ : $m×n$ 대각 행렬 (가장 중요한 정보: **특잇값**)
        - $V^T$ : $n×n$ 직교 행렬 (아이템 관련 정보)
    - Full SVD
        - 모든 특잇값과 벡터 사용
        - 계산 비용이 크고 노이즈 포함
    - Truncated SVD
        - 상위 k개의 특잇값과 해당 벡터만 사용
        - 노이즈 제거, 핵심 패턴만 학습
- SGD
    - SVD의 한계 보완
    - SGD의 목표 : 목적 함수
        - 관측된 평점과 예측 평점 사이의 오차를 최소화하는 것을 목적으로 함
        - 오차의 제곱합을 최소화하고, 정규화 항 추가
- NMF
    - 두개의 행렬 P와 Q^T의 모든 원소가 0이상이어야 한다는 제약 조건을 추가한 행렬 분해 기법
    - 장점 : 직관적 의미 부여 가능
- PMF
    - '누가 무엇을 좋아할지' 예측하는 데에 목표를 두무
    - 확률적 모델로 접근
    - 베이지안 추론 기법과 결합 -> 더 정교한 추천 모델로 평가 받기도 함  


**[결론]**  
- 메모리 기반 협업 필터링의 한계를 보완하기 위해, 모델 기반 협업 필터링을 사용
- 모델 기반 협업 필터링은 행렬 분해(MF)를 통해 유저와 아이템의 잠재 요인 벡터를 만들고, 잠재 요인 벡터 간의 내적 연산을 통해 평점을 예측함

### <span style="color: #000000;background-color: #ffdba4">**4. VAE 기반 협업 필터링**</span>
- 사용자의 아이템 상호작용 벡터를 입력으로 받아, 잠재 공간에서 재구성하도록 VAE를 학습하는 방식
- 구조
    - 인코더 : 사용자 이력을 잠재변수 z 분포로 매핑
    - 디코더 : z로부터 아이템 분포 복원
- 장점
    - 불확실성 반영
    - 희소한 상호작용 데이터에서도 강건

#### <span style="color: #000000;background-color: #ffc2e5">   **4-(1) MF의 한계와 딥러닝의 등장**</span> 
- MF의 근본적 한계
    - 선형성의 한계
        - 평점 예측이 단순한 잠재 벡터의 내적으로만 이루어짐
        - 해법 : 비선형 활성화 함수를 가진 신경망 사용
    - 불확실성 무시
        - 사용자 취향을 단 하나의 벡터값으로 표현
        - 해법 : 확률 분포로 취향을 모델링

-> 이 한계를 극복하기 위해 등장한 방식 중 하나가 VAE

#### <span style="color: #000000;background-color: #ffc2e5">   **4-(2) 왜 AutoEncoder(AE)가 아닌 VAE(Variational AE)인가?**</span> 
<U>**VAE는 불확실성을 모델링함으로써 추천 시스템의 고질적인 문제인 희소성에 대한 해답을 제시함**</U>

| 구분 | AutoEncoder (AE) | Variational AutoEncoder (VAE) |
| --- | --- | --- |
| **잠재 벡터 표현** | 단일 값 (Point Estimate) | **확률 분포** (평균 $\mu$, 분산 $\sigma$) |
| **데이터 불확실성** | 반영 불가 (단순 복원에 집중) | **데이터의 불확실성을 모델링**하여 안정성 확보 |
| **적용 시 장점** | 비선형성 학습 | 희소한 데이터에서 더욱 안정적이며, 일반화 성능 우수 |

#### <span style="color: #000000;background-color: #ffc2e5">   **4-(3) VAE 기반 CF(Mult-VAE)의 작동 구조**</span> 
- 사용자의 아이템 상호작용 벡터를 입력으로 받아, 잠재 공간에서 재구성하도록 학습하는 방식

| 구성 요소 | 역할 |
| --- | --- |
| **입력** ($x$) | 사용자의 아이템 상호작용 벡터 |
| **인코더** | 사용자 이력을 잠재변수 $z$ 분포로 매핑 |
| **잠재 벡터** ($z$) | 샘플링된 최종 취향 벡터 |
| **디코더** | $z$로부터 아이템 분포를 복원 |

#### <span style="color: #000000;background-color: #ffc2e5">   **4-(4) VAE의 비밀 병기 : 목적함수와 KL Divergence**</span> 
<U>**VAE의 핵심 : 목적 함수인 ELBO에 있으며, 이를 통해 두가지 목표를 동시에 최적화함**</U>

$\mathcal{L}(x; \theta, \phi) = \underbrace{E_{q_\phi(z|x)} [ \log p_\theta(x|z) ]}_{\text{재구성 오차 최소화 (Reconstruction Loss)}} - \underbrace{D_{\text{KL}}(q_\phi(z|x) || p(z))}_{\text{잠재 공간 정규화 (KL Divergence)}}$

1. 재구성 오차 최소화 (첫번째 항)
    - 목표 : 디코더가 잠재 벡터를 사용하여 원래 입력을 최대한 비슷하게 복원하도록 만듦.
2. KL Divergence 항(두번째 항)
    - 목표 : 인코더가 출력하는 잠재분포가 사전 분포와 유사하도록 강제
    - 역할
        - 이 항은 모델이 특정 데이터에 과적합하여 잠재 공간을 불규칙하게 만드는 것 방지
        - 잠재 공간을 넓고 부드럽게 유지  
        -> 희소성에 강한 이유

- VAE-CF의 주요 장점
    - 확률적 잠재 분포 모델링
        - 불확실성 반영
        - 노이즈에 자유로움
    - 희소한 상호작용 데이터에서도 살아남음
        - KL Divergence -> 잠재 공간 정규화
        - 희소성 문제에 효과적으로 대응
    - 비선형성 학습

### <span style="color: #000000;background-color: #ffdba4">**5. GNN 기반 사용자 선호 예측**</span>

#### <span style="color: #000000;background-color: #ffc2e5">   **5-(1) GNN(Graphic Neural Network, 그래프 신경망)**</span> 
**[그래프란?]**
- 노드(현실 세계의 개체)와 엣지(개체 간 관계나 상호작용)를 모아 구성한 자료구조
- 사용 이유
    - 추상적인 개념(관계, 상호작용..)을 다루기에 적합
    - Non-Euclidean space도 표현하고 학습 가능
        - 관계와 구조가 복잡하고 불규칙한 데이터를 모델링하고 학습하도록 하는 방법론

#### <span style="color: #000000;background-color: #ffc2e5">   **5-(2) GNN의 구조**</span> 
**[작동 구조]**
1. 그래프 구조 생성
    - 노드는 개별 객체, 엣지는 객체 간 관계나 연결성 의미
2. 메시지 패싱
    - 각 노드가 이웃 노드로부터 정보를 받아 전달
    1. 메시지 생성 - 이웃 노드로 전달할 메시지 생성
    2. 집계 - 받은 여러 이웃의 메시지를 하나로 합침
    3. 갱신 - 자신의 노드 상태 업데이트
3. 최종 출력

**[핵심 아이디어]**
- 내 이웃 노드의 정보를 나에게 전달/요약

#### <span style="color: #000000;background-color: #ffc2e5">   **5-(3) GNN 프로세스**</span> 



1. 그래프 구조 생성
    - 노드와 엣지를 정의한 후, 각각의 피처를 담아 표현
2. 메시지 패싱
    - 그래프에서 노드 간 정보를 전달하고 상호작용 하는 방법  
    - 단계(k개의 레이어로 구성된 GNN 모델은 k번 반복)   
        1) 메시지 생성 : 각 노드가 이웃에게 보낼 자신의 정보 생성   
        2) 집계 : 메시지 생성 이후, 여러 이웃 노드에서 받은 메시지를 하나의 벡터로 통합하는 단계    
        3) 갱신 : 집계된 이웃 정보와 자신의 이전 상태를 결합해 새로운 임베딩 계산

#### <span style="color: #000000;background-color: #ffc2e5">   **5-(4) LightGCN**</span> 
- 추천 시스템에 특화된 GNN, 복잡성을 제거하고 최적화한 버전
- 제거 요소
    1. 특징 변환 제거
        - 레이어마다 복잡했던 학습 파라미터 삭제
    2. 비선형 활성화 함수
        - ReLU와 같은 활성화 함수를 사용하지 않고, 이웃 정보를 합침 -> 주로 정규화된 합 혹은 평균 사용
    


#### <span style="color: #000000;background-color: #ffc2e5">   **5-(5) GNN의 장단점**</span> 
- 장점
    - 고차원 협업 시그널 포착
        - 나와 취향이 비슷한 사용자들과 내가 본 아이템과 연관된 다른 아이템들의 정보를 직접 반복해서 반영 가능
    - 정형화되지 않은 데이터 학습 가능
        - 관계로 학습 -> 관계망 속에서 정보를 효과적으로 전파 가능
    - 콜드 스타르 문제 완화에 효과적
        - 노드의 고유 특징과 주변 이웃과의 관계를 활용해 즉시 임베딩 생성 가능
- 단점
    - 과평탄화 문제
        - 모든 노드의 임베딩이 같은 값으로 수렴하는 현상
        - 레이어가 많아질수록 연결된 그래프 내의 모든 노드 간의 임베딩이 비슷해지는 문제 발생
    - 과압축 문제
        - GNN의 층이 깊어질수록 이웃의 범위는 급증하는데 고정된 크기의 벡터 하나에 모두 압축하려고 하면 정보가 손실될 위험
    - 계산 복잡
        - 수십억개의 노드와 엣지 포함 -> 막대한 자원과 메모리 요구
    - 해석의 어려움
        - 순서 반영 X

### <span style="color: #000000;background-color: #ffdba4">**6. Transformer 기반 추천 시스템**</span>

#### <span style="color: #000000;background-color: #ffc2e5">   **6-(1) 기존 추천 시스템의 순차적 한계**</span> 
**[CF의 순차적 한계]**
- 순서 무시 : 구매 순서를 반영 X
- 문맥 부족 : 구매 시기에 따른 중요도를 구분하기 어려움
- 예측의 한계 : 다음 행동 예측보다는 일반적인 선호도에 집중

**[GNN의 순차적 한계]**
- 지역적 정보 전파의 한계
    - 기본적으로 메시지 패싱을 통해 관계를 학습
    - [A->B->C]에서 C가 A의 정보를 받으려면 반드시 B를 거쳐야함 (정보 희석 위험)
- 장기 의존성 문제
    - 정의된 그래프 구조에 의존  


=> 따라서 사용자의 행동 시퀀스를 이해하고 다음 행동을 예측하는 트랜스포머 모델이 대안으로 등장

#### <span style="color: #000000;background-color: #ffc2e5">   **6-(2) Transformer : 사용자 행동을 순서로 접근**</span> 
<U>**트랜스포머는 시퀀스 내의 문맥과 핵심 의도를 파악하는 데 강력**</U>

1. 문맥에 따라 달라지는 동일 아이템의 의미
- 등산화 클릭 상황
- 문맥 A(캠핑 준비)
    - 행동 순서 : 텐트 -> 침낭 -> **등산화** 
    - 트랜스포머의 해석 : 등산화가 텐트, 침낭과 강하게 연산 -> 캠핑 장비로 해석됨
    - 다음 추천 : 캠핑용 코펠, 배낭, 랜턴
- 문맥 B (겨울 출퇴근 준비)
    - 행동 순서 : 롱패딩 -> 방한 장갑 -> **등산화**
    - 트랜스포머의 해석 : 등산화가 롱패딩, 방한 장갑과 강하게 연산 -> 방한 신발로 해석
    - 다음 추천 : 발열 내의, 울 양말, 넥워머

2. 시간 순서를 뛰어넘는 핵심 의도 포착
- 사용자의 행동이 길어질때, 최근 행동이 아니라 멀리 떨어진 과거의 행동이 다음 행동의 결정적 키가 될 수 있음
- 행동 순서(시간 순)
    1. 레이저 프린터 구매
    2. 가습기 검색
    3. 강아지 사료 클릭
    4. (약 20개의 서로 다른 행동 발생)
    5. 토너 검색
- 트랜스포머의 해석
    - 사용자가 토너를 검색했을 때, 모델은 과거 모든 행동을 동시에 참조
    - self attention 작동 : 토너라는 아이템은 1번의 레이저 프린터와 가장 강한 어텐션 관계
    - 다음 추천 : 레이저 프린터 전용 토너  

=> Transformer는 시간 순서나 거리에 얽매이지 않고, 전체 시퀀스 내에서 현재 의도와 가장 관련성이 높은 핵심 문맥을 정확히 주목하여 예측

#### <span style="color: #000000;background-color: #ffc2e5">   **6-(3) Transformer 기반 주요 추천 시스템 모델**</span> 
**[SASRec]**
- 단방향 어텐션으로 과거 정보만 활용하여 미래 행동을 추측하는 방식으로 훈련을 받은 추천 시스템 모델
- 방식 : 단방향 어텐션
- 어텐션 계산 : 현재 아이템은 오직 과거의 아이템들만 참조(미래 정보 차단)
- 목표 : 다음 아이템 예측
- 특징
    - 과거 정보만 활용하여 시간의 흐름을 명확하게 반영
    - 학습 과정에서 사용자의 다음행동을 예측하는 데에 초점
- 문제점
    - 계산량 많음 : 각 타임 step에 따른 loss모두 계산
    - 단방향성 : 이전에 등장한 값만 참고 가능


**[BERT4Rec]**
- SASRec의 두가지 문제점 보완
- 양방향 어텐션으로 과거 정보와 미래 정보를 활용하여 예측하는 방법을 훈련한 모델
- 방식 : 양방향 어텐션
- 어텐션 계산 : 시퀀스의 일부를 [MASK]토큰으로 일부러 가리고, 가려진 아이템이 '과거와 미래의 모든 아이템'을 참조 가능
- 목표 : 시퀀스 내 가려진 아이템 예측
- 특징
    - 아이템의 양방향 문맥을 모두 고려
    - 학습 과정에서 시퀀스 전체의 선호도를 파악한느 데에 초점

**[SASRec vs BERT4Rec]**
- 학습 방식의 차이
    - SASRec: 단방향 어텐션 → 최대한 현실과 유사한 방식으로 과거 데이터만 보고 학습
    - BERT4Rec: 양방향 어텐션 → 아이템 간의 숨겨진 복잡한 맥락을 훨씬 더 깊고 풍부하게 학습
