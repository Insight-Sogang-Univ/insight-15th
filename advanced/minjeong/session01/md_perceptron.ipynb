# 1. 인공신경망의 시작: 퍼셉트론

### **1-1. 인공신경망** <br>
##### (1) 인공신경망 <br>
= 뇌의 신경망을 모방한 모델    

- 입력 -> 가중치와 편항 적용 -> 활성화 함수 -> 출력    

##### (2) 퍼셉트론 <br>
= 초기 형태의 인공신경망, DL의 기원이 되는 선형 이진 분류기      

- 입력 -> 가중합 -> 활성화 함수 -> 출력    
    - 가중치: 입력값이 출력 값에 주는 중요도 결정하며, 가중치 값이 클수록 해당 입력 값의 중요도 높음     
    - 활성화 함수: 임계값 초과 여부에 따라 출력값 결정 (퍼셉트론은 계단 함수)       
    - 임계값: 활성화 함수에 사용되는 값으로, 계단 함수에서 임계값 넘으면 1 출력    
    - 편향: 계단 함수 구현한 식에서 임계값을 좌변으로 넘기면 편향으로 표현             

### **1-2. 단층 퍼셉트론과 한계** <br>
##### (1) 단층 퍼셉트론 <br>
= 입력층과 출력층으로만 이루어진 퍼셉트론     

##### (2) 논리 게이트 구현 <br>
= 디지털 회로에서 기본적인 논리 연산을 수행하는 데 사용되는 구성 요소       
- AND, OR, NOT, NAND, NOR, XOR 등 다양한 논리 게이트 조합해 복잡한 논리 회로 구성 가능         
    - AND: 입력 값이 모두 1인 경우에만 출력값 1       
    - OR: 입력값이 모두 0인 경우에 출력값 0, 나머지는 모두 1      
    - NAND: 입력값이 모두 1인 경우엠ㄴ 출력값 0, 나머지는 모두 1    
    - XOR: 입력값 두 개가 서로 다를 때에만 출력값 1            

##### (3) 단층 퍼셉트론 한계 <br>
- XOR 게이트 구현 불가능 (입력값과 가중치의 선형 결합으로만 이루어졌기 때문) => 곡선/비선형 영역으로 분리해야만 구현 가능             

### **1-3. 다층 퍼셉트론** <br>
- 중간에 층을 추가해 구성      
- XOR 문제보다 복잡한 문제 해결 시 중간에 수많은 은닉층 추가 가능        
- 두 개 이상의 은닉층을 가지는 MLP가 딥러닝의 시초 모델로 발전      
    - DNN: 은닉층이 2개 이상       
    - DL: 심층 신경망 학습 (가중치 스스로 찾음)

# 2. 딥러닝 모델은 어떻게 학습할까

### **2-1. 순전파와 역전파** <br>
##### (1) 학습 알고리즘 기본 구조 <br>
- 순전파를 통해 입력값으로부터 예측 -> 오차 계산해 예측 정확도 측정 [손실 계산] -> 역전파를 통해 입력층 방향으로 오차 전달 -> 가중치 업데이트 -> 오차값 0에 가까워질 때까지 반복        
##### (2) 연쇄 법칙 <br>
= 합성 함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다는 법칙       
- 입력값이 최종 출력에 미치는 영향을 계산하려면 각 층이 다음 층에 미친 영향 순서대로 곱하고, 각 단계가 연쇄적으로 얼마나 영향을 주었는지       
- 오차를 줄이는 데 손실에 얼마나 기여하는지 계산      
##### (3) 기울기 소실 문제 <br>
- 은닉층이 많고 각 단계마다 미분값이 1보다 작으면 기울기가 결국 0이 되는 문제 발생       
- 앞 층의 가중치가 거의 업데이트되지 않아 학습이 제대로 이루어지지 않음 <br>   
 
### **2-2. 손실 함수** <br>
= 모델이 정확한지 오차 계산하는 방법 [실제 값과 예측값 간 차이 계산에 사용하는 함수] <br>
##### (1) 손실 함수 종류 <br>
- MSE : 평균 제곱 오차 / 예측 값이 연속적이고 예측값과 실제 값 간 차이를 직접 측정하는 회귀에 적합     
- BCE : 이진 크로스 엔트로피 / 이진 분류 문제에 사용     
- CEE : 크로스 엔트로피 오차 / 다중 클래스 분류 문제에 적합      

### **2-3. 활성화 함수** <br>
##### (1) 활성화 함수 정의 <br>
= 각 층의 출력값에 비선형성을 추가해 신경망이 복잡한 패턴도 학습할 수 있게 하는 함수         
- 단층/다층 퍼셉트론에 따라 다르게 사용됨    
##### (2) 계단 함수 <br>
- 가중치: 각 입력 신호가 결과에 미치는 영향력 제어     
- 입력값과 가중치의 곱을 모두 더해 임계값 넘으면 1, 아니면 0 출력        

##### (3) 시그모이드 함수 <br>
- 입력 신호를 0~1 연속적인 값으로 변환하는 비선형 함수       
- 계단 함수와 달리, 출력값이 연속적으로 변화     

##### (4) ReLu 함수 <br>
- 입력이 0보다 크면 그대로, 0 이하면 0 출력하는 비선형 함수       
- 딥러닝에서 가장 많이 사용 (간단하고 효율적인 계산, 비선형성 제공, 양수 구간에서 미분값이 항상 1이므로 기울기 소실 문제 완화)

##### (5) Tanh 함수 <br>
- 시그모이드 함수 변형     
- -1~1 사이 출력 -> 중심이 0이면 가중치 업데이트가 안정적     
- 입력값 매우 크거나 작으면 기울기 소실 문제 발생 가능      

##### (6) softmax 함수 <br>
- 각 클래스에 대한 출력값 확률로 변환     
- 다중 클래스 분류 문제의 출력층에서 사용     

### **2-4. 경사 하강법** <br>
##### (0) 참고 <br>
- 배치: 매개변수 조정을 위해 사용되는 데이터의 묶음 (하이퍼파라미터)      
- 에포크: 모든 데이터셋을 학습하는 횟수       
- 이터레이션/스텝: 한 번의 에포크를 끝내기 위해 필요한 배치의 수      

##### (1) 배치 경사하강법 <br>
- 오차 구할 때 전체 데이터를 한 묶음으로 묶어 업데이터 횟수 적으나 메모리 많이 필요함       
- 항상 같은 데이터에 대해 경사를 구하므로 안정적으로 수렴      

##### (2) SGD <br>
= 배치 크기가 1인 경사 하강법      
- 랜덤으로 선택된 하나의 데이터에 대해서만 계산          
- 자원이 적어도 사용 가능하며, 변경폭이 불안정하거나 정확도 낮을 수 있음      

##### (3) 미니 배치 경사 하강법 <br>
- 배치 크기 적절히 지정해 매개 변수 값 조정         
- 가장 많이 사용   <br>

### **2-5. 옵티마이저** <br>
- 경사하강법 효율적&안정적으로 수행하여 효과적인 최솟값 찾아감      

##### (1) 모멘텀 <br>
- 경사하강법 + 관성 -> 경사하강법에서 계산된 접선 기울기에 바로 한 시점 전의 접선 기울기값 일정 비율 반영       

##### (2) RMSProp <br>
- 학습률 조정해 안정적이고 빠른 학습       
- 최근 기울기 참고     

##### (3) Adam <br>
- 모멘텀 + RMSProp       
- 방향과 학습률 두 가지 모두 고려

### **2-6. 데이터 증강 **<br>
= 학습 데이터셋에 다양한 변화를 가하는 것      
- 기존 데이터에서 다양한 변형을 만들어 학습 데이터를 늘려서 효과적으로 학습할 때 사용     
-> 데이터 부족 문제 해결, 모델의 견고함 향상, 과적합 감소       
- 이미지, 자연어 처리 등에 활용          
- 모델 성능 향상, 데이터 의존성 감소, 과적합 완화, 데이터 프라이버시 보호
