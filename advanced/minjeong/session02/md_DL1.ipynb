## 1. 합성공 신경망 (CNN) <br>
#### **1-1.CNN 사용하는 이유** <br>
- 픽셀을 일렬로 펼쳐 처리하는 MLP와 달리, CNN은 2차원 구조를 유지하며 주변 픽셀 관계와 시각적 특징 학습하기 때문      

| MLP | CNN |    
| ---- | ---- |     
|위치 변화에 따른 숫자 인식 성능 저하|Convolution [특징 학습] & Pooling [차원 축소] => 위치와 무관|            
|완전 연결 구조로 인한 파라미터 폭발 (가중치 수가 이미지 픽셀만큼 생성됨)|부분 연결 구조 & Convolution => 특정 피처에만 집중해 효율적|            
|1차원 벡터로 변환하는 과정에서 이미지 공간 정보 손실 큼|공간 정보 보존하며 학습 & 특징 추출 후 flatten하므로 손실 정도 낮음|            

#### **1-2. 합성곱과 풀링** <br>
(1) 특징 추출 - 합성곱, 풀링      
(2) 클래스 분류 - Fully Connected Layer        

**합성곱 layer** <br>
= 전체 이미지를 일정 크기의 필터로 훑으며 이미지 특징 학습       
- 필터/커널이 이미지 위 슬라이팅 -> 각 위치에서 요소별 곱셈 후 합계 -> 피처 맵에 결과 저장        

**Padding** <br>
= 필터 거치며 정보 손실 방지를 위해 이미지 바깥을 0으로 둘러싸는 기법      
- 이미지 크기 유지하며 가장자리 특징까지 반영         

**Stride** <br>
= 이미지를 몇 칸씩 점프하면서 훑을지 결정하는 값      
- 계산량 감소, 효율적으로 요약      
- stride 클수록 피처맵 작아짐      

**Pooling** <br>
= 피처맵에서 특징 추출을 위해 이미지를 격자 형태로 분할 후 해당 여역 내 가장 큰 값 / 평균 취하는 것        
- 모델 용량과 계산량 감소 & 노이즈 제거 & 과적합 방지       
- 정보 소실 위험         

**=>Convolution, Pooling으로 특징 추출. 연산 과정에서 하이퍼파라미터로 padding, stride 활용**  <br>

#### **1-3. CNN 아키텍처** <br>
**AlexNet** <br>
- 최초로 GPU를 이용해 학습한 대규모 CNN 모델     
- Conv. layer 5개 + Fully connected later 3개     
- ReLU 함수로 기울기 소실 문제 해소      
- 드롭아웃으로 과적합 방지       

**VGGNet** <br>
- 이미지 인식에 강한 깊은 층 CNN 모델     
- Conv. + Pool 여러 번 반복하며 필터 크기 3*3 고정    
- 전이 학습에 많이 사용 & 층의 깊이 늘려 정확도 높음 & 초반 연산량 많음     

**ResNet** <br>
- 잔차 연결 개념 도입해 깊은 신경망의 소실 문제 해결한 모델    
    - 잔차 연결: 이전 레이어 출력을 건너뛰고 다음 레이어 입력으로 직접 추가      

## 2. 이미지 딥러닝 응용 <br>
#### **2-1. 이미지 분류** <br>
= 이미지를 픽셀 값으로 인식해 사전 정의된 클래스 중 하나로 분류      

#### **2-2. 객체 탐지** <br>
= 이미지에서 객체와 객체를 둘러싸는 가장 작은 직사각형인 bounding box 찾는 작업   
- CNN으로 피처 맵 얻은 후, 객체의 클래스와 위치 좌표 예측하는 head를 올림     
- YOLO, Faster R-CNN, SSD       

#### **2-3. 이미지 캡셔닝** <br>
= 이미지 내 객체에 대한 판단 + 객체 간 관계 파악해 자연어로 표현      
- CNN으로 특징 추출 - Attention Network 통과하며 가중치를 두어 단어 추출 - RNN으로 문장 완성   

## 3. RNN <br>
#### **3-1. 순차 데이터** <br>
= 순서에 의미가 있어 순서가 달라지면 의미가 손상되는 데이터 ex. 자연어, 생물학적 서열 데이터 등      
- 요소의 순서 의존성: 데이터 구성 요소들의 순서 자체가 핵심 정보   
- 자기상관성: 이전 시점의 데이터가 다음 시점 데이터에 영향을 미침     

#### **3-2. RNN 구조와 원리** <br>
= 순환하는 구조를 가지는 순차 데이터 처리에 특화된 인공 신경망    
=> 시퀀스 데이터 입력 받아 순서 정보 유지하며 처리하고, 순화하는 은닉층이 매 시점의 은닉 상태 업데이트. 이전 시점 값을 현재 시점으로 넘김.     

**가중치 공유**    
= 동일한 가중치를 시퀀스의 다양한 시점에서 반복 적용 -> 시퀀스 데이터의 길이와 시점 위치에 상관 없이 작동         
- 학습 파라미터 수 감소 + 일반화 능력 향상     

#### **3-3. RNN 한계와 장기 의존성 문제** <br>

**RNN 한계점** <br>
- 전체 시퀀스 모두 읽은 후 역전파 -> 역전파 구하는 과정에서 미분값이 반복적으로 곱해진다는 특징으로 인해          
- 기울기 소실: 곱해지는 미분값 < 1 이면 기울기 0으로 수렵        
- 기울기 폭주: 곱해지는 미분값 > 1 이면 기울기 발산          
- 느린 훈련 시간     

**장기 의존성 문제**<br>
= 시퀀스 앞 부분의 중요한 정보 잊어 맥락 파악 능력 급격히 저하     
=> 더 길고 복잡한 시퀀스 데이터 안정적으로 다루는 새로운 대안 필요 (LSTM, GRU)      

## 4. LSTM, GRU <br>
#### **4-1. LSTM** <br>
- RNN의 기울기 소실 문제 해결하는 모델     
- 기억할 내용과 잊어버릴 내용을 선택해 중요한 정보들을 오래 가져감      

#### **4-2. GRU** <br>
- LSTM에서 발전, 별도의 메모리 셀 없이 게이트 수 줄요 간소화   
- Forget gate + Input gate -> Update gate        
- Reset gate        

#### **4-3. 비교 분석** <br>
|구분|LSTM|GRU|     
|---|---|---|     
|구조|입력, 삭제, 출력 게이트|업데이트, 리셋 게이트|       
|복잡도|파라미터 수 많음|적음|       
|학습 속도|느림|빠름|       
|메모리 사용량|큼|작음|       
|장기 의존성 처리|강력|약함|       
|연구/사례 축적|매우 많음|적음 - 최근 등장|       
|성능|데이터 충분하고 문제 복잡할 때|데이터 적거나 효율성 중요할 때|    
|사용|기계 번역, 언어 모델링, 장기 시계열 예측|음성 인식, 스트리밍 데이터, 실시간 비디오 분석|          

## 5. Seq2Seq <br>
= 한 시퀀스를 다른 시퀀스로 변화하는 딥러닝 모델     

#### **5-1. 기본 구조** <br>
- 아이템 입력 갯수 != 출력 갯수 여도 됨    
- 인코더-디코더 모델     
    - 인코더: 입력된 시퀀스 읽고 압축해 디코더에게 넘겨줄 문맥 정보 준비       
    - 디코더: 압축된 정보 바탕으로 원하는 시퀀스 생성      
- 컨텍스트 벡터: float형의 벡터, 인코더 중 마지막 시점의 은닉 상태     

#### **5-2. 한계 - 병목 현상** <br>
- 고정된 문맥 벡터에 소스 문장의 정보를 압축하면 병목 현상 발생 가능     
    - 고정된 길이에 입력 시퀀스의 모든 정보 담지 못해 손실 가능   
    - 긴 입력 시퀀스의 경우 정보 손실 심각하고, 모델 성능 하락 가능성    
- 기울기 소실 문제, 병렬화 불가능 문제 등
