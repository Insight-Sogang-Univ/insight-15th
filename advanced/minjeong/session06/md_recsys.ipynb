# **추시 기초** <br>
## 1. 추천 시스템 <br>
= 사용자 정보 데이터 분석해 개인 취향에 맞는 아이템 추천하는 알고리즘          

#### 1-1. 사용되는 데이터 종류와 특징 <br>

**명시적 데이터** <br>
= 고객이 입력한, 선호도를 직접 표현한 데이터      
- 사용자 취향이 정확하게 반영되지만 데이터를 얻기가 힘듦      

**암시적 데이터** <br>
= 고객 행동 기반으로, 선호도가 간접적으로 표현된 데이터    
- 높은 수치가 높은 선호도를 의미하지 않음    

#### 1-2. 추천 시스템 분류 <br>

- 내용 기반 필터링: 콘텐츠 유사도 기반으로 추천하는 방법       
- 협업 필터링: 다수 사용자에서 얻은 선호도 정보 기반으로 추천    
    - 사용자 기반 / 아이템 기반      
- 하이브리드 필터링: 2가지 이상의 추시 조합

## 2. 내용 기반 필터링 <br>

#### 2-1. 사용자 프로필 <br>
- 특성 가중치: 사용자가 가진 아이템 특성 가중치 평균 값 활용    
- 프로필 획득: 사용자 기호 파악, 설문조사, 자발적인 키워드 입력     
- 아이템에 대한 기존 평가로 분류 문제 해결하는 것과 같음   
=>  좋은 사용자 프로필 구성할수록 추시 정확도와 만족도 높아짐     

#### 2-2. 유사도 <br>

**TF-IDF** <br>
= 단어 빈도 (특정 문서 내 단어가 얼마나 자주 등장하는지)와 역문서 빈도 (전체 문서에서 특정 단어가 얼마나 자주 등장하는지)를 통해 다른 문서에 등장하지 않지만 특정 문서에서만 자주 등장하는 단어를 찾아 문서 내 단어 가중치 계산       
=> 높을수록 그 단어가 해당 문서에서 더 중요한 역할을 한다        


#### 2-3. 장단점 <br>
(+)     
- 아이템이 가진 피처 자체에만 집중하므로 다른 유저 평점에 영향 받지 않아, 콜드스타트 문제에서 자유로움    
- 아이템의 피처를 통해 추천하므로 추천 이유 설명 용이      

(-)    
- 정확한 피처 추출 못하면 정확도 낮아짐      
- 기존 아이템과 유사한 아이템을 추천해 새로운 장르 추천 어려움      

## 3. 협업 필터링 <br>

가정: 과거에 비슷한 선호도를 가진 사용자들은 앞으로도 비슷한 선호도 가질 가능성 높음      

#### 3-1. 메모리 기반 협업 필터링 <br>
= 사용자와 비슷한 취향을 가진 다른 사용자가 선호하는 아이템 추천        

- 사용자 기반: 추천하려는 타겟 유저와 유사한 사용자들의 평점 활용         
    - 아이템 수 > 사용자 수, 아이템이 수시로 바뀌는 환경, 다양성 높음      
- 아이템 기반: 추천하려는 타겟 아이템과 유사한 아이템의 평점 활용             
    - 사용자 수 > 아이템 수, 고정적 환경에 적합, 다양성 낮음, 사용자 유입 많은 대형 플랫폼에 유리          

#### 3-2. 메모리 기반 협업 필터링 한계 <br>
- 희소 행렬: 주어진 평점/선호도가 데이터가 적으면 성능 저하   
- 확장성 문제: 아이템/유저가 계속 늘어나면 확장성이 떨어짐     
- 콜드 스타트: 데이터가 아예 없는 경우, 기존의 평점이나 행동 데이터를 기반으로 유사도를 계산하기 때문에 추천 불가능      

#### 3-3. 모델 기반 협업 필터링 <br>
= 데이터로부터 패턴을 학습한 모델을 만들어 평점 예측    
= 관측된 평점 데이터를 학습해 관측되지 않은 평점 예측하는 방식      
<br>
- 잠재 요인: 사용자와 아이템 사이의 관계를 설명하는 눈에 보이지 않는 숨겨진 특성      
- 잠재 요인 모델: 사용자와 아이템을 보이지 않는 축을 사용해 나타낼 수 있고 이를 통해 평점 예측할 수 있다고 가정하는 모델     
<br>
- 행렬 분해: 유저 + 아이템의 잠재 요인의 내적으로 

# **딥러닝 기반 추시** <br>
## 1. VAE 기반 <br>
= 사용자의 아이템 상호작용 벡터를 입력으로 받아 잠재 공간에서 재구성하도록 VAE 학습하는 방식    
- 확률적 잠재 분포로 사용자 선호 모델링해 불확실성 반영       
- 희소한 상호작용 데이터에서도 강건       
- 사용자 이력을 잠재변수 z 분포로 매핑하는 인코더 / 아이템 분포를 복원하는 디코더     

#### 1-1. MF 한계와 딥러닝 등장 <br>
- 잠재 벡터의 내적으로만 이루어지는 선형성의 한계 -> 비선형 활성화 함수를 가진 신경망 사용   
- 사용자 취향을 하나의 벡터 값으로만 표현한다는 한계 -> 확률 분포로 취향 모델링       

#### 1-2. VAE인 이유 <br>
- 불확실성을 모델링해, 추시의 고질적 문제인 희소성에 대한 해답 제시       

#### 1-3. VAE 기반 CF 작동 구조 <br>
- 사용자의 아이템 상호작용 벡터를 입력으로 받아 잠재 공간에서 재구성하도록 학습     
- 확률적 잠재 분포 모델링      
- 희소한 상호작용 데이터에서도 살아남는다는 장점    
- 비선형적 패턴 학습해 추천 성능 극대화             

#### 1-4. 목적 함수와 KL Divergence <br>
- 목적 함수인 ELBO를 통해 두 목표 동시에 최적화      
- 재구성 오차 최소화: 디코더가 잠재 벡터로 원래 입력 최대한 비슷하게 복원    
- KL Divergence: 인코더가 출력하는 잠재 분포가 사전 분포와 유사하도록 함     


## 2. GNN 기반 <br>
= 관계 자체를 데이터로 보고 모델링     

#### 2-1. GNN 개념 <br>
- 그래프 = 노드 + 엣지      
- 그래프를 형성한다는 점에서 모티브를 얻음     
- 관계나 상호작용과 같이 추상적 개념을 다루기에 적합하고, 비유클리드 공간 데이터 표현하고 학습 가능     
=> 이웃 노드의 정보를 전달하고 요약하는 메시지 패싱으로 정보 전달    

#### 2-2. 구조와 프로세스 <br>
- 그래프 구조 생성 -> 메시지 패싱 -> 최종 출력       
    - 메시지 패싱: 메시지 생성 -> 집계 -> 갱신 (각 노드가 이웃 노드의 잠재 요인을 수집해 은닉 상태 업로드, 그리고 이웃 취향까지 반영한 최종 임베딩을 얻음)      

#### 2-3. LightGCN <br>
= 복잡성을 제거하고 최적화한 GNN의 한 종류        
- GNN의 특징 변환과 비선형 활성화 함수 제거       
- 정규화된 이웃 임베딩의 합만으로 구성      

#### 2-4. GNN 활용 사례 <br>
- 핀터레스트, 송금 네트워크 파악해 금융 사기 및 자금 세탁 방지, 우버 이츠 메뉴 추천      

#### 2-5. 장단점 <br>
(+) 고차원 협업 시그널 포착 가능, 정형화되지 않은 데이터 학습 가능, 콜드 스타트 문제 완화에 효과적                  
(-) 과평탄화 문제, 과압축 문제, 계산 복잡도, 해석의 어려움    

## 3. 트랜스포머 기반 <br>
= 순서와 문맥으로 사용자 이해       
#### 3-1. 기존 추시의 순차적 한계 <br>
- 구매와 클릭 순서를 고려하지 않으며 문맥 부족, 다음 행동을 예측하기보다 일반적 선호도 찾음     
- 사용자 행동 시퀀스를 이해하고 다음 행동을 예측하는 트랜스포머 모델이 대안으로 등장     

#### 3-2. 트랜스포머? <br>
- 시퀀스 내 문맥, 핵심 의도 파악하는 데 강력 (아이템 임베딩에 시퀀스 임베딩이 추가적으로 들어가기 때문)         

#### 3-3. 트랜스포머 기반 주요 추시 모델 <br>

**SASRec** <br>
= 단방향 어텐션으로 과거 정보만 활용해 미래 행동 추측하는 방식으로 훈련 받는 추시 모델     
- 단방향 어텐션 활용하며, 어텐션 계산 시 현재 아이템은 오직 과거 아이템들만 참조 가능     

**BERT4Rec** <br>
= 양방향 어텐션으로 과거 정보, 미래 정보를 활용해 예측하는 방법 훈련한 모델    
- 양방향 어텐션 활용하며, 시퀀스 일부를 마스킹하고 가려진 아이템이 과거와 미래의 모든 아이템을 참조하는 방식    
- 아이템의 양방향 문맥 모두 고려      

## 4. 오늘날의 추시 <br>
#### 4-1. LLM <br>
- 추천을 점수 계산보다는 문장으로 출력해 사용자 상황에 맞게 이유를 덧붙인 서술형 추천 가능      
- 메타 데이터에만 의존하기보다 텍스트 전반을 이해하므로 콜드 스타트 문제 해결 가능     
- 대화형 추천 가능      
<br>
- P5: 모든 추천 과제를 LLM이 풀 수 있도록 프롬프트 문제로 통일시킨 모델로, 컴퓨팅 자원과 소요 시간이 긺    
- TALLRec: 범용 LLM을 추천 작업에 맞게 효율적으로 튜닝하는 프레임워크    

#### 4-2. 인기 편향과 XAI <br>
- 인기 편향: 소수의 인기 아이템을 과도하게 노출하지만 다수의 비인기 아이템이 관심을 받지 못하는 현상     
- XAI: 추천 이유 설명할 수 있는지. 블랙박스 문제를 해결해 고객을 설득하고 추천 근거 제시   

#### 4-3. 컴퓨팅 자본, 콜드 스타트 문제 고려한 추시 도입 <br>
- 비용 문제: 딥러닝 모델은 GPU 필요하므로 연산량 증가하고, 막대한 서버비가 필요해, 선형 기반 추천 시스템과 병행 중     
- 후보군 생성 -> 순위 매기기
    - 후보군 생성: 선형 모델 사용해 사용자가 조금이라도 좋아할 만한 아이템 빠르게 골라내기   
    - 순위 매기기: 1단계에서 추려진 후보군 대상으로 사용자가 가장 좋아할 순서대로 정교하게 순위를 딥러닝 모델로 매김
