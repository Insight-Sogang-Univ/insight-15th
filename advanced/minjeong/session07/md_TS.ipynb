## 시계열 데이터란? <br>

#### 정의 <br>
= 순차적 시간 흐름에 따라 수집된 데이터     
- 연속한 관측치 -> 상관관계 존재 = 시퀀스 데이터     
- 시계열 분석의 목적: 미래 데이터 예측      

(1) 단변량 시계열 = 하나의 숫자가 시간에 따라 변하는 것     
- 시간에 따라 연속 측정된 단일 변수로 이루어진 시계열 데이터     
- 주식 가격, 온도 기록, 매출 데이터 등      

(2) 다변량 시계열 = 여러 숫자가 동시에 시간에 따라 변하는 것    
- 같은 시간에 기록된 둘 이상의 변수로 이루어진 시계열 데이터 -> 변수 간 상호 작용과 영향력 파악       
- 경제 지표, 기후 모델 등    

#### 특징 <br>
(1) 시간의존성 <br>
= 시간에 따라 변화     
- 데이터 순서를 바꾸면 의미가 달라지며, 과거 값이 현재 값에 영향    
- 시점 간 상관관계 고려     

(2) 계절성 & 추세 <br>
- 계절성 = 일정 주기에 따라 반복적으로 나타나는 데이터 변동      
- 추세 = 시간 흐름에 따라 나타나는 장기적 증가/감소 경향       

(3) 자기상관성 <br>
= 이전 시점 값이 이후 시점 값에 영향을 미침 = 자기자신끼리 상관관계를 갖는 특성     
-> 시간 지남에 따라 값들 간 일정한 패턴/반복 존재         
<br>
**자기상관성 때문에 회귀 분석 불가능** <br>
- 오차항 독립성 가정 위반으로 통계 결과 왜곡 발생     
- 허위 상관관계 문제     
- 예측 어려움 & 모델 복잡성 증가     

=> 시계열 데이터는 시계열 전용 모델 사용하거나 사전 처리 후 분석       


#### 분석 요인 <br>
(1) 추세 요인 <br>
= 데이터 전반에 걸쳐 시간에 따라 지속적으로 증가/감소하는 경향 ex. 연도별 매출 상승, 출산률 감소 등     

(2) 계절 요인 <br>
= **고정된 = 일정한 주기**로 반복되는 패턴 ex. 크리스마스 시즌 관련 용품 매출 증가하는 현상      

(3) 순환 요인 <br>
= 계절성보다 길고 **불규칙한 주기**로 반복되는 변동 ex. 경기 호황/불황      

(4) 불규칙 요인 <br>
= 추세, 계절성, 순환성 모두 제거한 뒤 남은 예측 불가능한 변동 ex. 자연재해, 팬데믹        

<br>

=> 데이터 전처리 - 정상성 판단 - 모델링 및 예측

## 시계열 데이터 전처리 <br>
#### 결측치/이상치 탐지 및 처리 <br>
(1) 결측치 <br>
= 특정 시점에서 값이 측정되지 않았거나 누락된 상태 -> 센서 오류, 수집 누락, 전송 지연 등      
- 추세 및 계절성 왜곡 가능하므로 적절한 대치와 보간 필요     
- 결측치 삭제, 단순 대체, 보간, 머신러닝      
    - 보간 = 주어진 데이터 포인트 값을 사용해 두 개 이상의 데이터 포인트 사이에서 누락된 값 추정      

(2) 이상치 <br>
= 값이 측정되었지만 측정값이 일반적이지 않은 경우    
- 측정 오류 or 실제로 특이한 이벤트 -> 사전 지식 필요       
- IQR, z-score, STL 분해      
- CART, 클러스터링 기반 이상 탑지    
- 오토 인코더     
- 삭제, 단순 대치, 보간     

#### EDA = 평활화 <br>
= 시계열 데이터에서 노이즈를 줄이고 전체적인 흐름/추세를 더 명확하게 파악하기 위해 데이터를 부드럽게 만드는 기술 <br>

**이동 평균 = MA** <br>
= 시계열의 부분집합으로 평균을 구해 평활화 효과를 얻는 방법 = 윈도우 (평균을 구할 구간의 크기) 내 평균 구하는 방법       
- 사용 목적: 노이즈 제거, 추세 파악, 예측의 기초 자료     
- 유형      
    - SMA (단순 이동평균) = 기간 내 모든 값의 단순 평균 -> 안정적 시계열에서 사용  
    - WMA (가중 이동평균) = 최근 값에 더 큰 가중치 부여 -> 최근 데이터가 더 중요한 경우     
    - EMA (지수 이동평균) = 최근 데이터에 지수적으로 더 많은 가중치 부여 -> 최근 값에 매우 빠르게 반응해야 하는 경우       

#### 요소 분해 = 시계열 분해 <br>
= 주어진 시계열 데이터를 개별 성분으로 분해하는 과정     
- 다양한 변동이 결합된 시계열 데이터에서 각 요소를 분리해 각 패턴을 명확히 이해하면 이후 예측이나 이상 탐지 작업 정교히 수행 가능     

<br>

**분해 모형 (1) 가법 모형** <br>
시계열 = 추세 + 계절성 + 불규칙성    
= 시계열 관측값은 변동 요인들의 가법으로 설명될 수 있다고 가정하며, 모든 요소의 영향이 일정함   
- 시계열 전체에서 계절성 진폭이 일정한 경우 적합        

**분해 모형 (2) 승법 모형** <br>
시계열 = 추세 * 계절성 * 불규칙성    
= 시계열 관측값은 변동 요인들의 승법으로 설명될 수 있다고 가정하며, 추세가 커질수록 계절성/불규칙성의 효과도 비례해서 커짐    
- 진폭이 시간에 따라 달라지는 경우 적합     

**분해 방법 (1) seasonal decompose - MA 기반** <br>
= 이동 평균으로 추세 추출한 후 잔여 성분에서 계절성 분리하는 방식     

**분해 방법 (2) STL - LOESS 기반** <br>
= 이동 평균 대신 LOESS 사용하는 방법으로, 추세와 계절성이 시간에 따라 변화하는 경우에도 유연하게 포착 가능하고 이상치에 강건한 분해 가능     
- Inner Loop, Outer Loop 이중 반복 구조로 구성       

## 정상성 판단 <br>
#### 정상성이란 <br>
= 시계열 데이터의 평균, 분산 등 통계적 특성이 시간에 따라 일정한 수준으로 유지되는 것 = 시간에 상관 없이 일정한 성질    
- 정상 시계열: 정상성 O / 비정상 시계열: 정상성 X      
- 정상 시계열은 예측이 가능한 것과 달리, 비정상 시계열은 예측이 어려워 수학적 처리를 거쳐 정상성 띠도록 변환      

#### 강한/약한 정상성 <br>
- 강한 정상성    
    - 시간을 언제로 잡든 데이터 확률 분포 자체가 똑같은 상태   
    - 평균, 분산, 왜도, 첨도 등 모든 통계적 특성이 일정     
    - 현실 데이터에서는 존재하기 어려움     
- 약한 정상성
    - 평균, 분산 일정    
    - 자기공분산이 시차에만 의존: 데이터 간 관계가 특정 시점이 아닌 시간 간격에 의해서만 결정    
    => 우리에게 필요한 데이터     

#### 정상성 판단 <br>
- 시각적: 평균을 중심으로 일정한 폭으로 진동하는 형태     
    - 추세가 있는 경우, 시간에 따라 데이터가 지속적으로 상승/하락    
    - 계절성 있는 경우, 특정 주기로 패턴이 반복됨      

- 통계: 그래프만으로는 시각적으로 정상성 여부를 판단하기 어려워 수치적 가설 검증 필요    
    - 단위근이 있는지 확인하는 과정: 과거 충격이 영구적으로 누적되는 특성으로, 대표적인 비정상성의 원인    

**ADF, KPSS** <br>
| ADF | KPSS |    
| ----- | ----- |     
| 데이터의 비정상성 가정 | 데이터의 정상성 가정 |          
| $H_0$: 단위근 존재 = 비정상 시계열 | $H_0$: 단위근 X = 정상 시계열 |          
| $H_0$ 기각으로 정상성 판단 |$H_0$ 채택으로 정상성 판단|          
| 추세를 단위으로 착각해 비정상으로 오판 가능 | 추세 때문에 평균 변하므로 비정상으로 판단 |          

#### 정상성 확보를 위한 방법 <br>
(1) 추세 제거     
- 시계열이 지속적으로 증가하거나 감소할 때, 연이은 관측 값들의 차이를 계산 (차분) 해 평균의 정상성 확보     

(2) 계절성 제거    
- 시계열이 주기적 패턴을 가질 때, 계절 차분 ($y_t-y_{t-m}$) 해 완전 랜덤한 데이터를 뽑아냄

(3) 분산 안정화      
- 시계열의 변동 폭이 점점 커지거나 작아질 때, 로그 변환을 취해 분산 정상성 확보       
- 음수 / 0 처리 불가능하며, 분산 변화가 비선형적이거나 복잡한 패턴을 가지는 경우 충분하지 않을 수 있음    

## 단변량 시계열 분석 <br>
#### 안정적 시계열 분석 <br>
(0) AR, MA 모델의 전제 조건    
- 정상성, 가역성 (MA 모델을 AR 형태로 재표현할 수 있는지 여부-> 특성 방정식이 0, 모든 해의 절댓값 > 1)

(1) 자기회귀 모델 = AR       
= 과거 자신의 값을 독립변수로 삼아 현재 값 예측하는 회귀 -> 시계열 데이터의 자기상관성 활용         
- $AR(p)$ : p차 자기 회귀 모형으로, 과거 p개 시점의 데이터의 선형 조합을 이용해 예측하는 모델     

(2) 이동평균 모델 = MA    
= 과거 시점의 잔차를 독립 변수로 삼아 현재 값 예측하는 모델 -> 과거의 예측 오차를 현재 시점에 반영     
- 평활화와 다른 개념     
- $MA(q)$ : 1차 이동 평균 모형으로, 과거 q개의 예측 오차의 선형 결합으로 예측하는 모델      

(3) ARMA 모형    
= $AR(p)$ 모형과 $MA(q)$ 모형 결합 = 과거 p개의 관측값과 과거 q개의 오차 활용해 예측    

(4) 자기 상관 함수 - 차수 결정 방법    
- 두 그래프로 현재 값과 과거 값의 관계 파악하여 적절한 차수 p, q 결정.      
- 자기 상관 계수 (ACF): ARMA(p,q) 모형에서 사용할 q개의 오차 결정법 -> 두 시점 사이 상관 관계 파악     
- 부분 자기 상관 계수 (PACF): ARMA(p,q) 모형에서 사용할 p개의 시점 결정법 -> 수지섬 사이의 직접적 영향만 파악       

#### 불안정 시계열 분석 <br>
- 비정상 시계열 데이터의 정상성 확보 -> 정상 시계열이 되면 ARMA로 안정적 시계열 분석      

(1) ARIMA      
= ARMA(p,q) + 차분      
- p = 0 : IMA(d,q) -> d번 차분하면 MA(q)     
- d = 0 : MRMA(p,q) -> 정상성 만족         
- q = 0 : ARI(p,d) -> d번 차분하면 AR(p)       

(2) SARIMA     
= 시계열 데이터의 계절성을 처리할 수 있는 모델 = ARIMA 모델에 계절성 변동을 반영한 모델      

## 다변량 시계열 분석 - VAR <br>
= 자신의 과거 값과 다른 변수들의 과거 값을 함께 사용하며 미래 예측하는 모델       
- 변수들이 서로 영향을 주고 받거나 상호 의존성이 중요한 경우 적합     
- 변수가 너무 많거나 관계가 명확하지 않고, 장기 예측 필요한 경우 적합 X       

(1) 충격 반응 함수     
= 한 변수의 충격이 다른 변수에 미치는 영향을 시간 흐름에 따라 분석      

(2) 예측오차 분산 분해      
= 예측 오차에 각 변수가 얼마나 기여하는지 분석       

## 딥러닝 활용한 시계열 분석 <br>
#### 등장 배경 <br>
- 데이터가 많으면 처리 속도가 느리거나 크기가 고정된 데이터만 가능하거나 비선형 패턴 포착이 어려운 전통 통계 모델의 한계       

#### RNN <br>
- 순차적 패턴이 중요하거나 (과거 정보가 미래 예측에 영향) 시계열 길이와 목적에 따라 모델 선택 가능한 경우 적합      

#### 트랜스포머 계열 <br>
- 인포머: 불필요한 연산 줄여 계산 복잡도를 대폭 낮춘 모델 -> 매우 긴 시계열 데이터 처리에 강력           
- 오토포머: 시계열 전통 기법인 요소 분해 개념을 DL 구조에 결합한 모델 -> 주기 패턴이 뚜렷한 경우 예측 성능 높음   
- TFT: 과거 데이터와 외부 변수를 섞어 예측 가능한 다목적 아키텍처 -> 예측에 가장 중요했던 변수 해석 가능           

#### 파운데이션 모델 <br>
= 거대한 데이터로 사전 학습해 다양한 문제를 해결할 수 있는 범용 모델     
- 사전 학습 후 파인 튜닝으로 특화       
- LLM 기반 모델, 시계열 전용 파운데이션 모델 등
