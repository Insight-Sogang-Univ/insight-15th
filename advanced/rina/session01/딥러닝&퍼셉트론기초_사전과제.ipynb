# 1. 인공신경망
- 개념: 우리 뇌의 신경망을 모방한 모델. 뉴런의 정보 전달 방식과 퍼셉트론의 연결 방식은 흡사함.
- 과정: 
1. 입력을 받아 각각 가중치와 편향을 적용.
2. 종합한 값을 활성화 함수에 넘김. 이때 임계값을 기준으로 미달이면 0에 가깝게 내보내고, 기준 통과면 유의미한 값으로 내보냄.
3. output 출력

# 2. 퍼셉트론
- 개념: 초기 형태의 인공신경망으로, 신경망(딥러닝)의 기원이 되는 선형 이진 분류기임.
- 과정: 
1. 입력값과 가중치를 곱한다.
2. 곱한 값들의 총합을 구한다.
3. 총합이 임계값을 넘으면 1, 넘지 않으면 0을 출력함.
인공신경망과 퍼셉트론의 차이?
- 퍼셉트론은 인공신경망의 가장 기초적인 초기 형태(단층)이고, 인공신경망은 이러한 퍼셉트론을 다층으로 쌓고 활성화 함수를 사용하여 복잡한 비선형 문제를 해결하는 확장된 개념.


** 용어 정리 **
1. 가중치: 입력값이 출력값에 주는 영향력(중요도)을 결정
2. 활성화 함수: 신호 활성화 결정을 담당하는 부분으로, 퍼셉트론의 성질은 활성화 함수로 **계단 함수**를 사용.
	- Step 외에도 sigmoid, Tanh, ReLU 등 다양한 활성화 함수 존재.
3. 임계값: 활성화 함수에 사용되는 값으로, 계단 함수에서는 가중합이 임계값을 넘으면 1, 넘지 못하면 0을 출력.
4. 편향: 위의 계단 함수를 구현한 식에서, 임계값을 좌변으로 넘기면 편향(b)으로 표현 가능함.
	- 임계값과 편향은 표현의 차이라고 보면 됨.


# 3. 단층 퍼셉트론
개념: 단순히 입력층과 출력층으로만 이루어진 퍼셉트론

## 3.1 논리 게이트
- 개념: 디지털 회로에서 기본적인 논리 연산을 수행하는 데 사용되는 구성요소.
- 게이트 연산에 쓰이는 것은 n개의 입력값과 하나의 출력값
- AND, OR, NOT, NAND, XOR 등 다양한 종류가 있으며, 이들을 조합해 복잡한 논리 회로를 구성할 수 있음.
- **단층 퍼셉트론을 이용하면 AND, NAND, OR 논리 게이트를 쉽게 구현할 수 있음.**

### 3.1.1 대표적인 논리 게이트
- 이때 앞서 편향 식에서 본 것과 같이 계단 함수에 의해 계산 결과가 양수면 출력값이 1, 음수면 0으로 봄.
1. AND 게이트
- 특징: 입력값이 모두 1인 경우에만 출력값이 1이 나오는 구조.

2. OR 게이트
- 특징: 입력값이 모두 0인 경우에 출력값이 0이고, 나머지 경우에는 모두 출력값이 1인 구조.

3. NAND 게이트
- 특징: 입력값이 모두 1인 경우에만 출력값이 0, 나머지 경우에는 모두 출력값이 1인 구조.

4. XOR 게이트
- 특징: 입력값 두 개가 서로 다른 값을 가지고 있을 때에만 출력값이 1인 구조.



## 3.2 단층 퍼셉트론의 한계
- **XOR 게이트를 구현할 수 없다**
- 이유: 단층 퍼셉트론은 직선 하나로 두 영역을 나눌 수 있는 선형 분류 문제에 대해서만 구현 가능. 
- 그러나 XOR 게이트는 직선이 아닌 곡선/비선형 영역으로 분리해야만 구현 가능

그래서 등장한 것이 **다층 퍼셉트론**


# 4. 다층 퍼셉트론
- XOR 게이트 문제: 퍼셉트론의 층을 더 쌓아서 해결 가능!
- 중간에 은닉층을 추가하여 XOR 문제보다 더 복잡한 문제 해결 가능

# 5. 딥러닝 모델 학습 방법

## 5.1 순전파와 역전파
- 학습 알고리즘의 기본 구조
1. 순전파: 인공신경망에서 입력 데이터가 입력층, 은닉층을 거쳐 출력층까지 순서대로 계산되며 예측값 도출.
2. 오차 계산을 통해 예측 정확도 측정
3. 역전파를 통해 출력층에서 다시 입력층 방향으로 오차 전달
4. 가중치 업데이트
5. 1~4 반복
- 이러한 순전파/역전파 과정은 **연쇄 법칙**을 통해 일어남.

## 5.2 연쇄 법칙
- 개념: 합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다는 법칙

- 손실함수(실제값과 예측값 사이의 오차를 수치화 하는 함수)에 대한 미분임

- 목적: 오차를 줄이기 위해서 각 가중치가 최종 손실에 얼마나 기여했는지 계산해야 하는데, 연쇄법칙으로 각 단계를 거슬러 올라갈 수 있음.

- 원리: 각 은닉층마다 손실에 주는 영향을 곱해서 최종 손실에 대한 값을 나타냄.
	- EX) A->B->C 구조에서 A가 C에 미친 영향은 B->C의 영향*A->B의 영향으로 분해해서 구할 수 있음.

- 미분값은 오차를 줄이기 위해 가중치를 어느 방향으로, 얼마나 바꿔야 하는지 알려주는 나침반임. 
	- EX) 오차가 크면, 미분값도 커짐->많이 틀렸으니 가중치를 크게 고쳐라

## 5.3 기울기 소실 문제
- 은닉층이 많고, 각 단계마다 미분값이 1보다 작으면 곱할 수록 기울기가 0이 됨.

- 그 결과 앞쪽 층의 가중치가 거의 업데이트되지 않아서 학습이 제대로 이뤄지지 않는 문제 발생.
	- 맨 앞 층의 가중치를 수정하려면 수많은 미분값을 곱해야 하는데 각 층 미분값이 1보다 작으면 미분값이 낮게 나오니까...


# 6. 손실함수
- 개념: 실제값(정답)과 예측값(출력값)의 차이를 계산하는데 사용되는 함수

## 6.1 손실함수의 종류
1. 평균 제곱 오차(MCE)
- 예측값이 연속적이고, 예측값과 실제값 간의 차이를 직접적으로 측정하는 회귀 문제에 적합.
2. 이진 크로스 엔트로피(BCE)
- 이진 분류 문제(로지스틱 회귀)에서 사용하는 손실함수
- MSE와 달리, BCE는 **얼마나 확신을 가지고 예측했는지**도 loss에 반영함.
	- 낮은 확률로 예측해서 맞추거나 높은 확률로 예측해서 틀리는 경우 loss가 더 큼!
- 실제 정답이 1일 때
	- 예측확률이 1에 가까울수록→ loss값은 0에 가까워짐 = 잘 맞힘
	- 예측 확률이 0에 가까울수록 → loss가 급격히 증가 (확신했지만 틀림)
- 실제 정답이 0일 때
	- 예측 확률이 0에 가까울수록 → loss값은 0에 가까워짐 = 잘 맞힘
	- 예측 확률이 1에 가까울수록 →  loss가 급격히 증가 (확신했지만 틀림)

3. 크로스 엔트로피 오차
- 다중 클래스 분류 문제에서 예측 확률이 실제 클래스와 얼마나 일치하는 지를 평가하는 데 적합함.
- 진짜 분포/ 예측 분포의 유사도를 측정하는데 유용.
- 이진 크로스 엔트로피와 유사한 형태이지만, 다중 분류에서 사용



# 7. 활성화 함수
- 개념: 각 층의 출력값에 비선형성을 추가해서 신경망이 복잡한 패턴도 학습할 수 있게 해주는 함수

|        | 단층 퍼셉트론       | 다층 퍼셉트              |
| ------ | ------------- | ------------------- |
| 활성화 함수 | 계단함수          | sigmoid,ReLU 등      |
| 출력값    | 0과 1(이진 출력)   | 연속적인 (확률 또는 실수)     |
| 특징     | 선형 문제만 풀 수 있음 | 복잡한 비선형적인 문제도 학습 가능 |

## 7.1 계단함수
- 개념: 임계값과 가중치의 곱을 모두 더한 뒤, 그 합이 임계값을 넘으면 1 또는 0을 출력하는 함수
- 임계값은 뉴런이 얼마나 쉽게 활성화 되는지를 제어, 가중치는 각 입력 신호가 결과에 미치는 영향력을 제어

## 7.2 시그모이드 함수
- 개념: 입력 신호를 0과 1 사이의 **연속적인 값**으로 변환하는 비선형 하뭇
- 계단함수와의 차이점: 출력값이 연속적이다!

## 7.3 ReLU 함수
- 입력이 0보다 크면 그대로, 0이하면 0을 출력하는 비선형 함수
**ReLU 함수를 가장 많이 사용하는 이유?**
1. 계산이 간단하고 효율적
2. 비선형성 제공
3. 기울기 소실 문제 완화->ReLU는 양수 구간에서 미분값이 항상 1이므로, 역전파가 안정적으로 전달됨.

## 7.4 Tanh(쌍곡탄젠트) 함수
- 개념: 시그모이드 함수의 변형으로, 출력값의 범위가 -1~1
- 장점: 시그모이드와 달리 출력값의 중심이 0이므로, 입력값의 중심을 0으로 맞추고 싶을 때 사용. 
	- 중심이 0이면 학습 과정에서 가중치 업데이트가 더 안정적으로 이루어짐.
- 단점: 기울기 소실 문제 발생

## 7.5 softmax 함수
- 개념: 각 클래스에 대한 출력값을 확률로 변환하는 함수
- 다중 클래스 분류 문제의 출력층에서 사용 가능

# 8. 경사하강법
- 개념: 오차를 가장 작게 만드는 가중치와 편향을 찾는 방법
- 딥러닝 학습 데이터가 방대하기 때문에 이 데이터를 어떻게 쪼개고 몇 번 반복해서 학습할 지 결정하기 위해 **배치, 에포크, 이터레이션** 필요
- 배치: 학습 과정에서 가중치와 같은 매개변수를 조정하기 위해 사용하는 데이터 묶음. 
	- 사용자가 임의로 정할 수 있는 하이퍼파라미터
	- 배치 크기가 크면 학습이 느려지고, 메모리가 부족해질 수 있음
	- 배치 크기가 작으면, 가중치가 자주 업데이트되어 학습이 불안정해질 수 있음
- 에포크: 모든 데이터셋을 학습하는 횟수
- 이터레이션: 한 번의 에포크를 끝내기 위해서 필요한 배치의 숩
	- 1 Epoch에서 다루는 데이터의 수= **배치의 크기*이터레이션(step)**

## 8.1 배치 경사하강법
- 특징: 
1. 오차를 구할 때 전체 데이터를 한 묶음으로 묶음
2. 전체 데이터를 통해 학습시키기 때문에, 업데이트 횟수가 적음 (1 Epoch 당 1 업데이트)
3. 전체 데이터를 모두 한 번에 처리하므로 메모리가 많이 필요
4. 항상 같은 데이터 (전체)에 대해서 경사를 구하기 때문에 수렴이 안정적

## 8.2 배치 크기가 1인 확률적 경사하강법
- 기존의 긴 계산시간을 보완하고자 등장. **랜덤으로 선택된 하나의 데이터에 대해서만 계산**
- 특징: 자원이 적은 컴퓨터에서도 쉽게 사용 가능
- 매개변수의 변경폭이 불안정하고, 정확도가 낮을 수 있음

## 8.3 미니 배치 경사하강법
- 앞서 두 경사하강법의 문제를 보완하기 위해 등장
- 배치 크기를 적절히 지정하여 해당 데이터 개수만큼에 대해 매개 변수 값을 조정
- 전체 데이터를 계산하는 것보다는 빠르고, 배치 크기가 1인 경사 하강법보다 안정적


# 9. 옵티마이저
- 경사하강법에서 최소값을 더 효율적이고 안정적으로 찾기 위해 사용
- 종류에는 모멘텀, 알엠에스프롭, 아담 등 많은 방법 존재

## 9.1 모멘텀(Momentum)
- 변수가 **가던 방향**으로 계속 가도록 **속도를 추가**한다
- 즉, **경사하강법+관성**
- 경사하강법에서 계산된 접선의 기울기에 바로 한 시점 전의 접선의 기울기값을 일정 비율만큼 반영
<br>-> 따라서 local minima에 빠지더라도 관성의 힘으로 넘어올 수 있음.

## 9.2 알엠에스프롭(RMSProp)
- 기존 경사하강법에서 기울기가 큰 방향에서는 너무 많이, 작은 방향에서는 너무 작게 이동하는 문제를 보완
- 각 학습률을 조정하여 안전하고 효율적으로 학습하게 함
- 최근 기울기를 참고하여 **급한 경사에서는 속도를 줄이고, 완만한 경사에서는 속도를 유지 or 증가**

## 9.3 아담(Adam)
- RMSProp과 Momentum을 합친, **방향과 스텝사이즈를 적절하게 합친 방법**-> 방향과 학습률 두 가지 모두 잡기 위한 방법
- 각 파라미터의 1차 모멘트(기울기의 평균)와 2차 모멘트(기울기의 제곱 평균)을 사용하여 학습률을 조정
