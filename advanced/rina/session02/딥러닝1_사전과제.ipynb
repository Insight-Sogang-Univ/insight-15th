# 1. CNN이란?
- 개념: 이미지 공간정보를 유지한 상태로 학습이 가능한 딥러닝 알고리즘
- *이미지 처리에 MLP가 아니라 CNN을 사용하는 이유?
	- CNN은 이미지의 2차원 구조를 유지한 채 주변 픽셀 관계와 시각적 특징을 학습->이미지 구조 유지
	- MLP는 위치, 크기, 회전 등의 변화에 취약한 것과 달리 CNN은 덜 민감
	- MLP의 경우 완전 연결 구조로 인해 파라미터가 급증하지만, CNN은 부분연결 구조&Convolution 연산으로 이미지 특정 피쳐에만 집중
	- MLP는 차원 flatten으로 공간 정보가 손실되지만, CNN에서는 특징을 추출한 후 flatten하기 때문에 손실 정도가 낮다

# 2. CNN 모델 및 구조
## 2.1 CNN 모델
- 이미지의 **위치 별 특징을 추출**하는 부분과 **클래스를 분류하는 부분**으로 나눌 수 있음.
- 특징 추출(Feature learning)영역: Convolution Layer와 Pooling Layer를 여러 겹 쌓는 형태로 구성
	- Convolution Layer: 입력 데이터에 필터를 적용 후 활성화 함수를 반영하는 필수 요소
	- Pooling Layer: Convolution Layer 다음에 위치하며, 이미지 공간 크기를 줄여주는 중요한 특징을 강조하는 역할
- 이미지 분류를 하는 영역: CNN모델의 마지막에 위치하고 Fully Connected Layer로 구성
- +) 특징 추출 영역과 이미지를 분류하는 영역 사이에 3차원을 1차원 배열 형태로 만드는 **Flatten** 위치함

## 2.2 CNN구조
- Convolution과 Pooling을 이용해 특징 추출 진행
- 연산 과정에서 출력(feature map)의 크기와 정보 보존 정도를 조절하기 위한 하이퍼파라미터로 padding, stride가 활용
- Convolution 연산: 
	- 1. 필터(커널)이 이미지 위를 슬라이딩
	- 필요시 padding 적용. 
	- Stride 값을 지정하여 이미지를 몇 칸씩 점프할건지 결정
	- 2. 각 위치에서 요소별 곱셈 후 합계 계산
	- 3. 결과를 Feature Map에 저장
- Pooling 연산: 피처맵에서 특징을 추출하기 위해 이미지를 격자 형태로 분할한 뒤, 해당 영역 안에서 가장 큰 값 또는 평균값을 취하는 것

# 3. 이미지 딥러닝 응용
## 3.1 이미지 분류
- 이미지를 수치 데이터(픽셀 값)로 인식-> 사전 정의된 클래스 중 하나로 분류

## 3.2 객체 탐지
- 한 이미지에서 객체와 이 객체를 둘러싸는 가장 작은 직사각형으로 정의되는 bounding box를 찾는 작업
- 모델 적용 방식: CNN을 통해 Feature Map을 찾고, 그 위에 Classificaition Head, Regression Head를 올림

## 3.3 이미지 캡셔닝
- 이미지 내에 있는 객체에 대한 판단뿐만 아니라 객체들 간의 관계를 파악하고 자연어 형태로 알맞게 표현
- 모델 적용 방식: 이미지 특징을 추출하기 위한 CNN과 문장으로 나타내기 위한 RNN
- 사용 단계: 
	- 1. CNN을 이용한 특징 추출
	- 2. Attention Network를 이용하여 단어 생성
	- 3. RNN을 이용하여 문장 완성

# 4. RNN
RNN에 대해 알아보기 전, 순차 데이터에 대해 알고 가야함.

## 4.1 순차 데이터란?
- 개념: 순서에 의미가 있으며, 순서가 달라질 경우 의미가 손상되는 데이터
- 예시: 음서 및 오디오 데이터, 자연어, 생물학적 서열 데이터, 비디오 데이터
- 특징: 데이터를 구성하는 요소들의 순서 자체가 핵심 정보, 이전 시점의 데이터가 다음 시점의 데이터에 영향을 미침
- 문제: 기존의 머신러닝 기법으로는 순차 데이터의 특징을 고려하기 힘들다

		그래서 등장한 것이 RNN!

## 4.2 RNN의 구조와 원리
- RNN 개념: 순환하는 구조를 가진, 순차 데이터 처리에 특화된 인공 신경망
- 특징
	- 시퀀스 데이터를 입력받아 순서 정볼르 유지하며 처리한다
	- 순한하는 은닉층이 매 시점의 은닉 상태를 업데이트 한다
	- 이전 시점의 값을 현재 시점으로 넘겨준다
- 구성요소: 시간 t에서의 입력, 시간 t에서의 은닉상태(이 시점의 메모리), 시간 t에서의 출력, 입력 가중치, 순환 가중치, 출력 가중치
- 정보 흐름: 
	- t-1 시점: 입력이 들어가서 은닉상태를 업데이트 하고 출력을 내보냄
	- t 시점: t-1 시점에서 계산된 은닉상태를 다음 시점으로 그대로 전달
	- **새로운 입력**과 **전달받은 과거 정보**를 함께 사용하여 **현재 시점의 은닉상태**를 업데이트하고 **출력**을 계산
- 핵심원리: 동일한 가중치를 시퀀스의 다양한 시점에서 반복적으로 적용
- 장점: 학습 파라미터 수 감소, 일반화 능력 향상

## 4.3 RNN의 한계 및 장기 의존성 문제
- 문제: 기울기 소실, 기울기가 무한대로 발산, 느린 훈련 시간
- 장기 의존성 문제: 시퀀스 앞 부분의 중요한 정보를 잊어버려 맥락 파악 능력이 급격히 저하되는 현상
<br> -> **장기 기억력에 문제 발생**

		이를 해결하기 위해 LSTM, GRU 등장!

# 4.4 LSTM
- 개념: 앞의 정보가 뒤로 충분히 전달되지 못하는 RNN의 **기울기 소실 문제를 해결**하기 위한 모델 구조
- 원리: 장기기억과 단기기억 및 출력을 따로 분리해서 관리
- 특징
	- **기억할 내용과 잊어버릴 내용을 선택**해서 중요한 정보들을 오래 가져간다.
	- RNN과 달리 (cell state: 장기 기억)와 ht(hidden state: 단기기억)의 2개의 순환되는 층을 사용한다.
	- Gate를 통해 필요한 정보만 통과한다.
		- Forget gate, Input gate, Ouput gate
	- Final memory cell: 현재 정보를 얼마나 기억할지 계산

- 장점: 기울기 소실 문제 효과적으로 완화
- 단점: 복잡한 구조, RNN보다 학습 파라미터가 많아짐

## 4.5 GRU
- 개념: 메모리 셀 없이 게이트 수를 줄여 구조를 간소화
- LSTM과 달리 **기억과 행동을 하나로 통합**하여 관리
- Update Gate: 과거vs현재 비율 조절->"과거를 얼마나 유지하고, 새 정보를 얼마나 반영할까?"
- Reset Gate: 과거를 무시할지 결정->강한 새로운 입력이 오면 과거 감정 무시
- 장점: 성능과 효율성 사이의 균형이 좋다
- 단점: 긴 시퀀스 처리에서의 한계는 여전히 존재

# 5. Seq2Seq
- 개념: 입력된 시퀀스(문장, 시계열 데이터 등)를 다른 도메인의 시퀀스로 변환하는 모델
	- 대표적으로 기계 번역에서 사용됨
- 특징: 
	- 아이템(단어)의 입력 개수와 출력개수가 같지 않아도 된다
	- 인코더와 디코더라는 모듈을 갖고 있어, Encoder-Decoder 모델이라고도 불린다
- 구조: **인코더(Encoder)** 와 **디코더(Decoder)**
- 인코더: 입력 시퀀스를 받아서 하나의 고정된 표현으로 변환
	- 출력: 컨텍스트 벡터
- 디코더: 컨텍스트 벡터를 받아서 출력 시퀀스를 생성
	- **입력을 그대로 따라하는게 아니라 새로운 결과를 만들어냄**
- 한계: 고정된 Context Vector에 소스 문장 정보 압축으로 **병목 현상** 발생
