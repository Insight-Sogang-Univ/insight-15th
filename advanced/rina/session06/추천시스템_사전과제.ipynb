# 1. 추천 시스템
- 개념: 사용자의 정보 데이터를 분석, 개인의 취향에 맞는 아이템을 추천하는 알고리즘

## 1.1. 추천시스템 종류
- **내용 기반 필터링**: 콘텐츠를 **유사도**를 기반으로 사용자에게 추천하는 방법
	- 사용자가 이용하거나 선택한 아이템과 유사한 아이템을 추천하는 방식

- **협업 필터링** : 서비스 내에 있는 **다수의 사용자로부터 얻은 선호도 정보**를 기반으로 사용자에게 추천하는 방법
<br>-> 두 명의 사용자가 비슷한 관심사를 가지고 있다면, 한 사용자의 데이터를 바탕으로 다른 사용자에게 추천해 주는 방식

	1. 사용자 기반(User-Based Filtering): 특정 사용자와 **유사한 취향의 사용자**가 선호한 아이템 추천
	2. 아이템 기반(Item-Based Filtering): 사용자가 선호한 아이템과 **유사한 아이템**을 추천

- **하이브리드 필터링** : 2가지 이상의 다양한 종류의 **추천시스템 알고리즘을 조합**하여 만들어진 알고리즘

# 2. 내용 기반 필터링
## 2.1 사용자 프로필(사용자 취향) 
1. 직접 지정형
- : 사용자에게 자신이 선호하는 아이템의 특징을 명시적으로 지정하도록 하는 방법
	- ex) 에이블리 가입하면 패션 선호도 체크

2. 간접 지정형
- : 사용자의 과거 행동 이력에 기반하여 사용자의 프로필을 작성하는 방법
	- ex) SF영화를 즐겨본 이력이 존재하는 유저에게 인터스텔라를 추천해주는 경우

- 사용자의 프로필을 획득하는 방식
	1. 특성 가중치는 사용자가 가지고 있는 아이템 특성 가중치의 평균 값을 활용한다
	- 예시) 유저가 시청한 영화

	| **영화** | **액션** | **로맨스** | **SF** | **공포** |
	| ------ | ------ | ------- | ------ | ------ |
	| 어벤져스   | 0.9    | 0.1     | 0.5    | 0.0    |
	| 인터스텔라  | 0.3    | 0.2     | 0.9    | 0.0    |
	| 노트북    | 0.0    | 0.9     | 0.0    | 0.1    |

	- **사용자 프로필 = 본 영화들의 특성 벡터 평균**
	- 새로운 영화가 들어왔을 때, 그 영화의 특성 벡터와 **유저 프로필 벡터의 유사도(코사인 유사도 등)**을 비교해서 추천 여부 결정

	2. 사용자의 기호를 따로 파악하거나, 설문조사 또는 자발적인 키워드 입력 등으로 프로필을 획득한다
	3. 아이템에 대한 유저의 기존 평가를 바탕으로 분류문제를 해결하는 것과 같다

## 2.2 유사도
- 추천 시스템을 이해하는데 중요한 수학적 개념
1. 영상, 이미지, 텍스트 등의 형태의 아이템 정보 데이터를 **벡터화**한 이후, **아이템 벡터 간의 유사도**를 측정
2. 측정된 유사도를 바탕으로 **고객이 선호하는 아이템과 비교**하여 **유사도가 높은 아이템을 선별**

- 유사도를 이용해 추천 시스템을 만드는 과정
	1. 인풋 데이터 만들기
		- 아이템 정보 활용: 영화의 장르, 제작, 국가, 감독 등이 feature(특징)
		- 테이블 형태로 데이터를 정리 -> 행: 영화 제목, 열: 아이템의 feature, 셀: feature 값 입력
		- 원-핫 인코딩으로 텍스트를 숫자로 변환

	2. 유저 정보 활용
		- 시청 기록을 통한 간접적 정보나 직접 입력한 프로필 정보로 유저의 취향 파악
		- 유저를 n개의 feature로 표현 -> 유저와 아이템을 동일한 기준의 feature로 표현하여 벡터 계산 가능하게 함
	
	3. 벡터의 내적과 유사성
		- 벡터 내적: 두 벡터의 동일 위치 값들끼리 곱한 후 모두 더한 값
		- **내적 값이 클수록 두 벡터의 유사성이 높음**
		- 만약 A, B의 내적값이 A,C의 내적값보다 크다면 B 영화가 사용자의 선택을 받을 가능성이 높을 것!

## 2.3 TF-IDF
- TF(Term Frequency): 특정 문서 내에 특정 단어가 얼마나 자주 등장하는 지를 의미하는 **단어 빈도**
- IDF(Inverse Document Frequency): 전체 문서에서 특정 단어가 얼마나 자주 등장하는지를 의미하는 **역문서 빈도**

- TF-IDF를 이용해서 **다른 문서에서는 등장하지 않지만 특정 문서에서만 자주 등장하는 단어**를 찾아<br> **문서 내 단어의 가중치**를 계산할 수 있음!

- 예시: "사랑"이라는 단어는 모든 로맨틱 코미디 리뷰에서 자주 나오지만, "집착광공"이라는 단어는 특정 리뷰에만 나올 수 있음. 
<br> 그럴 때 "집착광공"의 **IDF 값**이 높아짐!

- TF-IDF는 **각 단어가 그 문서에서 "얼마나 중요한지"를 나타내는 값**
	- TF-IDF 값이 높을 수록 그 단어가 다른 문서들보다 **해당 문서에서 더 중요한 역할**을 함을 의미

- TF-IDF 계산 
	- TF(d,t): 특정 문서 d에서의 특정 단어 t의 등장 횟수
	- DF(t): 특정 단어 t가 등장한 문서의 수
	- $$idf(d,t) = log(\frac{n}{1+df(t)})$$
	- $$TF-IDF = TF(d,t) \times IDF(d,t)$$

## 2.4 내용 기반 필터링의 장단점
- 장점: 아이템이 가지고 있는 **feature 자체에만 집중**하기 때문에 다른 유저의 평점 등에 영향을 받지 않음<br>
→ **콜드스타트 문제**에서 자유로움
- item의 feature를 통해서 추천을 하기 때문에 추천이 된 이유를 설명하기 용이함

- 단점: item의 feature를 추출해야 하고 이를 통해서 추천하기 때문에 **정확한 feature를 추출하지 못하면 정확도가 떨어짐**
- 기존의 item과 유사한 item 위주로만 추천하기에 **새로운 장르의 item을 추천하기 어려움**

# 3. 협업 필터링(Collaborative Filtering)
- 가정: 과거에 비슷한 선호도를 가진 사용자들은 앞으로도 비슷한 선호도를 가질 가능성이 높다

## 3.1 메모리 기반 협업 필터링(이웃 기반 협업 필터링)
- 사용자 또는 아이템 간 유사도를 기반으로 함
- <u>사용자 A와 비슷한 취향을 가진 사용자가</u> 선호하는 아이템을 추천
- 접근법: 1. 사용자 기반 2. 아이템 기반

- 사용자 기반: "당신과 비슷한 고객이 이 상품을 구매했습니다."
	- 유저 간의 유사도가 높을 수록 가중치를 부여 
	- target user와 **유사한 사용자들의 평점**을 활용
	- 같은 그룹의 다른 유저가 선호하는 아이템을 추천함

- 아이템 기반: "이 상품을 선택한 다른 고객들은 다음 상품도 구매했습니다."
	- 추천하고자 하는 target item과 유사한 아이템의 평점을 활용

- 사용자 기반 접근법의 실제 활용: 아이템이 자주 바뀌고 사용자 수가 적은 환경 → 뉴스, 소셜 미디어, 음악 초기 모델에서 주로 활용됨.
- 아이템 기반 접근법의 실제 활용: 사용자 유입이 많은 대형 플랫폼에 유리 → 아마존, 넷플릭스 등이 IBCF 채택

- 한계
	1. 희소 행렬 문제: 주어진 평점/선호도 데이터가 적을 경우, 성능이 저하됨(Sparsity)
	2. 확장성 문제: 아이템/유저가 계속 늘어날 경우, 확장성이 떨어짐 (Scalability)
	3. 콜드 스타트 문제: 협업 필터링은 **기존의 평점/행동 데이터를 기반**으로 유사도를 계산하는데, 데이터가 아예 없는 경우 추천이 불가능함 
		- 신규 유저 유입, 신규 아이템 등록 상황에서 발생
		- 해결방법 
		1. 직접 지정형 프로필 활용 
		2. Content-Based Filtering과 혼합하여 아이템 특성 기반으로 보완. 
		3. 인구통계학적 정보 활용

## 3.3 모델 기반 협업 필터링
- 등장배경: 메모리 기반 CF의 한계를 보완하기 위해 등장
- 단순히 유사도를 계산하는 것이 아니라, **데이터로부터 패턴을 학습한 모델**을 만들어 평점을 예측함
	- **관측된 평점 데이터를 학습(Training)** 하여, **관측되지 않은 평점을 예측(Inference)** 하는 방식!

- Model-based CF는 **데이터의 숨겨진 패턴을 학습**하여 추천의 정확도와 확장성을 동시에 높이는 방식이다.

## Latent Factor(잠재요인)
- 사용자와 아이템 사이의 관계를 설명해주는, 눈에 보이지 않는 숨겨진 특성(요인)

- 취향, 성향, 특징 등 실제 데이터에는 없지만 관측된 데이터(평점 패턴)을 분석하면서 수치로 추정할 수 있는 값

- **Latent Factor Model**: 데이터 속에 숨겨진 공통적 특징(잠재된 요인)을 찾아내어, 복잡한 데이터 간의 관계를 간략화 하는 모델링
	- **핵심기법: 행렬분해** -> 주어진 Matrix를 **사용자-잠재요인, 아이템-잠재요인**으로 각각 분해하여 학습

- Latent Factor Model 특징:
	- 유저와 아이템을 **동일한 잠재 공간의 벡터**로 표현
	- 둘 사이의 **거리, 각도로 숨겨진 취향 유사도**를 파악함
	- **축(차원)이 무엇을 의미**하는지, **몇 개의 차원**이 존재하는지 사람은 알 수 없음 <br>-> **모델이 데이터를 잘 설명할 수 있는 새로운 표현 축을 스스로 학습함**

## 행렬분해 (Matrix Factorization)
- 개념: **복잡한 하나의 행렬**을 특정한 구조를 가진 **여러 개의 더 단순하고 작은 행렬**들의 곱으로 쪼개어 표현하는 선형대수학 기법

- 이걸 추천시스템에 적용하면 **원래 평점 행렬 R을 사용자의 잠재요인 행렬(U)과, 영화 잠재요인 행렬(V)**로 분해하는 것!

- 이 방법으로 **평점을 예측** 할 수 있음

- 예시

|     | 영화A | 영화B | 영화C |
| --- | --- | --- | --- |
| 유저1 | 5   | ?   | 1   |
| 유저2 | 4   | 2   | ?   |
| 유저3 | ?   | 5   | 1   |

- 여기서 "?"는 아직 안 본 영화임
- **목표**: 비어있는 "?" 평점을 예측하는 것!
- 여기서 latent factor model은 "유저와 영화는 몇 개의 숨겨진 취향 축으로 표현 가능하다"고 가정함<br> **축1= 액션 선호, 축2=로맨스 선호로 정의**

- 그러면 유저를 벡터로 표현 가능: u1=[5,1] => 액션 좋아하고 로맨스 싫어함

- 영화도 벡터로 표현: VA=[1,0] VB=[0,1] => 각각 액션, 로맨스 영화

- **U와 V를 내적하여 평점을 예측**할 수 있다!
	- 유저 1이 영화 A를 얼마나 좋아할까? -> 5*1+1*0 = 5 -> 높은 평점 예상
	- 유저 1이 영화 B를 얼마나 좋아할까? -> 5*0+1*1=1 -> 낮은 평점 예상

# 4. 딥러닝 기반 추천 시스템
## 4.1. VAE 기반 협업 필터링
- 사용자의 아이템 상호작용 벡터(예: 구매/시청 이력)를 입력으로 받아, 잠재 공간에서 재구성하도록 VAE를 학습하는 방법
- 등장배경: MF의 한계
	1. 선형성의 한계
	2. 불확실성 무시
		- 모델이 자신의 예측을 얼마나 확신하는지 모름
- 구조
	- 입력(x): 사용자의 아이템 상호작용 벡터
	- 인코더: 사용자 이력을 잠재변수 z 분포로 매핑
	- 잠재 벡터(z): 샘플링된 최종 취향 벡터
	- 디코더: z로부터 아이템 분포를 복원
- 장점
	- 확률적 잠재 분포로 사용자 선호를 모델링 하여 **불확실성을 반영함**
	- 희소한 상호작용 데이터에서도 강건함
	- 비선형성 학습

- AE가 아니라 VAE를 선택한 이유
	1. AE의 접근 방식: "A의 취향은 (액션 0.9, 로맨스 0.1) 벡터 하나일 것이다!" (단일 값 추정)
		- 문제점: 데이터가 부족하면 이 단일 추정치가 불안정하고 틀리기 쉬워 **노이즈에 취약**

	2. VAE의 접근 방식: "A의 취향은 (평균 액션 0.8, 표준편차 0.2)를 따르는 분포일 것이다." (분포 추정)	
		- 그냥 "latent 공간 자체를 **정규분포처럼 부드럽게**만들자" 라고 이해하면 됨
		- 강점: 취향의 불확실성(데이터 부족으로 인한 오차 범위)까지 모델링에 포함시켜 안정성을 높임

### 4.1.1 VAE의 핵심: 목적 함수와 KL Divergence
- VAE의 목표: Reconstruction Loss(재구성 오차) 최소화, KL Divergence
- Reconstruction Loss(재구성 오차)
	- 목표: 잠재벡터 z를 이용하여 원래 입력(x)를 최대한 잘 복원하기

- KL Divergence: $$DKL(q(z∣x)∥p(z))$$
	- q(z∣x): 인코더가 만든 분포
	- p(z): 우리가 원하는 이상적인 분포 -> 보통 표준정규분포 사용
	- **즉, latent 공간을 전체적으로 평균0, 분산1 근처로 잘 정리해라는 뜻**

- KL Divergence의 역할: latent 공간이 정규분포를 따르면서 공간이 **부드럽고 연결된 구조**가 됨
<br>->그 결과 중간값도 의미를 가지면서 "고양이 같은 강아지 상" 이라는 **희소한 데이터에도 안정적인 예측** 가능


# 5. GNN 기반 사용자 선호 예측
- 등장배경: VAE는 **관계나 순서** 라는 풍부한 정보를 활용하지 못함
- GNN은 사용자와 아이템의 상호작용을 ‘누가(User) 무엇을(Item) 좋아했다’의 집합으로 보는 것을 넘어, <br>이들이 **서로 연결된 거대한 관계망(Graph)**을 형성한다는 점에서 모티브를 얻음

-  즉, 사용자-아이템 상호작용은 본질적으로 그래프(Graph)임
	- 그래프: **노드와, 노드를 잇는 선(엣지)**를 모아 구성한 자료
	- 노드: 현실 세계의 개체
	- 엣지: 개체 간 관계나 상호작용

- 그래프를 사용하는 이유
	1. 관계, 상호작용 같이 다소 추상적인 개념을 다루기에 적합
	2. Non-Euclidean space도 표현하고 학습할 수 있음

## 5.1 GNN의 프로세스
1. 그래프 구조 생성
사용자와 영화 정보가 노드로 생성됨
사용자가 영화를 보고 평점을 준 것은 상호작용이니까 평점은 엣지가 됨

2. 메시지 패싱: 각 노드가 이웃 노드로부터 정보를 받아 전달하고 상호작용 하는 것
- 보통 하나의 GNN 레이어에서 ‘메시지 생성 - 집계 - 갱신’ 의 세 단계로 이루어짐
- **메시지 생성**: 각 노드가 이웃에게 보낼 자신의 정보를 생성. 이때, 이웃에게 보내는 메시지느 임베딩 되어 있음
- **집계**: 메시지 생성 이후, 여러 이웃 노드에서 받은 메시지를 하나의 벡터로 통합하는 단계
- **갱신**: 집계된 이웃정보와 자신의 이전 상태를 결합해 새로운 임베딩을 계산
3. 최종 출력

## 5.2 Light GCN
- GNN은 메시지 계산 방법이나 aggregation 방식에 따라 아키텍처가 결정됨
- 주요 GNN 아키텍처로는 GCN, GRN, GAT, GraphSAGE가 존재
- LightGCN: GCN의 한 종류이며, 크게는 GNN의 한 종류
- 등장배경: 추천 시스템의 핵심은 사용자와 아이템의 상호작용, 즉 관계 그 자체인데 <br>복잡한 연산이 오히려 순수한 ‘관계’ 정보를 왜곡하고 학습을 방해할 수 있다는 아이디어에서 등장
- LightGCN 특징
    - 특징 변환 제거
    - 비선형 활성화 함수 제거: 임베딩 전파 과정에서 이미 비선형성이 충분히 반영됨 + 과적합 방지
- LightGCN은 오직 정규화된(Normalized) 이웃 임베딩의 합(Aggregation)만으로 구성됨 -> 매우 가볍고 효율이 좋음

## 5.3 GNN의 장단점
- 장점: 
    - 고차원 협업 시그널 포착
    - 정형화되지 않은 데이터 학습 가능
    - 콜드 스타트 문제 완화에 효과적
- 단점
    - 과평탄화 문제
    - 과압축 문제
    - 계산 복잡도 증가
    - 해석의 어려움


# 6. Transformer 기반 추천 시스템
- 등장배경
	1. CF의 순차적 한계
	- 순서 무시, 문맥 부족, 예측의 한계

	2. GNN의 순차적 한계
	- 지역적 정보 전파의 한계
	- 장기 의존성 문제

- Transformer: 사용자 행동을 순서 (Sequence)로 접근

- Transformer는 **시퀀스 내의 문맥과 핵심 의도를 파악**하는 데 강력!

- Transformer의 문맥 파악 예시
	- 문맥 A (캠핑 준비):
	- 행동 순서: [클릭: "텐트"] → [클릭: "침낭"] → [클릭: "등산화"]
	- Transformer의 해석: "등산화"가 "텐트", "침낭"과 강하게 연관(Attention)됩니다. 이때 "등산화"는 '캠핑 장비'로 해석됩니다.
	- 다음 추천: "캠핑용 코펠", "백패킹 배낭", "랜턴"

- Transformer의 핵심의도 파악 예시
	- 트랜스포머는 사용자의 행동이 길어질 때, 가장 최근의 행동이 아니라 **멀리 떨어진 과거의 행동**도 참고
	- 행동 순서 (시간 순):
    - **[구매: "레이저 프린터"]**
    - [검색: "가습기"] (계절성 탐색)
    - [클릭: "강아지 사료"] (반려동물 용품)
    - ... (약 20개의 서로 다른 행동 발생) ...
    - **[검색: "토너"]**
	- 사용자가 "토너"를 검색했을 때, 모델은 1~4번의 모든 과거 행동을 동시에 참조함
