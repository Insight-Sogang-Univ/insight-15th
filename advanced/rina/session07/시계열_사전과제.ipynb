# 1. 시계열 데이터란
## 1.1 시계열 데이터 정의
- 순차적인 시간의 흐름에 따라 수집된 데이터
- 연속한 관측치는 서로 **상관관계**가 존재하며, 순서가 중요한 **“시퀀스 데이터”**로 간주됨

- 단변량 시계열: **하나의 숫자**가 시간에 따라 변하는 것
	- 시간에 따라 연속적으로 측정된 단일 변수로 이루어진 시계열 데이터
	- 시간의 흐름에 따른 패턴과 트렌드 이해

- 다변량 시계열: **여러 숫자가 동시에** 시간에 따라 변하는 것
	- 같은 시간에 기록된 둘 이상의 변수로 이루어진 시계열 데이터 
	- 변수 간 상호 작용 및 영향력 파악

- 시계열 분석: 시계열 데이터를 분석하고 모델링하는 통계적인 방법과 기법
	- 데이터 **구조 파악, 예측, 이상치 탐지**를 포괄하는 개념

- 시계열 분석 목적: **미래 데이터 예측**

## 1.2 시계열 데이터 특징
1. 시간 의존성
- 시계열 데이터는 시간에 따라 변화한다

2. 계절성 & 추세
- 시간에 따라 장기적으로 증가/감소하는 추세 + 일정 주기로 반복되는 패턴
	- 계절성: 일정 주기 (일, 주, 월, 년 등)에 따라 반복적으로 나타나는 데이터의 변동<br> 
            → 주기적인 패턴을 고려한 예측 모델링 가능
	- 추세: 시간의 흐름에 따라 나타나는 장기적인 증가/감소 경향 <br>
            → 전반적인 방향성 파악 O

3. 자기 상관성
- 이전 시점의 값이 이후 시점의 값에 영향을 미친다
	- 자기 상관성: **자기자신**끼리 **상관관계**를 갖는 특성

- 한 시점의 값이 과거의 값과 통계적으로 관련 있는 성질 → 시간이 지남에 따라 값들 간에 **일정한 패턴이나 반복**이 존재한다는 것

- **자기상관성** 때문에 회귀 분석이 불가능하다!
	- 시계열 데이터를 일반 회귀 분석에 사용하면 다음과 같은 세 가지 문제 발생

	1. 오차항 독립성 가정 위반 -> 통계 결과 왜곡
	- **오차들이 서로 독립적 즉, 어제와 오늘은 독립적** 이어야 한다는 회귀분석의 기본 가정 위반

	2. 허위 상관관계 문제
	- 자기 상관성이 있는 두 시계열을 그냥 OLS 회귀하면 **잘못된 상관관계** 가 나타남
	- 실제로 **두 변수 간 인과관계가 없어도** , 공통된 시간 추세로 인해 높은 상관관계 나타남
	- 해결방안: 각 시계열 변수에서 **추세 제거**

	3. 예측의 어려움 & 모델 복잡성 증가
	- 자기상관성은 예측 신뢰도를 낮추고, 더 복잡한 시계열 모델을 요구하게 만든다

- **따라서 시계열 데이터는 시계열 전용 모델을 사용하거나, 사전 처리(차분, 추세 제거 등) 후 분석해야 한다.**

## 1.3 시계열 데이터의 분석요인
1. 추세 요인
	- 데이터 전반에 걸쳐 시간에 따라 지속적으로 증가/감소하는 경향
	- 시각적 표현: 꾸준히 증가하거나 감소하는 선형 그래프
<br>ex. 연도별 매출 상승, 출산률 감소 등

2. 계절 요인
	- **일정한 주기** (계절, 월, 요일 등 정기적인 주기)로 반복되는 패턴
	- 시각적 표현: 주기적인 변동을 나타내는 그래프 ex. 크리스마스 시즌에 관련 용품 매출이 증가

3. 순환 요인
	- 계절성보다 더 길고 **불규칙한 주기**로 반복되는 변동
	- 시각적 표현: 길게 상승과 하강을 반복하는 파형 그래프 ex. 경기 호황과 불황기

- 계절성과 순환성 구분 방법
	- **고정된 주기**를 가지고 반복되는 패턴: **계절성**
	- **주기가 불규칙하거나 장기적인 경우**, 경제적 요인처럼 외부 요인에 의해 발생하는 경우 **순환성**

4. 불규칙 요인
- 추세, 계절성, 순환성을 모두 제거한 뒤 남은 예측 불가능한 변동 
- 노이즈, 우연적 사건, 일시적인 변동, 이상치 등을 모두 포함
- 시각적 표현: 불규칙한 변동을 나타내는 그래프 ex. 자연재해, 팬데믹

## 1.4 시계열 분석의 Work Flow
1. 데이터 전처리
2. 정상성 판단
3. 모델링 및 예측

# 2. 시계열 데이터 전처리
- 전처리 필요성: 시계열 데이터의 특성(시간 의존성/추세&계절성/자기상관성)이 존재! 이러한 특성이 왜곡되지 않도록 필요
- 전처리 프로세스 
데이터 확인/시각화
결측치 확인 → 결측치 처리(삭제/단순 대치/보간/이동평균 대치 등)
이상치 탐지 → 이상치 처리(삭제/단순 대치/보간/이동평균 대치 등)
평활화 진행 → 이동평균 이용
요소 분해 → 가법 모형, 승법 모형

## 2.1 시계열 데이터의 결측치
- 특정 시점에서 값이 측정되지 않았거나 누락된 상태
- 결측치가 존재하면 **추세, 계절성이 왜곡**되므로 적절한 대치와 보간이 필요

- 결측치 처리 방법
	- 삭제/단순 대치/보간/머신러닝 기법 등 다양한 방법 존재

- 보간법
	- 주어진 데이터 포인트의 값을 사용하여 두 개 이상의 데이터 포인트 사이에서 **누락된 값을 추정**하는 수학적 기법
	- 결측치/이상치 구간이 짧거나 신호의 패턴이 단순한 경우에 보간 정확도 높음
	- 종류
		- 선형 보간법
		- 다항식 보간법
		- 스플라인 보간법
		- 조각별 상수 보간법

- 시계열 데이터의 이상치: 값이 측정되었으나, 측정된 값이 **일반적이지 않은 경우**
	- 단순히 이상한 값이 아니라 신호의 중요한 의미일 가능성도 있으므로 신중한 접근 필요 

- 이상치 탐지 방법
	- IQR Rule-based Anomaly Detection
	- Z-score
	- STL 분해

- 이상치 탐지 모델
	- 분류 및 회귀 트리 (CART)
	- 클러스터링 기반 이상 탐지
	- 오토 인코더

- 이상치 처리 방법
	- 이상치를 결측치로 간주 → 결측치와 같은 방법으로 처리

## 2.2 EDA(평활화)
- 시계열 데이터에서 노이즈를 줄이고, 전체적인 흐름이나 추세를  더 명확하게 파악하기 위해 데이터를 부드럽게 만드는 기술
	- 노이즈: 원래 데이터 안에 자연스럽게 섞여 있는 **설명 불가능한 랜덤 변동(오차)**
	- 우리가 관측한 데이터 = 신호(signal) + 노이즈(noise)

- 이동평균(Moving Average, MA)
	- N=Moving Window의 길이
	- N이 클수록 긴 기간의 데이터를 평균에 포함→ **노이즈가 더 많이 제거되어 보다 매끄러운 곡선 생성**

	- 사용 목적
		- 노이즈 제거: 일시적인 급등/급락을 평탄화
		- 추세 파악: 상승/하강 흐름을 쉽게 식별
		- 예측의 기초자료: 시계열 예측(ARIMA 등)의 기반이 되기도 함

- 이동평균 유형
	- 단순 이동평균: 특정 기간의 데이터를 단순 평균으로 계산
	- 가중 이동평균: 최근 데이터에 더 높은 가중치 부여 -> 최근 변화 더 민감하게 반영
	- 지수 이동평균: 최근 데이터 포인트에 더 많은 가중치 두면서 **지수적으로 감소하는 가중치**를 적용하여 평균을 계산하는 방법
		- 과거의 모든 데이터 고려

- 이동평균을 이용한 대치
	- 결측치나 이상치 처리 기법은 아니지만 결측치(이상치) 주변 n개의 값을 이동평균 계산 후, <br>그 값으로 결측치(이상치)를 대치하는 식으로 결측치/이상치 처리에 활용되기도 함

## 2.3 요소분해(시계열 분해)
- 주어진 시계열 데이터를 개별 성분(추세성, 계절성, 순환성, 불규칙성)으로 분해하는 과정
- Y  t ​   =  𝑓 f(  𝑇 𝑡 , 𝑆 𝑡 , 𝐶 𝑡 , 𝑅 𝑡 T  t ​  ,S  t ​  ,C  t ​  ,R  t ​  ) 
	- T  t ​   : 추세성분 (시점 t에서의 추세요소) 
	- S  t ​   : 계절성분 (시점 t에서의 계절요소) 
	- C  t ​   : 순환성분 (시점 t에서의 순환요소) 
	- R  t ​   : 불규칙성분 (시점 t에서의 불규칙요소)
	- 일반적으로 순환성은 장기간에서만 고려되므로 C t  = 0 으로 설정함

	- 예시) 아이스크림 가게의 매출 분석
	- Tt=장기적인 우상향
	- St=여름에는 잘 팔리고 겨울에는 상대적으로 덜 팔리는 주기적 패턴
	- Ct=몇 년 주기로 크게 출렁이는 경제 흐름
	- Rt= 어느날 갑자기 유명 먹방 유튜버가 방문해서 매출 폭발
	- 이때 매출=f(Tt, St,Ct, Rt)

- 요소분해를 하는 이유
	- 시계열 데이터 = 추세/계절성/순환성/불규칙성 등 다양한 변동이 결합된 형태임
	- 각 요소를 분리하면 각각의 패턴을 명확히 이해 가능
	- 이후 예측, 이상 탐지 등 분석 작업을 정교하게 수행할 수 있음

#### 분해 "모형"(수학적 형태로의 분해)
1. 가법 모형
	$$y_t = T_t + S_t + R_t$$
	- 시계열 = 추세 + 계절성 + 불규칙성
	- 시계열 관측값은 변동요인들의 가법(+)으로 설명될 수 있다 가정
	- **각 변동 요인들이 독립적으로 더해지며, 모든 요소의 영향이 일정함**
	- 시계열 전체에서 계절성의 진폭이 일정할 때 적합하다

2. 승법모형
$$log(y_t) = log(T_t) + log(S_t) +log(R_t)$$
<br>*로그 변환을 통해 가법 형태로 만들 수 있다
	- 시계열 = 추세 x 계절성 x 불규칙성
	- 시계열 관측값은 변동요인들의 승법(*)으로 설명될 수 있다 가정
	- **추세가 커질수록 계절성이나 불규칙성의 효과도 비례해서 커짐**
	- 진폭이 시간에 따라 달라질 때에 적합하다

- 예시)
	- 가법모형 : 여름철 한국에 방문하는 관광객은 **매년 평균 10만명**씩 증가한다
	- 승법모형: 여름철 한국에 방문하는 관광객은 **매년 평균 10%**씩 증가한다

#### 분해 "방법"(구현 알고리즘)
1. seasonal_decompose - MA 기반 분해
- 원리: 전통적인 시계열 분해 기법, 이동 평균을 통해 추세를 추출한 뒤 잔여 성분에서 계절성을 분리하는 방식
- 작동방식
	1. 추세추출: 중심 이동평균을 적용하여 추세성분 Tt 추정
	2. Detrending: 원 시계열에서 추세를 제거하여 detrended-시계열 생성
	- 𝐷 𝑡 = 𝑌 𝑡 − 𝑇 𝑡
	3. 계절성 추정: 디트렌드 시계열에서 동일한 계절 인덱스에 해당하는 값들의 평균을 취해 계절 패턴 St 도출. 이때 연간 평균이 0이 되도록 정규화
	4. 잔차 산출
	- R  t ​  =Y  t ​-T  t -S  t 

2. STL-LOESS 기반 분해
- 원리: 이동평균 대신 LOESS(Locally Estimated Scatterplot Smoothing)를 사용한다. <br>추세와 계절성이 시간에 따라 변화하는 경우에도 유연하게 포착할 수 있으며, 이상치에 강건한 분해가 가능하다.

- 작동 방식
	- Inner Loop와 Outer Loop의 이중 반복 구조로 구성됨
	- Inner Loop(반복 수렴)
	1. 추세 제거: 𝑌 𝑡 − 𝑇 𝑡
	2. Cycle-subseries 추출 후 각 subseries에 LOESS 적용
	3. Low-pass filter로 저주파 성분 제거
	4. 계절 성분 St 확정
	5. 계절성 제거: 𝑌 𝑡 − 𝑆 𝑡
	6. LOESS로 추세 Tt갱신
	- -> 통산 2회 반복으로 수렴
	
	- Outer Loop
	- 잔차 크기 기준으로 강건성 가중치를 부여해 이상치의 영향력을 줄인 후 Inner Loop 재실행

# 3. 정상성 판단
## 3.1 정상성이란
- 시계열 데이터의 **평균, 분산 등 통계적 특성**이 시간에 따라 일정한 수준으로 유지되는 것
	- **'시간에 상관없이' 일정한 성질을 띄고 있다**

- 정상성 여부에 따라 정상 시계열, 비정상 시계열로 나눠짐
- 통계 분석에서 정상성이 중요한 이유
	- **예측이 가능하다**: 과거의 패턴과 성질이 미래에도 똑같이 유지 될것이라 가정할 수 있기 때문에 통계모델링에 바로 적용하기 좋음

## 3.2 강한 정상성과 약한 정상성
- 강한 정상성: 완벽하지만 비현실적인 기준
- 강한 정상성 개념: 시간을 언제로 잡든, 데이터의 확률 분포 자체가 똑같은 상태
- 특징: 평균, 분산 뿐만 아니라 데이터의 왜도, 첨도 등 모든 통계적 특성이 일정
- 한계: 현실의 데이터는 잡음과 외부 요인이 항상 존재하기 때문에, 이런 상태는 거의 존재하지 않음

- 약한 정상성: 실무에서 쓰는 현실적인 기준
- 특징
	- 평균이 일정: 시간이 지나도 일정한 수준을 유지
	- 분산이 일정: 데이터의 오르내림 변동 폭이 일정한 수준을 유지
	- 자기공분산이 시차에만 의존: 데이터 간의 관계가 특정 시점이 아니라 시간의 간격에 의해서만 결정
		- 즉, 언제 측정하느냐가 중요한 것이 아닌, **얼마나 떨어졌느냐**만 중요하다는 것

## 3.3 정상성 판단하기
1. 시각적으로 정상성 판단하기
- 정상성을 띄는 시계열 데이터는 그래프를 봤을 때, **일정한 기준선(평균)을 중심으로 일정한 폭으로 진동**하는 형태를 보임

- 비정상 시계열의 대표적인 특징
	- 추세가 있는 경우: 시간에 따라 데이터가 지속적으로 상승하거나 하락(ex.주식 가격, 물가 상승률)
	- 계절성이 있는 경우: 1년 등의 특정 주기로 패턴이 반복됨 (ex. 월별 아이스크림 판매량)

- 추세가 뚜렷한 시계열을 정상 시계열로 변환하는 법
	- **차분** 한 결과로 해석 가능

2. 통계적 검정
- 필요성: 그래프만으로는 시각적으로 정상성 여부를 판단하기 어렵다. 이에 따른 수치적인 가설 검증이 필요함.
- 이때 대부분의 시계열 검정은 **단위근**이 있는지를 확인하는 과정을 거친다.
	- 단위근: 과거 충격이 영구적으로 누적되는 특성을 의미 - 대표적인 비정상성의 원인으로 존재
	- 시계열이 Random하게 움직이게 만들어 평균으로 회귀하지 않고, 분산이 시간에 따라 증가하며 예측이 어려워짐.

- 단위근으로 정상/비정상 시계열을 판단하는 과정
$$y_t = \alpha y_{t-1} + e_{t}$$ 
	- y  t ​   ,  𝑦 𝑡 − 1: 시계열의 관측값 
	- 𝛼  : AR(1) 계수 (자기회귀 계수) → 이 모델에서 특성방정식의 근(단위근)을 뜻함 
	- 𝑒 𝑡  : 백색 잡음. 외부 충격을 뜻함.

- 𝛼=0일때 (완전한 백색잡음)
	- 과거 값 𝑦 𝑡 − 1의 영향을 **전혀 받지 않음** 
	- 오직 외부 충격(오차항 𝑒𝑡)만으로 시계열이 생성됨
	- 오늘의 값은 어제와 아무 관련이 없고, 오로지 외부 요인에 의해 무작위로 결정된다<br>
→ 완전한 정상성을 가짐

- ∣α∣<1일때 (정상 시계열)
	- 현재 값이 이전 값의 영향을 받지만, 시간이 지날수록 과거 충격의 영향은 **점점 줄어듦(희석됨)**
	- 외부 충격의 영향이 시간이 지날수록 점차 사라지고 시계열은 평균 수준으로 돌아온다!<br>
→ 정상 시계열로 간주 

- α=1일때 (비정상 시계열), 단위근이 있음
	- yt는 지금까지의 오차의 누적과 같음
	- 외부 충격으로 인한 변동이 계속 유지되어 한 번 충격을 받으면 원상태로 돌아가지 못함
	- 시계열이 일정한 분포나 상태로 수렴하지 않음 → 예측 불가능

- 단위근 검정 방법
	- ADF
		- "이 시계열이 비정상이 아님을 증명하자" 는 관점에서 출발
		- 시계열이 과거 값에 얼마나 의존하는지를 측정
		- 과거 값의 영향이 시간이 지남에 따라 점점 사라지면 정상 시계열, <br>누적되어 사라지지 않으면 단위근이 존재하는 비정상 시계열로 판단
	- KPSS
		- "이 시계열이 정상임을 반증하자"는 관점에서 출발
		- ADF의 검정력이 낮다는 한계를 보완하기 위해 함께 사용
		- KPSS는 시계열이 일정한 평균 수준을 유지하는지를 측정
		- 평균이 시간에 따라 흔들리지 않으면 정상, 평균 자체가 계속 변하고 있으면 비정상으로 판단

## 3.4 정상성 확보를 위한 여러가지 방법들
1. 추세 제거
	- 시계열이 지속적으로 증가하거나 감소할 때 적용
	- **차분**: 연이은 관측값들의 차이를 계산하는 것

2. 계절성 제거
	- 시계열이 주기적인 패턴을 가질 때 적용
	- **계절 차분** t시점의 𝑦값에서 t-m시점의 y값을 빼는 것 (m: 주기의 길이)

3. 분산 안정화
	- 시계열의 변동 폭이 점점 커지거나 작아질 때 적용
	- **로그변환**: 분산의 정상성 확보하기 위한 방법

# 4. 단변량 시계열 분석
## 4.1 안정적 시계열 분석
1. 자기 회귀 모델(Autoregressive Model, AR)
<br>: 과거의 ‘자기 자신 데이터(순위)’가 미래 값을 설명해줄 것이라고 가정하는 것이 자기 회귀

- 자귀회귀: 
과거 자신의 값들을 독립변수(x)로 삼아 현재 값(y)을 예측하는 회귀
<br>→ 시계열 데이터의 ‘자기상관성’을 활용 

- $$AR(p):  y_{t}=c +\phi _1y_{t-1}+\phi_2y_{t-2}+...+\phi _py_{t-p}+\varepsilon _t$$
	- p차 자기 회귀 모형
	- 과거 p개 시점의 데이터의 선형 조합을 이용해 예측하는 모델

2. 이동 평균 모델(Moving Average, MA)
- 앞서 나온 이동평균과 완전히 다른 개념
- **어제의 충격이 오늘의 결과에 영향**을 준다고 가정하는 것이 바로 이동 평균의 핵심

- 이동평균모델:  과거 시점의 잔차(ε)를 독립변수로 삼아 현재 값을 예측하는 모델<br>
→ 과거의 잘못 예측한 것(예측 오차, 충격)을 현재 시점에 반영

- $$MA(q): y_t = \mu +\varepsilon _t+ \theta _1\varepsilon_{t-1}+\theta_2\varepsilon_{t-2}+...+\theta_q\varepsilon_{t-q}$$
	- q차 이동 평균 모형
	- **과거 q개의 예측 오차**의 **선형 결합**으로 예측하는 모델

#### AR, MA 모델의 전제 조건
1. 정상성
	- AR과 MA 모형은 안정적 시계열 분석으로 확률적 안정성을 전제로 추정<br>
→ 모델 구조도 정상성을 만족해야 함
	- AR 모형의 정상성 조건: 모든 해(z) 의 절대값>1
	- MA 모형의 정상성 조건: 잔차(<=> 백색잡음)들은 정상성을 이미 만족하기 때문에, <br>잔차들의 조합으로 이뤄진 MA 모형도 항상 정상성 만족

2. 가역성
	- 가역성: MA 모형을 AR(∞) 형태로 재표현할 수 있는지 여부
	- 필요성: MA 모형에서는 현재 시점의 값을 과거 시점의 오차항 ε으로 예측 <br>→ 문제 : 오차항은 직접 관찰이 불가능<br> → 오차항(ε )도 관측 가능한 과거값( 𝑦 𝑡 − 𝑛 )으로 표현할 수 있어야 안정적
	- AR 모형의 가역성 만족 조건?
		- AR 모형은 처음부터 AR형태로 표현되어 있기 때문에 가역성 체크할 필요 X
	- MA모형의 가역성 만족 조건: 모든 해(z)의 절대값>1

#### ARMA 모형
- AR(p) 모형과 MA(q) 모형을 결합한 모형 
- 과거 p개의 관측값 + 과거 q개의 오차를  활용해서 예측하는 방법
- 조건: AR(p) 모형의 정상성 + MA(q) 모형의 가역성 → 이 둘의 조건 모두 만족해야 함

#### 차수 결정 방법: 자기 상관 함수
- 두 그래프를 통해 현재 값이 과거 값과 어떤 관계가 있는지 파악하고 이를 통해 AR, MA, ARMA의 적절한 차수(p, q)를 결정

- 자기 상관 계수(Auto Correlation Function, ACF)
	- ARMA(p, q)모형에서 사용할 q개의 오차 결정 방법
	- 두 시점 사이의 상관 관계 파악 (간접 영향까지 모두 파악)

- 부분 자기 상관 계수 (Partial Autocorrelation Function, PACF)
	- ARMA(p, q) 모형에서 사용할 p개의 시점 결정 방법
	- 두 시점 사이의 직접적인 영향만 파악

## 4.2 불안정 시계열 분석
- 불안정 시계열 모델은 안정적 시계열 모델(ARMA)에 정상성 확보 과정(차분, 계절성 차분)이 추가된 형태

1. ARIMA (Autoregressive Integrated Moving Average)
$$ ARMA(p, q) + \text{차분} \rightarrow ARIMA(p,d,q) $$

- ARIMA 모형의 특수한 경우
1. 백색잡음: ARIMA(0,0,0)
2. 자기회귀: ARIMA(p,0,0) ⇔ AR(p)
3. 이동평균: ARIMA(0, 0, q) ⇔ MA(q)
4. 랜덤워크: 상수가 없는 ARIMA(0,1,0)
5. 표류를 포함하는 랜덤워크: 상수가 있는 ARIMA(0,1,0)

2. SARIMA (Seasonal ARIMA)
- 시계열 데이터의 계절성을 처리할 수 있는 모델
- ARIMA 모델에 계절성 변동을 반영한 모델

- 하이퍼파라미터
	- (p,d,q): 비계절 ARIMA 파라미터
		- p → AR의 차수
		- d → 차분의 정도
		- q → MA의 차수
	- (P,D,Q): 계절 ARIMA 파라미터
		- 𝑃→ 계절성 AR의 차수
		- D → 계절성 차분의 정도
		- Q → 계절성 MA의 차수
	- m: 계절성 주기의 길이

# 5. 다변량 시계열 분석
## 5.1 VAR(Vector Autoregression)

- 핵심 아이디어: 자신의 과거 값과 다른 변수들의 과거 값을 함께 사용하여 미래를 예측하는 모델
- 단변량 VAR: X는 자기 자신의 과거만 사용
- 다변량 VAR:  X와 Y가 서로의 과거를 함께 사용

- VAR 모델을 활용한 분석방법
	1. 충격 반응 함수: 한 변수의 충격이 다른 변수에 미치는 영향을 시간 흐름에 따라 분석
	2. 예측오차 분산 분해: 예측 오차에 각 변수가 얼마나 기여하는지 분석

# 6. 딥러닝을 활용한 시계열 분석
- 등장 배경: 전통 통계 모델의 한계

1. 순환 신경망(RNN 계열)
	- RNN/LSTM/GRU

- 적합한 경우:
	- 순차적 패턴이 중요한 시계열 데이터
	- 과거 정보가 미래 예측에 영향을 주는 문제
	- 시계열 길이와 목적에 따라 모델 선택 가능

- 부적합한 경우
	- 매우 긴 시계열(수천~수만 step)에서는 학습이 어려울 수 있음
	- 병렬 처리 효율이 낮음
	- Transformer 계열 대비 학습 속도가 느린 편

2. Transformer 계열
1. Informer(장기 예측 특화)
	- 특징: 불필요한 연산을 줄여 계산 복잡도를 대폭 낮춘 모델
	- 매우 긴 시계열 데이터 처리에 강력함

2. Autoformer(주기성 포착 특화)
	- 특징: 시계열의 전통적 기법인 '요소 분해(추세/계절성)' 개념을 딥러닝 구조에 직접 결합한 모델
	- 연간 계절성 등 주기 패턴이 뚜렷하게 나타나는 판매량 예측에 압도적인 성능을 보임

3. TFT(Temporal Fusion Transformer - 실무 활용도 1위)
	- 특징:  과거 데이터뿐만 아니라 다양한 외부 변수(날씨, 휴일 유무 등)를 섞어서 예측할 수 있는 다목적 아키텍처
	- 어떤 변수가 예측에 가장 중요했는지 어텐션 스코어를 통한 해석이 가능해, 설명력이 중요한 유통/비즈니스 수요 예측에 가장 많이 쓰임

3. Foundation Model
- 거대한 데이터로 사전 학습하여 다양한 문제를 해결할 수 있는 범용 모델
- 특징
	- 특정 문제에 적용 안됨
	- 범용적
	- 사전 학습 후 파인 튜닝으로 특화
