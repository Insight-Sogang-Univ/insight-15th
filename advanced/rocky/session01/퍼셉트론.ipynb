## 인공 신경망
- 우리 뇌의 신경망을 모방한 모델

1. 입력들을 받아 각각 가중치와 편량을 적용
2. 종합한 값을 활성화 함수에 넘김. 이때 활성화 함수는 임계값을 기준으로 삼음
3. output 출력

## 퍼셉트론
- 퍼셉트론(Perceptron)은 초기 형태의 인공신경망으로, 신경망(딥러닝)의 기원이 되는 선형 이진 분류기(Linear Binary Classifier)

### 퍼셉트론 출력값을 구하는 절차
⇒ 입력 → 가중합 → 활성화 함수 → 출력 흐름을 따라감
1. 입력값과 가중치를 곱함 →  ${x_1*w_1}$
2. 곱한 값들의 총합을 구함 → ${x_1*w_2+x_1*w_2+x_3*w_3 +...+x_n*w_n}$
(편향을 더해줄지는 옵션)
3. 총합이 임계값(${\theta}$**)**을 넘으면 $1$, 넘지 않으면 $0$을 출력
(이 경우, 활성화 함수는 Step 함수)

→ Training(학습): 원하는 출력값을 내보내도록 가중치를 조정해가는 작업

## 단층 퍼셉트론
### 단층 퍼셉트론: 단순히 입력층과 출력층으로만 이루어진 퍼셉트론


- 위의 인공신경망 예시에서 살펴보았던 과정은 단층 퍼셉트론
- 단층 퍼셉트론은 값을 보내는 단계와, 값을 받아서 출력하는 단계로만 이루어짐

### 논리 게이트 구현
- 단층 퍼셉트론을 이용하면 AND, NAND, OR 논리 게이트를 쉽게 구현 가능
- 논리 게이트란?
    - 디지털 회로에서 기본적인 논리 연산을 수행하는 데 사용되는 구성 요소
    - 게이트 연산에 쓰이는 것은 n 개의 입력값과 하나의 출력값
    - AND, OR, NOT, NAND, NOR, XOR 등 다양한 종류가 있으며, 이러한 논리 게이트를 조합하여 복잡한 논리 회로를 구성할 수 있음

### 대표적 논리 게이트
#### 1. AND 게이트
- 입력값이 모두 1인 경우에만 출력값이 1이 나오는 구조


- 가중치와 편향 값을 w1, w2, b라고 가정. 
- AND 게이트를 만족하는 두 개의 가중치와 편향 값은?
    - ⇒ [w1, w2, b] = [0.5, 0.5, -0.7], [0.5, 0.5, -0.8], [1.0, 1.0, -1.0] 등 다양한 조합이 가능
    - y = 0.5 * x1 + 0.5 * x2 + (-0.7)

#### 2. OR 게이트
- 입력값이 모두 0인 경우에 출력값이 0이고, 나머지 경우에는 모두 출력값이 1인 구조


- 만족하는 가중치와 편향 예시:  [0.6, 0.6, -0.5]

#### 3. NAND 게이트
- 입력값이 모두 1인 경우에만 출력값이 0, 나머지 경우에는 모두 출력값이 1인 구조


- 만족하는 가중치와 편향 예시:  [-0.5, -0.5, +0.7]

#### 4. XOR 게이트

- 입력값 두 개가 서로 다른 값을 갖고 있을 때에만 출력값이 1인 구조




## 단층 퍼셉트론의 한계
#### → XOR 게이트를 구현할 수 없음
- 퍼셉트론은 입력값과 가중치의 선형 결합으로 이루어진 구조이기 때문
- 단층 퍼셉트론은 직선 하나로 두 영역을 나눌 수 있는 선형 분류 문제에 대해서만 구현 가능
- ex) $y = x_1 w_1 + x_2 w_2 + \dots + x_n w_n$





#### XOR 게이트는 직선이 아닌 곡선/비선형 영역으로 분리해야만 구현 가능
- 따라서 퍼셉트론 하나로는 XOR 구현 불가


- -> 퍼셉트론을 여러 계층으로 쓰면 해결.

---
## 다층 퍼셉트론
- XOR 게이트는 기존의 AND, NAND, OR 게이트를 조합하여 생성 가능


- 중간에 층을 추가하여 (입력층 + 하나 이상의 은닉층 + 출력층)으로 구성
-  XOR 문제보다 더욱 복잡한 문제를 해결할 땐 중간에 수많은 은닉층을 추가 가능



- 두 개 이상의 은닉층을 갖는 MLP
    - 딥러닝의 시초 모델로 발전
    - **심층 신경망 (Deep Neural network, DNN)** = 은닉층이 2개 이상인 신경망
    - **딥러닝 (Deep Learning)** = 심층 신경망을 학습(가중치를 스스로 찾도록) 시킴


# 딥러닝 모델 학습방법

## 순전파와 역전파
### 학습 알고리즘의 기본 구조
1. 순전파를 통해 입력값으로부터 예측 수행
2. 오차 계산을 통해 예측 정확도 측정
3. 역전파를 통해 출력층에서 다시 입력층 방향으로 오차 전달
4. 가중치 업데이트
5. 1~4를 반복…
- 이 과정을 계속 반복하다 보면 오차값이 0에 가까워지게 됨

### 연쇄법칙
- 합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다는 법칙
- 딥러닝에서는 입력값이 여러 은닉층을 거친 후 최종 출력값이 나옴
- 이때 입력값이 최종 출력에 얼마나 영향을 미쳤는지 계산하려면 각 층이 다음 층에 미친 영향을 순서대로 곱해나가야 함
- 이렇게 하면 각 단계가 연쇄적으로 결과에 얼만큼 영향을 줬는지 알 수 있음

> #### 연쇄법칙이 필요한 이유
- 딥러닝 모델을 학습시키려면 오차를 줄여야 함
- 이를 위해 역전파 과정에서 각 가중치가 최종 손실에 얼마나 기여했는지를 계산해야 하는데, 이때 연쇄법칙을 사용
- 딥러닝 모델에는 은닉층이 여러 개 존재하는데, 각 은닉층마다 손실에 주는 영향을 곱해서 최종 손실에 대한 값을 나타냄


- A → B → C로 이어지는 3층 구조를 예로 들면,
- A는 B에 영향을 주고
- B는 C에 영향을 줌
- 이때 A가 C에 미친 영향은 직접 계산할 수 없고, 위처럼 분해해서 구해야 함


## 손실 함수(Loss Function)
- 손실 함수는 모델의 예측이 얼마나 틀렸는지를 수치로 표현하는 방법
### 손실함수의 종류
> #### 1. 평균 제곱 오차


- 예측 값이 연속적이고, 예측 값과 실제 값 간의 차이를 직접적으로 측정하는 회귀 문제에 적합
- ex) 모델이 예측한 집 가격과 실제 집 가격 간의 차이를 최소화하는 집 가격 예측

> #### 2. 이진 크로스 엔트로피


- BCE는 이진 분류 문제(로지스틱 회귀)에서 사용하는 손실 함수
1. 실제 정답이 1일 때
- 수식에서 $y=1$을 대입하면 $L=-\log\hat{y}$만 남음
- 예측 확률이 1에 가까울수록 → loss값은 0에 가까워짐 = 잘 맞힘
- 예측 확률이 0에 가까울수록 → loss가 급격히 증가 (확신했지만 틀림)

2. 실제 정답이 0일 때
- 예측 확률이 0에 가까울수록 → loss값은 0에 가까워짐 = 잘 맞힘
- 예측 확률이 1에 가까울수록 →  loss가 급격히 증가 (확신했지만 틀림)

MSE는 단순히 loss의 크기만 보는 반면, BCE는 얼마나 확신을 가지고 예측했는지도 loss에 반영한다는 차이가 있음

- -> 낮은 확률로 예측해서 맞추거나 높은 확률로 예측해서 틀리는 경우 loss가 더 큼

> #### 3. 크로스 엔트로피 오차



- 다중 클래스 분류 문제에서 예측 확률이 실제 클래스와 얼마나 일치하는 지를 평가하는 데 적합
- 두 확률 분포(진짜 분포 / 예측 분포)의 유사도를 측정하는 데에 유용
- 2번 손실 함수와 거의 유사한 형태이지만, 다중 분류에서 사용

## 활성화 함수
- 신경망 각 층은 기본적으로 아래와 같은 계산을 함
- **입력값 × 가중치 + 편향  →  활성화 함수  →  출력값**
- 이때  활성화 함수가 없으면, 층을 아무리 많이 쌓아도 결국 선형 계산의 반복에 불과하기에 앞서 봤던 XOR 같은 복잡한 패턴을 학습할 수 없음
- **활성화 함수는 각 층의 출력값에 비선형성을 추가해서 신경망이 복잡한 패턴도 학습할 수 있게 해주는 함수**
- 활성화 함수는 단층/다층 퍼셉트론에 따라 다르게 사용



> ### 계단 함수


- **$\mathbf{h(x) = \begin{cases} 1 & (x \geq 0) \\ 0 & (x < 0) \end{cases}}$**
- 입력값과 가중치의 곱을 모두 더한 뒤, 그 합이 특정 기준(임계값)을 넘으면 1, 아니면 0을 출력하는 함수
- 입력 신호: 입력값과 가중치의 곱
- 출력 신호: 0 또는 1

> ### 시그모이드 함수


- **$\mathbf{f(x) = \frac{1}{1+e^{-x}}}$**
- 입력 신호를 0과 1 사이의 연속적인 값으로 변환하는 비선형 함수

> ### ReLU 함수


- **f(x)=max(0,x)**
- 입력이 0보다 크면 그대로, 0 이하면 0을 출력하는 비선형 함수
- 단순하지만 딥러닝에서 가장 많이 사용되는 활성화 함수

> ### Tanh(쌍곡탄젠트)함수


- **$\mathbf{f(x) = \tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}}$**
- 시그모이드 함수의 변형으로, 출력값의 범위가 -1 ~ 1
- 시그모이드(0~1)와 달리 출력값의 중심이 0이기 때문에, 입력값의 중심을 0으로 맞추고 싶을 때 사용
-  중심이 0이면 학습 과정에서 가중치 업데이트가 더 안정적으로 이루어지는 장점이 있음
- 다만 시그모이드와 마찬가지로, 입력값이 매우 크거나 작으면 미분값이 0에 가까워져 기울기 소실 문제가 발생할 수 있음

> ### Softmax 함수


- **$\mathbf{f(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}}$**
- 각 클래스에 대한 출력값을 확률로 변환하는 함수
- 다중 클래스 분류 문제의 출력층에서 사용 가능
- 출력값이 모두 0~1 사이이고, 전체 합이 1이 되기 때문에 각 클래스에 속할 확률을 직접 읽을 수 있음
- ex) 고양이/강아지/새를 분류할 때 → [고양이: 0.7, 강아지: 0.2, 새: 0.1] 

## 옵티마이저 (Optimizer)
- 경사하강법을 더 효율적이고 안정적으로 수행하기 위해서 필요
- 옵티마이저를 통해 더 효과적인 최솟값을 찾아갈 수 있음


- 모멘텀(momentum), 알엠에스프롭(RMSProp), 그리고 아담(Adam)을 중심으로 학습 예정

> ### 모멘텀(Momentum)
**“변수가 가던 방향으로 계속 가도록 속도를 추가한다”**
- 모멘텀은 관성이라는 물리학 법칙을 응용한 방법
- 경사하강법 + 관성
- 경사하강법에서 계산된 접선의 기울기에 바로 한 시점 전의 접선의 기울기값을 일정 비율만큼 반영
- → 따라서 언덕에서 공이 내려올 때, 중간의 local minima에 빠지더라도 관성의 힘으로 넘어올 수 있음



> ### 알엠에스프롭(RMSProp)
- 기존의 경사하강법 문제: 기울기가 큰 방향에서는 너무 많이, 작은 방향에서는 너무 작게 이동
-  따라서 각 학습률을 조정해서, 안정적이고 빠르게 학습

- 산을 내려가는 상황에서의 RMSProp
    - 산을 내려갈 때, 기존 경사 하강법은 무조건 현재 기울기를 보고 내려감
- 그러나 **RMSProp**에서는 최근 기울기를 참고해서 감
    - 급한 경사에서는 속도를 줄이고
    - 완만한 경사에서는 속도를 유지하거나 증가

    

> ### 아담(Adam)
**RMSProp과 Momentum을 합친, “방향과 스텝 사이즈를 적절하게 합친 방법”**
- 방향과 학습률 두 가지를 모두 잡기 위한 방법
- 각 파라미터의 1차 모멘트(기울기의 평균)와 2차 모멘트(기울기의 제곱 평균)을 사용하여 학습률을 조정
