# NLP (Natural Langauge Processing)
## 자연어 
- 인간이 소통을 위해 자연스럽게 발생시켜 사용하는 언어(한국어, 영어 등)
- 컴퓨터가 이해하는 프로그램 언어(Python, Java)와 대비되는 개념
- 즉, 우리가 일상에서 사용하는 언어와 같은 개념

## 자연어 vs 프로그래밍 언어 
- **프로그래밍 언어 (C, Python):** 엄격한 규칙, 논리적, **일대일 대응**. 에러가 나면 돌아가지 않음
- **자연어 (Natural Language):** 인간이 소통을 위해 자연발생적으로 사용하는 언어. 모호함, 유연함, **일대다 대응**

## NLP
- NLP(Natural Langauge Processing)란, 인간의 언어(자연어)의 의미를 분석하고, 컴퓨터가 처리할 수 있도록 하는 인공지능(AI) 기술 분야
- **NLP의 역할:** 이 모호한 '자연어'를 컴퓨터가 이해할 수 있는 '숫자(데이터)'로 변환하여 유의미한 가치를 창출하는 기술
- NLP는 비정형 데이터인 글자를 컴퓨터가 계산할 수 있는 '숫자(Vector)'로 바꾸고, 그 안에서 의미를 추출하는 모든 과정을 포함

# 통계적 언어 모델 (SLM)
통계 기반 언어 모델(SLM)은 분포 가설(Distributional Hypothesis)이라는 아주 중요한 아이디어에 근거함

- 분포 가설: "비슷한 문맥에서 함께 나타나는 단어들은 비슷한 의미를 가진다"는 가설
- 어떤 단어의 주변에 자주 등장하는 단어들을 보면 그 단어의 의미를 짐작할 수 있다는 뜻
- 이런 분포가설을 설명해주는 기본 개념이 '**조건부 확률**’ 과 ‘**카운트기반 확률**’

> ### **조건부 확률**

- “나는 버스를” 까지만 말한 상황에서 그 다음에 올 단어는 “탔다” 가 자연스럽게 이어지는 경우가 더 많음
- 이렇게 더 높은 확률로 올 수 있는 것을 선택
- 조건부 확률을 통해 앞에 ‘버스를’이 오면 ‘탔다’가 올 확률을 활용하는 방식으로 진행

> ### **카운트 기반**

- 이전 단어 시퀀스의 **등장 빈도**를 통해 다음 단어의 확률을 계산
- 카운트 기반의 통계적 언어 모델은 직관적이지만 명확한 한계를 가짐
- 바로 **희소 문제(Sparsity Problem)**
- 모델은 이전에 경험한 적 있는 조합에 대해서만 잘 예측 가능
- “예쁘고 멋진 인사이트” 같은 조합이 한 번도 안 나왔으면, 이 문장 다음에 올 단어가 14기인지 15기인지 모델이 예측 불가

SLM은 새로운 문장 조합에 약하다 → 그래서 새로운 문장 조합에도 적용할 수 있는 신경망 모델이 필요



## N-gram
> N-gram이 나오게 된 배경
- SLM은 희소하게 등장한 것에 대해서는 잘 예측을 못하고, 문장이 길어질수록 다음 단어를 계산하기 어려운 문제를 가짐
- 전체 문장 대신에 문장의 일부만 보는게 어떨까? 해서 나온 것이 N-gram


- uni: 1개씩, bi: 2개씩, tri: 3개씩, … N: N개씩 보는 것
- 해당 예시는 N=4일 때, spreading 다음에 올 단어를 예측하기 위해 N-1개를 봐서 계산

> N-gram의 한계
1. 희소 문제
    1. 데이터 부족: 훈련 데이터에서 나타나지 않은 단어 조합(N-gram)은 확률이 0이 되어, 해당 단어 시퀀스를 예측하거나 평가할 수 없음
    2. N의 크기(n-gram)가 커질수록 훈련 데이터에서 해당 조합을 찾기 어려워져 희소 문제가 심해짐
2. 문맥 제한
    1. 앞선 N-1개의 단어만 고려하므로, 문장 앞부분에 나온 정보를 뒷부분 결정에 반영하지 못함

## Perplexity(PPL)
언어 모델이 문장을 얼마나 "혼란스러워하는지(perplexed)"를 수치로 나타낸 것
- 문장 W가 희소하게 등장할수록 PPL이 커지고, 단어 개수 N이 커질수록 PPL이 낮아짐
- 문장 W가 희소하게 등장하면 모델이 훈련 데이터에서 더 적게 봤으므로, 예측할 때 더 혼란스럽기 때문에 W가 희소하게 등장할수록 PPL이 커짐
- 단어 개수 N이 커지면 함께 보는 단어들이 더 많아짐 -> 희소한 부분이 있더라도 그 안에 예측이 쉬운 단어가 같이 있어 PPL을 낮춰줌

- → PPL이 낮다 = 모델이 예측에 덜 혼란스러움 = 모델이 문맥을 잘 이해함 = 모델의 성능이 좋음
--- 
- **신경망 기반 언어 모델**: 단어와 문맥 간의 관계를 학습하여 일반화하는 능력이 뛰어남
- 훈련 데이터에서 직접 보지 못한 문장이 등장하더라도 훨씬 유연하게 대처 가능

## 신경망 언어 모델 (NLM)
통계적 언어 모델의 한계를 극복하기 위해 신경망 언어 모델이 등장
### 임베딩


- **워드 임베딩(Word Embedding)**
- 신경망 모델(NLM)은 모든 단어를 거대한 다차원 지도 위의 한 점(좌표)으로 보냄
    - **통계 모델:** '축구'와 '농구'는 그냥 완전히 다른 이름
    - **신경망 모델:** '축구'와 '농구'를 지도 위에 찍어보니 아주 가까운 곳에 위치   
        -  둘 다 '운동’이고, '구기 종목'이고, '단체 스포츠’라는 특징을 공유하기 때문

- 통계 모델은 훈련 데이터에 "재밌는 축구를 한다"라는 문장이 없으면 확률을 0으로 처리

### NNLM 
1. 학습 데이터에서 "재밌는 농구를 한다"를 학습
2. "축구"와 "농구"가 유사한 벡터임을 인지
3. "재밌는 축구를 한다"라는 문장을 처음 봐도 높은 확률을 부여
> **핵심 포인트:** 
- NNLM은 단어를 '기호'에서 '의미를 가진 숫자(벡터)'로 바꿈으로써, 데이터에 없는 문맥도 어느 정도 예측할 수 있는 **일반화 능력**을 얻음


# 딥러닝 기반 언어모델
## BERT
Bidirectional Encoder Representations from Transformers
- **Bidirectional :** 텍스트의 앞뒤 문맥을 동시에 고려해서
- **Encoder :** 입력값을 벡터로 바꿔주는
- **Transformers :** 딥러닝 모델
- -> 한 단어를 이해할 때 그 단어의 좌우 문맥을 동시에 보고 이해
- -> encoder 기반

> ### BERT의 특징
- 양방향: 문장을 읽을 때 앞뒤를 동시에 고려
- → 한 단어를 이해할 때 그 단어의 좌우 문맥을 동시에 보고 이해
- 트랜스포머 기반: 트랜스포머 구조의 **encoder** 부분만 사용
- 사전 학습 + 파인 튜닝: 큰 데이터로 먼저 일반적인 언어 능력을 배우고, 나중에 특정 과제에 맞게 살짝 조정
    - 이때 사전학습은 **Masked Language Model(MLM)**과 **Next Sentence Prediction(NSP)**를 통해 진행


- BERT의 기본 구조
- transformer의 encoder를 여러 개 쌓아 올린 구조


> ### 처리과정
**input → Embedding → 트랜스포머 인코더 12개(or 24개) → Output**
- BERT는 입력 문장을 embedding한 뒤, 여러 개의 encoder를 통과하면서 각 단어의 의미를 문맥적으로 풍부하게 만듦
- BERT는 encoder만 있기 때문에 문장 이해를 목적으로 함

> ### 사전 학습
1. Masked Language Model(MLM)


- 빈 칸 채우기 방식으로 학습
- → 자연스럽게 문맥을 이해하고 추론하는 능력 향상

2. Next Sentence Prediction(NSP)
- 두 문장에 대해, 이 문장들이 서로 앞뒤 관계인지, 아니면 그냥 무작위 위치에 있는 두 개의 문장인지를 학습
- 이를 통해 **두 문장이 이어지는 논리 흐름** 학습 가능

## GPT
- GPT는 decoder 기반
- decoder는 미래의 단어를 볼 수 없게 mask로 가림
- 그래서 GPT 또한 지금까지의 단어를 바탕으로 다음 단어를 예측
- 즉, 왼쪽의 단어들”만” 보고 오른쪽에 올 단어를 예측

## LLM
'Large Language Model'의 줄임말로, 우리말로는 '거대 언어 모델'
- 방대한 양의 데이터를 학습하여 자연어 및 기타 유형의 콘텐츠를 이해하고 생성하여 광범위한 작업을 수행할 수 있는 기초 모델의 범주
- ex) ChatGPT,LLaMA, PALM, Bard, Gemini 등

> ### LLM의 작동 방식
- LLM은 크게 '학습'과 '예측'이라는 두 가지 핵심 단계를 통해 작동
1. 대규모 학습 (Training)
- LLM은 인터넷의 웹페이지, 책, 기사 등 수십억 개의 문장으로 이루어진 방대한 데이터를 학습
-  이 과정에서 단순히 정보를 암기하는 것을 넘어, 단어와 단어 사이의 관계, 문맥적 의미, 문법 구조, 그리고 세상의 지식까지 언어의 패턴을 스스로 터득

2. 예측 및 생성 (Prediction & Generation)
- 학습이 완료된 LLM은 사용자의 질문이나 지시(프롬프트)를 받으면, 학습한 패턴을 기반으로 다음에 올 가장 확률 높은 단어를 순서대로 예측하며 응답을 생성

> ### LLM의 한계
- **hallucination** → GPT에서 발생했던 유명한 문제였지만 지금은 해결됨. 환각처럼 그럴 듯하지만 틀린 정보를 말 하는 것을 의미
- **업데이트 비용** → LLM은 파라미터 수가 엄청나게 많기 때문에 새로운 정보를 반영할 때 계산량이 많음

## RAG
1. RAG의 필요성
- 외부에서 최신 정보를 가져와서 답변하게 하기 위한 것
- 질문에 대해 모델이 모르면 검색해서 답하는 구조
2. RAG의 구조


- **질의 인코더**(Query encoder): 사용자의 질문을 벡터로 바꿈
- **지식 검색기**(Knowledge Retriever): 인코딩된 질문을 바탕으로 외부 지식에서 관련 내용을 검색
- **지식 증강 생성기**(Knowledge Augmented Generator): 검색한 내용을 참고해서 답변을 생성

3. RAG의 장점
- 최신 정보를 검색해서 가져올 수 있으니 “**최신 정보 반영 가능**”이 가장 큰 장점
- 모르는 것을 검색해서 지식을 보완할 수 있으므로 몰라서 답을 이상하게 하는 **hallucination 문제도 감소**
- 이에 따라 자연스럽게 **정보 정확성도 증가**

## LangChain
1. LangChain
- LLM+RAG를 직접 구현하는 것은 복잡하니까, 이러한 모델 pipeline을 쉽게 구성할 수 있도록 도와주는 역할을 하는 것이 **LangChain 프레임워크**
- LLM을 기반으로 앱 제작 시 필요한 구성 요소를 쉽게 연결
- 파이썬과 JavaScript 라이브러리 제공 → 다양하게 활용 가능
- → LangChain: LLM 앱을 빠르게 조립할 수 있는 레고 블럭

> ### LangChain의 특징
- 추상화: LLM 서비스를 만들 때 필요한 각종 작업을 간결하게 표현(간소화)
- 표준화: 비슷한 기능을 갖추고 있는 요소들을 똑같은 형식을 갖춘 컴포넌트로 표준화
- 체이닝: 자주 활용하는 주요 컴포넌트를 쉽게 연결해서 LLM 서비스의 로직을 쉽게 파악 (Input → 컴포넌트 → Output)

2. LangGraph
> ### LangChain의 한계


- LangChain은 중간에 조건에 따라 다른 작업을 수행하거나, 결과가 마음에 들지 않을 때 이전 단계로 돌아가 다시 실행하는 등의 복잡한 로직을 구현하기 어려움
- 이러한 한계를 극복하기 위해 등장한 것이 LangGraph
### LangGraph
- LangGraph는 비선형적인 작업 형태로, 사이클 (Cycle), 루프 (Loop)를 만들 수 있음
-  에이전트는 보통 한 번에 완벽한 답을 찾지 못하기 때문에, 사이클을 통해 검색과 검증을 반복하며 완벽한 답을 찾아냄
- **노드 (Node)**: 특정 작업을 수행하는 단위로, 랭체인의 컴포넌트를 그대로 가져와 노드로 사용 할 수 있음  →  **‘무엇을 할 것인가’**
    
    ex) Tool - 웹 검색, 계산기 등 / Agent - 추론+Tool 사용
    
- **엣지 (Edge)**: 노드와 노드를 연결하는 선으로, 어떤 노드에서 다음 노드로 이동할지를 결정
    - 특히 ‘**조건부 엣지**(Conditional Edge)’를 통해 "만약 A 조건이면 B 노드로, C 조건이면 D 노드로 가라" 와 같은 **분기 로직**(if-else문)을 쉽게 구현 가능 → **‘다음에 어디로 갈 것인가’**



# sLM (small Language Model)
- 소규모 언어 모델(sLM)은 자연어 콘텐츠를 처리, 이해 및 생성할 수 있는 AI 모델
-  대규모 언어 모델(LLM)보다 규모와 범위가 작음
- sLM은 수백만~수십억 개 수준의 매개변수를 가지며, LLM은 수천억~수조 개의 매개변수를 가짐
- 매개변수는 모델이 학습을 통해 얻는 내부 값으로, 모델의 동작과 수행 방식에 영향을 줌
- sLM은 LLM보다 더 작고 효율적이어서 메모리와 연산 자원을 적게 사용
- 모바일 앱처럼 리소스가 제한된 환경, 또는 데이터 네트워크 없이 사용자 쿼리에 대한 오프라인 추론을 통해 응답을 생성해야하는 상황에 적합

### 모델 압축 
1. 가지치기
- 가지치기는 신경망에서 중요도가 더 낮거나 중복되거나 불필요한 매개변수를 제거
- 일반적으로 가지치기되는 매개변수에는 뉴런 간의 연결에 해당하는 숫자 가중치(이 경우 가중치는 0으로 설정됨), 뉴런 자체 또는 신경망의 계층이 포함

2. 양자화
- 양자화는 고정밀 데이터를 저정밀 데이터로 변환
- 양자화는 계산 부하를 줄이고 추론 속도를 높일 수 있음

3. 저차원 분해
- 큰 가중치 행렬을 더 작은 저차원 행렬로 분해하는 것
- 더 간결한 근사치를 사용하면 매개변수 수가 줄어들고 계산 횟수가 줄어들며, 복잡한 행렬 연산을 단순화 가능

4. 지식 증류
- 지식 증류는 사전 학습된 "교사 모델"의 학습 내용을 "학생 모델"로 이전하는 것


- 학생 모델은 교사 모델의 예측과 일치시키고 기본 추론 과정을 모방하도록 학습
- 이처럼 더 큰 모델의 지식이 더 작은 모델로 기본적으로 '증류'됨

### sLM의 장점
- 성능적 측면
    - **효과적인 성능**: 작은 모델을 해당 업무에 맞게 fine-tuning하면 필요한 기능만 빠르고 안정적으로 수행 가능
    - **지연 시간 감소**: 모델 크기가 작기 때문에 계산량 ↓
    - **보안:** 작은 모델이라서 기업 내부 서버나 개인 기기에서 직접 실행하기가 상대적으로 쉬움 → 민감한 데이터를 외부 클라우드 API로 보내지 않아도 됨
- 비용적 측면
    - **접근성**: sLM은 고성능 GPU 서버가 없어도 사용 가능→ 진입 장벽 ↓
    - **효율성**: 저장 공간, 메모리, 연산 자원을 적게 사용 → 같은 작업을 처리할 때 더 적은 자원으로 결과 도출
    - **에너지 효율**: 계산량이 적으므로 전력 소비 ↓
    - **비용 절감**: 모델이 작기 때문에 추론 비용, 서버 비용, GPU 비용, API 호출 비용 ↓

### sLM의 한계
- **편향**: 더 큰 모델에 존재하는 편향까지 함께 전달 가능 → 성능 저하로 이어질 수 있음
- **제한된 일반화 & 복잡한 작업에서의 성능 저하**: 특정 작업에 대해 미세 조정하기 때문에 포괄적인 지식이 필요한 작업에 취약
- **할루시네이션(환각)**: 정보가 적고 추론 능력이 낮으므로 당연히 발생 가능

### LLM과 sLM의 결합 (결합 추론)


- 복잡하고 어려운 작업에는 LLM, 가볍고 간단한 작업에는 SLM을 적절히 배분해서 사용
- 이를 통해 LLM 수준의 높은 정확도를 달성하면서도 LLM 토큰 사용량을 최대 90%까지 절약 가능
