# 생성 모델  <br>
## 1. 생성 모델 개념 <br>
#### 분류 모델과 생성 모델 차이 <br>
- 생성 모델 = 학습 데이터를 학습해 분포를 따르는 유사한 데이터 생성하는 모델       
    - 데이터 자체의 확률 분포를 모델링하려는 학습 목표    
    - 이상한 입력에 대해 판단해 거부하는 능력     

- 분류 모델 = 입력된 데이터가 어떤 레이블에 속할지 결정하는 모델     
    - 데이터 X가 주어졌을 때 정답이 Y일 확률 계산하려는 목표     
    - 본 적 없는 데이터가 들어와도 어떻게든 기존에 배운 라벨로 결론 내림    


#### 생성 모델 분류 <br>
확률 밀도를 직접 계산할 수 있는지에 따른 분류      
(1) 명시적 확률 밀도 모델 = 데이터 확률 분포 값을 직접 수치화해 출력 가능한 모델         
(2) 암시적 확률 밀도 모델 = 데이터 확률 분포를 따르는 데이터를 생성할 수 있는 모델           
    - 직접 샘플링       
    - 간접/반복 샘플링     

## 2. AE, VAE <br>
#### AE (Auto Encoder) <br>
= 입력과 동일한 출력을 만들고자 하는 신경망      
- 차원 축소, 특징 추출, 노이즈 제거, 이상 탐지 등 데이터 복원 / 특성 학습에 많이 사용     
(1) 압축 (인코더): 입력 데이터를 아주 작은 바늘 구멍으로 통과시키고자 핵심만 남기고 깎아냄    
(2) 병목 통과: 가장 중요한 정보만 남은 잠재 벡터         
(3) 복원 (디코더): 핵심 정보만을 단서로 원래의 커다란 이미지로 복원      

<br>

- 잠재 공간을 <u>**특정 점**</u>으로만 데이터를 보내므로, 학습 시 사용하지 않은 곳에 점을 찍으면 디코더는 엉뚱한 결과 출력    

=> 차원 축소 과정에서 정보의 중요도를 고민하며 본질 학습 가능     

#### VAE (Variational AE) <br>
= 데이터의 잠재 구조 학습해 원본과 닮은 새로운 데이터 생성하는 모델    
- AE와 달리 <u>**잠재 벡터를 가우시안 분포**</u>로 학습해, 어떤 점을 뽑아도 그럴 듯한 이미지 생성    
- 이미지 생성, 텍스트 생성, 신호 처리, 이미지 보간 등에 주로 사용     

(1) 분포 예측 (인코더): 입력 데이터가 존재할 확률이 높은 구역으로 매핑     
(2) 샘플링 및 재파라미터화: 예측한 분포에서 실제 디코더에 전달한 잠재 벡터를 하나 뽑음
(3) 생성 (디코더): 샘플링된 잠재 벡터를 단서로 새로운 데이터 생성    


## 3. GAN (Generative Adversarial Network) <br>
#### 개념 <br>
= 생성적 적대 신경망 = 생성자 신경망과 판별자 신경망이 경쟁하며 훈련을 통해 작업을 점점 더 정교하게 수행하는 신경망 모델     
    - 생성자: 진짜 분포에 가까운 가짜 분포 생성        
    - 판별자: 표본이 가짜/진짜 분포에 속하는지 결정     

=> 실제 데이터 분포에 가까운 데이터 생성         

#### 구조 <br>
- 생성자: 제어할 수 있는 무작위 노이즈인 잠재 변수를 입력 받아 이와 유사한 가짜 데이터 생성해, 판별자가 가짜를 진짜로 믿게 함     
- 판별자: 실제 데이터와 생성자가 만든 가짜 데이터를 모두 입력 받아 진짜일 확률 출력 [진짜는 1, 아니면 0]      

=> 최소-최대 게임을 두 신경망이 수행하는 구조     

#### 장단점 <br>

(+) <br>
- 진짜 같은 가짜 생성 가능      
- 판별자 통과하며 선명하고 깨끗한 샘플 생성 가능    
- 생성 속도 빠름    
- 잠재 공간에서 생성해 이미지가 부드럽게 변하는 보간 가능    

(-) <br>
- 학습 불안정성     
- 훈련 초기 문제: 그래디언트 소멸    
- 모드 붕괴: 생성자가 판별자를 속이기 쉬운 특정 패턴 하나만 만드는 문제    
- 지표 부재: 학습이 잘 되는지 보여주는 손실 곡선이 없음    
- 역매핑 불가능      

#### 적용 사례 <br>
- 엔비디아 가짜 이미지 생성, 워싱턴대 가짜 오바마 연설 영상, 페이스북 눈 감은 사진을 뜬 사진으로

## 4. 확산 모델 <br>
#### 확산 모델 개념 <br>
= 입력 이미지에 노이즈를 조금씩 추가해 완전한 노이즈로 만들어 거꾸로 복원하는 과정을 학습해 입력 이미지와 유사한 확률 분포를 가진 결과 이미지 생성하는 모델    

=> 순확산, 역확산을 거쳐 만든 결과 이미지의 확률 분포를 원래 데이터의 확률 분포와 최대한 유사하게 만듦     

#### 구조 <br>

(1) 순확산: 데이터에 점진적으로 노이즈 추가하는 과정을 거쳐, 원래 데이터를 점차 무작위 노이즈로 변형하고자 함      
(2) 역확산: 노이즈 데이터에서 원본 데이터를 재구성하며, 순확산 과정에서 추가된 노이즈를 제거해 원래 데이터 복원        

#### 장단점 <br>
(+) <br>
- 정교하고 다양한 이미지 생성하며 고해상도 확장에 유리     
- 손실 곡선 확인할 수 있어 학습 안정성 높음     
- 다양성 확보    

(-) <br>
- 반복 계산으로 느린 생성 속도     
- 고해상도 데이터 처리를 위한 컴퓨팅 자원과 계산 비용     
- 노이즈 분포와 데이터의 형태가 동일해야 한다는 구조적 제약    

# AI 최적화 <br>
## 1. 전이 학습과 딥러닝 경량화 <br>
#### 개념 <br>
= 데이터/문제 유형이 달라도 배운 지식을 가져와 새로운 상황에서 성능을 높임 (도메인, 태스크 중 하나라도 다를 때)    
    - 도메인: 어떤 데이터를 다루는지    
    - 태스크: 데이터로 무엇을 예측하는지     
    - 소스: 학습 완료된 모델이 있는 곳     
    - 타겟: 실제로 풀고 싶은 문제     
= 소스 도메인과 타겟 도메인이 다르거나, 소스 태스크와 타겟 태스트가 다를 때, 소스 도메인과 소스 태스크에서 얻은 지식으로 타겟 도메인 위의 타겟 태스크 성능 향상        
- 딥러닝 한계: 데이터 부족, 계산 시간 및 비용 증가, 과적합 등 => 이미 학습된 내용을 가져다 써야 함     

#### 도메인과 태스크 상태에 따른 분류 <br>
(1) 귀납적 전이학습 = 소스와 타겟의 태스크가 다르고 타겟 도메인에 라벨 존재 <br>
(2) 트랜스듀서형 전이학습 = 소스와 타겟의 태스크가 같지만 도메인이 다름  <br>
(3) 비지도 전이학습 = 소스와 타겟의 태스크가 다르고 소스와 타겟 모두 라벨 없음 <br>

#### 전이 대상에 따른 분류 <br>
(1) 인스턴스 = 소스 데이터 중 쓸 만한 샘플 골라 가중치 부여 [데이터 레벨 전이] <br>
(2) 특징 = 소스와 타겟의 공통 특징을 추출하는 표현 방식 전이 [표현 레벨 전이] <br>
(3) 파라미터 = 모델 가중치/하이퍼파라미터 초기값 -> 현재 딥러닝에서 주류 <br>
(4) 관계 = 데이터 간 관계/구조 전이 <br>

#### Pre-training & Fine-tuning <br>
- 트랜스포머 이전에는 주로 CNN의 ImageNet 가중치 가져오는 수준이었으나 트랜스포머 이후 사전 학습된 거대 모델 등장으로, 이를 용도에 맞게 조정하는 방법에 대한 고민이 생김 <br>

#### 파인튜닝 한계 극복 <br>
(1) PEFT <br>
= 최소한의 파라미터만 건드려 최대의 성능 내기     
= 기존 모델 파라미터 고정하되 별도의 작은 파라미터만 추가 학습     

(2) Zero-Shot & Few-Shot <br>
= 학습 데이터가 없어도 언어 이해 능력만을 기반으로 바로 추론 & 극소수 예시만으로 새로운 태스크 수행    

#### 전이학습 한계 <br>
- Negative Transfer: 소스와 타겟의 도메인이 너무 달라 전이가 성능을 떨어트릴 ㅅ 있음     
- Catastrophic Forgetting: 파인 튜닝 시 새 데이터에 너무 치우치면 모델이 원래 가지던 일반 능력 잃음   
- Bias Transfer: 사전 학습 데이터에 담긴 편향이 그대로 담기는 것     
- Transferability 측정 어려움: 어떤 태스크가 얼마나 관련되는지 사전에 정량 측정 어려움     

#### 사용 사례 <br>
- 컴퓨터 비전: 의료 영상 분석, 자율 주행    
- NLP: 법률 문서 분석, 기업 특화 챗봇    
- 의료 및 신약: 단백질 구조 예측   
- 제조: 공장 불량 검출     

#### 딥러닝 경량화 <br>
= 대규모/복잡한 AI 모델 가볍게 변환하는 기술과 방법      
- 가지치기: 중요도가 낮은 파라미터 제거     
- 양자화: 모델 가중치를 표현하는 숫자의 정밀도 줄임     
- 지식 증류: 이미 학습된 모델을 작고 가벼운 모델에 전달     
- 모델 구조 경량화: 모델 설계부터 가볍게 설계된 모델 사용      
- 하드웨어 가속화: 연산 빠르게 수행하는 전용 h/w 활용     

## 2. AI Alignment <br>
#### 개념 <br>
= AI가 인간 의도, 목표, 윤리에 부합하도록 행동하도록 하는 연구 분야      

(1) Outer Alignment: 목표를 제대로 설정했는지    
(2) Inner Alignment: 모델 내부적으로 다른 목표를 설정했는지 확인 = 해석 가능성     

- 가치 정렬 모호성: 어떤 가치?      
- 확장 가능한 감독: 인간이 이해하지 못하는 답변에 대한 피드백 방식?     

#### RHLF <br>
= 인간 피드백 기반 강화 학습        
= 어떤 답변을 선호하는지 학습하며 답을 잘 하도록       
- SFT: 사전 학습 모델이 인간 지시 따르는 방법을 배우는 과정               
- RM: 인간 선호도에 따른 채점기 만드는 과정      
- RL: 보상 모델 기준으로 답변에 점수 매겨 파라미터 업데이트 = 모델 최적화       

=> 사람에게 더 유용하고 적절한 답 생성하도록 유도    

## 3. 프롬프트 엔지니어링 <br>
#### 개념 <br>
= AI가 더 좋은 답변 내도록 입력 설계하는 기술       
- 파인튜닝보다 비용이 낮고 속도가 빠르지만 성능 한계 있음    

#### 주요 기법 <br>
(1) Zero-shot / Few-shot 프롬프팅: 예시 없이 / 몇 개만 보여주고 질문      
(2) Chain of Thought: 풀이 과정을 단계별로 생각하게 만듦      
(3) 역할 부여     
(4) Instruction 프롬프팅: 최대한 구체적으로 지시 (출력 형식, 길이, 톤)     
(5) Self-Consistency: 같은 질문을 여러 번 해 가장 많이 나온 답 채택     
(6) RAG = 검색 증강 생성: 최신 정보/도메인 지식을 외부 DB에서 검색해 프롬프트에 함께 넣음    

#### 한계 <br>
- 프롬프트 주입 공격: 프롬프트를 조작해 AI를 엉뚱하게 동작시킴    
- 환각: 없는 정보를 그럴 듯하게 지어내는 현상 완전히 막기 어려움     
- 모델 의존성: 모델마다 다른 프롬프트에 대한 반응     
- 성능 한계: 파인 튜닝 수준의 성능을 내기 어려움
