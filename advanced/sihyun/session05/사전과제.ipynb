# 분류 VS 생성
- SVM, 로지스틱 회귀 분석 등은 분류 모델
- 생성 모델은 학습 데이터를 학습해, 학습 데이터의 분포를 따르는 유사한 데이터를 생성하는 모델

## 상세
1. 분류 모델은 각 샘플에 정답 레이블이 존재
- 입력된 데이터가 어떤 레이블에 속할지 결정하는 모델
- 한계: 본 적 없는 이상한 데이터가 들어와도 모델은 어떻게든 기존에 배운 라벨 중 하나로 결론을 내려야 함

2. 생성 모델은 샘플에 레이블을 지정하는 데는 관심 X
- 데이터 자체의 확률분포를 학습하여 학습 데이터와 유사한 새로운 데이터를 만들어내는 모델
- 특징: 불합리하거나이상한 압력에 대해서는 "이건 세상에 존재할 확률이거의 0이야"라고 판단하며 거부할 수 있는 능력을 갖춤

### 생성 모델의 분류
1. 명시적 확률 밀도 모델
- 모델이 데이터의 확률 분포인 P(X)의 값을 직접적으로 수치화하여 출력할 수 잇는 모델
-- 수학적으로 확률 값을 정확하게 계산할 수 있음(정확히 계산 가능한 모델)
-- 확률 값을 정확히 구하기는 복잡해서 그 근삿값이나 하한선을 계산(근사적 밀도 모델)

2. 암시적 확률 밀도 모델
- 데이터의 확률 분포 P(X)가 무엇인지 수치로 알려주지는 않지만 그 분포를 따르는 데이터를 생성할 수 잇는 보델
- 직접 샘플링: 신경망을 한 번 통과시키는 것만으로 데이터를 바로 만들어냄
- 간접/반복적 샘플링: 여러 단계를 거쳐 점진적으로 데이터를 완성

## AE, VAE
1. AE
- 오토 인코더로 입력과 동일한 출력을 만드는 것을 목적으로 하는 신경망
- 순서
1. 압축: 방대한 정보를 가진 입력 데이터를 아주 작은 바늘구멍으로 통과시키기 위해 핵심만 남기고 깎아냄
2. 병목 통과: 중요한 정보만 남은 상태에서 좁은 구간을 잠재 백터로 넘김
3. 복원 단계: 바늘 구멍을 통과한 핵심 정보만을 단서로 삼아, 다시 원래의 커다란 이미지로 복원
- 한계: 데이터를 잠재 공간의 특정 점으로만 보냄, 학습할 때 쓰지 않은 빈공산의 점을 찍으면 디코더는 무엇을 그려야 할지 몰라 엉뚱한 결과를 출력

2. VAE
- 데이터의 잠재적인 구조를 학습하여, 원본과 닮았지만 새로운 데이터를 생성해내는 모델
- AE의 문제를 해결하기 위해 잠재 벡터를 점이 아닌 가우시안 분포로 학습
- 잠재 공간을 빈틈없이 촘촘하게 채워, 어떤 점을 뻡아도 그럴듯한 이미지를 생성
- 순서
1. 분포 예측단계: 입력 데이터를 특정 점이 아닌 데이터가 존재할 확률이 높은 '구역'으로 매핑
- 인코더는 '이 데이터는 잠재 공간 내에서 평균이 𝜇이고 퍼짐 정도가 𝜎 영역에 위치할 확류리 높다고 정의하는 역할
2. 샘플링 및 재파라미터화
- 예측한 분포에서 실제 디코더에 전달할 잠재 벡터를 하나 뽑는 단계
3. 생성 단계
- 샘플링된 잠재 벡터를 단서로 사용하여 새로운 데이터를 생성

3. GAN
- 생성자 신경망과 판별자 신경망이 서로 적대적으로 경쟁하면서, 훈련을 통하여 자신의 작업을 점점 더 정교하게 수행하는 신경망 모델
- 목적: 생성자는 진짜 분포와 가까운 가짜 분포를 생성하는 것, 판별자는 표본이 가짜 분포에 속하는지 진짜 분포에 속하는지를 결정하는 것

- 구조
1. 생성자
- 입력: 우리가 직접 제어할 수 있는 무작위 노이즈를 입력받음
- 출력: 실제 데이터와 유사한 가짜 데이터를 생성
- 목표: 판별자를 속여서 가짜를 진짜로 믿게 만드는 것
2. 판별자
- 입력: 실제 데이터와 생성자가 만든 가짜 데이터를 모두 입력받음
- 출력: 입력 데이터가 진짜일 확률을 출력
- 목표: 진짜는 1, 가짜는 0으로 정확히 분류하는 모델 수행

- 장단점
1. 장점
- 진짜 같은 가짜를 생성할 수 있음
- 압도적인 선명도: VAE가 수학적 평균을 취하며 흐릿한 이미지를 만드는 것과 달리 GAN은 판별자를 통과하여 매우 선명하고 깨끗한 샘플을 생성함
- 빠른 생성 속도: 확산 모델처럼 여러 단꼐를 거치지 않고, 신경망을 한 번만 통과하면 결과물이 바로 나옴
- 잠재 공간의 부드러움: 학습된 잠재 공간에서 주 덤 사이를 이동하며 생성하면, 이미지가 부드럽게 변하는 보간이 가능

2. 단점
- 학습의 불안정성 (진동현상, 끝없는 숨바꼭질)
- 훈련 초기 문제 (그래디언트 소멸)
- 모드 붕괴: 생성자가 판별자를 속이기 가장 쉬운 측정 패턴 하나만 주구장창 만드는 문제
- 지표의 부재: 객관적인 손실 곡선이 없음
- 역매핑 불가능: 잠재 변수 Z에서 데이터 X를 만드는 법은 배우지만, 이미지를 넣었을 때 특징 Z를 추출하는 과정이 기본 구조에는 없음

## 확산 모델
- 개념
1. 입력 이미지에 노이즈를 조금씩 추가하여 와넞ㄴ한 노이즈로 만든 뒤
2. 이를 다시 거꾸로 복원하는 과정을 학습 (노이즈 제거)
3. 입력 이미지와 유사한 확률 분포를 가진 결과 이미지를 생성하는 모델

- 목적
1. 목표: 순확산과 역확산 단계를 거쳐 생성된 결과 이미지의 확률 분포를 원래 데이터의 확률 분포와 최대한 유사하게 만드는 것
2. 모호성 모델링: 모호한 요청에 대해 모델은 완전한 노이즈 상태에서 시작해 데이터 구조의 아주 작은 부분을 환각하여 구체적인 유니콘 이미지를 완성해 나감

### 순확산
- 데이터에 점진적으로 노이즈를 추가하는 과정
- 원래 데이터를 점차적으로 무작위 노이즈로 변형시키는 것
1. 시작: 순확산 과정은 원본데이터로 시작
2. 노이즈 추가: 원본 데이터에 점진적으로 노이즈를 추가
3. 결과: 충분히 많은 단계가 지나면 원본 데이터는 완전한 무작위 노이즈로 변함

### 역확산
- 노이즈 데이터에서 원본 데이터를 재구성하는 과정
- 순확산 과정에서 추가된 노이즈를 제거하여 원래의 데이터를 복원하는 것
1. 시작: 역확산 과정은 완전한 노이즈 데이터에서 시작
2. 노이즈 제거: 각 단계에서 어떤 노이즈가 추가되었을지 예측하며 점진적으로 노이즈 제거
3. 결과: 수십번의 반복적인 디노이징 과정을 거쳐 점진적으로 이미지 복원


- 장점
1. 압도적인 샘플 품질
2. 학습의 안정성
3. 다양성 확보

- 단점
1. 느린 생성 속도
2. 높은 계산 비용
3. 구조적 제약

## 전이 학습
- 출발점 데이터랑 도착점 데이터가 다르거나
- 출발점 과제랑 도착점 과제가 다를 때
- 출발점에서 배운 걸 가져다가 도착점 문제를 더 잘 풀게 만드는 것

- 도메인과 테스크의 상태에 따른 구분
1. 귀납적 전이학습
- 소스와 타겟의 task가 다르고 타겟 도메인에 라벨이 존재할 때

2. 트랜스듀서형 전이학습
- 소스와 타겟의 task는 같지만 도메인은 다를 대

3. 비지도 전이학습
- 소스와 타겟의 task가 다르고 소스와 타겟 모두 라벨이 없을 때

- 무엇을 전이하느냐
1. 인스턴스: 소스 데이터 중 타겟에 도움이 될만한 샘플을 골라 가중치를 부여
2. 특징: 소스와 타겟의 공통적인 특징을 추출할 수 잇는 '표현 방식'을 전이
3. 파라미터: 모델의 가중치나 하이퍼파라미터 초기값을 그대로 가져옴 (가장 주류)
4. 관계: 데이터 간의 논리적 관계나 구조적 지식을 전이

## Pre-tuning & Fine-tuning
- 트랜스포머 이전에는 사전학습된 좋은 모델이 없었음
- 이후 엄청난 데이터로 사전 학습된 거대 모델들이 등장하기 시작했고 전이 학습의 질문자체가 바뀌게 됨

### PEFT
- 최소한의 파라미터만 건드려서 최대의 성능을 내자는 전략으로 원래 모델의 파라미터는 고정하고 별도의 작은 파라미터만 추가적으로 학습
-- 모델의 모든 계싼을 바꿔 기존 판단을완전히 지우는 게 아니라 작은 보정값만 더해서 최종 판단인 결과를 바꾸는 방식

### Zero-shot & Few-shot
- fine tuning 없이도 모델이 새로운 태스크를 수행할 수 있었음
- Zero shot은 학습데이터가 없어도 언어 이해 능력만을 기반으로 바로 추론을 수행하는 것
- Few shot은 극소수의 예시만으로 새로운 테스크를 수행하는 것 -> 모델의 파라미터를 업데이트하지 않는다는 점에서 기존 파인튜닝과 근본적으로 다음

- 한계
1. 소스와 타게스이 도메인이 너무 다르면 전이가 오히려 성능을 떨어트림
2. 파인튜닝 과정에서 새 데이터에 너무 치우치면 모델이 원래 가지고 잇던 일반 능력을 잃어버림
3. 사전 학습 데이터에 담긴 편향이 그대로 넘어옴
4. 어떤 두 테스크가 얼마나 관련되어 있는지를 사전에 정량적으로 측정하기 어려움

- 딥러닝 경량화
-- 대규모 및 복잡한 인공지능 모델을 가벼운 형태로 변환하는 기술과 방법임
1. 가지치기 
모델의 파라미터 중 중요도가 낮은 가중치를 제거하여 모델의 크기를 줄이고 연산속도를 높임

2. 양자화
모델의 가중치와 활성화 함수 값을 표현하는 비트 수를 줄여 연산량과 메모리 사용량을 절감하는 기술로 모델의 가중치를 표현하는 숫자의 정밀도를 줄이는 방법

3. 지식 증류
크고 복잡한 미리 학습된 지식을 작고 가벼운 모델에게 전달하는 방식

4. 모델 구조 경량화
모델 설계 단계부터 연산 효율성을 극대화하도록 구조를 최적화하는 방법으로 처음부터 가볍게 설계된 모델을 쓰는 방법

5. 하드웨어 가속화
특정 연산을 빠르게 수행할 수 잇는 전용 하드웨어를 활용하는 단계로 AI 연산에 트고하된 하드 웨어를 쓰는 방법

## AI Alignment
- 단순히 성능을 높이는 단계를 넘어 인간의 의도와 목표, 윤리적 가치에 부합하도록 행동하도록 만드는 연규
1. Outer Alignment: 목표를 제대로 설정했는가
- 모델이 보상을 극대롸하기 위해 인간의 의도와 무관한 꼼수 부리는 현상을 해결

2. Inner Alignment: 모델이 내면에서 무엇을 최적화하는가
- 모델의 실제 내부 동작을 해석 가능하게 만드는 해석 가능성 연구 

- 어렵게 만드는 요인
1. 가치 정렬의 모호성
-- 인간의 가치는 고정되어 있지 않고 사람마다 다름

2. 확장 가능한 감독
-- AI가 인간보다 똑똑해지면 인간이 AI의 답변이 맞는지 틀린지 판단 어렵기에 AI가 엄청 복잡한 코드를 짜면 맞는지 사람이 검토하기 어렵기에 어떻게 피드백 줄 것인지 고민해야 함 

### RHLF
- 문장은 잘 생성하나 인산이 원하는 답을 이해하지는 못하는 경우가 있었기에 인간의 피드백을 기반으로 모델을 강화학습으로 조정하는 방법임
- 사람의 피드백을 활용해 더 바람직한 응답을 생성하도록 강화학습으로 미세 조정하는 방식

## 프롬프트 엔지니어링
- AI가 더 좋은 답변을 내도록 입력을 설계하는 기술임
- 모델은 그대로 두고 입력만 잘 설계해서 원하는 답을 끌어내는 것으로 비용이 적고 누구나 쉽게 시도할 수 잇음

- 주요 기법
1. Zero-shot: 예시없이 바로 질문
2. Few shot: 예시를 몇 개보여주고 질문
3. CoT: 생각의 사슬로 풀이 과정을 단계별로 생각하게 만드는 기법
4. 역할부여: 특정 역할을 맡겨서 그 역할에 맞는 답변을 끌어내는 기법
5. 명확한 지시: 구체적으로 지시
6. 여러번 물어봐서 가장 많이 나온 답을 채택하는 기법
7. RAG: 모르는 최신 정보나 특정 도메인 지식을 외부 데이터 베이스에서 검색해 프롬프트에 함께 넣어주는 방식

-한계
1. 프롬프트 주입 공격: 악의적인 사용자가 프롬프트 조작해서 AI를 엉뚱하게 동작시키는 보안 문제
2. 할루시네이션: 그럴듯한 거짓말로 RAG를 활용하면 어느 정도 완화할 수 있지만 완전한 해결책 X
3. 모델 의존성: 모델이 업데이트되면 기존 프롬프트가 갑자기 안 먹힐 수 잇음
4. 성능한계: 특화된 도메인이나 높은 정확도가 필요한 작업에서는 파인튜닝이 필요할 수 잇음
