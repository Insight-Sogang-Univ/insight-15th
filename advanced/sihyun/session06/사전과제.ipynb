# 추천 시스템
- 사용자의 정보 제이터를 분석, 개인의 취향에 맞는 아이템을 추천하는 알고리즘

## 사용되는 데이터의 종류와 특징
1. 명시적 데이터
- 고객들에 의해 입력된 데이터, 사용자가 선호도를 직접적으로 표현한 데이터
- 명백한 데이터를 얻기 어려움
- 사람의 취향은 항상 변함

2. 암시적 데이터
- 고객들의 행동을 기반으로 수집된 데이터, 사용자의 선호도가 간접적으로 표현된 데이터
- 사용자가 남기는 발자취이나 높은 수치가 높은 선호도를 의미하지는 않음


1. 콘텐츠 기반 필터링
- 콘텐츠의 유사도를 기반으로 사용자에게 추천하는 방법
- 사용자가 이용하거나 선택한 아이템과 유사한 아이템을 추천하는 방식

2. 협업 필터링
- 서비스 내에 있는 다수의 사용자로부터 얻은 선호도 정보를 기반으로 사용자에게 추천하는 방법
-- 사용자 기반
-- 아이템 기반

3. 하이브리드 필터링
- 2가지 이상의 다양한 종류의 추천 시스템 알고리즘을 조합하여 만들어진 알고리즘
- 콘텐츠 기반과 협업 필터링 각각의 단점을 보완하기 위해 제시된 방식

### 추천 시스템에 사용되는 데이터
1. 직접 지정형: 사용자에게 자신이 선호하는 아이템의 특징을 명시적으로 지정하도록 하는 방법
2. 간접 지정형: 사용자의 과거 행동 이력에 기반하여 사용자의 프로필을 작성하는 방법

-- 유사도: 추천 시스템을 이해하는 중요한 수학적 개념
1. 영상, 이미지, 텍스트 등의 형태로 존재하는 아이템 정보 데이터를 벡터화 후 아이템 벡터 간의 유사도를 측정
2. 측정된 유사도를 바탕으로 고객이 선호하는 아이템과 비교하여 유사도가 높은 아이템을 선별

### TF-IDF
- 특정 문서 내에 특정 단어가 얼마나 자주 등장하는 지를 의미하는 단어 빈도와 전체 문서에서 특정 단어가 얼마나 자주 등장하는지를 의미하는 IDF를 통해 다른 문서에서는 등장하지 않지만 특정 문서에서만 자주 등장하는 단어를 찾아 문서 내 단어의 가중치 계싼

계산: $ TF(d,t) \times IDF(d,t)$
장점
- 아이템이 가지고 있는 피처 자체에만 집중하기에 다른 유저의 평점 등에 영향을 받지 않음
- 아이템의 피처를 통해서 추천하기에 추천이 된 이유를 설명하기 용이

단점
- 아이템의 피처를 추출해야 하고 이를 통해서 추천하기 때문에 정확한 피처를 추출하지 못하면 정확도가 떨어짐
- 기존의 아이템과 유사한 아이템 위주로만 추천하기에 새로운 장르의 아이템을 추천하기 어려움

## 협업 필터링
- 과거의 비슷한 선호도를 가진 사용자들은 앞으로도 비슷한 선호도를 가질 가능성이 높다

1. 메모리 기반 협업 필터링: 사용자 또는 아이템 간 유사도를 기반으로 비슷한 취향을 가진 사용자가 선호하는 아이템을 추천
- 한계
1. 희소 행렬 문제: 주어진 평점/선호도 데이터가 적을 경우 성능이 저하됨
2. 확장성 문제: 아이템/유저가 계속 늘어날 경우 확장성이 떨어짐
3. 콜드 스타트 문제: 기존의 평점, 행동 데이터를 기반으로 유사도를 계산하는데 데이터가 아예 없는 경우 추천이 불가능

- 해결방법
1. 직접 지정형 프로필 활용: 가입시 선호 장르/아이템 직접 선택
2. 콘텐츠 베이스 필터링과 혼합하여 아이템 특성 기반으로 보완
3. 인구 통계학적 정보 활용

- 사용자 기반: 당신과 비슷한 고객들이 다음 상품도 구매했다~
-- 유저 간의 유사도가 높을수록 가중치를 부여하는 방식

- 아이템 기반: 이 상품을 선택한 다른 고객들은 다음 상품도 구매했다
-- 추천하고자 하는 타겟 아이템과 유사한 아이템의 평점을 활용

(유저 베이스 VS 아이템 베이스)
1. 유저 베이스
- 추천 정확성 & 효율성: 아이템 수가 사용자 수보다 많은 환경에서 유리
- 안정성: 아이템이 수시로 바뀌는 환경에 적합
- 다양성: 특정 사용자가 독특한 아이템을 평가한 경우, 뜻밖의 독창적인 아이템 추천 가능
- 실제 활용: 아이템이 자주 바뀌고 사용자 수가 적은 환경

2. 아이템 베이스
- 추천 정확성 & 효율성: 사용자 수가 아이템 수보다 많은 환경에서유리
- 안정성: 아이템 리스트 벼노하가 적고 고정적인 환경에서 적합
- 다양성: 유사한 아이템 위주로 추천되어 다양성과 참신성은 부족
- 계산 속도: 사용자별 개별 계싼이 불필요 해 빠르고 안정적
- 실제 활용: 사용자 유입이 많은 대형 플랫폼에 유리

## 모델 기반 협업 필터링
- 단순히 유사도를 계산하는 것이 아니라 데이터로부터 패턴을 학습한 모델을 만들어 평점을 예측
-> 관측된 평점 데이터를 학습하여 관측되지 않은 평점을 예측하는 방식

(잠재 요인)
- 사용자와 아이템 사이의 관계를 설명해주는, 눈에 보이지 않는 숨겨진 특성

Latent Factor Model: 차원이 무엇을 의미하는지 몇개의 차원이 존재하는지 알 수 없음

- 주어진 데이터로는 알수 없는 사용자 - 아이템 간의 숨겨진 특성을 찾기 위해 주어진 Matrix를 사용자-잠재요인, 아이템-잠재요인으로 각각 분해하여 학습하는 모델
- 차워느이 수는 모델이 알아서 결정하며 평점 패턴을 분석해서 유저와 아이템의 숨겨진 성향을 수치로 추정

- factor model: 유저와 아이템을 구분없이 동일한 차원의 벡터 공간에 좌표로 표현하는 방식



## 행렬 분해
- 유저 아이템 행렬에는 유저의 취향 + 아이템의 특성이 녹아 있으나 결과값만 볼 수 잇음
- 이 행렬을 분해하여 평점을 유저의 잠재 요인과 아이템의 잠재 요인의 내적으로 설명할 수 잇으면 이로써 평점을 예측 가능

-- 입력 행렬(R): 노래/유저
- 사용자가 모든 노래를 듣거나 평점 주지 않았을 것이기에 R은 희소 행렬 가능성 높은

-- $X^T$: 사용자의 잠재 요인 행렬
-- $Y$: 아이템의 잠재 요인 행렬

- MF 수행 방식
1. SVD: 수학적 분해 기법
2. SGD: 최적화 알고리즘
3. NMF: 제약 조건 기반 모델
4. PMF: 확률적 모델

## 딥러닝 기반 추천 시스템
### VAE 기반 협업 필터링
사용자의 아이템 상호작용 벡터를 입력 받아 잠재 공간에서 재구성하는 방식
- 인코더: 사용자 이력을 잠재변수 z 분포로 매핑
- 디코더: z로부터 아이템 분포를 복원

- 장점:
1. 기존 MF처럼 단일 포인트 벡터가 아니라 확률적 잠재 분포로 사용자 선호를 모델링하여 불확실성 반영
2. 희소한 상호작용 데이터에서도 강건
3. 비선형성 학습

- 한계:
1. 선형성: 평점 예측이 단순한 잠재 벡터와 내적으로만 이루어짐
2. 불확실성 무시: 사용자의 취향을 단 하나의 벡터값으로만 표현

- VAE는 취향의 불확실성까지 모델링에 포함시켜 안정성을 높임

- 작동 구조:
1. 입력: 사용자의 아이템 상호작용 벡터
2. 인코더: 사용자 이력을 잠재변수 z 분포로 매핑
3. 잠재벡터: 샘플링된 최종 취향 벡터
4. 디코더: z로부터 아이템 분포를 복원

- 공식:
\mathcal{L}(x; \theta, \phi) = \underbrace{E_{q_\phi(z|x)} [ \log p_\theta(x|z) ]}_{\text{재구성 오차 최소화 (Reconstruction Loss)}} - \underbrace{D_{\text{KL}}(q_\phi(z|x) || p(z))}_{\text{잠재 공간 정규화 (KL Divergence)}}

1. 재구성 오차 최소화: 디코더가 잠재 벡터를 활용해서 원래 입력 x를 ㅗ치대한 비슷하게 복원
2. KL: 인코더가 출력하는 잠재분포와 사전분포가 비슷하게 강제해서 모델이 특정 데이터에 과적합하여 잠재 공간을 불규칙하게 만드는 것을 방지

### GNN 기반 사용자 선호 예측
: 관계나 순서라는 풍부한 정보를 활용
- 그래프를 활용하는 이유
1. 관계, 상호작용과 같이 다소 추상적인 개념을 다루기에 적합
2. 비유클리디안 공간도 표현하고 학습 가능

- 프로세스


- GNN은 메시지 패싱을 수행하는데 이때, 각 노드는 자신의 이웃 노드들의 정보를 수집해서 자신의 은닉 상태를 업데이트함
- K번 반복하면서 더 멀리 있는 이웃의 정보까지 반영

1. 그래프 구조 생성: 노드와 엣지를 정의 후 각각의 피처를 담아 표현하는 것이 첫번째 단계
2. 메시지 패싱: 그래프에서 노드 간 정보를 전달하고 상호작용하는 방법
- 보통 하나의 GNN 레이어에서 '메시지 생성- 집계 - 갱신' 으로 이루어짐

-> 정규화된 이웃 임베딩의 합만으로 구성되어 CF의 본질에 충실하게 그래프 구조와 전파만 남김

##### GNN 장단점
- 장점:
1. 고차원 협업 시그널 포착: 나와 취향이 비슷한 사용자들과 내가 본 아이테모가 연관된 다른 아이템들의 정보를 직접 반복하여 반영
2. 정형화되지 않은 데이터 학습 가능: GNN은 관걔를 학습하기에 각 노드가 정형화된 데이터로 정의되지 않아도 관계망 속에서 정보를 효과적으로 전파
3. 콜드 스타트 문제 와화에 효과적

- 단점:
1. 과평탄화 문제: 모든 노드와 임베딩에 같은 값으로 수렴하는 현상
2. 과압축: GNN의 층이 깊어질수록 집계하는 이웃의 범위가 기하급수적으로 커짐
3. 계산 복잡도: 실제 세계에서는 수십억개의 노드와 엣지가 포함된 그래프에 GNN을 적용해야 하므로 막대한 자원 요구
4. 해석이 어려움 (순서 반영 X)

1. SARec

- 문제점: 계산량이 많으며 단방향성임

- 특징: 아이템의 양방향 문맥을 모두 고려하고 학습 과정에서 시쿼스 전체의 선호도를 파악하는 데 더 초점
