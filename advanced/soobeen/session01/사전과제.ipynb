# **1-1. 인공신경망**  
- 인공신경망: 우리 뇌의 신경망을 모방한 모델  
    - 입력들에 **가중치**와 **편향** 적용  
    - 임계값 기준으로 활성화 함수의 출력값 정해짐  
    - 결과값 출력  
    <br>
- 퍼셉트론: 초기 형태의 인공신경망/ 신경망의 기원이 되는 선형 이진 분류기  
    - '입력 -> 가중합 -> 활성화 함수 -> 출력'의 흐름 따라감  
    - 학습: 원하는 출력값을 내보내도록 *가중치를 조정*해가는 작업  
    - 임계값과 편향: 표현의 차이  

# **1-2. 단층 퍼셉트론과 그 한계**  
> ## 1. 단층 퍼셉트론  
: 단순히 입력층과 출력츠으로만 구성된 퍼셉트론   
<br>

> ## 2. 논리 게이트 구현  
: AND, NAND, OR 논리 게이트 쉽게 구현 가능  
* 논리 게이트란?  
    - 디지털 회로에서 기본적인 논리 연산을 수행하는데 사용되는 구성 요소 
    - n 개의 입력값과 *하나의 출력값* 사용  
    - 논리 게이트 조합 -> 복잡한 논리 회로 구성 가능  
<br>

> ## 3. 대표적인 논리 게이트  
**1. AND 게이트**  
: *입력값이 모두 1인 경우*에만 출력값이 1이 나오는 구조  
**2. OR 게이트**  
: *입력값이 모두 0*인 경우에 *출력값이 0*이고, 나머지 경우에는 *모두 출력값이 1*인 구조  
**3. NAND 게이트**  
: *입력값이 모두 1*인 경우에만 **출력값이 0**, 나머지 경우에는 *모두 출력값이 1*인 구조  
**4. XOR 게이트**  
: 입력값 두 개가 *서로 다른 값*을 갖고 있을 때에만 출력값이 1인 구조  
<br>

> ## 4. 단층 퍼셉트론의 한계  
-> **XOR 게이트 구현 불가**  
* 퍼셉트론은 입력값과 가중치의 선형 결합으로 구성 -> 단층 퍼셉트론은 직선 하나로 두 영역을 나눌 수 있는 **선형 분류 문제**에 대해서만 구현 가능  
=> 직선이 아닌 *곡선/비선형 영역*으로 분리해야만 구현 가능(여러 계층의 퍼셉트론으로 구현 가능)   



# **1-3. 다층 퍼셉트론**  
- 중간에 층 추가(입력층+하나 이상의 은닉층+출력층)로 구성  
- 복잡한 문제 해결 시 *중간에 수많은 은닉층 추가 가능*  
<br>
* 두 개 이상의 은닉층을 갖는 **MLP** -> 딥러닝의 시초 모델로 발전  
    - 심층 신경망 = 은닉층이 2개 이상인 신경망  
    - 딥러닝 = 심층 신경망을 학습 시킴

# **2-1. 순전파와 역전파**  
> ## 1. 학습 알고리즘의 기본 구조  
1. **순전파**를 통해 **입력값으로부터 예측** 수행  
2. **오차 계산**을 통해 **예측 정확도** 측정
3. **역전파**를 통해 출력층에서 다시 입력층 방향으로 **오차 전달**
4. **가중치 업데이트**
5. 1~4를 반복  
-> 반복 시 *오차값이 0에 근사*  
(순전파/ 역전파의 과정은 *연쇄법칙*을 통해 발생)  
<br>

> ## 2. 연쇄 법칙  
: 합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다는 법칙  
-> 각 층이 다음 층에 미친 영향ㅇ을 *순서대로* 곱해나가야 함  
<br>
- **연쇄법칙의 필요성**  
: 오차 감소를 위해   
    * 각 은닉층마다 손실에 주는 영향을 곱해서 최종 손실에 대한 값을 나타냄  
<br>

> ## 3. 기울기 소실 문제  
: 기울기들을 계속해서 곱했는데, 결국 기울기가 0이 되는 문제 발생  
-> 앞쪽 층의 가중치가 거의 업데이트 되지 않아, 학습이 제대로 이뤄지지 않음  
<br>

> ## 4. 전체적인 흐름 정리  
1. 순전파=예측하기  
2. 손실게산=얼마나 틀렸는지 측정하기  
3. 역전파=책임 추적하기(연쇄법칙 사용)  
4. 가중치 업데이트 = 수정하기  




# **2-2. 손실함수**  
> ## 1. 손실함수란?   
: 모델의 예측이 얼마나 틀렸는지 수치로 표현하는 방법  
-> 실제값과 예측값의 차이를 계산하는데 사용  
* 회귀 -> 평균 제곱오차 사용  
* 분류 -> 크로스엔트로피 오차 사용 
<br>

> ## 2. 손실 함수의 종류  
### 1. 평균 제곱 오차  
: 예측 값이 연속적이고, 예측 값과 실제 값 간의 차이를 직접적으로 측정하는 회귀 문제에 적합  
### 2. 이진 크로스 엔트로피  
: 이진 분류 문제(로지스틱 회귀)에서 사용하는 손실 함수   
-> **얼마나 확신을 가지고 예측했는지**도 loss에 반영  
     * 낮은 확률로 예측해서 맞추거나 높은 확률로 예측해서 틀리는 경우 loss가 더 큼!  
### 3. 크로스 엔트로피 오차  
: 다중 클래스 분류 문제에서 **예측확률이 실제 클래스와 얼마나 일치하는 지를 평가**  
-> 두 확률 분폰 (진짜/예측)의 *유사도* 측정에 유용  


# **2-3. 활성화 함수**  
> ## 1. 활성화 함수란?  
: 각 층의 출력값에 비선형성을 추가해서 신경망이 복잡한 패턴도 학습할 수 있게 해주는 함수  
* 가중치: 입력값이 결과에 얼마나 영향을 미치는지를 결정하는 값 
* 편향: 활성화 함수의 출력값을 조정하는 값 -> 기준점 역할 수행  
<br>

> ## 2. 계단함수  
: 입력값과 가중치의 곱을 모두 더한 뒤, 그 합이 특정 기준(임계값)을 넘으면 1, 아니면 0을 출력하는 함수  
-> **임계값**: 뉴런이 얼마나 쉽게 활성화되는 지를 제어 
(가중치는 각 입력 신호가 결과에 미치는 영향력을 제어)  
<br>

> ## 3. 시그모이드 함수  
: 입력 신호를 **0과 1 사이의 연속적인 값**으로 변환하는 비선형 함수  
-> 계단 함수와 출력값의 형태가 다름  
<br>

> ## 4. ReLU 함수  
: 입력이 0보다 크면 그대로, 0 이하면 0을 출력하는 비선형 함수  
-> 계산이 간단하고 효율적, 비선형성 제공, 기울기 소실 문제 완화 의 장점 가짐  
<br>

> ## 5 쌍곡탄젠츠 함수  
: **시그모이드 함수의 변형**으로, 출력값의 범위가 -1 ~ 1  
-> 입력값의 중심을 0으로 맞추고 싶을 때 사용  
- 기울기 소실 문제 발생 가능  
<br>

> ## 6. Softmax 함수 
: 각 클래스에대한 출력값을 **확률로 변환** 하는 함수  
- **다중 클래스 분류 문제의 출력층**에서 사용 가능

- 옵티마이저를 통해 더 효고적인 최솟값 찾아갈 수 있음

: **관성**이라는 물리학 법칙을 응용한 방법  
* 경사하강법에서 계산된 접선의 기울기에 바로 *한 시점 전의 접선의 기울기값*을 일정 비율만큼 반영  
<br>

> ## 2. 알엠에스프롭  
* 기존의 경사하강법 문제?

     -> *학습률* 조정으로 안정적이고 빠르게 학습하도록  
<br>
