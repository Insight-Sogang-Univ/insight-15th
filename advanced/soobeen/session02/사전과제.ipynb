># **1. 합성곱 신경망**  
### **픽셀**  
- 입력 이미지의 가장 기본적인 데이터 단위  
- 이미지의 각 점에서 밝기/색상을 나타내는 값  
- 이미지 데이터 저장 시 -> *2차원 배열*로 표현  
<br>
### **채널**  
**색공간**  
- 픽셀의 색을 숫자로 표현하는 방식  
- 대표적인 색공간: 그레이스케일, RGB, HSV  
<br>

### **CNN 개념 설명**  
- 원본 이미지를 **픽셀 단위**로 구분  
- 각 픽셀에 대해 RGB 값이 적절히 조합되어 하나의 픽셀의 색상 정보 나타냄  
<픽셀 정보의 학습법>
- 컴퓨터는 하나의 완전한 이미지를 형성하는지 모르기 때문에 -> **순서가 뒤섞인 채로 학습**  
=> 정보의 특징 소실  
따라서, *이미지 공간 정보*를 유지한 상태로 학습이 가능한 방식이 CNN

## 1-1. 이미지 처리에 MLP 아닌 CNN 사용하는 이유  
- 이미지는 픽셀 간의 **공간적 관계**가 중요한 반면 MLP는 픽셀을 일렬로 처리하기 때문 -> 이미지 구조의 유지를 위해서는 CNN 사용해야함  
<br>
### 1) 위치, 크기, 회전 등의 변화에 취약  
- convolution으로 특징 추출 -> pooling으로 차원 축소 
=> 위치와 무관  
### 2) 파라미터 폭발  
- MLP의 경우 완전연결 구조로 인한 *파라미터 급증*  
-> 과적합 가능성 상승, 이미지가 클수록 느린 학습  
- But, CNN에서는 이미지의 특정 피쳐에만 집중  
### 3) 공간 정보 손실  
- MLP: 이미지를 1차원 벡터로 바로 변환   
- 2차원 이상의 이미지 데이터를 1차원으로 flatten하는 과정에서 이미지의 공간 정보가 많이 소실  
- CNN: 다차원의 이미지 그대로 입력받아 -> 공간 정보를 보존하면서 학습  
- 특징을 어느 정도 추출한 후 flatten하기 때문에, 그 손실 정도 낮음  


## 1-2. 합성곱과 풀링  
* 특징 추출 영역 
     - convolution layer: 입력 데이터에 필터를 적용한 후 활성화 함수를 반영하는 핊수 요소  
     - pooling layer: conolution Layer 다음에 위치하며, 이미지의 공간 크기를 줄여 중요한 특징을 강조  
* 이미지 분류하는 영역: CNN 모델의 마지막에 위치 + Fully connected layer로 구성  
* Flatten: 3차원을 1차월 배열로 변환  
<br>

### 1. Convolution layer 
: 전체 이미지를 일정 크기의 필터로 훑으면서 이미지의 *특징* 학습  
1. 필터 또는 커널이 이미지 위를 슬라이딩  
2. 각 위치에서 요소별 곱셈 후 합계 계산  
3. 결과를 feature map에 저장  
<br>

### 2. Padding   
: filter를 거치며 정보를 잃어버리는 것을 막기 위해 이미지 바깥을 0으로 둘러싸는 기법  
<br>

### 3. Stride   
: 이미지를 몇 칸씩 점프하면서 훑을 것인지 결정하는 값  
-> stride가 커질수록 feature map의 크기 작아짐  
<br>

### 4. Pooling  
: 피처맵에서 특징을 추출하기 위해 이미지를 격자 형태로 분할한 뒤, 해당 영역 안에서 가장 큰 값 또는 평균값을 취하는 것  



## 1-3. 대표적인 CNN 아키텍처  
- AlexNet -> VGGNet -> ResNet  
- AlexNet: 최초로 GPU 이용해 학습한 대규모 모델  
- VGGNet: **이미지 인식**에 강한 깊은 층의 모델  
- ResNet: 잔차 연결 개념을 도입해 *깊은 신경망의 소실 문제를 해결*한 모델

> # **3. 과거를 기억하는 신경망: RNN**  
## 3-1. 순차 데이터란?  
: *순서에 의미*있으며, 순서가 달라질 경우 의미가 손상되는 데이터  
1. 음성 및 오디오 데이터  
2. 자연어  
3. 생물학적 서열 데이터  
4. 비디오 데이터  
<br>

* **기존 정형 데이터와의 차이점**  
- 요소의 **순서 의존성**: 데이터를 구성하는 요소들의 순서 자체가 핵심 정보  
- **자기상관성**: 이전 시점의 데이터가 다음 시점의 데이터에 영향을 미침  
-> 과거의 정보를 기억하고, 순서의 의미를 학습할 수 있는 특별한 신경망 필요  
<br>

## 3-2. RNN의 구조와 원리 
**순환 신경망**  
1. **시퀀스 데이터**를 입력받아 **순서 정보를 유지**하며 처리  
2. 순환하는 **은닉층**이 매 시점의 **은닉 상태**를 업데이트   
3. 이전 시점의 값을 현재 시점으로 토스  
-> 은닉 상태의 순환을 통해 과거의 정보를 다름 시점으로 게속해서 전달  
<br>

**RNN 아키텍처**  
- One to One/ One to Many/ Many to Many  
<br>

**RNN 핵심원리: **가중치공유****   
- *동일한 가중치*를 시퀀스의 다양한 시점에서 반복 적용  
- RNN이 시퀀스 데이터의 *길이와 시점 위치에 상관없이* 효과적으로 작동하게 하는 핵심 원리  
<why?>  
- 학습 파라미터 수 감소  
- 일반화 능력 향상   
<br>

## 3-3. RNN의 한계 및 장기 의존성 문제 
: Rnn 구조의 특성 상 전체 시퀀스를 모두 읽은 후 역전파 발생 -> 역전파 구하는 과정에서 연쇄 법칙에 의해 미분값이 반복적으로 곱해짐  
- **기울시 소실** : 기울기가 0으로 수렴  
- **기울기 폭주** : 기울기의 무한대 발산 
- **느린 훈련 시간**: 전체 시퀀스 한번에 병렬 처리 불가능  
<br>
### **장기 의존성 문제**  
: 시퀀스 앞 부분의 중요한 정보를 잊어버려 맥락 파악 능력이 급격히 저하되는 현상  
-> 모델의 '장기 기억력'에 문제 발생  
따라서, Rn의 기억력 한계를 극복하고, 더 길고 복잡한 시퀀스 데이터를 안정적으로 다루고, 중요한 정보는 오래/불필요한 정보를 잊는 새로운 모델의 필요성 대두  

> # **4. 똑똑하게 기억하고 잊는 법: LSTM&GRU**   
## 4-0. LSTM & GRU  
### **LSTM**  
- 핵심 기억인 셀 상태와 은닉 상태인 엘리트로 구성  
-> 진짜 장기 기억과 단기 기억 및 출력을 **따로 분리해서 관리**  
=> 매우 정교하지만 복잡  
<br>
### **GRU** 
: 모든 기억과 행동이 하나의 상태로 통합됨  
-> LSTM보다 훨씬 **단순하고 즉각적인 반응**(정교함 다소 감소)  
<br>

## 4-1. 게이트로 정보의 흐름을 제어하다: LSTM 
* LSTM: RNN의 **기울기 소실 문제**를 해결하기 위한 모델 구조  
     - 기억학 내용과 잊어버릴 내요을 선택해 중요한 정보를 오래 가져감  
     - 2개의 순환되는 층 사용  
     - GATE를 통해 필요한 정보들만 통과  
     - 현재 정보를 얼마나 기억할지 계산 
<br>

## 4-2. 더 단순하고 효율적인 구조: GRU  
: 별도의 메모리 셀 없이 **게이트 수를 줄여** 구조를 간소화  
     - 성능과 효율성 사이의 균형 좋음  
     - but 긴 시퀀스 처리에서는 한계 여전히 존재  
<br>

## 4-3. 언제 무엇을 선택?  
* **LSTM이 적합한 경우**  
     - 긴 시퀀스에서 문맥 이해가 중요한 작업   
     - 데이터 양이 충분하고 복잡한 패턴을 학습해야 하는 경우  
     - 모델 성능이 계산 효율보다 더 중요한 경우  
     - 기계 번역, 언어 모델링, 장기 시계열 예측  
* **GRU 적합한 경우**  
     - 자원이 제한적이거나 학습 속도 중요한 경우  
     - 데이터 양이 적고 과적합 위험이 있는 경우  
     - 실시간 예측이 필요한 응용  
     

> # **5. 문장을 입력받아 문장 출력: Seq2Seq**  
## 5-0. Seq2Seq 알아보자  
- 인코더: 의미를 정리하는 사람  
- 디코더: 그걸 받아서 결과를 만드는 사람  
<br>
### **인코더**  
: 핵심 의도 = 컨텍스트 벡터  
### **디코더**  
: 인코더가 전달한 컨텍스트 벡터를 받아 결과 생성  
- 인코더의 요청 그대로가 아닌 **새로운 결과**를 만들어낸다는 점  
<br>
* **특징**  
- 입력과 출력의 길이 달라도 됨  
- 인코더와 디코더 분리됨  
- 문맥 벡터 통과  
<br>
### **컨텍스트 벡터**  
: 인코더 중 마지막 시점의 은닉 상태  
     - 입력 시퀀스의 요약 정보처럼 작동  
     - 벡터의 차원 직접 설정 가능  

<br>

## 5-2. Seq2Seq의 한계: 병목 현상  
: 고정된 컨텍스트 벡터에 소스 문장의 정보 압축 -> 병목 현상 발생
