> # **1. NLP**  
: 자연어  
### 자연어란?  
    인간이 소통을 위해 자연스럽게 발생시켜 사용하는 언어  
- 프로그래밍 언어와 달리 일대다 대응의 언어
- 역할: 컴퓨터가 이해할 수 있는 '숫자(데이터)'로 변환하여 유의미한 가치를 창출하는 기술  

**<자연어의 특징>**  
1. 이산적 데이터  
2. 순차적 데이터  
3. 중의성과 모호성  
4. 희소성  
5. 가변성과 진화

## (2)N-gram  
<등장배경> 
- 등장 빈도가 낮은 낮은 단어에 대해 잘 *예측*하지 못함  
- 문장이 길어질수록 다음 단어 계산이 어려움  
<br>
## 한계  
1. 희소 문제 
2. 문맥 제한  
<br>

## (3)Perplexity(PPL)  
- PPL: 언어 모델이 문장을 얼마나 "혼란스러워하는지' 수치화 한 것 

> # **3. 신경망 언어 모델 (NLM)**  
    통계적 언어 모델의 한계를 극복하기 위해 등장 
- 모든 단어를 거대한 다차원 지도 위의 한 점(좌표)로 보냄-> *워드 임베딩*  



> # **4. 딥러닝 기반 언어모델**  
## (1) Transformer의 등장  
**<RNN과 LSTM의 한계>**
- 병렬화 불가  
- 기울기 소실 및 폭주  
- 장기 의존성 문제  
-> 신경망의 깊이를 깊게 만든 것을 불가능하게 하는 원인  
  
-> *언어 모델의 표현력* 높아지게 하여 현재의 딥러닝 기반 언어모델 생성  
<br>

### 전이 학습의 2단계 
1. 사전학습  
2. 미세조정  
<br>  

## (2) BERT  
=  Bidirectional Encoder Representations from Transformers  
- 양방향 작동 
<br>

### [BERT의 특징]  
- 양방향  
- 트랜스포머 기반  
- 사전학습+파인 튜닝  
<br>

**BERT의 학습 과정**  
: encoder 기반  
- transformer의 encoder를 여러 개 쌓아 올린 구조  
<br>
**[처리 과정]**  
input → Embedding → 트랜스포머 인코더 12개(or 24개) → Output  
<br>
**[사전학습]**  
(1) Masked Language Model(MLM)  
: *빈 칸 채우기 방식*으로 학습  
(2) Next Sentence Prediction(NSP)  
: *두 문장이 이러지는 논리 흐름* 



## (3)GPT  
: decoder 기반  
- 지금까지의 다넝를 바탕으로 다음 단어 예측  
- CPT-3 부터 few-shot learning 도입  


## (4) LLM 
**특징**  
1. Scaling Law  
2. 창발적 능력  
3. RLHF: "똑똑한 AI를 '착한 AI'로 길들이기  
4. 패러다임의 전환: "튜닝(Tuning)에서 프롬프트로  
<br>
<LLM의 작동 방식>  
    학습과 예측을 통해 작동  
(1) 대규모 학습   
(2) 예측 및 생성 
<LLM 한계>  
- hallucination  
- 업데이트 비용  





## (5) RAG  
  
- 질의 인코더   
- 지식 검색기  
- 지식 증강 생성기  

<CAG>  
Credibility-Aware Generation  
- RAG의 한계를 보완하기 위해 제안된 프레임워크   
<br>
[CAG의 구조]
질문 입력 → 정보 검색 → 각 정보의 신뢰도 평가 → 높은 신뢰도 정보 우선 사용 → 최종 답변 생성  




## (6) LangChain  
    LLM 앱을 빠르게 조립할 수 있는 레고 블럭  

<LangChain의 특징>  
- 추상화, 표준화, 체이닝  
  
- 비션형적인 작업 형태  
- 사이크, 루프 생성 가능
