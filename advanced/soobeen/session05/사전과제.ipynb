> # **생성모델**  
## 1.1 분류 모델 v.s. 생성 모델  
### 분류모델  
분류 모델은 입력 데이터가 어떤 정답 레이블에 속하는지 판단하는 모델이다.  
즉, 주어진 데이터가 고양이인지, 개인지처럼 **정답을 구분하는 것**에 초점

### 생성모델  
    생성 모델은 데이터 자체의 분포를 학습하여, 기존 데이터와 유사한 새로운 데이터를 만들어내는 모델

## 1.2 생성 모델의 분류  
    생성 모델은 확률 분포를 직접 계산할 수 있는지에 따라 크게 **명시적 모델**과 **암시적 모델**로 구분  
    

### 명시적 모델  
: 데이터의 확률 분포를 직접 계산하거나 근사적으로 계산하는 모델  
- 확률 값을 수치로 표현할 수 있음
- 대표 예시: VAE
- 정확한 확률 계산이 어렵다면 근삿값을 사용함
<br>

### 암시적 모델  
: 확률 분포를 직접 계산하지는 않지만, 그 분포를 따르는 데이터를 생성할 수 있는 모델  
- 분포 공식은 알 수 없어도 샘플 생성은 가능함
- 대표 예시: GAN, 확산 모델
- 실제 데이터와 비슷한 결과물을 만드는 데 초점이 있음  


> # **2. AE, VAE**  
## 2.1 AE  
    입력 데이터를 압축한 뒤, 다시 원래 데이터와 비슷하게 복원하는 모델  

- 주로 **차원 축소, 특징 추출, 노이즈 제거, 이상 탐지**에 활용
<br>

### 구조  
크게 **인코더**, **잠재 벡터**, **디코더**로 구성  
- **인코더:** 입력 데이터를 중요한 정보만 남기도록 압축
- **잠재 벡터:** 압축된 핵심 특징이 담긴 공간
- **디코더:** 압축된 정보를 바탕으로 원래 데이터를 복원  
-> AE는 새로운 데이터를 생성하기보다는, 입력 데이터의 중요한 특징을 잘 압축하고 복원하는 데 목적  
=> 생성 모델의 기초 구조로 볼 수 있지만, 본격적인 생성 능력은 제한적


## 2.2 VAE  
    AE의 구조를 기반으로 하면서, 잠재 벡터를 하나의 고정된 점이 아니라 **확률 분포**로 표현하는 모델  
-> 원본과 비슷하지만 새로운 데이터를 생성  
### **AE와의 차이**
AE는 입력 데이터를 잠재 공간의 특정한 점으로 압축  
반면 VAE는 평균과 분산을 이용해 잠재 공간을 확률적으로 표현

## 2.3 GAN  
    **생성자**와 **판별자**가 서로 경쟁하면서 학습하는 생성 모델  
-> 생성자는 진짜 같은 가짜 데이터를 만들고, 판별자는 그것이 진짜인지 가짜인지 구별  

### 구조  
두 개의 신경망으로 구성  
- **생성자:** 무작위 노이즈를 입력받아 가짜 데이터를 생성함
- **판별자:** 실제 데이터와 가짜 데이터를 비교하여 진짜 여부를 판단함  
-> 생성자는 판별자를 속이도록 학습하고, 판별자는 더 정확하게 구별하도록 학습  
<br>
### **장점**
- 매우 선명하고 현실적인 결과물을 만들 수 있음
- 한 번의 신경망 통과로 빠르게 데이터를 생성할 수 있음
- 이미지 생성, 영상 합성 등에 효과적으로 사용됨

### **단점**
- 학습이 불안정함
- 생성자와 판별자의 균형을 맞추기 어려움
- 특정 결과만 반복적으로 생성하는 문제가 생길 수 있음
- 학습이 잘 되고 있는지 판단할 명확한 지표가 부족함  
<br>

## 2.4 확산 모델  
    원본 데이터에 노이즈를 조금씩 추가한 뒤, 다시 그 노이즈를 제거하는 과정을 학습하는 모델  
-> 노이즈에서 출발해 점진적으로 원본과 비슷한 데이터를 만들어내는 방식  
<br>

### **순확산**  
    원본 데이터에 점진적으로 노이즈를 추가하는 과정  
- 시작점은 깨끗한 원본 데이터
- 여러 단계를 거치며 노이즈가 조금씩 추가됨
- 마지막에는 완전한 노이즈 상태가 됨  

<br>

### **역확산**  
    완전한 노이즈에서 출발해 점진적으로 노이즈를 제거하는 과정  
- 시작점은 무작위 노이즈
- 각 단계에서 제거해야 할 노이즈를 예측함
- 반복적인 복원을 통해 최종 이미지를 생성함



> # **2. AI 최적화와 정렬**

## 0. AI 최적화와 정렬의 필요성  
    AI 모델이 점점 커지고 성능이 좋아지면서, 단순히 잘 작동하는 것을 넘어 효율성과 안전성이 중요해짐  
-> AI가 더 잘 학습하도록 만드는 최적화와, 인간의 의도에 맞게 작동하도록 만드는 정렬이 핵심 과제가 됨  
<br>

### **AI 발전 흐름**  
    AI 연구는 모델이 더 복잡한 문제를 해결할 수 있도록 발전해 왔음  
- 퍼셉트론 → 역전파 → CNN → RNN → 트랜스포머 → LLM → 생성모델의 흐름으로 발전함
- 모델 규모가 커지면서 성능은 향상되었지만, 학습 비용과 통제 문제가 함께 증가함
- 따라서 효율적인 학습 방법과 인간 의도에 맞는 조정 방법이 필요해짐  

<br>

---

## 2.1 전이학습  
    기존에 학습된 모델의 지식을 새로운 문제에 재사용하는 학습 방법  
-> 처음부터 다시 학습하지 않고, 이미 배운 지식을 활용해 효율적으로 성능을 높이는 방식  
<br>

### **전이학습의 개념**  
    대규모 데이터로 학습된 모델을 다른 작업에 맞게 활용하는 방법  
- 학습 시간과 비용을 줄일 수 있음
- 적은 데이터로도 좋은 성능을 낼 수 있음
- 이미지, 언어, 음성 등 다양한 AI 분야에서 활용됨  

<br>

### **전이학습의 분류**  
    기존 모델을 얼마나 수정하느냐에 따라 방식이 나뉨  
- **특징 추출 방식**: 기존 모델의 대부분을 고정하고, 마지막 부분만 새로 학습함
- **미세조정 방식**: 기존 모델의 일부 또는 전체를 새로운 데이터에 맞게 다시 학습함
- **도메인 적응 방식**: 기존 데이터와 새로운 데이터의 차이를 줄이도록 모델을 조정함  

<br>

### **Pre-Training**  
    대규모 데이터로 모델이 기본적인 패턴과 표현을 먼저 학습하는 단계  
- 모델이 언어, 이미지, 음성 등의 일반적인 구조를 배움
- 이후 다양한 작업에 활용할 수 있는 기초 능력을 형성함
- 많은 데이터와 연산 자원이 필요함  

<br>

### **Fine-tuning**  
    사전 학습된 모델을 특정 목적이나 데이터에 맞게 추가로 학습시키는 단계  
- 특정 과제에 맞게 모델의 성능을 조정함
- 비교적 적은 데이터로도 성능 향상이 가능함
- 과도하게 학습하면 기존 지식을 잃거나 과적합될 수 있음  

<br>

---

## 2.2 AI Alignment  
    AI가 인간의 의도, 가치, 목표에 맞게 행동하도록 조정하는 과정  
-> 성능이 좋은 AI를 넘어서, 사용자의 의도에 맞고 안전하게 작동하는 AI를 만드는 것이 핵심  
<br>

### **AI Alignment의 필요성**  
    AI가 예상과 다른 방향으로 답하거나 유해한 결과를 만들지 않도록 조정해야 함  
- 사용자의 질문 의도를 정확히 반영해야 함
- 부적절하거나 위험한 결과를 줄여야 함
- 인간이 원하는 기준에 맞는 답변을 생성해야 함  

<br>

### **RLHF**  
    사람의 피드백을 바탕으로 AI의 답변 품질을 개선하는 학습 방법  
- AI가 여러 답변을 생성함
- 사람이 더 적절한 답변을 선택하거나 평가함
- 그 피드백을 바탕으로 AI가 더 선호되는 답변을 하도록 학습함  

<br>

### **RLHF의 의미**  
    AI가 단순히 정답을 맞히는 것뿐만 아니라, 사람이 더 좋다고 판단하는 답변을 학습하게 함  
- 답변의 자연스러움과 유용성을 높일 수 있음
- 사용자의 의도에 맞는 결과를 생성하는 데 도움을 줌
- 사람의 평가 기준에 따라 편향이 반영될 수 있음  

<br>

## 2.3 프롬프트 엔지니어링  
    AI가 원하는 결과를 더 정확하게 생성하도록 입력 문장을 설계하는 방법  
-> 같은 모델이라도 질문을 어떻게 구성하느냐에 따라 답변의 품질이 달라짐  
<br>

### **프롬프트 엔지니어링의 개념**  
    AI에게 역할, 조건, 형식, 예시를 명확히 제시하여 원하는 답변을 유도하는 과정  
- 질문의 목적을 구체적으로 작성함
- 원하는 답변의 형식과 길이를 명시함
- 포함할 내용과 제외할 내용을 분명히 제시함  

<br>

### **주요 방법**  
    입력 구조를 체계적으로 구성해 AI가 문제를 더 잘 이해하도록 만듦  
- **역할 부여**: AI가 어떤 관점에서 답해야 하는지 정함
- **조건 제시**: 분량, 말투, 형식, 필수 내용을 지정함
- **예시 제공**: 원하는 답변의 방향을 보여줌
- **단계적 지시**: 복잡한 문제를 나누어 처리하게 함  

<br>

### **장점과 한계**  
    별도의 모델 학습 없이도 결과 품질을 개선할 수 있지만, 입력 방식에 따라 결과가 달라질 수 있음  
**장점**  
- 빠르고 간단하게 AI의 활용도를 높일 수 있음
- 사용자의 목적에 맞는 답변을 유도하기 쉬움  
<br>

**한계**  
- 프롬프트가 모호하면 결과도 불안정해질 수 있음
- 모델 자체의 지식이나 성능 한계를 완전히 해결할 수는 없음  

<br>
