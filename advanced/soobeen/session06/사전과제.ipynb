> # **1. 추천 시스템이란?**  
    사용자의 정보 데이터를 분석, 개인의 취향에 맞는 아이템을 추천하는 알고리즘  
<br>

## 1-1. 사용되는 데이터의 종류와 특징  
**1) 명시적 데이터**
    고객들에 의해 입력된 데이터. 사용자가 선호도를 직접적으로 표현한 데이터  
-> 사용자의 취향 정확 반영하지만, 데이터 자체를 얻기 힘듦  
**2) 암시적 데이터**  
    고객들의 행동을 기반으로 수집된 데이터. 사용자의 선호도가 간접적으로 표현된 데이터  
-> 데이터를 얻기 easy, 그러나 취향이나 선호도를 직접적으로 알 수 없다  



## 1-2. 추천시스템의 분류  
**1) 내용 기반 필터링**  
- 콘텐츠의 *유사도*를 기반으로 사용자에게 추천  
**2) 협업 필터링**  
- 서비스 내에 있는 *다수의 사용자로부터 얻은 선호도 정보*를 기반으로 사용자에게 추천하는 방법(사용자 기반/아이템 기반)  
**3) 하이브리드 필터링**  
- 2가지 이상의 다양한 종류의 *추천시스템 알고리즘을 조합*하여 만들어진 알고리즘
- 콘텐츠 기반 필터링과 협업 필터링 각각의 단점들을 보완하기 위해 제시된 방식

> # **2. 내용 기반 필터링**  
## 2-1. 사용자 프로필  
- 사용자에게 자신이 선호하는 아이템의 특징을 명시적으로 지정하는 '직접 지정형'과 사용자의 과거 행동 이력에 기반하여 사용자의 프로필을 작성하는 '간접 지정형'으로 구분  
<br>
### **사용자의 프로필을 획득하는 방식**  
1. 특정 가중치는 사용자가 가지고 있는 아이템의 특성 가중치의 평균 값 활용
2. 사용자의 기호 따로 파악 or 설문조사 또는 자발적인 키워드 입력 -> 프로필 획득  
3. 아이템에 대한 유저의 기존 평가를 바탕으로 분류 문제 해결  
-> (명시+암묵)을 *혼합하여 사용*하는 것이 일반적

## 2-2. 유사도  
    추천시스템을 이해하는데 중요한 수학적 개념  
1. 아이템 정보 데이터 벡터화-> 아이템 벡터간의 *유사도* 측정  
2. 측정 유사도 바탕으로 고객이 선호하는 아이템과 비교 -> 유사도가 높은 아이템 선별  
<br>

### TF-IDF  
  
    특정 문서 내에 특정 단어가 얼마나 자주 등장하는지 TF와 전체 문서에서 특정 단어가 얼마나 자주 등장하는지 IDF를 통해서 -> "다른 문서에서는 등장하지 않지만 특정 문서에서만 자주 등장하는 단어"를 찾아서 문서 내 단어의 가중치를 계산하는 방법

    각 단어가 그 문서에서 얼마나 중요한지를 나타내는 값
-> 높을수록 그 단어가 다른 문서들보다 해당문서에서 더 중요한 역할을 함을 의미

## 2-3. 내용 기반 필터링의 장단점  
- 아이템이 가진 특징 자체에만 집중하기 때문에 **콜드스타트 문제**에서 자유로움  
- But, 정확한 특징을 추출하지 못한다면 정확도가 떨어짐  
### Cold start problem
    추천 시스템이 새로운 유저들에 대한 정보 수집 불충분 -> 해당 유저들에게 적절한 제품 추천 불가한 문제  
    

> # **3. 협업 필터링**  
## 3-1. 메모리 기반 협업 필터링 (이웃 기반 협업 필터링)  
- 사용자 또는 아이템 간의 *유사도*를 기반으로  
- 비교적 초기에 개발된 알고리즘  
- 사용자 A와 *비슷한 취향을 가진 사용자*가 선호하는 아이템 추천
<br>

### 1. 사용자 기반 접근  
    유저 간의 유사도 높을수록 가중치 부여 -> 같은 그룹의 다른 유저가 선호하는 아이템 추천  
  


### 2. 아이템 기반 접근    
    추천하고자하는 타켓 아이템과 유사한 아이템의 평점을 활용
  
<br>

### User-based VS Item-based 비교  
**UBCF**  
- 추천 정확성&효율성: 아이템 수가 사용자 수보다 많은 환경에서 유리
- 아이템이 수시로 바뀔 때 적합 -> 안정적  
- 특정 아이템이 독특한 아이템을 추천했다면 *뜻밖의 독창적인 아이템 추천 가능* -> 다양성과 참신성이 강점   
ex) 뉴스, 소셜 미디어, 음악 초기 모델   
<br>

**IBCF**  
- 추천 정확성 & 효율성: 사용자 수가 아이템 수보다 많은 환경에서 유리  
- 아이템 리스트의 변화가 적고 *고정적인* 환경에서 안정적  
- 유사한 아이템 위주로 추천되어 다양성과 참신성 상대적으로 부족  
- 사용자별 개별 게산이 불필요 -> 빠르고 안정적
ex) 아마존, 넷플릭스 등  


## 3-2. 메모리 기반 협업 필터링의 한계  
### 1. 희소 행렬 문제  
    주어진 평점/선호도 데이터가 적을 경우, 성능 저하  
- 실제 유저-아이템 행렬 *대부분의 원소는 비어있음*  
- NBCF를 적용하려면 적어도 sparsity ratio(행렬 전체 원소 중 비어있는 원소의 비율)가 99.5%를 넘지 않게  
<br>

### 2. 확장성 문제  
    아이템/유저가 계속 늘어날 경우, 확장성 떨어짐
- 아이템 i에 대한 평점 예측하게 위해 -> 아이템 i에 대해 평가를 한 모든 유저와의 유사도 구해야함  
-> 따라서 KNN CF 이용
<br>

### 3. 콜드 스타트 문제
- 데이터 아예 없는 경우 추천 불가  
- 신규 유저 유입 or 신규 아이템 등록  
-> 직접 지정형 프로필 활용하거나  
-> content-Based Filtering과 혼합하여 아이템 특성 기반을 보완  
<br>



## 3-3. 모델 기반 협업 필터링  
    메모리 기반 CF의 한계 보완하기 위해 등장한 방식 -> 단순히 유사도를 계산하는 것이 아니라, *데이터로부터 패턴을 학습한 모델*을 만들어 평점 예측
### 행렬 분해  
- 수학적 분해 기법: 행렬 대수학 기반(결측치 처리 필요)  
- 최적화 알고리즘: 경사 하강법 기반 알고리즘 -> 희소성 문제 우회  
- 제약 조건 기반 모델: 분해된 행렬에 제약조건(비음수) 추가 -> 잠재 요인의 해석력 높임  
- 확률적 모델: 평점 예측 확률 분포 사용 -> 불확실성까지 모델링

> # **2. GNN 기반 사용자 선호 예측**  
## 2-1. GNN  
### 그래프  
    노드와 엣지를 모아 구성한 자료 구조  
-> 노드=현실 세계의 개체/ 엣지= 개체 간 관게나 상호작용(가중치 포함 가능)  



## 2-2. GNN의 구조  
1. 그래프 구조 생성
2. 메시지 패싱(메시지 생성 -> 집계 -> 갱신)  
3. 최종 출력  
-> 내 이웃 노드의 정보를 나에게 전달하자!!

## 2-3. GNN의 프로세스  
- GNN은 그래프 입력 받아 *메시지 패싱* 수행  
- 각 노드는 *자신의 이웃 노드들의 정보를 수집*하여 임베딩 업데이트


### GNN의 장점  
1. 고차원 협업 시그널 포착  
2. 정형화 되지 않은 데이터 학습 가능  
3. Cold start 문제 완화에 효과적  
<br>

### 단점  
1. 과평탄화 문제  
2. 과압축 문제  
3. 계산 복잡도  
4. 해석의 어려움
