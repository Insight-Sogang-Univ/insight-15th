**퍼셉트론의 출력값을 구하는 절차**

⇒ **`입력 → 가중합 → 활성화 함수 → 출력`** 흐름을 따라감

1. 입력값과 가중치를 곱합니다 →  ${x_1*w_1}$
2. 곱한 값들의 총합을 구합니다 → ${x_1*w_2+x_1*w_2+x_3*w_3 +...+x_n*w_n}$
(편향을 더해줄지는 옵션이에요)

<span style="color:hotpink; background-color:grey">**편향은 모델이 얼마나 쉽게 활성화될지(기준점을 얼마나 높이거나 낮출지)를 조절하는 역할을 합**</span>

3. 총합이 임계값(${\theta}$**)**을 넘으면 $1$, 넘지 않으면 $0$을 출력합니다.
(이 경우, 활성화 함수는 Step 함수)

※ 활성화 함수: 임계값(${\theta}$) 초과 여부에 따라 출력값을 결정. 
     사용하는 활성화 함수에 따라 출력값이 달라질 수 있음.

→ **Training(학습)**: 퍼셉트론이 우리가 원하는 정답을 출력할 수 있도록, 입력과 결과 사이의 관계를 파악하고 가중치($w$)와 편향($b$)을 컴퓨터가 스스로 조금씩 수정해 나가는 과정을 의미

<span style="color:white; background-color:hotpink">연쇄법칙</span>

- 연쇄 법칙은 합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있다는 법칙
- 딥러닝에서는 입력값이 여러 은닉층을 거친 후 최종 출력값이 나오는데, 이때 입력값이 최종 출력에 얼마나 영향을 미쳤는지 계산하려면 각 층이 다음 층에 미친 영향을 순서대로 곱해나가야 함
- 이렇게 하면 각 단계가 연쇄적으로 결과에 얼만큼 영향을 줬는지 알 수 있겠죠?→ 이게 바로 연쇄법칙

- 왜필요함? / 오차를 줄이기위해. 이를 위해 역전파 과정에서 각 가중치가 최종 손실에 얼마나 기여했는지를 계산해야 하는데, 이때 연쇄법칙을 사용

  




- 딥러닝 모델의 성능을 높이려면, 모델의 오차를 계산해 순전파와 역전파를 활용해서 오차를 줄이는 것이 필요!
- 손실 함수 = 오차값과 실제값을 수치로 표현
- 연쇄법칙을 통해 구하는 기울기 = ‘손실 함수를 줄이기 위해 가중치를 어떻게 바꿀까?’를 알려주는 신호

<span style="color:white; background-color:hotpink">기울기 소실 문제 (Vanishing Gradient Problem)</span>



<span style="color:white; background-color:hotpink">손실 함수 (Loss Function)</span>

- 손실 함수는 모델이 정확한지 오차를 계산하는 방법 
⇒ 실제값(정답)과 예측값(출력값)의 차이를 계산하는 데에 사용되는 함수

<span style="color:white; background-color:hotpink">손실 함수의 종류</span>





(1) 실제 정답이 1일 때

수식에서 $y=1$을 대입하면 $L=-\log{\hat{y}}$ 만 남습니다.

- 예측 확률이 1에 가까울수록 → loss값은 0에 가까워짐 = 잘 맞힘
- 예측 확률이 0에 가까울수록 → loss가 급격히 증가 (확신했지만 틀림)
- 예시
    
    실제 레이블이 고양이(1)인데 고양이일 예측 확률이 0.1이라면, 모델이 고양이가 아니라고 확신하며 틀린 것이므로 −log(0.1)로 계산되어 손실 값이 큽니다.
    

(2) 실제 정답이 0일 때

- 예측 확률이 0에 가까울수록 → loss값은 0에 가까워짐 = 잘 맞힘
- 예측 확률이 1에 가까울수록 →  loss가 급격히 증가 (확신했지만 틀림)

  
마이너스 붙는 이유: log 함수의 특성 때문입니다! 확률값은 항상 0~1 사이인데, 0~1 사이의 수를 log에 넣으면 결과가 항상 음수가 나옵니다. 손실값은 양수여야 "오차가 크다/작다"를 표현할 수 있으니까, 앞에 마이너스를 붙여서 양수로 바꿔주는 것입니다!





<span style="color:white; background-color:hotpink">활성화 함수</span>

신경망의 각 층은 기본적으로 이런 계산을 합니다.
입력값 × 가중치 + 편향  →  활성화 함수  →  출력값
(wx+b를 생각하시면 편해요!)
이때  활성화 함수가 없으면, 층을 아무리 많이 쌓아도 결국 선형 계산의 반복에 불과하기에 앞서 봤던 XOR 같은 복잡한 패턴을 학습할 수 없습니다. → 활성화 함수의 필요성!!
다시 말해 활성화 함수는 각 층의 출력값에 비선형성을 추가해서 신경망이 복잡한 패턴도 학습할 수 있게 해주는 함수





계단 함수와의 차이점? 출력값의 형태!!
계단 함수는 0 또는 1만 출력하는 반면, 시그모이드 함수는 x값에 따라 출력값이 연속적으로 변화함. 덕분에 "확실히 맞다/틀리다"가 아니라 얼마나 확실한지를 표현할 수 있음.



  
시그모이드와 마찬가지로, 입력값이 매우 크거나 작으면 미분값이 0에 가까워져 기울기 소실 문제가 발생할 수 있음

배치 경사하강법의 특징
전체 데이터를 통해 학습시키기 때문에, 업데이트 횟수가 적음 (1 Epoch 당 1 업데이트)
전체 데이터를 모두 한 번에 처리하므로 메모리가 많이 필요
항상 같은 데이터 (전체)에 대해서 경사를 구하기 때문에 수렴이 안정적

배치 크기가 1인 확률적 경사하강법(Stochastic Gradient Descent, SGD)

기존의 경사하강법의 단점?

## 2-6 데이터 증강(Data Augmentation)

현실의 데이터는 다양하게 변형되고 조금씩 바뀌는 경우가 많습니다.
따라서 기존 데이터에서 다양한 변형을 만들어 학습 데이터를 늘린다면, 더욱 효과적인 학습이 가능하겠죠?
⇒ 이때 필요한 것이 데이터 증강


- 학습 데이터 셋에 일부러 다양한 변화를 가하는 것
- 목적: 데이터 부족 문제 해결, 모델의 견고함(Robustness) 향상, 과적합 감소
- 적용 분야
- 비전(이미지): 뒤집기(Flip), 회전(Rotation), 부분 자르기(Crop), 확대/축소, 밝기 조정 등
자연어 처리(NLP): 동의어 대체, 랜덤 삽입/삭제, 문장 섞기, 번역 후 역번역 등
- 주의점: 변형된 데이터가 원본의 본질적인 특성을 잃지 않아야 하며, 태스크에 적절한 증강 방식을 선택해야 합니다.

데이터 증강은,
- 모델 성능의 향상
- 데이터 의존성 감소
- 과적합 완화
- 데이터 프라이버시 보호
에 도움이 됨!
