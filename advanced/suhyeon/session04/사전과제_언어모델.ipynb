## <span style="color:crimson; background-color:white">NLP와 언어 모델</span>


프로그래밍 언어는 일대일 대응이고 자연어는 일대다 대응이다..

NLP: natural languge processing; 자연어를 분석하고, 컴퓨터가 처리할 수 있도록 하는 인공지능 기술
- 예시: 음성 인식, 내용 요약, 번역, 감정 분석, 텍스트 분류, 챗봇
- 글자를 숫자 벡터로 바꾸고, 의미를 추출하는 모든 과정

과정
1. 데이터 수집
    - 웹 크롤링, API호출, 데이터셋 다운로드 등
2. 데이터 전처리
    - 모든 NLP 모델의 기초작업. 모델이 이해할 수 있는 형태로 텍스트 정제
3. 텍스트 표현
    - 텍스트를 수치로 표현- 머신러닝/딥러닝 입력 가능하게 변환  
  
얘네도 종류가 다양하구나~
4. 모델 학습 및 예측  
  

5. 후처리 및 응용
    - 출력 다듬기: 단어 복원, 마침표 추가, 오타 제거 등
    - 응용 시스템: 챗봇, 검색 엔진, 추천 시스템, 음성 비서
    - 평가 지표: 정확도, F1, ROUGE, Perplexity 등 사용

  
형태소 분석.. 국문학과 같아서 묘하다

자연어의 특징
1. 이산적 데이터(연결되지 않고 개별적으로 떨어짐)
    - 특징: 이미지나 소리-연속적 신호, 언어는 단어라는 독립적이고 딱 끊어지는 단위로 구성됨
    - NLP의 숙제: 컴퓨터가 이해할 수 있도록 이산적인 단어를 연속적인 벡터(임베딩)로 변환하는 과정이 반드시 필요  
2. 순차적 데이터
    - 특징: 단어가 나열되는 순서에 따라 문장의 의미가 완전히 결정
        - 예) 사자가 사람을 먹다 / 사람이 사자를 먹다 (근데 사자가 사람먹을 확률이 높을듯)
    - NLP는: 데이터의 앞뒤 관계와 위치 정보를 유지하며 처리할 수 있는 모델 구조가 必
3. 중의성과 모호성
    - 특징: 동일한 형태의 단어라도 문맥에 따라 뜻이 천차만별
        - 예) 언어 "모델"과 패션 "모델"은 다른 단어 ->주변 단어를 봐야 뜻을 앎
    - NLP는; 단어 하나가 아닌 문맥을 전체적으로 파악해야
4. 희소성
    - 특징: 단어의 조합은 무한에 가깝지만, 우리가 가진 데이터에 나타나는 조합은 극히 일부
        - Zipf의 법칙: 경험적 법칙으로, 데이터셋에서 빈출 순위가 k번째인 항목의 빈도가 1위 항목 빈도의 1/k에 비례
    - NLP는: 한 번도 본 적 없는 문장 들어와도 *단어 간 유사도를 이용해* 확률적으로 추론할 수 있는 일반화 능력이 중요
5. 가변성과 진화
    - 특징: 언어는 고정 X 신조어, 은어, 오타, 이모지등 야민정음 문법 파괴괴
    - NLP는: 고정된 규칙 X 데이터로부터 유연하게 패턴 학습 가능해야
    

<span style="color:white; background-color:hotpink">언어 모델</span>

- 가장 넓은 의미에서 언어 모델은 인간이 말하고 쓰는 방식을 모방하는 시스템
- 언어 모델은 <span style="color:hotpink; background-color:grey">**단어 시퀀스(문장)**</span>에 확률을 할당하여, 가장 자연스러운 단어 시퀀스를 찾아냄  
    -> 언어 모델이 문장의 자연스러움을 판단하게 하는 보편방법: 예측도 올리기.
- 언어 모델(2) : 통계 기반, 인공신경망 기반

### <span style="color:crimson; background-color:white">통계적 언어 모델(SLM)</span>
Statisic Language Model

- SLM은 *분포 가설*에 근거
    -  <span style="color:black; background-color:yellow">비슷한 문맥에서 함께 나타나는 단어들은 비슷한 의미를 가진다.</span>  
      
    그럼 벡터화햇을 때 그래프의 비슷한 공간에 있게 될듯

- 조건부 확률
    - 조건부 확률을 통해 앞에 ‘버스를’이 오면 ‘탔다’가 올 확률을 활용하는 방식으로 진행

- 카운트 기반
    - 이전 단어 시퀀스의 등장 빈도를 통해 다음 단어의 확률을 계산. 
        - 희소 문제: 모델은 이전에 경험한 적 있는 조합에 대해서만 잘 예측할 수 있음  
        -> 새로운 문장 조합에도 적용할 수 있는 신경망 모델이 필요

<span style="color:crimson; background-color:white">N-gram</span>  
N-gram이 나오게 된 배경:  
SLM은 희소하게 등장한 것에 대해서는 잘 예측을 못하고, 문장이 길어질수록 다음 단어를 계산하기 어렵다.  
전체 문장 대신에 문장의 일부만 보는게 어떨까? 해서 나온 것이 N-gram



한계:
- 희소 문제(ㅇㄴ 희소 문제 해결하려고 만들엇는데 그럼에도 불구하고 계속잇군..)  
- 문맥 제한: 앞선 N-1개의 단어만 고려하므로, 문장 앞부분에 나온 정보를 뒷부분 결정에 반영하지 못함

<span style="color:crimson; background-color:white">Perplexity(PPL)</span>  
- PPL: 언어 모델이 문장을 얼마나 "혼란스러워하는지(perplexed)"를 수치로 나타낸 것
- 문장 W가 희소하게 등장할수록 PPL이 커지고, 단어 개수 N이 커질수록 PPL이 낮아진다. 

→ PPL이 낮다 = 모델이 예측에 덜 혼란스러움 = 모델이 문맥을 잘 이해함 = 모델의 성능이 좋음

<span style="color:hotpink; background-color:grey">**전이 학습**</span>

1) 사전 학습(pre-training..GPT)
- 방법: 레이블이 없는 방대한 데이터(인터넷 전체)를 스스로 학습하게 함

2) 미세 조정(파인 튜닝)
- 방법: 사전 학습된 모델 위에 우리가 해결하고 싶은 특징 작업(예: 감성 분석, 번역)의 데이터를 아주 조금만 더 학습시킴
- 결과: 적은 데이터로도 특정 분야의 전문가가 된 모델 탄생

<span style="color:white; background-color:hotpink">BERT</span>

### <span style="color:cyan; background-color:grey">**사전 학습**</span>

(1) MLM (masked language model)  
  

- **빈 칸 채우기 방식** 으로 학습
- “I go to school.”이라는 문장으로 예를 들면, 이 중에 무작위로 go를 빈칸을 뚫어서 I __ to school 이라고 만들고, 이 빈칸에 어떤 단어가 들어갈 지 모델이 예측하면서 학습

- Input 텍스트의 15%정도의 단어를 랜덤으로 마스킹
    - 빈칸을 뚫는 방식, 랜덤 단어로 바꾸는 방식, 그대로 유지하는 방식을 섞어서 사용

(2) Next Sentence Prediction(NSP)
    - 두 문장에 대해, 이 문장들이 서로 앞뒤 관계인지, 아니면 그냥 무작위 위치에 있는 두 개의 문장인지를 학습
    - 두 문장이 이어지는 논리 흐름을 학습

### <span style="color:crimson; background-color:white">5. sLM (small Language Model)</span>

모델 압축
- 가지치기
- 양자화 (계산 부하 줄이고 추론 속도 높임)
- 저차원 분해
- 지식 증류
