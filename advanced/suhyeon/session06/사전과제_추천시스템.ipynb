## 추천 시스템

#### ① 명시적 데이터 (Explicit Data)
- 고객들에 의해 입력된 데이터. 사용자가 선호도를 직접적으로 표현한 데이터
    - ex) 평점, 좋아요, 투표, 스케일 점수
- 모든 사용자가 리뷰나 평점, 별점 남기지 x
- 사람들 취향은 시간따라 변함

#### ② 암시적 데이터 (Implicit Data)  
- 행동 기반 데이터. 간접적 데이터
    - ex) 조회수, 영상시청 기간, 구매 기록, 페이지 체류 기간
- 발자취. 사용자가 서비스를 이용하면 로그데이터나 검색/클릭 기록은 무조건 남는다.
- 데이터 얻기 쉽지만, 직접적인 데이터는 아님



#### ① 내용 기반 필터링 (Content-Based Filtering)
- 콘텐츠의 유사도를 기반으로 사용자에게 추천하는 방법
- 사용자가 이용하거나 선택한 아이템과 유사한 아이템을 추천하는 방식



#### ② 협업 필터링 (Collaborative Filtering)
- 서비스 내에 있는 **다수의 사용자로부터 얻은 선호도 정보**를 기반으로 사용자에게 추천하는 방법
- 스포티파이가 사용하는 필터링!
- 두 명의 사용자가 비슷한 관심사를 가지고 있다면, 한 사용자의 데이터를 바탕으로 다른 사용자에게 추천해 주는 방식, 사용자 간의 선호도를 서로 고려하여 많은 선택 사항들로부터 아이템을 걸러내거나 선택
    1. 사용자 기반 (User-Based Filtering): 특정 사용자와 유사한 취향의 사용자가 선호한 아이템 추천
    2. 사용자 기반 (User-Based Filtering): 특정 사용자와 유사한 취향의 사용자가 선호한 아이템 추천  
  

#### ③ 하이브리드 필터링 (Hybrid Filtering)





### 2. 내용 기반 필터링

사용자 프로필

1. 직접 지정형  
: 사용자에게 자신이 선호하는 아이템의 특징을 명시적으로 지정하도록 하는 방법  
ex) OTT 서비스 가입시, 제시하는 몇 개 영화들에 대한 선호도를 체크해야 하는 경우  

2. 간접 지정형 
: 사용자의 과거 행동 이력에 기반하여 사용자의 프로필을 작성하는 방법  
ex) SF영화를 즐겨본 이력이 존재하는 유저에게 인터스텔라를 추천해주는 경우  



1. 특성 가중치는 사용자가 가지고 있는 아이템 특성 가중치의 평균 값을 활용
2. 사용자의 기호를 따로 파악하거나, 설문조사 또는 자발적인 키워드 입력 등으로 프로필을 획득
3. 아이템에 대한 유저의 기존 평가를 바탕으로 분류(Classification)문제를 해결하는 것

#### 2-2. 유사도
- 유사도 : 추천시스템을 이해하는데 중요한 수학적 개념
    1. 영상, 이미지, 텍스트 등의 형태로 존재하는 아이템 정보 데이터를 벡터화한 이후, 아이템 벡터간의 유사도를 측정
    2. 측정된 유사도를 바탕으로 고객이 선호하는 아이템과 비교하여 유사도가 높은 아이템을 선별

- 전체 문서에서 특정 단어가 얼마나 자주 등장하는지를 의미하는 역문서 빈도(IDF)

#### 3-2. 메모리 기반 협업 필터링의 한계

### 1. 희소 행렬 문제
- 주어진 평점/선호도 데이터가 적을 경우, 성능이 저하됨 (Sparsity)
    - 실제로 유저-아이템 행렬 대부분의 원소는 비어 있음 (sparse matrix, 희소 행렬)
- 확장성 문제
    - 아이템/유저가 계속 늘어날 경우, 확장성이 떨어짐 (Scalability)
- 콜드 스타트 문제
    - 협업 필터링은 기존의 평점/행동 데이터를 기반으로 유사도를 계산하는데, 데이터가 아예 없는 경우 추천이 불가능
    
| 유형 | 상황 | 설명 |
| --- | --- | --- |
| **User Cold Start** | 신규 유저 유입 | 평가 이력이 없어 유사 유저 탐색 불가 |
| **Item Cold Start** | 신규 아이템 등록 | 아무도 평가하지 않아 유사도 계산 불가 |



#### 3-3. 모델 기반 협업 필터링 (Model-Based Collaborative Filtering)

메모리 기반 CF(NBCF)의 한계(Sparsity, Scalability, Cold Start)를 보완하기 위해 등장한 방식으로, 단순히 유사도를 계산하는 것이 아니라, 데이터로부터 패턴을 학습한 모델을 만들어 평점을 예측함

메모리 기반 협업 필터링은 관측된 평점 정보만으로 유사도를 계산하고 추천을 수행하지만, 이는 숨겨진 선호 요인이나 아이템의 특성을 반영하지 못하고, 단순히 유사도에 의존하기 때문에 결측이 많은 환경이나 복잡한 상황에서 정확도가 떨어짐...

#### 행렬 분해 (Matrix Factorization)
- User-Item Matrix (유저-아이템 행렬)에는 유저의 취향+아이템의 특성이 녹아 있지만, 그 결과값만 볼 수 있다.

## Part 2. 딥러닝 기반 추천 시스템

### 1. VAE 기반 협업 필터링 (Variational Autoencoder CF)


- 사용자의 아이템 상호작용 벡터(예: 구매/시청 이력)를 입력으로 받아, 잠재 공간에서 재구성하도록 VAE를 학습하는 방식입니다.
- 구조:  
인코더: 사용자 이력을 잠재변수 z 분포로 매핑  
디코더: z로부터 아이템 분포를 복원  

#### 1-1. MF의 한계와 딥러닝의 등장

| MF의 근본적 한계 | 내용 | 딥러닝의 해법 |
| --- | --- | --- |
| **선형성의 한계** | 평점 예측($\hat{r}_{u, i}$)이 단순한 잠재 벡터의 내적(Dot Product)으로만 이루어집니다. | 비선형 활성화 함수(ReLU 등)를 가진 신경망을 사용하여 복잡한 패턴을 학습합니다. |
| **불확실성 무시** | 사용자 취향을 단 하나의 벡터값(Point Estimate)으로만 표현합니다. | **확률 분포**로 취향을 모델링하여 데이터의 불확실성을 반영합니다. |

#### 1-2. 1-2. 왜 AutoEncoder(AE)가 아닌 VAE(Variational AE)?

→ VAE는 불확실성을 모델링함으로써 추천 시스템의 고질적인 문제인 희소성(Sparsity)에 대한 해답을 제시

**문제 상황 비유: 데이터 희소성과 불확실성**

사용자 A가 본 영화는 5편에 불과합니다. (매우 희소한 데이터 - 즉 희소 행렬 문제!)

- **AE 접근 방식:** "A의 취향은 **(액션 0.9, 로맨스 0.1)** 벡터 하나일 것이다!" (단일 값 추정)
    - **문제점:** 데이터가 부족하면 이 단일 추정치가 불안정하고 틀리기 쉬워, **노이즈에 취약**합니다.
- **VAE 접근 방식:** "A의 취향은 (평균 액션 0.8, 표준편차 0.2)를 따르는 분포일 것이다." (분포 추정)
    - **강점:** 취향의 불확실성(데이터 부족으로 인한 오차 범위)까지 모델링에 포함시켜 안정성을 높입니다.

| 구분 | AutoEncoder (AE) | Variational AutoEncoder (VAE) |
| --- | --- | --- |
| **잠재 벡터 표현** | 단일 값 (Point Estimate) | **확률 분포** (평균 $\mu$, 분산 $\sigma$) |
| **데이터 불확실성** | 반영 불가 (단순 복원에 집중) | **데이터의 불확실성을 모델링**하여 안정성 확보 |
| **적용 시 장점** | 비선형성 학습 | 희소한 데이터에서 더욱 안정적이며, 일반화 성능 우수 |






#### 1-3. VAE 기반 CF (Mult-VAE)의 작동 구조

| 구성 요소 | 역할 |
| --- | --- |
| **입력** ($x$) | 사용자의 아이템 상호작용 벡터 |
| **인코더** | 사용자 이력을 잠재변수 $z$ 분포로 매핑 |
| **잠재 벡터** ($z$) | 샘플링된 최종 취향 벡터 |
| **디코더** | $z$로부터 아이템 분포를 복원 |

### 2. GNN 기반 사용자 선호 예측
VAE와 같은 딥러닝 잠재 요인 모델도 ‘관계’나 ‘순서’라는 풍부한 정보를 활용하지 못함  
이 ‘관계’ 자체를 데이터로 보고 모델링

-> 

#### 2-1. GNN (Graphic Neural Network, 그래프 신경망)



GNN은 사용자와 아이템의 상호작용을 ‘누가(User) 무엇을(Item) 좋아했다’의 집합으로 보는 것을 넘어, 이들이 서로 연결된 거대한 관계망(Graph)을 형성한다는 점에서 모티브를 얻음



- 관계, 상호작용과 같이 다소 추상적인 개념을 다루기에 적합
- Non-Euclidean space도 표현하고 학습할 수 있음


**GNN 작동 구조**

1. **[그래프 구조 생성]** 노드는 개별 객체, 엣지는 객체 간 관계나 연결성을 의미함
2. **[메시지 패싱]** 각 노드가 이웃 노드로부터 정보를 받아 전달
    1. **[메시지 생성]** 이웃 노드로에 전달할 메시지 생성
    2. **[집계]** 받은 여러 이웃의 메시지를 하나로 합침 
    3. **[갱신]** 자신의 노드 상태를 업데이트함
3. **[최종 출력]**




이 과정을 K번 반복하여 더 멀리 있는 이웃의 정보까지 반영 -> 이를 통해 이웃들의 취향까지 반영한 최종 임베딩







메세지 패싱





타겟 노드 선정 -> 메세지 생성 -> 집계 -> 갱신
