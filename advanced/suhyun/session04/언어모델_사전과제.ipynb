# **NLP와 언어모델**

## 1. NLP (Natural Langauge Processing)
- 자연어: 인간이 소통을 위해 자연스럽게 발생시켜 사용하는 언어
    - 이산적 데이터: 언어는 >단어<라는 독립적인 단위로 구성 -> 연속적인 벡터로 변환하는 과정 필요
    - 순차적 데이터: 단어의 순서에 따라 의미 결정 -> 데이터의 위치 정보를 유지하며 처리할 수 있는 모델 구조 필요
    - 중의성과 모호성: 동일한 단어여도 문맥에 따라 뜻이 다름 -> 문맥 파악 기술 필요 
    - 희소성: 단어의 조합은 무한에 가깝지만, 우리가 가진 데이터에 나타나는 조합은 극히 일부 -> 처음 보는 문장이어도 유사도를 통해 추론할 수 있는 일반화 능력 필요
    - 가변성과 진화: 언어는 수시로 변동 ->  데이터로부터 유연하게 패턴을 학습하는 머신러닝 방식 필요
- 프로그래밍 언어: 컴퓨터가 이해하는 프로그램 언어(Python, Java)
- NLP: 자연어의 의미를 분석하고, 컴퓨터가 처리할 수 있도록 하는 인공지능(AI) 기술 분야
    - 비정형 데이터인 글자를 컴퓨터가 계산할 수 있는 '숫자(Vector)'로 바꾸고, 그 안에서 의미를 추출
<br>
## 2. 언어 모델
- 인간이 말하고 쓰는 방식을 모방하는 시스템
- 단어 시퀀스(문장)에 확률을 할당하여, 가장 자연스러운 단어 시퀀스를 찾아내는 모델
    - 이전 단어가 주어졌을 때 다음 단어를 예측하도록 훈련시킴

# **2. 통계적 언어 모델 (SLM)**

## 1. SLM(Statistic Language Model)
- 분포 가설(Distributional Hypothesis)에 근거
    - 분포 가설: 비슷한 문맥에서 함께 나타나는 단어들은 비슷한 의미를 가짐
        - 조건부 확률: 앞에 단어가 있을 때 뒤에 특정 단어가 올 확률 활용
        - 카운트 기반: 이전 단어 시퀀스의 등장 빈도를 통해 다음 단어의 확률을 계산
            - **희소 문제**: 모델이 이전에 경험한 적 있는 조합에 대해서만 잘 예측할 수 있음
<br>
## 2. N-gram
  
- SLM은 문장이 길어질수록 다음 단어 계산 어려움 -> 전체 문장 대신에 문장의 일부만 보도록 함
- N-gram의 한계: 
    - 희소 문제: 
        - 데이터 부족: 훈련 데이터에서 나타나지 않은 단어 조합(N-gram)은 확률이 0이 되어, 해당 단어 시퀀스를 예측하거나 평가할 수 없음
        - N의 크기(n-gram)가 커질수록 훈련 데이터에서 해당 조합을 찾기 어려워져 희소 문제가 심해짐
    - 문맥 제한:
        - 앞선 N-1개의 단어만 고려하므로, 문장 앞부분에 나온 정보를 뒷부분 결정에 반영하지 못함
<br>
## 3. Perplexity(PPL)
- PPL: 언어 모델이 문장을 얼마나 "혼란스러워하는지(perplexed)"를 수치로 나타낸 것
    - 문장 W가 희소하게 등장할수록 PPL이 커지고, 단어 개수 N이 커질수록 PPL이 낮아짐

# **3. 신경망 언어 모델 (NLM)**

- 워드 임베딩(Word Embedding): 모든 단어를 거대한 다차원 지도 위의 한 점(좌표)으로 보냄
      
- NNLM은 단어를 '기호'에서 '의미를 가진 숫자(벡터)'로 바꿈으로써, 데이터에 없는 문맥도 어느 정도 예측할 수 있는 일반화 능력을 얻게 됨 
 
| 특징        | 통계적 모델 (N-gram)   | 신경망 모델 (NNLM)       |
| --------- | ----------------- | ------------------- |
| 미학습 단어 조합 | 확률 0 (예측 불가)      | 유사 단어를 통해 확률 추론 가능  |
| 저장 공간     | 데이터가 늘수록 기하급수적 증가 | 가중치(Weight)만 저장하면 됨 |
| 의미 파악     | 불가능 (단순 빈도수)      | 가능 (임베딩 벡터 간 거리 계산) |



# **4. 딥러닝 기반 언어모델**

## 1. Transformer
- RNN과 LSTM의 한계:
    - 병렬화 불가
    - 기울기 소실 및 폭주
    - 장기 의존성 문제
- Transformer: 인코더와 디코더가 여러 층으로 구성 => 신경망을 깊게 쌓으면서,  언어 모델의 표현력(capacity)이 높아져 현재의 딥러닝 기반 언어모델을 만들 수 있었음
<br>
## 2. BERT(Bidirectional Encoder Representations from Transformers)
- 텍스트의 **앞뒤 문맥**을 동시에 고려해서 입력값을 벡터로 바꿔주는 딥러닝 모델
- 트랜스포머 구조의 encoder 부분만 사용
- BERT 학습:
      
    - 처리과정: input → Embedding → 트랜스포머 인코더 12개(or 24개) → Output
    - 사전학습: 
        - Masked Language Model(MLM)
              
            - 랜덤 마스킹, 빈칸 채우기 방식 
        - Next Sentence Prediction(NSP)
            - 두 문장에 대해, 이 문장들이 서로 앞뒤 관계인지, 아니면 그냥 무작위 위치에 있는 두 개의 문장인지를 학습
- BERT 이후 모델:
    - RoBERTa: BERT의 사전 훈련 프로세스를 개선하여 성능을 향상시킨 실전형 모델
    - ALBERT (A Lite BERT): BERT의 경량화 버전으로, 가볍지만 성능은 유지
<br>
## 3. GPT
- **decoder 기반**, 지금까지의 단어를 바탕으로 다음 단어를 예측
    - zero-shot: 예시 없이 문제 지시만 주고 task 수행
    - one-shot: 1개의 예시만 보고 task 수행
- GPT vs BERT

| 항목                 | BERT                     | GPT                         |
| ------------------ | ------------------------ | --------------------------- |
| 방향성                | 양방향 (Bidirectional)      | 단방향 (Left-to-Right)         |
| 어떻게?               | 한 단어를 예측할 때 앞뒤 문맥을 모두 참고 | 다음 단어를 예측할 때 앞 단어만 참고       |
| 사용 목적              | 문장의 이해(인식)               | 문장의 생성(생산)                  |
| 대표 태스크             | 문장 분류, 질의응답, NER, NSP 등  | 텍스트 생성, 요약, 번역, 대화 등        |
| Self-Attention 마스크 | 제한 없음 (모든 단어 참조 가능)      | Masked Attention (미래 단어 가림) |

## 4. LLM
  
- 대규모 학습: 방대한 데이터를 학습 -> 어와 단어 사이의 관계, 문맥적 의미, 문법 구조, 그리고 세상의 지식까지 언어의 패턴을 스스로 터득
- 예측 및 생성: 프롬프트를 받으면 학습한 패턴을 기반으로 다음에 올 가장 확률 높은 단어를 순서대로 예측하며 응답을 생성
- LLM의 한계: 
    - hallucination: 그럴 듯하지만 틀린 정보를 말 하는 것
    - 업데이트 비용: LLM은 파라미터 수가 엄청나게 많기 때문에 새로운 정보를 반영할 때 계산량이 많음
<br>
## 5. RAG
- 질문에 대해 모델이 모르면 검색해서 답하는 구조
  
- 질의 인코더(Query encoder): 사용자의 질문을 벡터로 바꿈
- 지식 검색기(Knowledge Retriever): 인코딩된 질문을 바탕으로 외부 지식에서 관련 내용을 검색
- 지식 증강 생성기(Knowledge Augmented Generator): 검색한 내용을 참고해서 답변을 생성
- RAG의 장점: 최신 정보 반영 가능, 환각 감소, 정보 정확성 증가
- CAG: 문서의 신뢰도를 스스로 판단하고 활용할 수 있도록 훈련시키는 기법
    - 질문 입력 → 정보 검색 → **각 정보의 신뢰도 평가** → 높은 신뢰도 정보 우선 사용 → 최종 답변 생성
<br>
## 6. LangChain
- LLM 앱을 빠르게 조립할 수 있는 레고 블럭
- LangChain의 특징
    - 추상화: LLM 서비스를 만들 때 필요한 각종 작업을 간결하게 표현(간소화)
    - 표준화: 비슷한 기능을 갖추고 있는 요소들을 똑같은 형식을 갖춘 컴포넌트로 표준화
    - 체이닝: 자주 활용하는 주요 컴포넌트를 쉽게 연결해서 LLM 서비스의 로직을 쉽게 파악 (Input → 컴포넌트 → Output)
- LangGraph:
      
    - 비선형적인 작업 형태로, 사이클 (Cycle), 루프 (Loop)를 만들 수 있음
    - 노드 (Node): 특정 작업을 수행하는 단위로, 랭체인의 컴포넌트를 그대로 가져와 노드로 사용 할 수 있음
    - 엣지 (Edge): 노드와 노드를 연결하는 선으로, 어떤 노드에서 다음 노드로 이동할지를 결정
- LangSmith:
    - 대규모 언어 모델(LLM) 기반 애플리케이션을 위한 통합 개발 및 운영 플랫폼
    - 핵심 기능:
        - 디버깅 및 추적: LLM 애플리케이션의 모든 실행 과정을 단계별로 시각화
        - 테스트 및 평가: 미리 준비된 질문과 정답 데이터셋을 만들어, 개발 중인 LLM 애플리케이션의 성능을 자동으로 평가
        - 모니터링: 실제 사용자들이 애플리케이션을 어떻게 사용하는지 모니터링
        - 프롬프트 허브: 다양한 프롬프트를 저장하고, 버전을 관리하며, 쉽게 테스트할 수 있는 공간 제공

# 5. sLM (small Language Model)

- sLM: 자연어 콘텐츠를 처리, 이해 및 생성할 수 있는 AI 모델, 대규모 언어 모델(LLM)보다 규모와 범위가 작음
- 모델 압축:
    - 가지치기: 신경망에서 중요도가 더 낮거나 중복되거나 불필요한 매개변수를 제거
    - 양자화: 고정밀 데이터를 저정밀 데이터로 변환
    - 저차원 분해: 큰 가중치 행렬을 더 작은 저차원 행렬로 분해하는 것
    - 지식 증류: 사전 학습된 "교사 모델"의 학습 내용을 "학생 모델"로 이전하는 것
- sLM의 장점:
    - 효과적인 성능
    - 지연 시간 감소
    - 보안: 민감한 데이터를 외부 클라우드 API로 보내지 않아도 됨
    - 접근성: sLM은 고성능 GPU 서버가 없어도 사용 가능→ 진입 장벽 ↓
    - 효율성: 저장 공간, 메모리, 연산 자원을 적게 사용 → 같은 작업을 처리할 때 더 적은 자원으로 결과 도출
    - 비용 절감: 모델이 작기 때문에 추론 비용, 서버 비용, GPU 비용, API 호출 비용 ↓
- sLM의 한계:
    - 편향: 더 큰 모델에 존재하는 편향까지 함께 전달 가능 → 성능 저하로 이어질 수 있음
    - 제한된 일반화 & 복잡한 작업에서의 성능 저하: 특정 작업에 대해 미세 조정하기 때문에 포괄적인 지식이 필요한 작업에 취약
    - 할루시네이션(환각): 정보가 적고 추론 능력이 낮으므로 당연히 발생 가능
-  LLM과 sLM의 결합 (결합 추론)
