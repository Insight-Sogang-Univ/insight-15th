# *Part 1. 추천시스템 기초*

# **1. 추천 시스템** 

## 1. 추천 시스템의 개념과 특징
- 추천 시스템: 사용자의 정보 데이터를 분석, 개인의 취향에 맞는 아이템을 추천하는 알고리즘
- 사용되는 데이터:
    - 명시적 데이터: 고객들에 의해 직접적으로 입력된 데이터
    - 암시적 데이터: 고객들의 행동을 기반으로 수집된 데이터
<br>
## 2. 추천시스템의 분류
- 내용 기반 필터링: 콘텐츠의 유사도를 기반으로 사용자에게 추천하는 방법
- 협업 필터링: 서비스 내에 있는 다수의 사용자로부터 얻은 선호도 정보를 기반으로 사용자에게 추천하는 방법
    - 사용자 기반: 특정 사용자와 **유사한 취향의 사용자**가 선호한 아이템 추천
    - 아이템 기반: 사용자가 선호한 아이템과 **유사한 아이템**을 추천
- 하이브리드 필터링: 2가지 이상의 다양한 종류의 추천시스템 알고리즘을 조합하여 만들어진 알고리즘

# **2. 내용 기반 필터링**

## 1. 사용자 프로필
- 직접 지정형: 사용자가 선호하는 아이템의 특징을 명시적으로 지정하도록 하는 방법
- 간접 지정형: 사용자의 과거 행동 이력에 기반하여 사용자의 프로필을 작성하는 방법
- 프로필 획득 방법:
    - 특성 가중치: 사용자가 가지고 있는 아이템 특성 가중치의 평균값
    - 사용자의 기호를 따로 파악하거나 설문조사 또는 자발적인 키워드 입력
    - 사용자의 기존 평가를 바탕으로 분류 문제를 해결하는 것돠 비슷함
    => 다양한 방식 혼합해서 사용
<br>
## 2. 유사도
- 유사도: 
    1. 영상, 이미지, 텍스트 등의 형태로 존재하는 아이템 정보 데이터를 벡터화한 이후, 아이템 벡터간의 유사도 측정
    2. 측정된 유사도를 바탕으로 고객이 선호하는 아이템과 비교하여 유사도가 높은 아이템 선별
- 유사도 사용 과정:
    1. 인풋 데이터 생성: 아이템 정보 활용 -> 테이블 형태로 정리 -> 숫자로 변환
    2. 유저 정보 활용: 유저의 취향 파악 후 유저를 n개의 피처로 표현
    3. 벡터의 내적과 유사성: 벡터 내적 후 유사도 측정
- **TF-IDF(Term Frequency - Inverse Document Frequency)**
  
- 단어 빈도(TF): 특정 문서 내에 특정 단어가 얼마나 자주 등장하는지
- 역문서 빈도(IDF): 전체 문서에서 특정 단어가 얼마나 자주 등장하는지
=>"다른 문서에서는 등장하지 않지만 특정 문서에서만 자주 등장하는 단어"를 찾아서 
문서 내 단어의 가중치 계산
<br>
## 3. 내용 기반 필터링의 장단점
- 장점:
    - 아이템이 가지고 있는 feature 자체에만 집중하기 때문에 다른 유저의 평점 등에 영향을 받지 않음→ 콜드스타트 문제에서 자유로움
    - item의 feature를 통해서 추천을 하기 때문에 추천이 된 이유를 설명하기 용이
- 단점: 
    - 아이템의 정확한 feature를 추출하지 못하면 정확도가 떨어짐
    - 새로운 장르의 item을 추천하기 어려움

# **3. 협업 필터링**

## 1. 메모리 기반 협업 필터링 (Memory-Based Collaborative Filtering)
- 사용자 기반: 특정 사용자와 타 사용자 간의 유사도를 측정한 뒤 가장 유사도가 높은 TOP-N 사용자를 추출해 그들이 선호하는 아이템을 추천
  

- 아이템 기반: 사용자들이 그 아이템을 좋아하는지/싫어하는지의 평가 척도가 유사한 아이템을 추천하는 기준이 되는 알고리즘
  

- 비교:

| 구분               | User-based CF (UBCF)                             | Item-based CF (IBCF)                    |
| ---------------- | ------------------------------------------------ | --------------------------------------- |
| **기준**           | 나와 비슷한 **사용자**를 찾아 추천                            | 내가 좋아한 것과 비슷한 **아이템**을 찾아 추천            |
| **추천 정확성 & 효율성** | 아이템 수가 사용자 수보다 **많은 환경**에서 유리                    | 사용자 수가 아이템 수보다 **많은 환경**에서 유리           |
| **안정성**          | 아이템이 **수시로 바뀌는 환경**에 적합                          | 아이템 리스트 변화가 적고 **고정적인 환경**에 적합          |
| **다양성**          | 특정 사용자가 독특한 아이템을 평가한 경우, **뜻밖의 독창적인 아이템** 추천 가능  | 유사한 아이템 위주로 추천되어 **다양성과 참신성은 상대적으로 부족** |
| **계산 속도**        | 사용자 유사도를 계속 계산해야 하므로 상대적으로 불안정할 수 있음             | 사용자별 개별 계산이 불필요해 **빠르고 안정적**            |
| **적합한 환경**       | 아이템이 자주 바뀌고 사용자 수가 적은 환경                         | 사용자 유입이 많고 아이템 구성이 비교적 안정적인 대형 플랫폼      |
| **실제 활용 예시**     | 뉴스, 소셜 미디어, 음악 초기 모델, 알 수도 있는 친구 추천, 음악 추천 서비스 등 | 아마존, 넷플릭스 등 대형 추천 플랫폼                   |


<br>

## 2. 메모리 기반 협업 필터링 한계:
    - 희소 행렬 문제: 주어진 평점/선호도 데이터가 적을 경우, 성능이 저하됨 (Sparsity)
        => model-based CF 이용  
    - 확장성 문제: 아이템/유저가 계속 늘어날 경우, 확장성이 떨어짐 (Scalability)
        => KNN CF 이용  
    - 콜드 스타트 문제: 기존의 평점/행동 데이터를 기반으로 유사도를 계산하는데, 데이터가 아예 없는 경우 추천이 불가능  
        => 직접 지정형 프로필 활용/Content-Based Filtering 과 혼합/인구통계학적 정보 (나이, 성별 등) 활용
<br>    

## 3. 모델 기반 협업 필터링 (Model-Based Collaborative Filtering)  
- 관측된 평점 데이터를 학습(Training) 하여, 관측되지 않은 평점을 예측(Inference) 하는 방식  
- 잠재 요인 (Latent Factor): 사용자와 아이템 사이의 관계를 설명해주는, 눈에 보이지 않는 숨겨진 특성(요인)  
- Latent Factor Model: 사용자-아이템 간의 숨겨진 특성(feature)을 찾기 위해 주어진 Matrix를 사용자-잠재요인 아이템-잠재요인으로 각각 분해하여 학습하는 모델  
  
=> 유저와 아이템을 동일한 잠재 공간의 벡터로 표현하여, 둘 사이의 거리·각도로 숨겨진 취향 유사도를 파악하는 모델  
- 행렬 분해:   
  
- 행렬 분해 방식:
    - Orthgonality Factorization(Singular Vector Decomposition: SVD) 특이값 분해: 
        - 행렬 대수학 기반
        - 𝐴: 원래의 𝑚×𝑛 행렬 (사용자-아이템 평점 행렬)
        - 𝑈: 𝑚×𝑚 직교 행렬 (사용자 관련 정보)
        - Σ: 𝑚×𝑛 대각 행렬 (가장 중요한 정보: 특잇값)
        - 𝑉𝑇: 𝑛×𝑛 직교 행렬 (아이템 관련 정보)
    - Stochastic Gradient Descent (SGD)
        - 관측된 평점과 예측 평점 사이의 오차를 최소화하는 것을 목적으로 함
    - Non-Negatice Matrix Factorization (NMF)
        - 모든 원소가 0 이상이어야 한다는 제약 조건을 추가한 행렬 분해 기법=>직관적 의미 부여 가능
    - Probabilistic Factorization (PMF): 누가 무억을 좋아할 지 예측하는 것이 목적

# **1. VAE 기반 협업 필터링**

- MF의 한계:
    - 선형성의 한계: 평점 예측이 단순한 잠재 벡터의 내적으로만 이루어짐 => 비선형 활성화 함수
    - 불확실성 무시: 사용자 취향을 단 하나의 벡터값으로만 표현 => 확률 분포 => VAE 사용하는 이유!
- 작동 구조:

    | 구성 요소         | 역할                       |
    | ------------- | ------------------------ |
    | **입력 (x)**    | 사용자의 아이템 상호작용 벡터         |
    | **인코더**       | 사용자 이력을 잠재변수 (z)의 분포로 매핑 |
    | **잠재 벡터 (z)** | 샘플링된 최종 취향 벡터            |
    | **디코더**       | (z)로부터 아이템 분포를 복원        |

- 목적 함수: ELBO(Evidence Lower Bound)
    - 재구성 오차 최소화: 디코더가 잠재 벡터 𝑧를 사용하여 원래 입력 𝑥를 최대한 비슷하게 복원하도록 만듦
    - KL Divergence: 인코더가 출력하는 잠재 분포 𝑞(𝑧∣𝑥)가 사전 분포 𝑝(𝑧)(표준 정규 분포)와 유사하도록 강제
- VAE-CF의 주요 장점:
    - 확률적 잠재 분포 모델링 (불확실성 반영)
    - 희소한 상호작용 데이터에서도 살아남음
    - 비선형성 학습

# **2. GNN 기반 사용자 선호 예측**

## 1. GNN
- 그래프
      
    - 노드(Node)와 그 노드를 잇는 선(Edge)을 모아 구성한 자료 구조
    - 추상적인 개념 다루기에 적합
    - 비유클리드 고간의 데이터도 표현/학습 가능
- GNN의 작동 구조:
    1. **[그래프 구조 생성]** 노드는 개별 객체, 엣지는 객체 간 관계나 연결성을 의미함
    2. **[메시지 패싱]** 각 노드가 이웃 노드로부터 정보를 받아 전달
        1. **[메시지 생성]** 이웃 노드로에 전달할 메시지 생성
        2. **[집계]** 받은 여러 이웃의 메시지를 하나로 합침 
        3. **[갱신]** 자신의 노드 상태를 업데이트함
    3. **[최종 출력]**
- GNN의 프로세스:
  
1. 그래프 구조 생성: 노드와 엣지를 정의한 후, 각각의 피처를 담아 표현
2. 메시지 패싱: 그래프에서 노드 간 정보를 전달하고 상호작용하는 방법
    - 메시지 생성 - 집계 - 갱신 3단계
        - 메시지 생성: 각 노드가 이웃에게 보낼 자신의 정보를 생성
        - 집계: 메시지 생성 이후, 여러 이웃 노드에서 받은 메시지를 하나의 벡터로 통합하는 단계
        - 갱신: 집계된 이웃 정보와 자신의 이전 상태를 결합해 새로운 임베딩을 계산
<br>
## 2. LightGCN
- 추천 시스템에 특화된 GNN
- GNN의 구성 요소 특징 변환(Feature Transformation) 제거, 비선형 활성화 함수 (Non-linear Activation) 제거
=> 정규화된(Normalized) 이웃 임베딩의 합(Aggregation)만으로 구성
- 장점:
    - 고차원 협업 시그널 (High-order Collaborative Signals) 포착: 나와 취향이 비슷한 사용자들(이웃)과 내가 본 아이템과 연관된 다른 아이템들의 정보를 직접 반복해서 반영 가능
    - 정형화되지 않은 데이터 학습 가능
    - Cold Start 문제 완화에 효과적
- 단점: 
    - 과평탄화 (Over-smoothing) 문제: 모든 노드의 임베딩이 같은 값으로 수렴하는 현상
    - 과압축 (Oversquashing) 문제: 고정된 크기의 벡터 하나에 모두 압축하려고 하면 정보가 손실될 수 있음
    - 계산 복잡도 ↑
    - 해석의 어려움

# **3. Transformer 기반 추천 시스템**

# 1. 기존 추천 시스템의 순차적 한계
- CF의 순차적 한계: 구매 순서 무시, 문맥 파악 부족, 예측의 한계
- GNN의 순차적 한계: 장기의존성 문제
=> CF와 GNN은 사용자의 일반적인 선호도, 관계를 파악하는 데 뛰어나지만, 행동의 ‘순서’와 ‘문맥’, 특히 동적인 의도를 파악하는 데 한계가 있음
<br>
# 2. Transformer: 사용자 행동을 시퀀스로 접근
- 시퀀스 내의 문맥과 핵심 의도 파악에 유리
  
- 문맥에 따라 달라지는 동일 아이템의 의미: 같이 클릭한 다른 아이템들의 의미에 따라 해석이 달라짐
- 시간 순서를 뛰어넘는 '핵심 의도'의 포착: 시간 순서나 거리에 관계없이 현재 의도와 가장 관련성이 높은 핵심 문맥에 주목
<br>
# 3. Transformer 기반 주요 추천 시스템 모델
- SASRec
    - 단방향 어텐션으로 과거 정보만 활용하여 미래 행동을 예측하는 방식으로 훈련을 받은 추천 시스템 모델
    - Training Action Sequence (학습용 행동 순서)
    - Embedding Layer (임베딩 레이어)
    - Self Attention Layer (셀프 어텐션 레이어)
    - Prediction Layer (예측 레이어)
    - 단점: 단방향성, 계산량 많음
- BERT4Rec
    - 양방향 어텐션으로 과거 정보와 미래 정보를 활용하여 예측하는 방법을 훈련한 모델
    - 입력 (Input): 사용자 행동 시퀀스 중 일부를 [mask] 토큰으로 의도적 치환
    - 임베딩 레이어 (Embedding Layer)
        - 아이템 임베딩: 아이템 고유 의미 벡터
        - 포지셔널 임베딩: 아이템 '순서' 정보 벡터
    - 트랜스포머 블록: 양방향(Bi-directional) 셀프 어텐션 적용
    - 양방향 문맥 모두 고려, 시퀀스 전체의 선호도 파악에 초점


# **4. 오늘날의 추천 시스템**

## 1. LLM
- LLM: 추천을 점수 계산이 아닌 문장으로 출력
- 대화형 추천: 자연어 요청을 이해하고 추천 가능
- Zero-shot 추천 : 학습 데이터에 없던 새로운 아이템도 아이템의 ‘텍스트 설명’만으로 그럴싸한 추천 가능 => 콜드스타트 문제의 강력한 해결책
- P5: 모든 추천 과제(랭킹, 다음 아이템 예측, 설명 생성 등)를 LLM이 풀 수 있도록 '하나의 거대한 텍스트 프롬프트' 문제로 통일시킨 모델
- TALLRec: 범용 LLM을 추천 작업에 맞게 효율적으로 튜닝하는 프레임워크

## 2. 인기 편향과 XAI 문제

- 설명가능성: 추천 이유 설명 가능해야 함
