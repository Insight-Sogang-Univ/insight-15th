## Attention의 등장 배경 
### **Seq2Seq의 병목 현상**  
: 기존 Seq2Seq 모델은 입력 문장의 모든 정보를 하나의 **고정된 context vector**에 압축  
-> 입력 문장이 길어질수록 *중요한 정보가 손실되는 문제* 발생  
-> *매번 같은 Context Vector만 참고*하므로 출력 단어마다 필요한 정보 다르게 반영하기 어려움  
=> Attention의 등장  

## Attention 장점  
: 중요한 정보에 더 집중하는 메커니즘  
     - 입력값의 모든 부분을 똑같이 참고하지 X  
     -> **현재 시점에서 가장 중요한 부분에 더 높은 가중치를 부여하여 활용**  
     - 출력 시점마다 **새로운 context vector를 동적으로 계산** -> *병렬화*  
    1) 시간 경과에 대한 유연성  
        - 모든 단어에서 다른 모든 단어로 직접 연결 가능  
        - 순서적 거리에 제약 없음  
    2) 공간에 대한 유연성  
        - 픽셀 간 거리와 무관하게 전체 이미지의 *전역적인 관계* 학습 가능  
    3) 병렬화  
        - self-attention에서 문장 내 모든 단어의 관계를 한번에, 독립적으로 계산 가능  

## 1-2. Attention is All you Need  
: Transformer 구조를 처음 발표한 Google Brain의 논문  
<br>
     - 인코더부터 디코터까지 *오직 Attention만 사용하는 모델* 개발  
     - self-attention에 기반한 Transformer 모델 제안  
     - 병렬화 가능  
     - 파급효과: LLM의 탄생, NLP 평정, 타 분야로의 확장  
    

> # **2. Self-Attention**  
## 2-1. Query, Key, Value
: 모델 안의 수많은 정보 내에서 *가장 관련 있는 정보를 효과적으로 찾아내기* 위해 역할 분할  
<br>
### Q : Query  
: '질문' 또는 '요청'으로, 당장 내가 알고 싶거나, 초점을 맞추고 있는 대상  
### K : Key  
: 검색 대상이 되는 **모든 정보들이 달고 있는 '이름표' 또는 '색인'**  
### V : Value  
: 키와 한 쌍으로 묶여 있는 **실제 내용물**  


## 2-2. 기존 Cross-Attention vs. Self-Attention  
### **Cross - Attention**  
     하나의 시퀀스가 완전히 다른 시퀀스를 참고하여 정보를 만들어내는 메커니즘  
1. Query 만들기  
2. Key, Value 불러오기  
3. 유사도 계산  
4. Weighted Sum  
5. Decoder가 다음 단어를 예측  
<br>

### **Self-Attention**  
     문장 자체를 뜯어보며 내부의 관계를 파악  
1. 단어의 프로필 만들기  
2. 단어 간 관계 점수 계산  
3. 중요도 배분  
4. 새로운 벡터 생성  



## 2-3. Scaled Dot-Product Attention  
### **Dot-Product Attention**  
    Attention Score를 계산하는 가장 기본적인 방법 중 하나
<br>

### **Scaled Dot-Product Attention**
<기존의 문제 해결 방법>  
: 계산된 값의 **스케일링** 진행  
-> 문장 내 다른 단어들과의 관계를 반영한 *문맥적 표현* 생성  

> # **3. Multi-Head Attention**  
## 3-1. Why Multi-Head Attention?
    한 단어와 다른 단어 간의 관계를 여러 차원으로 나누어 **병렬로** 학습 
-> Attention 스코어를 병렬로 여러 번 계산

## 3-2. Multi-Head Attention 작동 방식  
1. 분할- 각 헤드에 단어 벡터 할당  
2. 병렬 어텐션 계산 - 각 헤드들이 각자 분석 수행  
3. 결합 및 최종 투영  
    **결합**  
        - 각 헤트의 벡터들을 순서대로 이어 붙여 **하나의 거대한 벡터** 생성  
    **최종투영**  
        - 위의 나열된 상태의 정보들을 잘 융합하고 최종적으로 정리하기 위해, 다른 가중치 행렬을 곱해주는 과정  

> # **4. Transformer 전체 아키텍처**  
## 4-1. 트랜스포머의 전처리 단계  
    텍스트를 입력받고, 그 텍스트를 토큰화하고-> 토큰을 임베딩 -> Positional Encoding  
### 1.토큰화  
: 입력 텍스트를 모델이 처리할 수 있는 단위로 나누는 단계  
### 2. 임베딩  
: **숫자의 벡터**로 변환  
-> 단어의 의미를 *공간적으로 표현*  
### 3. Positional Encoding  
: 단어들의 **순서 정보**를 임베딩 벡터에 추가  


## 4-2. 트랜스포머의 Encoder와 Decoder  
### 1. Encoder  
    입력 문장을 이해하고 요약된 의미 벡터로 변환하는 역할  
: 전처리 과정 및 **인코더 레이어**로 구성  
<Feed-Forward Layer (FFN)>  
- 입력 벡터의 차원을 확장하고 비선형 변환을 적용하여 새로운 표현을 생성하는 신경망 구조  
<br>

### 2. Decoder  
    Encoder가 분석한 입력 문장의 의미 벡터를 받아서 출력 문장을 순차적으로 생성하는 역할  
: 전처리 과정, 디코더 레이어, 선형 레이어와 소프트맥스 레이어로 구성
