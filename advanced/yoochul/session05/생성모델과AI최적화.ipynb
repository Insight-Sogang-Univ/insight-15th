## 생성모델
---



### 분류모델과 생성모델

이전세션에서 해왔던 것들이 분류 모델입니다. 정답 레이블이 있고, 입력된 데이터가 어떤 레이블에 속할지 결정하는 모델로 알면 됩니다. 반면에 생성 모델은 샘플에 레이블 지정에 관여하지 않고, 데이터 자체의 확률 분포를 학습해 학습 데이터와 유사한 새로운 데이터를 만들어 냅니다.

둘의 차이를 예시를 들어 설명하자면 



생성모델은 메뚜기 월드를 만드는 박명수라고 생각하시면 됩니다. 유재석이라는 사람을 학습해서 이에 맞는 새로운 노래를 만드는 것. 그것이 생성모델이라고 이해하시면 되고,



분류모델은 이와달리 정답을 찾아주는 것으로 수능시험 보듯이 가장 정답에 가까운 것을 맞추는 모델이라고 생각해주시면 됩니다.


---

### 생성모델의 분류

- AE

오토인코더(AE) 는 입력과 동일한 출력을 만드는 것을 목적으로 합니다.

구조는 정보를 압축시켜 잠재 백터로 만들고 이를 다시 디코더로 재복원 시킵니다.

저는 이부분을 궁금하면서 제일 이해가 안되는 부분이 그냥 원본 데이터를 쓰면 되지 왜 압축시켜서 데이터를 일부분 남기고 이를 왜 다시 복원시키는지 이해가 안됬습니다. 찾아본 결과로는 원본 그대로 쓰는거는 크기도 크고 이상치도 많아서 몇가지의 정보를 제거하고 쓰는 것 입니다. 예를 들어 고양이 두마리를 학습 시켰을때





색깔은 쓸데없는 정보일것입니다. 그래서 중요한 정보인 귀가 2개이고 4발로 걷는다 이런 정보들만 남기는 것이 오토인코더입니다.

---

- VAE

Variational Auto Encoder(VAE)는 데이터의 잠재적인 구조를 학습해 원본과 비슷한 새로운 데이터를 생성합니다. AE와 차이점은 AE는 점으로 학습하고 VAE는 분포로 학습한다는 점입니다. 따라서 잠재공간이 AE와는 달리 빈 공간이 거의 없습니다.

---

- GAN

GAN(Generative Adversarial Network)은 생성자 신경망과 판별자 신경망이 서로 경쟁하고 훈련하며 점점 더 정교하게 수행합니다.

예시를 들어보겠습니다.

투수(생성자) 오타니와 타자(판별자) 쇼헤이가 있습니다.





투수는 타자를 속이기 위해 직구와 최대한 비슷한 변화구를 던지려 노력합니다.
타자는 반대로 이런 직구와 변화구를 구분하려고 최대한 노력합니다.
이렇게 서로 경쟁하며 점점 더 정교하게 수행하는 신경망 모델이 GAN입니다.
생성자는 실제와 가까운 데이터를 생성하고 판별자는 진짜와 가짜를 점점 더 정확히 판별해나가려 합니다.

이를 통해 빠르게 진짜같은 가짜를 만들고 학습된 잠재 공간에서 보간이 가능하다는 장점이 있지만, 학습이 불안전하고, 좋은지 나쁜지 판별할수 있는 지표가 없다는 단점이 있습니다.

- 확산모델

확산 모델(Diffusion Model)은 입력 이미지에 노이즈를 조금씩 추가해 완전한 노이즈로 만든 뒤 이를 다시 거꾸로 복원해 나갑니다. 어떤 모호한 개념을 하나하나 구체화 해가며 결과를 출력합니다.

이때 순확산과 역확산을 사용하는데 순확산은 데이터에 노이즐 추가하는 과정, 역확산은 노이즈를 제거하는 과정입니다.

# AI 최적화

### 전이학습

- 전이학습이란 무엇일까?

이전에 딥러닝 모델들을 이용해 AI를 만드는 과정에서 데잍터가 부족하고, 비용이 증가하는 등의 문제들을 겪습니다. 이를 위해 이미 어디선가 학습된 내용을 가져다가 쓰는 것이 전이학습입니다.

이때 도메인과 태스크 ,소스와 타겟을 알아야 하는데
도메인은 어떤 데이터를 다루는가
태스크는 뭘 하는가 (뭘 예측할건가)
소스는 출발점
타겟은 도착점
으로 단어뜻과 거의 비슷하게 생각하시면 됩니다. 데이터나 문제 유형이 달라도 배운 지식을 가져와 새로운 상황에서 성능을 높이는 방법입니다.



이번에는 현우진 T에게 수학을 배운다고 가정을 해보겠습니다. 저희는 수능을 보기 전까지 확원 또는 학교와 같이 각자의 환경에서 현우진 선생님에게 수학을 배웁니다. 그 와중에 실전 모의고사를 풀기도 합니다. 이 문제가 수능과 절대로 같지는 않겠지만, 이떄 배운 지식들을 통해 수학문제를 푸는것과 유사하다고 볼 수 있습니다.

- 전이학습의 분류

이런 전이학습은 3가지로 분류합니다.
소스와 타겟의 태스크가 다르고 도메인의 라벨이 존재하는 귀납적 전이학습

스스와 타겟의 태스크는 같지만, 도메인은 다른 트랜스듀서형 전이학습

소스와 타켓의 태스크가 다르고 도메인의 라벨도 없는 비지도 전이학습이 있습니다.

-  Pre-training & Fine-tuning 

트랜스포머의 등장 이후, 판이 완전히 바뀌었습니다. 엄청난 양의 데이터로 사전 학습된 거대 모델들을 이제 어떻게 자기 용도에 맞게 조정하는 것으로 트렌드가 바뀌었습니다.
GPT나 Gemini 혹은 클로드와 같은 파운데이션 모델들을 자기에 맞게끔 조정합니다.

### AI Alignment

AI Alignment는 모델의 성능을 높일 뿐 아니라 의도,목표,그리고 가치에 부합하도록 행동하게 만드는 연구입니다. 저번에 서영이가 세종대왕이 맥북던진걸 설명했던 것처럼 사실이 아닌 정보를 지어내거나 혹은 편향된 정보로 답변하는 문제를 줄이기 위해 세가지 지표를 도입해 해결하려 했습니다.



요즘은 발전해서 그렇지 않습니다~

정렬은 크게 두가지로 하는데 

Outer Alignment : 사람과 같이 어떤 행위를 하기 위해서 방법을 가리지 않는것으로 생각하면 됩니다. 문제를 풀어라 했는데 정답여부와 상관없이 푸는것이 그 예시일 수 있습니다.

Inner Alignment : 훈련 중에는 인간의 감시를 피하기 위해 정렬된 척 행동하다가, 실제 배포된 후(감시가 없을 때)에는 자신의 원래 목표를 수행하는 '기만적 행동(Deceptive Alignment)'이 나타날 수 있어요.

이를 극복하기 위해서 RLHF(Reinforcement Learning from Human Feedback) 를 통해 진행합니다. 어떤 대답이 정확한지를 넘어서 어떤 대답이 더 인간이 선호하는지를 알 수 있습니다. (어쩐지 요즘 억빠가 심하더라) 하지만 이는 인간이 좋아할만한 대답을 골라한다는 단점이 있습니다.



그리고 프롬프트 엔지니어링을 통해 더 좋은 답변을 내도록 설계하기도 합니다.

생각을 단계별로 시키거나, 역할을 부여하고, 지시를 명확하게 하며, 여러번 물어봐서 가장 많이 나온 답을 채택하는등의 방법이 있지만, 거짓말을 완전히 없앨수 없고  Fine-tuning 수준의 성능을 내기 어렵다는 한계가 있습니다.
