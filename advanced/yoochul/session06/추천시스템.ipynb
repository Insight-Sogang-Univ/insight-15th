# 추천시스템



### 추천시스템이란?



먼저 정의를 알아보겠습니다. 추천시스템이란 사용자의 정보 데이터(명시적 데이터, 암시적 데이터)를 분석, 사용자의 취향에 맞는 아이템을 추천하는 알고리즘으로 위에 점메추표를 바탕으로 예시를 들자면, 개인의 취향을 바탕으로 사용자가 다이어트 중이고 최근에 다른음식들을 먹어봤으니 오늘은 연어셀러드가 좋을거 같습니다. 와 같이 추천해주는 알고리즘을 추천시스템이라 합니다.



---

### 추천시스템의 분류

#### 내용기반 필터링

내용기반 필터링은 개인의 취향을 반영해서 추천을 해주는 것입니다.



지금 제가 데이식스의 노래를 듣는 중이라서 옆에 데이식스 노래들과 이와 비슷한 음악들이 알고리즘에 뜨는 것을 예시로 들 수 있습니다.

다음으로 내용기반 필터링을 할때 사용되는 데이터들을 알아보겠습니다.


- 사용자 프로필
사용되는 데이터는 직접,간접 지정형 두가지로 나뉩니다. 직접은 사용자가 말해주는것으로 핸드폰으로 옷을 살때 키와 몸무게등을 어플에서 물어보는 경우를 생각하시면 되고, 간접은 유튜브 알고리즘 같이 과거행동을 기반으로 알 수 있는것입니다.

크게 3가지 방법이 있는데 이들을 예시와 함께 알아보겠습니다.



무신사를에서 옷을 팔때 

1. 특성 가중치를 활용해 사용자가 구입했던 아이템들을 바탕으로 추천



사용자가 무신사에서 스트릿, 캐주얼, 스포츠 스타일 상품을 자주 구매했다면, 시스템은 해당 사용자가 이러한 스타일을 선호한다고 판단합니다.

이후 새로운 스트릿·캐주얼 계열 상품이 등록되면, 사용자의 취향과 유사도가 높다고 판단하여 추천할 수 있습니다.

2. 사용자의 정보를 직접 입력받아 프로필을 만듭니다.



무신사에서는 사용자가 키·몸무게·선호 스타일 등을 입력할 수 있습니다. 사용자가 자신의 체형 정보를 입력하면, 비슷한 체형의 사용자들이 만족했던 사이즈나 스타일을 기반으로 상품을 추천할 수 있습니다.

3. 사용자의 기존 평가를 바탕으로 분류 문제처럼 추천합니다.



사용자가 특정 상품에 “사이즈가 잘 맞습니다”, “핏이 마음에 듭니다” 같은 리뷰를 남기거나 별점 5점을 줬다면 시스템은 이를 긍정적인 평가로 판단하고 이후 비슷한 스타일이나 핏의 상품이 들어왔을 때, 해당 사용자가 좋아할 가능성이 높다고 판단하여 추천할 수 있습니다.

- 유사도

유사도는 추천시스템을 이해하는데 중요한 **수학적** 내용입니다.

유사도를 크게 두가지 방법으로 측정하는데 먼저 간단하게 거리를 이용해서 가까이 있는 것들끼리 비슷하다고 묶는 유클리드 거리를 활용한 방법과

각도를 이용하는 코사인 유사도 방법이 있습니다. 코사인 유사도는 거리와 상관없이 각도를 재기 때문에 만약 두 점이 원점을 지나는 직선을 동시에 통과한다면 두 점을 같다고 판단할 것입니다.

이 방법들을 활용해 기존의 영상혹은 이미지 또는 텍스트들을 백터로 변환한 값들의 유사도를 측정합니다.


- TF-IDF

TF-IDF (Term Frequency - Inverse Document Frequency)
* $TF(d,t)$ : 특정 문서 $d$ 안에서 단어 $t$가 등장한 횟수
* $DF(t)$ : 단어 $t$가 등장한 전체 문서의 수

TF-IDF는 어떤 단어가 특정 문서에서는 많이 등장하지만, 전체 문서에서는 자주 등장하지 않을수록 중요한 단어라고 판단합니다.

예를 들어 인사이트 노션 전체에서는 “데이터”라는 단어가 거의 모든 페이지에 등장할 것입니다. 따라서 “데이터”의 $DF$ 값은 매우 커지게 되고, 흔한 단어라고 판단되어 가중치가 낮아집니다.

반면 심화세션의 추천 시스템 페이지에서는 “추천”이라는 단어가 여러 번 등장할 수 있지만, 전체 노션에서는 상대적으로 적은 문서에만 등장할 수 있습니다. 이 경우 해당 페이지에서의 $TF$ 값은 크고, 전체 문서 기준 $DF$ 값은 상대적으로 작기 때문에 “추천”이라는 단어의 TF-IDF 값이 높아집니다.

즉 TF-IDF는 단순히 많이 나온 단어를 중요한 단어로 보는 것이 아니라, “특정 문서를 얼마나 잘 대표하는 단어인가”를 기준으로 가중치를 계산하는 방식입니다.

#### 협업 필터링

햡업 필터링은 내용기반 필터링과 다르게 여러 사용자의 선호도를 활용해서 추천을 해줍니다.





제가 이직로그란 웹툰을 읽는다면 이 웹툰을 같이 즐기는 사용자 중 90%의 사용자가 십이지 소녀를 좋아한다고 같이 추천해주는 네이버 웹툰이 대표적인 예시입니다.

- 메모리 기반 협업 필터링
메모리 기반 협업 필터링은 사용자나 아이템 간의 유사도를 기반으로 추천하는 방식입니다. 나와 비슷한 취향을 가진 사람들이 좋아했거나 내가 좋아한것과 비슷한것을 추천합니다

예를 들어 네이버 웹툰에서 내가 “이직로그”를 읽고 있다면, 이 웹툰을 좋아한 다른 사용자들이 많이 본 “십이지 소녀”를 추천해주는 방식이 대표적인 예시입니다.

메모리 기반 협업 필터링은 크게 사용자 기반과 아이템 기반으로 나뉘는데. 위에 말한것중 비슷한 취향을 가진 사람들이 종하했던걸 추천하는게 사용자 기반이며, 내가 좋아한것과 비슷한 것을 추천하는것이 아이템 기반입니다.

하지만 이런 메모리 기반 협업 필터링은 아래의 3가지 한계가 있습니다.

첫 번째는 희소 행렬 문제입니다. 실제 추천 시스템에서는 사용자 수와 아이템 수가 매우 많기 때문에, 대부분의 사용자들은 전체 아이템 중 극히 일부만 평가합니다. 따라서 비어있는 데이터가 너무 많아 유사도를 정확하게 계산하기 어려워집니다.

두 번째는 확장성 문제입니다. 사용자와 아이템이 계속 증가할수록 모든 유사도를 계산하는 비용이 매우 커지게 됩니다.

세 번째는 콜드 스타트 문제입니다. 새로 가입한 사용자나 새롭게 등록된 아이템은 데이터가 거의 없기 때문에 추천을 수행하기 어렵습니다.

- 모델 기반 협업 필터링

메모리 기반의 한계를 극복하기 위해 등장한 모델 기반 협업 필터링입니다. 관측된 데이터로 부터 학습해 관측되지 않은 데이터들을 예측합니다.모델 기반은 메모리 와 다르게 잠재요인들을 학습한다는 차별점이 있습니다. 이중 잠재요인은 취향과 특성과는 눈에 보이지 않는 데이터들을 의미합니다.

---

- 행렬분해

사용자-아이템 평점 행렬을 사용자 잠재 요인 행렬과 아이템 잠재 요인 행렬로 분해하여, 비어있는 평점을 예측하는 방식입니다.

예를 들어 사용자가 액션 영화 성향이 높고, 특정 영화 역시 액션 성향이 강하다면 두 잠재 요인의 유사도를 바탕으로 높은 평점을 예측할 수 있습니다. 즉 사용자의 숨겨진 취향과 아이템의 숨겨진 특징을 학습하여 추천을 수행하는 방법입니다. 이 행렬분해 총 4가지 방법으로 나뉩니다.

- SVD (Singular Value Decomposition)

SVD는 사용자-아이템 행렬을 여러 개의 작은 행렬로 분해하여, 데이터의 핵심 패턴만 추출하는 방식입니다.
넷플릭스에서 어떤 사용자가 액션 영화와 SF 영화를 자주 본다면, 시스템은 이러한 숨겨진 취향 패턴을 학습할 수 있습니다. 이후 비슷한 성향의 새로운 영화를 추천하는 데 활용됩니다.

- SGD (Stochastic Gradient Descent)

SGD는 예측 평점과 실제 평점 사이의 오차를 줄이는 방향으로 사용자 벡터와 아이템 벡터를 반복적으로 업데이트하는 방식입니다.

유튜브에서 사용자가 실제로 좋아요를 누른 영상은 백예린의 노래인데 모델이 예측한 것은 빅뱅의 영상이라면 좀 더 차분한 노래들을 넣어가며 그 차이를 줄입니다.

- NMF (Non-Negative Matrix Factorization)

NMF는 행렬의 모든 값이 0 이상이 되도록 제한하는 행렬 분해 방식입니다.

모든 취향을 반영해 아예 관심이 없다면 0 하나만 본다면 1로 가정해 모든 값을 0이상으로 제한해 한사람의 취향을 직관적으로 표현합니다.

- PMF (Probabilistic Matrix Factorization)

PMF는 확률 기반으로 평점을 예측하는 행렬 분해 방식입니다.

쇼핑몰 추천 시스템에서 사용자가 특정 신발을 좋아할 가능성을 단순 점수 하나로 예측하는 것이 아니라, “높은 확률로 좋아할 것 같다”처럼 불확실성까지 함께 고려하여 추천합니다.

### VAE 기반 모델

VAE 기반 협업 필터링은 사용자의 행동 데이터를 잠재 공간에 확률 분포 형태로 표현하여 추천하는 방식입니다.

예를 들어 어떤 사용자가 넷플릭스에서 액션 영화는 많이 봤지만 로맨스 영화는 몇 개만 봤다고 가정해보겠습니다. 기존 방식은 이 사용자의 취향을 하나의 고정된 벡터로 표현하지만, VAE는 “액션을 좋아할 가능성이 높고, 로맨스도 어느 정도 좋아할 수 있다”처럼 불확실성까지 함께 학습합니다.

따라서 데이터가 적은 사용자라도 보다 안정적인 추천이 가능하며, 희소한 데이터 환경에서도 강한 성능을 보입니다.

---

### GNN 기반 모델

GNN은 사용자와 아이템 사이의 관계 자체를 그래프로 표현하여 학습하는 방식입니다.

예를 들어 배달 앱에서 사용자 A가 치킨집 여러 곳을 자주 주문했고, 비슷한 사용자들이 특정 떡볶이 가게도 많이 주문했다면, GNN은 이러한 연결 관계를 그래프 형태로 학습합니다.

즉 단순히 “무엇을 좋아했는가”뿐 아니라, “누가 무엇과 어떻게 연결되어 있는가”까지 함께 학습하여 추천 성능을 높입니다.

---

### Transformer 기반 모델

Transformer 기반 추천 시스템은 사용자의 행동 순서와 문맥을 함께 고려하여 추천하는 방식입니다.

예를 들어 사용자가 쇼핑몰에서 “텐트 → 침낭 → 등산화” 순서로 상품을 조회했다면, 시스템은 현재 사용자의 목적이 캠핑 준비라고 이해할 수 있습니다. 반대로 “패딩 → 장갑 → 등산화” 순서였다면 겨울 방한 용품을 찾고 있다고 해석할 수 있습니다.

즉 Transformer는 단순히 사용자가 좋아했던 아이템만 보는 것이 아니라, 행동의 순서와 문맥까지 이해하여 다음 행동을 예측하는 데 강점을 가집니다.
