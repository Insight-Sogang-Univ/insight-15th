# 시계열 데이터



### 시계열데이터란?


순차적인 시간의 흐름에 따라 수집된 데이터이며, 순서가 중요한 시퀀스 데이터로 간주됩니다.



동물의 숲에서 무를 판다고 가정을 해봅시다.
월요일에는 92벨, 화요일에는 105벨, 수요일에는 128벨, 목요일에는 87벨, 금요일에는 142벨, 이렇게 날짜의 흐름에 따라 가격이 변화하는 데이터를 시계열 데이터라고 합니다.

이런 시계열 데이터는 크게 단변량과 다변량으로도 나눌 수 있습니다.
단변량은 시간에 따라 연속적으로 측정된 단일 변수로 이루어진 시계열 데이터이고
다변량은 같은 시간에 기록된 둘 이상의 변수로 이루어진 시계열 데이터입니다.

---

### 시계열 데이터의 특징

시계열 데이터는 크게 3가지 특징이 있습니다
1. 시간의존성



시계열 데이터는 시간 순서에 따라 관측되서 순서를 바꾸면 의미가 바뀝니다.
예를들어 제 키는 지금 182cm 이고 중2때 175cm입니다. 이를 보면 제가 중2때 비해 신체적으로 성장했다고 해석 할 수 있지만, 반대로 지금 175cm 이고 중2때 182cm 라고 가정한다면 늙어서 키가 줄었구나라고 해석됩니다. 

---

2. 계절성과 추세



일정 주기에 따라 반복적으로 변동되는 계절성을 띄며, 시간의 흐름에 따라 나타나는 장기적인 증가/감소 경향인 추세를 볼 수 있습니다.

봄 사랑 벚꽃 말고 라는 봄노래는 봄이 될때마다 멜론차트에서 엄청난 상승세를 기록하며 차트 100위 안에 드는 계절성을 볼 수 있습니다. 

하지만 시간이 지남에 따라 점점 낮아지면서 2017년 최고 18위까지 올라갔던 순위가 2021년 이후 6,70위 권을 기록하며 점점 내려가는 추세를 보입니다.

---

3. 자기상관성

이전 시점의 값이 이후 시점의 값에 영향을 미칩니다. 시간의존성에서 예시를 들었던 키처럼 10초전의 저의키가 지금 제 키와 엄청난 영향을 미친다고 볼 수 있습니다.
이러한 특성때문에 데이터끼리 독립적이지 못해 일반 회귀에서 많은 문제를 야기하기도 합니다.

---

### 시계열 데이터의 분석요인

이러한 특성을 기반으로 분석요인들을 크게 4가지로 정의할 수 있습니다.

---

1. 추세 요인



추세 요인은 데이터 전반에 걸쳐 증가하거나 감소하는 경향을 말합니다.

---

2. 계절 요인

계절요인은 일정한 주기 반복되는 패턴을 봅니다.

---

3. 순환 요인

계절성보다 더 길고 불규칙한 주기로 반복되는 변동을 순환요인이라 합니다.

---

4. 불규칙 요인



배달관련 결제횟수나 금액이 증가하지만 2019년 초반에 급증하는걸 볼 수 있습니다. 코로나 때문에 그런것인데요. 이처럼 팬대믹 혹은 자연재해로 인해 발생하는 예측 불가능한 변동을 나타냅니다.

### 정상성 판단

- 정상성이란?



시간의 상관없이 일정한 성질을 띄는 것을 말합니다. 평일의 저의 기상시각과 같습니다. 저는 매일 어제 7:30에 기상했고, 담주 화요일에도,수요일도,목요일도 똑같을 겁니다.

- 강한 정상성과 약한 정상성

강한 정상성은 시간을 언제로 잡든, 데이터의 확률 분포 자체가 똑같은 상태로 현실에서 일어나기 힘든 상태를 말합니다.

약한 정상성은 실무에서 쓰이는 현실적 기준으로 평균과 분산이 일정하지만, 자기공분산이 시차에만 의존하는 상태입니다.

---

- 정상성의 판단 - 시각과 통계



시각적으로 봤을때 평균을 중심으로 진동하는 형태를 보이면 정상성을 띈다고 볼 수 있습니다. 위 그림과 같이 0을 중심으로 진동하는 형태를 띄며 정상성을 띈다고 해석합니다.

통계적으로는 단위근(Unit Root)이 있는지를 확인하는 과정을 거친다.

여기서 단위근을 구할때

오일러 공식(Euler's Formula)은 다음과 같습니다.

$$
e^{i\theta}=\cos\theta+i\sin\theta
$$

즉 복소수는 극형식으로

$$
z=r(\cos\theta+i\sin\theta)=re^{i\theta}
$$

형태로 표현할 수 있습니다.

이 식을 이용해서 복소수를 회전의 형태로 표현할 수 있으며
$$
z^n=1
$$
을 만족하는 n차 단위근(Unit Root)을 구합니다.

---

또 이를 검정하는 방법은 두가지로 나뉩니다.

ADF 검정은 데이터가 비정상 시계열이라고 가정하고 시작하며, 과거 값의 영향이 계속 누적되는지 확인하여 정상성을 판단합니다. 일반적으로 p-value가 작으면 정상 시계열로 판단합니다.

반대로 KPSS 검정은 데이터가 정상 시계열이라고 가정하고 시작하며, 평균과 구조가 시간에 따라 일정하게 유지되는지 확인합니다. 일반적으로 p-value가 작으면 비정상 시계열로 판단합니다.

### 단변량 시계열 분석

- AR / MA

자기 회귀 모델(AR)은 자기 자신 데이터가 미래 값을 설명해줄 것이라고 가정하고 회귀모델을 사용하는것입니다.



제가 좋아하는 키움인데요 작년도 제작년도 꼴등이니까 올해도 꼴등이지 않을까 라고 예측합니다.

---

이동 평균 모델(MR)

과거 시점의 잔차를 독립변수로 삼아 현재 값을 예측하는 모델로 과거의 잘못 예측한 것을 현재 시점에 반영하는 모델입니다.

- AR,MA의 전제조건

둘다 정상성과 가역성 조건들을 만족해야 모델을 사용할 수 있습니다.

AR의 특성방정식을 통해서 정상성을 확인하고, MR은 특성방정식을 통해서 가역성을 확인한 후 모델을 사용합니다.

---

- ARMA

위애서의 정상성과 가역성을 만족한다면, ARMA (Autoregressive Moving Average) 모형을 사용할 수 있습니다.

이 모델은 AR(p) 모형과 𝑀𝐴(𝑞)모형을 결합한 모형
과거 p개의 관측값 + 과거 q개의 오차를  활용해서 예측하는 방법을 사용합니다.

---

- 차수 결정 방법 : 자기 상관 함수 (ACF, PACF)

자기상관계수(ACF, Auto Correlation Function)

ACF는 현재 데이터가 과거 데이터와 얼마나 관련 있는지를 확인하는 지표입니다.

PACF는 중간 시점들의 영향을 제거한 뒤, 순수한 직접 관계만 측정하는 지표입니다. 즉 현재 값과 특정 시차 값 사이의 순수한 영향력만 측정하는 방식입니다.

### 불안정 시계열 분석

불안정 시계열 분석은 크게 아래 2가지로 볼 수 있습니다.

- ARIMA 

AR(Autoregressive) : 자기회귀, I(Integrated) : 차분, MA(Moving Average) : 이동평균을 활용하는 모델입니다.

즉,과거 값 + 과거 오차 + 차분을 이용하여 미래를 예측하는 모델입니다

---

- SARIMA



하지만 ARIMA는 계절성을 처리할 수가 없습니다시계열 데이터의 계절성을 처리할 수 있는 모델로 ARIMA에서 계절성도 처리할 수 있게끔 발전한 모델입니다.

---

### 다변량 시계열 분석

- VAR



AR 모형은 자기 자신의 과거 데이터만 사용하여 미래를 예측하는 방식입니다. 예를 들어 이번 주 프로미스나인 유튜브 조회수는 지난주 프로미스나인 조회수 흐름만 보고 예측하는 방식입니다.

반면 VAR 모형은 여러 변수의 과거 데이터를 함께 사용하여 미래를 예측하는 방식입니다. 예를 들어 이번 주 프로미스나인 화제성은 지난주 화제성뿐 아니라 팬 커뮤니티 언급량, 음원 순위, 쇼츠 조회수 등 다른 변수들의 과거 데이터까지 함께 사용하여 예측합니다.

즉 AR은 자기 자신의 과거만 사용하는 단방향 구조이고, VAR은 여러 변수들이 서로 영향을 주고받는 관계까지 함께 고려하는 모델입니다.

### 딥러닝을 활용한 시계열 분석

RNN



RNN 계열은 이전 시점의 정보를 현재 시점으로 전달하는 순환 구조 기반 시계열 딥러닝 모델입니다. 대표적으로 RNN, LSTM, GRU가 있으며, 과거 정보의 흐름을 학습하는 데 사용됩니다. LSTM과 GRU는 장기 의존성 문제를 개선한 구조이지만 긴 시계열에서는 학습 효율이 떨어질 수 있습니다.

---

Transformer



Transformer 계열은 Attention 메커니즘을 활용하여 긴 시계열의 전역 패턴과 복잡한 관계를 학습하는 모델입니다. Informer, Autoformer, TFT 등이 대표적이며, 긴 시계열·주기성·외생 변수 처리에 강점을 가집니다. 병렬 연산이 가능해 대규모 데이터 처리에 효율적이지만 구조가 복잡하고 학습 비용이 큽니다.

---

Foundation Model

Foundation Model은 대규모 데이터로 사전 학습된 범용 시계열 모델입니다. Fine-tuning이나 Prompting을 통해 다양한 시계열 문제에 활용할 수 있으며, LLM 기반 방식과 시계열 전용 모델 방식으로 발전하고 있습니다. 대표적으로 TimeGPT, TimesFM, Chronos 등이 존재합니다.
