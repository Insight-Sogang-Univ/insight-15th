### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> 학습 방법: 순전파와 역전파 </span>

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> 손실 함수 </span>

모델이 정확한지 오차 계산 = 실제값(정답)과 예측값(출력값)의 차이 계산!

|손실 함수|사용 상황|표현식|예|
|-----|--------|-----------|---------|
|MSE|회귀|$L = \frac{1}{n} \sum_{i=1}^{n}(y_i - \hat{y}_i)^2$|연속형 변수 예측|
|BCE|이진 분류|$L = -(y \log(\hat{y}) + (1 - y) \log(1 - \hat{y}))$|정상/불량 진단 필터링 등등|
|CEE|다중 클래스 분류|$L = -\sum_{i=1}^{C} y_i \log(\hat{y}_i)$|이미지 분류 등|

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> 활성화 함수</span>

> 각 층의 출력값에 비선형성을 추가해서 신경망이 복잡한 패턴도 학습할 수 있게 해주는 함수

활성화 함수 없을 때

2층: W₂(W₁x + b₁) + b₂ <br>
   = (W₂W₁)x + (W₂b₁ + b₂) <br>
   = Wx + b   ← 1층이랑 완전히 같음 <br>
100층 쌓아도 결국 Wx + b 하나. 직선만 그릴 수 있음 <br>

활성화 함수 있을 때

2층: W₂ · σ(W₁x + b₁) + b₂ <br>
σ 때문에 전개가 안 됨 → 각 층이 진짜로 다른 변환을 함 → 구불구불한 경계 표현 가능 <br>

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 계단함수 </span>


> 입력 신호: 입력값과 가중치의 곱 <BR>
> 출력 신호: 1과 0

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 시그모이드 함수 </span>



> 입력 신호를 특정 범위 (0과 1 사이)로 변환 <BR>
> 비선형, 연속적 변화

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;">ReLU 함수 </span>


> 입력이 0을 넘으면 그 입력값을 그대로 출력하고, 0 이하면 0을 출력하는 비선형 함수 <br>
> 양수 구간에서 미분값 1 -> 역전파 안정적 전달

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> Tanh 함수 </span>


> 시그모이드 함수 변형 (0-1 -> -1-1) <BR>
> 입력값의 중심을 0으로 맞춤 -> 가중치 업데이트가 더 안정적

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> Softmax 함수 </span>


> 각 클래스에 대한 출력값을 확률로 변환 <br>
> 다중 클래스 분류 문제의 활성화 함수 <BR>

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> 학습 과정에 필요한 파라미터</span>

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 배치 </span>
> 학습 과정에서 가중치와 같은 매개변수를 조정하기 위해 사용하는 데이터의 묶음

- 너무 크면 학습 느려지고 메모리 부족
- 너무 작으면 학습 불안정...

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;">에포크 </span>
> 데이터셋을 학습하는 횟수

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> iteration </span>
> 한 번의 에포크를 끝내기 위해서 필요한 배치의 수 <BR>
> 1 epoch에서 다루는 데이터의 수 = 배치의 크기 * 이터레이션

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> 경사하강법 </span>

> Batch size = Data size

> - 업데이트 횟수 적음
> - 메모리가 많이 필요
> - 수렴 안정적

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> SGD </span>

> 배치 크기 1, 랜덤으로 선택된 하나의 데이터에 대해서만 계산

> - 메모리 적게 필요
> - 변경폭 불안정, 정확도 떨어질 수 o

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;">미니 배치 경사하강법 </span>

> 배치 크기 적절히 지정 ($2^n$)

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> 옵티마이저</span>
> 경사 하강법을 효율적이고 안정적으로 수행하기 위해 필요한 것



##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 모멘텀 </span>

> 경사하강법 + 관성 <br>
> 경사하강법으로 계산한 접선 기울기에 한 시점 전의 접선 기울기값을 일정 비율만큼 반영

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> RMSProp </span>

> 급하면 좀 천천히, 완만하면 유지 혹은 빠르게 <br>
> 학습률을 적절히 조정하자~
