## <span style="background-color:#ffdef1; color:#ff009b; padding:4px; border-radius:5px;"> NLP </span>

자연어: 인간이 소통을 위해 쓰는 언어~

NLP: 자연어 -> 컴퓨터 언어로 변환하는 기술

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 특징 </span>

- 이산적 데이터
- 순차적 데이터
- 중의성 o
- 희소성
- 가변성

언어 모델: 단어 시퀀스에 확률을 할당하여 가장 자연스러운 단어 시퀀스를 찾아내는 모델

## <span style="background-color:#ffdef1; color:#ff009b; padding:4px; border-radius:5px;"> SLM(통계 기반 모델) </span>

> 분포 가설[비슷한 문맥에서 함께 나타나는 단어들은 비슷한 의미를 가짐]에 근거

`조건부 확률`과 `등장 빈도 기반 확률`을 이용

그러나 희소 문제[데이터 부족으로 인해 언어 정확히 모델링 x] 존재

-> 새로운 문장 조합에 약하다!

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> N-gram </span>
> 희소 문제 완화를 위해 전체 문장 대신 문장 일부만 보자!

문장을 n개 단위로 잘라 토큰화 -> 앞 n-1개의 단어만 참고


##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 한계</span>
- 그래도 훈련 데이터에서 나타나지 않은 단어 조합은 여전히 예측 불가...
- 앞선 n-1개 단어만 봐서 문맥 반영 x


### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> Perplexity(PPL) </span>
> 언어 모델이 문장을 얼마나 혼란스러워하는지 수치로 나타냄 (희소하게 등장할수록 커지고 단어 개수가 많아질수록 낮아짐)

## <span style="background-color:#ffdef1; color:#ff009b; padding:4px; border-radius:5px;"> NLM(신경망 언어 모델) </span>

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> 임베딩</span>



단어를 차원에다가 놓음!


### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> BERT </span>
> Bidirectional Encoder Representations from Transformers: 트랜스포머의 인코더 부분을 활용해 만든 양방향 언어 모델


입력 문장 임베딩 -> 인코더로 입력 문장 이해 -> 의미 벡터 변환

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;">  Masked Language Model(MLM) </span>
`빈칸 채우기` 방식으로 언어 문맥 학습!

> 사전 훈련을 위해 들어가는 Input 텍스트 중 15% 정도의 단어를 랜덤으로 마스킹

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;">  Next Sentence Prediction(NSP) </span>
> 두 문장의 관계를 학습


### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> GPT </span>
> 트.포의 디코더 기반 모델: 지금까지의 단어를 바탕으로 다음 단어 예측!

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> few-shot learning </span>

- zero-shot: 예시 없이 문제 지시만 주고 task 수행
- few-shot: 몇 개의 예시만 보고 task 수행

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> LLM </span>
> 이 모든 걸 포괄하는 거대한 언어 모델 (디코더 기반)


- 대규모 학습
- 예측 및 생성

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> RAG </span>
> 질문에 대해 llm 모델이 모르면 문서 찾아서 같이 답변

- 최신 정보 반영 가능
- 환각 현상 방지


##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 구조 </span>

1. 질의 인코더
> 사용자의 질문을 이해하기 위한 언어 모델: 질문을 벡터 형태로 인코딩
2. 지식 검색기
> 질문 바탕으로 외부 지식 베이스에서 관련 정보 검색
3. 지식 증강 생성기
> 검색된 지식을 활용하여 질문에 대한 답변 생성

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> Langchain </span>
> RAG pipeline 쉽게 구성할 수 있도록 도와줌!

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 특징 </span>
1. 추상화
> LLM 서비스 만들 때 필요한 각종 작업 최소화
2. 표준화
> 비슷한 기능 요소 형식 표준화
3. 체이닝
> 자주 활용하는 컴포넌트 쉽게 연결

+ 랭체인의 선형적 한계 보완 -> 랭그래프 (사이클 형태!)

## <span style="background-color:#ffdef1; color:#ff009b; padding:4px; border-radius:5px;"> sLM </span>
> 소규모 언어 모델 (작고 효율적~)


### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> 모델 압축 </span>
##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 가지치기 </span>

> 신경망에서 중요도가 더 낮거나, 중복되거나, 불필요한 매개변수 제거!

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 양자화 </span>

> 고정밀 데이터 -> 저정밀 데이터

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 저차원 분해 </span>

> 큰 가중치 행렬을 저차원 행렬로 분해 (연산 단순화!)

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 지식 증류 </span>
> 사전 학습된 교사 모델의 학습 내용 -> 학습 모델
