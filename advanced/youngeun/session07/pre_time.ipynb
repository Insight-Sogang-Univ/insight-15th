## <span style="background-color:#ffdef1; color:#ff009b; padding:4px; border-radius:5px;"> 시계열 데이터 특징 </span>

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> 요소 분해 </span>

> 주어진 시계열 데이터를 개별 성분(추세성, 계절성, 순환성, 불규칙성)으로 분해하는 과정


$Y_t$ = $f$( $T_t, S_t, C_t, R_t$)

- $T_t$ : 추세성분 (시점 t에서의 추세요소)
- $S_t$ : 계절성분 (시점 t에서의 계절요소)
- $C_t$ : 순환성분 (시점 t에서의 순환요소)
- $R_t$ : 불규칙성분 (시점 t에서의 불규칙요소)
</aside>

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 가법 모형 </span>
$$
y_t = T_t + S_t + R_t
$$

> 시계열 = 추세 + 계절성 + 불규칙성

`각 변동 요인 독립적으로 더해지고, 모든 요소 영향 일정`

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 승법 모형 </span>

$$
y_t = T_t * S_t * R_t
$$

> 시계열 = 추세 * 계절성 * 불규칙성

`추세가 커질수록 게절성/불규칙성 효과 비레해서 커짐`

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 통계로 판단: 단위근 검정 </span>

단위근? 과거 충격이 영구적으로 누적되는 특성

##### 1. ADF(Augmented Dicky-Fuller) 검정

> **단위근(Unit root)** 존재 여부를 통해 시계열의 정상성 판단

- **귀무가설(H₀):** ‘시계열에 단위근이 존재한다**’ → 비정상 시계열**
- **대립가설(H₁):** ‘시계열에 단위근이 존재하지 않는다’ **→ 정상 시계열**

##### 2. KPSS(Kwiatkowski-Phillips-Schmidt-Shin) 검정**

> **단위근 기반**의 검정이지만 **귀무가설이 반대**

- **귀무가설(H₀):** ‘시계열에 단위근이 존재하지 않는다’ **→ 정상 시계열**
- **대립가설(H₁):** ‘시계열에 단위근이 존재한다**’ → 비정상 시계열**

# <span style="background-color:#FFFFE0; color:#CD5C5C; padding:4px; border-radius:5px;"> 단변량 시계열 </span>
## <span style="background-color:#ffdef1; color:#ff009b; padding:4px; border-radius:5px;"> 안정적 시계열 분석 </span>

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> 자기 회귀 모델 (AR) </span>

> 과거 자신의 값들을 독립변수로 삼아 현재의 값을 예측하는 회귀 모델

시계열 데이터의 `자기 상관성` 활용

$AR(p):  y_{t}=c +\phi _1y_{t-1}+\phi_2y_{t-2}+...+\phi _py_{t-p}+\varepsilon _t$

$p$차 자기 회귀 모형: 과거 $p$개 시점의 데이터의 선형 조합을 이용해 예측하는 모델

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> 이동 평균 모델 (MA) </span>
> 과거 시점의 잔차($\varepsilon$)를 독립변수로 삼아 현재 값을 예측하는 모델

$MA(q): y_t = \mu +\varepsilon _t+ \theta _1\varepsilon_{t-1}+\theta_2\varepsilon_{t-2}+...+\theta_q\varepsilon_{t-q}$

$q$차 이동 평균 모형: 과거 q개의 예측 오차의 선형 결합으로 예측하는 모델

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> AR, MA 모델 전제 조건 </span>
##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 정상성 </span>
> AR과 MA 모형은 안정적 시계열 분석으로 확률적 안정성을 전제로 추정


##### **AR 모형**의 정상성 조건
> 특성 방정식: $1 - \phi_1 z - \phi_2 z^2 - \cdots - \phi_p z^p = 0$

: 모든 해의 절댓값 > 1 -> AR의 정상성 확보


##### **MA 모형**의 정상성 조건
> 잔차는 정상성 이미 만족 -> MA는 항상 만족


##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 가역성 </span>
> MA 모형을 AR(∞) 형태로 재표현할 수 있는지 여부

##### **MA 모형**의 가역성 만족 조건
> 특성 방정식: $1 + \theta_1 z + \cdots + \theta_q z^q = 0$

: 모든 해의 절댓값 > 1 ⇒ MA의 가역성 확보

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> ARMA </span>

$Y_t = \underbrace{\phi_1 Y_{t-1} + \phi_2 Y_{t-2} + \cdots + \phi_p Y_{t-p}}_{\text{AR(p): 자기회귀}} + \underbrace{\varepsilon_t + \theta_1 \varepsilon_{t-1} + \theta_2 \varepsilon_{t-2} + \cdots + \theta_q \varepsilon_{t-q}}_{\text{MA(q): 이동평균}}$

> $AR(p)$ 모형과 $MA(q)$ 모형을 결합한 모형

: 과거 $p$개의 관측값 + 과거 $q$개의 오차를  활용해서 예측하는 방법

### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> 차수 결정법: 자기상관함수 </span>
> 두 그래프를 통해 현재 값이 과거 값과 어떤 관계가 있는지 파악하고 이를 통해 AR, MA, ARMA의 적절한 차수(p, q)를 결정
##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 자기 상관 계수 (ACF) </span>
> ARMA(p, q)모형에서 사용할 q개의 오차 결정 방법

: 두 시점 사이의 **상관 관계** (간접 영향) 파악

# <span style="background-color:#FFFFE0; color:#CD5C5C; padding:4px; border-radius:5px;"> 다변량 시계열 분석 </span>
### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> VAR </span>
> 자신의 과거 값과 다른 변수들의 과거 값을 함께 사용하여 미래를 예측하는 모델


$X_t = a_{11} X_{t-1} + a_{12} Y_{t-1} + ε_{x,t}$

-> X와 Y가 서로의 과거를 함께 사용

사용
- 여러 경제 지표가 서로 영향을 주고받는 경우 (GDP, 소비, 투자 등)
- 금융 시장에서 여러 자산의 가격이 상호작용하는 경우
- 변수 간 상호 의존성이 중요한 경우

불가
- 변수가 너무 많은 경우 (파라미터 폭발)
- 변수 간 관계가 명확하지 않은 경우
- 장기 예측이 필요한 경우

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 충격 반응 함수 (IRF) </span>
> 한 시계열에 특정 시점에서 충격이 발생할 때 다른 시계열에 시간에 따라 어떤 영향을 주는지 분석

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> 예측 오차 분산 분해 </span>
> 각 변수가 미래의 움직임을 예측하는 데에 어느 정도의 기여를 하는지 평가하는 방법

## <span style="background-color:#ffdef1; color:#ff009b; padding:4px; border-radius:5px;"> 딥러닝 시계열 분석 </span>
### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> 순환 신경망 (RNN...) </span>

| 모델 | 장점 | 단점 | 사용 예시 |
| --- | --- | --- | --- |
| **RNN** | 간단, 빠름 | 장기 의존성 학습 불가 | 짧은 시계열 (일주일 주가) |
| **LSTM** | 장기 의존성 학습 가능 | 복잡, 느림 | 분기별 재무제표 예측 |
| **GRU** | LSTM보다 빠름, 적당한 성능 | LSTM보다 장기 의존성 약함 | 실시간 센서 데이터 예측 |


### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> Transformer 계열 </span>

| 모델 | 언제 사용? | 사용 예시 | 장점 | 단점 |
| --- | --- | --- | --- | --- |
| **Vanilla Transformer** | 짧은 시계열 | 30일 주가 예측 | 구조 단순 | 긴 시계열 처리 어려움 |
| **Informer** | 긴 시계열 | 전력 수요 예측 | 계산 효율적 | 짧은 시계열 과적합 |
| **Autoformer** | 주기성 뚜렷 | 월별 판매량 예측 | 주기 포착 우수 | 구조 복잡 |
| **TFT** | 외생 변수 많음 | 유통 수요 예측 | 해석 가능 | 학습 시간 오래 걸림 |
### <span style="background-color:#ccffd0; color:#0b8c16; padding:4px; border-radius:5px;"> 파운데이션 모델 </span>
> 거대한 데이터로 사전 학습하여 다양한 문제를 해결할 수 있는 범용 모델

##### <span style="background-color:#c9cfff; color:#000b6e; padding:4px; border-radius:5px;"> LLM 기반 </span>
> 대규모 언어 모델의 패턴 인식 능력을 시계열에 활용하는 접근법
