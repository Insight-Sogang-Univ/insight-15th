## **1. 퍼셉트론**

### 1) 인공신경망
- 개념: 우리 뇌의 신경망을 모방한 모델

### 2) 퍼셉트론  
- 개념: 초기 형태의 인공신경망으로, 딥러닝의 기원이 되는 선형 이진 분류기  
- 출력값을 구하는 절차: 입력 -> 가중합 -> 활성화 함수 -> 출력  
- 단층 퍼셉트론: 단순히 입력층과 출력층으로만 이루어진 퍼셉트론
    - 논리 게이트 구현 가능  
    - 한계: XOR 게이트 구현 불가능 
<br>
- 다층 퍼셉트론: 입력층 + 하나 이상의 은닉층 + 출력층

## **2. 딥러닝 모델의 학습 방법**

### 1) 순전파와 역전파  
- 학습 알고리즘의 기본 구조
    - 순전파를 통해 입력값으로부터 예측 수행  
    - 오차 계산을 통해 예측 정확도 측정  
    - 역전파를 통해 출력층에서 다시 입력층 방향으로 오차 전달  
    - 가중치 업데이트  
- 연쇄 법칙: 합성함수의 미분은 각 함수의 미분의 곱으로 나타낼 수 있음.  
- 기울기 소실 문제: 은닉층의 개수가 많아질수록 은닉층의 영향력이 작아지는 문제(앞쪽 층의 가중치가 거의 업데이트 되지 않아서 학습이 제대로 이루어지지 않음.)  
<br>
<흐름>
1. 순전파: 예측하기  
2. 손실 계산: 예측값과 실제 정답을 비교해서 오차 계산  
3. 역전파: 거꾸로 거슬러 올라가며 각 가중치가 손시에 얼마나 기여했는지 계산  
4 가중치 업데이트: 모델 정확도 향상을 위해 가중치 수정

### 2) 손실 함수 (Loss Function)  
- 개념: 실제값과 예측값의 차이를 계산하는 데 사용되는 함수  
- 종류
    - 평균 제곱 오차(MSE): 예측값이 연속적이고 예측값과 실제값 간의 차이를 직접적으로 측정하는 회귀 문제에 적합함.  
    - 이진 크로스 엔트로피(BCE): 이진 분류 문제(로지스틱 회귀)에서 사용, 얼마나 확신을 가지고 예측했는지도 loss에 반영함.  
    - 크로스 엔트로피 오차(CEE): 다중 클래스 분류 문제에서 예측 확률이 실제 클래스와 얼마나 일치하는 지를 평가함. 

### 3) 활성화 함수 (Activation Function)  
- 개념: 각 층의 출력값에 비선형성을 추가해서 신경망이 복잡한 패턴도 학습할 수 있게 해주는 함수  
- 가중치(Weight): 입력값이 결과에 얼마나 영향을 미치는 지를 결정하는 값  
- 편향(Bias): 활성화 함수의 출력값을 조정하는 값으로, 기준점 역할을 수행  
- 계단함수(Heaviside Step Function)  
<br>
  
-> 입력값과 가중치의 곱을 모두 더한 뒤, 그 합이 임계값을 넘으면 1, 아니면 0을 출력하는 함수, 주로 단층 퍼셉트론에서 사용됨.  
<br>
- 시그모이드 함수  
<br>
  
-> 입력 신호를 0과 1 사이의 연속적인 값으로 변환하는 비선형 함수  
<br>
- ReLu 함수  
<br>
  
-> 입력이 0보다 크면 그대로, 0 이하면 0을 출력하는 비선형 함수  
    - 계산이 간단하고 효율적  
    - 비선형성 제공  
    - 기울기 소실 문제 완화  
<br>
- Tanh(쌍곡탄젠트)함수 
<br>
  
-> 시그모이드 함수의 변형으로, 출력값의 범위가 -1 ~ 1임.  
<br>
- Softmax 함수   
<br>
  
-> 각 클래스에 대한 출력값을 확률로 변환하는 함수, 다중 클래스 분류 문제의 출력층에서 사용 가능  

### 4) 경사하강법 (Gradient Descent)  
- 개념: 손실 함수를 최소화하여 모델을 학습시키는 방법  
- 배치(Batch): 학습 과정에서 가중치와 같은 매개변수를 조정하기 위해 사용하는 데이터의 묶음  
- 에포크(Epoch): 모든 데이터셋을 학습하는 횟수  
- 이터레이션(Iteration)/스텝(Step): 한 번의 에포크를 끝내기 위해서 필요한 배치의 수  
- 1 Epoch에서 다루는 데이터의 수 = 배치의 크기 * 이터레이션(step)

(1) 배치 경사하강법  
- 가장 기본적인 경사하강법으로, 오차를 구할 때 전체 데이터를 한 묶음으로 묶음.  
- 전체 데이터를 통해 학습시키기 때문에 업데이트 횟수가 적음.  
- 전체 데이터를 모두 한 번에 처리하므로 메모리가 많이 필요함.  
항상 같은 데이터 (전체)에 대해서 경사를 구하기 때문에 수렴이 안정적임.  

(2) 배치 크기가 1인 확률적 경사하강법  
- 매개변수 값을 조정할 때 전체 데이터가 아니라, 랜덤으로 선택된 하나의 데이터에 대해서만 계산함.  
- 하나의 데이터에 대해서만 메모리에 저장하면 되므로 자원이 적은 컴퓨터에서도 쉽게 사용이 가능함.  
- 매개변수의 변경폭이 불안정하고, 정확도가 낮을 수 있음.  

(3) 미니 배치 경사하강법  
- 배치 크기를 적절히 지정하여 해당 데이터 개수만큼에 대해 매개 변수의 값을 조정함.  
- 전체 데이터를 계산하는 것보다는 빠르고, 배치 크기가 1인 경사 하강법보다 안정적임.  

### 5) 옵티마이저 (Optimizer)  
- 모멘텀: 변수가 가던 방향으로 계속 가도록 속도 추가  
- 알엠에스프롭(RMSProp): 최근 기울기를 참고해서, 급한 경사에서는 속도를 줄이고 완만한 경사에서는 속도를 유지하거나 증가  
- 아담(Adam): 각 파라미터의 1차 모멘트와 2차 모멘트를 사용하여 학습률을 조정
