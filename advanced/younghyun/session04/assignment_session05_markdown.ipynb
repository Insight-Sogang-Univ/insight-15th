## **1. NLP와 언어 모델**

### (1) NLP (Natural Language Processing)  
- 자연어: 인간이 소통을 위해 자연스럽게 발생시켜 사용하는 언어(한국어, 영어 등)  
    - 이산적 데이터  
    - 순차적 데이터  
    - 중의성과 모호성  
    - 희소성  
    - 가변성과 진화  

### (2) 언어 모델  
- 가장 넓은 의미에서 언어 모델은 인간이 말하고 쓰는 방식을 모방하는 시스템  
- 단어 시퀀스에 확률을 할당하여, 가장 자연스러운 단어 시퀀스를 찾아내는 모델  
- 크게 통계 기반 / 인공 신경망 기반으로 구분됨.

---

## **2. 통계적 언어 모델 (SLM)**

### (1) Statistic Language Model  
- 분포 가설(비슷한 문맥에서 함께 나타나는 단어들은 비슷한 의미를 가짐)에 근거함.  
- 조건부 확률  
- 카운트 기반: 이전 단어 시퀀스의 등장 빈도를 통해 다음 단어의 확률을 계산  
    - 한계: 희소 문제(Sparsity Problem)

### (2) N-gram  
- 배경: SLM은 희소하게 등장한 것에 대해서는 잘 예측을 못하고, 문장이 길어길수록 다음 단어를 계산하기 어려움.  
- uni: 1개씩, bi: 2개씩, tri: 3개씩, … N: N개씩 보는 것  

  

- 한계  
    - 희소 문제  
    - 문맥 제한

### (3) Perplexity(PPL)  
- 언어 모델이 문장을 얼마나 혼란스러워하는지를 수치로 나타낸 것  
- 문장 W가 희소하게 등장할수록 PPL이 커지고, 단어 개수 N이 커질수록 PPL이 낮아짐.

## **3. 신경망 언어 모델 (NLM)**  
- 배경: 통계적 언어 모델의 한계 극복  
- 임베딩: 토큰화된 각 단위는 임베딩 단계에서 숫자 벡터로 변환됨.  

| 특징 | 통계적 모델 (N-gram) | 신경망 모델 (NNLM) |
|---|---|---|
| 미학습 단어 조합 | 확률 0 (예측 불가) | 유사 단어를 통해 확률 추론 가능 |
| 저장 공간 | 데이터가 늘수록 기하급수적 증가 | 가중치(Weight)만 저장하면 됨 |
| 의미 파악 | 불가능 (단순 빈도수) | 가능 (임베딩 벡터 간 거리 계산) |

## **4. 딥러닝 기반 언어 모델**

## (1) Transformer의 등장  
- RNN 셀을 반복 사용하는 것 대신, 인코더와 디코더를 여러 층으로 깊게 쌓은 것  

  
- 신경망을 깊게 쌓으면서, 언어 모델의 표현력이 높아져 현재의 딥러닝 기반 언어모델을 만들 수 있었음.

#### [전이학습]  
- 사전 학습(Pre-training)
    - 방법: 레이블이 없는 방대한 데이터(인터넷 전체)를 스스로 학습하게 함. (Self-supervised Learning)  
    - 결과: 언어의 구조, 상식, 문맥을 통째로 이해하는 범용적 뇌를 가진 모델(Base Model) 탄생  
- 미세 조정(Fine-tuning)  
    - 방법: 사전 학습된 모델 위에 우리가 해결하고 싶은 특정 작업의 데이터를 아주 조금만 더 학습시킴.  
    - 결과: 적은 데이터로도 특정 분야의 전문가가 된 모델 탄생

### (2) BERT  
- Bidirectional : 텍스트의 앞뒤 문맥을 동시에 고려해서  
- Encoder : 입력값을 벡터로 바꿔주는  
- Transformers : 딥러닝 모델  
- 특징  
    - 양방향: 문장을 읽을 때 앞뒤를 동시에 고려  
    - 트랜스포머 기반: 트랜스포머 구조의 encoder 부분만 사용  
    - 사전 학습 + 파인 튜닝: 큰 데이터로 먼저 일반적인 언어 능력을 배우고, 나중에 특정 과제에 맞게 살짝 조정  

      

#### [처리 과정]  
input → Embedding → 트랜스포머 인코더 12개(or 24개) → Output  
- BERT는 입력 문장을 임베딩 한 뒤, 여러 개의 인코더를 통과하면서 각 단어의 의미를 문맥적으로 풍부하게 만듦.  

#### [사전 학습]
- Masked Language Model(MLM): 빈칸 채우기 방식 학습  
- Next Sentence Prediction(NSP): 두 문장에 대해 서로 앞뒤 관계인지 무작위 위치에 있는 두개의 문장인지 학습

### (3) GPT  
- decoder 기반, 지금까지의 단어를 바탕으로 다음 단어 예측  
- few-shot learning  
    - zero-shot: 예시 없이 문제 지시만 주고 task 수행
    - one-shot: 1개의 예시만 보고 task 수행

### (4) LLM (Large Language Model)  
- 방대한 양의 데이터를 학습하여 자연어 및 기타 유형의 콘텐츠를 이해하고 생성하여 광범위한 작업을 수행할 수 있는 기초 모델의 범주  
- ex) ChatGPT,LLaMA, PALM, Bard, Gemini등  

#### [작동 방식]  
- 대규모 학습 (Training): 단순히 정보를 암기하는 것을 넘어, 단어와 단어 사이의 관계, 문맥적 의미, 문법 구조, 세상의 지식까지 언어의 패턴을 스스로 터득  
- 예측 및 생성 (Prediction & Generation): 학습이 완료된 LLM은 사용자의 질문이나 지시를 받으면, 학습한 패턴을 기반으로 다음에 올 확률이 가장 높은 단어를 순서대로 예측하며 응답을 생성 

#### [한계]  
- hallucination: 환각처럼 그럴 듯하지만 틀린 정보를 말 하는 것을 의미 (지금은 문제 해결됨.)
- 업데이트 비용: LLM은 파라미터 수가 엄청나게 많기 때문에 새로운 정보를 반영할 때 계산량이 정말 많음.

### (5) RAG (Retrieval Augmented Generation)  
- 질문에 대해 모델이 모르면 검색해서 답하는 구조   
    - 질의 인코더(Query encoder): 사용자의 질문을 벡터로 변환
    - 지식 검색기(Knowledge Retriever): 인코딩된 질문을 바탕으로 외부 지식에서 관련 내용을 검색
    - 지식 증강 생성기(Knowledge Augmented Generator): 검색한 내용을 참고해서 답변을 생성  
- 장점  
    - 최신 정보 반영 가능  
    - hallucination 문제 감소 -> 정보 정확성 증가  
- CAG (Credibility-Aware Generation)  
    - RAG의 한계를 보완하기 위해 제안된 프레임워크  
    - 잘못된 컨텍스트가 모델 성능에 미치는 부정적 영향을 최소화하고, 모델이 문서를 단순히 활용하는 수준을 넘어 문서의 신뢰도를 스스로 판단하고 활용할 수 있도록 훈련시키는 기법  
    - 구조: 질문 입력 → 정보 검색 → 각 정보의 신뢰도 평가 → 높은 신뢰도 정보 우선 사용 → 최종 답변 생성

### (6) LangChain  
#### 1) LangChain  
- LLM + RAG를 직접 구현하는 것이 복잡하므로 이러한 모델 pipeline을 쉽게 구성할 수 있도록 도와주는 역할  
- LLM을 기반으로 앱 제작 시 필요한 구성 요소를 쉽게 연결  
- 파이썬과 JavaScript 라이브러리 제공  

- 특징  
    - 추상화: LLM 서비스를 만들 때 필요한 각종 작업을 간결하게 표현  
    - 표준화: 비슷한 기능을 갖추고 있는 요소들을 똑같은 형식을 갖춘 컴포넌트로 표준화  
    - 체이닝: 자주 활용하는 주요 컴포넌트를 쉽게 연결해서 LLM 서비스의 로직을 쉽게 파악

#### 2) LangGraph  
- LangChain의 한계: 중간에 조건에 따라 다른 작업을 수행하거나, 이전 단계로 돌아가 다시 실행하는 등의 복잡한 로직을 구현하기 어려움.  
- 비선형적인 작업 형태로, 사이클과 루프를 만들 수 있음.  
    - 노드(node): 특정 작업을 수행하는 단위  
    - 엣지(edge): 노드와 노드를 연결하는 선 

#### 3) LangSmith  
- LLM 기반 애플리케이션을 위한 통합 개발 및 운영 플랫폼  
- 핵심 기능  
    - 디버깅 및 추적: LLM 애플리케이션의 모든 실행 과정을 단계별로 시각화
    - 테스트 및 평가: 미리 준비된 질문과 정답 데이터셋을 만들어, 개발 중인 LLM 애플리케이션의 성능을 자동으로 평가
    - 모니터링: 실제 사용자들이 애플리케이션을 어떻게 사용하는지 모니터링 (API 호출 횟수, 지연시간, 토큰 사용량 등을 추적해 최적화)
    - 프롬프트 허브: 다양한 프롬프트를 저장하고, 버전을 관리하며, 쉽게 테스트할 수 있는 공간 제공

## **5. sLM (small Language Model)**

#### 1) sLM  
- 자연어 콘텐츠를 처리, 이해 및 생성할 수 있는 AI 모델 (LLM보다 규모와 범위가 작음.)  
- 수백만 ~ 수십억 개 수준의 매개변수를 가짐.  
- 더 작고 효율적이어서 메모리와 연산 자원을 적게 사용함. -> 모바일 앱처럼 리소스가 제한된 환경, 또는 데이터 네트워크 없이 사용자 쿼리에 대한 오프라인 추론을 통해 응답을 생성해야하는 상황에 적합

#### 2) 모델 압축  
- (1) 가지치기
    - 신경망에서 중요도가 더 낮거나 중복되거나, 불필요한 매개변수 제거
    - 가지치기되는 매개변수에는 뉴런 간의 연결에 해당하는 숫자 가중치(이 경우 가중치는 0으로 설정됨), 뉴런 자체 또는 신경망의 계층이 포함됨.  

      

- (2) 양자화  
    - 고정밀 데이터를 저정밀 데이터로 변환 -> 계산 부하를 줄이고 추론 속도를 높일 수 있음.  

      

- (3) 저차원 분해  
    - 큰 가중치 행렬을 더 작은 저차원 행렬로 분해하는 것 -> 매개변수 수가 줄어들고 계산횟수가 줄어들어 복잡한 행렬 연산을 단순화 가능  

      

- (4) 지식 증류  
    - 사전 학습된 교사 모델의 학습 내용을 학생 모델로 이전하는 것  

      

#### 3) sLM의 예시
- DistilBERT, Gemma, GPT-4o mini, Llama, Ministral, Phi 등

#### 4) sLM의 장점  
- 성능적 측면  
    - 효과적인 성능: 작은 모델을 해당 업무에 맞게 fine-tuning하면 필요한 기능만 빠르고 안정적으로 수행 가능
    - 지연 시간 감소: 모델 크기가 작기 때문에 계산량 ↓
    - 보안: 작은 모델이라서 기업 내부 서버나 개인 기기에서 직접 실행하기가 상대적으로 쉬움. → 민감한 데이터를 외부 클라우드 API로 보내지 않아도 됨.  

- 비용적 측면
    - 접근성: sLM은 고성능 GPU 서버가 없어도 사용 가능→ 진입 장벽 ↓
    - 효율성: 저장 공간, 메모리, 연산 자원을 적게 사용 → 같은 작업을 처리할 때 더 적은 자원으로 결과 도출
    - 에너지 효율: 계산량이 적으므로 전력 소비 ↓
    - 비용 절감: 모델이 작기 때문에 추론 비용, 서버 비용, GPU 비용, API 호출 비용 ↓

#### 5) sLM의 한계  
- 편향: 더 큰 모델에 존재하는 편향까지 함께 전달 가능 → 성능 저하로 이어질 수 있음.  
- 제한된 일반화 & 복잡한 작업에서의 성능 저하: 특정 작업에 대해 미세 조정하기 때문에 포괄적인 지식이 필요한 작업에 취약
- 할루시네이션(환각): 정보가 적고 추론 능력이 낮으므로 당연히 발생 가능

#### 6) LLM과 sLM의 결합  
- 복잡하고 어려운 작업에는 LLM, 가볍고 간단한 작업에는 SLM을 적절히 배분해서 사용 -> LLM 수준의 높은 정확도를 달성하면서도 LLM 토큰 사용량을 최대 90%까지 절약할 수 있음.
