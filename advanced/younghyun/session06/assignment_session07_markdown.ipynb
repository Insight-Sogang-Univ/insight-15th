# **<span style="background-color:#FFF59D;">추천 시스템 기초</span>**

## **<span style="background-color:#FCE4EC;">1. 추천 시스템이란?</span>**

#### 사용자의 정보 데이터를 분석, 개인의 취향에 맞는 아이템을 추천하는 알고리즘

### **<span style="background-color:#E3F2FD;">1.1 사용되는 데이터의 종류와 특징</span>**

① 명시적 데이터 (Explicit Data)  
- 고객들에 의해 입력된 데이터, 사용자가 선호도를 직접적으로 표현한 데이터  
- 당시의 사용자의 취향이 정확하게 반영되지만, 데이터 자체를 얻기 힘듦.  
- ex) 평점, 좋아요, 투표, 스케일 점수  

② 암시적 데이터 (Implicit Data)
- 고객들의 행동을 기반으로 수집된 데이터, 사용자의 선호도가 간접적으로 표현된 데이터  
- 데이터를 얻기 굉장히 쉽지만, 사용자의 취향이나 선호도를 직접적으로 알 수 없음.  
- ex) 조회수, 영상 시청기간, 구매 기록, 페이지 체류 기간 

### **<span style="background-color:#E3F2FD;">1.2 추천 시스템의 분류</span>**

① 내용 기반 필터링 (Content-Based Filtering)
- 콘텐츠의 유사도를 기반으로 사용자에게 추천하는 방법
- 사용자가 이용하거나 선택한 아이템과 유사한 아이템을 추천하는 방식  

② 협업 필터링 (Collaborative Filtering)
- 서비스 내에 있는 다수의 사용자로부터 얻은 선호도 정보를 기반으로 사용자에게 추천하는 방법
    - 사용자 기반: 특정 사용자와 유사한 취향의 사용자가 선호한 아이템 추천
    - 아이템 기반: 사용자가 선호한 아이템과 유사한 아이템 추천

③ 하이브리드 필터링 (Hybrid Filtering)
- 2가지 이상의 다양한 종류의 추천시스템 알고리즘을 조합하여 만들어진 알고리즘

## **<span style="background-color:#FCE4EC;">2. 내용 기반 필터링(Content-Based Filtering)</span>**

### **<span style="background-color:#E3F2FD;">2.1 사용자 프로필</span>**

추천 시스템에 사용되는 데이터  
- 직접 지정형: 사용자에게 자신이 선호하는 아이템의 특징을 명시적으로 지정하도록 하는 방법  
- 간접 지정형: 사용자의 과거 행동 이력에 기반하여 사용자의 프로필을 작성하는 방법

사용자의 프로필을 획득하는 방식  
- 특성 가중치는 사용자가 가지고 있는 아이템 특성 가중치의 평균값 활용  
- 사용자의 기호를 따로 파악하거나, 설문조사, 자발적인 키워드 입력 등  
- 아이템에 대한 유저의 기존 평가를 바탕으로 분류 문제를 해결하는 것과 같음.

### **<span style="background-color:#E3F2FD;">2.2 유사도 (Similarity)</span>**

- 유사도
1. 영상, 이미지, 텍스트 등의 형태로 존재하는 아이템 정보 데이터를 **벡터화**한 이후, 
**아이템 벡터간의 유사도**를 측정
2. 측정된 유사도를 바탕으로 **고객이 선호하는 아이템과 비교**하여 **유사도가 높은 아이템을 선별**

#### TF-IDF(Term Frequency - Inverse Document Frequency)  
- 특정 문서 내에 특정 단어가 얼마나 자주 등장하는 지를 의미하는 단어 빈도(TF) 와 
전체 문서에서 특정 단어가 얼마나 자주 등장하는지를 의미하는 역문서 빈도(IDF)를 통해서 
"다른 문서에서는 등장하지 않지만 특정 문서에서만 자주 등장하는 단어"를 찾아서 
문서 내 단어의 가중치를 계산하는 방법

### **<span style="background-color:#E3F2FD;">2.3 내용 기반 필터링의 장단점</span>**

- 장점
    - 다른 유저의 평점에 영향을 받지 X -> 콜드 스타트 문제에서 자유로움. 
    - 추천이 된 이유 설명 용이  

- 단점 
    - 정확한 feature를 추출하지 못하면 정확도가 떨어짐.  
    - 새로운 장르의 item을 추천하기 어려움.

    *Cold start probelm: 추천 시스템이 새로운 유저들에 대한 충분한 정보가 수집된 상태가 아니라서, 해당 유저들에게 적절한 제품을 추천해주지 못하는 문제

## **<span style="background-color:#FCE4EC;">3. 협업 필터링(Collaborative Filtering)</span>**

가정 - 과거에 비슷한 선호도를 가진 사용자들은 앞으로도 비슷한 선호도를 가질 가능성이 높다.

### **<span style="background-color:#E3F2FD;">3.1 메모리 기반 협업 필터링</span>**

- 사용자 또는 아이템 간 유사도를 기반으로 함.  
- 비교적 초기에 개발된 알고리즘  
- 사용자 A와 비슷한 취향을 가진 사용자가 선호하는 아이템을 추천
    - 사용자 기반 접근: 유저 간의 유사도가 높을수록 가중치를 부여하는 방식, 같은 그룹의 다른 유저가 선호하는 아이템을 추천함.
    - 아이템 기반 접근: 추천하고자 하는 target item과 유사한 아이템의 평점을 활용

### **<span style="background-color:#E3F2FD;">3.2 메모리 기반 협업 필터링의 한계</span>**

- 희소 행렬 문제: 주어진 평점/선호도 데이터가 적을 경우, 성능이 저하됨.  
- 확장성 문제: 아이템/유저가 계속 늘어날 경우, 확장성이 떨어짐.  
- 콜드 스타트 문제: 협업 필터링은 기존의 평점/행동 데이터를 기반으로 유사도를 계산하는데, 데이터가 아예 없는 경우 추천이 불가능함. 
    - User Cold Start: 신규 유저 유입 -> 평가 이력이 없어 유사 유저 탐색 불가
    - Item Cold Start: 신규 아이템 등록 -> 아무도 평가하지 않아 유사도 계산 불가
    - 해결 방법: 직접 지정형 프로필 활용, Content-Based Filtering과 혼합하여 아이템 특성 기반으로 보완, 인구통계학적 정보 활용

### **<span style="background-color:#E3F2FD;">3.3 모델 기반 협업 필터링 (Model-Based Collaborative Filtering)</span>**

- 메모리 기반 CF의 한계를 보완하기 위해 등장한 방식으로, 단순히 유사도를 계산하는 것이 아니라, 데이터로부터 패턴을 학습한 모델을 만들어 평점을 예측함.  
- 관측된 평점 데이터를 학습(Training) 하여, 관측되지 않은 평점을 예측(Inference) 하는 방식 -> 데이터 뒤에 숨겨진 패턴 학습 
- 잠재 요인 (Latent Factor): 사용자와 아이템 사이의 관계를 설명해주는, 눈에 보이지 않는 숨겨진 특성(요인)
- Latent Factor Model: 주어진 데이터로는 알 수 없는 사용자-아이템 간의 숨겨진 특성(feature)을 찾기 위해 주어진 Matrix를 사용자-잠재요인 아이템-잠재요인으로 각각 분해하여 학습하는 모델

-> 유저와 아이템을 동일한 잠재 공간의 벡터로 표현하여, 둘 사이의 거리/각도로 숨겨진 취향 유사도를 파악하는 모델

#### [행렬 분해 (Matrix Factorization)]

| 구분 | 접근 방식 | 특징 |
|---|---|---|
| SVD | **수학적 분해 기법** | 행렬 대수학에 기반하여 행렬을 분해합니다. 희소 행렬에 적용 불가하여 결측치 처리가 필요합니다. |
| SGD | **최적화 알고리즘** | 경사 하강법 기반의 알고리즘으로, 관측된 평점에 대해 학습을 진행하여 희소성 문제를 우회적으로 해결합니다. |
| NMF | **제약 조건 기반 모델** | 분해된 행렬에 비음수(Non-Negative)라는 제약 조건을 부여하여 잠재 요인의 해석력을 높입니다. |
| PMF | **확률적 모델** | 평점 예측에 확률 분포를 사용하여 불확실성까지 모델링하는 통계적 접근법입니다. → VAE와 연결 |

# **<span style="background-color:#FFF59D;">딥러닝 기반 추천 시스템</span>**

## **<span style="background-color:#FCE4EC;">1. VAE 기반 협업 필터링 (Variational Autoencoder CF)</span>**

- 사용자의 아이템 상호작용 벡터를 입력으로 받아, 잠재 공간에서 재구성하도록 VAE를 학습하는 방식  
- 구조
    - 인코더: 사용자 이력을 잠재변수 z 분포로 매핑
    - 디코더: z로부터 아이템 분포를 복원

- 장점
    - 기존 MF처럼 단일 포인트 벡터가 아니라 확률적 잠재 분포로 사용자 선호를 모델링 하여 불확실성을 반영
    - 희소한 상호작용 데이터에서도 강건함.

### **<span style="background-color:#E3F2FD;">1.1 MF의 한계와 딥러닝의 등장</span>**

- 선형성의 한계: 평점 예측이 단순한 잠재 벡터의 내적으로만 이루어짐.
    -> 비선형 활성화 함수를 가진 신경망을 사용하여 복잡한 패턴 학습

- 불확실성 무시: 사용자 취향을 단 하나의 벡터값으로만 표현
    -> 확률 분포로 취향을 모델링하여 데이터의 불확실성 반영

- VAE 등장

### **<span style="background-color:#E3F2FD;">1.2 VAE</span>**

- 불확실성을 모델링함으로써 추천 시스템의 고질적인 문제인 희소성에 대한 해답 제시

| 구분 | AutoEncoder (AE) | Variational AutoEncoder (VAE) |
|---|---|---|
| 잠재 벡터 표현 | 단일 값 (Point Estimate) | **확률 분포** (평균 μ, 분산 σ) |
| 데이터 불확실성 | 반영 불가 (단순 복원에 집중) | **데이터의 불확실성을 모델링**하여 안정성 확보 |
| 적용 시 장점 | 비선형성 학습 | 희소한 데이터에서 더욱 안정적이며, 일반화 성능 우수 |

### **<span style="background-color:#E3F2FD;">1.3 VAE 기반 CF의 작동 구조</span>**

- VAE 기반 CF는 사용자의 아이템 상호작용 벡터를 입력으로 받아, 잠재 공간에서 재구성하도록 학습하는 방식  
    - 입력: 사용자의 아이템 상호작용 벡터
    - 인코더: 사용자 이력을 잠재변수 $z$ 분포로 매핑
    - 잠재 벡터: 샘플링된 최종 취향 벡터
    - 디코더: $z$로부터 아이템 분포를 복원

### **<span style="background-color:#E3F2FD;">1.4 목적 함수와 KL Divergence</span>**

- VAE의 핵심은 목적 함수인 ELBO(Evidence Lower Bound)에 있으며, 이를 통해 두 가지 목표를 동시에 최적화함.  
- KL Divergence의 역할: 잠재 공간을 정규 분포 근처로 압축하여 부드럽게 만듦. → 이 덕분에 데이터가 부족한 유저(희소한 데이터)라도 주변의 풍부한 데이터(정보가 많은 유저)로부터 정보를 자연스럽게 연결받아 안정적인 예측 가능

- VAE-CF의 주요 장점  
    - 확률적 잠재 분포 모델링 (불확실성 반영)
    - 희소한 상호작용 데이터에서도 살아남음.
    - 비선형성 학습

## **<span style="background-color:#FCE4EC;">2. GNN 기반 사용자 선호 예측</span>**

### **<span style="background-color:#E3F2FD;">2.1 GNN (Graphic Neural Network, 그래프 신경망)</span>**

- 그래프 (Graph): 노드(Node)와 그 노드를 잇는 선(Edge)을 모아 구성한 자료 구조
    - 노드: 현실 세계의 개체
    - 엣지: 개체 간 관계나 상호작용을 나타냄.

- 관계, 상호작용과 같이 다소 추상적인 개념을 다루기에 적합함.
- Non-Euclidean space도 표현하고 학습할 수 있음.

### **<span style="background-color:#E3F2FD;">2.2 GNN의 구조</span>**

[핵심 아이디어]: 내 이웃 노드의 정보를 나에게 전달, 요약
1. **[그래프 구조 생성]** 노드는 개별 객체, 엣지는 객체 간 관계나 연결성을 의미함
2. **[메시지 패싱]** 각 노드가 이웃 노드로부터 정보를 받아 전달
    1. **[메시지 생성]** 이웃 노드로에 전달할 메시지 생성
    2. **[집계]** 받은 여러 이웃의 메시지를 하나로 합침 
    3. **[갱신]** 자신의 노드 상태를 업데이트함
3. **[최종 출력]**

### **<span style="background-color:#E3F2FD;">2.3 GNN의 프로세스</span>**

(1) 그래프 구조 생성: 노드와 엣지를 정의한 후, 각각의 피처를 담아 표현  
(2) 메시지 패싱 (Message Pasing): 그래프에서 노드 간 정보를 전달하고 상호작용하는 방법  
    - 메시지 생성: 각 노드가 이웃에게 보낼 자신의 정보 생성  
    - 집계: 여러 이웃 노드에서 받은 메시지를 하나의 벡터로 통합  
    - 갱신: 집계된 이웃 정보와 자신의 이전 상태를 결합해 새로운 임베딩 계산

### **<span style="background-color:#E3F2FD;">2.4 LightGCN (Light Graph Convolutional Networks)</span>**

- GCN 중에서도 추천 시스템에 특화된 GNN으로, 복잡성을 제거하고 최적화한 버전  
- GNN의 구성 요소 2가지를 제거함.  
    - 특징 변환 제거
    - 비선형 활성화 함수 제거  

-> LightGCN은 오직 정규화된(Normalized) 이웃 임베딩의 합(Aggregation)만으로 구성, CF의 본질에 충실하게 ‘그래프 구조’와 ’전파’라는 뼈대만 남긴, 매우 가볍고 효율이 좋은 모델

### **<span style="background-color:#E3F2FD;">2.5 GNN의 장단점</span>**

- 장점  
    - 고차원 협업 시그널 포착
    - 정형화되지 않은 데이터 학습 가능
    - Cold Start 문제 완화에 효과적

- 단점  
    - 과평탄화 문제
    - 과압축 문제
    - 계산 복잡도 증가
    - 해석의 어려움

## **<span style="background-color:#FCE4EC;">3. Transformer 기반 추천 시스템</span>**

### **<span style="background-color:#E3F2FD;">3.1 기존 추천 시스템의 순차적 한계</span>**

- CF의 순차적 한계  
    - 순서 무시  
    - 문맥 부족  
    - 예측의 한계  

- GNN의 순차적 한계  
    - 지역적 정보 전파의 한계  
    - 장기 의존성 문제

### **<span style="background-color:#E3F2FD;">3.2 Transformer: 사용자 행동을 순서(Sequence)로 접근</span>**

행동 순서가 중요한 이유 -> 의도 파악(Transformer는 시퀀스 내의 문맥과 핵심 의도를 파악하는 데 강력함.)

### **<span style="background-color:#E3F2FD;">3.3 Transformer 기반 주요 추천 시스템 모델</span>**

- SASRec: 단방향 어텐션으로 과거 정보만 활용하여 미래 행동을 추측하는 방식으로 훈련을 받은 추천 시스템 모델  
    - 과거 정보만 활용하여 시간의 흐름 명확하게 반영
    - 문제점: 계산량이 많고 단방향성  




<br>

- BERT4Rec: 양방향 어텐션으로 과거 정보와 미래 정보를 활용하여 예측하는 방법을 훈련한 모델
    - 학습 과정에서 시퀀스 전체의 선호도를 파악하는 데 더 초점이 맞춰져있음.
