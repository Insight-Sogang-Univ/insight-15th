# 생성모델

## 1. 생성모델의 개념

### 1.1. 분류모델 vs. 생성모델
1. 분류 모델
* 분류 모델을 학습시킬 때, 훈련 데이터의 각 샘플에는 정답 레이블이 있음 > 어떤 데이터가 정답에 속할 지를 결정하는 모델
* 학습 목표: P(Y∣X) 모델링 (데이터 X가 주어졌을 때 정답 Y일 확률 계산).
* 한계 : 본 적 없는 이상한 데이터(예: 외계인 그림)가 들어와도, 모델은 어떻게든 기존에 배운 라벨(고양이 혹은 개) 중 하나로 결론을 내려야만 합니다.
2. 생성모델 
* 주어진 학습 데이터를 학습해, 학습 데이터의 분포를 따르는 유사한 데이터를 생성하는 모델 = 정답 지정에 관심없음.
* 학습 목표: P(X) 모델링 (세상에 존재할 법한 데이터 X 자체의 확률 분포 파악)
* 특징: 불합리하거나 이상한 입력(Unreasonable input)에 대해서는 "이건 세상에 존재할 확률이 거의 0이야"라고 판단하며 거부할 수 있는 능력을 갖춥니다.

### 1.2 생성모델의 분류
##### 생성 모델은 "확률 밀도(Probability Density)를 직접 계산할 수 있는가?"에 따라 명시적 모델과 암시적 모델로 분류
1. 명시적 확률 밀도 모델
* 모델이 데이터의 확률 분포인 P(X)의 값을 직접적으로 수치화하여 출력할 수 있는 모델
    1. 정확히 계산 가능한 모델 (Tractable Density)
        * 수학적으로 확률값을 정확하게 계산 가능 
        * ex. 자기 회귀 모델 =  픽셀 하나하나를 순서대로 예측하며 전체 확률을 곱해서 구함
    2. 근사적 밀도 모델 (Approximate Density)
        * 확률값을 정확히 구하기 복잡해서, 근삿값이나 하한선을 계산
        * ex. VAE = 정확한 확률 대신 '변이적 하한(ELBO)'이라는 수치를 최대화하는 방식으로 학습
2. 암시적 확률 밀도 모델
* 데이터의 확률 분포 P(X)가 무엇인지 수치로 알려주지는 않지만, 그 분포를 따르는 데이터를 생성(Sampling)할 수는 있는 모델
    1. 직접 샘플링 : 신경망을 한 번 통과시키는 것만으로 데이터를 바로 만들어냄
        * ex. GAN
    2.간접/반복적 샘플링 (마르코브 체인) : 여러 단계를 거쳐 점진적으로 데이터를 완성
        * ex. 
            * GSN : 데이터의 확률 분포를 직접 계산하는 대신, 데이터 분포로 수렴하는 마르코프 연쇄의 전이 연산자를 학습해, 다음에 어떤 데이터를 생성할지 결정하는 규칙 > 딥러닝으로 학습시킴
            *확산 모델 : 노이즈를 여러 단계에 걸쳐 조금씩 제거하며 이미지를 생성

## 2. AE, VAE
### 2.1 AE
1. Auto Encoder(AE) : 입력과 동일한 출력을 만드는 것을 목적으로 하는 신경
    * 차원 축소, 특징 추출, 노이즈 제거, 이상 탐지 등 데이터 복원이나 특성 학습에 많이 사용
    * 정답이 없다는 것이 인코더 디코더 와의 차이!! = 비지도 학습
2. 구조와 작동 : 넣고, 줄이고, 다시 키우기!
    1. 압축 단계 (Encoder) = 방대한 정보를 가진 입력 데이터를 아주 작은 바늘구멍(잠재 벡터)으로 통과시키기 위해 핵심만 남기고 깎아냄
        * 인코더가 압축한 z를 Latent Vector(잠재 벡터) = 중요한 정보가 압축되어 있음
    2. 병목 통과 (Bottleneck) = 가장 중요한 정보만 남은 상태. 이 좁은 구간을 잠재 벡터(Latent Vector)라고 부름.
        * 잠재 벡터(z): 원본 데이터를 저차원으로 압축하여 함축된 정보를 저장하고 있는 벡터
    3. 복원 단계 (Decoder) = 바늘구멍을 통과한 핵심 정보만을 단서로 삼아, 다시 원래의 커다란 이미지로 복원해냅니다.
        * 디코더: 잠재표현을 풀어서 입력을 재복원하여 출력하는 네트워크 (입력과 출력이 같은지 보는거!)
3. 왜 이케 줄였다 늘렸다 함? : 그냥 복붙하는게 아니라 어느 정보가 진짜 중요한 정보인지를 고민하게 만들어서 본질 학습을 시키기 위해서!

### 2.2. VAE
1. Variational Auto Encoder(VAE)
    * 데이터의 잠재적인 구조를 학습하여, 원본과 닮았지만 새로운 데이터를 생성해내는 모델
    * AE의 결정적 한계: AE는 데이터를 잠재 공간의 특정 점으로만 보냄. 학습할 때 쓰지 않은 '빈 공간'의 점을 찍으면 디코더는 무엇을 그려야 할지 몰라 엉뚱한 결과를 출력함
    * VAE의 해결책: 잠재 벡터를 점이 아닌 가우시안 분포(범위)로 학습 > 어떻게든 채워서 그럴듯한 이미지 생성하게 만듦
2. AE vs. VAE
* 입력 데이터를 잠재 공간(Latent Space)에 어떤 형태로 배치하느냐?
    1. AE의 잠재 공간: 입력 데이터를 잠재 공간의 어떤 하나의 고정된 수치(점)로 매핑
        * 한계; 빈공간 노이즈 출력
    2. VAE의 잠재 공간: 잠재 코드 값이 평과 분산으로 표현되는 어떤 가우시안 분포(확률 분포)
        * 구조적 특징 
            1. 연속성
            2. 조밀함 -> 정규분표 주변에 모이도록 강제함
        * 장점: 자연스럽고 새로운 결과물 만들 수 있음!
3.  VAE의 구조와 작동 : 찾고, 뽑고, 다시 그리기!
* VAE는 데이터를 고정된 점이 아닌 확률 분포로 변환하여 학습
    * [Step 1] 분포 예측 단계 (Encoder) : 입력 데이터를 특정 점이 아닌, 데이터가 존재할 확률이 높은 '구역'으로 매핑
    * [Step 2] 샘플링 및 재파라미터화: 예측한 분포에서 실제 디코더에 전달할 잠재 벡터를 하나 뽑는 단계
        * 인코더가 찾은 범위안에서 무작위로 점하나(잠재벡터)를 추출함
    * [Step 3] 생성 단계 (Decoder) : 샘플링된 잠재 벡터를 단서로 사용하여 새로운 데이터를 생성 
4. 왜 굳이 확률분포로 하냐면? VAE는 생성이 목적이기 때문에!!! 그럴듯한 데이터 만들어내게 하기 위해서!

## 3. GAN
### 3.1 GAN의 개념
1.  GAN(Generative Adversarial Network) =  생성자 신경망과 판별자 신경망이 서로 적대적으로 경쟁하면서, 훈련을 통하여 자신의 작업을 점점 더 정교하게 수행하는 신경망 모델
2. 목적
    * 생성자(Generator)의 목적은 진짜 분포에 가까운 가짜분포를 생성
    * 판별자(Discriminator)의 목적은 표본이 가짜 분포에 속하는지 진짜 분포에 속하는지를 결정
    * **궁극적인 목적은 “실제 데이터의 분포”에 가까운 데이터를 생성> 판별자가 진짜/가짜를 한 쪽으로 판단하지 못하는 경계를 최적 솔루션으로 간주**

### 3.2 GAN 구조
* GAN은 두 개의 신경망이 최소-최대 게임(MinMax Game)을 수행하는 구조
    * 생성자
        * 입력: 무작위 노이즈
        * 출력: 가짜 데이터
        * 목표: 판별자를 속여서 가짜를 진짜(1)로 믿게 만드는 것
    * 판별자
        * 입력: 실제데이터랑 생성자가 만든 가짜 데이터 모두
        * 출력:  입력 데이터가 진짜일 확률(0 ~ 1)을 출력
        * 목표: 진짜는 1로, 가짜는 0으로 정확히 분류하는 분류 모델의 역할을 수행

### 3.3 GAN 장단점
1. 장점
    * 진짜 같은 가짜 생성
    * 선명도가 높음
    * 생성 속도 빠름
    * 잠재공간이 부드러움 = 보간이 가능
2. 한계
    * 학습이 불안정함 
    * 훈련 초기 = 그래디언트 소멸
    * 모드 붕괴 : 생성자가 판별자를 속이기 가장 쉬운 특정 패턴(예: 숫자 1) 하나만 주구장창 만드는 문제
    * 지푣의 부재
    * 역매핑 불가능

## 4. 확산 모델
### 4.1 확산모델 개념
1. 확산 모델: 입력 이미지에 노이즈를 조금씩 추가하여 완전한 노이즈로 만든 뒤,이를 다시 거꾸로 복원하는 과정을 학습하여 (노이즈 제거), 입력 이미지와 유사한 확률 분포를 가진 결과 이미지를 생성하는 모델
2. 목적: 순확산과 역확산 단계를 거쳐 생성된 '결과 이미지'의 확률 분포를 '원래 데이터'의 확률 분포와 최대한 유사하게 만드는 것

### 4.2. 확산 모델의 구조: 순확산과 역확산
* 순확산 (Forward Diffusion Process): 데이터에 점진적으로 (정규분포를 가진)노이즈를 추가하는 과정
    * 시작 : 원본 데이터에서 시작
    * 노이즈 추가: 원본 데이터에 점진적으로 노이즈를 추가
    * 결과 : 충분히 많은 단계가 지나면, 원본 데이터는 완전한 무작위 노이즈
* 역확산 (Reverse Diffusion Process) :  노이즈 데이터에서 원본 데이터를 재구성하는 과정
    * 시작 : 역확산 과정은 완전한 노이즈 데이터에서 시작
    * 노이즈 제거: 각 단계에서 "어떤 노이즈가 추가되었을까?”를 예측하며 점진적으로 노이즈를 제거
    * 결과 : 수십 번의 반복적인 디노이징 과정을 거쳐 점진적으로 이미지를 복원

### 4.3 확산 모델의 장단점
1. 장점
    * 압도적인 샘플 품질
    * 학습의 안정성
    * 다양성 확보
2. 단점
    * 느린 생성 속도
    * 높은 계산 비용
    * 구조적 제약

# AI 최적화
### 도입
1. 문제 1 — 이 거대한 모델을 어떻게 써먹지? = "이미 잘 학습된 모델을 가져다가 내 용도에 맞게 조금만 손보자!" → 전이학습 + 경량화
2. 문제 2 — 이 모델이 진짜 사람한테 도움이 되긴 해? = "모델이 사람의 의도랑 가치에 맞게 행동하도록 맞춰 주자!" → AI 정렬(Alignment)

## 1. 전이학습과 딥러닝 경량화
1. 전이학습? 딥러닝 모델 한계 해결하려구
2. 전의학습 정의
    * Domain은 "어떤 데이터를 다루고 있는가"입니다.  (데이터)
    * Task는 "그 데이터로 무엇을 예측하는가"입니다. (과제, 해야 하는 일)
    * Source는 전이학습의 출발점
    * Target은 전이학습의 도착점
* 따라서 전의학습은 데이터나 문제 유형이 달라도 다른 상황에서 배운 지식을 가져와서 새로운 상황에서 성능을 높이는 것 = Domain과 Task 중에 하나라도 다를 때 의미가 생김

## 2. 전이학습의 분류
1. 귀납적 전이학습:소스와 타겟의 task가 다르고 타겟 도메인에 라벨이 존재할 때
2. 트랜스듀서형 전이학습 :소스와 타겟의 task는 같지만 도메인이 다를 때
3. 비지도 전이학습: 소스와 타겟의 task가 다르고 소스와 타겟 모두 라벨이 없을 때
* 무엇을 전이하느냐에 따라서도 다름
     1. 인스턴스(Instance): 소스 데이터 중 타겟에 도움이 될만한 샘플을 골라 가중치를 부여. (데이터 레벨의 전이)
     2. 특징(Feature): 소스와 타겟의 공통적인 특징을 추출할 수 있는 '표현 방식'을 전이
     3. 파라미터(Parameter): 모델의 가중치(W)나 하이퍼파라미터 초기값을 그대로 가져옴 > **현재 딥러닝에서 주류!!!**
     4. 관계(Relation): 데이터 간의 논리적 관계나 구조적 지식을 전이

### 3. 전이학습의 현재 패러다임 
* 트랜스포머 등장 이후, 사전 학습된 거대 모델들이 등장하면서 어케 내 용도에 맞게 조정할지에 대한 문제가 대두되었음.
1. Foundation Model(파운데이션 모델): GPT나 Gemini처럼 수만 가지 일을 할 수 있는 지능의 기반이 되는 거대 모델들
2.  Fine-tuning: 파운데이션 모델을 가져왔으면, 내 용도에 맞게 조정해야 함.
### 4. 파인 튜닝의 한계를 넘어서
* PEFT:  “최소한의 파라미터만 건드려서 최대의 성능을 내자”는 전략으로, 원래 모델의 파라미터는 고정하고 별도의 작은 파라미터만 추가적으로 학습 = 모델의 모든 계산을 바꿔 기존 판단을 완전히 지우는 게 아니라, 작은 보정값만 더해서 최종 판단인 결과를 바꾸는 방
* Zero-shot Transfer: 학습 데이터가 없어도, 언어 이해 능력만을 기반으로 바로 추론을 수행하는 것
* Few-shot Transfer(In-Context Learning): 극소수의 예시만으로 새로운 태스크를 수행하는 것 > **이게 잘 되려면 예시와 입력을 잘 설계하는 것이 중요함 = 프롬프트 엔지니어링과 이어짐**
### 5. 전이학습의 한계
* Negative Transfer: 도메인이 너무 다르면 오히려 전이가 성능을 떨어트립니다. 
* Catastrophic Forgetting:Fine-tuning 과정에서 새 데이터에 너무 치우치면, 모델이 원래 가지고 있던 일반 능력을 잃어버립니다. 
* Bias Transfer: 사전 학습 데이터에 담긴 편향이 그대로 넘어오게 되는 것
* Transferability 측정의 어려움: 어떤 두 Task가 얼마나 관련되어 있는지를 사전에 정량적으로 측정하기 어려움
### 7. 딥러닝 경량화
* 성능은 최대한 유지하면서 모델을 가볍게 만들
* 가지치기 = 모델 안에 있는 수많은 파라미터 중에서 별로 중요하지 않은 것들을 그냥 잘라내는 방법
    * 파라미터 값이 0에 가까울수록  → 모델 판단에 영향을 거의 주지 않기 때문에 → 그냥 없애버리면 결과적으로 모델이 가벼워지고 속도도 빨라짐
* 양자화 = 모델의 가중치를 표현하는 숫자의 정밀도를 줄이는 방법
* 지식 증류 = 크고 똑똑한 모델(Teacher)의 지식을 작은 모델(Student)에게 가르쳐 주는 방법
* 모델 구조 경량화 = 처음부터 가볍게 설계된 모델을 쓰는 방법
* 하드웨어 가속화 = 소프트웨어적으로 모델을 줄이는 게 아니라, AI 연산에 특화된 하드웨어를 쓰는 방법

## 2. AI Alignment
### 2.1 AI Alignment
1. **AI Alignment(인공지능 정렬)**는 단순히 모델의 성능을 높이는 단계를 넘어, AI가 **인간의 의도(Intent), 목표(Goal), 그리고 윤리적 가치(Values)에 부합하도록 행동**하게 만드는 연구 분야!!
    * *요즘 정렬의 기준이 되는 세가지 지표는 Helpful(도움이 되는가) / Honest(정직한가) / Harmless(무해한가)*
2. Outer Alignment vs Inner Alignment
    1. Outer Alignment: "목표를 제대로 설정했는가?"
    * 모델이 보상을 극대화하기 위해 인간의 의도와는 무관한 '꼼수'를 부리는 현상을 해결하기 위해 진행
    2.  Inner Alignment: "모델이 내면에서 무엇을 최적화하는가?"
    * 모델이 내부적으로는 다른 목표를 최적화해 버리는 문제
3. 정렬을 어렵게 만드는 요인
    * 가치 정렬의 모호성 : 가치가 사람마다 다름... 기술적 난제로 이어짐
    * 확장 가능한 감독 : 인간이 이해하지 못하는 답변에 대해 어떻게 피드백을 줄 것인가?

### 2.2 RLHF
* 모델이 문장은 잘 생성하지만, 사람이 원하는 답을 이해하지는 못해 원하는 답변을 제공하지 못하는 Outer Alignment 문제
1. RLHF(Reinforcement Learning from Human Feedback): 이를 해결하기 위해 인간의 피드백을 기반으로 모델을 강화학습으로 조정하는 방법
* RLHF는 SFT → Reward Model → RL 의 3단계로 진행
    * SFT: 모범 답안 학습
    * RM: 인간의 선호도 학습
    * RL: 모델 최적화 (높은 점수를 받도록 계속 연습해 봐.)
* 모델이 단순히 “정확한 답”을 맞히는 것을 넘어서, “사람에게 더 유용하고 적절한 답”을 생성하도록 유도
* 주의: 정렬의 부작용 = 보상해킹 => 기만적 행동 가능성 있음

## 3. 프롬포트 엔지니어링
*  AI가 더 좋은 답변을 내도록 입력을 설계하는 기술
* 주요 기법
    1. Zero-shot → 예시 없이 바로 질문
    2. Few-shot → 예시를 몇 개 보여주고 질문
    3. Chain-of-Thought (CoT) — 생각의 사슬 = AI한테 답만 내놓으라고 하지 않고, 풀이 과정을 단계별로 생각하게 만드는 기법
    4. Role Prompting — 역할 부여 = 특정 역할을 맡겨서 그 역할에 맞는 답변을 끌어내는 기법
    5. Instruction Prompting — 명확한 지시 = AI한테 원하는 걸 최대한 구체적으로 지시하는 기법
    6. Self-Consistency = 같은 질문을 여러 번 해서 가장 많이 나온 답을 채택하는 기법
    7.  RAG (Retrieval-Augmented Generation) — 검색 증강 생성 = AI가 모르는 최신 정보나 특정 도메인 지식을 외부 데이터베이스에서 검색해서 프롬프트에 함께 넣어주는 방식
