# 시계열 데이터의 정의
## 시계열 데이터란?
* 순차적인 시간의 흐름에 따라 수집된 데이터
* 연속한 관측치는 서로 상관관계가 존재하며, 순서가 중요한 시퀀스 데이터로 간주
* 단순 시간-수치로 이루어진 데이터 뿐만 아니라, 단어의 순서가 중요한 NLP 데이터에도 적용됨!
* **핵심: 과거 데이터가 미래에 어떤 영향을 미치느냐?**

## 단변량 vs. 다변량 시계열
1. 단변량 시계열 : 하나의 숫자가 시간에 따라 변하는 것
* 단일 변수로 이루어진 시계열 데이터
* 아리마 지수평활법
2. 다변량 시계열 : 여러 숫자가 동시에 시간에 따라 변하는 것
* 같은 **시간**에 **둘 이상의 변수**로 이루어진 시계열 데이터
* 전이함수모형 등을 사용함

## 시계열의 데이터의 특징
1. 시간의존성 = 시계열 데이터는 시간에 따라 변함
2. 계절성 = 일정주기마다 반복적으로 나타나는 데이터 변동
3. 추세 = 시간의 흐름에 따라 나타나는 장기적인 증가/감소 경향
4. 자기 상관성 = 자기자신끼리 상관관계를 갖는 특성 = 독립적이지 않음
    * 이 때문에 회귀분석이 불가능함 !
        * 오차항 독립성 위반
        * 허위 상관관계 문제 > 추세 제거를 통해서 해결 가능
        * 에측이 어려워지고, 모델 복잡성 증가(일반적인 회귀가 안되니까 AR,ARIMA등의 시계열 모델로 대체해야함!)

## 시계열 데이터의 분석요인
1. 추세요인 : 전반적으로 증가/감소하는 경향
2. 계절요인 : 일정한 주기로 반복되는 패턴
3. 순환요인 : 계절성보다 더 길고 불규칙하게 반복되는 변동
4. 불규칙요인 : 앞선 3가지를 다 제거하고도 남은 예측 불가능한 변동

# 시계열 분석의 워크 플로우
## 데이터 전처리
* 시계열 데이터 만의 특징이 존재하기 때문에 왜곡되지 않도록 전처리가 필요하다.
* 과정
    1. 데이터 확인/ 시각화
    2. 결측치 확인 → 결측치 처리(삭제/단순 대치/보간/이동평균 대치 등)
        * 결측치란 값이 측정되지 않았거나 누락된 상태
        * 처리 방법
        1. 보간법: ㄷ누락된 값을 추정하는 수학적 기법 (선형보간,다항식보간,스플라인보간,조각별 상수보간 등 다양함)
        2. 단순 삭제 머신러닝 등등 더 존재함
    3. 이상치 탐지 → 이상치 처리(삭제/단순 대치/보간/이동평균 대치 등)
        * 값은 있는데 일반적이지 않은 경우 > **사전지식 필요**
        * 탐지 : IQR,Z-score,STL분해
        * 탐지 모델 : 분류 및 회귀트리(CART), 클러스터링 기반 이상 탐지, 오토인코더
        * 처리는 결측치랑 동일
    4. 평활화(EDA) 진행 → 이동평균 이용
        * 평활화란 시계열 데이터에서 노이즈를 줄이고, 전체적인 흐름이나 추세를 더 명확하게 파악하기 위해서 데이터를 부드럽게 만드는 기술
        * 이동평균: 시계열의 부분집합으로 평균을 구하여 평활화 효과를 얻는 방법 = 윈도우(평균을 구할 구간의 크기) 내의 평균을 구하는 방법
        * 목적 : 노이즈 제거, 추세 파악, 예측의 기초자료
        * 유형, SMA, WMA, EMA
    5. 요소 분해 → 가법 모형, 승법 모형
        * 주어진 시계열 데이터를 개별 성분(추세성, 계절성, 순환성, 불규칙성)으로 분해하는 과정
        * 일반적으로 순환성은 장기간에서만 고려되므로 순환성분은 0으로 설정함
        * 왜 해요? : 시계열 데이터 = 추세/계절성/순환성/불규칙성 등 다양한 변동이 결합된 형태 > 각 요소를 분리하면 각각의 패턴을 명확히 이해 가능
        * **가법모형** : 각 변동 요인들이 독립적으로 더해지며, 모든 요소의 영향이 일정함 (시계열 전체에서 계절성의 진폭이 일정할 때 적합)
        * **승법모형** (로그 변환해서): 시계열 관측값은 변동요인들의 승법(*)으로 설명될 수 있다 가정 = 추세가 커질수록 계절성이나 불규칙성의 효과도 비례해서 커짐

## 정상성 판단
* 정상성: 시계열 데이터의 평균, 분산 등 통계적 특성이 시간에 따라 일정한 수준으로 유지되는 것 
* 정상성이 있는것을 정상시계열이라고 한다. > 이게 있어야 예측이 가능하다. 
1. 강한 정상성 - 완벽하지만 비현실적인 기준 = 언제나 확률 분포가 똑같음
2. 약한 정상성 - 실무에서 쓰는 현실적인 기준 = 평균, 분상ㄴ이 일정하고, 자기 공분산이 시차에만 의존함. **이게 우리가 필요한 것**
* '약 정상성’을 띠는 시계열 데이터는 시간이 흘러도 평균, 분산이 변하지 않으며 자기공분산이 시차에만 의존한다!
3. 정상성 판단하기 
    1. 시각적으로 판단하기 - 추세나 계절성이 있는경우 > 근데 이것도 차분하면 해석 가능
    2. 통계정 검정 => 시각적으로만 알기 어려워서, 수치적인 가설 검증이 필요함. = 단위근이 있는지 확인해야 함. 
    * 단위근 = 과거 충격이 영구적으로 누적되는 특성을 의미 - 대표적인 비정상성의 원인으로 존재
    * **이러한 단위근을 검정하는 것이 ADF와 KPSS**
        1. ADF : 단위근의 존재 여부를 검정하여 시계열의 정상성을 판단하는 방법. "이 시계열이 비정상이 아님을 증명하자" 는 관점에서 출발
        * 과거 값의 영향이 시간이 지남에 따라 점점 사라지면 정상 시계열, 누적되어 사라지지 않으면 단위근이 존재하는 비정상 시계열로 판단
        2. KPSS : 이 시계열이 정상임을 반증하자" 는 관점에서 출발하는 검정 = ADF 검정력이 낮아서 함께 사용
        * 평균이 시간에 따라 흔들리지 않으면 정상, 평균 자체가 계속 변하고 있으면 비정상으로 판단
4. 정상성을 어떻게 확보할까
    1. 추세제거 = 시계열이 지속적으로 증가/감소할 때 적용
        * 차분 = 연이은 관측값들의 차이를 계산
    2. 계절성 제거 = 시계열이 주기적인 패턴을 가질 때 적용
        * 계절 차분 -> 평균이 시간에 따라 흔들리지 않으면 정상, 평균 자체가 계속 변하고 있으면 비정상으로 판단
    3. 분산 안정화 = 시계열의 변동 폭이 점점 커지거나 작아질 때 적용
        * 로그 변환 => 비율적인 변화 강조 가능 > 분산을 적절하게 만들 수 있음

## 모델링 및 예측
### 안정적 시계열 분석
1. 자귀회귀모델 : 자기 자신의 데이터가 미래 값을 설명해줄 것이라고 가정하는 것 = 자기상관성을 활용
    * AR(p): P개 시점의 선형 조합을 이용해 예측하는 모델
    * AR(1): y의 현재값은 1 시점 이전의 y값의 영향을 받는다고 해석!
2. 이동평균 모델 : 과거 시점의 잔차를  독립변수로 삼아 현재 값을 예측하는 모델
* 이떄 잔차는 실제 - 예측값 이며, 모두 백색 잡읍이면서 서로 독립적이다.
    * MA(q) : q차 이동평균 모형 = 과거 q개의 예측 오차의 선형 결합으로 예측하는 모델
    * MA(1) : y의 현재값은 1 시점 이전의 오차의 영향을 받는다
* 전제 조건 
    1. 정상성 : AR과 MA 모형은 안정적 시계열 분석으로 확률적 안정성을 전제로 추정 + 모델 구조까지!!!
        * AR 모형의 정상성 조건 : 모든 해(z)의 절댓값 > 1 ⇒ AR의 정상성 확보
        * MA 모형의 정상성 조건 : 잔차들의 조합으로 이뤄진 MA 모형은 항상 정상성 만족!
    2. 가역성 : MA 모형을 AR 형태로 재표현할 수 있는지 여부 = 오차항도 관측 가능한 과거값으로 표현할 수 있어갸 안정적이어어서!
        * AR 모형의 가역성 조건 : 체크할 필요 없음 
        * MA 모형의 가역성 조건 : 모든 해(z)의 절댓값 > 1 ⇒ MA의 가역성 확보
3. ARMA 모형 : AR과 MA 모형을 결합한 것 즉, AR(p) 모형의 정상성 + MA(q) 모형의 가역성 → 이 둘의 조건 모두 만족해야 함!
4. 차수 결정 방법 : 자기 상관 함수 = 두 그래프를 통해 현재 값이 과거 값과 어떤 관계가 있는지 파악하고 이를 통해 AR, MA, ARMA의 적절한 차수(p, q)를 결정
* 자기 상관 계수 : 간접영향 까지 모두 파악
* 부분 자기 상관 계수 : 시점 결정하는거라서 직접적인 영향만 파악

### 불안정 시계열 분석
* 프로세스 : 비정상 시계열 데이터 > 정상성 확보 > 정상시계열 > ARMA를 이용해 안정적 시계열 분석 ! 즉 안정적 시계열 모델에 정상성 확보 과정이 추가된 형태임
1. ARIMA : 차분, 변환을 통해 AR,MA, ARMA로 정상화! 랜덤워크이거나 표류를 포함하는 랜덤워크일 떄 독특하다
2. SARIMA : 시계열 데이터의 계절성을 처리할 수 있는 모델 = 아리마 모델에 계절성 변동을 반영한 모델
* 하이퍼파라미터 결정법
    1. 차분, 계절 차분 결정: 단위근 검정으로 정상성 확보 여부 판단 후 결정
    2. AR,MA 차수 결정 : ACF,PACF로 범위를 추정한 뒤 그래프로 살펴보고, AIC,BIC 같은 정보 기준으로 그리드 서치 후, 최종적으로 선택한다

## 다변량 시계열 분석
1. VAR : 자신의 과거 값과 다른 변수들의 과거 값을 함께 사용하여 미래를 예측하는 모델
* 적합한 경우 : 변수들이 서로 영향을 주고 받을 때, 상호 의존성이 중요할 떄
2. VAR를 활용한 분석 방법
    1. 충격 반응 함수 = 한 변수의 충격이 다른 변수에 미치는 영향을 시간 흐름에 따라 분석
    2. 예측오차 분산 분해 = 예측 오차에 각 변수가 얼마나 기여하는지 분석

## 딥러닝을 활용한 시계열 분석
1. 순환신경망 계열 
* 적합한 경우 = 순차적 패턴 중요할 때, 과거 정보가 미래 예측에 영향을 줄 때 등
2. 트랜스포머 계열
* Informer (장기 예측 특화)
    * 특징 : 불필요한 연산을 줄여 계산 복잡도를 대폭 낮춘 모델
    * 실무 강점 : 매우 긴 시계열 데이터 처리에 강력함
* Autoformer (주기성 포착 특화)
    *특징 : 시계열의 전통적 기법인 '요소 분해(추세/계절성)' 개념을 딥러닝 구조에 직접 결합한 모델
    * 실무 강점 : 연간 계절성 등 주기 패턴이 뚜렷하게 나타나는 판매량 예측에 압도적인 성능을 보임
* TFT (Temporal Fusion Transformer - 실무 활용도 1위)
    * 특징 : 과거 데이터뿐만 아니라 다양한 외부 변수(날씨, 휴일 유무 등)를 섞어서 예측할 수 있는 다목적 아키텍처
    * 실무 강점 : 어떤 변수가 예측에 가장 중요했는지 어텐션 스코어를 통한 해석이 가능해, 설명력이 중요한 유통/비즈니스 수요 예측에 가장 많이 쓰임
3. foundation 모델 = 거대한 데이터로 사전 학습하여 다양한 문제를 해결할 수 있는 범용 모델
    1. LLM 기반 모델
        1. Without Adaptation
        2. Adapt LLM
        3. Adapt to LLM
    2. 시계열 전용 파운데이션 모델
