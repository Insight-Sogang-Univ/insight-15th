## 1.통계란?

### 1-1 통계에 대해서

모집단 전체에게 물어보는 것은 비현실적인 방법이기 때문에 보통 일부 샘플(표본 집단)만 추출하여 분석
1. 샘플자체에 대한 관심 -> 기술 통계
        기술통계란? 
            수치와 시각화를 통해 샘플을 요약하여 표현하는 방법론
2. 샘플의 모집단에 대한 관심 -> 추론 통계
        추론 통계란?
            추정과 검정을 통해 샘플로부터 모집단을 추론하는 방법론

### 1-2 자료의 척도

- 자료:
    - 양적자료:수치로 표현됨
        - 구간 척도: 값들 사이에 순서 존재, 그 간격이 일정하여 연산 가능
            ex) 시간, 온도
        - 비율 척도: 임의의 측정 단위로 구성됨
            ex) 시간, 무게
    - 질적자료:수치로 표현되더라도 연산의 의미를 가지지 못함
        - 명목 척도: 레이블이나 이름의 요소의 속성을 식별 
            ex) 성별(남:0, 여:1)
        - 서열 척도: 명목형의 분류 속성 + 항목 간의 순서 또는 순위
            ex) 만족도, 직급 체계

## 2.기술통계

### 2-1 기술통계란?

- 데이터를 있는 그대로 설명하고 이해하는데 중점을 둔 통계법
    ex) 평균 매출, 불량률, 연령 분포

### 2-2 3가지 주요한 도구 집합

1. 중심화 경향: 데이터의 중심적인 경향을 나타내는 기술 통계량
    - 평균: 데이터 값의 총합을 데이터의 개수로 나눈 값 / 절사 평균은 이상치를 잘라 평균을 계산한 것
            `series.mean()`
    
    - 중앙 값: 데이터를 크기 순서대로 일렬로 정렬시 한가운데 위치한 값
            `series.median()`
    
    - 최빈값: 데이터에서 가장 자주 나타나는 값
            `series.mode()`

2. 자료의 산포: 데이터가 퍼져있는 정도를 설명하는 기술 통계량
    - 범위: 데이터의 최대값과 최소값의 차이, 이상치에 민감함
            `series.max() - series.min()`
    
    - 분산: 데이터 값들이 평균으로 부터 퍼진 오차의 제곱의 평균
            `series.var()`
    
    - 표준편차: 분산의 제곱근
            `series.std()`
    
    - 사분위수: 데이터의 표본을 네개의 동일한 크기로 나눈 값 -> Q1,Q2,Q3 각각 하위 25%, 하위 50%, 하위 75%
            `series.quantile(0.25)
            np.percentile(series, 75) -> numpy로 실행하는 법`
    
    - 사분위간 범위: Q3-Q1은 데이터의 중앙 50%가 퍼져있는 범위이다. 위 그래프의 사각형 맨아래부터 순서대로 Q1, Q2, Q3
    - 변동계수: 데이터의 상대적인 변동성을 나타내는 지표
            - 변동계수 = (표준편차/평균)*100


3. 자료의 정리 및 시각화

cf) 첨도와 왜도
    - 첨도: 확률 분포의 꼬리가 두꺼운 정도
    - 왜도: 확률 변수의 확률 분포 비대칭성을 나타내는 지표

## 3.확률과 분포

### 3-1 평균의 불확실성
    - 평균은 이상치를 제외하더라도 완전히 모든 표본을 대표한다고만 볼수는 없음
    - 그 이외에도, 표본의 불확실성, 모수추정, 추정의 현실성, 구간 추정의 오류 관리, 적극적 주장 검증, 가설검증의 오류 관리 등이 존재

### 3-2 확률분포
    - 확률변수는 모든 경우의 수를 특정 값으로 대응시키는 함수
    - 확률분포는 확률 변수가 특정한 값을 가질 확률을 나타내는 함수
        - 이산 확률 분포: 확률 변수가 유한하거나 셀 수 있는 값을 가지는 경우
        - 연속 확률 분포: 확률 변수가 연속적인 값을 가지는 경우

### 3-3 대표적인 분포 모형

    - 정규분포: 
        - 매개변수로 평균, 표준편차가 존재하며, 평균을 기준으로 대칭 
        - 평균을 기준으로 대칭, 표준편차가 클수록 곡선이 평평해짐
        - 표본의 크기가 커질수록 표본 평균의 분포는 정규분포에 가까워짐

    - Z분포:
        - 평균이 0이고 표준편차가 1이 되도록 맞춘 정규분포
        - 많은 데이터가 따르는 대표적인 확률분포

    - T분포:
        - 매개변수는 자유도
            - 자유도: 통계에서 얼마나 많은 정보가 독립적으로 변할 수 있는지
        - 정규분포보다 꼬리가 두꺼움, 평균을 기준으로 대칭
        - 모평균을 추정할때 사용
        - 모집단의 분산을 모를때, 모분산을 추정하여 만들어 표본의 크기가 작을수록 불확실성이 커짐(표본 크기 30미만일시 사용)

    - 카이제곱분포:
        - 매개변수는 자유도
        - 0에서 시작하여 오른쪽으로 긴 꼬리를 지님
        - 자유도의 증가에 따라 중심이 오른쪽으로 이동, 대칭적으로 변함
        - 표준 정규 분포를 따르는 변수들의 제곱 합

    - F분포:
        - 매개변수는 자유도(df1, df2)
        - 성질은 카이제곱분포와 동일
        - 두 독립적인 카이 제곱 분포의 비율 -> 두 집단 간 분산 차이 비교

### 3-4 분포모형을 알아야하는 이유

    - 가설 검증을 위한 통계량 산출에 사용
    - 표본 추출의 과정에서 인위적인 조작 여부 확인 -> 표본을 random하지 않게 샘플링을 했다는것

## 4.추론통계 - 추정

### 4-1 추론통계란?

- 수집한 데이터를 기반으로 전체 정보를 추론하고 결론을 도출하는 방법
    - 추정: 모집단을 대표할 수 있는 일부를 표본으로 추출, 통계량을 이용하여 모수를 예측
        - 모수: parameter을 뜻하며, 관심의 대상이 되는 미지의 참 값
        - 추정량: 표본 정보에 의존하는 확률변수
        - 추정치: 표본이 추출된 후 표본자료를 이용하여 계산한 통계량의 특정 값
    - 검정: 표본을 추출하여 얻은 표본 통계량으로 모집단 모수에 대한 가설의 평가

### 4-2 점추정

- 어떤 모수를 표본에 기초하여 계산된 하나의 값을 추정하는 것

### 4-3 구간추정

- 모수의 참값이 포함되리라고 기대하는 추정치를 일정한 범위와 신뢰도를 가지고 나타내는 방법
- 모수는 고정된 값이기 때문에, 표본을 뽑아서 구간을 측정하기 때문에 일정확률로 특정 구간에 존재한다고 볼수 없다.

## 5.추론통계 - 검정

### 5-1 가설 검정의 절차

- 가설 검정은 표본의 정보를 사용하여 가설의 합당성여부를 판정

    1. 가설설정
        - 귀무가설 H0: 기각하고자 하는 명제(통용되는 가설)
        - 대립가설 H1: 내가 주장하고자 하는 명제
    
    2. 유의 수준 설정
        - 결과를 판단할 때 허용할 수 있는 오류의 최대치(약 5% 설정)
        - 1종, 2종 오류
            - 1종 오류: 귀무가설이 실제로 참이었는데 기각할 확률 -> 유의 수준
            - 2종 오류: 귀무가설이 실제로 거짓인데 기각하지 못할 확률
    
    3. 검정통계량 산출
        - Z값: 표준정규분포를 사용하여 얻은 값
            - $Z=(X−μ)/σ$ (X: 관측값, μ: 평균, σ: 표준편차)
            - 큰 표본 크기에 대한 가설검증 때 사용
        
        - T값: T분포를 사용하여 얻은 값
            - 작은 표본크기에 주로 사용, 불확실성을 잘 반영
            - $T = \frac{X - \mu}{s / \sqrt{n}}$ (X: 표본 평균, μ: 설정값, s: 표본표준편차, n: 표본 크기)
            - 두 표본 평균 간의 차이 검정
        
        - F값: F분포를 사용하여 얻은 값
            - 두 표본 분산 사이의 차이를 비교하는데 사용
            - F분포는 두개의 카이제곱 분포된 변수들의 비율로 정의
            - $F = \frac{S_{bet}^2}{S_{wit}^2}$ (표본 평균 간 퍼진정도/표본 내에서 퍼진 정도)
            - 두 개 이상의 그룹 간 평균 차이 검정할때 사용 -> F가 1에 가까울수록 분산 유사
    
    4. 가설 기각/채택 판단
        - 유의확률
            - 가설 검정에 쓰일 수 있는 척도. 유의 확률이 유의 수준보다 작을시 귀무 가설 기각

### 5-2 데이터의 자료형에 따른 검정 방법

#### 1. 상관분석
        `plt.scatter(df['colx'], df['coly'])
         sns.pairplot(df)`
        - 두 변수간 피어슨 상관계수가 얼마나 유의미한지 검증
        - 공분산: 두 변수가 함께 어떤 방향으로 변화하는지와 그 크기를 표현
            - Cov(X,Y):
                - 양수: X와 Y는 같이 커짐
                - 음수: X가 커지면 Y는 작아짐
                - 0 : 선형적인 관계는 없으나, 비선형적인 관계 존재할 수 있음
                - `np.cov(df['col1'], df['col2'])`

        - 상관계수: 두 변수 간의 선형적인 관계를 나타내는 통계적 지표
            - -1, 1사이의 값을 가짐
                - Corr(X,y):
                    - 양수: 양의 상관관계
                    - 음수: 음의 상관관계
                    - `df[['col1', 'col2']].corr()`

#### 2. 범주형-수치형 : F-test, T-test
        `sns.barplot(x = 'col1', y = 'col2', data = df)`
        - 이때 검은 선은 신뢰구간을 의미한다
        - T-test: 범주가 2개 -> t값은 두 평균 간 차이를 표준 오차로 나눈 값
            - 두개씩 따로따로 묶어서 진행
        - F-test: 범주가 3개 이상 -> T-test로 할시 묶을 가짓수가 너무 많아지므로 분산 분석을 사용
            - 집단 간 분산 / 집단 내 분산으로 계산

#### 3. 범주형 - 범주형 : Chi2-test
        `from statsmodels.graphics.mosaicplot import mosaic
        mosaic(df, ['col_행','col_열'])
        plt.axhline(표시할 값, color = '원하는색')`
        - 시각적으로 범주별 차이가 명확하게 보여야함
        - Chi2-test:
            - 두 범주형 변수의 독립성을 검정하는 법
            - 기대 빈도와 관측 빈도의 차이의 제곱을 기대 빈도로 나눈 합을 통해 카이제곱 통계량을 구함
            - 해석은 자유도의 2배 가량으로 기준을 잡음
            - 범주형 변수의 자유도는 범주의 수 -1
            - `pd.crosstab(df['col_행'], df['col_열'])`
