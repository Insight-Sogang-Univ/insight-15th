## 1.머신러닝과 모델링

### 1-1 머신러닝

- 데이터에서 패턴을 발견하는 것과 이 패턴을 바탕으로 새로운 데이터에 대해 예측하는 것을 가능하게 함
a. 머신러닝의 종류
    - 지도학습: 
        - 정답 label이 있는 데이터 사용 
        - 기계의 예측값과 데이터의 정답과 같아지도록 학습
        - 회귀, 분류가 지도학습에 해당된다 / 사진을 보고 강아지, 고양이인지를 판별
    - 비지도학습:
        - 정답 label이 없는 데이터 사용
        - 데이터의 패턴 혹은 유사성을 학습
        - 군집화 / 데이터의 특성을 학습해 스스로 패턴을 파악하여 분류하는 것
    - 강화학습: 
        - 시행착오를 반복해 보상을 최대화 하는 행동 학습
        - 자율주행, 알파고

### 1-2 모델링
a. 모델이란?
    - 모델은 학습을 마친 인공지능 그 자체를 의미함

    - 패턴이 저장되어 있으며, 새로운 데이터를 넣었을 때 예측기 역할을 함

b. 모델링이란?
    - 예측기를 만들어가는 전체적인 과정
c. 모델링의 과정
    - 설계(데이터 특성에 맞추어 어떤 알고리즘으로 학습할지) -> 훈련(데이터를 기계에 보여주어 정답을 맞추도록 수정) -> 검증(훈련시킨 모델이 새로운 데이터에 대해 정답을 잘 맞히는지 테스트)

### 1-3 지도학습
a. 지도학습이란?
    - 정답 레이블을 함께 주고 학습을 시키는 방법
    - 예측값과 레이블이 같아지도록 학습

b. 지도 학습의 과정
    - labeled data:

        - 독립변수(feature) : 예측에 사용되는 입력 변수

        - 종속변수(target/label) : 예측하고자 하는 대상 변수

        - train set(학습시킬때 사용하며, 독립변수를 통해 예측값, 종속변수가 같아지도록 훈련시킴), test set(학습된 모델을 검증)으로 labeled data를 나눔 -> 정답 label을 모델이 외우는 것 방지
c. 회귀와 분류
    - 회귀: 주어진 데이터를 기반으로 정답을 잘 맞히는 함수를 찾는 문제
    - 분류: 기존데이터가 어떤 레이블에 속하는지 패턴을 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별하는 것
    - 회귀 vs 분류: 회귀는 종속변수가 연속형, 분류는 종속변수가 범주형

## 2.회귀

### 2-1 회귀 개요

- 데이터에서 패턴을 찾아내어 미래 값을 예측하는 것
- 종속변수가 연속적이어야 함

### 2-2 선형 회귀

a. 선형회귀란?
    - 데이터를 가장 잘 설명하는 회귀선을 찾는 과정
    - 현재 데이터를 요약하는 동시에, 미래의 값을 예측하는 도구가 됨
    - 단순선형회귀: x가 한개, 다중선형회귀: x가 여러개

b. 가중치와 편향의 이해
    - 가중치 w:
        - 독립변수 x가 종속 변수 y에 주는 영향력의 크기
        - 그래프에서는 기울기로 나타남
    - 편향 b:
        - 독립변수 x가 0일때 나오는 기본값, y절편으로 나타남
        - 데이터에 더 유연하게 최적화될 수 있도록 위아래로 움직여주는 역할


c. 단순선형회귀
    - 하나의 독립변수만 존재하는 선형 회귀
    - 데이터를 직선 형태로 표현하는 경우 가장 데이터를 잘 설명하는 $y=wx + b$ 를 찾는것
    - 잔차를 줄여가며 w와 b를 찾아나감

d. 잔차
    - 실제값과 예측값 간 차이

### 2-3 최적의 w와 b를 찾는 두가지 방법

- 눈으로만 봐서 최적의 가중치와 편향을 알수 없기 때문에 사용하는 방법

a. 최소제곱법
    - 최적의 회귀선을 찾는 방식으로, 잔차의 제곱의 합이 최소가 되도록 함

b. 경사하강법
    - 목적함수: 달성하고 싶은 목표
    - 손실함수; 개별 데이터에 대한 오차를 측정하는 함수
    - 비용함수: 전체 데이터셋에 대한 손실함수를 합산하거나 평균을 내어 전체적인 성적 평가

    - 손실함수와 경사하강법:

        - $E = \frac{1}{n} \sum_{i=1}^{n} [t_i - y_i]^2$

        - 잔차 제곱의 평균(손실함수)을 최소화하는 W, b를 찾는 것이 목표이다.

        - cost가 최소화가 되는 지점은 접선의 기울기가 0이 되는 지점이자, 미분값이 0이 되는 지점

        - 이때, 현재의 W에 접선의 기울기를 구해 학습률을 곱한 값을 빼서 새로운 W로 사용

    - 학습률: 보폭의 크기라고 생각할 수 있다.

        - Learning Rate가 너무 작다면, 모델이 수렴할때 까지 너무 오랜 시간이 걸림

        - Learning Rate가 너무 크다면, 모델은 수렴하지 못하고 발산

        - 적당한 학습률이 좋음

    - Local Minima 문제:
        - 모든 비용함수가 하나의 최솟값을 가지는 것은 않음

        - Global Minima를 구해야하지만, Local Minima에만 머물고있을수 있음


    - 해결법:
        - 기존 경사하강법: 현재 기울기만 기준으로 움직임

        - 이전 기울기와 방향을 고려하도록 관성을 부여해주는것(모멘텀 부여)

        - Local Minima를 모멘텀을 통해 탈출할 수 있도록 도와줌

### 2-4 다중선형회귀

- 독립 변수가 두개 이상인 경우를 뜻함

- 독립변수가 두개인 경우, 3차원 공간에 표현되며, 회귀선은 평면이됨

a. 다중공선성
    - 독립변수간의 강한 상관관계로 인해 회귀 모델이 불안정해지는 문제

    - 어떤 독립변수가 종속변수에 얼마나 영향을 미치는지 정확하게 구분이 어려워짐

b. 다중공선성 확인 방법
    - 상관계수
        - 상관계수를 통해 상관관계의 정도를 수치화

        - -1,1사이의 값을 가짐

        - 양의 상관관계: 한 변수가 증가하면 다른 변수도 증가

        - 음의 상관관게: 한 변수가 증가하면 다른 변수는 감소

        - 0: 두 변수는 직접적인 연관성이 없다

    - VIF지수(분산 팽창 인수)

        - 회귀 모델의 결정계수를 사용하여 계산

        - VIF가 높으면 다중 공선성 존재한다 판단

        - VIF = 1: 다른 변수와 상관관계가 전혀 없음

        - VIF < 5: 일반적으로 다중 공선성 문제가 없음

        - VIF > 5: 다중공선성의 징후가 있음

        - VIF > 10: 다중공선성이 심함

c. 다중공선성 대처 방법
    - 변수 제거
    - 변수 변환
        - 변수들끼리의 연산을 통해 새로운 변수 생성
    - 규제 선형 모델 활용
        - 릿지, 라쏘, 엘라스틱넷 등을 통해 복잡도 감소
    - PCA
        - 데이터의 차원을 축소하는 통계적 기법

        - 상관 있는 변수들을 묶어서 더 적은 수의 대표 축으로 바꾸는 것

        - 데이터 전처리 -> 주성분찾기 -> 차원 축소의 과정을 거쳐 실행

### 2-5 규제선형모델

a. 규제선형 모델이란?
    - overfitting 되지 않도록 규제를 가하고자 등장한 모델 : 선형회귀모델에서 계수의 크기를 조정
    - 특정 독립변수에 과하게 의존하여 발생

b. 규제 선형모델의 종류
    - L2규제: W의 제곱에 대해 페널티를 부여하는 방식

    - L1규제: W의 절댓값에 대해 페널티를 부여하는 방식

    - L1+L2규제: 엘라스틱넷 회귀

c. 릿지 회귀(L2 규제)
    - $\text{Cost Function} = \text{RSS} + \alpha \sum_{j=1}^{p} w_j^2$
    - 회귀계수의 제곱을 페널티로 주어, 큰 회귀 계수일수록 큰 페널티를 받음.

d. 라쏘 회귀(L1 규제)
    - 절댓값을 페널티로 추가하여, 회귀계수의 크기와 상관 없이 같은 페널티를 받음

e. 엘라스틱넷(L1+L2)
    - L1, L2규제를 동시에 사용하여, 불필요한 변수는 제거하고, 남은 변수들은 적절한 크기로 유지
    - but 계산량이 증가해 속도가 느림

## 3.모델 평가 방법

### 3-1 성능 평가 지표

a. 평균제곱오차: 예측값과 실제값 사이의 오차 제곱의 평균(MSE)

    - 큰 오차에 대해 민감하게 반응 -> 이상치에 민감함

b. 평균절대오차: 예측값과 실제값 사이의 절대값 오차의 평균(MAE)

    - 모든 오차를 동일하게 취급 -> 이상치가 존재해도 성능 준수

### 3-2 변수 유의성 평가

a. T-검증
    - 독립변수의 회귀계수가 유의미한지 검정하는데 사용

    - 귀무가설: 독립변수가 종속변수에 영향을 주지 않음(회귀계수 = 0)

    - 대립가설: 독립변수가 종속변수에 영향을 줌 (회귀계수 != 0)


    - t-값 계산
$t = \frac{\hat{\beta} - 0}{SE(\hat{\beta})}$
$\hat{\beta}$ : 추정회귀계수, $SE(\hat{\beta})$ : 표준오차
    
    - p-value 확인 및 판단
        - p-value < 0.05 : 해당 변수 유의미
        - p-value >= 0.05 : 해당변수는 무의미
