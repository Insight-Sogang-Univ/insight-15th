## 1.분류란?

### 1-1 지도학습: 회귀 vs 분류

- 회귀: 연속적인 숫자값을 예측하는 것
    - 주어진 데이터를 기반으로 fit한 회귀선을 찾는 문제


- 분류: 입력된 데이터를 주어진 항목들로 나누는 것
    - 기존 데이터가 어떤 레이블에 속하는지 판별하는 것

### 1-2 이진분류와 다중분류

- 이진분류: 예측하고자 하는 변수 어떤 기준에 대하여 참 또는 거짓의 값만을 가질때
- 다중분류: 예측하고자 하는 변수가 가질 수 있는 값이 3개 이상일 때

## 2.분류 모델

### 2-1 로지스틱 회귀

- 이진분류 문제를 푸는 대표적인 알고리즘으로, 특정 클래스에 속할 확률 추정

- 출력값을 0에서 1사이로 변환해주는 것을 의미한다.

- [로지스틱 회귀의 필요성]


    - 값이 0아니면 1일 시에는, 위와 같이 연속적이지 않은 그래프로 나타나지지만, 여기서 다중 선형 회귀를 적용하면 제대로 분류가 안됨
    - 따라서 이를 제대로 분류하기 위해서는 S자 형태로 그려지는 함수가 존재해야함

- [시그모이드 함수]

    - 시그모이드 함수는 출력이 0과 1사이의 값을 가지면서 S자 형태로 그려지는 함수.
    - 출력값이 특정 값 이상이면 1, 이하면 0으로 정해서 이진 분류 문제를 풀때 사용한다.

- [시그모이드 함수의 가중치]
    - 선형 회귀 모델에 시그모이드 함수를 적용하여 비선형성을 준 모델.

    - but odds를 적용하면, 모델의 결과를 선형적으로 해석할 수 있음

    - 가중치 w가 커질수록, 경사가 가파라짐 / 가중치 b가 커질수록 그래프가 왼쪽으로 이동

### 2-2 의사결정 나무

- 조건에 따라 데이터를 분류하며, 최종적으로 데이터가 순수한 label의 집합으로 구성될때 까지 분류를 반복하는 것

a. 의사 결정나무의 구성 요소

    - root node, parent node, child node, binary tree, leaf nodes, edge, heigh, depth

b. CART(이진분할 알고리즘)

    - 데이터셋을 임계값을 기준으로 두 child로 나누는 알고리즘
    - 불순도(지니 계수)가 낮아지는 방향으로 나눔
        - 불순도란?
            - 분류하려는 데이터 집합에서 서로 다른 클래스가 섞인 정도
            - 통계적 분산 정도를 정량화해서 표현한 값

c. CART 알고리즘의 주요 단계
    - 임계값 설정: 입계값을 기준으로 데이터를 그룹으로 나눔
    - 불순도 감소 알고리즘: 클래스를 정확하게 구분해줄 수 있는 분류기준을 찾아 불순도가 낮은 쪽으로 가지 형성

d. 실제 학습시 고려해야 할 것들: 모수 설정, 차이점 시각화, Prunning
    - 하이퍼 파라미터 설정
    - 시각화: 분류가 잘 이루어졌는지 확인
    - pruunning(가지치기) : 불필요한 노드를 지우기(과적합 방지)


### 2-3 SVM(Support Vector Machine)

- SVM: 클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘 = "데이터 분리 초평면"중 가장 거리가 먼 초평면을 선택

a. SVM의 구성
    - Support vector: 데이터 중에서 결정 경계와 가장 가까이에 있는 데이터들의 집합
    - Decision Boundary: 데이터 분류의 기준이 되는 경계 -> 데이터로부터 가장 멀리 떨어져 있는게 좋음
    - Margin: 결정 경계에서 서포트 벡터까지의 거리
    - Hyperplane: n차원 공간의 n-1차원 평면
    - Slack Variables: 완벽한 분리가 불가능할때 분류를 위해 허용된 오차

b. SVD의 장단점
    - 장점: 데이터가 적을때 효과적, 새로운 데이터 입력에 대해서 연산량 최소화, 정확성 높음, 과적합 가능성 낮음, 노이즈 영향 적음
    - 단점: 데이터 전처리 과정이 중요, 데이터가 많아질수록 최적화를 위한 평가 과정이 많아져 속도가 느림.

### 2-4 KNN (K-Nearest Neighbor)

- KNN: 비슷한 특성을 가진 데이터끼리 서로 가까이 있는다는 점을 이용한 분류 알고리즘
    - 학습이 필요하지 않음

- 계산 순서
    1. 데이터 준비: 학습 과정이 없어, 벡터와 레이블로 이루어짐
    2. K 값 설정: 보통 홀수개로 설정 - 동점 방지하기 위해서
    3. 거리 계산: 새로운 데이터가 주어지면, 모든 데이터 간의 거리를 계산
    4. 가장 가까운 K개의 이웃 선택: 가장 작은 거리 값을 가진 K개의 데이터 선택
    5. 분류하기: K개의 이웃 중 가장 많이 등장하는 클래스가 예측 결과

- 최적의 K 선택하는 법
    - 데이터 개수의 제곱근으로 설정
    - K를 너무 크게 설정하면 과소 적합 발생
    - K를 너무 작게 설정하면 노이즈 존재 가능성

- KNN의 장단점
    - 장점: 범주 기준 없어도 분류 가능, 바로 예측값 구할 수 있음, 구현이 간단, 추가 데이터 처리 용이, 원본 데이터 전체를 그대로 저장
    - 테스트 데이터 개수에 따라 시간이 오래 걸림, K값 결정 어려움, 수치아니면 유사도 정의 어려움, 이상치에 영향을 크게 받음

- 앙상블이란?
    - 여러개의 개별 분류 모델들을 결합하는 머신러닝 기법
    - 보팅 = 다른 모델 병렬로 사용, 배깅 = 같은 모델 병렬로 사용, 부스팅 = 같은 모델 직렬로 사용

## 3.분류 평가 지표

### 3-1 혼동 행렬

- 혼동행렬: 분류 모델의 예측 결과를 정확한 에측과 잘못된 예측으로 구분하여 나타낸 표

| 실제 | 예측 | 의미 |
|------|------|------|
| TP | 정답 맞춤 | 실제 정답을 맞게 예측 |
| TN | 정답 맞춤 | 실제 오답을 맞게 예측 |
| FP | 틀림 | 실제는 틀린데 맞다고 예측 |
| FN | 틀림 | 실제는 맞는데 틀리다고 예측 |

- 정확도 = 맞춘 문제수 / 전체 문제수 ((TP+TN)/(TP+TN+FP+FN))

- 정밀도 = 모델이 맞다고 판단한것 중 실제로 맞은 비율 (TP / (TP+FP))

- 재현도 = 실제로 맞아야 하는 것 중 모델이 맞춘 비율 (TP / (TP+FN))

- 이때, 정밀도와 재현도의 그래프에서 두 value 값이 만나는 지점을 임계값으로 정할 때, 예측 오류최소화 가능

### 3-2 F1-Score

- 정밀도와 재현율의 조화 평균

- 정밀도와 재현율 간의 균형을 효과적으로 평가하기 위해서

### 3-3 ROC / AUC Curve

a. ROC Curve: 얼마나 분류가 잘 되었는가를 보여주는 그래프
    - TPR: 참인 것들 중에 참이라고 예측한 비율
    - FPR: 거짓인 것들 중에 참이라고 잘못 예측한 비율

b. AUC Curve: ROC와 x축 사이의 면적(적분값)
    - 모델의 성능을 숫자로 나타낼수 있음
    - 1에 가까울수록 분류 성능이 좋은 것

## 4.하이퍼파라미터 최적화

### 4-1 하이퍼파라미터 최적화

- 하이퍼파라미터: 학습 시작 전에 사용자가 직접 설정하는 변수
- 하이퍼파라미터 최적화: 적절한 하이퍼파라미터를 찾아 모델 성능을 향상 시키는 것

### 4-2 하이퍼파라미터 최적화 방법

- Grid Search:
    - 정해진 범위에서 hyperparameter를 모두 순회하며 가장 좋은 성능을 내는 값 찾기
    - 장점: 최적해를 정확히 찾을 수 있다.
    - 단점: 시간이 너무 오래 걸림
    - 적용: 넓은 번위, 큰 step을 활용해 범위를 좁힘

- Random Search:
    - 정해진 범위에서 Hyperparameter를 무작위로 탐색
    - 장점: 속도가 조금 더 빠름
    - 단점: 무작위이기 때문에 정확도가 떨어짐

- Bayesian Optimization
    - 사전 정ㅂ를 바탕으로 최적 htperparameter 값을 확률적으로 추정하여 탐색
    - 방법
        - 초기값 몇개를 무작위로 실험
        - 그 결과를 바탕으로 어떤 파라미터 조합이 좋을지 예측
        - 다음 실험할 조합을 가장 좋을 것 같은 위치로 선택
        - 반복
