# 연관분석

## 1.연관 분석이란?

### 1-1 연관 분석의 정의

- 연관 분석: 상품이나 서비스를 구매하는 등 일련의 거래나 사건 안에 존재하는 항목 간의 일정한 연관 규칙을 발견하는 분석 - 비지도 학습
    - ex) OTT 서비스, 금융기관 등 

    - 연관분석 알고리즘: Apriori, FP-Growth



### 1-2 연관 규칙 분석

- 연관규칙의 형태 - A를 구매할때 B 또한 구매 할 것이다. -> if-then : A -> B로 표시
    - 조건절: 만일 ~ 라면
    - 결과절: 그 뒷부분에 해당
    - 아이템 집합:조건절 또는 결과절을 구성하는 아이템 집합 -> 여러개의 아이템들은 상호 배반 이어야함


- 연관규칙 = 인과관계는 알 수 없음
    - 상관관계를 파악할 뿐, 인과 관계는 알 수 없음.


--- 

## 2.주요 개념

### 2-1 연관규칙의 조건

- 두 품목이 함께 구매한 경우의 수가 일정 수준 이상이어야 함

- 품목 A를 포함하는 거래 중 품목 B를 구입하는 경우의 수가 일정 수준 이상이어야 함

    - 지지도:전체 거래 항목 중 A,B가 동시에 포함하는 거래의 비율
        - 지지도가 낮을 시, 우연 이거나 실행 가능한 인사이트 도출의 어려움 존재
    
    - 신뢰도: 품목 A를 포함하는 거래 수 A, B 가 동시에 포함하는 거래의 비율
        - 의미 없는 연관 규칙, 추천실패로 이루어짐


### 2-2 연관 분석의 평가 측도

- 향상도: 품목 A가 주어지지 않았을 때의 품목 B의 확률 대비, 품목 A가 주어졌을때의 품목 B의 확률의 증가 비율

    - A와 B가 우연히 함께 등장할 확률과 실제 함께 등장할 확률을 비교하는 것
        - 독립일때: A를 샀다는 사실이 B를 살 확률에 영향을 주지 않으므로 향상도는 
        - 독립이 아닐때: A를 샀다는 사실 B에 영향을 주거나, B가 A에 영향을 주는 것
        -  $$P(A∩B) > P(A)\cdot P(B)$$ = 양의 연관성
        -  $$P(A∩B) > P(A)\cdot P(B)$$ = 음의 연관성
        - 비율(향상도): 1초과 = 우연적 기회보다 높은 확률, 1 = 독립, 1미만: 우연적 기회보다 낮은 확률

## 3.활용 사례

### 3-1 넷플릭스 Case

- 데이비드 핀쳐 감독의 영화는 시청 완료율이 높다 / 영국판 <하우스오브카드>가 인기 있었다 / 영국판 <하우스 오브 카드>를 본 사람들은 케빈 스페이시가 출연한 영화나 데이비드 핀쳐 감독의 영화를 보았다.

    - 지지도: 영국판과 케빈 스페이시 콘첸츠를 동시에 시청한 사람의 비중이 높다. - 사용자 기반 취향 파악
    - 신뢰도: A가 데이비드 핀처 감독, B가 시청완료 율일시 B|A의 확률이 높았다 - 핀처감독은 몰입도가 높아 리스크가 낮다.
    - 향상도: A가 시리즈 시청, B가 케빈 스페이시 콘텐츠 시청일시, A,B의 향상도는 1 초과 - 두 요소가 결합되면 반응이 좋을 것이다.

## 4.연관 분석 알고리즘

### 4-1 Apriori 알고리즘

- 무차별 탐색의 비효율 때문에 생겨난 알고리즘 이다.

- Apriori 알고리즘 소개: 상위 조합에서부터 차례로 스캔하면서 특정 조합이 자주 않으면, 후속 조합들까지 후보에서 배제
    - 장점: 하나의 조합만 검사하고도 파생된 나머지 조합 drop이 가능 - 연산량 감소

- Apriori알고리즘 프로세스: 빈발 항목 집합을 찾고, 단계적으로 조합을 생성하고 필터링
    - 단일 항목 집합 생성: 모든 상품에 대해 지지도 게산 후 최소 지지도 넘는 항목만 다음 단계로
    - 2개 항목 집합: 1단계의 상품들중 2개 항목의 조합을 만듬 - 이에 대해 지지도 계산후 최소 지지도 안넘는 조합 제거
    - 더 큰 항목 집합 생성: 이전 단계에서 남은 조합 바탕으로 3개 이상의 항목 조합 만듬 - 제거 - 조합생성 반복
    - 최소 빈발 항목 선정

- Apriori 알고리즘의 장단점
    - 장점: 수많은 상품 연관 구매 패턴 발견 - 이해분석이 용이
    - 단점: 데이터가 커질수록 속도가 느려짐

### 4-2 FP Growth 알고리즘

- 여전히 느리긴함

- FP-Growth 알고리즘 소개
    - FP-Growth는 FP-Tree구조를 이용하여 Apriori를 구현함 = Tree, Array, Linked_List 합친 구조
    - 모든 거래에서 빈도 높은 아이템 순서로 정렬 후, 부모노드 중심으로 자식 노드 추가해주며 tree생성

- FP-Growth 알고리즘 프로세스
    - 데이터셋 확인
    - 빈도순 재정렬: 남은 항목들을 빈도순으로 재정렬 
    - FP-트리 생성: 동일한 항목은 기존 노드에 연결, 새로운 항목은 새로운 노드로 추가
    - 빈발 항목 집합 추출: FP-트리를 바탕으로 특정 항목을 기준으로 부분 트리를 만들어 빈발 항목 집합 추출
    - 연관 규칙 도축: 빈발 항목 집합을 바탕으로 연관 규칙 생성 - 지지도, 신뢰도, 향상도의 지표를 계산해 가장 유용한 규칙 선택

- FP-Growth 알고리즘 장단점
    - 장점:
        - Apriori보다 훨씬 빠르다
        - 후보 Itemset 생성할 필요 없이, Tree만 구성하면 됨.
    - 단점:
        - 메모리를 효율적으로 사용하지 않음
        - 설계하기 어렵고, 무조건 FP-Tree가 무조건 만들어져야 가능

# 경영기초

## 1.문제 정의와 구조화

### 1-1 문제 정의

- 현상 vs 문제:
    - 현상: 관찰된 사실
    - 문제: 해결해야할 질문

- 문제란 단순히 나쁜 상황을 나열하는 것이 아닌, 의사결정을 내려야하는 구체적인 질문임

#### SMART 원칙
    - Specific: 구체적이고 범위가 명확함 - 무엇을 어디까지 풀지 정의
    - Measurable: 수치로 측정 가능 - 성공 여부를 숫자로 판단 가능
    - Action-oriented: 실행 지향적/결과 중심적 - 행동을 이끌어내고 성과에 집중
    - Relevant: 관련성 - 조직의 목표와 관련
    - Time-bound: 시한성 - 마감 기한과 의사 결정 시점이 존재



### 1-2 문제 구조화

- 문제 뒤에는 Sub-issues가 존재

- MECE:
    - Mutually Exclusive: 겹치는 부분이 없음
    - Collectively Exhaustive: 누락된 부분 부재

- MECE를 위한 규정
    - 논리적 일관성: 한 층위에서는 하나의 기준으로만 나눠야 함
    - 통찰력: 문제를 해결하는 데 도움이 되는 기준으로 나눠야 함
    - 상호배타성: 조각들이 서로 겹치지 않아야 함

## 2.지표

### 2-1 지표

- 수치적으로 정보들을 설명하기 시작한 것을 지표라고 부름
- 방향이나 목적, 기준 따위를 나타내는 표지.
- 보통 지표는 분모 대비 분자로 표현됨.
    - 효율성: 산출/투입 - ex) 노동생산성 = 생산량 / 노동시간
    - 효과성: 달성 결과 / 목표 - ex) 목표 달성률 = 실제매출/목표 매출
    - 수익성: 이익 / 자원 - ex) ROE = 순이익 / 자기자본
    - 성장성: 증가분 / 기존 규모 - ex) 매출 성장률 = (당기매출 - 전기매출)/전기매출
    - 안정성: 보유자산 / 부채 - ex) 부채비율 = 부채/자기자본


### 2-2 지표 정의

- 절대적인 수치가 들어갈 수 없을 시, Proxy Metric 사용.
    - 추상적이여서, 가정과 정의, 논의가 필요한 지표


### 2-3 북극성 지표 L1/L2 Metrics

- 북극성 지표: 기업이 가장 중요하게 생각하는 지표
- L1/L2 지표:
    - L1: 전사적인, 고수준의 성과 지표
    - L2: L1 지표의 세부 사항을 측정하는 지표


### 2-4 AARRR

- 사용자 운영 과정에서 자주 활용하는 데이터 분석 모델로, 사용자 증가를 위한 5가지 로드맵을 구성

1. Acquisition: 주로 신규 고객을 어떻게 획득할지

2. Activation: 사용자 활성화 - ex) 회원가입 10일 내 최소 7명의 친구를 가진 신규 측정자
    - Sign-up: 유저가 제품에 들어온 시점
    - Setup Moment: 유저가 제품의 핵심가치를 경험할 준비 완료
    - A-HA! Moment: 유저가 핵심가치를 처음 경험한 순간
    - Habit Moment: 유저가 제품 사용에 대한 습관 형성

3. Retention: 신규 사용자를 장기 사용자로 전환 시키는 것에 초점 - 익일 유지율, 1주차 유지율 등

4. Revenue: 사용자의 소비를 유도해 매출을 높이는데 유도하는 것 - 고객의 평상가치(LTV)를 지속적으로 높여나가는것 + 고객 유치 비용 줄이는 것

5. Referral: 어떻게 충성도 높은 사용자를 보이지 않는 브랜드 홍보자로 만들 것인지 - 바이럴 지수, 순 추천 고객지표

## 3.지표 분석 방법

### 3-1 코호트 분석

- 코호트: 동질 집단
    - 정의된 시간 범위 내에서 공통된 특성이나 경험을 공유하는 사용자 집단. 
    - 분석 전에 데이터 세트의 데이터를 관련 그룹으로 나누는 행동 분석
    - 시간에 기반하는 세그먼트 분류

- 코호트 분석 해석 방법:
    - 세로축: 코호트 별 비교
    - 대각선: 시점별 비교
    - 가로축: 시간의 경과에 따른 변화 관찰

### 3-2 퍼널 분석

- 유저가 접속해서 떠날 때까지 유저의 행동 추적 (깔때기로 불림)
- ex) 유입률 높이기, 전환율 높이기, 퍼널 단계 없애기, 순서 바꾸기 등
