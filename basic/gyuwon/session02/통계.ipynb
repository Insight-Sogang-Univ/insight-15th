> ##### ***자료의 척도***

**자료**: 설명과 해석을 위하여 수집되고 분석되며, 요약되는 사실과 수치

* 양적 자료 : 수치로 표현되는 자료
    
    - **구간(등간)척도** : 값들 사이 순서 O, 그 간격 일정해 덧셈과 평균 연산 가능(곱셈/나눗셈x) 0의 의미 X

    - **비율척도** : 절대영점과 임의의 측정단위로 구성하고, 동간성이 유지되는 척도. 값이 0이면 그 변수에 대해 아무것도 존재하지 않는 상태

* 질적 자료 : 연산의 의미를 갖지 못하는 자료

    - **명목척도** : 레이블이나 이름으로 요소의 속성을 식별하려고 하는 척도

    - **서열척도** : 명목형의 분류 속성과 더불어, 항목간의 순서 또는 순위를 나타내는 특성을 지니는 척도

#### **2. 기술통계**

> ***기술통계란?***

데이터를 있는 그대로 설명하고 이해하는 데 중점을 둔 통계 방법

> ***기술통계의 3 요소***

**(1) 중심화 경향** : 데이터의 중심적인 경향을 나타내는 기술 통계량 

* 평균(mean)

    - 데이터 값의 초합을 데이터 개수로 나눈 값

    - 총합을 사용하므로 이상치에 민감할 수 있음 -> 절사 평균

    - series.mean()

* 중앙값(median)

    - 데이터를 크기 순서대로 일렬로 정렬했을 때, 한가운데에 위치하는 값

    - 평균에 비해 이상치에 덜 민감함

    - series.median()

* 최빈값(mode)

    - 데이터에서 가장 자주 나타나는 값

    - 주로 이산형 데이터, 데이터의 분포를 이해하는 데 사용

    - series.mode()

---

**(2) 자료의 산포** : 데이터가 퍼져있는 정도를 설명하는 기술 통계량

* 범위(range)

    - 데이터의 최대값과 최소값의 차이를 나타내는 지표로 데이터의 전체 변동 범위를 나타냄

    - 양쪽 극단값을 이용하기에 이상치에 매우 민감

    - series.max() - series.min()

* 분산(variance)

    - 데이터 값들이 평균으로부터 얼마나 멀리 퍼져있는지를 나타내는 지표

    - 분산값이 크다 = 데이터가 많이 퍼져있다

    - $\frac{\sum_{i=1}^{N}(x_i - μ)^2}{N}$

    - series.var()

* 표준편차(standard deviation)

    - 분산에 제곱근을 취한 값 : 분산의 단위는 관측값 단위의 제곱이기에 대상 데이터와 일치하지 않아 제곱근을 씌워 단위 일치 시켜주는 것.

    - $\sqrt{\frac{\sum_{i=1}^{N}(x_i - μ)^2}{N}}$

    - series.std()

* 사분위수(quartile)

    - 데이터의 표본을 네 개의 동일한 크기로 나눈 값(하위 25, 50, 75, 100 퍼센트)

    - 
        series.quantile(0.25)
        series.quantile(0.5)   
        series.quantile(0.75)  
        np.percentile(series, 75)
    

* 사분위간 범위(IQR)

    - Q3에서 Q1 값을 제외한 범위, 즉 데이터의 중앙 50%가 퍼져 있는 범위 -> 이상치를 탐지할 때 유용(Tukey Fence)

    - 아래쪽/위쪽 이상치 : Q1 –1.5 x IQR 보다 작은 값/Q3 + 1.5 x IQR 보다 큰 값

* 변동계수(coefficient of variation)

    - 데이터의 절대적인 변동성이 아닌 상대적인 변동성을 나타내는 지표
    
    - 변동계수 = (표준편차/평균) * 100

**(3) 자료의 정리 및 시각화**

* 히스토그램
* 박스플롯
* 파이차트

---

    cf) 첨도와 왜도

* 첨도 : 확률분포의 꼬리가 두꺼운 정도를 나타내는 척도

    - series.kurt()


* 왜도 : 비대칭도, 실수 값 확률 변수의 확률 분포 비대칭성을 나타내는 지표

    - 왜도의 값은 양수나 음수가 될 수 있으며 정의되지 않을 수 있음

    - 왜도가 양수일 때는 확률밀도함수의 오른쪽 부분에 긴 꼬리를 가지며 자료가 왼쪽에 더 많이 분포해 있다는 의미

    - 반대로 왜도가 음수일 경우에는 확률밀도함수의 왼쪽 부분에 긴 꼬리를 가지며 중앙값을 포함한 자료가 오른쪽에 더 많이 분포해 있음

    - 평균과 중앙값이 같으면 왜도는 0
    
    - series.skew()

#### **3. 확률과 분포**

> ***추론통계학의 논리***

* 표본의 불확실성 : 운 없이 이상한 표본만 뽑힌 것 아닌가?
    -> **표본분포(Sampling Distribution)**

* 모수 추정 : 그 표본의 통계량이 정말로 모수를 잘 반영하는가?
    -> **점추성(point Estimation)**

* 추정의 현실성: 점추정 값이 모수와 정확히 일치하는 게 말이되나?
    -> **구간추정(Interval Estimation)**

* 구간 추정의 오류 관리 : 이 구간이 모수를 포함하지 못할 가능성 관리
    -> **신뢰수준(Confidence Level)**

* 적극적 주장 검증 : 특정 주장의 옳고 그름을 직접 판단할 수 없나?
    -> **가설검정(Hypothesis Testing)**

* 가설검정의 오류 관리 : 가설이 틀렸다 했는데 내 결론이 틀린거면?
    -> **유의수준(Significance Level, 제 1종 오류)**


> ***확률분포***

* 확률변수 : 무작위한 세계에서 어떤 특정 사건이 발생할 때 나타날 수 있는 모든 경우의 수를 특정 값으로 대응시키는 함수

* 확률분포(Probability distribution) : 확률변수가 특정한 값을 가질 확률을 나타내는 함수 의미
    
    - 이산확률분포 : 확률변수가 이산적인(discrete) 값을 가지는 경우
    
    - 연속확률분포 : 확률변수가 연속적인(continuous) 값을 가지는 경우


> ***대표적인 분포 모형***

* **정규분포(Normal Distribution)**

    - 평균 기준으로 대칭인 종 모양이며 표준편차 클수록 곡선 평평해짐(데이터가 넓게 퍼짐)

    - 가장 잘 알려진 연속확률분포로 많은 현상에서 발견됨

    - 모집단이 정규분포를 따를 때 표본평균도 정규분포를 따름, BUT 중심극한정리에 따르면 표본의 크기가 커질수록 표본평균의 분포는 모집단의 분포와 상관없이 정규분포에 가까워짐 -> 큰 수의 법칙

* **Z 분포(표준정규분포)**
    - 평균이 0이고 표준편차가 1이 되도록 맞춘 표준적인 정규분포
            


   > 📌**추출한 표본이 ‘정규분포를 따르는 모집단’으로부터 왔다고 가정하면 아래의 표본분포들을 통해 통계적 추론 가능**

* **T 분포(T-Distribution)**

    - 정규분포와 같이 평균을 기준으로 대칭인 종 모양이지만 꼬리가 두꺼움
            
    - 자유도가 커짐에 따라 표준정규분포 N(0,1) 에 수렴
    
    - 표본으로 모평균 혹은 모평균 간의 차이를 추정할 때 사용됨
            
    - 표본 크기가 작을 때 정규분포보다 더 정확한 결과 제공

> 📌 **자유도** : 얼마나 많은 정보가 독립적으로 변할 수 있는가를 나타내는 숫자

* **카이제곱분포(Chi-Square Distribution)**

    - 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포

    - 자유도가 증가함에 따라 중심이 오른쪽으로 이동하고 모양이 더 대칭적으로 변화하며, 정규분포에 가까워짐
    
    - 독립적인 표준 정규분포를 ᄄᆞ르는 변수들의 제곱 합 -> 표본분산의 분포와 관련된 지표
    
    - 독립성 검정, 적합도 검정 등에 사용됨

* **F 분포(F-Distribution)**

    - 카이제곱분포와 형태는 비슷. 
    
    - 두 독립적인 카이 제곱 분포의 비율로 두 집단 간 분산 차이를 비교할 때 사용됨

#### **4. 추론통계**

> 수집한 데이터를 기반으로 미래의, 혹은 조사하지 못한 전체 정보를 추론하고 결론을 도출하는 방법을 다루는 분야 (ex. 통계적 가설검정, 신뢰구간)

##### (1) **추정**

> 모집단을 대표할 수 있는 일부를 표본으로 추ᅟᅮᆯ하여 분석한 통계량을 이용해 모집단	의 모수를 예측하는 과정

* **점추정** : 어떤 모수를 표본에 기초하여 계산된 하나의 값으로 추정 -> 오차에 대한 정보가 전혀 고려되지 않음

* **구간추정** : 추정하고자 하는 모수가 존재하리라 예상하는 구간을 정하여 추정



    - 모수 : 모집단의 특성에 관계되는 평균, 비율 또는 표준편차 등 관심의 대산이 되는 미지의 참값


* 추정량 : 표본정보에 의존하는 확률변수로서 모수를 추정하는 데 사용되는 표본 통계량

* 추정치 : 표본이 추출된 후 표본자료를 이용하여 계산한 통계량의 특정한 값

##### **(2) 검정**

> 표본을 추출하여 얻은 표본통계량으로 모집단 모수에 대한 새로운 예상/가설 등이 옳다고 판단할 수 있는지를 평가하는 과정

##### 📌 **가설검정** : 모집단의 실제 값이 얼마가 된다는 주장과 관련해, 표본의 정보를 사용해서 가설의 합당성 여부를 판정하는 과정을 의미

> **<1>** 가설 설정
* 귀무가설 H0(기각하고자 하는 명제) 과 대립가설 H1(내가 주장하고자 하는 명제)을 설정

> **<2>** 유의 수준 설정
* 연구자가 결과를 판단할 때 허용할 수 있는 오류의 최대치(보통 5% -> 귀무가설이 실제로 참인데도 불구하고, 5%의 확률로 귀무가설을 잘못 기각할 수 있다는 의미)

    - 1종 오류(알파) : 귀무가설이 실제로 참이었는데 기각할 확률

    - 2종 오류(베타) : 귀무가설이 실제로 거짓인데 기각하지 못할 확률
        
    - 1종 오류가 더 치명적 -> 유의수준 = 1종 오류의 확률
    
 > **<3>** 검정통계량 산출
* 가설 검정의 결과를 판단하는 데 사용되는 값
    
* Z값 
    - 표준정규분포를 사용하여 얻은 값으로 특정 데이터가 평균으로부터 얼마나 떨어져 있는지를 표준편차 단위로 나타냄
                        
    - 큰 표본 크기(30 이상)에 대한 가설 검정
                        
    - Z 값이 크면 해당 데이터가 평균으로부터 멀리 떨어져 있으며 때로는 비정상적이거나 이상치일 가능성이 있음 의	미

* T 값

    - T 분포를 사용하여 얻은 값으로, 작은 표본 크기에 대한 				가설 검정(T검정)에 주로 사용됨

    - 정규분포와 유사하지만 꼬리 부분이 더 두꺼워 작은 표				본에서 발생할 수 있는 불확실성을 보다 잘 반영

    - 모집단의 평균에 대한 가설 검정/두 표본 평균 간 차이 				검정

    - T 값의 절대값이 크면 모집단 평균으로부터 멀리 떨어				져 있어 관측된 표본 평균이 모집단 평균과 유의미한 차이				가 있을 가능성이 높다는 의미

* F 값
    
    - F 분포(두 개의 독립적인 카이제곱 분포된 변수들의 비				율)를 사용하여 얻은 값으로 두 표본 분산 사이 차					이 비교할 때 사용
    
    - 분산 분석(ANOVA)에서 두 개 이상의 그룹 간 평균의 				차이를 검정
    
    - F 값은 두 표본의 분산의 비이기에 값이 1에 가까우면 				두 표본의 분산이 서로 비슷하다는 의미



> **<4>** 가설 기각/채택 판단

* 유의확률(p-value) : 유의확률(p)가 유의수준(알파)보다 작을 시에 귀무가설을 기각, 클 시에는 귀무 가설을 기각하지 못함

##### **데이터의 자료형에 따른 검정 방법**

* **[수치형-수치형] (피어슨)상관분석 : 두 변수 간 피어슨 상관계수가 얼마나 유의미한지를 검정하는 방법**

    - 귀무가설 : 두 변수간 선형 상관관계가 존재하지 않는다
    
    - 대립가설 : 두 변수간 선형 상관관계가 존재한다.
    
    - 코드 : 
    ```python
    import scipy.stats as spst
    spst.pearsonr(df['col1'], df['col2'])
    ```

    - 공분산(covariance) : 두 변수가 함께 어떤 방향으로 변화하는지			와 그 크기를 표현하는데 사용됨
    
    ```python 
    np.cov(df['col1'], df['col2'])
    ```

    - 상관계수(correlation) : 두 변수 간의 선형적인 관계를 나타내는 						지표로 –1 ~ 1 의 값을 가짐 인과관계는 한 변수가 다른 변수를 발생시키는 관계로 두 변수가 수치적으로 연	관되어있는 관계인 상관관계와 다름

    ```python
    df[['col1', 'col2']].corr()
    ````

* **[범주형 – 수치형] F-test(ANOVA), T-test**
    - **T-test**
        
        - 범주가 2개일 때 두 평균 간 차이를 표준 오차로 나눈 t 값
        
        - 귀무가설 : 두 그룹의 평균에는 차이가 없다
        
        - 대립가설 : 두 그룹의 평균에는 차이가 있다

    - **F-test(ANOVA)**

        - 범주가 3개 이상일 때 (전체 평균과 각 집단의 평균을 가지고 만			든 분산)/(각 집단의 평균과 개별 값으로 만든 분산)으로 계산

        - 귀무가설 : 각 그룹 별 ~~평균은 차이가 없다

        - 대립가설 : 각 그룹 별 ~~평균은 차이가 있다
		
        - 주의 : anova 는 전체 평균 대비 각 그룹간 차이 유무만을 알 			수 있기에 하나가 튀고 나머지가 비슷하더라도 귀무가설을 기각하			는 결과가 나올 수 있다.

* **[범주형 – 범주형] Chi2-test**

    - 귀무가설 : ~여부와 ~ 여부는 관련이 없다

    - 대립가설 : ~여부와 ~ 여부는 관련이 있다

    - 기대빈도와 관측빈도 간 차이가 크다면 둘이 독립적이지 않을 가			능성이 크기에 밑의 수식으로 카이제곱 통계량 계산 가능
    
    - 카이제곱 통계량은 범주가 많아질수록 커지기에 자유도의 2배 가			량을 기준으로 2배 이상이면 차이가 존재하는 것으로 판단함
    
    - 카이제곱 변수에서의 자유도는 (x변수의 자유도) * (y변수의 자유			도)로 계산
