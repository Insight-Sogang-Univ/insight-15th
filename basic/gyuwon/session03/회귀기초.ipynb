### **1. 머신러닝과 모델링**

> #### **머신러닝**

=> 인공지능의 한 분야로, 모델이라는 소프트웨어를 학습하여 새로운 데이터를 예측하거나 결정을 내릴 수 있도록 하는 기술

##### <**머신러닝의 종류**>

* **지도 학습(Supervised Learning)** : 문제와 정답을 모두 알려주고 공부시키는 방법(회귀, 분류)

* **비지도 학습(Unsupervised Learning)** : 답을 가르쳐주지 않고 공부시키는 방법(군집화)

* **강화 학습(Reinforcement Learning)** : 보상을 통해 상은 최대화, 벌은 최소화하는 방향으로 행위를 강화하는 학습(알파고, 자율주행)

---

> #### **모델링(Modeling)**

##### **(1) 모델이란?**

=> 학습의 결과물(학습을 마친 인공지능 그 자체)

##### **(2) 모델링이란?**

=> 예측기(공식)를 만들어가는 전체적인 과정, 아무것도 모르는 기계가 주어진 데이터 속에서 스스로 패턴을 찾아내 오차가 작은 최적의 공식을 완성할 때까지 학습하는 과정

##### **(3) 모델링의 과정**

* **설계** : 데이터의 특성에 맞추어 어떤 알고리즘으로 학습할지 결정하는 단계

* **훈련** : 데이터를 기계에게 계속 보여주며 정답을 맞히도록 수정하는 단계

* **검증** : 훈련시킨 모델이 새로운 데이터에서도 정답을 잘 맞히는지 테스트하는 단계

---

> #### **지도 학습**

##### **(1) 지도 학습이란?**

=> 정답 레이블을 함께 주고 학습 시키는 방법. 예측값(prediction)과 정답(label) 이 최대한 같아지는 방향으로 학습함

##### **(2) 지도 학습의 과정**

📌 ***지도 학습을 위해선 반드시 데이터가 labeled data 형태여야함***

* **labeled data** : 독립변수와 종속변수가 모두 존재하는 데이터, 즉 입력 데이터와 그에 대응하는 출력값이 있는 데이터

    * 독립변수 : 머신러닝에서 예측에 사용되는 입력 변수

    * 종속변수 : 머신러닝에서 예측하고자 하는 대상 변수

* **labeld data 를 training set 와 test set 로 나눈다** 

    * train set : 모델을 학습(훈련)시킬 때 쓰이는 데이터

    * test set : 학습된 모델을 검증하기 위해 사용하는 데이터 -> **unseen data**

    * 왜? : train data 로 학습시킨 모델이 새로운 데이터, 즉 unseen data 에 대해 얼마나 에측을 잘 수행하는지 공평하게 평가하기 위해서!

* **training set 를 통해 모델을 학습**

    * 독립변수를 통해 추정한 예측값이 주어진 종속변수와 같아지도록 모델을 훈련시킴

    * 훈련이 수만 번 반복되면 예측값과 실제 정답 사이의 차이가 줄어들어 정답 확률이 높아짐


* 학습 후 컴퓨터는 데이터 사이 통계적인 상관관계들을 하나의 함수 형태로 가지게 됨 
-> 학습 과정에서 본 적 없는 새로운 데이터가 입력으로 들어와도, 그동안 학습했던 규칙에 따라 가장 확률이 높은 정답을 출력하게 됨

### **2. 회귀(Regression)**

##### **회귀 : 주어진 데이터(X)를 기반으로 정답(Y)를 잘 맞히는 함수를 찾는 문제**

* 종속변수가 연속적일 때 사용

    - 범주형, 즉 불연속적이면 분류를 사용

    ###### Cf) **분류 : 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별하는 것**

> #### **선형 회귀**

##### **(1) 선형 회귀란?**

* 선형 회귀는 회귀선(ex. 직선 y = wx+b) 을 찾는 과정

* 회귀선은 현재 데이터를 요약하는 동시에 아직 관측되지 않은 미래 값을 예측하는 도구

* 단순선형회귀(x가 한 개)와 다중선형회귀(x가 여러개)로 나눌 수 있음

##### **(2) 가중치와 편향의 이해**

📌 ***y = wx + b 기준***

* **가중치 w(weight)**

    - 독립변수 x 가 종속변수 y 에 주는 영향력의 크기 = 기울기 = 회귀계수

* **편향 b(bias)**

    - 독립변수 x 가 0일 때 나오는 기본값 = y 절편

    - 모델이 데이터에 더 유연하게 최적화될 수 있도록 위아래로 움직여주는 역할

##### **(3) 단순선형회귀**

* 하나의 독립변수(x) 만 존재하는 선형 회귀

* 데이터를 직선형태(하나의 독립 변수)로 표현하는 경우

* 데이터를 가장 잘 설명하는 y = wx + b 를 찾는 과정에 포함

* 잔차를 줄여나가며 직선을 결정짓는 적절한 회귀계수와 편향을 찾아감

    - **잔차** : 실제값과 예측값 간 차이

---

> #### **최적의 w 와 b를 찾는 두 가지 방법**

##### **(1) 최소제곱법(Least Squares Method)**

=> 최적의 회귀선을 찾는 방식 중 하나 = **최소자승법(Ordinary Least Square)**

* 잔차의 제곱의 합(RSS)이 최소가 되는 지점으로 최적의 회귀선을 구하는 방식

    - 잔차를 제곱하는 이유 : 

        - 양수-음수의 상쇄 방지 : 제곱하지 않으면 더하는 과정에서 상쇄됨<br>
        - 큰 잔차에 더 큰 벌을 줘서 예측값을 강하게 조정할 수 있음

        - 경시하강법 등의 알고리즘에서 미분 가능한 형태가 되기 위해

##### **(2) 경사하강법(Gradient Descent)**

=> 최소제곱법을 통해 구한 함수가 복잡하여 최솟값을 구하기 어려울 떄 사용하는 방법

| **구분** | **정의** | **방향** |
| --- | --- | --- |
| 목적함수 | 최적화하려는 전체 대상 | 최대화/최소화 |
| 손실함수 | **데이터 한 개**에 대한 오차 | 최소화 |
| 비용함수 | **전체 데이터**에 대한 평균 오차 | 최소화 |

***<손실함수와 경사하강법>***

$$E = \frac{1}{n} \sum_{i=1}^{n} [t_i - y_i]^2$$

- $t_i$ : $i$번째 데이터포인트 실제값(정답)

- $y_i$ : $i$번째 예측값

- $E$: Error
- $n$: 데이터 샘플의 총 개수

    이 식에 y=Wx +b 를 대압해 최종적으로 아래와 같은 손실함수를 얻을 수 있다

$$E(W, b) = \frac{1}{n} \sum_{i=1}^{n} [t_i - y_i]^2 = \frac{1}{n} \sum_{i=1}^{n} [t_i - (Wx_i + b)]^2$$

목표 : E를 최소화하는 W, b를 찾는 것

* 1. W로 x 축을 구성



* 2. 임의의 W 값에서 E 값을 살펴본다면 지금 경우보다 W 값이 왼쪽으로, 즉 E 값이 그래프에서 최소 지점으로 이동하도록 해야함



* 3. 미분 계수가 양수일 때, W 는 왼쪽(감소)으로 가야하는 핵심 아이디어를 담아 W 값을 반복적으로 업데이트



* 4. W 값이 최솟값에 도달하면 미분값이 0이 되어 "현재 값-0" 으로 더 이상 변화하지 않고 멈추게 됨



    손실 함수 그래프의 최솟값은 검정 점, 파란 점에서의 미분값은 그래프의 접선의 기울기

* 만약 현재 값이 최솟값의 왼쪽에 있다면 기울기(미분 값)가 음수이기 때문에 "현재 값 - 미분 값(음수)" 를 하면 현재 값이 커지게 됨 -> 오른쪽으로 점 이동

* 만약 현재 값이 최솟값의 오른쪽에 있다면 기울기(미분 값)가 양수이기 때문에 "현재 값 - 미분 값(양수)"를 하면 현재 값이 작아지게 됨 -> 왼쪽으로 점 이동

* 이 과정을 반복하다가 최솟값에 도달하면 미분값이 0이 되기에 "현재 값 - 0"으로 더이상 변화하지 않고 멈추게 됨

> ***<학습률(α: learning rate)>***

단순히 미분값만 뺴면 오히려 최솟값에서 멀어지는 경우 발생 가능

그럼 얼만큼을 빼야하는가? -> ***Learning Rate***

* Learning Rate 가 너무 작다면, 모델이 수렴할 때 까지 너무 오랜 시간이 걸리게 됨

* Learning Rate 가 너무 크다면, 모델은 수렴하지 못하고 발산(= 손실 값이 계속해서 커지는 것)하게 됨.

> ***<Local Minima 문제>***

실제 이동량은 작은 기울기에서 조금씩, 큰 기울기에서 많이 이동 -> 작은 기울기에서는 업데이트가 조금씩 이루어지기에 local minima 문제 발생

    📌 특히 딥러닝의 경우 비용함수가 정확히 어떤 모양인지 예측할 수 없어 local minima 에 빠지는 경우 발생

> ***<해결책 - 모멘텀>***

기존 경사하강법은 이전 기울기를 다 잊어버림, 즉 현재 기울기만 기준으로 움직임

    => 이전의 기울기와 방향을 고려하도록 관성을 부여해주자!

* **작동방식**

    - 빨간색 화살표 : 경사하강법이 이동하려고 하는 방향

    - 모멘텀 화살표 : 이전 단계의 이동을 반영한 모멘텀 방향(관성을 부여한 화살표)

    - 파란색 화살표 : 실제 이동 방향

* **모멘텀의 장점**

    - 기울기에 관성을 부여하여 작은 기울기는 쉽게 넘어갈 수 있도록 만든 것

    - 지역 최소값을 탈출할 수 있게 함

    - 모멘텀을 사용하지 않으면 아주 작은 언덕도 빠져나오지 못할 수 있음 + 기울기가 매우 작은 구간을 빠져나오는데 아주 오랜 시간 걸림

> #### **다중선형회귀**

$$
y = W_1 x_1 + W_2 x_2 + ... + W_n x_n + b 
$$

* **다중선형회귀** : 독립 변수가 두 개 이상인 경우

* 독립 변수 $x_i$ 가 2개인 경우 위와 같이 3차원 공간에 표현되고, 회귀선은 평면이 됨

* 단순선형회귀와 마찬가지로 최소제곱법/경사하강법을 사용할 수 있음, 잔차 = 관측치와 평면간의 차

##### **(1) 다중공선성(Multicollinearity)**

**다중공선성** : 독립 변수들 간의 강한 상관관계로 인해 회귀 모델이 불안정해지는 문제

##### **(2) 다중공선성 확인 방법**

> ***<상관계수, $r$>***

변수 간의 통계적 관계인 상관관계의 정도를 수치로 나타낸 것

* 상관관계

    - $1 ≤ r ≤ 1$  사이의 값을 가짐

    - $r>0$: 양의 상관관계, 한 변수가 증가하면 다른 변수도 증가하는 경향

    - $r>0$: 음의 상관관계, 한 변수가 증가하면 다른 변수는 감소하는 경향

    - $r=0$: 두 변수 사이에 직접적인 연관성이 없다

* 히트맵 혹은 corr() 함수를 통해 상관계수를 확인할 수 있음

    ```python
    import seaborn as sns
    import matplotlib.pyplot as plt
    sns.heatmap( df.corr(), annot = True)
    plt.show()
    ```

> ***<VIF지수>***

VIF 가 높으면 다중공선성이 존재한다고 판단

* VIF = 1: 해당 독립 변수는 다른 변수와 상관관계가 전혀 없음을 의미

* VIF < 5: 일반적으로 다중공선성 문제가 없다고 간주

* VIF > 5: 다중공선성의 징후가 있을 수 있음

* VIF > 10: 다중공선성이 심각하다고 간주되며, 이 변수를 포함한 회귀 모델은 불안정할 가능성 큼 -> 해당 변수를 제거하거나, 변수 간의 상관관계를 줄이기 위한 조치

> ***<다중공선성 대처 방법>***

* 변수 제거

* 변수 변환

    - 변수들을 더하거나 뺴서 새로운 변수 생성

    - 독립변수를 더하거나 뺴더라도 문제가 없는 경우

* 규제선형모델 활용

    - 릿지, 라쏘, 엘라스틱넷 등의 방법을 통해 모델의 복잡도 줄일 수 있음

* PCA(주성분 분석)

    - 데이터의 차원을 축소하는 데에 사용되는 통계적 기법, 고차원 데이터에서 중요한 정보를 최대한 보존하면서 데이터의 복잡성을 줄이고자 함, 데이터 차원이 너무 높으면 분석과 시각화 모두 어려움

    - ***데이터의 방향성을 찾아 변환하고 중요한 정보만 남겨 차원을 축소***

---

> #### **규제선형모델**

##### **(1) 규제선형모델이란?**

* 모델이 학습 데이터에 과적합(overfitting) 되지 않도록 규제를 가하고자 등장한 모델
    
    - 선형회귀 모델에선 특성에 곱해지는 계수(기울기)의 크기를 조정하는 것

* 기존 선형 모델에서는 최소제곱법과 같이 RSS(잔차 제곱합)를 최소화하는 방식으로 비용 함수 설정 -> 오히려 회귀 계수가 쉽게 커지게 되어 과적합이 발생할 가능성이 커짐

* 여러 독립 변수를 사용하여 예측하는 다중회귀에서도 모델이 복잡해져 과적합 가능성 증가




비용함수는 학습 데이터의 잔차 오류값을 최소화하는 RSS 최소화 방법(최소제곱법)과 과적합을 방지하기 위해 ***회귀 계수 값이 커지지 않도록 하는 방법***이 서로 균형을 유지해야함

##### **(2) 규제선형모델의 종류**

* ##### **L2 규제 : W(회귀계수) 의 제곱에 대해 페널티를 부여하는 방식** - **릿지(Ridge)회귀**

    - $$\text{Cost Function} = \text{RSS} + \alpha \sum_{j=1}^{p} w_j^2$$

    - 회귀계수의 **제곱**을 페널티로 주기 때문에, 큰 회귀계수일 수록 훨씬 강한 페널티 받음

    - 모든 계수의 크기가 작아지지만, 규제를 통해 변수가 제거되진 않음(회귀계수가 0이 되지 않음)
<br>


* ##### **L1 규제 : W(회귀계수) 의 절댓값에 대해 페널티를 부여하는 방식** - **라쏘(Lasso)회귀**

    - $$\text{Cost Function} = \text{RSS} + \alpha \sum_{j=1}^{p} |w_j|$$

    - 회귀계수의 **절대값**을 패널티로 주기 때문에, 회귀계수의 크기와 상관없이 같은 패널티 받음

    - 작은 계수가 0이 되기도 하기에 변수가 완전히 사라질 수 있음
<br>

* ##### **L2 규제 + L1 규제 : 엘라스틱넷(Elastic Net) 회귀**

    - $$\text{Cost Function} = \text{RSS} + \alpha_1 \sum_{j=1}^{p} |w_j| + \alpha_2 \sum_{j=1}^{p} w_j^2$$

    - L1 의 변수 선택 기능(일부 계수 삭제)과 L2의 규제 기능(큰 계수를 줄임)을 동시에 활용하는 모델

    - L1 은 상관관계가 높은 변수 중 일부만 선택하는 경향이 있음 -> L2 를 함꼐 적용하면 상관관계가 높은 변수를 함께 고려하면서 규제 가능

### **3. 모델 평가방법**

> #### **성능 평가 지표**

##### **(1) 평균 제곱 오차(MSE)**

예측값과 실제값 사이의 **오차 제곱의 평균**

$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 
$$

##### **(2) 평균 절대 오차(MAE)**

모델의 예측값과 실제값 사이의 **절대값 오차의 평균**

$$
\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|
$$

##### **(3) MSE 와 MAE 의 차이**

* MSE 는 큰 오차에 대해 민감하게 반응함(제곱이기 때문) -> 이상치에 민감

* MAE 는 모든 오차를 동일하게 취급 -> 이상치가 존재하는 경우에도 전반적인 예측 성능 유지 가능

---

> #### **변수 유의성 평가**

##### **(1) T-검정**

* 귀무가설 : 회귀계수(a)가 0이다 -> 독립변수가 종속변수에 영향을 주지 않는다
* 대립가설 : 회귀게수(a)가 0이 아니다 -> 독립변수가 종속변수에 영향을 준다

    <t-값 계산>

* $$ t = \frac{\hat{\beta} - 0}{SE(\hat{\beta})}$$

    - $\hat{\beta}$ : 추정된 회귀계수
    - $SE(\hat{\beta})$ : 표준오차
    - t-분포를 따름.

    <P-value 확인 및 판단>

- **p-value < 0.05** → 유의 수준 5%에서 $H_0$ 기각 → **해당 변수는 유의미함**

- **p-value ≥ 0.05** → $H_0$ 기각 불가 → **해당 변수는 유의미하지 않음**

- **회귀계수가 0이면 해당 독립변수는 의미 없음**

- **p-value가 작을수록** 해당 변수는 종속 변수에 **더 큰 영향을 줄 확률이 높아짐**

- **t-값이 크면** 귀무가설을 기각할 가능성이 높아짐
