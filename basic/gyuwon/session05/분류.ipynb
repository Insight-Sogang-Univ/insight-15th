### 1. 분류란?

#### (1) 지도 학습 : 회귀 vs 분류 

##### 회귀(regression) : 주어진 데이터(X) 를 기반으로 **정답(Y)를 잘 맞추는** 함수
> 데이터가 연속형 변수를 예측하기 위해 사용될 때

##### 분류(classificiation) : 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 **레이블을 판별**하는 것
> 데이터가 범주형 변수를 예측하기 위해 사용될 때

* 이진 분류(Binary Classification) : 예측하고자 하는 변수의 어떤 기준에 대하여 참 또는 거짓의 값만을 가질 때

* 다중 분류(Multiclass Classification) : 예측하고자 하는 변수가 가질 수 있는 값이 3개 이상일 때

### 2. 분류 모델

#### (1) 로지스틱 회귀

> 독립 변수의 선형 조합(선형회귀)에 로지스틱 함수를 적용하여 출력값을 0에서 1사이로 변환해주는 것

📌 **회귀 모델이지만 직선($y=wx+b$)을 사용하지 않음**

1. x 와 y의 관계를 표현하기 위해서는 직선을 표현하는 함수가 아니라 S자 형태로 표현할 수 있는 함수 필요

    - 시그모이드 함수는 입력값이 작거나 클 때는 천천히 변화하고, 중간값 근처에서는 급격히 변함 -> 모델이 예측할 때 급격한 변화를 방지하고, 현실적인 확률값 제공

2. 예측값은 0과 1사이의 값을 가져야 하는데, 직선으로 분류 작업을 수행하면 값이 양/음의 무한대 처럼 범위를 벗어날 수 있게 됨

##### ***시그모이드 함수***

= 로지스틱 함수 : 출력이 0과 1사이의 값을 가지면서 S 자 형태로 그려지는 함수

***출력값이 특정값 이상이면 1(True), 특정값 이하면 0(False)***

시그모이드 함수가 $H(x)=\frac{1}{1+e^{-(wx+b)}}=sigmoid(wx+b)=\sigma(wx+b)$일 때,

    인공지능은 주어진 데이터에 적합한 가중치 w와 b를 계산해서 찾음

#### (2) 의사결정나무

> 조건에 따라 데이터를 분류하며 최종적으로 데이터가 순수한 label 의 집합으로 구성될 때까지 분류를 반복하는 분석 방법(스무고개)

##### ***구성요소***

- **Root Node**: 가장 상위에 위치한 노드

- **Parent Node**: 주어진 노드의 상위에 있는 노드

- **Child Node**: 하나의 노드로부터 분리된 2개 이상의 노드들

- **Binary Tree**(이진 트리): Tree 중에서 children이 최대 2개인 tree

- **Leaf Nodes**: 자식마디가 없는 노드, 모델에서 label에 해당

- **Edge**: Parent node와 child node를 연결하는 연결선

- **Height**: 특정 노드에서 가장 멀리 있는 leaf node까지의 경로에 있는 edge 개수

- **Depth:** Root node에서 특정 노드에 도달하기 위한 edge의 수

##### ***CART 알고리즘***

CART : 가장 널리 사용되는 의사결정 나무 알고리즘으로, 데이터셋을 임계값 기준으로 두 child 로 나누는 알고리즘

    불순도(지니 계수) : 분류하려는 데이터 집합에서 서로 다른 클래스(범주)가 섞여 있는 정도. 임계값을 나눌 때 이 불순도가 낮아지는 방향으로 나눔

1. **임계값 설정** : t1~t4 의 임계값을 기준으로 데이터를 그룹으로 나눔



2. **불순도 감소 알고리즘** : 데이터를 어떤 기준으로 분류했을 때 동일한 객체들로만 잘 모아지게 할 수 있는지를 고려하여 분류기준을 찾아 불순도 알고리즘을 사용해 불순도 낮은 쪽으로 가지를 형성해나감

#### (3) SVM(Support Vector Machine)

> 클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘 => 데이터를 분리하는 초평면(Hyperplane) 중에서 데이터들과 가장 거리가 먼 초평면을 선택하여 분리하는 지도 학습 기반의 이진 선형 분류 모델

-> 명확하게 분류할 수 있는 데이터 집단에서 뛰어난 성능 보이며 고차원 공간에서도 효과적으로 사용 가능

##### ***SVM 의 구성***

- **Support vector** (서포트 벡터): 데이터 중에서 결정 경계와 가장 가까이에 있는 데이터들의 집합

- **Decision Boundary** (결정 경계): 데이터 분류의 기준이 되는 경계

- **Margin** (마진): 결정 경계에서 서포트 벡터까지의 거리(여유 공간)

- **Hyperplane** (초평면): $n$차원 공간의 ($n-1$)차원 평면

- **Slack Variables** (슬랙 변수/여유변수): 완벽한 분리가 불가능할 때 분류를 위해 허용된 오차에 대한 변수

##### ***최적의 선을 찾는 방법***

> 결정 경계는 데이터로부터 가장 멀리 떨어져 있는 것이 좋음 -> Margin(거리)가 가장 큰 경우를 선택함으로써 최적의 선을 찾는다

##### ***SVD 장단점***

| **장점** | **단점** |
| --- | --- |
| 서포트 벡터만을 이용해서 결정 경계를 생성하므로 데이터가 적을 때 효과적 | 데이터 전처리 과정이 중요 |
| 새로운 데이터가 입력되면 전체 데이터 포인트와의 거리를 계산하지 않고 서포트 벡터와의 거리만 계산하면 되기 때문에 연산량 최소화 | 데이터가 많아질수록 최적화를 위한 평가 과정이 많아져서 속도가 느림 |
| 정확성이 뛰어나며, 비선형 모델 분류 가능 |  |
| 다른 모형보다 과대 적합의 가능성이 낮고, 노이즈의 영향이 적음 |  |

#### (4) KNN(K-Nearest Neighbor)

> 데이터로부터 거리가 가까운 k개의 다른 데이터 레이블을 참조하여 분류하는 알고리즘(학습 필요 X, 별도의 모델없이 데이터만을 이용해 새로운 데이터가 왔을 때 classify 하는 것)

##### ***KNN 알고리즘 계산 순서***

1. **데이터 준비**

    * KNN은 미리 학습하는 과정이 없어 데이터 준비. 각 데이터는 특징 벡터와 레이블로 이루어짐

2. **K 값 설정**

    * 원하는 K 값 설정. K값은 가장 가까운 이웃의 개수를 나타내며 보통 홀수개로 설정(짝수로 설정하면 동점 발생, 그래도 동점 발생했다면 동점 발생한 클래스들 중 더 가까운 이웃에 해당하는 클래스로 분류/동점 발생한 클래스들 중 무작위 하나 선택)

        - 최적의 K 선택하기 위해 학습용 데이터 개수의 제곱근으로 설정

        - 너무 크면 주변의 점과 근접성 떨어져 과소적합

        - 너무 작으면 이상값, 잡음 데이터와 이웃될 수 있음

3. **거리 계산**

    * 새로운 데이터(예측하려는 데이터)가 주어지면 이 값과 기존 모든 데이터 간의 거리 계산(유클리드 거리, 맨해튼 거리)

4. **가장 가까운 K개의 이웃 선택**

    * 계산된 거리 중에서 가장 작은 거리 값을 가진 K개의 데이터를 선택

5. **분류하기**

    * K개의 이웃 중 가장 많이 등장하는 클래스가 예측 결과

### 3. 분류 평가 지표

#### (1) 혼동 행렬

> 분류 모델의 예측 결과를 정확한 예측과 잘못된 예측으로 구분하여 나타낸 표(행렬)

| 구분 | 분류 값 | 설명 |
| --- | --- | --- |
| 예측이 정확한 경우 | **True Positive (TP)** | 정답이 참, 참으로 예측 |
|  | **True Negative (TN)** | 정답이 거짓, 거짓으로 예측 |
| 예측이 틀린 경우 | **False Positive (FP)** | 정답이 거짓, 참으로 예측 (예측이 틀림)  |
|  | **False Negative (FN)** | 정답이 참, 거짓으로 예측 (예측이 틀림)  |

##### ***혼동 행렬을 이용한 분류 모델 평가 지표***

* 정확도 : 모든 가능한 예측 중 참인 비율

    - (TP+TN) / (TP+TN+FP+FN)

* 정밀도 : 참이라고 예측한 경우 실제 참의 비율

    - (TP) / (TP + FP)

* 재현도 : 실제로 참인 경우 중 참으로 예측하는 비율

    - (TP) / (TP + FN)

    📌 정밀도와 재현도는 Trade-off 관계

Threshold(참/거짓을 가르는 분류기준)가 낮아지면 모델이 참으로 예측하는 경우 많아지기에 재현도는 증가하지만 정밀도의 분모가 커지기에(FN) 정밀도는 낮아지게 된다

##### ***정밀도/재현도 그래프***

> 정밀도 & 재현도 그래프에서 두 value 값이 만나는 지점을 Threshold로 정하면, 예측 오류를 최소화할 수 있음

- 각 점은 분류 모델이 해당 임계값에서 precision과 recall value를 어떻게 예측했는지 표시
- 모든 임계값에 대해 점을 그려서 모델의 예측 성능을 시각적으로 파악할 수 있음
- **두 value 값이 만나는 지점을 임계값으로 정할 때, 예측 오류를 최소화**할 수 있음

#### (3) ROC/AUC Curve

##### ***ROC Curve***

> 얼마나 분류가 잘 되었는가를 보여주는 그래프

* **True Positive Rate (TPR)** : 참인 것들 중에 참이라고 예측한 비율 (=재현율)

* **False Positive Rate (FPR)** : 거짓인 것들 중에 참이라고 잘못 예측한 비율 

##### ***AUC Curve***

> ROC 와 x 축 사이의 면적(적분값)으로 1에 가까울 수록 분류 성능이 좋음

### 4. 하이퍼파라미터 최적화

#### (1) 하이퍼파라미터

> 학습 시작 전에 사용자가 직접 설정하는 변수. 하이퍼파라미터 최적화란 적절한 하이퍼파라미터를 찾아 모델 성능을 향상시키는 것

---

#### (2) 하이퍼파라미터 최적화 과정/방법

> 과정

1. Hyperparameter 탐색 범위 설정 (최적 값을 찾고 싶은 하이퍼파라미터의 범위를 설정)

2. 평가 지표 계산 함수(성능 평가 함수) 정의(탐색하려는 Hyperparameter를 인수로 받아 평가지표 값을 
계산해주는 함수를 정의)

3. 1단계에서 샘플링한 Hyperparameter 값을 사용하여 검증 데이터로 정확도 평가

4. 위 단계를 특정 횟수 반복하며, 정확도 결과를 보고 하이퍼파라미터의 범위 좁힘

> 방법

- Grid Search: 정해진 범위에서 Hyperparameter를 모두 순회

    - 장 : 범위가 넓고 step 이 작을수록 꼼꼼하게 전 범위를 탐색해 최적 해를 정확히 찾을 수 있음

    - 단 : 시간이 오래 걸림

- Random Search: 정해진 범위에서 Hyperparameter를 무작위로 탐색

    - 장 : 속도가 Grid Search 보다 빠름

    - 단 : 무작위라는 한계 때문에 정확도가 떨어짐.

- Bayesian Optimization: 사전 정보를 바탕으로 Hyperparameter 값을 확률적으로 추정하며 탐색

    - 초기 값 몇 개를 무작위로 실험

    - 그 결과를 바탕으로 어떤 파라미터의 조합이 좋을 지를 확률 모델로 예측

    - 다음 실험할 조합을 가장 좋을 것 같은 위치로 선택

    - 결과를 반영하고 다시 반복
