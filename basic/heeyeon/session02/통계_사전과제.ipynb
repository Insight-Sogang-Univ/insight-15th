# 기술통계 <br>
>기술통계란, 데이터 샘플을 그대로 설명하고 이해하려는 시도. 평균, 불량률, 연령 분포 등 
## 기술통계의 3요소 <br>  
  
1.**중심화경향** 데이터의 중심을 나타냄.  
평균:총합/데이터 개수 중앙값: 데이터 순서상 정중앙값 최빈값: 가장 자주 나타나는 값   
*series.median()*  

2.**자료의산포**  
데이터가 퍼진 정도.  
 -범위: 최대값-최소값 (Range) 이상치가 있을 경우 결과물이 왜곡될 수 있다.  
 -분산: 값이 평균으로부터 얼마나 퍼져있는지. 항상 양수이며 값이 클수록 데이터가 널리 퍼져있음   
 $ \frac{\sum_{i=1}^{N}(x_i - μ)^2}{N} $  
 -분산에 제곱근을 취함으로서 좀더 직관적인 지표로 변환  
$ \sqrt{\frac{\sum_{i=1}^{N}(x_i - μ)^2}{N}}$   

3.**자료의 정리 및 시각화**  
>첨도와 왜도를 통해 확률분포의 전체 양상을 확인한다. (꼬리가 두꺼운 정도와, 한방향으로 중앙값이 치우친 정도)  
 
### *series.kurt(), series.skew()*

# 확률분포 <br>
## 1. 정규분포 (Normal Distribution)
> 평균을 중심으로 좌우가 대칭인 종 모양의 분포이다. 많은 자연현상과 사회현상이 따르는 대표적인 분포
    1. 평균 = 중앙값 = 최빈값
    2. 표준편차가 클수록 분포가 퍼짐 

## 대표 분포 모형 <br>
## 1. Z 분포  

- **모집단 자체가 갖는 분포라기보다는, 정규분포의 통계적 성질을 이용해 가설 검정이나 구간 추정 같은 통계적 추론을 하기 위해 만들어진 도구**  

z = (표본평균 - 모평균) / (표준편차 / √n)

## 2. 이산 확률분포 (Discrete Probability Distribution)
- 확률변수가 정수 값을 가질 때 사용하는 분포
  - **이항분포**: 성공/실패처럼 두 결과가 반복되는 실험, 모집단 자체가 갖는 분포라기보다는, 정규분포의 통계적 성질을 이용해 가설 검정이나 구간 추정 같은 통계적 추론을 하기 위해 만들어진 도구  
  - **포아송분포**: 일정 시간/공간에서 발생하는 사건 수  


## 3. 연속 확률분포 (Continuous Probability Distribution)  
- 확률변수가 **연속적인 값**을 가질 때 사용  
- 대표 예
  - 정규분포: 모집단의 분포를 몰라도 표본의 크기가 충분히 크다면(n>=30) 포본평균의 분포는 근사적으로 정규분포를 따른다
  - t분포: 자유도가 커짐에 따라 표준정규분포 N(0,1)에 수렴
  - 카이제곱분포 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포
  - F분포: - 두 독립적인 카이 제곱 분포의 비율  
  두 집단 간의 분산 차이를 비교하는 데 사용됨

# 가설검정 (Hypothesis Testing)

모집단에 대한 **가설이 맞는지 통계적으로 판단하는 과정**  
표본을 추출하여 얻은 표본통계량으로 모집단 모수에 대한 새로운 예상·가설 등이 옳다고 판단할 수 있는지를 평가하는 과정  

---

## 1. 가설 설정

- **귀무가설(H0)**  
  → 차이가 없다 / 효과가 없다: 반박하고자 하는 기존 가설

- **대립가설(H1)**  
  → 차이가 있다 / 효과가 있다: 내가 새롭게 주장하고자 하는 가설

예시  

*마포구와 서대문구 인구의 평균 키는 같다.8

- H0 : 마포구 평균키 = 서대문구 평균키  
- H1 : 마포구 평균키 ≠ 서대문구 평균키  

---

## 2. 유의수준 설정 (α)

- 귀무가설을 **잘못 기각할 허용 확률**
>유의 수준을 0.05로 설정한다는 것은, "귀무가설이 실제로 참인데도 불구하고, 5%의 확률로 귀무가설을 잘못 기각할 수 있다"는 것을 의미

보통 사용하는 P값 
- 0.05
- 0.01

---

## 3. 검정통계량 계산

데이터로부터 계산한 통계값
사용되는 통계량

- **Z 값**: Z값 또는 Z-점수는 표준 정규 분포(평균이 0이고 표준편차가 1인 분포)를 사용하여 얻은 값
- **t 값**: T값은 T분포를 사용하여 얻은 값으로, 작은 표본 크기에 대한 가설 검정(T검정)에 주로 사용
- **F 값**: F값은 F분포를 사용하여 얻은 값으로, 주로 두 표본 분산 사이의 차이를 비교하는 데 사용

---

## 4. 유의확률 (p-value)

- 귀무가설이 맞다고 가정했을 때, 현재 데이터가 나타날 확률
즉 우연히 대립가설이 맞을 확률 (1종 오류)
가설 검정에 쓰일 수 있는 또 다른 척도입니다. 유의확률은 검정통계량에서 유도할 수 있으며 유의 확률(p)가 유의 수준(α)보다 작을 시에 귀무가설을 기각

*판단 기준 도식*
p-value < α → 귀무가설 기각
p-value ≥ α → 귀무가설 채택

# 1종 오류와 2종 오류

## 1종 오류 (Type I Error)
- **참인 귀무가설을 기각하는 오류**
>실제로 차이가 없는데 차이가 있다고 판단하는 것 


확률 = α

---

## 2종 오류 (Type II Error)
- **거짓인 귀무가설을 채택하는 오류**
>귀무가설이 실제로 거짓인데 기각하지 못할 확률  


## 가설검정 과정 <br>

1. 가설 설정  
- 귀무가설(H0) : 기각하고자 하는 명제로 설정. 보통 “~는 차이가 없다”라는 명제
- 대립가설(H1) : 내가 주장하고자 하는 명제로 설정. 보통 “~는 차이가 존재한다” 

2. 유의수준 설정 (α) 연구자가 결과를 판단할 때 허용할 수 있는 오류의 최대치로, 주로 0.05, 

3. 검정통계량 계산  
- Z, t, F 등
   해당 데이터가 평균으로부터 멀리 떨어져있는지, F분포를 사용하여 얻은 값으로, 주로 두 표본 분산 사이의 차이가 얼마나 되는지 확인 가능

4. p-value 계산: 가설 검정을 하기 전에 유의 수준 α를 정해야 합니다. 일반적으로 α=0.05로 설정/ 이 유의 수준은 "5%까지의 오류 가능성을 감수하고 귀무가설을 기각할 준비가 되어 있다"는 의미

5. 결론 도출  
   - 귀무가설 기각 / 채택
