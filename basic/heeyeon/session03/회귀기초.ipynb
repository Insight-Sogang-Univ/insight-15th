# 머신러닝
머신러닝이란 모델을 학습하여 새로운 데이터를 예측하는 기술. 
할 수 있는 것
-패턴 발견
-새로운 데이터 예측

## 머신러닝의 종류 
|지도학습|비지도학습|강화학습|
|---|---|---|
|정답이 있는 데이터|정답이 없는 데이터|보상|
|기계의 예측값이 정답과 같아지게|데이터 패턴 학습|시행착오 반복, 보상 최대화|

지도: 문제, 정답 모두 알려줌
비지도: 답을 가르쳐주지 않음
강화: 보상 통해 행위 강화

>모델란 학습의 결과물로 예측기(공식)을 만들어가는 전체적인 과정 설계, 훈련, 검증을 거침

### 지도학습

독립변수와 종속변수 모두 존재
EX) 대표적인 예로 연속적 숫자값을 예측하는 회귀, 입력된 데이터를 항목들로 나누는 분류가 있음 
>회귀란, 주어진 데이터 X기반으로 정답 Y를 맞히는 함수, 회귀선을 찾는 문제



# 선형회귀
종속변수가 불연속적일 땐 회귀가 아닌 분류를 사용. 
선형, 비선형, 규제 등의 다양한 기법이 존재
*"회귀선(ex. 직선 y=wx+b)"을 찾는 과정*  

가중치: 독립변수가 종속변수에 주는 영향력의 크기, 즉 기울기  

편향: 독립변수가 0일 때 나오는 기본값, 그래프에선 Y절편

## 단순회귀

>하나의 독립변수만 존재하는 선형회귀
데이터를 직선 형태로 표현, 적절한 회귀계수와 편향을 찾아간다.


잔차:선형회귀를 통해 X에따른 예측값 Y를 추론할 수 있다.
여기서 실제값과 예측값 간 차이를 잔차라고 부르며, 이 차이를 줄여가는 과정이 회귀 추론의 주요점이다. 

**최적의 W와 B를 찾는 방법?**  
-최소제곱법:(최소자승법) RSS 가 최소가 되는 지점을 찾음
-경사하강법:위 방법이 어려울 경우 사용, 잔체 제곱을 줄이려는 목적은 그대로이다. 
목적함수가 원하는 최적화=비용함수의 최소화  
시험결과의 총합 최대화=시험 감점의 최소화    
$E = \frac{1}{n} \sum_{i=1}^{n} [t_i - y_i]^2$  
잔차를 제곱한 뒤 합하는 손실함수  

- $t_i$ : $i$번째 데이터포인트 실제값(정답)  
- $y_i$ : $i$번째 예측값  
- $E$: Error  
- $n$: 데이터 샘플의 총 개수  

학습률 보폭, LOCAL MINIMA 작은 기울기에선 나오기 어려운 문제 등이 있다.
그에 대한 해결법으로 *모멘텀*: 현재 기울기만을 기준으로 움직이며 이전의 기울기와 방향을 고려하여 관성을 부여해줌.  

-지역 최솟값 탈출 가능  
-작은 기울기는 쉽게 넘어갈 수 있음

## 다중선형회귀
독립 변수들 간의 강한 상관관계가 있어 전체 회귀모델이 불안정해지는 문제. 추정 결괏값이 부정확해지고, 해석력이 떨어짐. EX) 주택의 크기와 방의 개수는 두 변수간의 강한 상관관계가 있음. 
확인 방법
1. 상관계수 **sns.heatmap(df.corr(), annot=True)**
2. vif 지수 회귀모델의 결정계수 r square을 사용 

대처방법
1.변수 제거
2.변수변환
3.규제선형모델 활용



# 규제선형모델
>규제선형모델이란: 모델이 과적합되면 특정 변수에 집착하게 되어 전체 균형이 깨짐. 학습 데이터에는 완벽히 들어맞지만 새로운 데이터에서는 성능이 크게 떨어진다. 따라서 회귀계수 값이 너무 커지지 않도록 서로 균형을 유지해야한다.  

-릿지 회귀: 회귀계수의 제곱에 대해 페널치를 부여  
-라쏘회귀: 회귀계수의 절댓값에 대해 페널티 부여  
-두가지 모두 사용하는 것이 엘라스틱넷 회귀  

# 모델 평가 방법
## 성능지표평가
>평균제곱오차  

$\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 $  
>평균절대오차  

$\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$  

>mse와 mae의 차이  

mae는 모든 오차를 동일하게 취급하므로 이상치가 존재해도 전반적 성능 유지 가능 

## 변수유의성평가
>t 검정
독립변수의 회귀계수가 유의미한지 검정하는 데 사용됨 
$t = \frac{\hat{\beta} - 0}{SE(\hat{\beta})}$  

t 값이 클수록 기존 귀무가설을 기각할 가능성이 높아짐 
pvalue는 별도로 확인하여 0.05보다 작아 대립가설이 유의한지 꼭 확인하여야 한다.
