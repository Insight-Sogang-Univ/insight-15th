# 회귀와 분류의 차이
## 연속 숫자값을 예측하는 것이 회귀라면 입력된 데이터를 특정 항목으로 나누는 것이 분류이다.   

| |회귀|분류|
|---|---|---|
|종속변수|주택가격|대출상환여부|
|독립변수|주택 평수|고객의 특성|  

>분류는 패턴을 알고리즘으로 인지후 새로운 과측 데이터를 레이블 판별(0/1) 한다.

# 분류 모델의 네 가지 종류와 각 모델
1. 로지스틱 회귀: 일반적인 선형회귀와 달리, 이진 분류 문제를 해결하는 데 적합함. *이메일의 스팸 여부*  
단순선형회귀는 일부 분류문제에 적합하지 않다. 
1-(2) 시그모이드 함수: 출력이 0과 1 사이 값, s 자 형태로 그려짐   
> $H(x)=\frac{1}{1+e^{-(wx+b)}}=sigmoid(wx+b)=\sigma(wx+b)$  
적합한 w와 b를 계산해내서 찾는다.  

2. 의사결정 나무: 조건에 따라 순차적으로 데이터를 분류하도록 반복함.  
- **Root Node**: 가장 상위에 위치한 노드
- **Parent Node**: 주어진 노드의 상위에 있는 노드
- **Child Node**: 하나의 노드로부터 분리된 2개 이상의 노드들
- **Binary Tree**(이진 트리): Tree 중에서 children이 최대 2개인 tree
- **Leaf Nodes**: 자식마디가 없는 노드, 모델에서 label에 해당
- **Edge**: Parent node와 child node를 연결하는 연결선
- **Height**: 특정 노드에서 가장 멀리 있는 leaf node까지의 경로에 있는 edge 개수
- **Depth:** Root node에서 특정 노드에 도달하기 위한 edge의 수  

3. 서포트 벡터 머신: 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘, 데이터들과 가장 거리가 먼 초평면을 선택  
 > 서포트 벡터란 데이터 중에서 결정경계와 가까이 붙은 집합
 결정경계란 분류 기준의 경계가 됨   
4. k-nearest neighbor  
유사 특성으로 묶인 데이터끼리 가까이 붙어있다는 점을 이용, 
가장 가까운 k개의 데이터를 선택하는 과정이 중요한데, 일반적으로는 학습용 데이터 개수의 제곱근으로 설정한다. 
 *knn은 어떠한 학습도 선행되지 않고, 별도 모델 없이 즉석에서 주변 데이터 활용하는 것이 큰 특징*

# 분류 평가 지표
1. 혼용 행렬  
> 정확한 예측과 잘못된 예측으로 행렬 구분  

| 구분 | 분류 값 | 설명 |
| --- | --- | --- |
| 예측이 정확한 경우 | **True Positive (TP)** | 정답이 참, 참으로 예측 |
|  | **True Negative (TN)** | 정답이 거짓, 거짓으로 예측 |
| 예측이 틀린 경우 | **False Positive (FP)** | 정답이 거짓, 참으로 예측 (예측이 틀림)  |
|  | **False Negative (FN)** | 정답이 참, 거짓으로 예측 (예측이 틀림)  | 
*정확도: 모든 가능한 예측 기준 정밀도: 참 예측 경우 기준 재현도: 실제 참 기준*
2. F1-score: 정밀도와 재현율의 조화 평균  
머신러닝 모델의 성능 평가 주요 지표/정밀도와 재현율 간 균형 평가  
3. ROC / AUC Curve 
얼마나 분류가 잘 되었는지, tpr(참인 것들 중에 참이라고 예측한 비율) 고려

# 하이퍼파라미터 최적화  
> 하이퍼 파라미터란, 모델학습 과정에 반영되는 사용자 직접 설계 변수로, 학습 시작 전 사용자가 미리 변수를 설정하는 것이다.  

탐색 범위를 설정하고, 평가 지표를 계산하고, 샘플링한 값을 사용하여 검증 데이터를 정확도 평가하여 이 모든 단계를 특정 횟수 반복한다. 
- Grid search 정해진 범위에서 모두 순회/넓은 범위, 큰 step을 활용해 범위를 좁힘
- Random search 정해진 범위에서 무작위 순회 / 무작위의 한계로 정확도가 떨어짐, 사용 빈도수 낮음
