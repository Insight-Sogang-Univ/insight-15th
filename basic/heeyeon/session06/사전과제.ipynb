# 거리 / 유사도 정의와 측정 방법 

>데이터 간 얼마나 비슷한 지 판단하는 기준이다.

- 유클리드 거리: 직선 거리  
$d(\mathbf{x}, \mathbf{y}) = \sqrt{\sum_{i=1}^n (x_i - y_i)^2}$  
- 맨해튼 거리: 좌표 축을 따라가는 이동 거리
- 코사인 유사도: 방향 즉 각도 기준

**거리가 작을수록, 유사도가 클수록 더욱 비슷하다.**

# 주성분 분석(PCA)의 목적  

> 주성분 분석은 차원을 축소하는 기법으로, 목적은 데이터의 차원을 줄여 속도를 높이거나, 중요한 정보만 남기거나, 시각화를 쉽게하는 것이다. 즉 고차원 데이터를 저차원으로 축소하는 것이다.  


군집 간 거리계간 방법에는 크게 여섯가지가 있다. 가장 가까운 두점 간 거리를 연결하는 단일 연결법, 가장 먼 두점을 연걸하는 완전 연결법, 모든 점 쌍 거리의 평균인 평균 연결 법, 중심 연결법, 두 군집의 중앙값 간 거리인 중앙 연결법, 제곱합이 증가하지 않는 최소값을 선택하는 Ward 연결법이다.  

*차원이란 변수의 개수, 차원 축소 목적은 변수의 개수를 줄이는 것*

**핵심은 분산이 큰 방항-정보가 많은 방향으로 데이터를 압축하는 것**

# 4가지 군집화 알고리즘의 특징과 과정  
### 계층적 군집화  
>트리 구조, 군집 개수를 미리 정하지 않아도 된다.  
가까운 데이터부터 계속합치는 bottom up 방식  
 

### k-means  
>가장 많이 쓰이며, k 즉 군집 개수를 미리 정해야 한다.  


방법은 1. 중심점을 설정하고 2. 가까운 중심으로 데이터를 할당하고 3. 중심을 재계산한다. 이와 같은 방법을 반복해서 재계산하면 되지만, 단점은 초기값에 민감하다는 것이다.   

### DBSCAN  
> 밀도 기반, 군집 개수를 자동으로 결정하며 노이즈 즉 이상치를 탐지하는 기능이 있다.  


일정 거리 내 점이 많으면 군집을 형성하며, 이상치 처리에 특히 강점이 있다.  

### GMM  
>확률 기반 군집화로, 데이터라 여러개의 정규분포로 구성된다고 가정한다.  

각 데이터가 군집에 속할 확률을 계산하며, k means 보다 타원형 군집이 가능하는 등 더욱 유연하다

# 군집화 평가 방법 (실루엣 계수, 던 인덱스)  
- 실루엣 계수: 값의 범위가 -1에서 1까지. 1에 가까울수록 잘 군집되고 좋다. 0근처는 모호하고, 음수는 잘못 군집되었다는 지표가 된다  
- 던 인덱스: 군집 간의 거리/군집 내부의 거리로 계산한다. 군집은 서로 멀고 내부가 촘촘해야 좋으므로 값이 클수록 더욱 좋다.
