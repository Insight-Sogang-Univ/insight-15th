# 통계

    - 표본에서 기술통계로 얻어진 모집단의 특징에 관한 유용한 정보를 추출, 추측하는 것

## 자료의 척도
- 질적자료 : 수치로 표현 But 연산의 의미 X
    - 명목척도 : 레이블이나 이름으로 요소의 속성을 식별하려고 하는 척도
    - 서열척도 : 명목형의 분류 속성과 더불어, 항목 간의 순서 또는 순위를 나타내는 특성을 지니는 척도
- 양적자료 : 수치로 표현
    - 구간(등간)척도 : 값들 사이에 순서가 있고, 그 간격이 일정하여 덧셈이나 평균과 같은 연산이 가능한 척도
        - But 곱셈, 나눗셈 X
    - 비율척도 : 절대영점과 임의의 측정단위로 구성하고, 동간성이 유지되는 척도
        - 0 값은 그 변수에 대해 아무것도 존재하지 않는 상태

## 기술통계
- 데이터를 있는 그대로 설명하고 이해하는 데 중점을 둔 통계 방법
- 기술 통계의 세 가지 도구 집합
    1. 중심화 경향 : 데이터의 중심적인 경향을 나타내는 기술 통계량
        - 평균 (mean)
            - 데이터 값의 총합을 데이터 개수로 나눈 값
            - 이상치에 민감
        - 중앙값 (median)
            - 데이터를 크기 순서대로 일렬로 정렬했을 때, 한가운데에 위치하는 값
            - 평균에 비해 이상치에 덜 민감
        - 최빈값 (mode)
            - 데이터에서 가장 자주 나타나는 값
            - 주로 이산형 데이터, 데이터의 분포 이해에 도움

    2. 자료의 산포
        - 범위 (range)
            - 데이터의 최대값과 최소값의 차이
            - 이상치에 매우 민감
        - 분산 (variance)
            - 데이터 값들이 평균으로부터 얼마나 멀리 퍼져있는지 나타내는 지표
            - 분산 값이 크다 -> 데이터가 많이 퍼져 있다
            > $\frac{\sum_{i=1}^{N}(x_i - \mu)^2}{N}$
        - 표준편차 (standard deviation)
            - 분산에 제곱근을 취한 값
            > $\sqrt{\frac{\sum_{i=1}^{N}(x_i - \mu)^2}{N}}$
        - 사분위수 (quartile)
            - 데이터의 표본을 네 개의 동일한 크기로 나눈 값
            - 사분위간 범위 (IQR)
                - Q3에서 Q1 값을 제외한 범위, 즉 데이터의 중앙 50%가 퍼져 있는 범위
                - 이상치를 탐지할 때 유용
        - 변동계수 (coefficient of variation)
            - 데이터의 상대적인 변동성을 나타내는 통계적 지표
            - (표준편차/평균) * 100

    3. 자료의 정리 및 시각화
        - 히스토그램, 박스플롯, 파이차트 등


### 확률분포
- 확률변수가 특정한 값을 가질 확률을 나타내는 함수
- 이산확률분포 : 확률 변수가 이산적인(discrete) 값을 가지는 경우
- 연속확률분포 : 확률 변수가 연속적인(continuous) 값을 가지는 경우
- 대표적인 분포 모형
    - 정규분포
        - 형태
            - 종(bell) 모양
            - 평균을 기준으로 대칭
            - 표준편차가 클수록 곡선이 평평해짐 (데이터가 넓게 퍼짐)
        - 매개변수
            - 평균(μ)
            - 표준편차(σ)
        - 특징
            - 가장 잘 알려진 (연속)확률분포
            - 많은 자연 현상과 사회 현상이 정규분포를 따르는 것으로 알려져 있음
        - 큰 수의 법칙
            - 표본의 크기가 커질수록 표본 평균이 모평균에 근사하는 현상
        - 중심극한정리
            - n개의 평균의 분포는 n이 적당히 크다면 정규분포에 가까워진다는 정리
    - Z분포 (표준정규분포)
        - 평균이 0이고 표준편차가 1이 되도록 맞춘 표준적인 정규분포
    - T분포 (T-Distribution)
        - 형태
            - 종(bell) 모양, 정규분포보다 꼬리가 두꺼움
            - 평균을 기준으로 대칭
            - 자유도가 커짐에 따라 표준정규분포 N(0,1)에 수렴
        - 매개변수
            - 자유도(df)
        - 특징
            - 표본으로 모평균(혹은 모평균 간의 차이)을 추정할 때 사용되는 분포
            - 표본 크기가 작을 때 정규분포보다 더 정확한 결과를 제공
    - F분포 (F-Distribution)
        - 형태
            - 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포 
            - 자유도가 증가함에 따라 중심이 오른쪽으로 이동하고 모양이 더 대칭적으로 변화, 정규분포에 가까워짐
        - 특징
            - 두 독립적인 카이 제곱 분포의 비율
            - 두 집단 간의 분산 차이를 비교하는 데 사용됨


## 추론 통계 - 추정
- 수집한 데이터를 기반으로 미래의, 혹은 조사하지 못한 전체 정보를 추론하고 결론을 도출하는 방법을 다루는 분야
- 추정 : 모집단을 대표할 수 있는 일부를 표본으로 추출하여 분석한 통계량을 이용해서 모집단의 모수를 예측하는 과정
    - 점추정 : 추정하고자 하는 모수를 하나의 수치로 추정
    - 구간추정 : 추정하고자 하는 모수가 존재하리라 예상하는 구간을 정하여 추정
- 검정 : 표본을 추출하여 얻은 표본통계량으로 모집단 모수에 대한 새로운 예상·가설 등이 옳다고 판단할 수 있는지를 평가하는 과정

## 추론 통계 - 검정
- 가설 검정의 절차
    1. 가설 설정
        - 귀무가설 H0와 대립가설 H1를 설정
    2. 유의 수준 설정
        - 유의수준 : 연구자가 결과를 판단할 때 허용할 수 있는 오류의 최대치
        - 1종 오류의 확률
            - 1종 오류(𝛼) : 귀무가설이 실제로 참이었는데 기각할 확률
            - 2종 오류(𝛽) : 귀무가설이 실제로 거짓인데 기각하지 못할 확률
    3. 검정통계량 산출
        - 가설 검정의 결과를 판단하는 데 사용되는 값
        - 데이터의 분포나 특징에 따라 T값, F값 등 어떤 통계량을 사용할지 결정
    4. 가설 기각/채택 판단
        - 유의확률(p-value)
            - 유의 확률(p)가 유의 수준(α)보다 작을 시에 귀무가설 기각, 클 시에는 귀무 가설 기각 X

- 데이터의 자료형에 따른 검정 방법
    1. [수치형 - 수치형] (피어슨)상관분석
        - 피어슨 상관 분석
    2. [범주형 - 수치형] F-test(ANOVA), T-test
        - T-test : 범주가 2개일 때
        - F-test(ANOVA) : 범주가 3개 이상일 때
    3. [범주형 - 범주형] Chi2-test
