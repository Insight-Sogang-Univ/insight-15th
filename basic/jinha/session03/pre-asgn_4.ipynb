# 머신러닝
- 인공지능의 한 분야로, 모델이라는 소프트웨어를 학습하여 새로운 데이터를 예측하거나 결정을 내릴 수 있도록 하는 기술
- 데이터에서 패턴을 발견하는 것과 이 패턴을 바탕으로 새로운 데이터에 대해 예측하는 것을 가능하게 함

## 머신러닝의 종류
- 지도학습
    - 문제와 정답을 모두 알려 주고 공부시키는 방법
    - 정답(label)이 있는 데이터 사용
    - 기계의 예측값이 데이터의 정답과 같아지도록 학습
    - 예측, 회귀, 분류
- 비지도 학습
    - 정답(label)이 없는 데이터 사용
    - 데이터의 패턴 혹은 유사성 학습
    - 군집화
- 강화 학습
    - 시행착오를 반복하여 보상을 최대화하는 행동 학습
    - 알파고, 자율 주행

## 모델링
- 모델
    - 학습의 결과물
    - 데이터를 학습하며 찾아낸 패턴이 저장된 상태
    - 새로운 데이터를 넣었을 때 답을 내놓는 예측기 역할
- 모델링
    - 모델을 만들어가는 전체적인 과정
- 모델링 과정
    1. 설계 : 데이터의 특성에 맞추어 어떤 알고리즘으로 학습할지 결정하는 단계
    2. 훈련 : 데이터를 기계에게 계속 보여주며 정답을 맞히도록 수정하는 단계
    3. 검증 : 훈련시킨 모델이 새로운 데이터에서도 정답을 잘 맞히는지 테스트하는 단계

## 지도 학습

### 지도 학습의 과정
1. labeled data 
    - 독립변수(feature)와 종속변수(target/label)가 모두 존재하는 데이터
    - 입력 데이터와 그에 대응하는 출력값이 있는 데이터 
        - 독립변수 : 머신러닝에서 예측에 사용되는 입력 변수
        - 종속변수: 머신러닝에서 예측하고자 하는 대상 변수
2. labeled data를 training set과 test set으로 나눔
    - train set: 모델을 학습(훈련)시킬 때 쓰이는 데이터
    - test set: 학습된 모델을 검증하기 위해 사용하는 데이터
3. training set을 통해 모델을 학습
    - 독립변수(feature)를 통해 추정한 예측값(prediction)이 주어진 종속변수(target/label)와 같아지도록(가까워지도록) 모델을 훈련

### 회귀와 분류
- 회귀
    - 연속적인 숫자값을 예측
    - 주어진 데이터(X)를 기반으로 정답(Y)를 잘 맞히는 함수를 찾는 문제
- 분류
    - 입력된 데이터를 주어진 항목 나누는 것
    - 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별하는 것
- 회귀와 분류의 차이
    - 예측하고자 하는 값, 즉 종속 변수가 연속형이면 회귀, 범주형이면 분류를 사용

## 회귀
- 선형회귀
    - 데이터를 가장 잘 설명하는 회귀선을 찾는 과정
- 가중치와 편향
    - 가중치 w (Weight)
        - 독립변수 $x$가 종속변수 $y$에 주는 영향력의 크기
        - 그래프의 기울기
        - 회귀계수
    - 편향 $b$ (Bias) 
        - 독립변수 $x$가 0일 때 나오는 기본값
        - 그래프의 y절편
- 단순선형회귀
    - 하나의 독립변수($x$)만 존재하는 선형 회귀
    - 데이터를 직선 형태(하나의 독립 변수)로 표현하는 경우
    - 데이터를 가장 잘 설명하는 $y=wx+b$을 찾는 과정에 해당
- 잔차
    - 실제값과 예측값 간 차이를 바로 잔차
    - 회귀선을 잘 찾는 과정은 잔차를 줄여나가는 과정

### 최적의 가중치와 $w$와 $b$를 찾는 방법
1. 최소제곱법
    - 잔차의 제곱의 합(Sum of Squared Residuals, RSS라고도 함)이 최소가 되는 지점
2. 경사하강법
    - 손실 함수(Loss)를 최소화하기 위해 기울기(gradient)의 반대 방향으로 파라미터를 조금씩 업데이트하는 최적화 알고리즘
    - 현재 위치에서 기울기(미분값)를 계산해 가장 빠르게 감소하는 방향을 찾음
    - 이를 반복하면서 손실 함수의 최소값에 점점 가까워짐

### 다중선형회귀

1. 다중공선성
    - 독립 변수들 간의 강한 상관관계로 인해 회귀 모델이 불안정해지는 문제
    - 각 변수의 회귀계수 추정이 부정확, 해석력 감소
    - 확인 방법
        1. 상관계수
            - 변수 간의 통계적 관계인 상관관계의 정도를 수치로 나타낸 것
        2. VIF지수 (분산 팽창 인수)
            - 회귀 모델의 결정계수 R-Square를 사용하여 계산
    - 대처 방법
        1. 변수 제거
        2. 변수 변환
        3. 규제선형모델 활용
        4. PCA(주성분 분석)


### 규제선형모델
- 모델이 학습 데이터에 과적합(overfitting) 되지 않도록 규제를 가하고자 등장한 모델
- 오류는 최대한 줄이면서도, 특정 변수에 집착하지 않도록 균형을 맞춰 과적합을 방지
- 모델의 종류
    1. L2규제: W(회귀계수)의 제곱에 대해 페널티를 부여하는 방식, 릿지(Ridge)회귀
    2. L1규제:  W(회귀계수)의 절댓값에 대해 페널티를 부여하는 방식, 라쏘(Lasso) 회귀
    3. L2규제 + L1규제**: 엘라스틱넷(Elastic Net) 회귀**

### 모델 평가 방법
- 성능 평가 지표
    1. 평균 제곱 오차 (Mean Squared Error, MSE)
    2. 평균 절대 오차 (Mean Absolute Error, MAE)
    3. MSE와 MAE의 차이
- 변수 유의성 평가
    - T-검정
        - 독립 변수의 회귀계수가 유의미한지 검정하는 데 사용
