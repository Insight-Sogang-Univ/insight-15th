# 회귀
- 연속적인 데이터에서 패턴을 찾아내는 통계적 방법
- 선형회귀, 비선형 회귀, 릿지 회귀, 라쏘 회귀, 다항 회귀 등
- 데이터 요약 및 예측, 시계열 모델링 및 변수 간 인과관계 발견 등

## 단순/다중 선형회귀분석
- 단순선형회귀분석
    - 하나의 독립변수만 존재
- 다중선형회귀분석
    - 두 개 이상의 독립변수 존재
- 최소제곱법을 통해 최적의 회귀선을 찾음

        - F-검정, T-검정
3. 모델 성능 평가
    - R-squared
    - 회귀 분석 후 검증 가능한 가설 검토

## 선형 회귀
### 다중선형회귀의 수식적 이해
> $y_i = \beta_0+\beta_1X_{1i}+\beta_2X_{2i}+...+\beta_kX_{ki}+\varepsilon_i$
- 상수항과 오차항 제외하면
> $y=w_0x_0+w_1x_1+\cdot\cdot\cdot+w_mx_m=\sum_{i=0}^mw_ix_i=w^Tx$
- 종속변수 y에 대한 독립변수들의 가중평균

### 다중선형회귀의 기본 가정 6가지
1. 선형성
    > $E(y|X) = \beta_1+\beta_2X_2+...+\beta_nX_n$
    - 종속변수와 설명변수 간의 관계가 선형적이어야 한다
2. 독립성
    - 각각의 설명변수가 서로 선형독립적이어야 한다
    - 선형독립적이지 않은 경우 = 다중공선성이 있다
3. 오차항의 평균은 0이다 (0에 가까워야 한다)
    > $E(\varepsilon_i|X_i) = 0$
    - 오차항 : 실제 값과 예측 값의 차이
4. 등분산성
    > $Var(\varepsilon_i|X_i) = \sigma^2$
    - 오차항의 분산이 일정해야 한다
    - 오차항의 분산이 일정하지 않은 경우 = 이분산성
5. 오차항은 자기상관되어 있지 않다
    > $Cov(\varepsilon_i, \varepsilon_j|X_i) =0$
    - 오차항의 공분산은 항상 0이다
6. 정규성
    > $\varepsilon_i \sim N(0, \sigma^2)$
    - 오차항은 정규분포를 따른다

### 회귀분석 평가 방법
1. 시각화
2. 통계지표
    1) 모델 유의성 검정
        - F-statistic, Prob(F-statistic)
        - T-statistic, Prob(T-statistic)
        - F-검정, T-검정 비교!
    2) 모델 성능 평가 
        - 결정 계수 (R-squared, $R^2$)
            - $0 ≤ R^2 ≤ 1$
            - $SST = SSE + SSR$
            - $R^2 = SSR / SST$
        - 조정된 결정 계수 (Adjusted R-squared, Adjusted $R^2$)
        - AIC, BIC(SC)



## 비선형회귀
1. 다항식 회귀 모델
    - 독립 변수에 거듭제곱 항을 추가하여 데이터의 곡선적(비선형적) 특성을 모델링
2. 지수 회귀 모델
    - 종속 변수가 지수적으로 변화하는 관계를 모델링할 때 사용
    - 종속 변수 y에 로그를 취한 후 선형회귀 적용
        - 원래의 데이터 형태로 복구하기 위해 예측 값에 지수함수 적용
3. 로그 회귀모델
    - 종속 변수와 독립 변수 간의 관계가 로그 함수를 통해 더 잘 표현될 때 사용
4. 스플라인 회귀
    - 데이터를 구간별로 나누고 각 구간에서 다른 선형 또는 비선형 함수를 적용하여 예측하는 방식
