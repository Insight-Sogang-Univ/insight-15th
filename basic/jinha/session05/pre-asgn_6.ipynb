# 분류

## 지도학습 : 회귀 vs 분류
- 회귀
    - 연속적인 숫자값 예측
    - 주어진 데이터(독립변수)를 기반으로 정답(종속변수)을 맞추는 함수
- 분류
    - 입력된 데이터를 주어진 항목들로 나누는 것
    - 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에, 새롭게 관측된 데이터에 대한 레이블을 판별하는 것
- 회귀와 분류의 차이
    - 회귀: 데이터가 연속형 변수를 예측하기 위해 사용될 때
    - 분류: 데이터가 범주형 변수를 예측하기 위해 사용될 때

## 분류 모델
1. 로지스틱 회귀
    - 이진 분류 문제를 푸는 대표적인 알고리즘
    - 샘플이 특정 클래스에 속할 확률 추정 -> 기준치에 따라 분류
    - 독립 변수의 선형 조합(선형 회귀)에 로지스틱 함수를 적용하여 출력값을 0에서 1사이로 변환해 주는 것
    - 시그모이드 함수 = 로지스틱 함수
        - 출력이 0과 1의 사이 값을 가지면서 S자 형태로 그려지는 함수
        - 출력값이 특정값 이상이면 1(True), 특정값 이하면 0(False)
        - 가중치 W : 그래프의 기울기 변화
        - 가중치 b : 그래프의 위치 변화

2. 의사결정나무
    - 조건에 따라 데이터를 분류하며, 최종적으로 데이터가 순수한 label의 집합으로 구성될 때까지 분류를 반복
    - 구성 요소
        - Root Node: 가장 상위에 위치한 노드
        - Parent Node: 주어진 노드의 상위에 있는 노드
        - Child Node: 하나의 노드로부터 분리된 2개 이상의 노드들
        - Binary Tree(이진 트리): Tree 중에서 children이 최대 2개인 tree
        - Leaf Nodes: 자식마디가 없는 노드, 모델에서 label에 해당
        - Edge: Parent node와 child node를 연결하는 연결선
        - Height: 특정 노드에서 가장 멀리 있는 leaf node까지의 경로에 있는 edge 개수
        - Depth: Root node에서 특정 노드에 도달하기 위한 edge의 수
    - CART (이진분할) 알고리즘
        - 가장 널리 사용되는 의사결정나무 알고리즘
        - 데이터셋을 임계값을 기준으로 두 child로 나누는 알고리즘
            - 임계값은 어떻게 나누는가? -> 불순도(지니 계수)가 낮아지는 방향
                - 불순도 : 분류하려는 데이터 집합에서 서로 다른 클래스(범주)가 섞여 있는 정도
        - 주요 단계
            1. 임계값 설정
            2. 불순도 감소 알고리즘
        - 실제 학습 시 고려 사항
            1. Hyperparameter 설정
            2. 시각화
            3. Prunning (가지 치기)

3. SVM (Support Vector Machine)
    - 클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘
    - 데이터를 분리하는 초평면 중에서 데이터들과 가장 거리가 먼 초평면을 선택하여 분리하는 지도 학습 기반의 이진 선형 분류 모델
    - 구성 요소
        - Support vector (서포트 벡터): 데이터 중에서 결정 경계와 가장 가까이에 있는 데이터들의 집합
        - Decision Boundary (결정 경계): 데이터 분류의 기준이 되는 경계
        - Margin (마진): 결정 경계에서 서포트 벡터까지의 거리(여유 공간)
        - Hyperplane (초평면): $n$차원 공간의 ($n-1$)차원 평면
        - Slack Variables (슬랙 변수/여유변수): 완벽한 분리가 불가능할 때 분류를 위해 허용된 오차에 대한 변수

4. KNN (K-Nearest Neighbor)
    - 데이터로부터 거리가 가까운 K개의 다른 데이터 레이블을 참조하여 분류하는 알고리즘
    - 계산 순서
        1. 데이터 준비
        2. K값 설정
            - 가장 가까운 이웃의 개수
            - 보통 홀수 개로 설정
            - 학습용 데이터 개수의 제곱근으로 설정
        3. 거리 계산
            - 새로운 데이터와 기존 모든 데이터 간의 거리 계산
        4. 가장 가까운 K개의 이웃 선택
        5. 분류하기
            - K개의 이웃 중 가장 많이 등장하는 클래스가 예측 결과
    - 별도의 모델없이 데이터만을 이용해서, 새로운 데이터가 왔을 때 즉석에서 주변 데이터들을 이용하여 분류

## 분류 평가 지표
1. 혼동 행렬
    - 분류 모델의 예측 결과를 정확한 예측과 잘못된 예측으로 구분하여 나타낸 표
    - 예측 성공 여부(T/F) + 예측값(P/N)으로 구분하여 총 4종류의 경우의 수
        - TP : 정답이 참, 참으로 예측
        - TN : 정답이 거짓, 거짓으로 예측
        - FP : 정답이 거짓, 참으로 예측
        - FN : 정답이 참, 거짓으로 예측
    - 혼동 행렬을 이용한 분류 모델 평가 지표
        - 정확도 (Accuracy): 모든 가능한 예측 중 참인 비율
        - 정밀도 (precision): 참이라고 예측한 경우 실제 참의 비율
        - 재현도 (recall): 실제로 참인 경우 중 참으로 예측하는 비율

2. F1-Score
    - 정밀도(precision)와 재현율(recall)의 조화 평균
    > $$
\text{F1 Score} = \frac{2}{\frac{1}{\text{Precision}} + \frac{1}{\text{Recall}}}
$$

    > $$
= \frac{2 \times \text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}
$$

3. ROC / AUC Curve
    - ROC Curve
        - Receiver Operating Characteristic
        - 분류가 얼마나 잘 되었는지를 모여 주는 그래프
        - 최상단으로 붙어 있는 ROC 커브일수록 더 좋은 이진분류기
    - AUC Curve
        - Area Under the Curve
        - ROC와 x축 사이의 면적 (적분값)

## 하이퍼파라미터 최적화
- 하이퍼파라미터
    - 모델 학습 과정에 반영되는 값
    - 학습 시작 전에 사용자가 직접 설정하는 변수
- 하이퍼파라미터 최적화
    - 튜닝을 거쳐 적절한 하이퍼 파라미터를 찾음으로써 모델 성능을 향상시키는 것
    - 과정
        1. Hyperparameter 탐색 범위 설정 (최적 값을 찾고 싶은 하이퍼파라미터의 범위를 설정)
        2. 평가 지표 계산 함수(성능 평가 함수) 정의(탐색하려는 Hyperparameter를 인수로 받아 평가지표 값을 계산해주는 함수를 정의)
        3. 1단계에서 샘플링한 Hyperparameter 값을 사용하여 검증 데이터로 정확도 평가
        4. 위 단계를 특정 횟수 반복하며, 정확도 결과를 보고 하이퍼파라미터의 범위 좁힘
    - 방법
        1. Grid Search
            - 정해진 범위에서 하이퍼파라미터를 모두 순회하며 가장 좋은 성능을 내는 값을 찾는 기법
        2. Random Search
            - 정해진 범위에서 하이퍼파라미터를 무작위로 탐색해 가장 좋은 성능을 내는 값을 찾는 기법
        3. Beyesian Optomization
            -  사전 정보를 바탕으로 최적 하이퍼파라미터 값을 확률적으로 추정하며 탐색하는 기법
