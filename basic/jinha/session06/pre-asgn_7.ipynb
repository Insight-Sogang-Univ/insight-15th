# 군집화를 위한 데이터 준비
## 2-1. 군집화의 주요 고려사항
1. 변수 유형 이해
    - 군집화에 사용될 피처의 종류와 특성 명확하게 이해
    - 변수 종류가 연숙형인지 명목형인지, 변수 개수가 몇 개인지에 따라 적합한 알고리즘이 달라지기 때문
2. 거리 / 유사도 정의와 측정
    - 군집화는 데이터 간의 거리 또는 유사도를 기반으로 그룹을 형성
    - 거리 측정 방법에 대한 정의가 중요
3. 차원 축소
    - 모델 성능 향상 위해 차원 축소 고려해야 함
    - 유사한 변수 통합, 불필요한 변수 제거 등

## 2-2. 변수 유형 이해
- 변수 유형별 군집화 고려사항
    1. 연속형 변수
        - 거리 측정: 유클리디안 거리, 맨하탄 거리 등
        - 적합한 알고리즘 : K-means, 계층적 군집화 등
        - 주의사항 : 스케일링 필수
    2. 명목형 변수
        - 거리 측정 : 해밍 거리(Hamming distance), 자카드 거리 등
        - 적합한 알고리즘 :  K-modes, 계층적 군집화 (특별한 거리 함수 사용)
        - 처리 방법 : 원-핫 인코딩 또는 더미 변수 변환
    3. 혼합형 변수
        - 연속형과 명목형이 섞인 데이터
        - 적합한 알고리즘 : K-prototypes, Gower distance 기반 계층적 군집화
        - 처리 방법 : 각 변수 유형에 맞는 거리 함수를 조합하여 사용

## 2-3. 거리 / 유사도 정의와 측정
- 군집화에서 데이터 포인트나 군집 간의 유사성/비유사성을 수치로 계산하는 방법
- 거리가 가까울수록 유사하고, 멀수록 다르다고 판단
1. 유클리디안 거리
    - 가장 일반적인 거리 측정 방법
    - 직선 거리 의미
    - K-means에서 기본적으로 사용
    > $ d(\mathbf{x}, \mathbf{y}) = \sqrt{\sum_{i=1}^n (x_i - y_i)^2} $
2. 맨하탄 거리
    - 각 차원의 차의의 절댓값의 합
    - 이상치에 덜 민감
    > $ d(\mathbf{x}, \mathbf{y}) = \sum_{i=1}^n |x_i - y_i| $
3. 코사인 유사도
    - 벡터 간의 각도를 측정
    - 텍스트 데이터나 고차원 데이터에 효과적
    > $ \text{similarity} = \frac{\mathbf{x} \cdot \mathbf{y}}{||\mathbf{x}|| \times ||\mathbf{y}||} $
- 6가지 연결 방법
    - 계층적 군집화에서 두 군집 간의 거리를 계산하는 다양한 방식
    - 군집 간 거리 측정 기준이 달라 서로 다른 군집화 결과를 만들어 냄
    1. Single linkage(단일 연결법/최단 연결법)
        - 군집 A와 B 사이의 가장 가까운 두 점 간 거리
        - 고립된 군집을 찾는데 중점
        - 이상치에 취약
    2. Complete linkage(완전 연결법/최장 연걸법)
        - 군집 A와 B 사이의 가장 먼 두 점 간 거리
        - 군집들의 내부 응집성에 중점을 둔 방법
        - 이상치에 민감
    3. Average linkage(평균 연결법)
        - 군집 A와 B의 모든 점 쌍 사이 거리의 평균
        - 모든 항목에 대한 거리 평균을 구함
        - 이상치에 덜 민감
        - 계산량 불필요하게 많아질 수 있다는 단점
    4. Centroid Method(중심 연결법)
        - 각 군집의 중심점 간의 거리
        - 두 군집이 결합될 때 새로운 군집의 평균은 가중평균을 통해 구함
        - 시간 오래 걸림
        - 덴드로그램에서 거리 축이 위에서 아래로 줄어드는 invision이 발생할 가능성이 있음
    5. Median(중앙 연결법)
        - 두 군집의 중앙값 사이의 거리
        - 극단값에 영향 덜 받음
        - 기하학적 구조 파악하기 어렵다는 단점
    6. Ward's Procedure(Ward 연결법)
        - 군집의 병합 후 군집 내 SSE(오차제곱합)의 증가분이 최소인 것을 선택

## 2-4. 차원 축소
### 차원의 저주
- 피처 공간이 너무 커서 알고리즘이 데이터를 효과적/효율적으로 훈련할 수 없는 현상
- 고차원일수록 모든 점들이 비슷하게 멀어져 거리 기반 알고리즘의 분별력이 감소 -> 차원 축소가 필요
### 차원 축소
- 차원 : 변수의 개수
- 차원 축소 : 변수의 개수를 줄이는 것
- 목표
    - 고차원 데이터를 저차원 공간에 투영하면서도 핵심 정보는 가능한 지켜내도록 함
    - 노이즈를 줄여 패턴 학습에 유리하게 함
- 유형
    - 선형 투영 : 고차원 공간에서 저차원 공간으로 선형적으로 데이터를 투영
    - 매니폴드 학습(비선형 차원 축소) : 유클리드 거리가 아닌 데이터 포인트들 사이의 곡선 거리를 고려하며 학습
- PCA
    - 고차원 데이터를 저차원으로 축소하는 선형 투영 기법
    - 다차원 데이터에서 가장 정보량(분산)이 큰 방향을 찾고, 그 방향을 기준으로 데이터를 투영하여 차원 축소
    - 상관관계가 높은 피처들 결합해 더 적은 수의 피처로 데이터 표현
    - 새로 파생된 성분 : 주성분
    - 주의사항 
        - 사용 전 데이터 표준화 수행
        - 결측값 있으면 사용 불가이므로 결측값 처리
        - 직관적 해석이 어려움
        - 차원축소 과정에서 일부 정보 손실 가능성


# 3. 군집화 알고리즘
## 3-1. 계층적 군집화
- 데이터 간의 유사성을 기반으로 트리 구조를 형성하며, 상향식 또는 하향식 방식으로 군집을 형성해 나가는 방법
- 특징
    - 군집의 개수를 사전에 설정 X, 클러스터링이 종료된 후 원하는 군집 개수 선택 가능
    - 트리 구조를 어떻게 구성하냐에 따라 응집형과 분리형으로 구분
        - 상향식 = 응집형 계층적 군집화
        - 하향식 = 분리형 계층적 군집화
- 응집형 계층적 군집화
    - 모든 데이터 포인트를 각각 하나의 클러스터로 간주하고 시작, 점차 가까운 클러스터들 병합
    - '가장 가까운 클러스터'를 어떻게 정의하느냐에 따라 다양하게 나뉨
    - 단일 연결(가장 가까운 두 샘플의 거리)과 완전 연결(가장 먼 두 샘플의 거리) 방식이 대표적

## 3-2. k-means
- 데이터를 정해진 개수(K)의 그룹으로 나누되, 각 그룹의 중심점과의 거리가 가장 가까운 데이터끼리 묶는 군집화 알고리즘
- 가장 일반적으로 사용됨
- 단계
    1. 데이터 표본들 중에서 랜덤하게 $k$개의 중심점(centroid)을 초기 클러스터 중심으로 선택
    2. 각 표본들을 가장 가까운 중심점 $\mu^{(j)}, j \in \{ 1, …, k \}$에 할당
    3. 각 클러스터에 할당된 표본들의 데이터 평균을 계산하여 중심점(centroid)을 이동
    4. 클러스터 할당이 변하지 않거나 사용자가 지정한 허용 오차 또는 최대 반복 횟수(max_iter)에 도달할 때까지 2, 3번을 반복
- 장점
    - 직관적, 구현 쉬움
    - 대용량 데이터 적용 가능
- 단점
    - 초기 CENTROID 값에 민감함
    - 군집 수(K) 결정이 어려움
    - 이상치에 민감

## 3-3. DBSCAN
-  밀도가 높은 지역의 데이터를 하나의 군집으로 묶고, 밀도 기준을 만족하지 못하는 점은 군집에 포함시키지 않는 군집화 알고리즘
- 특징
    - 명시적으로 이상치를 지정하여 클러스터링에서 제외
    - 이상치로부터 비교적 자유로움
    - 어떤 데이터가 여러 군집의 특정 거리 내에 있으면 가장 밀도가 높은 군집과 그룹화
- 단계
    1. 샘플 분류
        - 핵심 샘플 (Core Point)
            - 특정 반경 ε 안에 있는 이웃 샘플이 지정된 개수(MinPts) 이상인 경우
        - 경계 샘플 (Border Point)
            - ε 이내에 MinPts보다 이웃이 적지만 다른 핵심 샘플의 반경 ε 안에 있는 경우
        - 잡음 샘플 (Noise Point)
            - 위 두 조건에 해당하지 않는 모든 샘플
    2. 클러스터 생성
    3. 경계 샘플 할당
- 하이퍼 파라미터
    - eps : 두 포인트가 이웃이 되기 위한 포인트 사이의 최대 거리
    - min_samples : 한 포인트가 군집이 되기 위한 eps 거리 내 최소 포인트 개수

## 3-4. Gaussian Mixture Model (GMM)
- 데이터가 여러 다른 모양의 가우시안 분포로 구성되었다고 가정하고, 각 분포를 클러스터로 인식하는 군집화 방법
- 기본 가정
    - 관측된 데이터는 특정 가우시안 확률 분포에 의해 생성됨
    - 전체 데이터셋에는 여러 개의 다변량 가우시안 분포가 섞여 있음
    - 개별 데이터는 우도에 따라 K개의 가우시안 분포 중 하나에 속함
- 진행 과정
    1. 주어진 전체 데이터셋의 분포 확인
    2. 전체 데이터 셋은 서로 다른 정규 분포 형태의 확률 분포 곡선으로 구성되어 있다고 가정
    3. 전체 데이터 셋을 구성하는 여러 개의 정규분포 곡선 추출, 개별 데이터가 이 중 어떤 정규분포에 속하는지 결정
        - 이때 각각의 분포가 하나의 군집이 됨
- 장점
    - K-means보다 유연하게 다양한 데이터 세트에 잘 적용될 수 있음
    - k-means가 잘 작동하지 않는 타원형 분포나, 중첩된 군집 구조에서도 좋은 성능을 낼 수 있음
- 단점
    - 군집화를 위한 수행시간이 오래 걸림
    - 가정한 분포(가우시안 분포)에 맞지 않은 데이터일 경우 계산 복잡도가 높아지고 성능이 저하될 수 있음

# 4. 군집화 평가 방법
## 4-1. 외부 평가와 내부 평가
1. 외부 평가
    - 데이터에 정답 레이블 존재하는 경우 사용
    - 군집화 결과와 실제 레이블을 비교하여 얼마나 유사한지 평가
2. 내부 평가
    - 정답 레이블 없는 경우 자용
    - 군집 내부 응집도와 군집 간 분리도를 기반으로 군집의 품질 정량적으로 측정
    - 알고리즘 별 내부 평가 지표 및 목표
        - K-means 알고리즘 : SSE(Sum of Squared Errors)를 최소화
        - GMM : 로그 우도(모델이 주어진 데이터를 얼마나 잘 설명하는지 나타내는 지표)를 최대화하기

## 4-2. 실루엣 계수
- 클러스터 내 샘플들이 얼마나 조밀하게 모여 있는지 측정하는 도구
> $s^{(i)} = \frac{b^{(i)} - a^{(i)}}{\max \left\{ b^{(i)}, a^{(i)} \right\}}$
- 해석
    - 분리도 > 응집도 : 실루엣 계수 1에 가까워짐
    - 분리도 = 응집도 : 실루엣 계수 0
    - 분리도 < 응집도 : 실루엣 계수 -1에 가까워짐

## 4-3. Dunn Index
- 클러스터 간 최소 거리(분리도)와 클러스터 내 최대 거리(응집도)의 비율을 계산해 클러스터링의 품질 평가하는 지표
> $ \text{Dunn Index} = \frac{\min\limits_{i \ne j} \; d(C_i, C_j)}{\max\limits_{1 \le k \le K} \; \delta(C_k)}$
- 해석
    - 값이 클수록 좋은 군집화
    - 일반적으로 0 이상의 값, 무한대까지 가능
    - 실루엣 계수와 종합적 평가 권장
- 한계
    - 군집 수가 많아질수록 계산 비용 증가
    - 극단적 값에 큰 영향 받음
    - 이상치에 민감하게 반응
