# 1. 연관 분석
## 1-1. 연관 분석의 정의
- 상품이나 서비스를 구매하는 등 일련의 거래나 사건 안에 존재하는 항목 간의 일정한 연관 규칙을 발견하는 분석
- 비지도 학습에 해당

## 1-2. 연관규칙 분석
- 형태
    - if(조건절)-then(결과절)의 형식
    - A -> B로 표현
    - 아이템 집합 : 조건절 또는 결과절을 구성하는 아이템들의 집합
    - 상관관계를 파악할 뿐, 인과관계는 알 수 없음

# 2. 주요 개념
## 2-1. 적절한 연관규칙의 조건
- 두 품목을 함께 구매한 경우의 수가 일정 수준 이상이어야 한다 (일정 이상의 지지도)
    - 지지도
        - 전체 거래항목 중 품목 A와 품목 B를 동시에 포함하는 거래의 비율
        - 조건절이 발생하는 확률
- 품목 A를 포함하는 거래 중 품목 B을 구입하는 경우의 수가 일정 수준 이상이어야 한다 (일정 이상의 신뢰도)
    - 신뢰도
        -  품목 A를 포함하는 거래 수 중 품목 A와 품목 B를 동시에 포함하는 거래의 비율

## 2-2. 연관 분석의 평가 척도
- 향상도
    - A와 B가 우연히 함께 등장할 확률과 실제 함께 등장할 확률을 비교
    - 향상도 > 1 : 우연적 기회보다 높은 확률 (두 품목이 서로 양의 상관관계)

=> 임의의 규칙 1이 규칙 2보다 효과적인 규칙이려면, 지지도, 신뢰도, 향상도 세 지표가 모두 커야 함

# 3. 연관 분석 알고리즘
## 3-1. Apriori 알고리즘
- 상위 조합에서부터 차례로 스캔 -> 특정 조합이 자주 발생하지 않는다면 이의 결과물로 탄생한 후속 조합까지 모두 후보에서 배제
- 장점
    - 하나의 조합만 검사하고도 파생된 다른 조합까지 후보에서 배제
    - 시간과 연산량 단축 가능
- 프로세스
    1. 단일 항목 집합 생성
        - 모든 개별 상품에 대해 지지도 계산
        - 최소 지지도 넘는 항목들만 다음 단계
    2. 2개 항목 집합 생성
        - 1단계에서 선별된 상품들로 2개 항목 조합 생성
        - 조합에 대해 지지도 계산, 최소 지지도 넘지 않는 조합 제거
    3. 더 큰 항목 집합 생성
    4. 최종 빈발 항목 집합 선정
        - 마지막으로 남은 빈발 항목 집합들을 바탕으로 연관 규칙 생성
        - 각 규칙에 대해 지지도, 신뢰도, 향상도 계산하여 가장 유용한 규칙 선택
## 3-2. FP Growth 알고리즘
- Apriori의 연산 속도 문제 해결
- FP-Tree 구조 이용하여 Apriori를 효과적으로 구현
- 모든 거래를 확인하여 각 아이템마다 지지도 계산, 최소 지지도 이상의 아이템만 선택
- 프로세스
    1. 데이터셋 확인
    2. 빈도순 재정렬
    3. FP-트리 생성
    4. 빈발 항목 집합 추출
    5. 연관 규칙 도출
- 장점
    - 속도가 빠름
    - 후보 item set 생성할 필요 없이 Tree만 구성하면 됨
- 단점
    - 대용량 데이터셋에서는 메모리 비효율적
    - 설계 어려움, FP-Tree 만들어져야 지지도 계산 가능



# 경영 기초
## 1. 문제 정의와 구조화
### 1-1. 문제 정의
- 문제 : '현재의 상태'와 '바라는 상태' 사이의 격차
- '의사결정'을 내려야 하는 구체적인 질문으로 치환되었을 때 비로소 문제가 된다
- SMART 원칙
    - S : Specipic
    - M : Measurable
    - A : Action-oriented
    - R : Relevant
    - T : Time-bound

### 1-2. 문제 구조화
- MECE
    - 문제를 쪼갤 때 지켜야 할 철칙
    - Mutually Exclusive (상호배타적) : 겹치는 부분이 없어야 한다
    - Colloectively Exhaustive (전체포괄적) : 누락된 부분이 없어야 한다
    - 좋은 MECE를 위한 규정들
        1. 논리적 일관성
        2. 통찰력
        3. 상호 배타성

## 2. 지표
### 2-1. 지표
- 방향이나 목적, 기준 따위를 나타내는 표지
- 측정하고 싶은 것을 숫자로 표현한 것
- 대개 분모 대비 분자로 표현

### 2-2. 지표의 정의
- 절대적인 수치를 구할 수 없다면?
- Proxy Metric을 설정
    - 프로덕트마다 나름의 가정과 정의, 논의가 필요한 지표
    - 대체물(Proxy)를 정하여 측정
    - 정답이 별도로 존재하지 않음

### 2-3. 북극성 지표 & L1/L2 Metrics
- 북극성 지표 : 기업이 가장 중요하게 생각하는 지표
- L1/L2 지표
    - L1 : 전사적인, 고수준의 성과 지표 (전략 관점)
    - L2 : L1 지표의 세부 사항을 측정하는 지표 (전술 관점)
- KPI
    - Key performance Indicator (핵심성과지표)

### 2-4. AARRR (해적 지표)
- 사용자 운영 과정에서 자주 활용하는 데이터 분석 모델
- 사용자 증가를 위한 5가지 로드맵으로 구성
1. Acquistion
    - 사용자 획득
    - 사용자가 여러 채널을 통해 앱 혹은 웹사이트에 방문하는 것을 의미
    - 조회수, 가입 사용자 수 등 지표를 활용해 측정
2. Activation
    - 사용자 활성화
    - 사용자가 앱 또는 웹사이트에 첫 방문하여 주요 기능을 직접 사용하는 것을 의미
    - 제품의 성격에 따라 다양한 기준으로 측정
    - 4단계로 구성
        - Sign-up: 유저가 제품에 들어온 시점
        - Setup Moment: 유저가 제품의 핵심가치를 경험할 준비 완료
        - A-HA! Moment: 유저가 핵심가치를 처음 경험한 순간
        - Habit Moment: 유저가 제품 사용에 대한 습관 형성
3. Revenue
    - 수익 단계
    - 사용자가 제품 혹은 서비스를 유료 구매하기로 결정한 단계
    - 제품 성격에 따라 측정 지표 달라딤
    - 고객의 평생 가치 향상, 고객 유치 비용 감소 필요
4. Referral
    - 사용자가 상품 혹은 서비스에 만족해 타인에게 이를 추천
    - 측정 지표 : ‘순 추천고객 지수(NPS)’, ‘바이럴 지수(Viral Coefficient)’

## 3. 지표 분석 방법
### 3-1. 코호트 (Cohort) 분석
- 코호트
    - 정의된 시간 범위 내에서 공통된 특성이나 경험을 공유하는 사용자 집단
    - 특정 기간에 특정 특성을 공유하는 사용자 집단
- 분석 전에 데이트 세트의 데이터를 관련 그룹으로 나누는 일종의 행동 분석
- 시간(날짜)에 기반하는 사용자 세그먼트를 분류한다고 이해
- 주로 사용자 유지와 이탈을 체크하는 데 많이 사용

### 3-2. 퍼널 (Funnel) 분석
- 유저가 접속해서 떠날 때까지 유저의 행동을 추적
- 개선 방법 예시
    1. 유입률(Acquisition) 높이기
    2. 전환율(Activation) 높이기
    3. 퍼널 단계 없애기
    4. 퍼널 순서 바꾸기
    5. 퍼널 단계 추가하기
    6. 코호트로 나누어 보기
