### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.1 자료의 척도 </span>
#### 1. 자료의 **개념** : 설명과 해석을 위해 수집/분석/요약되는 사실과 수치
#### 2. 자료의 **구분**
- **질적 자료**
    > 수치로 표현되어도 연산의 의미 X <br>

    - **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">명목 척도</span>** : 레이블, 이름으로 요소의 속성을 식별
        - ex : 펀드유형(국내주식:1, 국내주식:2, 채권: 3), 성별(남:0, 여:1)
    - **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">서열 척도</span>** : 명목형의 분류 속성 + 항목 간의 순서/순위를 나타내는 특성
        - ex : 만족도(좋음, 보통, 나쁨), 직급체계(사원-대리-과장-차장-부장)
- **양적 자료** 
    > 수치로 표현 & 연산의 의미 O <br>

    - **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">구간(등간)척도</span>** : 순서O, 간격이 일정해 연산(덧셈, 평균 등) 가능 / 곱셈, 나눗셈은 의미 X
        - ex : 온도(섭씨, 화씨), IQ, 시간
    - **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">비율 척도</span>** : 절대영점, 임의의 측정단위로 구성, 동간성이 유지됨 / 유의미한 0을 가짐 (절대적인 부재)
        - ex : 거리, 높이, 무게, 시간

- **구분하기**
    1. **수치형 자료 구분하기**
        - 덧셈 가능 유무 : 불가능하면 명목척도 <-> 서열척도
        - 0값의 의미 유무 : 유의미하면 비율척도 <-> 구간척도
    2. **척도 V.S. 종류**
        - 자료의 종류 : 데이터의 본질적인 성격을 구분 (범주:명목,순서/ 수치:이산,연속)
        - 자료의 척도 : 데이터를 측정하는 방식의 규칙을 구분 (질적:명목,서열/질적:구간,비율)

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 2. 기술통계 </span>
> 데이터의 특성/패턴을 있는 그대로 설명하고 이해하는 데 중점을 둔 통계 방법
> **3가지의 주요한 도구**를 가짐

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.1 도구1 : 중심화 경향 </span>
#### **1. 평균**
- 개념 : 데이터 값의 총합을 데이터의 개수로 나눈 값. 가장 대표적인 중심 경향 지표
- `series.mean()`
- 특징
  - 이상치에 민감
  - 절사평균 : 데이터의 양 끝을 일정 비율 잘라낸 후 평균을 계산
#### **2. 중앙값**
- 개념 : 데이터를 크기 순서대로 정렬했을 때, 한 가운데에 위치하는 값 (짝수개일 경우, 가운데 값 2개의 평균)
- `series.median()`
- 특징
  - 평균에 비해 이상치에 덜 민감, 평균의 대안적 지표
#### **3. 최빈값**
- 개념 : 데이터에서 가장 자주 나타나는 값
- `series.mode()`
- 특징
  - 주로 이산형 데이터에서 활용됨, 데이터 분포 이해에 도움

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.2 도구2 : 자료의 산포  </span>
> 데이터가 퍼져있는 정도를 설명하는 기술 통계량
#### **1. 범위(range)**
- 개념 : 데이터의 최소값 - 최대값의 차이를 나타내는 지표 -> 데이터의 전체 변동범위 파악
- `series.max()-series.min()`
- 특징
  - 양쪽 극단값을 활용 -> 이상치에 매우 민감
#### **2. 분산 (variance)**
- 개념 : 데이터 값들이 평균으로부터 얼마나 퍼져있는지를 설명하는 지표
- `series.var()`
- 의미 해석 : 분산값이 크다 = 데이터가 많이 퍼져있다
#### **3. 표준편차**
- 개념 : 분산에 제곱근을 취한 값
- `series.std()`
- 제곱근을 취하는 이유
  - 분산 단위는 기존 관측값 단위의 제곱임. -> 대상 데이터의 단위와 일치시키기 위함 (직관성을 위해)
#### **4. 사분위수**
- 개념 : 데이터의 표본을 4개의 동일한 크기로 나눈 값
    ```python
    series.quantile(0,25) #25 per (Q1, 하위 사분위수)
    series.quantile(0,50) #50 per (Q2, 중앙값)
    series.quantile(0,75) #75 per (Q3, 상위 사분위수)

    np.percentile(series,75) #넘파이로도 실행 가능
    ```
#### **5. 사분위간 변위 (IQR)**
- 개념 : Q3에서 Q1을 제외한 범위 -> 데이터의 중앙 50%가 퍼져있는 범위
- 특징
  - 이상치 탐지 시 유용
  - Tukey Fence 방식에서, 아래쪽 이상치는 `Q1 - 1.5*IQR` 보다 작은 값으로 / 위쪽 이상치는 `Q3 + 1.5*IQR` 보다 큰 값으로 정의
  - boxplot에서 박스로 나타나는 부분에 해당
#### **6. 변동계수**
- 개념 : 데이터의 상대적인 변동성을 나타내는 통계적 지표
- 공식
    $$
    변동계수 = (표준편차/평균) * 100
    $$
### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.3 도구3 : 자료의 정리 및 시각화 </span>
> 지난 시간에 다뤘던 방법들 : 히스토그램, 박스플롯, 파이차트 등

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.4 첨도 & 왜도 </span>
#### **1. 첨도**
- 개념 : 확률분포의 꼬리가 두꺼운 정도를 나타내는 척도
- `series.kurt`
#### **2. 왜도**
- 개념 : 실수 값 확률 변수의 비대칭성을 나타내는 지표 -> 비대칭도
- `series.skew`
- 특징 : 양/음수 모두 가능, 정의되지 않을 수도 있음
- 해석
  - 왜도 > 0 : 확률밀도함수의 오른쪽 꼬리가 길다 -> 자료가 왼쪽에 더 많이 분포돼 있다
  - 왜도 < 0 : 확률밀도함수의 왼쪽 꼬리가 길다 -> 자료가 오른쪽에 더 많이 분포돼 있다
  - 왜도 = 0 : 평균값 = 중앙값

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 3. 확률과 분포 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.1 추론통계학을 향한 질문들 </span>
#### 1. 표본의 불확실성
- 운이 나빠 이상한 표본만 뽑힌 건 아닌가? => **표본분포**로 확인
#### 2. 모수 추정
- 그 표본의 통계량이 정말로 모두를 잘 반영하는가? => **점추정**으로 확인
#### 3. 추정의 현실성
- 점추정 값이 모수와 정확히 일치하는 건 비현실적이다 => **구간추정**으로 확인
#### 4. 구간추정의 오류 관리
- 구간 추정 시의 구간이 틀릴 가능성도 있지 않나? => **신뢰수준**으로 확인
#### 5. 적극적 주장 검증
- 특정 주장이 맞는지 직접 판단을 내릴 수는 없나? => **가설검정**으로 판단
#### 6. 가설검증의 오류 관리
- 내 결론이 틀렸을 가능성도 있지 않나? => **유의수준**으로 확인

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.2 확률분포 </span>
#### **1. 확률 변수**
1. 개념 : 임의의 세계에서 어떤 특정 사건이 발생할 때 **나타날 수 있는 모든 경우의 수를 특정 값으로 대응**시키는 것

#### **2. 확률 분포**
1. 개념 : 확률변수가 특정한 값을 가질 확률
2. 종류 (확률변수 속성에 따른 범주)
    - **이산확률분포**
      - 확률변수가 **이산적인 값**을 갖는 경우 (즉, 유한하거나 / 셀 수 있거나 / 특정값을 선택할 수 있는 경우)
      - 종류 : 베르누이분포. 이항분포, 기하분포, 음이항분포, 초기하분포, 포아송분포
    - **연속확률분포**
      - 확률변수가 **연속적인 값**을 갖는 경우 (즉, 무한한 범위를 갖거나 / 특정값을 선택할 수 없는 경우)
      - 종류 : 정규분포, T분포, F분포, 감마분포, 지수분포, 카이제곱분포, 균등분포

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.3 분포 모형과 활용 </span>
#### **1. 정규분포**
1. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">정규분포</span>**
   - 형태 : 종 모양, 평균 기준 대칭, 표준편차가 클수록 곡선은 평평해짐 (=데이터가 넓게 퍼져있다)
   - 매개변수 : 평균(μ), 표준편차(σ)
   - 특징
     - 가장 대표적인 (연속)확률분포
     - 많은 자연현상 & 사회현상이 정규분포를 따름
     - 모집단의 분포가 정규분포 -> 표본평균의 분포도 정규분포
   - **큰 수의 법칙** : 표본의 크기가 커질 수록 표본 평균이 모평균에 근사 => 통계의 정확도를 위해 데이터는 다다익선
   - **중심극한정리** : 모집단의 분포 형태와 관계없이, 표본의 크기(n)이 적당히 크다면 표본 평균들의 분포가 정규분포에 수렴한다
     - 적당히 큰 n >= 30
     - 모집단의 분포를 몰라도, 표본의 크기가 충분히 클 경우 -> "표본평균의 분포는 근사적으로 정규분포를 따른다"라고 간주 가능!

2. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">Z분포</span>**(표준정규분포)
   - 개념 : 평균이 0이고 표준편차가 1이 되도록 맞춘 표준적인 정규분포

#### **2. 표본 분포**
> **정규분포를 따르는 모집단**에서 추출한 표본(sample) 통계량들이 따르는 분포 = **표본 분포**<br><br>
> 모집단 자체가 갖는 분포 X<br>
> 가설 검정, 구간 추정 등 **통계적 추론**을 위해 만들어진 도구<br>

1. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">T분포</span>**
   - 형태 : 종 모양, 정규분포에 비해 두꺼운 꼬리, 평균 기준 대칭, 자유도가 커질 수록 표준정규분포 N(0,1)에 수렴
   - 매개변수 : 자유도(df)
     - 개념 : 얼마나 많은 정보가 독립적으로 변할 수 있는가를 나타내는 숫자
     - T분포의 자유도는 표본 크기에 비례 -> 자유도가 높을 수록 정규분포에 가까워지고 정확도가 올라감
     - 너무 많은 자유도를 가질 경우, 오버피팅 문제 발생하므로 적절한 자유도 선택이 중요
   - 특징
     - 목적 : 표본으로 모평균(혹은 모평균 간 차이)를 추정하기 위해
     - **표본 크기가 작을 때**(표본 크기 <30), 정규분포보다 정확한 결과 제공

2. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">카이제곱분포</span>**
   - 형태 : 0에서 시작, 오른쪽으로 긴 꼬리(비대칭 분포)
   - 매개변수 : 자유도(df)
   - 특징
     - 목적 : 독립성 검정, 적합도 검정 등
     - 독립적인 표준 정규 분포를 따르는 변수들의 제곱 합 -> **표본분산의 분포**와 관련된 지표

3. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">F분포</span>**
   - 형태 : 0에서 시작, 오른쪽으로 긴 꼬리(비대칭 분포)
   - 매개변수 : 자유도 (df1,df2)
   - 특징
     - 목적 : **두 집단 간 분산 차이** 비교
     - 두개의 독립적인 카이제곱분포의 비율

#### **3. 분포 모형의 중요성**
1. 분포 모형은 가설 검정을 위한 **검정통계량 산출**에 사용됨
2. 분포 모형을 통해 표본추출의 과정에서 **인위적인 조작**이 있었는지 확인 가능
   - 인위적 조작 : **random sampling**이 이뤄지지 않은 것 (=모집단으로부터 랜덤하게 표본이 추출되지 않음)

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 4. 추론통계 - 추정 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.1 추론통계의 개념 </span>
#### **1. 개념**
- 수집한 데이터를 기반으로, 전체정보를 추론하고 결론을 도출하는 방법을 다루는 분야 (통계적 가설검정, 신뢰구간 등)
  
#### **2. 추정**
> 모집단을 대표할 수 있는 **일부를 표본으로 추출**해, 분석한 통계량을 이용해 모집단의 **모수를 예측하**는 과정<br><br>
> **추정량** : 표본정보에 의존하는 확률변수, 모수 추정에 사용되는 표본통계량 <br>
> **추정치** : 표본 추출 후 표본자료를 이용해 계산한 통계량의 특정 값 <br>

- **점추정** : 추정하고자 하는 모수를 하나의 수치로 추정
- **구간 추정** : 추정하고자 하는 모수가 존재할 것이라 예상되는 구간을 정해 추정
- **모수**(*parameter*) : 모집단의 특성에 관계되는 평균, 비율, 표준편차 등 관심 대상이 되는 미지의 참값
- **통계량**(*statistic*) : 모집단으로부터 추출된 표본에 기초해 계산되는 수치함수 (표본에서 계산되는 평균, 중앙값, 사분위수, 분산, 표준편차 etc)

#### **3. 검정**
> 표본을 추출해 얻은 표본통계량으로, 모집단 모수에 대한 새로운 예상, 가설 등이 옮다고 판단할 수 있는지 평가하는 과정

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.2 추정의 종류 </span>
#### **1. 점추정**
- 어떤 모수를 표본에 기초해 계산된 하나의 값으로 추정하는 것
#### **2. 구간추정**
- 모수의 참 값이 포함되리라고 기대하는 추정치를 일정 범위, 신뢰도를 갖고 나타내는 것
- 예시
  - "구간추정 95%" : 동일한 조사를 100번 반복했을 때, 그 중 95개의 신뢰구간이 실제 모수를 포함할 것으로 기대된다

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 5. 추론통계 - 검정 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 5.1 가설 검정 </span>
#### **1. 가설 검정의 개념**
1. 정의
   - 모집단의 실제 값이 얼마가 된다는 주장과 관련해, 표본의 정보를 사용해 **가설의 합당성 여부를 판정**하는 과정
2. 과정
   - `가설 설정 - 유의 수준 설정 - 검정통계량 산출 - 가설 기각/채탁 판단`

#### **2. 가설 검정 절차**
1. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">가설 설정</span>**
   - 귀무가설 H0, 대립가설 H1 설정
     - **귀무 가설** : (현재 통용되고 있는) 기각하고자 하는 명제 ("~는 차이가 없다")
       - 지지될 때 위험성이 비교적 약함
     - **대립 가설** : (새롭게) 주장하고자 하는 명제 ("~는 차이가 존재한다")
       - 지지될 때 위험성이 비교적 강함

2. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">유의 수준 설정</span>**
   - **유의 수준**
     - 결과를 판단할 때 **허용할 수 있는 최대치의 오류** (주로 5%(0.05)) -> '*결정이 얼마나 신뢰될 만한지*' 판단하는 기준
     - = **1종 오류의 확률**을 의미
       - **1종 오류(𝛼)** : 귀무가설이 실제로 참이었는데 기각할 확률 -> 더 치명적
       - **2종 오류(𝛽)** : 귀무가설이 실제로 거짓인데 기각하지 못할 확률

3. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">검정통계량 산출</span>**
    > **검정통계량** : 가설 검정의 결과를 판단하는 데 사용되는 값
    > 데이터의 분포, 특징에 따라 어떤 통계량(T값, F값 등...)을 사용할지 결정함 <br>

   - **Z값**
     - 정의 : 표준정규분포를 사용해 얻은 값
       - 특정 데이터가 평균으로부터 얼마나 떨여져있는지를 표준편차 단위로 나타냄
     - 계산 (X=관측값, μ=평균, σ=표준편차)
       -    $$ Z=(X−μ)/σ $$
     - 사용 경우 : 큰 표본 크기(보통 >=30)에 대한 가설 검정
     - 해석
       - 큰 Z값의 절댓값 -> 데이터가 평균으로부터 멀리 떨어져있다 (or 비정상/이상치)
       - 임계점 찾기
         - Z = 1.96일 때 유의수준 0.05(=신뢰수준 95%)
         - Z = 2.58일 때 유의수준 0.01(=신뢰수준 99%)
         - Z값이 임계점보다 크다면 귀무가설 기각 (신뢰구간= 1-유의 수준(α))
  
   - **T값**
     - 정의: T분포를 사용하여 얻은 값으로, 표본이 작을 때의 불확실성을 반영한 지표
     - 개념: 관측된 표본 평균이 가설로 설정한 모집단 평균과 얼마나 떨어져 있는지 나타냄
     - 계산 ($\bar{X}$=표본 평균, $\mu$=비교 대상 값, $s$=표본 표준편차, $n$=표본 크기)
       - $$T = \frac{\bar{X} - \mu}{s / \sqrt{n}}$$
     - 사용 경우
       - 모집단의 평균에 대한 가설 검정
       - 두 그룹 간의 평균 차이를 검정
     - 해석
       - 큰 T값의 절댓값: 표본 평균이 모집단 평균과 유의미하게 차이 날 가능성이 높음을 의미
       - 임계점 찾기: T분포는 **자유도**(df)에 따라 모양이 변하므로, T분포표를 참조하여 임계점을 결정
       - 표본 크기가 커질수록 T값은 Z값과 비슷해짐 (Z값에 수렴)

   - **F값**
     - 정의: F분포를 사용하여 얻은 값으로, 두 개 이상의 분산(흩어진 정도)을 비교하는 지표
     - 개념: 집단 간의 차이(분산)가 집단 내의 우연한 차이(분산)보다 얼마나 큰지를 나타냄
     - 계산
       - $$F = \frac{\text{표본 평균 간 퍼진 정도}}{\text{표본 내에서 퍼진 정도}}$$
     - 사용 경우
       - 두 개 이상의 그룹 간 평균 차이를 검정 (ANOVA)
     - 해석
       - F값이 1에 가깝다면: 그룹 간의 차이가 별로 없음

4. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">가설 기각/채택 판단</span>**
   - **유의확률**(p-value)
     - 가설검정에 쓰이는 또 다른 척도 
     - 검정통계량에서 유도 가능
   - 판단
     - `유의 확률(p) < 유의 수준(α)` -> 귀무가설 기각
     - `유의 확률(p) >= 유의 수준(α)` -> 귀무가설 기각 X

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;">5.2 데이터 자료형에 따른 검정 방법 </span>

#### 1. **[수치형-수치형] (피어슨)상관분석**
> **두 수치형 변수 간 피어슨 상관계수가 얼마나 유의미한지**를 검정

1. **시각화**로 알아보기
   - `plt.scatter(df['colx'], df['coly'])` : 산점도를 통해 눈으로 직접 관계를 확인합니다.
   - `sns.pairplot(df)` : 모든 수치형 변수 간의 관계를 한꺼번에 확인합니다. (데이터가 많으면 시간이 오래 걸림)

2. **주요 개념**
   - **공분산** (Covariance): 두 변수가 함께 변하는 방향과 크기를 나타냄
     - $Cov(X,Y) > 0$ : 양의 관계 (함께 커짐)
     - $Cov(X,Y) < 0$ : 음의 관계 (하나가 커지면 하나는 작아짐)
     - $Cov(X,Y) = 0$ : 선형적 관계 없음 (하지만 비선형 관계는 있을 수도 있음)
   - **상관계수** (Correlation): 공분산을 표준화하여 -1과 1 사이의 값으로 나타낸 지표
     - 상관관계 != 인과관계(원인과 결과) <br><br>
        | 상관관계 | 인과관계 |
        | --- | --- |
        | 두 변수가 수치적으로 연관되어있는 관계 | 한 변수가 다른 변수를 발생시키는 관계 |

3. **코드 예시**
    ```python
    import scipy.stats as spst
    spst.pearsonr(df['col1'], df['col2']) # 결과: (상관계수 r, p-value)의 튜플 형태
    ```
4. **해석 방법** (절대적 기준은 아님)
   - $|r| < 0.3$: 상관관계 약함
   - $|r| < 0.5$: 상관관계 중간
   - $|r| > 0.5$: 상관관계 강함

#### 2. **[범주형-수치형] F-test(ANOVA), T-test**
> **범주에 따라 수치 데이터의 평균에 차이**가 있는지 분석

1. **시각화** (sns.barplot)
   - 코드 예시
        ```python
        sns.barplot(x = 'col1', y = 'col2', data = df) # col1의 범주 갯수는 많아도 괜찮음
        plt.show()
        ```
   - 막대 위의 검은 선은 각 평균의 95% 신뢰구간을 나타냄
   - 해석: 두 막대의 평균 차이가 크고 신뢰구간이 거의 겹치지 않으면 대립가설을 지지(차이가 있음)할 가능성이 높습니다.

2. **T-test** (범주가 2개일 때)
   - 독립표본 T-test (ttest_ind): 남/여, 민초/반민초 등 서로 독립된 두 집단 비교
     - independent : 두 범주가 서로 독립적인 경우
   - 대응표본 T-test (ttest_rel): 약 먹기 전/후 등 같은 집단의 변화 비교
   - 단측 검정: p-값을 2로 나누어 판단
     - 양측 검정 : 두 그룹 간 “차이”가 존재하는가 확인
     - 단측 검정 : 두 그룹 간 차이의 ”한 쪽“만 보고 싶을 때 사용하는 방법

3. **F-test(ANOVA)** (범주가 3개 이상일 때)
   - **개념**: "모든 MBTI별 수명 평균이 같을까?" 처럼 여러 집단을 한 번에 비교합니다.
   - **계산**: $F = \text{집단 간 분산} / \text{집단 내 분산}$

#### 3. **[범주형 - 범주형] Chi2-test**
> **두 범주형 변수가 서로 독립적인지(=관련이 없는지)** 분석

1. **시각화**
   - 칸의 너비와 높이로 범주별 비율 차이를 시각화 
   - 코드 예시
      ```python
      from statsmodels.graphics.mosaicplot import mosaic

      mosaic(df, ['col_행','col_열'])
      plt.axhline(표시할 값, color = '원하는색')
      ```

2. **카이제곱 검정**
- **기대빈도**: "두 변수가 아무 상관이 없다면(독립적이라면) 이 정도 숫자가 나오겠지?" 하고 예상되는 값
- **관측빈도**: 실제로 조사해서 나온 값
- **해석** : `관측빈도 - 기대빈도의 차이`가 크면 "둘은 서로 관련이 있다(독립적이지 않다)"고 판단
- **자유도** : ($df$): $(x \text{범주 수} - 1) \times (y \text{범주 수} - 1)$
  - 카이제곱 통계량 > 자유도의 약 2배 : 유의미한 차이가 있다고 봄

1. **코드 예시**
   ```python
   # 교차표를 작성
   table = pd.crosstab(df['col1'], df['col2'])

   spst.chi2_contingency(table) 
   # 결과: (통계량, p-값, 자유도, 기대빈도)
   ```
