# 회귀 기초

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 1. 머신러닝과 모델링 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.1 머신러닝 </span>
#### 1. **머신러닝의 개념**
1. **개념**
    - 인공지능의 한 분야
      - 참고 : 딥러닝 ⊂ 머신러닝 ⊂ 인공지능
    - <span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;"> 모델이라는 소프트웨어를 학습해 새로운 데이터를 예측하거나 결정을 내일 수 있도록 하는 기술</span>
      - **데이터에서 패턴을 발견**하고, 이 패턴을 바탕으로 **새로운 데이터에 대해 예측**할 수 있게 함!

#### 2. **머신러닝의 종류**
1. **지도 학습**
   - 문제, 정답을 모두 알려주고 공부를 시키는 방법
   - 즉, <span style="color:#0052ef">정답(label)이 있는 데이터</span>를 사용
   - 기계의 **예측값**, 데이터의 **정답**이 **같아지도록** 학습시킴
   - 예 : 회귀, 분류 등

2. **비지도 학습**
   - 답을 가르쳐주지 않고 공부를 시킴
   - 즉, <span style="color:#0052ef">정답(label)이 없는 데이터</span>를 사용
   - 데이터의 **패턴, 유사성**을 학습시킴
   - 예 : 군집화, 밀도추정, 차원축소 등

3. **강화 학습**
    - <span style="color:#0052ef">보상을 통해, 상은 최대화 벌은 최소화하는 방향</span>으로 행위를 강화
    - 시행착오를 반복해 **보상을 최대화하는 행동**을 학습하게 함
    - 예 : 알파고, 자율 주행 등

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.2 모델링  </span>
#### 1. **모델의 개념**
1. **개념** : 학습의 **결과물** or 학습을 마친 **인공지능 그 자체** (예측기)를 의미

2. **역할** : 데이터를 학습하며 찾아낸 **"패턴"이 저장된 상태**로, **새로운 데이터를 넣었을 때 답을 내놓음**

#### 2. **모델링의 개념**
1. **개념** : 예측기(공식)을 만들어가는 전체적 **과정**

2. **역할** : 아무것도 모르는 기계가, 1) 데이터 속 **패턴**을 찾아내 2) 오차가 적은 **최적의 공식을 완성**할 때까지 학습하는 과정!

#### 3. **모델링의 과정** (설계-훈련-검증)
1. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">설계</span>** : 데이터 특성에 맞춰, **어떤 알고리즘으로 학습할지 결정**

2. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;"> 훈련</span>** : 데이터를 기계에게 계속 보여주며, **정답을 맞히도록 수정**하는 단계

3. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;"> 검증</span>** : 훈련시킨 모델이 **새 데이터에서도 정답을 잘 맞히는지 테스트**하는 단계

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.3 지도 학습 </span>
#### 1. **개념**
- **정답 레이블**을 주고 학습을 시키는 방법
- 학습 방향 : **예측값**(prediction) <-> **정답**(label)이 같아지도록!

#### 2. **과정**
1. <span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;"> labeled data</span>를 준비<br>
    - <span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;"> labeled data</span> : **독립변수**(예측에 사용되는 입력 변수), **종속변수**(예측하고자 하는 대상 변수)가 모두 존재하는 데이터임
2. labeled data를 `training set` / `test set`(unseen data)으로 분류<br><br>
    - 데이터를 분류하지 않으면, 모델이 암기를 해서 정답을 맞추는 것인지, 패턴을 찾아내 제대로 학습한 것인지 구별할 수 없기 때문에, 반드시 사전에 분류해야함
3. `training set`을 통해 모델을 학습<br><br>
    - **독립변수**를 통해 추정한 **예측값**이 주어진 **종속변수**와 가까워지도록 모델을 훈련<br><br>
        - **오차**(예측값과 실제 정답 사이의 차이)를 계산해, 오차가 줄도록 **가중치**를 수정해 나가는 것을 반복!
        - 이를 통해 데이터 사이의 **통계적 상관관계를 하나의 함수 형태**로 가지게 됨
        - 학습이 완료된 후, 별도의 새로운 데이터를 사용해, **정확도**를 측정

#### 3. **회귀 vs 분류**
> 지도학습의 대표적인 예 : 회귀, 분류<br>

![회귀와 분류](https://img.notionusercontent.com/s3/prod-files-secure%2F8d7ae215-ed29-46ba-88cb-25d0d6cd52f9%2Fe66c1bc3-6a0e-4a4c-bce6-8fb8334bc0d8%2FUntitled.png/size/w=860?exp=1773556561&sig=KHQuFWv_ootyNr_0b-OUgZ0DNPn5Pxv_Cr1szj7sEHw&id=31701cc3-93ff-814d-b291-fe733af2b593&table=block&userId=836047a6-7920-4aef-9a23-1cd5c348bafa)

1. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;"> 회귀 (regression)</span>**
    - <span style="color:#0052ef">주어진 데이터를 기반으로 정답을 잘 맞추는 **함수**를 찾는 문제</span>
    - 예측하고자 하는 값(종속변수)이 **연속형**일 경우
    - <span style="color:#888888"> 예 : *주택 가격 예측, 음식 배달 시간 예측*

2. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;"> 분류 (classification)</span>**
    - <span style="color:#0052ef">기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤, 새롭게 관측된 데이터에 대한 **레이블을 판별** </span>
    - 예측하고자 하는 값(종속변수)이 **범주형**일 경우
    - <span style="color:#888888"> 예 : *당뇨병 **여부** 예측, 대출 상환 **여부** 예측*


## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 2. 회귀 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.1 개요 </span>
1. **개념** : <span style="color:#0052ef"> 통계, 머신러닝의 주요 개념으로, **데이터에서 패턴을 찾아내, 미래 값을 예측**하는 것
2. **사용 경우** : 종속변수가 **연속형**일 때 (범주형 -> 분류)
3. <span style="color:#888888"> **다양한 기법** : 선형회귀, 비선형회귀, 규제회귀(릿지, 라쏘, 엘라스틱), 다항회귀 등</span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.2 선형회귀 </span>
#### 1. **개념**
1. **개념** : <span style="color:#0052ef"> 데이터를 가장 잘 설명하는 회귀선을 찾는 과정</span>
2. **회귀선**의 역할 : 현재 데이터를 요약, 미래값을 예측
3. **종류** : **단순선형회귀**(x가 1개), **다중선형회귀**(x가 여러개)
    - 예시 : **시간**(독립변수)에 따른 **주가**(종속변수) 예측
  
#### 2. **가중치와 편향**
> 선형회귀의 기본식 : $y = wx + b$ 
<br>
1. <span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;"> **가중치** $w$ (Weight) </span>
    - 독립변수가 종속변수에 주는 **영향력의 크기**
    - <span style="color:#0052ef"> = 기울기, 회귀계수</span>
2. <span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;"> **편향** $b$ (Bias) </span>
    - **독립변수가 0**일 때 나오는 **기본값**
    - <span style="color:#0052ef"> = y절편</span>
    - 역할 : 모델이 데이터에 더 유연하게 최적화될 수 있도록, **위아래로 조절**

#### 3. **단순선형회귀**
![단순선형회귀의 예시](https://upload.wikimedia.org/wikipedia/commons/b/be/Normdist_regression.png)

1. 형태 : **하나의 독립변수**만 존재 (=**직선형태**로 표현하는 경우)
    - $$ y = wx + b $$
1. 과정의 **목적** : **잔차**를 줄여나가며, 직선을 결정하는 **최적의 회귀계수($w$), 편향($b$)을 탐색**

#### 4. **잔차**(Residual)
1. **개념** : <span style="color:#0052ef"> 실제값과 예측값 간 차이</span>
    - <span style="color:#ff0060> **다중선형회귀**에서의 잔차 : 관측치와 평면간의 차이 </span>
    - 회귀선을 잘 찾는 과정 = 잔차를 줄여나가는 과정
2. **오차 vs 잔차**
    - <span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;"> 오차</span> : `실제값 - 실제 평균값` 간의 차이
    - <span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">잔차</span> : `실제값 - 예측값` 간의 차이

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.3 최적의 w, b 구하기 </span>
> 눈으로만 봐서는 최적의 회귀선을 구별하기 어려움. -> 최소제곱법, 경사하강법을 활용!
<br>

#### 1. <span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">**최소제곱법**</span> (=최소자승법)
1. **개념** : **잔차의 제곱의 합(RSS)이 최소**가 되는 지점
    - *Sum of Squared Residuals*
2. `회귀선 - 실제 데이터`와의 차이를 최소화히기 위한 시도
3. 잔차를 제곱하는 이유
    - 양수-음수의 상쇄 방지
    - 큰 잔차에 더 큰 별을 주어 예측값을 강하게 조절
    - 미분가능한 형태가 되기 위해 (경사하강법의 경우)

#### 2. <span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">**경사하강법**</span>
> 최소제곱법을 통해 구한 함수가 복잡해 최솟값을 구하기 어려울 경우 사용<br>
    > **잔차 제곱을 최소화**하려는 목적이 동일함
<br>
1. 경사하강법에서 등장하는 **함수**들<br><br>
   1. <span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">**목적함수**</span>
        - **최적화하려는 전체 대상**을 수학적으로 표현한 함수
        - 방향 : 케이스별로, 어떤 방향으로 최적화할지 달라짐
   2. <span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">**손실함수**</span>
        - **개별 데이터에 대한 오차**를 측정하는 함수
        - 방향 : 최소화
        - 예측값 $\hat{y}$ - 실제값 $y$의 차이를 계산 
          - -> 한 개의 데이터 포인트에 대해 모델이 얼마나 틀렸는지 평가
        - 예 : 회귀분석(평균제곱오차(MSE), 평균절대오차(MAE)), 분류문제(크로스 엔트로피)
   3. <span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">**비용함수**</span>
        - **전체 데이터셋에 대한 손실 함수의 합산 or 평균** => 모델의 전체적 성적을 평가
        - 방향 : 최소화 (보통 목적함수의 방향과 같은 것으로 봄 - 예외존재)

2. **<span style="background-color:#ffb9d3; color:#000000; padding:0px; border-radius:5px;">손실함수와 경사하강법</span>**
    - 손실함수 도출하기
![손실함수와 경사하강법1.jpg](<attachment:손실함수와 경사하강법1.jpg>)
      - 잔차를 제곱한 뒤 합하는 형태
    - E를 최소화하는 $W$,$b$ 찾기
      - $W$로 x축을 구성한 경우, 최적의 $W$를 찾아가는 과정
  ![손실함수와 경사하강법2.jpg](<attachment:손실함수와 경사하강법2.jpg>)
        - (참고) 임의의 점이 왼측에서 시작하는 경우
![손실함수와 경사하강법4.jpg](<attachment:손실함수와 경사하강법4.jpg>)
      - $b$로 x축을 구성한 경우, 최적의 $b$를 찾아가는 과정
![손실함수와 경사하강법3.jpg](<attachment:손실함수와 경사하강법3.jpg>)

1. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">학습률</span>**(α, Learning Rate)
    - **개념** : 오차의 최솟값을 찾아나가기 위해 $W$ or $b$를 이동시키는 정도(**보폭**) 
    - **크기** : *적당한 보폭* (너무 작으면 모델 수렴까지 너무 오랜시간이 소요, 너무 크면 모델이 발산됨(손실값이 계속해 커짐))
    - 설정 방법
      - 학습률은 개발자가 직접 설정하는 하이퍼파라미터임
      - 초기에 작은 값부터 시도해, 실험을 통해 조정해감
      - 자동으로 찾는 방법도 있음 (Learning Rate Scheduler, Grid Search, Ramdom Search 등)

2. **Local Minima** 문제
    - **개념** : <span style="color:#0052ef"> 모든 비용함수가 **하나의 최솟값을 가지지는 않아** 생기는 문제</span>
      - **MSE, RMSE 의 비용함수**에서는 함수가 **볼록함수**라는 것이 증명되었으나, 아닌 경우에는 주의해서 탈출해야함. (다양한 해결 알고리즘 참고!)
    - 작은 기울기에서는 업데이트가 조금씩 이루어지기 때문에 L.M에서 탈출하기 어려움
    - **Local Minima vs Global Minima**


3. **해결법 : <span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">모멘텀</span>**
    - **개념** : 기존 경사하강법과 달리, <span style="color:#0052ef">이전의 기울기와 방향을 기억해 고려하도록 **관성**을 부여한 방식</span>
    - **장점** : 기울기에 관성을 부여해, **작은 기울기(Local Minima)는 쉽게 빠져나갈 수 있음**
    - 모멘텀을 적용한 경사하강법의 식

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.4 다중선형회귀 </span>

#### 1. **개념**
![다중선형회귀의 형태](https://user-images.githubusercontent.com/31824102/35181410-0e137ae6-fdb9-11e7-8e87-1f3560c5c9c5.PNG)
1. **형태** : **독립변수가 두 개 이상**인 경우
    - $$ y = W_1 x_1 + W_2 x_2 + ... + W_n x_n + b $$
2. 독립 변수가 **2개**인 경우 -> <span style="color:#ff0060">3차원 공간에 표현, 회귀선 : 평면</span>
    - *변수가 3개 이상이어도, 비슷한 느낌으로 차원이 쭉 확장됨*
3. **최소제곱법, 경사하강법** 사용 가능 (*최소제곱법으로 안되면 경사하강법으로 파악*)
    - 다중선형회귀에서의 잔차 : 관측치와 평면간의 차이

#### 2. **다중공선성의 개념** (Multicollinearity)
1. **개념** : <span style="color:#0052ef">**독립변수들 간 강한 상관관계**로 인해 회귀 모델이 불안정해지는 문제</span>
2. **영향**
    - 어떤 독립변수가 종속변수에 얼마나 영향을 미치는지 정확하게 구분하기 어려워짐
    - 즉, 각 변수의 회귀계수 추정이 부정확해짐
    - -> 해석력이 떨어짐

#### 3. **다중공선성 확인 방법**
1. **상관계수** (correlation coefficient, $r$)
    - **개념** : 변수간 상관관계의 정도를 수치화한 것
    - **범위** : $-1 ≤ r ≤ 1$ => 양의 상관관계, 음의 상관관계
    - 확인 방법
      - heatmap, corr() 활용
        ```python
        sns.heatmap(df.corr(),annot=True)
        plt.show()
        ```
      - pairplot 활용
        ```python
        sns.pairplot(df, hue="컬럼명", diag_kind="kde", corner=False)
        plt.show()
        ```

2. **VIF 지수**(분산 팽창 지수)
   - **개념** : **R-Square**(회귀 모델의 결정계수)를 활용해 계산
   - **해석**
     - VIF = 1 : 해당 독립변수는 다른 변수와 상관관계가 전혀 없음
     - VIF < 5 : 다중공선성 문제 없음
     - VIF > 5 : 다중공선성 징후가 있을 수 있음, 주의 필요
     - VIF > 10 : 다중공선성이 심각해 회귀모델이 불안정할 가능성이 큼, 조치가 필요

#### 4. **다중공선성 대처 방법**
1. **변수 제거**

2. **변수 변환**
    - 방법 : 변수들을 더하거나 빼 **새로운 변수**를 생성
    - 사용 경우 : 독립변수를 더하거나 뺴도 문제가 없을 경우에만
    - 예 : `남편의 수입` <-> `아내의 수입` : 높은 상관관계 -> 두개를 더해 `가족 수입`이라는 새 변수 생성

3. **규체선형모델** 활용
    - 릿지, 라쏘, 엘라스틱넷 등을 통해 모델의 복잡도 감소

4. **PCA** (주성분 분석)
    - 데이터 차원 축소에 사용되는 통계적 기법
    - 특징 : 고차원 데이터에서 중요 정보를 최대한 보존하며 데이터의 복잡성을 줄여줌
    - 방법 : 상관있는 변수들을 묶어, **더 적은 수의 대표 축**으로 바꾸기 <br>
      - 즉, 데이터의 방향성을 찾아 변환, 중요한 정보만 남겨 차원을 축소

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.5 규제선형모델 </span>


#### 1. **개념**
- **모델이 과적합되지 않도록, 규제를 가하고자 등장한 모델**
  - 선형회귀 모델에서는 특성에 곱해지는 **계수(기울기)의 크기를 조정**하는 것
- 과적합의 위험
  - **기존 선형 모델** : 잔차 제곱합의 평균을 최소화하는 방식으로 비용함수를 설정
    - 회귀 계수가 쉽게 커짐 -> 과적합 발생 가능성 증가
  - **다중 선형 회귀** : 모델이 복잡해짐 -> 과적합 발생 가능성 증가
  - "회귀 계수가 커진다" = 특정 독립 변수(특징)에 과하게 의존한다
- **비용함수의 목표** : RSS 최소화(최소제곱법) + 회귀계수가 너무 커지지 않도록 균형 유지가 중요
  - 즉, 오류는 최소화하되, 특정 변수에 너무 집착하지 않도록!
- <span style="color:#0052ef">**적용 방식** : 비용 함수의 $W$(회귀계수)에 alpha 값으로 **페널티**를 부여해, 회귀 계수 크기를 감소시켜 과적합을 개선</span>

#### 2. **규제선형모델의 종류**
0. **개요**
> 모든 계수에 대해 균등한 페널티를 부과하여 모델의 복잡성을 줄이기 위해 규제는 스케일링이 꼭 필요함<br>
> `StandardScaler`와 `Min-Max scaler`을 주로 이용함

| 규제 방식 | 관련 회귀                | 페널티 부여 대상       | 특징 및 효과                                                        |
|-----------|--------------------------|------------------------|---------------------------------------------------------------------|
| **L2 규제**   | **릿지 회귀** (Ridge)             | 회귀계수($W$)의 제곱   | 계수의 크기를 전체적으로 감소, 모델의 복잡도를 낮춤               |
| **L1 규제**   | **라쏘 회귀** (Lasso)             | 회귀계수($W$)의 절댓값 | 중요하지 않은 변수의 계수를 0으로 만들어 변수 선택 효과가 있음. |
| **L1 + L2**   | **엘라스틱넷 회귀** (Elastic Net) | 절댓값 + 제곱          | 라쏘와 릿지의 장점을 결합 -> 다중공선성 문제를 완화.         |

<br>

1. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">릿지회귀</span>**(L2 규제)<br><br>
    - **방식** : 회귀계수는 **크기에 따라 다른 강도의 패널티**를 부여 (회귀계수의 제곱을 패널티로 주기 때문)
      - 따라서 큰 회귀계수일 수록 강한 패널티를 받음
      - 모든 계수의 크기가 작아지나, 모든 변수가 제거되지는 않음
    - **장점** : 모든 변수를 유지하며, 크기를 적절히 줄여 과적합을 방지할 수 있음! (변수 간 상관관계 높아도 좋은 성능)
    - **사용 경우** : <span style="color:#0052ef"> *예측 변수가 많을 때, **다중공선성**이 존재할 때(변수 선택이 필요 X)*
    - 수식
$$
\text{Cost Function} = \text{RSS} + \alpha \sum_{j=1}^{p} w_j^2
$$

2. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">라쏘회귀</span>**(L1 규제)<br><br>
    - **방식** : 모든 회귀계수에 **같은 강도의 패널티** 부여 (회귀계수의 절대값을 패널티로 주기 때문)
      - 따라서 작은 회귀계수의 경우 제거될 수 있음
    - **장점** : 비중이 낮은 변수는 아예 사라져 중요한 변수만 남음! (변수 간 상관관계 높으면 성능 감소)
    - **사용 경우** : <span style="color:#0052ef">*예측 변수 수가 많은데 **일부만** 중요할 때, 모델의 해석을 **간단**하게 유지하고싶을 때*
    - 수식
$$
\text{Cost Function} = \text{RSS} + \alpha \sum_{j=1}^{p} |w_j|
$$
  
3. **<span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">엘라스틱넷</span>**(L1+L2 규제)<br><br>
    - **방식** : **변수 선택 기능**(L1,일부 계수 삭제)과 **규제 기능**(L2,큰 계수를 줄임)을 **동시**에 활용
      - 예 : 작은 회귀계수를 0으로 만들어 불필요한 변수를 제거하고(L1), 남은 기둥을 균형 잡히게 눌러 크기를 조절함(L2)
    - **장점** : 안정적인 예측(모든 변수가 균형있게 줄어듦) + 희소성(변수제거)를 동시에 달성 가능! (더 나은 성능 but 느림)
    - **사용 경우** : <span style="color:#0052ef">*예측 변수 수가 많고, **중요한 변수**를 선택하면서 **다중공선성**을 관리할 때*
    - 수식
$$
\text{Cost Function} = \text{RSS} + \alpha_1 \sum_{j=1}^{p} |w_j| + \alpha_2 \sum_{j=1}^{p} w_j^2
$$

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 3. 모델 평가방법 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.1 성능 평가 지표 </span>
1. **평균제곱오차(MSE)**
 - **개념** : `예측값 - 실제값 사이 오차를 제곱한 것`들의 평균
 - 수식
$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 
$$

2. **평균 절대 오차(MAE)**
- 개념 : `예측값 - 실제값 사이 오차의 절대값`들의 평균
- 수식
$$
\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|
$$

3. **MSE vs MAE 비교**
   
    | **MSE**            | **MAE**                                          |
    |----------------|----------------------------------------------|
    | 큰 오차에 민감 (제곱이라서) | 모든 오차를 동일하게 취급                    |
    | <span style="color:#0052ef">이상치에 민감  | <span style="color:#0052ef">이상치가 있어도 전반적인 예측 성능 유지 가능 |

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.2 변수 유의성 평가 </span>
#### 1. **T-검정**
1. **목적** : <span style="background-color:#fffa7d; color:#000000; padding:0px; border-radius:5px;">*독립 변수의 회귀계수가 유의미한가?*</span> 검정
    - **귀무가설**(H0) : 회귀계수(a)가 0이다 -> *독립변수가 종속변수에 영향을 주지 않음*
    - **대립가설**(H1) : 회귀계수(a)이 0아니다 -> *독립변수가 종속변수에 영향을 줌*

2. **검정 방식**
   1. **t-값 계산**
      1. $\hat{\beta}$ : 추정된 회귀계수 | $SE(\hat{\beta})$ : 표준오차
      3. t-분포를 따름.
      4. 수식

            $$
            t = \frac{\hat{\beta} - 0}{SE(\hat{\beta})}
            $$

   2. **p-value 확인 & 판단**<br>
      - ` p-값 < 0.05` : H0 기각 -> 해당 변수의 회귀계수는 유의미
      - `p-값 > 0.05` : H0 기각 불가 -> 해당 변수의 회귀계수는 무의미
        - <span style="color:#0052ef"> p-값이 작을 수록, 해당 독립변수가 종속변수에 더 큰 영향을 줄 확률이 높음</span>
      - 회귀계수가 0 : 독립변수는 의미 X
      - <span style="color:#0052ef"> t값이 크면 귀무가설을 기각할 가능성이 높아짐</span>
