## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 1. 선형회귀 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.1 다중선형회귀 이해하기 </span>
- 가장 일반적인 형태
$$
y_i = \beta_0+\beta_1X_{1i}+\beta_2X_{2i}+...+\beta_kX_{ki}+\varepsilon_i
$$
  - 관심있는 $y$에 대해 영향을 줄 것 같은 **여러 독립변수**(설명변수)를 채택해 선형적으로 식을 작성한 것
- *위 식에서 상수항과 오차항을 제거하면...*
$$
y=w_0x_0+w_1x_1+\cdot\cdot\cdot+w_mx_m=\sum_{i=0}^mw_ix_i=w^Tx
$$
  - 즉, **선형회귀식** = <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 종속변수에 대한 독립변수들의 **가중평균**과 같군! </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.2 다중선형회귀의 기본 가정 </span>
> 선형 회귀를 사용하기 위해 아래 **6가지 가정**이 만족돼야함<br>
> 만족 X시, 신뢰성 하락 + 심하면 분석 불가능!!
<br>

#### <span style="color:blue"> **1. 6가지 가정** </span>

1. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">**[가정 1] 선형성**</span>
   1. **수식**
      1. $E(y|X) = \beta_1+\beta_2X_2+...+\beta_nX_n$
   2. **의미** : 종속변수 - 설명변수 관계가 선형적이어야 한다
<br><br>
2. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">**[가정 2] 독립성**</span>
   1. **수식**
      1. $X_i, X_j \ is \ linearly\ independent \ for\ all \ i,j$
   2. **의미** : 각각의 설명변수는 서로 선형독립적이어야 한다
   3. **특징** : **다중공선성**을 예방하기 위함
      1. <span style="color:#0052ef">다중공선성 : **독립변수들 간 강한 상관관계**로 인해 회귀 모델이 불안정해지는 문제</span>
<br><br>
3. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">**[가정 3] 오차항의 평균**</span>
   1. **수식**
      1. $E(\varepsilon_i|X_i) = 0$
   2. **의미** : 오차항의 평균은 0이거나, 0에 가까워야 한다
   3. **오차항**이란? : `실제 값` - `예측 값` 간의 차이 
      1. 오차항은 **백색잡음**이라고도 하는데, 백색잡음은 특정 패턴 없이 랜덤하게 변하는 신호로서, 한 쪽에 치우치게 되면 모델이 평균적으로 오차항만큼 잘 못 예측할 수 있음. (올바른 패턴 반영 X)
<br><br>
4. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">**[가정 4] 등분산성**</span>
   1. **수식**
      1. $Var(\varepsilon_i|X_i) = \sigma^2$
   2. **의미** : 오차항의 분산이 일정해야한다
      1. 분산이 일정하지 않다 => *이분산성*
      2. 이분산성 시 -> 모델의 예측의 정확성이 구간마다 달라질 수 있음
   3. **확인 방법**
      1. 잔차를 도표화해서 확인하기, 검정(Brown-Forsythe, Breusch-Pagan)
<br><br>
5. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">**[가정 5] 오차항의 자기상관 여부**</span>
   1. **수식**
      1. $Cov(\varepsilon_i, \varepsilon_j|X_i) =0$
   2. **의미** : 오차항은 자기상관되어있지 않다, 즉 **오차항의 공분산(Cov)는 항상 0**이어야 한다
      1. **공분산** : 두 변수간의 상관된 정보를 나타냄. 즉, 두 변수가 어떻게 함께 변하는지 측정
         1. 공분산값은 상관계수와 같은 방식으로 해석함
      2. **자기상관** : 한 변수의 현재 값이 과거값과 상관관계를 갖는다.
         1. 시계열 자료에서 자주 나타나는 편
         2. 데이터를 예측하지 않고, 이전 데이터의 패턴을 학습해 반복 오류를 만들 가능성이 높음
   3. **확인 방법**
      1. 더빈-왓슨 검정
      2. ACF, PACF 함수 찍어보기
<br><br>
6. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">**[가정 6] 정규성**</span>
   1. **수식**
      1. $\varepsilon_i \sim N(0, \sigma^2)$
   2. **의미** : 오차항은 정규분포를 따른다
      1. 회귀계수들의 신뢰구간, p-값 공식이 이 가정을 기반으로 만들어졌기 때문
   3. 특징
      1. 위배되어도 다중선형회귀분석 결과에 큰 영향은 주지 않음
   4. 확인 방법
      1. 샤피로-윌크 검정, 자퀴베라 검정 Q-Q plot 찍어보기
<br><br>
7. **모아보기**

    |       기본 가정       |          의미          | 회귀 분석 전 | 회귀 분석 후 |          검증 방법          |
    |:---------------------:|:----------------------:|:------------:|:------------:|:---------------------------:|
    | 선형성                | 종속-설명 관계 선형    |       O      |       O      | scattor plot                |
    | 독립성(다중공선성)    | 설명 변수는 선형독립적 |       O      |       O      | VIF 지수, 상관계수          |
    | 오차항 평균           | 0이거나 0에 가까워야함 |       X      |       O      | np.mean(residuals)          |
    | 등분산성              | 오차항 일정한 분산     |       X      |       O      | 잔차의 도표화               |
    | 오차항 자기상관 여부  | 자기상관 X             |     간접     |    O(최종)   | Durbin-Watson 검정(분석 후) |
    | 오차항 정규성         | 오차항 정규분포 따름   |       X      |       O      | Shapiro-Wilk 검정, Q-Q plot |

#### 2. **주의사항**

1. **언제 검증해야할까?**
   1. 가정마다 다름. -> 위 표 참고
   2. 오차항 = `실제 값` - `예측 값` 간의 차이 
      - 따라서, 오차항이 모델 학습 후 정해지기 때문에 모델 학습 전에는 확실히 알 수 없음
   3. 검증 진행 후, 위배됨을 확인했을 경우 모델 수정 -> 학습 -> 다시 검정
   4. 즉, 분석 전 데이터의 관계를 확인 (가정 1,2) -> 분석 후 오차항의 성질 확인 (가정 3-6)

2. **수정을 거듭해도 6개 가정이 모두 만족되지 않을 경우**
   1. 100% 만족은 어렵기 때문에 아래 기준으로 타협하기도 함
      1. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 데이터 행 수가 충분히 많을 경우 </span>($n$>=30 : 중심극한정리)
         1. 변수 한 개당 10-20개의 행은 필요
      2. <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 회귀 모델의 목적이 **예측**인경우</span> (비즈니스 분야)
         1. 가정이 조금 깨져도, **실제 예측 오차(MSE, RMSE)가 낮으면 그대로 사용가능**
         2. 다만, 목적이 설명(학술 분야)일 경우, 신뢰성 있는 p-value를 위해 가정을 지키는 것이 중요


### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.3 회귀분석 평가하기 </span>
> 회귀선을 데이터와 함께 시각화 or 통계지표 활용!
<br>

#### 1. 시각화
- **회귀선**을 통해 **전반적인 요약**을 볼 수 있음
- 다만 객관적인 비교는 어려우므로 **통계지표 확인을 병행**하는 것이 좋음

#### 2. 통계지표
1. **모델 유의성 검정**
    - <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">**F-statistic**</span>, Prob(F-statistic) (F-통계량, 유의확률)
      - **목적** : 회귀 **모델**이 유의미한지 검정 (= y-X 관계가 통계적으로 유의미한가)
      - **가설**
        - 귀무 : 모든 회귀계수가 0이다
        - 대립 : 적어도 하나의 회귀계수는 0이 아니다
      - **해석**
        - F통계량 값이 클 수록 -> 통계적으로 유의미 (0<F<1 : 의미X)
        - p-값 <= 0.05 -> 귀무가설 기각 (적어도 하나의 X가 y와 유의미한 관계이다)

    - <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">**T-statistic**</span>, Prob(T-statistic)
      - **목적** : **각 독립변수**가 종속변수에 유의미한 영향을 미치는지 검정
      - **가설**
        - 귀무 : 해당 변수의 회귀 계수가 0이다
        - 대립 : 해당 변수의 회귀 계수는 0이 아니다
      - **해석**
        - T 통계량 절대값의 유의확률(p-값)을 기준으로 가설 검정 수행
        - 예시
          - x1 : P>|t|이 0.05보다 작으면 -> 유의미한 영향 O
          - x2 : P>|t|이 0.05보다 크면 -> 유의미한 영향 X

2. **모델의 성능 평가**
    - **결정계수** (<span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;">R-squared</span>, $R^2$)
      - **개념** : 회귀 모델이 종속변수의 전체 변동성을 얼마나 잘 설명하는지 (모델이 데이터를 얼마나 잘 설명하는지 측정)
        - 평균값으로 데이터를 측정했을 때에 비해, 모델을 활용했을 때 오차가 얼마나 감소하는지를 수치화함
        - **SST** : 평균으로 예측했을 때의 총 오차 (극복해야하는 전체 변동성)
        - **SSE** : 모델이 예측하고도 여전히 남은 오차 (모델의 한계)
        - **SSR** : 평균에 비해 회귀 모델이 얼마나 더 잘 맞췄는지 (모델의 성과)
      - **수식**
        - $R^2 = SSR/SST = 1 - SSE/SST$
      - **해석**
        - $R^2$가 1에 가까울 수록 -> *모델이 데이터를 잘 설명한다*

    - **조정된 결정 계수** (Adjusted R-squared, Adjusted $R^2$)
      - 독립 변수의 개수 증가 -> $R^2$도 함께 증가
      - 따라서 변수의 개수 증가에 덜 민감하도록 조정한 것이 Adjusted $R^2$
      - **수식**
        - $\overline{R}^2 = \frac{SSR}{SST}\frac{n-1}{n-k}=1-(\frac{SSE/(n-k)}{SST/(n-1)})$<br><br>
        - (*$n$: 샘플 수, $k$: 독립변수의 개수*)
        - 변수의 개수($K$) 증가 -> 전체값 감소 : 변수 개수에 따른 설명력 추가 부분을 상쇄하는 효과

    - **AIC, BIC(SC)**
      - 두 지표 모두 **정보 기준**이라고 불림
      - **해석** : 값이 낮을 수록 좋은 평가
      - **AIC** : 비교적 작은 패널티 부과
      - **BIC(SC)** : 데이터에 양에 따라 더 강한 패널티 부과

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 2. 비선형회귀 </span>
> 데이터셋의 분포가 선형적이지 않을 때 사용하는 방식<br>
> <span style="color:blue"> 회귀는 데이터 간의 패턴을 함수 관계로 나타내는 것 -> 직선이 아닌 데이터셋에는 비선형 식을 적용하여 모델링 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.1 다항식 회귀 모델 (Polynomial Regression Model) </span>

1. **개념** : 독립 변수에 **거듭제곱 항**을 추가하여 데이터의 <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 곡선적(비선형적) 특성 </span>을 모델링하는 방식.

2. **수식**
    $$Y = a + bX + cX^2 + dX^3 + \dots$$

3. **특징**
   - <span style="color:blue"> 기존의 선형 모델에 $X^2, X^3$ 등을 포함하여 확장한 후 선형 회귀를 적용함. </span>
   - 선형 모델보다 회귀선을 데이터에 <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> fit하게 </span> 그릴 수 있으나, 변수가 너무 많으면 **과적합** 위험이 있음.
   - **성장률이나 감소율**이 시간에 따라 다르게 변화하는 곡선 데이터 패턴에 유용

4. **코드 예제**
    ```python
    from sklearn.preprocessing import PolynomialFeatures
    from sklearn.linear_model import LinearRegression
    import statsmodels.api as sm

    # 변수를 다항식으로 변환
    poly = PolynomialFeatures(degree=2, include_bias=False)  
    X_poly = poly.fit_transform(X)

    # statsmodels sm 활용
    X_poly_sm = sm.add_constant(X_poly)
    model = sm.OLS(Y, X_poly_sm).fit()
    Y_pred = model.predict(X_poly)
    ```

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.2 지수 회귀 모델 (Exponential Model) </span>

1. **개념** : 종속 변수가 <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 지수적으로 변화 </span>하는 관계를 모델링함.

2. **수식**
    $$Y = ae^{bX} \quad \rightarrow \quad \ln Y = \ln a + bX$$

3. **특징**
   - <span style="color:blue"> 종속 변수 $Y$에 로그를 취한 후 선형 회귀를 적용
   - 예측 시에는 다시 지수함수($e^x$)를 적용해 원래 형태로 복구 </span>
   - 값이 시간에 따라 **급격히 증가하거나 감소**하는 경우(기술 발전, 투자 수익률 등)에 적합

4. **코드 예제**
    ```python
    import numpy as np
    import matplotlib.pyplot as plt
    import statsmodels.api as sm

    # 지수함수 형태의 임의의 데이터 생성 
    np.random.seed(42)
    X = np.linspace(1, 10, 20).reshape(-1, 1)
    Y = 2 * np.exp(0.8 * X).flatten() + np.random.normal(0, 10, X.shape[0]) 

    # 종속변수 Y에 자연 로그 변환
    log_Y = np.log(Y) # log_Y: 로그 변환된 Y

    # 회귀 모델 학습 (log(Y) = a + bX)
    X_const = sm.add_constant(X)  
    model = sm.OLS(log_Y, X_const).fit()

    # 예측값 변환 (exp를 적용해 원래 스케일로 변환)
    log_Y_pred = model.predict(X_const)
    Y_pred = np.exp(log_Y_pred) # exp(지수 함수) 적용해서 로그 변환 되기 전의 Y로

    # 데이터 시각화
    plt.scatter(X, Y, label="Actual Data", color="blue")
    plt.plot(X, Y_pred, label="Exponential Fit", color="red")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.legend()
    plt.title("Exponential Regression")
    plt.show()
    ```

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.3 로그 회귀 모델 (Log Regression Model) </span>

1. **개념** : 변수 간의 관계가 <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 로그 함수 </span>를 통해 더 잘 표현될 때 사용

2. **수식**
    $$Y = a + b \log(X)$$

3. **특징**
   - <span style="color:blue"> 독립 변수 $X$에 로그를 취한 후 선형 회귀를 적용 </span>
   - 지수 회귀와 달리 예측된 값이 이미 $Y$이므로 별도의 처리가 필요 X
   - <span style="color:blue"> 초기에 빠르게 증가하다가 점차 증가율이 줄어드는 패턴(인구 성장, 감염병 확산 등)에 효과적! </span>

4. **코드 예제**
    ```python
    import numpy as np
    import matplotlib.pyplot as plt
    import statsmodels.api as sm

    # 로그함수 형태의 임의의 데이터 생성 
    np.random.seed(42)
    X = np.linspace(1, 10, 20).reshape(-1, 1)
    Y = 5 + 3 * np.log(X).flatten() + np.random.normal(0, 0.5, X.shape[0]) 

    # 독립 변수 X에 자연 로그 변환
    log_X = np.log(X)

    # 회귀 모델 학습 (Y = a + b * log(X))
    log_X_const = sm.add_constant(log_X) 
    model = sm.OLS(Y, log_X_const).fit()

    # 예측값 계산
    Y_pred = model.predict(log_X_const)

    # 데이터 시각화
    plt.scatter(X, Y, label="Actual Data", color="blue")
    plt.plot(X, Y_pred, label="Logarithmic Fit", color="red")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.title("Logarithmic Regression")
    plt.show()
    ```

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.4 스플라인 회귀 (Spline Regression) </span>

1. **개념** : 데이터를 <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 구간별</span>로 나누고 각 구간에 다른 함수를 적용하는 방식

2. **수식**
    $$Y = \begin{cases} a_1 + b_1X & \text{if } X < c_1 \\ a_2 + b_2X & \text{if } c_1 \leq X < c_2 \\ \dots \end{cases}$$

3. **특징**
   - 전체 범위를 여러 구간으로 나누고, **경계점**에서의 <span style="background-color:yellow; color:#000000; padding:0px; border-radius:5px;"> 연속성 </span>을 유지하며 분석
   - <span style="color:blue"> 데이터 패턴이 구간마다 서로 다르게 나타날 때 유용
   - 계절에 따른 판매량 변화 분석 등에 적합 </span>

4. **코드 예제**
    ```python
    # 스플라인 회귀 모델에 대한 간략한 버전의 코드. 
    # 전체 구간에 대해 한 종류의 회귀 모델만 사용 가능 (선형 또는 다항)
    # 첫 번째 구간에는 선형 모델을 사용, 두 번째 구간에는 다항 모델을 사용하는 것처럼
    # 구간별로 다른 모델을 사용하기 위해서는 반복문을 활용한 코드를 사용해야 힘. 

    import numpy as np
    import matplotlib.pyplot as plt
    import pandas as pd
    from sklearn.linear_model import LinearRegression
    from patsy import dmatrix
    import statsmodels.api as sm

    # 랜덤 데이터 생성
    np.random.seed(42)
    X = np.linspace(0, 10, 100)
    Y = np.sin(X) + np.random.normal(0, 0.2, size=X.shape)

    # X의 데이터에 스플라인 함수를 적용하여 새로운 데이터 X_spline을 만들고, 
    # 이 새로운 데이터로 선형 회귀를 진행하는 방식

    X_spline = dmatrix("bs(X, df=5, degree=3, include_intercept=False)", 
    {"X": X}, return_type='dataframe')
    # X: 데이터
    # df: 나눌 구간의 수+1 (df=5라면 4개의구간으로 구분)
    # degree: 사용할 회귀 모델의 차수 (degree=1이면 선형회귀, degree=2면 2차 다항 회귀)
    # include_intercept=False: 절편 제거(필요하면 True로)
    # return_type='dataframe': dataframe 형식으로 반환

    model_spline = sm.OLS(Y, X_spline).fit()
    Y_pred_spline = model_spline.predict(X_spline)

    # 시각화
    plt.scatter(X, Y, label="Actual Data", color="blue", alpha=0.5)
    plt.plot(X, Y_pred_spline, label="Cubic Spline Regression", color="red")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.legend()
    plt.title("Cubic Spline Regression Fit")
    plt.show()
    ```
