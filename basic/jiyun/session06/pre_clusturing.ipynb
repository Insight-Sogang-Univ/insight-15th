## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 1. 군집화 (Clusturing) </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.1 머신러닝 : 비지도학습 </span>
1. **머신러닝의 분류**
    - **머신러닝**: 컴퓨터가 스스로 학습할 수 있도록 알고리즘을 개발하는 분야로, 크게 **지도학습**과 **비지도학습**으로 구분됨.

    | 구분 | 지도 학습 (Supervised) | 비지도 학습 (Unsupervised) |
    | :--- | :--- | :--- |
    | **정답 여부** | **정답**(Label)이 있는 데이터 사용 | **정답**(Label)이 없는 데이터 사용 |
    | **학습 방식** | **예측값**과 정답이 일치하도록 학습 | 데이터 속 **패턴이나 유사도**를 학습 |
    | **주요 예시** | 회귀, 분류 | **군집화**, 연관분석 |
<br>

2. **비지도학습의 특징**
    - 정답 레이블 없이 **데이터 자체의 구조나 고유 패턴을 식별**
    - 명확한 정답이 없어 모델의 **성능을 객관적으로 측정하기 어려움**
    - 목표가 불명확한 상태에서 **새로운 특성을 찾거나 지식을 일반화**하는 데 적합

3. **비지도학습이 적합한 케이스**
    - <span style="color:blue">패턴이 아직 알려지지 않아</span> 정답을 알 수 없는 경우
    - <span style="color:blue">패턴이 계속 변하여</span> 정답을 고정하는 것이 의미가 없는 경우
    - <span style="color:blue"> 열린 문제를 해결하고 지식을 일반화</span>해야하는 경우
    - **예**: *고객 세그먼트 분석, 이상 거래 탐지(FDS), 자연어 처리 인사이트 도출 등*

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.2 군집화의 개념 </span>

1. **정의**
    - 데이터를 **비슷한 특성을 가진 그룹**(군집)으로 나누는 비지도 학습 기법
    - **군집**(Cluster): 유사한 데이터들의 집합이며, 서로 다른 군집 간에는 확실한 차이가 있어야 함

2. **목표**
    > 사전에 정해진 목표 없이 데이터셋의 고유 패턴을 발견하고 비슷한 특성을 지닌 개체를 그룹화함
    <br>
    - **<span style="background:yellow">응집도(Cohesion)</span> 최대화**: 동일 군집 내 데이터끼리는 **최대한 비슷**하게 구성
    - **<span style="background:yellow">분리도(Separation)</span> 최대화**: 서로 다른 군집은 최대한 명확히 **분리**되도록 함

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.3 군집화의 과정 </span>

1.  **피처 선택 or 추출**: 군집화에 사용할 데이터의 핵심 피처 선정<br><br>
2.  **군집화 알고리즘 선택**: 데이터 특성과 분석 목표에 따라 적합한 군집화 알고리즘 선택<br><br>
3.  **군집 유효성 검증**: 산출된 군집화 결과가 얼마나 타당한지 평가<br><br>
4.  **결과 해석**: 각 군집이 지닌 특성을 분석 & 해석

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 2. 군집화를 위한 데이터 준비 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.1 군집화의 주요 고려사항 </span>

> 군집화는 단발성이 아닌 <span style="backgroung:yellow">**반복적인 시도**를 통해 최적의 결과를 찾는 과정</span>임.<br>
> 효과적인 군집화를 위해 아래 **세 가지 핵심 사항**을 반드시 고려해야 함.
<br>

1.  **변수 유형 이해**
    1.  군집화에 사용될 피처의 종류(연속형, 명목형)와 개수에 따라 알고리즘 선택이 달라짐.
2.  **거리 / 유사도 정의**
    1.  데이터 간 거리, 유사도 기반 그룹 생성
    2.  거리를 측정하는 방식이 군집 형성의 핵심임. (변수 유형에 따라 다름))
3.  **차원 축소**
    1.  변수가 많아질수록 모델 효율/성능이 떨어질 수 있음 
    2.  -> **유사 변수를 통합**하거나 **불필요한 변수를 제거**하는 과정이 필요함.

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.2 변수 유형 이해 </span>

변수 유형에 따라 **거리 측정법**과 **알고리즘**이 결정됨

1.  **연속형 변수**
    - 예: *키, 몸무게, 소득 등 수치 데이터*
    - 측정법: 유클리디안, 맨하탄 거리 등
    - 알고리즘: K-means, 계층적 군집화
    - 주의사항: 변수 간 단위가 다를 경우 큰 값이 결과를 지배하므로 **스케일링이 필수!!**

2.  **명목형 변수**
    - 예: *성별, 지역, 직업 등 범주 데이터*
    - 측정법: 해밍 거리, 자카드 거리
    - 알고리즘: K-modes, 계층적 군집화(특별 거리 함수 사용)
    - 처리 방법: **원-핫 인코딩** 또는 더미 변수 변환 수행

3.  **혼합형 변수 (Mixed Variables)**
    - 연속형과 범주형이 섞인 형태
    - 알고리즘: **K-prototypes**, **Gower distance** 기반 계층적 군집화
    - 처리 방법 : 각 변수 유형에 맞는 거리 함수를 조합해 사용

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.3 거리 / 유사도 정의와 측정 </span>
> 데이터 포인트 / 군집 간의 유사성/비유사성을 수치로 계산하는 방법들<br>
> <span style="color:blue"> 거리가 가까울 수록 유사, 멀수록 다르다!! </span>
<br>

1. **주요 거리 측정 방법**
   - **유클리디안 거리(기본))**:  => 이번 세션에서 사용!!
     - 가장 일반적인 <span style="color:blue">직선거리 측정법</span>
     - K-means에서 기본적으로 사용
        $$d(\mathbf{x}, \mathbf{y}) = \sqrt{\sum_{i=1}^n (x_i - y_i)^2}$$

   - **맨하탄 거리** (Manhattan Distance)
     - 차원의 <span style="color:blue">절댓값 합</span>으로 계산
     - **이상치에 상대적으로 덜 민감함**
        $$d(\mathbf{x}, \mathbf{y}) = \sum_{i=1}^n |x_i - y_i|$$

   - **코사인 유사도** (Cosine Similarity)
     - 벡터 간의 <span style="color:blue">각도</span>를 측정
     - 텍스트 및 고차원 데이터 분석에 효과적
       $$\text{similarity} = \frac{\mathbf{x} \cdot \mathbf{y}}{||\mathbf{x}|| \times ||\mathbf{y}||}$$

2. **계층적 군집화의 6가지 연결 방법**
    > 계층적 군집화에서 두 군집 간의 거리를 계산하는 방식들<br>
    > <span style="color:blue"> 군집 간 거리 **측정 기준이 달라** -> **서로 다른 군집화 결과**를 만들어냄 </span>
    <br>

    | 연결 방법 | 정의 및 특징 |
    | :--- | :--- |
    | **단일 연결법 (Single)** | 군집 간 **가장 가까운 두 점**의 거리 기준. 고립 군집 탐색에 유리하나 **이상치에 매우 취약** |
    | **완전 연결법 (Complete)** | 군집 간 **가장 먼 두 점**의 거리 기준. **내부 응집성에 중점**을 두지만 이상치에 민감함 |
    | **평균 연결법 (Average)** | 모든 점 쌍 사이의 **거리 평균** 사용. **이상치에 비교적 덜 민감** but 계산량이 많음 |
    | **중심 연결 (Centroid)** | 각 군집 **중심점(Mean)** 간의 거리 측정, 가중평균으로 새 군집의 평균을 구함, **Inversion** 현상 발생 가능성 O |
    | **중앙 연결법 (Median)** | 데이터 개수와 상관없이 두 군집 **중심의 중점**을 새 대표값으로 설정, 극단값 영향 감소 |
    | **Ward 연결법 (Ward)** | 병합 시 **군집 내 오차제곱합(SSE) 증가량**을 최소화하는 방식, 가장 널리 쓰임 |
<br>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.4 차원 축소 </span>

1. **차원의 저주**
   - 피처가 늘어날수록 데이터 밀도가 급격히 낮아지는 현상
   - 고차원에서는 모든 점 사이의 거리가 비슷해져 **거리 기반 알고리즘의 분별력이 상실됨.**
   - 따라서 차원 축소가 필요!

2. **차원 축소** : 변수의 개수를 줄이는 것
   - **목표** : 고차원 데이터를 저차원 공간에 투명 + 핵심 정보는 가능한 최대로 지키기 => 노이즈가 줄어, 패턴 학습이 유리 
   - **유형**
     - 선형 투영 : 고차원 공간 -> 저차원 공간으로 선형적으로 데이터 투영
       - 주성분분석(**PCA**), 특잇값 분해(SVD), 랜덤 투영 등
     - 매니폴드 학습(비선형 차원 축소) -> 유클리드 거리가 아닌 데이터 포인트들 사이의 곡선 거리 고려해 학습
       - t-SNE, UMAP, Isomap 등

   - 다차원 데이터에서 가장 정보량(분산)이 큰 방향을 찾아, 그 방향으로 데이터를 투영해 차원을 줄이는 것!
   - 데이터의 **분산을 최대한 보존**하면서 차원을 줄이는 선형 투영 기법
   - 상관관계가 높은 피처들을 결합하여 새로운 **주성분**(PC) 생성

1. **주요 파라미터**
   1. 1. `n_components` (주성분 개수)
     - **정수** (예: `3`): 정확히 그 개수만큼 주성분 선택
     - **실수** (예: `0.95`): 그 비율의 분산을 설명하는 개수를 자동 계산
     - **None**: 모든 주성분 사용 (차원축소 없음)

   2. `whiten` (데이터 정규화)
    - **True**: 각 주성분을 단위분산으로 정규화
    - **False** (기본값): 정규화 미적용

   3. `random_state` (랜덤 시드)
<br>

1. **주요 매서드**
    | 메서드 | 설명 |
    | --- | --- |
    | `fit(X)` | PCA 모델 학습 (주성분 계산) |
    | `transform(X)` | 데이터를 주성분 공간으로 변환 |
    | `fit_transform(X)` | 학습과 변환을 한번에 수행 |
    | `inverse_transform(X)` | 원본 공간으로 역변환 (근사값) |
<br>

1. **주요 속성**
   1. `explained_variance_ratio_`
      1. 각 주성분이 설명하는 **분산 비율**
      2. 예: `[0.4, 0.3, 0.2]` → PC1이 40%, PC2가 30%, PC3가 20% 분산 설명

   2. `explained_variance_`
      1. 각 주성분의 **실제 분산값**
 
   3. `components_`
      1. **주성분 벡터** (원본 특성들의 가중합)
      2. 어떤 원본 특성이 각 주성분에 얼마나 기여하는지 표시

   4. `n_components_`
      1. **실제 선택된 주성분 개수** (분산 비율로 설정 시 유용)

2. **주의사항**
   - 피처의 스케일에 민감하므로 적용 전 **표준화를 반드시 수행**해야 함

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 3. 군집화 알고리즘 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.1 계층적 군집화 </span>

1. **개념**
    - 데이터 간 유사성을 기반으로 **트리 구조**(Dendrogram)를 형성하는 방법
    - 군집 개수를 사전에 설정할 필요가 없으며, 결과 도출 후 원하는 지점에서 잘라 군집 수를 선택 가능
    - **<span style="background:yellow">Dendrogram</span>**: 개체 간 병합 과정을 시각화한 트리. 수직축의 높이는 군집 간의 거리(비유사성)를 의미

2. **응집형(상향식) vs 분리형(하향식)**
    - **응집형**(agglomerative): 각 데이터를 개별 군집으로 시작하여 가장 가까운 것끼리 병합
    - **분리형**(divisive): 전체를 하나의 군집으로 시작하여 유사성이 낮은 데이터를 분할

3. **응집형 계층적 군집화**
   - 모든 데이터 포인트를 각각 하나의 클러스터로 간주하고 시작해 점차적으로 가까운 클러스터들을 병합해 나가는 방식    
     - **'가장 가까운 클러스터'**를 어떻게 정의하느냐에 따라 다양한 알고리즘으로 나뉨
     - **단일 연결**(가장 가까운 두 샘플의 거리)과 **완전 연결**(가장 먼 두 샘플의 거리) 방식이 대표적

   - **완전 연결을 사용한 응집형 군집화 과정**
     1. **거리 행렬 계산**: 모든 데이터 포인트 간의 거리를 계산하여 거리 행렬을 만듦
     2. **단일 클러스터 시작**: 각 데이터 포인트를 개별적인 클러스터로 초기화
     3. **가장 가까운 클러스터 병합**: 정의된 기준(가장 멀리 있는 샘플 간의 거리)에 따라, 가장 가까운 두 클러스터를 병합하여 새로운 클러스터를 만듦
     4. **거리 행렬 업데이트**: 새로 생성된 클러스터와 나머지 클러스터들 간의 거리를 다시 계산해 거리 행렬 업데이트합
     5. **반복**: 모든 데이터 포인트가 하나의 클러스터에 속할 때까지 3, 4번 과정을 반복

4. **계층적 군집화 시각화**
    - **덴드로그램**을 통해 클러스터들을 요약해 확인 가능
    - 계산량이 많아 대규모 데이터보다는 **소규모 데이터의 구조 시각화**에 유용

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.2 k-means </span>
1. **개념**
    - 데이터를 정해진 개수(K)의 그룹으로 나누고, 각 그룹의 **중심점**(Centroid)과 가까운 데이터를 묶는 알고리즘
    - **프로토타입 기반 군집화**: 연속형 데이터의 경우 평균값인 센트로이드를 대표점으로 사용

2. **작동 단계**
    - 초기 중심점 K개 랜덤 선택 → 유클리디안 거리 제곱을 기준으로 가까운 중심점에 샘플 할당 → 할당된 샘플들의 평균으로 중심점 이동 → 변화가 없을 때까지 반복
    - 목표: 클러스터 내 **제곱 오차합**(SSE) 최소화

3. **K-Means++ & 엘보우 방법**
    - **K-Means++**: 초기 중심점을 서로 멀리 떨어뜨려 초기화하는 개선된 방법임. 일관된 성능을 보장
    - **Elbow Method**: K값 변화에 따른 SSE 감소 폭이 급격히 줄어드는 지점(팔꿈치)을 최적의 K로 판단

    ```python
    from sklearn.cluster import KMeans
    kmeans = KMeans(n_clusters=3, init='k-means++', random_state=42)
    kmeans.fit(data)
    ```

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.3 DBSCAN </span>
1. **DBSCAN의 개념**
    - 데이터의 **밀도**(Density)를 기준으로 군집을 형성, 고밀도 영역은 군집/ 저밀도 영역은 노이즈로 간주
    - 기하학적이고 복잡한 분포를 가진 데이터 처리에 탁월!!

2. DBSCAN 알고리즘의 주요 단계
   1. **샘플 분류**
       - 핵심 샘플(**Core**): 반경(eps) 내 이웃이 최소 개수(min_samples) 이상인 점
       - 경계 샘플(**Border**): 이웃은 적지만 핵심 샘플의 반경 내에 있는 점
       - 잡음 샘플(**Noise**): 어디에도 속하지 않는 이상치
   2. **클러스터 생성**
      - 개별 핵심 샘플 또는 ε 이내에 있는 핵심 샘플을 연결한 그룹을 클러스터로 생성
   3. **경계 샘플 할당**

3. **장단점**
    - **장점**: 군집 수를 미리 정할 필요가 없고 **이상치 제거**에 강함
    - **단점**: 데이터 밀도가 균일하지 않거나 차원이 높아지면 성능이 저하

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 3.4 Gaussian Mixture Model </span>
1. **GMM의 개념**
    - 데이터가 여러 개의 **가우시안 분포**(정규분포)의 혼합으로 구성되었다고 가정하는 모델 기반 군집화
    - 데이터가 특정 분포에 속할 **확률**을 계산하여 군집을 할당

2. **EM 알고리즘 진행 과정**
    - **E(Expectation) 단계**: 현재 파라미터를 기준으로 데이터가 각 분포에 속할 확률(사후 확률) 계산
    - **M(Maximization) 단계**: 계산된 확률을 통해 분포의 파라미터(중심, 모양 등)를 재추정

3. **특징**
    - K-Means보다 유연하여 타원형이나 중첩된 군집 구조도 잘 처리함
    - 정규분포 가정이 필요하며 연산 시간이 다소 오래 걸림

```python
from sklearn.mixture import GaussianMixture
gmm = GaussianMixture(n_components=3, covariance_type='full', random_state=42)
gmm.fit(data)
```

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 4. 군집화 평가 방법 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.1 외부 평가와 내부 평가 </span>
1. **외부 평가** (External Evaluation)
    - **정답 레이블이 존재**할 때 사용함. 실제 정답과 군집 결과의 일치도를 측정함.
    - **ARI (Adjusted Rand Index)**: 무작위 일치 가능성을 배제한 지표로, 1에 가까울수록 완벽한 일치를 의미함.

2. **내부 평가** (Internal Evaluation)
    - 정답 레이블이 없을 때 **데이터 자체의 구조**로 평가함.
    - 군집 내 **응집도**는 높고, 군집 간 **분리도**는 높은 것을 지향함.
    - 지표: SSE(K-means), 로그 우도(GMM), 실루엣 계수, 던 지수 등.

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.2 실루엣 개수 (Silhouette Coefficient) </span>
1. **실루엣 계수 계산**
    - $s^{(i)} = \frac{b^{(i)} - a^{(i)}}{\max\{a^{(i)}, b^{(i)}\}}$
    - $a^{(i)}$: 동일 군집 내 다른 점들과의 평균 거리 (응집도).
    - $b^{(i)}$: 가장 가까운 타 군집 점들과의 평균 거리 (분리도).

2. **해석하기**
    - **1에 가까움**: 군집화가 매우 잘 됨 (분리도 > 응집도)
    - **0에 가까움**: 군집 경계가 모호
    - **-1에 가까움**: 데이터가 잘못된 군집에 할당
    - **평가 기준**: 전체 평균이 **0.7 이상**이면 우수한 성능으로 간주, **군집별 계수 분포가 고른 것**이 중요

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.3 Dunn Index </span>
1. **개념**
    - 군집 간 최소 거리(`분리도`) / 군집 내 최대 거리(`응집도`)의 비율
    - $\text{Dunn Index} = \frac{\text{min inter-cluster distance}}{\text{max intra-cluster distance}}$

2. **특징 및 한계**
    - 값이 **클수록** 군집 간 거리는 멀고 내부 응집은 잘 된 좋은 군집화 결과
    - 실루엣 계수와 함께 종합적으로 평가하는 것 권장
    - 계산 비용이 높고 이상치에 매우 민감하여 수치가 급격히 변할 수 있음
