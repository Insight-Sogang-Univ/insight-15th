# 연관 분석

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 1. 연관 분석의 개념 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.1 연관분석의 정의 </span>
1. **정의** : 상품이나 서비스를 구매하는 등 일련의 거래, 사건 안에 존재하는 항목 간의 일정한 **연관 규칙**을 발견하는 분석!
   1. *유통업에서는 연관 분석의 한 분야인 **장바구니 분석**이라고 많이 언급한다 함*

2. 특징
   1. 연관 규칙을 발견하는 것이기에, 다양한 분야에서 사용됨
      1. 시청 패턴 분석을 통한 맞춤 콘텐츠 추천, 거래 패턴 분석을 통한 사기 행위 탐지/맞춤 금융 상품 기획 드
   2. 정답이 없는 비지도학습
   3. 대표 알고리짐 : `Apriori`, `FP-Growth`

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.2 연관규칙 분석 </span>
1. **연관 규칙의 형태**
   - `if-then`의 형식 : "A를 구매했을 때, B 또한 구매할 것이다"
   - `A -> B`로 표현 (A : 조건절, B : 결과절)
   - 아이템 집합 : 조건절 or 결과절을 구성하는 아이템들의 집합 (**여러 아이템** 가능하나, **상호 배반**이어야 함!)

2. **주의사항** :  <span style="background-color:yellow">**상관관계를 파악할 뿐, 인과관계는 알 수 없다!!!**</span>
   - [이유 1] 실험 X 관찰데이터 O
   - [이유 2] 방정식 = 그냥 수학적 비율임
   - [이유 3] 상관관계 ≠ 인과관계
   - 즉, `A -> B`의 신뢰도가 0.8이라는 것은, "*A를 산 사람 중 80%는 B도 샀다*"

3. **비교 : 연관분석 vs 협업 필터링**
    > 연관 규칙 분석은 관점에 따라, **추천 시스템의 하위개념 or 초기 개념**으로 분류되기도 함!
    <br>

   - 협업 필터링의 장점 : **복잡한 수치** 계산 가능 (평가점수, 구매횟수 등)
   - 연관 분석의 장점 : 상품이 포함된 **비율만 계산** 가능 but 다양한 **세트상품** 구성 가능!

    |     | **상품기반 협업 필터링** | **연관분석(장바구니 분석)** |
    | --- | --- | --- |
    | **연산 방식** | 상관관계 기반(동일 방향으로 증감하는 강도) | 교집합 기반(장바구니에 같이 담겨있는 비율) |
    | **알고리즘 구조** | 상품마다 타 상품과 1:1 계산(특정상품의 구매데이터를 다른 상품의 구매데이터와 비교) | 상품마다 주문내역을 1:n 계산(특정상품이 포함된 구매데이터에서 다른 상품의 비율을 계산) |
    | **특징** | 사용자의 개인 선호도(평가점수, 구매횟수 등)에 따라 선호 상품 추출 가능 | 상품과 연관성이 높은 1개이상의 상품을 세트로 추출 가능 |
    | **추천 방식** | 타 상품의 유사도 순위대로 노출 | 연관지표가 높은 세트상품(1개이상) 노출 |

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 2. 연관 분석의 주요 개념 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.1 적절한 연관규칙의 조건 </span>
> [조건1] 일정 이상의 **지지도** : 두 품목이 함께 구매된 경우의 수가 일정 수준 이상<br>
> [조건2] 일정 이상의 **신뢰도** : 품목 A 포함 거래 중 품목 B를 구입하는 경우의 수가 일정 수준 이상

1. <span style="background-color:yellow">**지지도**</span>
   1. **개념** : 전체 거래항목 중 품목 A, B가 동시에 포함되는 거래의 비율
      1. 엄밀히 말하면 조건절이 발생하는 확률이라 함! / 근데 연관분석 알고리즘에서는 $P(A \cap B)$ 사용
   2. **낮은 지지도의 의미**
      1. 유연일 수 있는 규칙 포함 (낮은 통계적 신뢰도)
      2. 실행가능한 인사이트 도출의 어려움
   3. **의의** : 도출한 규칙이 *현실에서 의미 있는지*를 판단하는 첫번째 기준!

2. <span style="background-color:yellow">**신뢰도**</span>
   1. **개념** : 품목 A를 포함하는 거래 중 품목 A, B가 동시에 포함되는 거래의 비율
   2. 낮은 신뢰도의 의미
      1. 의미없는 연관 규칙
      2. 추천 실패 가능성이 높음

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.2 연관 분석의 평가 측도 </span>
1. <span style="background-color:yellow">**향상도(Lift)**</span>
   1. **개념** : [품목 A가 주어지지 않았을 때 품목 B의 확률] 대비 [품목 A가 주어졌을 때의 품목 B의 확률]의 증가 비율
      1. 즉, A&B가 우연히 함께 등장활 확률, 실제 함께 등장한 확률 비교
   2. 특징
      1. 독립일 때 기댓값
         1. A&B가 독립이면, A를 샀다는 사실이 B를 살 확률에 영향 X (lift=1)
         2. A&B가 독립이면, A(혹은 B)를 샀다는 사실이 B(혹은 A)를 살 확률에 영향 O (lift<1 or lift>1)
      2. 실제 동시 발생 확률
         1. 실제로 관측한 A와 B가 함께 구매되는 확률이, A&B가 함께 등장할 확률의 기댓값(우연적 기회)보다
            1. 크면 -> 양의 연관성 (A 사면 B도 잘 삼)
            2. 작으면 -> 음의 연관성 (A 산 사람은 B 잘 안 삼)

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.3 연관 분석의 평가 종합 </span>
> 규칙의 효과성을 판단하려할 때, **지지도, 신뢰도, 향상도 모두 클 경우에만** 결론을 내릴 수 있음.
<br>

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 4. 연관 분석 알고리즘 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.1 Apriori 알고리즘 </span>

1. **등장 배경** : 연관규칙을 찾기 위해 가능한 모든 조합을 찾는 방식의 비효율성

2. **개념** : 상위 조합부터 차례대로 스캔하며, 특정 조합이 자주 발생하지 않으면, 그 결과로 만들어진 후속 조합들까지 모두 후보에서 배제

3. **장단점**
   1. 하나의 조합만 검사하고도, 파생된 다른 조합들까지 후보에서 배제할 수 있음 -> 시간, 연산량 감소에 효과적!
   2. 데이터가 커질 수록 속도가 느려지긴 함

4. **프로세스**
   1. 최소 지지도 
   2. **단일 항목 집합 생성** : 최소 지지도를 넘는 항목들만 2단계로!
   3. **2개 항목 집합 생성** : 조합들의 지지도를 계산 하고 최소 지지도 넘지 않는 조합 제거!
   4. **더 큰 항목 집합 생성** 
      1. 남은 조합들을 바탕으로 3개 이상의 항목으로 구성된 조합 생성
      2. 이 중 최소 지지도를 넘지 않는 조합 제거, 새 조합 생성할 수 없을 때까지 3단계 반복
   5. **최종 빈발 항목 집합 생성**
      1. 마지막으로 남은 빈발 항목 집합들을 바탕으로 연관 규칙 생성
      2. 각 규칙의 지지도, 신뢰도, 향상도 계산해 가장 유용한 규칙 계산
         1. **빈발 항목 집합** : 일정한 최소 지지도를 넘는 항목들의 조합

5. **코드**
    ```python
    from mlxtend.frequent_patterns import apriori
    from mlxtend.frequent_patterns import association_rules

    # 최소 지지도를 설정하여 자주 함께 구매되는 상품 조합을 찾기
    frequent_itemsets = apriori(df, min_support=0.06, use_colnames=True)

    # 연관 규칙을 추출하고, 신뢰도 0.8 이상의 규칙만 추출
    rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.8)

    # 향상도를 기준으로 내림차순 정렬하여 상위 규칙을 확인
    top_rules = rules.sort_values(by='lift', ascending=False).head()
    ```

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 4.2 FP Growth 알고리즘 </span>

1. 등장 배경 : 데이터가 커질 수록 속도가 느려진다는 Apriori의 단점

2. **개념** : 빈발 품목 집합을 효율적으로 찾아냄. FP-Tree 구조를 이용해 Apriori를 효과적으로 구현, Tree, Array, Linked_List를 합친 구조

3. **장단점**
   1. 데이터를 두번만 스캔하면 도미 -> 훨씬 빠름
   2. 대용량 데이터셋에서 메모리를 효율적으로 사용 X, 설계가 어려움, FP-Tree가 만들어져야 지지도 계산 가능

4. **프로세스**
   1. 데이터셋 확인
   2. 빈도순 재정렬 -> f-list 생성
   3. **FP-Tree 생성** 
      1. 각 거래를 노드로 표현
      2. 동일한 항목은 기존 노드에 연결해 빈도 증가, 새로운 항목은 새로운 노드로 추가
   4. 빈발 항목 집합 추출 (FP-트리를 바탕으로 특정 항목 기준의 부분 트리를 만듦)
   5. 연관 규칙 도출, 지지도/신뢰도/향상도 계산해 가장 유용한 규칙 선택

5. **코드**
    ```python
    from mlxtend.frequent_patterns import fpgrowth
    from mlxtend.frequent_patterns import association_rules

    # 최소 지지도를 0.06으로 설정하여 빈발 항목 집합을 찾기
    frequent_itemsets = fpgrowth(df, min_support=0.06, use_colnames=True)

    # 연관 규칙을 추출하고, 신뢰도 0.8 이상의 규칙만 추출
    rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.8)
    ```

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 1. 문제 정의와 구조화 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.1 문제 정의 </span>

1. **개념 정의**
   1. 현상 v.s. 문제
      1. <span style="background-color:yellow">**문제** : 현재의 상태와 바라는 상태(=목표) 사이의 격차 -> 해결해야할 질문</span>
      2. 현상 : 관찰된 사실
   2. 즉, 나쁜 상황을 나열하는 것을 넘어, 의사결정을 내려야하는 **구체적인 질문으로 치환**된게 문제!

2. **<span style="background-color:yellow">SMART 원칙</span>**
   1. **S**pecific : 무엇을 어디까지 풀지 명확히 정의 -> 구체적 & 범위가 명확
   2. **M**easurable : 성공 여부를 숫자로 판단 가능 -> 수치로 측정 가능
   3. **A**ction-oriented : 행동을 이끌어내고 성과에 집중하는 것 -> 실행지향적, 결과중심적
   4. **R**elevant : 조직의 목표와 관련이 있어야 함 -> 관련성
   5. **T**ime-bound : 마감 기한과 의사결정 시점이 존재 -> 시한성

    > 즉, 좋은 문제 정의를 위해 SMART 원칙을 사용해 **좋은 목표**를 설정한다고 생각해도 됨

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 1.2 문제 구조화 </span>

> 커다란 문제를 해결 가능한 **하위 문제들로 분해**하는 과정!

#### 1. **MECE** (Mutually Exclusive, Collectively Exhaustive)
1. **개념** : 어떠한 개념을 누락과 중복 없이 전체를 포괄하도록 분류하는 것
   - Mutually Exclusive (상호 배타적) : 중복 방지
   - Collectively Exhaustive (전체 포괄적) : 누락 방지

2. **적용 예시**
   1. 좋은 예
      - 생물학적인 성 구분 : 남성 / 여성
      - 소비자를 나이대로 구분 : 직장인 / 학생
   2. 나쁜 예
      - 소비자를 맘대로 구분 : 남성 / 여성 / 직장인 / 학생
        - => 중복이 생김

3. **좋은 MECE를 위한 규칙**
   1. **논리적 일관성** : 한 층위에서는 하나의 기준으로만 분류
   2. **통찰력** : 문제 해결에 도움이 되는 기준으로만 분류
   3. **상호배타성** : 분류한 문제들이 서로 겹치지 않아야 분석 결과가 명확

## <span style="background-color:#fae0f0; color:#c0156d; padding:4px; border-radius:5px;"> 2. 지표 </span>

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.1 지표의 개념 </span>

1. **개념** : **<span style="color:blue">방향, 목적, 기준 따위를 나타내는 표지**
   1. 즉, 추상적인 개념과 관련되는 지수, 척도 등을 측정하기 위해 숫자로 표현한 것

2. **표현 방식**
   1. 대표적 : 분모 대비 분자
      1. $ \text{metrics} = \frac{\text{numerator}}{\text{denominator}} $
    <br>

      - 효율성 : 산출/투입 (ex.노동생산성 = 생산량/노동시간)
      - 효과성 : 달성 결과/목표 (ex.목표달성률 = 실제 매출/목표 매출)
      - 수익성 : 이익/자원 (ex.ROE=순이익/자기자본)
      - 성장성 : 증가분/기존 규모 (ex.매출 성장률=(당기매출-전기매출)/전기매출)
      - 안정성 : 보유자산/부채 (ex.부채비율 = 부채/자기자본)

    2. 예외 구조
       1. 절대치 지표 : 매출액, 직원 수, 생산량
       2. 지수형 지표 : 고객만족지수(CSI), 순추천지수(NPS)
       3. 복합 지표 : 균형성과표(BSC)처럼 여러 지표를 묶어서 본 것

### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.2 지표 정의 </span>

1. **<span style="color:blue">Proxy Metric</span>**
   1. 등장 배경 : 분모 분자에 넣을 절대적인 수치를 구할 수 없을 때
   2. 개념 : 추상적이어서 프로덕트마다 나름의 가정과 정의, 논의가 필요한 지표 (활성화, 인게이지 등)
      1. 측정할 수 없어, 대체물(Proxy)을 정해 측정한다는 의미!
      2. 따라서 설정 자체에 많은 논의가 필요 + 정답이 별도로 존재 X
   3. 스티비의 사례
<br>
2. **<span style="color:blue">북극성 지표 L1/L2 Metrics</span>**
   - 북극성 지표 : 기업이 가장 중요하게 생각하는 지표
   - L1 지표 : 전사적인 고수준의 성과 지표 (전략 관점)
   - L2 지표 : L1 지표의 세부 사항을 측정하는 지표 (전술 관점)
   - KPI (Key Performance Indicator | 핵심성과지표)
   - OKR (Objective Key Results) : 도전적인 목표를 설정하고 핵심 성과를 도출해내는 것
<br>
3. **<span style="color:blue">AARRR</span> (해적 지표)**
   - 사용자 운영 과정에서 활용하는 데이터 분석 모델로, 사용자 증가를 위한 5가지 로드맵을 구성

   1. **Acquisition**
      1. 사용자가 여러 채덜을 통해 우리 서비스에 방문하는 것을 의미 -> 고객 획득
      2. "신규고객을 어떻게 획득할지", "어디로부터 획득할지", "주로 어떤 채널을 통해 획득할지"
      3. 지표 예시
         1. CAC (Customer Acquisition Cost): 유저 획득 비용
         2. CTR (Cost Per Click): 클릭 한 건당 얼마 지출
         3. CPC (Cost Per Install): 인스톨 당 과금되는 상품
         4. CPL (Cost Per Lead): 유저 행동 당 드는 비용
         5. ROAS (Return On Ads Spending): 광고로 인한 매출액/광고비

   2. **Activation** (사용자 활성화)
      1. 사용자가 우리 서비스에 처음 유입되어 주요 기능을 직접 활용하는 것
      2. 예시 : 회원가입 3일 내 최소 1건 주문한 신규 사용자, 온라인 접속 시간 등
      3. 4단계
         1. Sign-up: 유저가 제품에 들어온 시점
         2. Setup Moment: 유저가 제품의 핵심가치를 경험할 준비 완료
         3. A-HA! Moment: 유저가 핵심가치를 처음 경험한 순간
         4. Habit Moment: 유저가 제품 사용에 대한 습관 형성

   3. **Retention** (유지 단계)
      1. 신규사용자를 장기 사용자로 전환시키는 것
      2. "장기 고객을 어떻게 양성할 것인지", "고객의 지속적인 구매행위를 증가시킬 수 있을지"
      3. 예시 : 익일 유지율, 2주차 유지율, 월 유지율, EDM 메일 개설율 등
      4. 지표 예시
         1. DN Retention: N일 안으로 재접속하는 경우
         2. MAU/DAU/WAU: 일/주/월마다 방문하는 횟수
         3. Stickness(DAU, MAU): 해당기간 동안 활성화 수 (*점성)

   4. **Revenue** (수익 단계)
      1. 사용자가 여러분의 제품, 서비스를 유료 구매하기로 결정한 단계
      2. 사용자의 소비를 유도해 매출 높이기 -> 최대한 사용자가 소비하게끔 / 매출을 늘리는 행위를 하도록 유도
      3. 주요 목표 : 고객의 평생 가치(LTV)를 지속적으로 높여가고, 고객 유지 비용은 계속 줄여나가야 한다는 것
      4. 지표 예시
         1. PU(Paying User): 전체 사용자 중에 결제를 한 사용자 비율
         2. Average Check: 구매 1회당 평균 금액 (Revenue/Transaction)
         3. ARPU(Average Revenue Per User): 인당 평균 매출
         4. ARPPU(Average Revenue Per Paying User): 결제자당 평균 매출
         5. LTV(Life Time Value): 유저가 진입부터 이탈까지 누적 수익

   5. **Referral** (추천 단계)
      1. 사용자가 제품/서비스에 만족했을 때 타인에게 추천하는 것
      2. 목표 : 충성도 높은 사용자를 invisible 브랜드 홍보자로 만드는 것
      3. 주요 목표 : 상품을 잠재 고격등이 어떻게 추천하게 만들지?          


### <span style="background-color:#e6fae0; color:#42c015; padding:4px; border-radius:5px;"> 2.3 지표 분석 방법 </span>

1. **<span style="background-color:yellow">코호트 (Cohort) 분석</span>**<br>
   1. **정의 및 개념**
      1. 동질 집단(Cohort)끼리 비교 및 대조를 진행하는 분석 방식
         1. 동일 집단 : 특정 시간 범위 내에서 공통된 특성이나 경험을 공유하는 사용자 집단
      2. 분석 전 데이터 세트를 관련 그룹으로 나누는 일종의 행동 분석법
      3. 시간(날짜)에 기반하여 사용자 세그먼트를 분류하는 개념
      4. 분류 예시 : 가입 날짜, 특정 이벤트 발생 시간, 성별 등

   2. **세그먼트(Segment)와의 차이점**
      1. 세그먼트는 전체 고객을 '특정 기준'으로 단순히 나눈 결과물
      2. 코호트는 '동일한 기간'과 '동일한 특성'을 동시에 충족하는 사람들을 분류
      3. 세그먼트와 달리 코호트는 반드시 '기간'이라는 **시점 개념을 포함**해야 함

   3. **주요 용도**
      1. 주로 사용자의 유지(Retention)와 이탈(Churn)을 체크에 사용
      2. 대표적인 예시 지표 : 재구매율과 같은 리텐션 지표

   4. **코호트 분석 해석 방법**
      - **세로축 (코호트별 비교):** 동일한 기간이 경과한 시점에서 각 코호트 간의 리텐션을 비교 (ex. 신규 가입자 증가 추세 확인) 
      - **대각선 (시점별 비교):** 특정 달력 날짜를 기준으로 전체적인 지표 변화를 확인하며, 전반적인 구매율 하락 등의 추세를 파악
      - **가로축 (시간 경과에 따른 변화):** 특정 코호트가 시간이 흐름에 따라 어떻게 변하는지 관찰 (ex. 가입 후 시간 경과에 따른 앱 활성률 저하 확인)
      - **인사이트 도출:** 가입자는 느는데 활성률이 떨어진다면, 획득은 원활하나 활성화 및 리텐션에 문제가 있다고 판단
<br>

2. **<span style="background-color:yellow">퍼널(Funnel) 분석</span>**<br>
   1. **개념 및 특징**
      1. 유저가 서비스에 접속하여 이탈할 때까지의 모든 행동 과정을 추적
      2. 사용자가 아래 단계로 갈수록 이탈하여 수가 줄어드는 모습이 깔때기와 유사
      3. 고객 여정을 시각화 -> 어느 단계에서 병목 현상이 발생하는지 파악

   2. **관련 용어**
      1. DA (Display Advertising): 유입을 유도하는 디스플레이 광고
      2. CRM (Customer Relationship Management): 고객 관계 관리를 통한 리텐션 강화

   3. **개선 방법 및 전략**
      1. **유입률 높이기:** 스타트 지점에 도달하는 절대적인 인원수를 늘림
      2. **전환율 높이기:** 각 단계에서 다음 단계로 넘어가는 비율을 개선함
      3. **단계 간소화:** 불필요한 퍼널 단계를 삭제하여 이탈 방지함 (예: 소셜 로그인을 통한 회원가입 절차 압축)
      4. **순서 변경 및 단계 추가:** 유저 경험 최적화를 위해 퍼널의 흐름을 재설계함
      5. **코호트별 분석:** 집단마다 성향이 다르므로 코호트로 나누어 퍼널을 세부 분석함
