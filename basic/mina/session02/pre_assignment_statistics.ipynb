### <span style="color: #000000;background-color: #ffdba4">**1. 통계란?**</span> 
<U><span style="color: #ffdba4">**다양한 비즈니스와 경제 상황을 이해하는 데 도움을 주는 숫자로 표현된 사실**</span> </U><BR>
*ex) 중간 주택 가격, 평균 출근 시간, 평균 연봉, 학위 취득 비율, 산업평균지수...*  


##### <span style="color: #000000;background-color: #ffc2e5">   **1-(1) 통계학**</span>
**자료를 수집, 분석, 표현, 해석하는 기술이자 과학** 
- 대상
    - 전 인구(모집단)을 대상으로 분석  
    - 일부 샘플(표본집단)을 추출하여 분석
- 방법론
    - 수치와 시각화를 통해 샘플을 요약하여 표현 -> <U>**기술통계**</U>  
    - 추론과 검정을 통해 샘플로부터 모집단을 추론 -> <U>**추론통계**</U>

##### <span style="color: #000000;background-color: #ffc2e5">   **1-(2) 자료의 척도**</span>
**자료**</span> : 설명과 해석을 위해 수집,분석,요약되는 사실과 수치  
- 양적 자료 : 수치로 표현되는 자료  
    - 구간(등간척도) - 값들 사이에 순서 존재, 연산(곱셈,나눗셈 X) 가능  
    - 비율척도 - 절대영점과 임의의 측정단위로 구성, 동간성 유지   
- 질적 자료 : 수치로 표현되더라도 연산의 의미를 가지지 못하는 자료  
    - 명목척도 - 레이블이나 이름으로 요소의 속성을 식별  
    - 서열척도 - 명목형 + 항목 간 순서 또는 순위를 나타냄

### <span style="color: #000000;background-color: #ffdba4">**2. 기술통계**</span> 
<U><span style="color: #ffdba4">**데이터를 있는 그대로 설명하고 이해하는 데 중점을 둔 통계 방법**</span> </U><BR>

##### <span style="color: #000000;background-color: #ffc2e5">   **2-(1)기술통계의 3요소**</span>
- 중심화 경향  
- 자료의 산포  
- 자료의 정리 및 시각화

##### <span style="color: #27baff">   **i. 중심화 경향**</span>
<U>**데이터의 중심적인 경향을 나타내는 기술 통계량**</U><BR>
- <U>평균(mean)</U>  
    - *데이터 값의 총합/데이터의 개수*  
    - 데이터의 중심 경향성을 나타내는 가장 대표적인 지표  
    - 총합 사용 -> 이상치에 민감  
        - <span style="color: #ff2b2b">   **절사평균**</span> : 이상치 민감 문제를 보완하기 위해 등장한 개념 (데이터의 양끝을 일정 비율 잘라낸 후 평균 계산)
- <U>중앙값(median)</U>
    - 데이터를 크기 순으로 일렬 정렬했을 때 **한가운데에 위치하는 값**
    - 데이터의 중심 경향성을 나타내는 대안적인 지표    
    - 이상치에 덜 민감  
- <U>최빈값(mode)</U>
    - **데이터에서 가장 자주 나타나는 값**  
    - 주로 이산형 데이터

##### <span style="color: #27baff">   **ii. 자료의 산포**</span>
<U>**데이터가 퍼져있는 정도를 설명하는 기술 통계량**</U>  
- <U>범위(range)</U>
    - 데이터의 **최댓값과 최솟값의 차이**를 나타내는 지표(데이터의 전체 변동 범위)  
    - 양쪽 극단값 이용 -> 이상치에 매우 민감
- <U>분산(variance)</U>
    - 데이터 값들이 평균으로부터 얼마나 멀리 퍼져있는지를 나타내는 지표  
    - 분산값이 크다 = 데이터가 많이 퍼져있다
- <U>표준편차(standard deviation)</U>
    - 분산에 제곱근을 취한 값(단위 일치)
- <U>사분위수(quartile)</U>
    - 데이터의 표본을 네 개의 동일한 크기로 나눈 값
        - Q1 = 하위 25%  
        - Q2 = 하위 50%(중앙값)  
        - Q3 = 하위 75%  
- <U>사분위간 범위(IQR)</U>
    - Q3-Q1  
    - 아래쪽 이상치 : $Q1-1.5*IQR$보다 작은 값  
    - 위쪽 이상치 : $Q3+1.5*IQR$보다 큰 값  
- <U>변동계수(coefficient of variation)</U>
    - 데이터의 **상대적인 변동성**을 나타내는 통계적 지표  
    - *변동계수 = (표준편차/평균)x100*  
- <U>첨도와 왜도</U>
    - 첨도 : 확률분포의 꼬리가 두꺼운 정도를 나타내는 척도  
    - 왜도 : 비대칭도, 실수 값 확률 변수의 확률 분포 비대칭성을 나타내는 지표  
        - 왜도>0 : 오른쪽으로 긴 꼬리 -> 자료가 왼쪽에 더 많이 분포되어 있다.  
        - 왜도<0 : 왼쪽으로 긴 꼬리 -> 자료가 오른쪽에 더 많이 분포되어 있다.
        - if 평균==중앙값 : 왜도=0

##### <span style="color: #27baff">   **iii. 자료의 정리 및 시각화**</span>
<U>**히스토그램, 박스플롯, 파이차트 등 시각적으로 나타냄**</U><BR>

### <span style="color: #000000;background-color: #ffdba4">**3. 확률분포**</span>
<U><span style="color: #ffdba4">**확률변수가 특정한 값을 가질 확률을 나타내는 함수**</span> </U><BR>
*확률 변수 : 랜덤한 세계에서 어떤 특정 사건이 발생할 때 나타날 수 있는 모든 경우의 수를 특정 값으로 대응시키는 함수*
- 이산확률분포 : 확률변수가 이산적인(유한한) 값을 가지는 경우
    - *베르누이분포,이항분포,기하분포,음이항분포,초기하분포,포아송분포*
- 연속확률분포 : 확률변수가 연속적인(무한한) 값을 가지는 경우
    - *정규분포,T분포,F분포,감마분포,지수분포,카이제곱분포,균등분포*

##### <span style="color: #000000;background-color: #ffc2e5">   **3-(1) 대표적인 분포 모형**</span>
- 정규분포  
- Z분포  
- T분포
- 카이제곱분포  
- F분포

##### <span style="color: #27baff">   **i. 정규분포**</span>
- 형태
    - 종모양  
    - 평균을 기준으로 대칭  
    - 표준편차가 클수록 곡선이 평평해짐
- 매개변수
    - 평균, 표준편차
- 특징
    - 가장 잘 알려진 연속확률분포  
    - 많은 자연,사회환경이 정규분포를 따름
<BR>
- 모집단의 분포가 정규분포를 따름 -> 표본평균의 분포도 정규분포를 따른다.
- 그러나 표본의 크기가 충분히 크다면(n>=30) 모집단의 분포와 상관없이 표본평균의 분포는 정규분포를 따른다 <span style="color: #ff0000"> (중심극한정리)</span>

##### <span style="color: #27baff">   **ii. Z분포(표준정규분포)**</span>
**평균이 0이고 표준편차가 1이 되도록 맞춘 표준적인 정규분포**
- 많은 데이터가 따르는 가장 대표적인 확률분포  
- <U>표본분포</U> : 정규분포를 따르는 모집단에서 추출한 표본통계량들이 따르는 분포
    - T분포, 카이제곱분포, F분포

##### <span style="color: #27baff">   **iii. T분포**</span>
- 형태
    - 종 모양, 정규분포보다 꼬리가 두꺼움  
    - 평균을 기준으로 대칭  
    - 자유도가 커짐에 따라 <U>표준정규분포 N(0,1)에 수렴</U>
- 매개변수
    - 자유도(df) :*얼마나 많은 정보가 독립적으로 변할 수 있는가*를 나타내는 숫자
- 특징
    - 표본으로 모평균(혹은 모평균 간의 차이)을 추정할 때 사용되는 분포  
    - 표본 크기가 작을 때 정규분포보다 더 정확

##### <span style="color: #27baff">   **iv. 카이제곱분포**</span>
- 형태
    - 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포  
    - 자유도 증가 -> 중심 : 오른쪽으로 이동, 모양 : 더 대칭적, 정규분포에 가까워짐
- 매개변수
    - 자유도(df)
- 특징
    - 독립적인 표준 정규 분포를 따르는 변수들의 제곱합  
    - **표본분산의 분포와 관련된 지표**  
    - 주로 독립성 검정, 적합도 검정 등에 사용

##### <span style="color: #27baff">   **v. F분포**</span>
- 형태
    - 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포  
    - 자유도 증가 -> 중심 : 오른쪽으로 이동, 모양 : 더 대칭적, 정규분포에 가까워짐
- 매개변수
    - 자유도(df1,df2)
- 특징
    - 두 독립적인 카이제곱분포의 비율   
    - 두 집단 간의 분산 차이를 비교하는 데에 사용

##### <span style="color: #000000;background-color: #ffc2e5">   **3-(2) 분포모형을 알아두어야 하는 이유**</span>
1. 가설검정을 위한 검정통계량 산출에 사용
2. 표본추출의 과정에서 인위적인 조작이 있었는지 판단

### <span style="color: #000000;background-color: #ffdba4">**4. 추론통계 - 추정**</span> 

##### <span style="color: #000000;background-color: #ffc2e5">   **4-(1) 추론통계란?**</span>
<U><span style="color: #ffc2e8">**수집한 데이터를 기반으로 미래의, 혹은 조사하지 못한 전체 정보를 추론하고 결론을 도출하는 방법을 다루는 분야**</span> </U><BR>
- **통계적 가설검정 / 신뢰구간** 등  
- <span style="color: #ff0000"> **추정**</span>
    - 모집단을 대표할 수 있는 일부를 표본으로 추출하여 분석한 통계량으로 모집단의 **모수**를 예측하는 과정
    - 추정량 : 표본정보에 의존하는 확률변수, 모수를 추정하는 데 사용되는 표본 통계량  
    - 추정치 : 표본이 추출된 후 표본자료를 이용하여 계산한 통계량의 특정한 값
- <span style="color: #ff0000">**검정**</span>
    - 표본을 추출하여 얻은 표본통계량으로 모집단 모수에 대한 새로운 예상,가설 등이 옳다고 판단할 수 있는지를 평가하는 과정

##### <span style="color: #000000;background-color: #ffc2e5">   **4-(2) 점추정**</span>
- 어떤 모수를 표본에 기초하여 계산된 **하나의 값**으로 추정하는 것  
- 오차에 대한 정보가 전혀 고려되지 않음
##### <span style="color: #000000;background-color: #ffc2e5">   **4-(3) 구간추정**</span>
- 모수의 참값이 포함되리라고 기대하는 **추정치를 일정한 범위와 신뢰도**를 가지고 나타내는 방법  
- 어느 정도 오차를 포함 -> 일반적으로 널리 사용

### <span style="color: #000000;background-color: #ffdba4">**5. 추론통계 - 검정**</span> 

##### <span style="color: #000000;background-color: #ffc2e5">   **5-(1) 가설 검정의 절차**</span>
**가설 검정이란?**  
<U>표본의 정보를 사용해서 가설('모집단의 실제값은 이럴 것이다')의 합당성 여부를 판정하는 과정</U>  
*가설 설정 -> 유의 수준 설정 -> 검정통계량 산출 -> 가설 기각/채택 판단*


**① 가설 설정**  
귀무가설(H0)과 대립가설(H1) 설정
- 귀무가설 : 기각하고자 하는 명제  
- 대립가설 : 주장하고자 하는 명제  

**② 유의수준 설정**  
<U>유의수준</U> : 연구자가 결과를 판단할 때 허용할 수 있는 오류의 최대치, 주로 0.05(5%)로 설정
- 1종 오류(α) - 귀무가설이 실제로 참인데 기각할 확률  <span style="color: #ff0000"> <- 더 치명적, **유의수준**</span>  
- 2종 오류(β) - 귀무가설이 실제로 거짓인데 채택할 확률  

**③ 검정통계량 산출**  
- 검정통계량 : 가설 검정의 결과를 판단하는 데 사용되는 값  
- 데이터의 분포나 특징에 따라 Z값, T값, F값 등을 정함  

**④ 가설 기각/채택 판단**  
- <U>유의확률(p-value)</U>
    - 가설 검정에 쓰일 수 있는 또 다른 척도  
    - 유의확률 < 유의수준 : 귀무가설 기각
    - 유의확률 > 유의수준 : 귀무가설 채택

##### <span style="color: #000000;background-color: #ffc2e5">   **5-(2) 데이터의 자료형에 따른 검정 방법**</span>


##### <span style="color: #27baff">   **i. [수치형-수치형] (피어슨) 상관분석**</span>
**피어슨 상관 분석**  
- 두 변수간 피어슨 상관계수가 얼마나 유의미한지를 검정하는 방법  
- 가설  
    - 귀무가설 : 두 변수간 선형 상관관계가 존재하지 않는다.  
    - 대립가설 : 두 변수간 선형 상관관계가 존재한다.  
- 공분산 : 두 변수가 함께 어떤 방향으로 변화하는지 & 그 크기를 표현하는 데 사용
    - cov(x,y)>0 -> x가 커지면 y도 커짐  
    - cov(x,y)<0 -> x가 커지면 y는 작아짐  
    - cov(x,y)=. -> 선형적인 관계는 없으나 비선형적인 관계는 존재할 수 있는 관계
- 상관계수 : 두 변수 간의 선형적인 관계를 -1과 1 사이의 값으로 나타내는 통계적 지표

##### <span style="color: #27baff">   **ii. [범주형-수치형] F-test(ANOVA), T-test**</span>
**1. T-test : 범주가 2개일 때**  
- *ex) 남/여, 인사이트 학회원인 사람/아닌 사람*  
- t값 : 두 평균 간 차이를 표준 오차로 나눈 값  
    - 보통 절댓값 2를 기준으로 그보다 크면 차이가 있으며, 절댓값이 클수록 차이가 크다
- 가설  
  - 귀무가설 : 두 그룹의 평균에는 차이가 없다  
  - 대립가설 : 두 그룹의 평균에는 차이가 있다  

**2. F-test(ANOVA) : 범주가 3개 이상일 때**
- F값 = $집단 간 분산 / 집단 내 분산$  
    - 보통 2-3 이상이면 차이가 있으며, 그 값이 클수록 차이가 크다
- 가설
    - 귀무가설 : 각 MBTI 그룹 별 평균 수명은 차이가 없다.
    - 대립가설 : 각 MBTI 그룹 별 평균 수명은 차이가 있다.

##### <span style="color: #27baff">   **iii. [범주형-범주형] Chi2-test**</span>
**Chi2-test**  
*Suppose* [인사이트 전체 학회원 100명을 대상으로 '경영학과 여부'와 '갈색머리 여부'에 대한 조사를 진행했다.]
- 가설
    - 귀무가설 : 경영학과 여부와 갈색머리 여부는 관련이 없다  
    - 대립가설 : 경영학과 여부와 갈색머리 여부는 관련이 있다  
- 카이제곱 통계량 
    - $x^2=∑ (O_i-E_i)^2/E_i $
        - $x^2$ = chi squared   
        - $O_i$ = 관측빈도  
        - $E_i$ = 기대빈도
    - 기준 : 자유도의 약 2배

##### <span style="color: #27baff">   **iv. [수치형-범주형]**</span>
- 범주형을 target, 수치형을 feature로 두고 모델을 만든 다음, 그 모델의 개수를 유의성 검정하여 그에 대한 p값을 보고 판단하는 방법 사용
