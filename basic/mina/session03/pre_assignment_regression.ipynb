### <span style="color: #000000;background-color: #ffdba4">**1. 머신러닝과 모델링**</span>
 


##### <span style="color: #000000;background-color: #ffc2e5">   **1-(1) 머신러닝(Machine Learning)이란?**</span>  
**인공지능의 한 분야로, 모델이라는 소프트웨어를 학습하여 새로운 데이털르 예측하거나 결정을 내릴 수 있도록 하는 기술**  
- 특징
    1. 데이터에서 패턴 발견
    2. 발견한 패턴을 바탕으로 새로운 데이터에 대해 예측
- 종류  

|종류|정답 여부| 방식 | 예시|    
|---|---|---|---|  
|지도학습| O | 기계의 예측값이 데이터의 정답과 같아지도록 학습| 회귀,분류|  
|비지도학습|X|데이터의 패턴 혹은 유사성 학습 | 군집화|  
|강화학습 | 보상| 시행착오를 반복하여 보상을 최대화하는 행동 학습|알파고, 자율주행|  



##### <span style="color: #000000;background-color: #ffc2e5">   **1-(2) 모델링(Modeling)**</span>  
##### <span style="color: #27baff">   **i. 모델이란?**</span>
**학습의 결과물, 학습을 마친 인공지능 그 자체**  
- 데이터를 학습하며 찾아낸 패턴이 저장된 상태
- 새로운 데이터를 넣었을 때 답을 내놓는 예측기 역할  
##### <span style="color: #27baff">   **ii. 모델링이란?**</span>
**예측기(공식)를 만들어가는 전체적인 과정**  
- 주어진 데이터 속에서 스스로 패턴을 찾아내 오차가 적은 최적의 공식을 완성할 때까지 학습하는 과정  
##### <span style="color: #27baff">   **iii. 모델링의 과정**</span>
1. **설계** : 데이터의 특성에 맞추어 어떤 알고리즘으로 학습할지 결정  
2. **훈련** : 데이터를 기계에게 계속 보여주며 정답을 맞히도록 수정
3. **검증** : 훈련시킨 모델이 새로운 데이터에서도 정답을 잘 맞히는지 테스트

##### <span style="color: #000000;background-color: #ffc2e5">   **1-(3) 지도 학습이란?**</span> 
**정답 레이블을 함께 주고 학습을 시키는 방법**  
*예측값과 정답이 최대한 같아지는 방향으로 학습*
- 과정  
    *데이터: labeled data의 형태*  
    1. labeled data : 독립변수(입력 변수)와 종속변수(대상 변수)가 모두 존재하는 데이터
    2. labeled data를 **training set**(훈련 시 쓰이는 데이터)과 **test set**(검증 시 쓰이는 데이터)으로 분류
    3. training set을 통해 모델 학습
        - 독립변수를 통해 추정한 예측값이 주어진 종속변수와 같아지도록 훈련

- 종류
    - **회귀**
        - 주어진 데이터를 기반으로 정답을 잘 맞히는 함수를 찾는 문제
        - 종속변수가 <U>연속형</U>
    - **분류**
        - 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별하는 것
        - 종속변수가 <U>범주형</U>

### <span style="color: #000000;background-color: #ffdba4">**2. 회귀**</span>
<U>**데이터에서 패턴을 찾아내 미래 값을 예측하는 것**</U>
- 종속변수 : 연속적
- 선형회귀 / 비선형회귀 / 규제회귀 / 다항회귀 등

##### <span style="color: #000000;background-color: #ffc2e5">   **2-(1) 선형회귀**</span>
##### <span style="color: #27baff">   **i. 선형회귀란?**</span>
- 데이터를 가장 잘 설명하는 <U>회귀선($y=wx+b$)</U>을 찾는 과정
    - 회귀선 : 현재 데이터 요약 + 미래 값 예측
- 단순선형회귀(x가 한개) / 다중선형회귀(x가 여러개)

##### <span style="color: #27baff">   **ii. 가중치와 편향의 이해**</span>
**기본 식 : $y=wx+b$**  
- 가중치 $w$(weight)
    - 독립변수 $x$가 종속변수 $y$에 주는 영향력의 크기(기울기)  
    - = 회귀계수
- 편향 $b$(bias)
    - 독립변수 $x$가 0일 때 나오는 기본값(y절편)  
    - 모델이 데이터에 더 유연하게 최적화될 수 있도록 위아래로 움직여주는 역할

##### <span style="color: #27baff">   **iii. 단순선형회귀**</span>
- <U>**하나의 독립변수**</U>만 존재하는 선형회귀  
- 데이터를 <U>**직선 형태**</U>로 표현하는 경우 
- 데이터를 가장 잘 설명하는 식을 찾는 과정  
- 잔차를 줄여나가며 적절한 회귀계수와 편향을 찾는다

##### <span style="color: #27baff">   **iv. 잔차(Residual)**</span>
<U>**실제값과 예측값 간 차이**</U>
- 잔차를 줄여나가며 데이터에 최적화된 $w$와 $b$를 찾음 -> 최적의 직선 $y=wx+b$를 찾음

##### <span style="color: #000000;background-color: #ffc2e5">   **2-(2) 최적의 $w$와 $b$를 찾는 방법**</span>
##### <span style="color: #27baff">   **i. 최소제곱법(Least Squres Method)**</span>
<U>**잔차의 제곱의 합이 최소**가 되는 지점으로 회귀선을 구하는 방식</U>  
회귀선과 실제 데이터와의 차이를 최소화하기 위한 시도

##### <span style="color: #27baff">   **ii. 경사하강법(Gradient Descent)**</span>
<U>**잔차의 제곱의 합이 최소**가 되는 지점으로 회귀선을 구하는 방식</U>  
최소제곱법으로 구한 함수가 복잡하여 최솟값을 구하기 어려울 때 사용  
<BR>
<span style="color: #2bba66">**(0) 목적함수/손실함수/비용함수**</span>
|구분|정의|방향|
|---|---|---|
|목적함수|최적화하려는 전체 대상| 최대화/최소화|
|손실함수| 데이터 한개에 대한 오차|최소화|
비용함수|전체 데이터에 대한 평균 오차|최소화|  

<span style="color: #2bba66">**(1) 손실함수와 경사하강법**</span>  
- 손실함수 :  
$E(W, b) = \frac{1}{n} \sum_{i=1}^{n} [t_i - y_i]^2 = \frac{1}{n} \sum_{i=1}^{n} [t_i - (Wx_i + b)]^2$  
-> 목표 : **$E$를 최소화하는 $W,b$ 찾기**   
<BR>
- 경사하강법
    - 최솟값을 가지는 손실함수 그래프에서 최솟값을 a라고 하고 그래프 위의 한 점 b가 있다고 생각해보자  
    - ① b가 a의 왼쪽에 있을 경우
        - 기울기(b에서의 미분값) = 음수  
        - b - 기울기 > b → b가 오른쪽으로 이동  
    - ② b가 a의 오른쪽에 있을 경우
        - 기울기 = 양수  
        - b - 기울기 < b → b가 왼쪽으로 이동  
    - ③ 반복하다가 최솟값에 도달하면 미분값=0 즉, b-기울기 = 0 → 더이상 변화하지 않고 멈춤
    - <U>최솟값의 왼쪽의 값은 점점 오른쪽으로 이동하고, 최솟값 오른쪽의 값은 왼쪽으로 이동하며 최종적으로 최솟값에 도달하게 되는 원리</U>  
<BR>

<span style="color: #2bba66"> **(2) 학습률(Learning Rate)** </span>  - '점을 <U>얼만큼</U> 이동시킬 것인가'
- **적당히**가 중요!
- Learning Rate가 너무 작으면 : 모델이 수렴할 때까지 많은 시간 소요  
- Learning Rate가 너무 크면 : 모델이 수렴하지 못하고 발산  
<BR>

<span style="color: #2bba66"> **(3) Local Minima 문제** </span>  
- 작은 기울기에서는 업데이트가 조금씩 이루어지므로 Local Minima에서 나오기 어려움  
- 특히 딥러닝의 경우, 비용함수의 모양을 예측할 수 없기 때문에 주의 필요  
<BR>

<span style="color: #2bba66"> **(4) 해결법 - 모멘텀** </span>  
- 현재 기울기만 고려하는 기존 경사하강법의 보안 -> <U>이전의 기울기와 방향을 고려하도록 **관성**을 부여하자!</U>  
- 경사하강법이 이동하려고 하는 방향과 이전 단계의 이동을 반영한 모멘텀 방향을 고려해 실제 이동 방향을 정함
- 장점
    - 작은 기울기는 쉽게 넘어갈 수 있음     
    - 지역 최솟값 탈출에 용이


##### <span style="color: #000000;background-color: #ffc2e5">   **2-(3) 다중선형회귀**</span>
$$y = W_1 x_1 + W_2 x_2 + ... + W_n x_n + b$$
- 독립 변수가 **두개 이상**인 경우     
- 독립 변수가 2개인 경우 : 3차원 공간, 회귀선은 평면 -> 변수가 늘어나면 차원이 확장
- 최소제곱법/경사하강법 사용 -> 잔차 = 관측치와 평면 간의 차이  


##### <span style="color: #27baff">   **i. 다중공선성(Multicollinearity)**</span>  
<U>**독립 변수들 간의 강한 상관관계로 인해 회귀 모델이 불안정해지는 문제**</U>  
- 다중공선성이 높은 경우 : 어떤 독립변수가 종속변수에 얼마나 영향을 미치는지 정확하게 구분하기 어려움 -> 확인&처리 필요!

##### <span style="color: #27baff">   **ii. 다중공선성 확인 방법**</span> 
<span style="color: #2bba66">**(1) 상관계수**</span> - 변수 간의 상관관계의 정도를 나타낸 수치
- 상관관계
    - 1 ≤ $r$ ≤1    
    - $r$ > 0 : 양의 상관관계 - 한 변수가 증가하면 다른 변수도 증가
    - $r$ < 0 : 음의 상관관계 - 한 변수가 증가하면 다른 변수는 감소
    - $r$ = 0 : 두 변수 사이에 직접적인 연관성이 없다.
- 상관계수 확인 방법
    - heatmap 혹은 corr() 함수
    - 산점도로 시각화  
<BR>

<span style="color: #2bba66">**(2) VIF지수 (분산 팽창 인수)**</span> - R-Square를 사용하여 계산
- VIF가 높다 => 다중공선성이 존재한다
    - VIF = 1 : 해당 독립변수는 다른 변수와 상관관계가 전혀 없음
    - VIF < 5 : 일반적으로 다중공선성 문제가 없다고 판단
    - VIF > 5 : 다중공선성의 징후, 주의
    - VIF > 10 : 다중공선성 심각
<BR>

##### <span style="color: #27baff">   **iii. 다중공선성 대처 방법**</span> 
**1. 변수 제거**  
**2. 변수 변환**  
- 변수들을 더하거나 빼서 새로운 변수 생성
- 독립변수를 더하거나 빼더라도 문제가 없는 경우  

**3. 규제선형모델 활용**  
- 릿지, 라쏘, 엘라스틱넷 등

**4. PCA(주성분 분석)**  
- 데이터의 차원 축소
- 상관 있는 변수들을 묶어서 더 적은 수의 대표 축으로 바꾸는 것
 

##### <span style="color: #000000;background-color: #ffc2e5">   **2-(4) 규제선형모델**</span>
! 과적합 모델은 학습데이터에는 완벽히 맞지만 새로운 데이터에서는 성능이 떨어짐  

##### <span style="color: #27baff">   **i. 규제선형모델이란?**</span>  
- 모델이 **과적합 되지 않도록 규제**를 가하고자 등장한 모델
    -> 선형회귀 모델에서는 특성에 곱해지는 계수(기울기)의 크기를 조정
- 잔차 제곱합을 최소화하는 방식 -> 회귀계수가 쉽게 커져 과적합 발생 가능성 ↑
    - 회귀계수가 커진다 = 특정 독립변수에 과하게 의존한다  
- <U>**최소제곱법과 회귀 계수 값이 커지지 않도록 하는 방법이 서로 균형을 유지해야한다!**</U>

##### <span style="color: #27baff">   **ii. 규제선형모델의 종류**</span>
- L2 규제 : 회귀계수의 제곱에 대해 패널티를 부여하는 방식, *릿지 회귀*
- L1 규제 : 회귀계수의 절댓값에 대해 패널티를 부여하는 방식, *라쏘 회귀*
- L2 규제 + L1 규제 : *엘라스틱넷 회귀*


##### <span style="color: #27baff">   **iii. 릿지 회귀(Ridge, L2 규제)**</span>  
$$\text{Cost Function} = \text{RSS} + \alpha \sum_{j=1}^{p} w_j^2$$
- 회귀계수의 제곱합에 패널티 부여
- 모든 계수의 크기가 작아지지만, 변수가 제거되지는 않는다
- 상황
    - 예측 변수가 많을 때 : 고차원 데이터에서 모든 변수에 작은 가중치를 부여하고자 할 때 유용  
    - 다중공선성 존재할 때 : 변수 선택이 필요하지 않은 상황

##### <span style="color: #27baff">   **iv. 라쏘 회귀(Lasso,L1 규제)**</span> 
$$\text{Cost Function} = \text{RSS} + \alpha \sum_{j=1}^{p} |w_j|$$
- 회귀계수의 절댓값 합에 패널티 부여
- 작은 계수는 사라지고 중요한 변수만 남는다
- 상황
    - 예측 변수 수가 많고, 그 중 일부만이 실제로 중요할 때
    - 모델의 해석을 간단하게 유지하고자 할 때 : 불필요 변수 제거

##### <span style="color: #27baff">   **v. 엘라스틱넷(Elastic Net, L1+L2)**</span> 
$$\text{Cost Function} = \text{RSS} + \alpha_1 \sum_{j=1}^{p} |w_j| + \alpha_2 \sum_{j=1}^{p} w_j^2$$
- L1의 변수 선택 기능(일부 계수 삭제)과 L2의 규제 기능(큰 계수를 줄임)을 동시에 활용하는 모델
- 상황
    - 예측 변수 수가 많고, 그 중 중요한 변수를 선택하면서도 다중공선성을 관리할 때



### <span style="color: #000000;background-color: #ffdba4">**3. 모델 평가 방법**</span>


##### <span style="color: #000000;background-color: #ffc2e5">   **3-(1) 성능 평가 지표**</span>
##### <span style="color: #27baff">   **i. 평균 제곱 오차(Mean Squared Eroor, MSE)**</span> 
$$\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$
- 예측값과 실제값 사이의 오차 제곱의 평균

##### <span style="color: #27baff">   **ii. 평균 절대 오차(Mean Absolute Eroor, MAE)**</span> 
$$\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$$
- 예측값과 실제값 사이의 절대값 오차의 평균

##### <span style="color: #27baff">   **iii. MSE와 MAE의 차이**</span>  
- MSE
    - 큰 오차에 민감하게 반응 (제곱하기 때문)  
    - 이상치에 민감
- MAE
    - 모든 오차를 동일하게 취급
    - 이상치가 존재해도 전반적인 예측 성능 유지

##### <span style="color: #000000;background-color: #ffc2e5">   **3-(2) 변수 유의성 평가**</span>
##### <span style="color: #27baff">   **i. T-검정**</span>
<U>**독립변수의 회귀계수가 유의미한지 검정**</U>  
- 귀무가설 : 회귀계수(a)가 0이다 -> 독립변수가 종속변수에 영향을 주지 않는다.
- 대립가설 : 회귀계수(a)가 0이 아니다 -> 독립변수가 종속변수에 영향을 준다. 
<BR>

<span style="color: #2bba66">**(1) t-값 계산**</span>  
$$t = \frac{\hat{\beta} - 0}{SE(\hat{\beta})}$$
- $\hat{\beta}$ : 추정된 회귀계수
- $SE(\hat{\beta})$ : 표준오차
- t-분포를 따름

<span style="color: #2bba66">**(2) p-value 확인 및 판단**</span>   
- p-value < 0.05 : 귀무가설 기각 -> 해당 변수는 유의미하다
- p-value ≥ 0.05 : 귀무가설 기각 불가 -> 해당 변수는 유의미하지 않다
- 회귀계수가 0이면 해당 독립변수는 의미 x
- p-value가 작을수록 해당 변수는 종속 변수에 더 큰 영향을 줄 확률 ↑
- t-값이 크면 귀무가설 기각할 가능성 ↑
