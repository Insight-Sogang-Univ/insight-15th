### <span style="color: #000000;background-color: #ffdba4">**1. 분류란?**</span>

#### <span style="color: #000000;background-color: #ffc2e5">   **1-(1) 지도학습 : 회귀 vs 분류**</span> 

- <U>**회귀**</U> : 주어진 데이터(독립변수)를 기반으로 정답(종속변수)을 잘 맞추는 함수
    - <span style="color : #ff0000">**연속형 변수 예측**</span>
- <U>**분류**</U> : 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별하는 것
    - <span style="color : #ff0000">**범주형 변수 예측**</span>
    - 대표적인 머신러닝 알고리즘
        - 로지스틱 회귀 : 독립변수와 종속변수의 선형 관계성에 기반해 분류
        - 결정 트리 : 데이터 균일도에 따른 규칙 기반의 분류
        - 서포트 벡터 머신 : 개별 클래스 간의 최대 마진을 효과적으로 찾아 분류
        - 최소 근접 알고리즘 : 근접 거리 기준으로 분류

#### <span style="color: #000000;background-color: #ffc2e5">   **1-(2) 이진 분류와 다중 분류**</span> 
- 이진 분류(Binary Classification) 
    - 예측하고자 하는 변수가 어떤 기준에 대하여 참 또는 거짓의 값만을 가질 때
- 다중 분류(Multiclass Classification)
    - 예측하고자 하는 변수가 가질 수 있는 값이 3개 이상일 때

### <span style="color: #000000;background-color: #ffdba4">**2. 분류모델**</span>

#### <span style="color: #000000;background-color: #ffc2e5">   **2-(1) 로지스틱 회귀 (Logistic Regression)**</span> 
<U>**독립 변수의 선형 조합(선형 회귀)에 로지스틱 함수를 적용하여 출력값을 0에서 1 사이로 변환해주는 것**</U>
- 이진 분류 문제에 적합!
    - 이진 분류를 단수 선형 회귀로 풀려고 하면 문제 발생 -> <U>출력이 0과 1사이의 값을 가지면서 S자 형태로 그려지는 함수를 이용해야 한다!</U>
##### <span style="color: #27baff">   **[시그모이드 함수]** <span>
<U>**출력이 0과 1 사이의 값을 가지면서 S자 형태로 그려지는 함수**</U>
- 출력값이 특정값 이상이면 1(True), 특정값 이하면 0(False)로 수렴 -> **이진 분류 문제에 사용하자!**  
<BR>
##### <span style="color: #27baff">   **[시그모이드 함수의 가중치]** <span>
- 시그모이드 함수가 $H(x)=\frac{1}{1+e^{-(wx+b)}}=sigmoid(wx+b)=\sigma(wx+b)$일 때, 인공지능은 주어진 데이터에 <U>**적합한 가중치 $w$와 $b$를 계산해서 찾는다**</U>
    
    1. 가중치 w의 변화에 따른 그래프 변화 (그래프의 기울기 변화)
    - 빨간색 선의 w = 0.5, 초록색 선의 w=1.0, 파란색 선의 w=2.0

      
    2. 가중치 b의 변화에 따른 그래프 변화 (그래프의 위치 변화)
    - 빨간색 선의 b = 0.5, 초록색 선의 b=1.0, 파란색 선의 b=2.0

#### <span style="color: #000000;background-color: #ffc2e5">   **2-(2) 의사결정나무 (Decision Tree)**</span> 
<U>**조건에 따른 데이터를 분류하며, 최종적으로 데이터가 순수한 label의 집합으로 구성될 때까지 분류를 반복하는 분석 방법**</U>
##### <span style="color: #27baff">   **[의사결정나무의 구성 요소]** <span>

- **Root Node** : 가장 상위에 위치한 노드
- **Parent Node** : 주어진 노드의 상위에 있는 노드
- **Child Node** : 하나의 노드로부터 분리된 2개 이상의 노드들
- **Binary Tree** : Tree 중에서 children이 최대 2개인 tree
- **Leaf Nodes** : 자식마디가 없는 노드, 모델에서 label에 해당
- **Edge** : Parent node와 child node를 연결하는 연결선
- **Height** : 특정 노드에서 가장 멀리 있는 leaf node까지의 경로에 있는 edge 개수
- **Depth** : Root node에서 특정 노드에 도달하기 위한 edge의 수

##### <span style="color: #27baff">   **[CART(Classification And Regression Tree;이진분할) 알고리즘]** <span>
<U>**가장 널리 사용되는 의사결정나무 알고리즘이자, 데이터셋을 임계값을 기준으로 두 child로 나누는 알고리즘**</U>
- 임계값 나누는 방법 : *불순도(지니 계수)가 낮아지는 방향*으로 나눈다
    - 불순도 : 분류하려는 데이터 집합에서 서로 다른 클래스가 섞여있는 정도
    - 지니 지수 
        - 통계적 분산 정도를 정량화해서 0~1 사이로 표현한 값 (0이면 가장 순수도가 높다)
        - 얼마나 불확실한지 (얼마나 많은 것이 섞여 있는지) 파악  
<BR>
- CART 알고리즘의 주요 단계
1. 임계값 설정
  
오른쪽 그림의 t_n의 임계값을 기준으로 하여 왼쪽 그림과 같이 데이터를 그룹(R_n)으로 나눈다.  
2. 불순도 감소 알고리즘
- 현재 집단에 어느정도의 다른 객체들이 섞여있는지 확인하고 불순도가 낮은 쪽으로 가지를 형성하게 하는 알고리즘

##### <span style="color: #27baff">**[고려사항 : 모수 설정, 차이점 시각화, Prunning]** <span>
- **Hyperparameter 설정**  
    - min_samples_split : 분할되기 위해 노드가 가져야 하는 최소 샘플 수
    - min_samples_leaf : 리프 노드가 가지고 있어야 하는 최소 샘플 수 
    - min_weight_fraction_leaf : min_samples_leaf와 같지만 가중치가 부여된 전체 샘플수에서의 비율
    - max_leaf_nodes : 리프 노드의 최대 개수
    - max_features : 각 노드에서 분할에 사용할 특성의 최대 수
- **시각화**
    - 시각화로 분류가 잘 이루어졌는지 확인

      
    
     왼쪽 그림 : 규제 X, 과적합  / 오른쪽 그림 : min_samples_leaf =4 -> 적절히 분류
- **Prunning (가지치기)**
    - 불필요한 노드 지우기
        - 노드가 너무 많아지면 과적합 확률 ↑
        - Prunning으로 하부 트리 제거 → 일반화 성능↑ , 깊이&결과의 개수 ↓

#### <span style="color: #000000;background-color: #ffc2e5">   **2-(3) SVM (Support Vector Machine)**</span> 
<U>**클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘**</U>
- 명확하게 분류할 수 있는 데이터 집단에서 뛰어난 성능을 보이며, 고차원 공간 (다수의 feature)에서도 효과적으로 사용 가능

##### <span style="color: #27baff">   **[SVM의 구성]** <span>
 

- Support vector (서포트 벡터) : 데이터 중에서 결정 경계와 가장 가까이에 있는 데이터들의 집합
- Decision Boundary (결정 경계) : 데이터 분류의 기준이 되는 경계
- Margin (마진) : 결정 경계에서 서포트 벡터까지의 거리(여유 공간)
- Hyperplane (초평면) : n차원 공간의 (n-1)차원 평면
- Slack Variables (슬랙 변수/여유변수) : 완벽한 분리가 불가능할 때 분류를 위해 허용된 오차에 대한 변수

##### <span style="color: #27baff">   **[최적의 선을 찾는 방법]** <span>
- 결정 경계는 데이터로부터 가장 멀리 떨어져 있는게 좋다.
- 따라서 SVM은 <U>**Margin이 가장 큰 경우를 선택**</U>함으로써 최적의 선을 찾는다!

##### <span style="color: #27baff">   **[SVM의 장단점]** <span>
- 장점
    - 서포트 벡터만을 이용하므로 데이터가 적을 때 효과적
    - 새로운 데이터가 입력되면 서포트 벡터와의 거리만 계산하면 되므로 연산량 최소화
    - 뛰어난 정확성
    - 비선형 모델 분류 가능
    - 낮은 과적합 확률, 적은 노이즈 영향
- 단점
    - 데이터 전처리 과정 중요
    - 데이터가 많아질수록 평가과정이 많아져서 속도 저하

#### <span style="color: #000000;background-color: #ffc2e5">   **2-(4) KNN (K-Nearest Neighbor)**</span> 
<U>**데이터로부터 거리가 가까운 k개의 다른 데이터 레이블을 참조하여 분류하는 알고리즘**</U> - k개의 이웃 설정
- 비슷한 특성을 가진 데이터끼리 서로 가까이 있다는 점 이용  
- 중요한 특징 : <span style="color: #ff0000"> **어떠한 학습이 필요하지 않다!** </span>
- 계산 순서
    1. 데이터 준비
        - 각 데이터는 특징 벡터와 레이블(어떤 클래스냐)로 이루어짐
    2. k값 설정
        - k는 보통 홀수개로 설정 (동점 발생 문제 방지)
    3. 거리 계산
        - 새로운 데이터가 주어지면, 이 값과 기존 모든 데이터 간의 거리 계산
        - 유클리드 거리, 맨해튼 거리 등 사용
    4. 가장 가까운 k개의 이웃 선택
    5. 분류
        - k개의 이웃 중 <U>가장 많이 등장하는 클래스가 예측 결과!</U> 
- 최적의 k를 선택하는 방법
    - 일반적으로, 학습용 데이터 개수의 제곱근으로 설정
    - k가 너무 크다 : 주변에 있는 점과의 근접성이 떨어져 과소적합 발생
    - k가 너무 작다 : 이상값, 잡음 데이터와 이웃 될 가능성이 있다.

##### <span style="color: #27baff">   **[K-NN 알고리즘 예시]** <span>

1. 6개의 기존 데이터(A~F)와 1개의 신규 데이터(N)이 있고, 주변에 있는 이웃의 개수를 k라고 하자
2. k=1이면 거리가 가장 가까운 C를 보고 N을 동그라미 그룹으로 분류
3. k=3이면, N과 거리가 세번째로 가까운 C,D,E를 보고 다수결에 따라 N은 세모 그룹으로 분류
- 3개 이상의 범주로 분류하는 문제에서는 분류될 범주와 데이터의 세부 값을 고려하여 적절한 k의 값 설정

##### <span style="color: #27baff">   **[K-NN의 장단점]** <span>
- 장점
    - 범주를 나눈 기준을 알지 못해도 데이터 분류 가능
    - 입력 데이터만 주어지면 바로 예측값을 구할 숭 있으므로 학습 과정 필요 없음
    - 이해가 쉽고 구현이 간단하다
    - 추가된 데이터 처리 용이
    - 훈련데이터를 요약하거나 일반화 X, 원본 데이터 전체를 그대로 저장
- 단점
    - 테스트 데이터의 개수에 따라 시간이 오래 걸림
    - 학습데이터의 양도 계산 시간에 영향을 미침
    - k 값 결정 어려움
    - 수치형 데이터가 아니면 유사도 정의 어려움
    - 이상치가 존재하면 성능에 큰 영향을 미침

### <span style="color: #000000;background-color: #ffdba4">**3. 분류 평가 지표**</span>

#### <span style="color: #000000;background-color: #ffc2e5">   **3-(1) 혼동 행렬**</span> 

<U>**분류 모델의 예측 결과를 정확한 예측과 잘못된 예측으로 구분하여 나타낸 표**</U>  
<BR>


- TP, TN : 예측을 올바르게 한 경우 (예측=실제 결과)
    - TP : 정답 참, 참으로 예측
    - TN : 정답 거짓, 거짓으로 예측
- FN, FP : 예측을 잘못한 경우 (예측 $\neq$ 실제 결과)
    - FP : 정답 거짓, 참으로 예측
    - FN : 정답 참, 거짓으로 예측
##### <span style="color: #27baff">   **[혼동 행렬을 이용한 분류 모델 평가 지표]** <span>
- **Accuracy (정확도)** : 모든 가능한 예측 중 참인 비율
    - (TP+TN)/(TP+TN+FP+FN)
    - 단점 : 정답 레이블의 비율이 불균형하면 정확도를 신뢰할 수 없다.
- **Precision (정밀도)** : 예측이 참일 경우 중 실제도 참인 비율
    - TP / (TP+FP)
    - 거짓을 참으로 판단한 정도 파악 가능
    - 정밀도가 높을수록, 거짓을 참으로 예측한 정도가 낮다.
- **Recall (재현도)** : 실제 참인 경우 중 예측도 참인 경우
    - TP / (TP+FN)
    - 참을 거짓으로 판단한 정도 파악 가능
    - 재현도가 높을수록, 참을 거짓으로 예측한 정도가 낮다.  
<BR>
- **Precision과 Recall의 Trade-off 관계**
    - 확률에 기반해서 Threshold가 넘으면 참, 미만이면 거짓으로 판단
    - 이 Threshold를 조정하면 Precision과 Recall을 조정할 수 있음
        - Threshold 낮춤 -> Positive 예측 늘어남 -> Recall ↑, Precision ↓
        - Threshold 높임 -> Positive 예측 늘어남 -> Recall ↓, Precision ↑

##### <span style="color: #27baff">   **[Recall & Precision 그래프]** <span>


- 각 점은 분류 모델이 해당 임계값에서 Precision과 Recall value를 어떻게 예측했는지 표시
- 모든 임계값에 대해 점을 그려서 모델의 예측 성능을 시각적으로 파악 가능
- <U>**정밀도 & 재현도 그래프에서 두 value 값이 만나는 지점을 Threshold로 정하면, 예측 오류를 최소화할 수 있다**</U>

#### <span style="color: #000000;background-color: #ffc2e5">   **3-(2) F1-Score**</span> 
- **정밀도와 재현율의 조화 평균**  
<BR>
$F1 Score = \frac {2}{\frac{1}{Precision} + \frac{1}{Recall}} = \frac {2*Precision*Recall}{Precision+Recall}$  
<BR>
- 정밀도와 재현율 간의 균형을 효과적으로 평가


#### <span style="color: #000000;background-color: #ffc2e5">   **3-(3) ROC /AUC Curve**</span> 

##### <span style="color: #27baff">   **[ROC Curve]** <span>
<U>**얼마나 분류가 잘 되었는가를 보여주는 그래프**</U>  

  
- True Positive Rate (TPR) : 참인 것들 중에 참이라고 예측한 비율(=Recall)
- False Positive Rate (FPR) : 거짓인 것들 중에 참이라고 잘못 예측한 비율
<BR>
- 완벽한 분류기는 나올 수 없기 때문에 현실적 한계를 고려하여 적절히 타혐한 최선의 ROC Curve를 도출해야 한다.

##### <span style="color: #27baff">   **[AUC Curve]** <span>
<U>**ROC와 x축 사이의 면적(적분값)**</U>  

  
- 모델의 성능을 0.5~1 사이의 숫자로 표현
- 1에 가까울수록 분류 성능이 좋은 것

#### <span style="color: #000000;background-color: #ffc2e5">   **3-(4) 다중 분류 평가 지표**</span> 
- 위 이진 분류 평가 지표를 사용해서 클래스별로 점수를 구한 뒤, 이를 적절히 평균 내리는 것

##### <span style="color: #27baff">   **[Macro average]** <span>
- 클래스별로 구한 평가 지표의 평균을 구해 전체적인 모델 성능을 평가하는 지표
- 모든 클래스에 동등한 가중치 부여

##### <span style="color: #27baff">   **[Weighted average]** <span>
- 각 클래스별로 구한 평가 지표를 클래스의 샘플 수로 가중 평균하여, 전체적인 모델 성능을 평가하는 지표
- 빈도가 높은 클래스에 더 큰 가중치 부여

##### <span style="color: #27baff">   **[Micro average]** <span>
- 모든 클래스의 예측 결과를 더하여 전체적인 성능을 평가하는 지표
- 모든 클래스에 동등한 가중치 부여

### <span style="color: #000000;background-color: #ffdba4">**4. 하이퍼파라미터 최적화**</span>

#### <span style="color: #000000;background-color: #ffc2e5">   **4-(1) 하이퍼파라미터 최적화**</span> 
- **하이퍼파라미터**
    - 모델 학습 과정에 반영되는 값으로, 학습 시작 전에 사용자가 직접 설정하는 변수
- **하이퍼파라미터 최적화**
    - Tuning을 거쳐 적절한 하이퍼파라미터를 찾음으로써 model 성능을 향상시키는 것  
    
<BR>

- 하이퍼파라미터 최적화 과정  

1. 하이퍼파라미터 탐색 범위 설정 (최적값을 찾고 싶은 하이퍼파라미터의 범위 설정)   
2. 평가 지표 계산 함수(성능 평가 함수) 정의 (탐색하려는 하이퍼파라미터를 인수로 받아 평가지표 값을 계산해주는 함수 정의)   
3. 1단계에서 샘플링한 하이퍼파라미터 값을 사용하여 검증 데이터로 정확도 평가   
4. 위 단계를 특정 횟수 반복하며, 정확도 결과를 보고 하이퍼파라미터의 범위 좁힘  

#### <span style="color: #000000;background-color: #ffc2e5">   **4-(2) 하이퍼파라미터 최적화 방법**</span> 

##### <span style="color: #27baff">   **[Grid Search]** <span>
<U>**정해진 범위에서 하이퍼파라미터를 모두 순회하며 가장 좋은 성능을 내는 값을 찾는 기법**</U>
- 장점 : 범위가 넓고 step이 작을수록 꼼꼼하게 전 범위 탐색 -> 최적해를 정확히 찾을 수 있다.
- 단점 : 시간이 너무 오래 걸림
- 적용 : 넓은 범위, 큰 step을 활용해 범위를 좁힘

##### <span style="color: #27baff">   **[Random Search]** <span>
<U>**정해진 범위에서 하이퍼파라미터를 무작위로 탐색해 가장 좋은 성능을 내는 값을 찾는 기법**</U>
- 장점 : 속도가 Grid Search보다 빠름
- 단점 : 정확도가 떨어짐 (다른 방법보다 사용 빈도가 적음)

##### <span style="color: #27baff">   **[Bayesian Optimization]** <span>
<U>**사전 정보를 바탕으로 최적의 하이퍼파라미터 값을 확률적으로 추정하며 탐색하는 기법**</U>
- 특징 : 'Gausain Process'라는 통계학을 기반으로 만들어진 모델로, 여러 개의 하이퍼파라미터들에 대해 'Aquisition Function'을 적용했을 때, 가장 큰 값이 나올 확률이 높은 지점을 찾아냄
- 방법
    - 초기값 몇개를 무작위로 실험
    - 그 결과를 바탕으로 어떤 파라미터의 조합이 좋을지 확률 모델로 예측
    - 다음 실험할 조합을 가장 좋을 것 같은 위치로 선택
    - 결과를 반영하고 다시 반복
