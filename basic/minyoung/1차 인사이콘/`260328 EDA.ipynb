```python
import pandas as pd

# 1. 숫자여야 할 컬럼을 강제로 숫자형으로 변환합니다. 
# 이때 'MCDONALDS' 같은 문자는 숫자로 바꿀 수 없으므로 강제로 결측치(NaN)로 만듭니다. (errors='coerce')
df['CHECK_IS_NUMERIC'] = pd.to_numeric(df['CARD_HOLDER_TOTAL_TRANSACTION_COUNT'], errors='coerce')

# 2. 강제로 NaN이 된 행 = 원래 문자가 들어있던 불량 데이터(열이 밀린 데이터)
bad_rows = df[df['CHECK_IS_NUMERIC'].isna()]
print(f"열 밀림 오류가 발생한 불량 데이터 개수: {len(bad_rows)}개")

# 3. 477만 개의 방대한 데이터 중 이 불량 데이터가 극소수라면, 과감히 삭제하는 것이 분석에 효율적입니다.
df_clean = df.dropna(subset=['CHECK_IS_NUMERIC']).copy()

# 4. 검사용으로 만들었던 임시 컬럼은 다시 지워줍니다.
df_clean = df_clean.drop(columns=['CHECK_IS_NUMERIC'])

print(f"정제 완료 후 남은 정상 데이터 개수: {len(df_clean)}개")
```

```python
# 1. 필수 식별자(ACCOUNT_ID)가 없는 행 삭제
df_clean = df.dropna(subset=['ACCOUNT_ID']).copy()

# 2. LTM 관련 수치형 결측치는 0으로 대치
ltm_cols = ['CARD_HOLDER_AVERAGE_LTM_SPEND', 'CARD_HOLDER_AVERAGE_LTM_TRANSACTION_COUNT']
df_clean[ltm_cols] = df_clean[ltm_cols].fillna(0)

# 3. 나머지 범주형(문자형) 결측치는 'Unknown'으로 대치
categorical_cols = [
    'TRANSACTION_TYPE', 'TRANSACTION_STATE', 'TRANSACTION_CITY',
    'MERCHANT_STATE', 'MERCHANT_CITY', 
    'MERCHANT_CATEGORY_LEVEL_1', 'MERCHANT_CATEGORY_LEVEL_2', 'MERCHANT_CATEGORY_LEVEL_3',
    'TRANSACTION_DESCRIPTION', 'CARD_HOLDER_STATE', 'CARD_HOLDER_CITY', 'CARD_HOLDER_GENERATION'
]
df_clean[categorical_cols] = df_clean[categorical_cols].fillna('Unknown')

# 결측치가 잘 처리되었는지 최종 확인
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```

```python
import pandas as pd

# 1. TRANSACTION_DATE를 판다스 datetime 형식으로 변환
df_clean['TRANSACTION_DATE'] = pd.to_datetime(df_clean['TRANSACTION_DATE'])

# 2. 트렌드 분석을 위한 파생 변수 추출 (월, 요일, 시간)
df_clean['MONTH'] = df_clean['TRANSACTION_DATE'].dt.month
df_clean['DAY_OF_WEEK'] = df_clean['TRANSACTION_DATE'].dt.day_name()
df_clean['HOUR'] = df_clean['TRANSACTION_DATE'].dt.hour

# 3. 월별 총 결제액 및 거래 건수 트렌드 요약 확인
monthly_trend = df_clean.groupby('MONTH').agg(
    TOTAL_SPEND=('GROSS_TRANSACTION_AMOUNT', 'sum'),
    TRANSACTION_COUNT=('TRANSACTION_ID', 'count')
).reset_index()

print("월별 결제 트렌드 요약:")
print(monthly_trend)
```

```python
# 결제 금액의 전반적인 분포와 기초 통계량 확인
print(df_clean['GROSS_TRANSACTION_AMOUNT'].describe())

# 1. 환불/취소로 추정되는 '음수' 결제 점검
refunds = df_clean[df_clean['GROSS_TRANSACTION_AMOUNT'] < 0]
print(f"환불/취소 추정 결제 건수: {len(refunds)}건")

# 2. 유효성 검증용 또는 전액 리워드 결제인 '0원' 결제 점검
zero_spend = df_clean[df_clean['GROSS_TRANSACTION_AMOUNT'] == 0]
print(f"0원 결제 건수: {len(zero_spend)}건")

# 3. 극단적으로 높은 고액 결제 (상위 1% 기준선 산출) 점검
upper_bound = df_clean['GROSS_TRANSACTION_AMOUNT'].quantile(0.99)
high_rollers = df_clean[df_clean['GROSS_TRANSACTION_AMOUNT'] > upper_bound]
print(f"상위 1% 초과 (기준액: {upper_bound:.2f}달러) 결제 건수: {len(high_rollers)}건")
```

```python
# 1. 0원 결제는 아예 제외 (수익성 분석에 불필요)
df_profit = df_clean[df_clean['GROSS_TRANSACTION_AMOUNT'] != 0].copy()

# 2. 일반 소비자 마케팅 및 CLV 분석용 메인 데이터셋 (상위 1% 이하 정상 결제 및 환불 포함)
df_regular = df_profit[df_profit['GROSS_TRANSACTION_AMOUNT'] <= upper_bound].copy()

# 3. 프리미엄/법인 타겟팅용 별도 데이터셋 (상위 1% 초과 고액 결제)
df_vip = df_profit[df_profit['GROSS_TRANSACTION_AMOUNT'] > upper_bound].copy()

print(f"일반 고객 분석용 데이터: {len(df_regular)}건")
print(f"VIP/법인 분석용 데이터: {len(df_vip)}건")
```
