## 통계 사전 과제

통계 전수조사 하지 않고 샘플을 구해 조사함  
1. 기술통계: 샘플 자체에 대한 관심
    - 수치와 시각화 -> 샘플을 요약하여 표현
2. 추론통계: 샘플의 모집단에 대한 관심
    - 추정과 검정 -> 샘플로부터 모집단 추론

### 자료의 척도

> **자료**: 설명과 해석을 위해 수집•분석•요약되는 사실, 수치  

- 질적 자료
    - 명목척도: 레이블, 이름으로 요소의 속성 식별
    - 서열척도: 명목형의 분류 속성 + 항목 간 순서 또는 순위 나타냄
- 양적 자료
    - 구간(등간)척도: 값 사이에 순서 O, 간격이 일정하여 덧셈, 평균과 같은 연산 가능
    - 비율척도: 절대영점과 임의의 측정단위 구성, 동간성 유지되는 척도

### 기술통계

> **기술통계**: 데이터를 있는 그대로 설명하고 이해하는 데 중점을 둔 방법

*3요소: 중심화 경향 / 자료의 산포 / 자료의 정리 및 시각화*  
1. 중심화 경향
데이터의 중심적인 경향을 나타내는 기술 통계량  
- 평균: 데이터 값의 총합을 데이터의 개수로 나눔 ```series.mean()```
    - 이상치에 민감 -> 절사 평균으로 보완(양끝 일정 비율 잘라냄)
- 중앙값: 데이터를 순서대로 정렬했을 때, 가운데에 위치하는 값 ```series.medium()```
    - 이상치에 덜 민감
- 최빈값: 데이터에서 가장 자주 나타나는 값 ```series.mode()```
    - 이산형 데이터, 데이터 분포 이해하기 유용

2. 자료의 산포
데이터가 퍼져있는 정도를 설명하는 기술 통계량  
- 범위: 최댓값과 최소값의 차이 ```series.max()-series.min()```
    - 이상치에 매우 민감
- 분산: 데이터 값이 평균으로부터 얼마나 퍼져있는지 ```series.var()```
    - $$\frac{\sum_{i=1}^{N}(x_i-\mu)^2}{N}$$

- 표준편차: 분산에 제곱근 취한 값 ```series.std()```
    - 기존 단위 맞춰주기 위해
    - $$\sqrt{\frac{\sum_{i=1}^{N}(x_i-\mu)^2}{N}}$$

---
- 사분위수: 데이터의 표본을 네 개의 동일한 크기로 나눈 값
    - Q1: 하위 25퍼센트 값, 하위 사분위수 ```series.quantile(0.25)```
    - Q2: 하위 50퍼센트 값, 중앙값 ```series.quantile(0.5)```
    - Q3: 하위 75퍼센트 값, 상위 사분위수 ```series.quantile(0.75)```, ```np.percentile(series,75)```
- 사분위간 범위 (IQR)
    - Tukey Fence: Q3-Q1, 데이터의 중앙50%가 퍼져있는 범위
    - 아래쪽 이상치: Q1-1.5*IQR 보다 작은 값
    - 위쪽 이상치: Q3+1.5*IQR 보다 큰 값
---
- 변동계수: 상대적인 변동성을 나타내는 통계적 지표
    - $$ 변동계수=(표준편차/평균)*100 $$
    - 평균 대비 변동성
---
- 첨도: 확률분포의 꼬리가 두꺼운 정도를 나타내는 척도 ```seires.kurt()```
- 왜도: 비대칭도, 실수 값 확률 변수의 확률 분포 비대칭성을 나타내는 지표 ```series.skew()```
    - 양수 -> 오른쪽 긴 꼬리
    - 음수 -> 왼쪽 긴 꼬리
    - 평균=중앙값 -> 왜도=0

### 확률과 분포

> 추론통계학의 의심들...

- 표본의 불확실성 -> 표본분포
- 모수 추정 -> 점추정
- 추정의 현실성 -> 구간추정
- 구간추정의 오류 관리 -> 신뢰수준
- 적극적 주장 검증 -> 가설검정
- 가설검정의 오류 관리 -> 유의수준

> 확률분포

**확률변수**: 랜덤한 세계에서 어떤 특정 사건이 발생할 때 나타날 수 있는 모든 경우의 수를 특정 값으로 대응시키는 함수  
**확률분포**: 확률변수가 특정한 값을 가질 확률을 나타내는 함수  
- 이산확률분포: 베르누이분포, 이항분포, 기하분포, 음이항분포, 초기하분포, 포아송분포
- 연속확률분포: 정규분포, T분포, F분포, 감마분포, 지수분포, 카이제곱분포, 균등분포

> 대표적인 분포 모형
1. 정규분포
- 종 모양, 평균 기준으로 대칭, 표준편차 클수록 평평해짐
- 매개변수: 평균, 표준편차
- 모집단이 정규분포 따르면 표본평균도 정규분포 따름
    - 중심극한정리: 표본 크기 커질수록 모집단 분포 상관 없이 표본평균 분포는 정규분포에 가까워짐

2. Z분포 (표준정규분포)
- 평균이 0이고 표준편차 1이 되도록 맞춘 표준적 정규분포
---
아래부터는 모집단 자체가 갖는 분포라기보다는, 정규분포의 통계적 성질을 이용해 통계적 추론을 하기 위해 만들어진 도구  
*가정: 표본이 '정규분포를 따르는 모집단'으로부터 왔다!*  

3. T분포
- 종모양, 정규분포보다 꼬리 두꺼움, 평균 기준으로 대칭
- 매개변수: 자유도(df) - 얼마나 많은 정보가 독립적으로 변할 수 있는가
- 표본으로 모평균(or 모평균 간 차이)을 추정할 때 사용
- 표본 크기 작을 때(거의 30 기준) 정규분포보다 정확한 결과 제공

4. 카이제곱분포
- 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포, 자유도 증가하면 중심 오른쪽으로 이동-점점 대칭으로 변화-정규분포에 가까워짐
- 매개변수: 자유도(df)
- 독립적인 표준 정규 분포를 따르는 변수들의 제곱 합 
    - 표본분산의 분포와 관련된 지표
- 독립성 검정, 적합도 검정 등에 사용

5. F분포
- 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포, 자유도 증가하면 중심 오른쪽으로 이동-점점 대칭-정규분포에 가까워짐
- 매개변수: 자유도(df1,df2)
- 두 독립적인 카이 제곱 분포의 비율
- 두 집단 간 분산 차이 비교하는 데 사용

---
분포모형은  
- 가설검정을 위한 검정통계량 산출에 사용
- 표본추출 과정에서 인위적인 조작 있었는지 확인 가능

### 추론 통계-추정

> 추론통계

수집한 데이터를 기반으로 전체 정보를 추론하고 결론을 도출하는 방법을 다룸  
통계적 가설검정, 신뢰구간  
- 추정: 모집단을 대표하는 일부를 표본으로 추출하여 분석한 통계량을 사용해 모집단의 모수를 예측하는 과정
    - 추정량: 표본 정보에 의존하는 확률변수로서 모수를 추정하는 데 사용되는 표본통계량
    - 추정치: 표본이 추출된 후 표본자료를 이용하여 계산한 통계량의 특정한 값
- 검정: 표본을 추출하여 얻은 표본통계량으로 모수에 대한 새로운 가설이 옳다고 판단할 수 있는지 평가하는 과정

1. 점추정
- 어떤 모수를 표본에 기초하여 계산된 하나의 값으로 추정

2. 구간추정
- 모수의 참 값이 포함되리라고 기대하는 추정치를 일정한 범위, 신뢰도를 가지고 나타냄
- 구간추정 95%의 의미: *동일한 조사를 100번 반복했을 때, 그중 95개의 신뢰구간이 실제 모수를 포함할 것으로 기대된다*

### 추론 통계 - 검정

> 가설 검정의 절차

가설 검정: 모집단의 실제 값이 얼마가 된다는 주장과 관련해, 표본의 정보를 사용해 가설의 합당성 여부 판정
- 가설 설정 -> 유의 수준 설정 -> 검정통계량 산출 -> 가설 기각or채택 판단

1. 가설 설정
- 귀무가설 H0, 대립가설 H1 설정
    - 귀무가설: 기각하고자 하는 명제 (현재 통용/위험성 약함)
    - 대립가설: 주장하고자 하는 명제 (새롭게 주장/위험성 높음)
2. 유의 수준 설정
- 유의수준: 허용할 수 있는 오류의 최대치, 5%
    - 1종 오류: 귀무가설이 참이었는데 기각할 확률
    - 2종 오류: 귀무가설이 거짓이었는데 기각하지 못할 확률
    - 1종 오류 더 치명적 -> 유의수준은 1종 오류의 확률
3. 검정통계량 산출
- 검정통계량: 가설 검정의 결과를 판단하는 데 사용되는 값
- **Z값**: 표준 정규 분포를 사용하여 얻은 값. 특정 데이터가 평균으로부터 얼마나 떨어져 있는지를 표준편차 단위로 나타냄.
    - $$ Z=\frac{X-\mu}{\sigma} $$
    - X는 관측값, mu는 평균, 시그마는 표준편차
    - 큰 표본 크기에 대한 가설 검정에 주로 사용
    - Z값의 절댓값 크면 -> 해당 데이터 평균으로부터 멀리 떨어져 있음. 비정상적일 가능 성이 있음.
    - 임계점 찾기: 양측검정의 경우 Z=1.96일 때(95%) 혹은 Z=2.58일 때(99%) -> Z값이 임계점보다 크면 귀무가설 기각
- **T값**: T분포 사용해서 얻은 값, 작은 표본 크기에 대한 가설 검정에 주로 사용
    - $$ T=\frac{X-\mu}{s/\sqrt{n}} $$
    - X는 표본 평균, mu는 연구자 설정 값, s는 표본 표준편차, n은 표본 크기
    - 주로 모집단의 평균에 대한 가설 검정 or 두 표본 평균 간 차이 검정할 때 사용
    - T값의 절댓값 크면 -> 해당 데이터가 모집단 평균으로부터 멀리 떨어져 있음. 관측된 표본 평균이 모집단 평균과 유의미한 차이가 있을 가능성 높음.
    - 임계점 찾기: 자유도에 따라 모양 달라짐. 같은 유의수준에 대해 t값은 z값보다 커지며, 자유도가 높아질수록 z값에 수렴.
- **F값**: F분포를 사용하여 얻은 값. 주로 두 표본 분산 사이의 차이를 비교하는 데 사용.
    - $$ F=\frac{s_{bet}^2}{s_{wit}^2} $$
    - 분자에는 표본 평균 간 퍼진 정도, 분모에는 표본 내에서 퍼진 정도를 대입
    - 주로 분산 분석(ANOVA)에서 두 개 이상의 그룹 간 평균의 차이를 검정하는 데 사용
    - 1에 가까우면 두 표본의 분산이 비슷하다는 의미
4. 가설 기각or채택 판단
- **유의확률(p-value)**: 유의 확률이 유의 수준보다 작을 시 귀무가설 기각, 클 시 귀무 가설 기각하지 못함 


> 데이터의 자료형에 따른 검정 방법


1. [수치형-수치형] 피어슨 상관분석  
두 변수간 피어슨 상관계수가 얼마나 유의미한지를 검정하는 방법  
- 가설
    - H0: 두 변수간 선형 상관관계가 존재하지 않는다.
    - Ha: 두 변수간 선형 상관관계가 존재한다.
- 공분산(covariance): 두 변수가 함께 어떤 방향으로 변화하는지와 그 크기를 표현 ```np.cov(df['col1'],df['col2'])```
    - Cov(X,Y)>0 -> X와 Y는 양의 관계
    - Cov(X,Y)<0 -> X와 y는 음의 관계
    - Cov(X,Y)=0 -> X와 Y는 양도 음도 아닌 관계
- 상관계수(correlation): 두 변수 간 선형적인 관계 나타내는 통계적 지표, -1~1 사이의 값 ```df[['col1','col2']].corr()```
    - Corr(X,Y)>0 -> 양의 상관 관계
    - Corr(X,Y)<0 -> 음의 상관관계

- 피어슨 상관계수
    - |r|<0.3 -> 상관관계 약함
    - |r|<0.5 -> 상관관계 중간
    - |r|>0.5 -> 상관관계 강함

2. [범주형-범주형] F-test(ANOVA), T-test  

**T-test**: 범주가 2개일 때
- 가설
    - H0: 두 그룹의 평균에는 차이가 없다.
    - Ha: 두 그룹의 평균에는 차이가 있다.  

- ``` spst.ttest_ind(groupa, groupb)```  

- 주로 절댓값 2보다 크면 차이 O, 2보다 클수록 차이 크다고 생각
- p값: 이 결과가 얼마나 믿음직한지, 보통 5% 기준  

**F-test(ANOVA)**: 범주가 3개 이상일 때  
- 집단 간 분산 / 집단 내 분산으로 계산
- 가설
    - H0: 각 MBTI 그룹 별 수명 평균은 차이가 없다.
    - Ha: 각 MBTI 그룹 별 수명 평균은 차이가 있다.   

- ``` spst.f_oneway(groupa,groupb,...,groupz)```  

- F값 2~3 이상이면 차이 O, 클수록 차이 큼  

3. [범주형-범주형] Chi2-test  
**Chi2-test**  
- 가설
    - H0: 경영학과 여부와 갈색머리 여부는 관련이 없다.
    - Ha: 경영학과 여부와 갈색머리 여부는 관련이 있다.
- 카이제곱 통계량: $(관측빈도-기대빈도)^2/기대빈도$ 를 쭉 더해줌
- 범주가 많아질수록 카이제곱 통계량 커짐
- 카이제곱 통계량이 자유도의 2배 이상이면 차이가 존재
