## 회귀 기초 사전과제

### 머신러닝과 모델링

> 머신러닝

인공지능의 한 분야  
모델이라는 소프트웨어를 학습하여 패턴 발견 -> 새로운 데이터 예측 or 결정  

- 머신러닝의 종류
    - **지도학습**: 정답이 있는 데이터 사용 -> 기계의 예측값이 데이터의 정답과 같아지도록 학습 ``` 회귀, 분류 ```
    - **비지도 학습**: 정답 X -> 데이터의 패턴, 유사성 학습 ```군집화```
    - **강화 학습**: 시행착오 반복하여 보상을 최대화하는 행동 학습 ``` 알파고, 자율주행```

> 모델링

**모델**: 학습의 결과물, 학습을 마친 인공지능  
- 데이터를 학습하며 찾아낸 패턴이 저장된 상태
- 새로운 데이터를 너헝ㅆ을 때 답을 내놓는 예측기 역할  
  
**모델링**: 예측기를 만들어가는 전체적인 과정  
  
**모델링 과정**  
1. 설계: 데이터의 특성에 맞춰 어떤 알고리즘으로 학습할지 결정하는 단계
2. 훈련: 데이터를 기계에 계속 보여주며 정답을 맞히도록 수정하는 단계
3. 검증: 훈련시킨 모델이 새로운 데이터에서도 정답 잘 맞히는지 테스트하는 단계
  
> 지도 학습  

정답 레이블을 함꼐 주고 학습을 시키는 방법  
예측갑소가 정답이 최대한 같아지는 방향  
- Labeled Data의 형태여야 함
    - Labeled Data: 독립변수, 종속변수가 모두 존재하는 데이터
  
**지도 학습의 과정**  
1. Labeled 데이터를 Training set과 Test set으로 나눔
2. Training set을 통해 학습: 독립변수를 통해 예측한 값이 종속변수와 같아지도록 훈련  

**회귀와 분류**  
  

- 회귀: 주어진 데이터를 기반으로 정답을 잘 맞추는 함수    
- 분류: 기존 데이터가 어떤 레이블에 속하는지 패턴을 인지한 뒤 새롭게 관측된 데이터에 대한 레이블을 판별  


### 회귀

> 회귀  

데이터에서 패턴을 찾아내 미래 값을 예측  
- 종속변수가 연속적
- 다양한 기법: 선형회귀, 비선형회귀, 규제회귀, 다항회귀  

> 선형회귀  

회귀선을 찾아내는 과정  
- $y=wx+b$를 찾아냄
- 현재 데이터 요약 + 미래 값을 예측
- 단순선형회귀 - x가 한 개 / 다중선형회귀 - x가 여러 개  

**가중치와 편향**
- $w$ 가중치: 독립변수 $x$가 종속변수 $y$에 주는 영향력의 크기, 기울기, 회귀계수
- $b$ 편향: 독립변수 $x$가 0일 때 나오는 기본값, y절편
  
**단순선형회귀**
- 하나의 독립변수만 존재
- 데이터를 직선 형태로 표현

**잔차**
- 실제값과 예측값의 차이
- 잔차를 줄여나감 = 회귀선을 잘 찾는 과정
- 오차: 실제값 - 실제 평균값 차이
- 잔차: 실제값 - 예측값 차이

> 최적의 $w$와 $b$를 찾는 두 가지 방법
  
1. 최소제곱법: 잔차의 제곱의 합(RSS)이 최소가 되는 지점으로 최적의 회귀선 구함
- 회귀선과 실제 데이터 차이를 최소화
  
2. 경사하강법
손실함수: $E=\frac{1}{n}\sum_{i=1}^{n} [t_i - y_i]^2$  
- 잔차를 제곱한 뒤 합하는 형태
- 이 식에 $y=wx+b$를 대입해 최종적으로 아래와 같은 손실함수 얻음
    - $E(W,b)=\frac{1}{n}\sum_{i=1}^{n} [t_i - (wx_i +b)]^2$
  
*E를 최소화하는 $w$, $b$를 찾아야 함!*  

학습: '오차의 최솟값'을 찾아감  
- 손실함수를 최소화하는 값을 찾아가며 '적당한' 보폭으로 내려가기  
- Local Minima, Global Minima가 존재하며, 작은 기울기에서는 Local에서 벗어나기 어려움 -> 탈출하는 방법에 대해 알고리즘 존재
  
모멘텀: 기존 경사하강법과 달리 이전의 기울기와 방향을 고려  
- 작은 기울기는 쉽게 넘어갈 수 있음
- Local Minima 탈출 가능  

> 다중선형회귀  

$y=W_1*x_1 + W_2*x_2 + ... + W_n*x_n + b$  
독립변수가 두 개 이상인 경우  
- 독립변수가 2개인 경우 3차원 공간에 표현, 회귀선은 평면
- 최소제곱법/경사하강법 사용 가능
- 잔차: 관측치와 평면간 차이
  
**다중공선성**  
독립변수들 간 강한 상관관계로 인해 회귀 모델 불안정해지는 문제  
  
- 확인 방법
    1. 상관계수: 변수 간 상관관계를 수치로 나타냄
    2. VIF 지수 (분산 팽창 인수): 회귀 모델의 결정계수 R-Square 사용해 계산
        - VIF가 5 이상일 때부터 다중공선성의 징후가 있다고 판단, 10 이상이면 심각하다고 간주

- 대처 방법
    1. 변수 제거
    2. 변수 변환: 변수들을 더하거나 빼서 새로운 변수 생성
    3. 규제선형모델 활용: 릿지, 라쏘, 엘라스틱넷 등의 방법을 통해 모델의 복잡도 줄임
    4. PCA(주성분 분석): 데이터의 차원 축소 - 데이터의 복잡성 줄임

> 규제선형모델  

모델이 학습 데이터에 과적합되지 않도록 규제를 가함  
- 최소제곱법과 과적합을 방지하기 위해 회귀 계수 값이 커지지 않도록 하는 방법이 *균형*을 유지해야 함
- 오류는 줄이면서, 특정 변수에 집착하지 않도록

**규제선형모델 종류**  
1. L2규제: 회귀계수의 제곱에 패널티 부여, 릿지 회귀
2. L1규제: 회귀계수의 절댓값에 패널티 부여, 라쏘 회귀
3. L2규제+L1규제: 엘라스틱넷 회귀
  
**릿지 회귀**  
$\text{Cost Function}=\text{RSS}+\alpha\sum_{j=1}^{p} w_j^2$  
- 비율이 균등하지 않지만 전체적으로 작아짐
- 큰 회귀 계수는 강한 패널티, 작은 회귀 계수는 적은 패널티
  
**라쏘 회귀**  
$\text{Cost Function}=\text{RSS}+\alpha\sum_{j=1}^{p} |w_j|$   
- 절댓값을 패널티로 줌, 크기에 상관없이 같은 패널티
- 변수가 아예 사라질 수도 있음
  
**엘라스틱넷**  
$\text{Cost Function}=\text{RSS}+\alpha_1\sum_{j=1}^{p} w_j^2+\alpha_2\sum_{j=1}^{p} |w_j|$  
- 안정적은 예측과 희소성 동시에 달성 가능
- 계산량 많아져서 속도 느림


### 모델 평가방법

> 성능 평가 지표  

1. 평균 제곱 오차: 예측값과 실제값 사이의 오차 제곱의 평균
    - $\text{MSE}=\frac{1}{n} \sum_{i=1}^{n}(y_i-\hat{y}_i)^2$
    - 큰 오차에 민감하게 반응, 이상치에 민감
2. 평균 절대 오차: 모델의 예측값과 실제값 사이의 절대값 오차의 평균
    - $\text{MAE}=\frac{1}{n} \sum_{i=1}^{n}|y_i-\hat{y}_i|$
    - 모든 오차를 동일하게 취급, 이상치 있어도 전반적인 예측 성능 유지

> 변수 유의성 평가  

- T-검정: 독립 변수의 회귀계수가 유의미한지 검정
    - p-value 5%를 기준으로 해당 변수가 유의미한지 아닌지 판단
    - t-값이 크면 귀무가설 기각 가능성 높아짐
