보통 `describe()` 를 통해 기본적인 기술 통계값을 확인할 수 있습니다.<br/>
여기서는 중간값, 결측치, 왜도, 첨도까지 보기 위해 다음과 같이 코드를 작성하겠습니다.

```python
# 중간값, 결측치, 왜도, 첨도를 표시합니다.
df_stats = df.describe().T # 전치시키는 연산. 행과 열의 위치를 바꿔줘요. 

median_results = []
skew_results = []
kurtosis_results = []
null_results = []

for idx, val in enumerate(df_stats.index):
    median_results.append(df[val].median()) # 중간값 추가
    skew_results.append(df[val].skew()) # 왜도 추가
    kurtosis_results.append(df[val].kurtosis()) # 첨도 추가
    null_results.append(df[val].isnull().sum()) # 결측치 추가

df_stats['median'] = median_results
df_stats['missing'] = null_results
df_stats['skewness'] = skew_results
df_stats['kurtosis'] = kurtosis_results
df_stats
```

OLS를 이용해 회귀분석을 할 때는 항상 상수항을 추가해야 합니다.</br>
add_constant 함수를 사용해서 상수항을 추가해 봅시다.

```python
X = df[['sqft_living']]
y = df[['price']]

# 상수항을 추가합니다
X = sm.add_constant(X, has_constant="add")
X.head()
```

fit() 함수를 통해 선형 모델에 적합시켜봅시다.

해당 결과를 result_model에 저장하고 summary() 함수를 통해서 결과를 확인합니다.

```python
# 모델 fit
model = sm.OLS(y, X)
result_model = model.fit()
result_model.summary()
```

주요 항목별 세부 내용은 다음과 같다.

- Dep. Variable : 종속 변수, 타겟값.
- Model : 학습 모델 / OLS(Ordinary Least Squares), 잔차 제곱합(손실)을 최소로 하는 파라미터를 선택하는 방법
- Method : Least Squares / 잔차 제곱합(손실)을 최소로 하는 파라미터를 선택하는 방법
- No. Observation : 데이터셋 크기
- Df Residuals : 데이터셋 크기 - 추정된 파라미터 수를 뺀 것
- Df Model : 독립 변수의 숫자
- Covariance Type : 공분산 종류(default: nonrobust)
- R-squared : 결정계수, 모델의 설명력(0~1 사이 값), 1에 가까울수록 모델의 설명력이 높음
- Adj. R-squared : 조정된 결정계수, 조정된 모델의 설명력
- Std. error : 계수의 표준 오차
- F-statistic : F value
- Prob (F-statistic): p-value
- Log-Likelihood : 최대 로그 우도
- 아카이케 정보 기준(Akaike's Information Criterion; AIC) : 모델의 성능지표로 작을수록 좋은 모델
- Coef : 계수값
- P>|t| : p-value

P>|t|(p-value) 항목의 유의 확률이 유의 수준 0.05보다 훨씬 작아 유의 수준 0.05 이하에서 귀무 가설은 기각됩니다.

즉, 모형 통계학적 유의성이 확보됩니다.

시각화를 통해 실제 타겟값과 모델을 통해 추정한 값, 잔차(residual)를 확인해 봅시다.

```python
# 잔차를 확인합니다
result_model.resid.plot()
plt.show()
```

# 다중 선형 회귀


앞에서는 단순히 변수 하나만 가지고 단순 선형 회귀 모형을 돌렸습니다.

다중 선형 회귀 모형은 단순 선형 회귀모형의 확장으로, 연속형 반응변수 하나에 "설명 변수가 둘 이상"인 모형을 말합니다.
설명 변수가 늘어나므로 추가 검토해야 할 문제는 설명 변수 정보 중첩으로 발생하는 다중공선성(multicollinearlity) 문제와,
너무 많은 설명 변수를 포함해 발생하는 차원의 저주 문제 등이 있습니다.

이번에는 범주형 변수를 설명 변수로 포함한 회귀모형에 대해서도 논의할 거예요.

id를 제외한 모든 독립 변수를 사용해서 다중 선형 회귀분석을 합니다.

그 후 선별한 독립 변수들로 다시 다중 선형 회귀분석을 해 봅시다.

>원래 뭐든 다 때려보고, 그 중 괜찮은 걸 고르는 것이 가장 속이 편합니다.

```python
# date의 경우 전처리를 한 column이 존재하기 때문에 제외
columns = list(df.columns)
not_used = ['id','date']

for item in not_used:
    columns.remove(item)
columns
```

```python
# 위의 columns 만 선택한 df 선언
df_reg = df[columns]

# constant 더하고 feature columns 뽑아내기
df_kc_reg = sm.add_constant(df_reg, has_constant='add')
feature_columns = list(df_kc_reg.columns.difference(['price']))

# 변수 선택
X = df_kc_reg[feature_columns]
y = df_kc_reg.price

# 회귀 모형
multi_linear_model = sm.OLS(y, X)
result_model_1 = multi_linear_model.fit()

# 결과
result_model_1.summary()
```

결과를 보면 모든 독립 변수들의 p>|t| (p-value) 유의 확률이 유의 수준 0.05보다 훨씬 작아 모형 통계학적 유의성이 확인됩니다.</br>
R-squared 값이 0.701 로써 약 70%가 모형에 의해 설명된다는 것을 의미합니다.

단순 선형 회귀와 달리 다중 선형 회귀분석에서는 모든 변수의 경우 **다중공선성(multicollinearlity)** 문제를 확인해야 한다.

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

df_vif = pd.DataFrame()
df_vif["VIF"] = np.round([variance_inflation_factor(df_reg.values, i) for i in range(df_reg.shape[1])], 2)

df_vif["features"] = df_reg.columns
df_vif.sort_values(by='VIF', ascending=False)
```

설명 변수가 다수이기 때문에 모형에 포함된 설명 변수의 정보가 중첩(상관관계를 가짐)으로,

**다중공선성 문제가 발생하는 것을 확인할 수 있습니다.**

다음 독립 변수들을 임의로 선택해서 다중 선형 회귀분석을 실시해볼게요.

또한, 반응 변수(price)에 자연로그를 취해 비대칭인 데이터의 분포를 완화한 상태에서 회귀분석을 합니다.

- bedrooms(연속형)
- sqrt_living(연속형)
- waterfront(범주형)
- view(범주형)
- sold-built_years : date - yr_built (연속형)

```python
df_reg.info()
```

```python
df_reg = df[['bedrooms','sqft_living','waterfront','view','sold-built_years','price']]

df_kc_reg = sm.add_constant(df_reg, has_constant='add') # 상수항 추가
feature_columns = list(df_kc_reg.columns.difference(['price'])) # price를 제외한 feature 컬럼들만 뽑아냅니다.

X = df_kc_reg[feature_columns]
y = np.log(df_kc_reg.price)

# 회귀 모형
multi_linear_model = sm.OLS(y, X)
result_model_2 = multi_linear_model.fit()
result_model_2.summary()
```

결과를 보면 모든 독립 변수들의 P >|t| (p-value) 유의 확률이 유의 수준 0.05보다 훨씬 작아 모형 통계학적 유의성이 확인되었습니다.

이제 다중공선성을 살펴봅니다.

```python
vif_df = pd.DataFrame()
vif_df["feature"] = X.columns # X의 컬럼 리스트를 feature 이름으로 넣기
vif_df["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])] # 각 변수별 VIF 계산 (상수항 포함)

vif_df.sort_values(by='VIF', ascending=False).reset_index(drop=True)
```

다중공선성 문제가 확실히 줄어들었음을 알 수 있습니다.

### (참고) 읽어보세요~. 상수항을 포함하여 VIF를 계산한 까닭

상수항 없이 다중공선성을 계산하면 함수 내부적으로 계산되는 $R^2$가 비정상적으로 높게 측정돼서 실제로는 문제가 없는 변수인데도 VIF가 높게 나오는 현상이 발생할 수 있다고 합니다.

## 4. 등분산성: 오차항의 분산이 일정해야 한다.

```python
# 잔차 패턴 확인
fitted = result_model_2.predict()

resid = result_model_2.resid
pred = result_model_2.predict(X)

fig = plt.scatter(pred, resid, s=3)
plt.axhline(y=0, color='r', linestyle='--')
plt.xlabel('Fitted values')
plt.ylabel('Residual')
plt.show()
```

```python
from scipy.stats import levene

# 예측값(fitted values)의 중앙값을 기준으로 데이터를 두 그룹으로 나눕니다.
median_fitted = np.median(result_model_2.predict())
group1 = result_model_2.resid[result_model_2.predict() < median_fitted]
group2 = result_model_2.resid[result_model_2.predict() >= median_fitted]

# Brown-Forsythe는 levene 검정에서 center='median'을 사용해야 합니다.
stat, p = levene(group1, group2, center='median')

# Brown-Forsythe 검정의 귀무가설은 "두 그룹의 분산이 같다", 즉 "잔차의 분산이 일정하다"입니다.
print(f'Brown-Forsythe Statistic: {stat:.4f}, p-value: {p:.4f}')
```

오차항의 분산 도표를 보면 분산의 정도가 일정하지 않다는 것을 느낄 수 있어요.  
Brown-Forsythe 검정에서도 p-value가 0.02이므로 귀무가설이 기각되고 대립가설이 채택되었네요.  
따라서 해당 가정은 만족하지 못했습니다...  

부동산 데이터의 경우, 데이터의 본질적인 특성 때문에 등분산성 가정을 완벽히 만족시키기는 어렵다고 합니다.  
이상치 제거, 스케일링, 변수 변환 등을 통해 이분산성을 최대한 완화해 볼 수는 있지만, 교과서처럼 완벽한 등분산성을 100% 맞추는 것은 아주 어렵다고 하네요..

## 6. 정규성 검정

```python
# Q-Q 도표
qqplot = sm.qqplot(result_model_2.resid, line="s")
plt.show() 
```

```python
# 샤피로-월크 검정
result_shapiro = stats.shapiro(result_model_2.resid)

print(f"F value : {result_shapiro[0]:.4f} / p-value : {result_shapiro[1]:.4f}")

if result_shapiro[1] < 0.05:
    print("p-value < 0.05 입니다.")
```

```python
# 관측치 개수 확인
print(f"관측치는: {len(result_model_2.resid)}")
```

H0 귀무가설: 데이터가 정규분포를 따른다<br>
H1 대립가설: 데이터가 정규분포를 따르지 않는다. <br>

샤피로-윌크 검정에서는 p-value < 0.05인데<br>
이러면 귀무가설을 기각해서 '데이터가 정규분포를 따르지 않는다'가 결론이 돼야 하지 않나요??<br>

정답입니다!! <br>
샤피로-윌크 검정에서 p-value가 0.05보다 작으면 통계적으로는 정규성을 만족하지 않아요.  
그런데, 이 경우는 관측치의 개수가 21613개죠.  
이처럼 데이터가 많은 경우는 p-value 하나만 보고 정규성을 만족하지 않는다고 하기보다는 Q-Q 플롯을 같이 보면서 판단해야 해요.  
- 샤피로 윌크 검정은 데이터의 양이 많을 때 특잇값에 영향을 많이 받습니다. 
- 사소한 비정규 데이터에도 p-value를 매우 작게 계산하기 때문에 샤피로 윌크 검정만 하기보다는 Q-Q plot도 살펴봐야 해요.

위에서 봤을 때, Q-Q 플롯이 직선과 대체로 일치했죠! <BR>
그래서 이 경우는 정규성을 만족한다고 할 수 있어요.

샤피로-윌크(Shapiro-Wilk) 같은 통계적 가설 검정은 샘플 사이즈에 매우 민감합니다.
- 샘플 사이즈가 너무 클 때: 데이터가 정규분포와 아주 미세하게만 달라도 p-값이 0.05보다 작게 나와 정규성이 없다고 결론 내릴 수 있습니다.
    - 샤피로 윌크 같은 통계 검정은 너무 예민해져 Q-Q plot을 통해 큰 규모를 전체적으로 판단하는 게 좋다고 합니다.
- 샘플 사이즈가 너무 작을 때: 데이터가 정규분포와 꽤 차이가 나더라도 p-값이 0.05보다 크게 나와 정규성이 있다고 결론 내릴 수 있습니다.
    - 시각적으로는 판단이 주관적일 수 있어서 통계 수치가 도움이 됩니다.
