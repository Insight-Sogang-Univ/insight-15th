## 분류 사전 과제

### 분류

> 지도 학습: 회귀 vs 분류  

**회귀**: 연속적인 숫자값을 예측  
- 주어진 데이터(X)를 기반으로 정답(Y)을 잘 맞추는 함수
- ex) 주택 가격 예측
- 예측하려는 값이 연속형 변수  
  
  
  
**분류**: 입력된 데이터를 주어진 항목들로 나눔  
- 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤, 새롭게 관측된 데이터에 대한 레이블을 판별
- 머신러닝 알고리즘
    - 로지스틱 회귀: 독립변수 - 종속변수의 선형 관계성에 기반
    - 결정 트리: 데ㅣ터 균일도에 따른 규칙 기반 분류
    - 서포트 벡터 머신: 개별 클래스 간 최대 마진을 효과적으로 찾아 분류
    - 최소 근접 알고리즘: 근접 거리 기준으로 분류
- ex) 대출상환여부 예측
- 예측하려는 값이 범주형 변수
  
  
  
  
> 이진 분류와 다중 분류  

**이진 분류**: 예측하고자 하는 변수 기준에 대하여 참 또는 거짓의 값만 가질 때  
**다중 분류**: 예측하고자 하는 변수가 가질 수 있는 값이 3개 이상


### 분류 모델

> 로지스틱 회귀  

독립 변수의 선형 조합에 로지스틱 함수를 적용하여 출력값을 0에서 1사이로 변환해주는 것  
- 이진 분류 문제 풀이
- 샘플이 특정 클래스에 속할 확률을 추정하여 기준치에 따라 분류
- 단순 선형회귀(직선)로 이진 분류를 하기 어렵기 때문에, 출력이 0과 1사이의 값을 가지면서 S자 형태로 그려지는 함수를 이용해야 함
- 시그모이드 함수: 입력값이 커지면 1에 수렴, 작아지면 0에 수렴
    - 시그모이드 역시 데이터에 적합한 가중치 $w$와 $b$를 찾아 모양 수정 가능
  
> 의사결정 나무  

조건에 따라 데이터를 분류하며, 최종적으로 데이터가 순수한 label의 집합으로 구성될 때까지 분류를 반복  

**구성 요소**  
  
  
- Root Node: 가장 상위 노드
- Parent Node: 주어진 노드의 상위에 있는 노드
- Child Node: 하나의 노드로부터 분리된 2개 이상의 노드
- Binary Tree: Tree 중 Children이 최대 2개인 Tree
- Leaf Nodes: 자식마디가 없는 최하단 노드, 모델에서 label에 해당
- Edge: Parent node와 Child node를 연결하는 연결선
- Height: 특정 노드에서 가장 멀리 있는 leaf node까지 경로에 있는 edge rotn
- Depth: Root node에서 특정 노드에 도달하기 위한 edge의 수
  
**CART 알고리즘**  
가장 널리 사용되는 의사결정나무 알고리즘  
데이터를 임계값을 기준으로 두 child로 나누는 알고리즘  
- 임계값은 불순도(지니 계수)가 낮아지는 방향으로 나눔
    - 불순도: 분류하려는 데이터 집합에서 서로 다른 범주가 섞여 있는 정도를 의미
  
**CART 알고리즘의 주요 단계**  
1. 임계값 설정
2. 불순도 감소 알고리즘
  
**실제 학습 시 고려할 것들**
1. Hyperparameter 설정
- `min_samples_split`: 분할되기 위해 노드가 가져야 하는 최소 샘플 수
- `min_samples_leaf`: 리프 노드가 가지고 있어야 하는 최소 샘플 수
- `min_weight_fraction_leaf`: `min_sample_leaf`와 같지만 가중치가 부여된 전체 샘플 수에서의 비율
- `max_leaf_nodes`: 리프 노드의 최대 개수
- `max_features`: 각 노드에서 분할에 사용할 특성의 최대 수
  
2. 시각화
3. 가지치기(Pruning): 하부 트리를 제거하여 과적합을 방지 
  
> SVM  

클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아냄  
데이터를 분리하는 초평면 중 데이터들과 가장 거리가 먼 초평면을 선택하여 분리하는 이진 선형 분류 모델  
  
**SVM 구성**
  

  
- Support vector: 데이터 중 결정 경계와 가장 가까이 있는 데이터들의 집합
- Decision Boundary: 데이터 분류의 기준이 되는 경계
- Margin: 결정 경계에서 서포트 벡터까지의 거리
- Hyperplane: n차원 공간의 (n-1)차원 평면, 결정 경계
- Slack Variables: 완벽한 분리가 불가할 때 분류를 위해 허용된 오차에 대한 변수
  
**최적의 선을 찾는 방법**  
- Margin이 가장 큰 경우!
- 결정 경계가 데이터로부터 멀리 떨어질수록 좋음
  
**장단점**  
- 장점
    - 데이터가 적을 때 효과적
    - 새로운 데이터 입력시 전체 데이터와 거리 계산하지 않고 서포트 벡터와의 거리만 계산하면 됨 -> 연산량 최소화
    - 정확성 뛰어남, 비선형 모델 분류 가능
    - 과대 적합 가능성 낮고 노이즈 영향 적음
- 단점
    - 데이터 전처리 과정 중요
    - 데이터가 많아질수록 최적화 위한 평가 과정 많아져서 속도 느림
  
> KNN  

거리가 가까운 k개의 다른 데이터 레이블을 참조하여 분류  
  
**계산 순서**  
1. 데이터 준비: 미리 학습 안함, 각 데이터는 특징 벡터와 레이블로 이루어짐
2. K값 설정: 원하는 K값 설정, K는 가장 가까운 이웃의 개수, 보통 홀수개로 설정(짝수면 동점 발생 가능!)
    - 동점 발생하면: 더 가까운 이웃에 해당하는 클래스로 분류하거나, 동점이 발생한 클래스 중 무작위로 하나 선택
3. 거리 계산: 새로운 데이터가 주어지면 이 값과 기존 모든 데이터 간 거리를 계산
4. 가장 가까운 k개의 이웃 선택
5. 분류: 이웃 중 가장 많이 등장하는 클래스
  
**최적의 K 선택**  
- 학습용 데이터 개수의 제곱근
- K를 너무 크게 설정하면 근접성 떨어져서 과소 적합
- K를 너무 작게 설정하면 이상값, 잡음 데이터와 이웃이 될 가능성
  
**장단점**  
- 장점
    - 범주를 나눈 기준 몰라도 분류 가능
    - 입력 데이터만 주어지면 학습 과정 필요 없음
    - 이해 쉽고 구현 간단
    - 추가된 데이터 처리 용이
    - 훈련 데이터를 요약하거나 일반화된 모델로 만들지 않고, 원본 데이터 전체를 그대로 저장
- 단점
    - 테스트 데이터의 개수에 따라 시간이 오래 걸림
    - 학습데이터의 양도 계산 시간에 영향 미침
    - K값 결정 어려움
    - 수치형 데이터 아니면 유사도 정의하기 어려움
    - 데이터 내 이상치가 존재하면 분류 성능에 큰 영향을 받음

### 분류 평가 지표

> 혼동 행렬  

분류 모델의 예측 결과를 정확한 예측과 잘못된 예측으로 구분하여 나타낸 표  
  
  
  
- 파랑: 예측을 올바르게 함
- 빨강: 예측을 잘못함
  
**혼동 행렬을 이용한 분류 모델 평가 지표**  
- 정확도: 모든 가능한 예측 중 참인 비율
    - 모델이 입력된 데이터에 대해 얼마나 정확하게 예측하는지
    - (TP+TN)/(TP+TN+FP+FN)
    - 단점: 정답 레이블 불균형 -> 모델의 정확도 신뢰 불가
- 정밀도: 참이라고 예측한 경우 실제 참의 비율
    - 거짓을 참이라고 판단한 정도
    - TP/TP+FP
- 재현도: 실제로 참인 경우 참으로 예측하는 비율
    - 참을 거짓으로 판단한 정도
    - TP/TP+FN
*정밀도와 재현도는 모델의 종류와 역할에 따라 중요도가 달라짐!*  
  
**정밀도 & 재현도 그래프**  

두 value값이 만나는 지점을 Threshold로 지정하여 예측 오류를 최소화  
  
  
  
> F1-Score  

정밀도와 재현율의 조화 평균  
$$F1 Score=\frac{2*Precision*Recall}{Precision+Recall}$$
  
- 정밀도와 재현율 간의 균형을 효과적으로 평가
  
> ROC/AUC Curve  

**ROC Curve**: 얼마나 분류가 잘 되었는가를 보여줌
  
  
  
- 좌상단으로 붙어있는 ROC 커브일 수록 더 좋은 이진분류기

  
**AUC Curve**: ROC와 x축 사이의 면적
- 모델의 성능을 숫자로 나타냄 (0.5~1)
- 1에 가까울수록 분류 성능 좋음
  
> 다중 분류 평가 지표  
- Macro average: 클래스별로 구한 평가 지표 평균
- Weighted average: 클래스별로 구한 평가 지표 가중 평균
- Micro average: 모든 클래스의 예측 결과를 더하여 전체적인 성능을 평가하는 지표  
  

### 하이퍼파라미터 최적화

> 하이퍼파라미터 최적화  

**하이퍼파라미터**: 학습 시작 전에 사용자가 직접 설정하는 변수  
**하이퍼파라미터 최적화**: 적절한 하이퍼파라미터를 찾아 모델 성능을 향상시키는 것  

**최적화 과정**  
1. 하이퍼파라미터 탐색 범위 설정
2. 평가 지표 계산 함수(성능 평가 함수) 정의
3. 1단계에서 샘플링한 하이퍼파라미터 값을 사용하여 검증 데이터로 정확도 평가
4. 위 단계를 특정 횟수 반복하여 정확도 결과를 보고 하이퍼파라미터의 범위 좁힘
  
> 최적화 방법  

1. Grid Search: 정해진 범위에서 하이퍼파라미터를 모두 순회하며 가장 좋은 성능을 내는 값을 찾는 기법
- `GridSearchCV` (scikit-learn)
- 장점: 범위가 넓고 step이 작을수록 꼼꼼하게 전 범위를 탐색해 최적해를 정확히 찾을 수 있음
- 단점: 시간이 너무 오래 걸림
- 적용: 넓은 범위, 큰 step을 활용해 범위 좁힘

2. Random Search: 정해진 범위에서 하이퍼파라미터를 무작위로 탐색해 가장 좋은 성능을 내는 값을 찾는 기법
- `RandomSearchCV` (scikit-learn)
- 장점: 속도가 Grid Search보다 빠름
- 단점: 무작위라는 한계 때문에 정확도 떨어짐

3. Bayesian optimization: 사전 정보를 바탕으로 최적 하이퍼파라미터 값을 확률적으로 추정하며 탐색하는 기법
- `bayes_opt` 
- 방법
    - 초기 값 몇 개를 무작위로 시험
    - 그 결과를 바탕으로 어떤 파라미터 조합이 좋을지 확률 모델로 예측
    - 다음 실험할 조합을 가장 좋을 것 같은 위치로 선택
    - 결과를 반영하고 다시 반복
