# 1. 자료의 척도

* 질적 자료: 수치로 표현되는 자료  
    - 명목 척도: 레이블이나 이름으로 요소의 속성을 식별하는 척도.  예) 성별
    - 서열척도: 항목 간의 순서 또는 순위를 나타내는 특성을 지닌 척도. 예) 사원->대리->과장

* 양적 자료: 수치로 표현되더라도 연산의 의미를 갖지 못하는 자료
    - 구간(등간)척도: 값들 사이에 순서가 있고, 그 간격이 일정하여 덧셈이나 평균 같은 연산이 가능한 척도<br>
    예) 온도, IQ, 시간
    <br>**단, 곱셈이나 나눗셈은 의미를 지니지 못한다.**
    <br>즉, 절대 0점이 없으므로, 20도가 10도보다 '두 배 더 따뜻하다' 라고 할 수 없다.

    - 비율척도: 절대영점(Absolute Zero, 없음을 의미)과 임의의 측정단위로 구성하고, 동간성이 유지되는 척도<br>
    ex. 거리, 높이, 무게, 시간<br>
    **절대적인 영점(Absolute Zero, '없음'을 의미)이 존재**하여 측정값 간의 비율 계산(곱하기, 나누기)과 사칙연산이 가능하다.<br>
    등간척도의 경우 0이 임의적(예: 섭씨 0도는 온도가 없다는 뜻이 아님)이라 배수 계산이 불가능


# 2. 기술 통계

개념: 수집한 데이터를 설명하고 이해하는데 중점을 둔 통계방법

## 2.1 3가지 핵심 요소: 중심화 경향, 자료와 산포, 자료의 정리 및 시각화

- 중심화 경향: 데이터의 중심적인 경향을 나타내는 기술 통계량. 
    - 평균-총합을 사용하기에, 이상치에 민감할 수 있음. series.mean()
    
    - 중앙값-평균에 비해 이상치에 덜 민감. series.media() 
    
    - 최빈값-주로 이산형 데이터, 데이터의 분포를 이해하는데 유용. series.mode()
    

- 자료의 산포: 데이터가 퍼져있는 정도를 설명하는 기술 통계량
    - 범위: 데이터의 최댓값과 최솟값의 차이를 나타냄. 양쪽 극단값을 이용하기에 **이상치에 매우 민감.** 

    - 분산: 데이터값이 평균으로부터 얼마나 멀리 퍼져 있는지 나타냄.

    - 표준편차: 분산의 단위는 관측값 단위의 **제곱**이기에 대상 데이터와 일치하지 않음. 따라서 제곱근을 씌워 단위를 일치시켜주는 것이 표준편차. series.std()

    - 사분위수: 데이터 표본을 네 개의 동일한 크기로 나눈 값. <br>1사분위수(Q1): 하위 25퍼센트, 2사분위수(Q2): 하위 50퍼센트, 3사분위수(Q3): 하위 75퍼센트에 위치한 값
        * 사분위간 범위(IQR): Q3에서 Q1 값을 제외한 범위, 즉 데이터의 중앙 50%가 퍼져 있는 범위. 이상치를 탐지할 때 유용함(Tukey Fence)
        <br>아래쪽 이상치: Q1-1.5*IQR 보다 작은 값. 
        <br>위쪽 이상치: Q3+1.5*IQR 보다 큰 값.
        
    - 변동 계수-데이터의 절대적인 변동성이 아닌 상대적인 변동성을 나타내는 지표.

# 3. 확률과 분포

확률분포: 확률 변수가 특정한 값을 가질 확률을 나타내는 함수

# 3.1 확률분포 종류

- 이산확률분포: 확률 변수가 이산적인 값을 가지는 경우. *이산적이다: 유한하거나 셀 수 있으며, 특정 값을 선택 가능 
- 연속확률분포: 확률 변수가 연속적인 값을 가지느 ㄴ경우. *연속적이다: 무한한 범위를 가지며, 특정 값을 선택하는 것이

# 3.2 대표적인 분포 모형

- 정규분포-형태: 평균을 기준으로 대칭, 표준편차가 클수록 곡선이 평평. 
    - 매개변수: 평균, 표준편차.
    - 모집단의 분포를 몰라도 표본의 크기가 충분히 크다면(n>=30)  표본 평균의 분포는 근사적으로 정규 분포를 따른다.
    - 관련 이론: 
        - 큰 수의 법칙: 표본 크기가 클수록 표본 평균이 모평균에 근사하는 현상
        - 중심극한정리: n이 적당히 크다면 n개의 평균의 분포는 정규 분포에 가까워짐

- Z분포(표준정규분포)- 평균이 0이고 표준편차가 1이 되도록 맞춘 표준적인 정규분포

## 3.2.2 표본분포 
- 표본분포: **정규분포를 따르는 모집단**에서 추출한 **표본(sample) 통계량**들이 따르는 분포

    - 목적: 모집단을 전부 조사하기 힘들 때, 같은 모집단에서 여러 번 표본을 뽑아 계산. 

    - 종류: T분포, 카이제곱 분포, F 분포

- T분포: 
    - 형태: 종모양, 정규분포보다 꼬리가 두꺼움. 평균을 기준으로 대칭. 자유도가 커짐에 따라 표준정규분포 N(0,1)에 수렴

    - 매개변수: 자유도(df1) 
        - *자유도란?: 얼마나 많은 정보가 독립적으로 변할 수 있는가를 나타내는 숫자. <br> ex) 4,6,8 이 있으면 평균은 6. 이때 4와 평균이 6이라는 정보만 주어지면 나머지 숫자는 8이 될 수밖에 없음. 이때 4,6이 자유롭게 정해질 수 있으므로 자유도는 2.

    - 특징: 표본 크기가 작을 때 정규분포보다 더 정확한 결과 제공->꼬리가 두꺼워서 평균이 멀리 벗어나는 경우도 더 가능성 있게 인정하기 때문.

    - 목적: 표본이 적을 때 모평균을 추정하기 위함.

- 카이제곱분포
    - 형태: 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포. 자유도가 증가함에 따라 중심이 오른쪽으로 이동하고 모양이 정규분포에 가까워짐

    - 매개변수: 자유도(df1)

    - 특징: 서로 독립적인 k개의 표준정규분포 확률 변수를 각각 제곱하여 합한 값->즉, 본분산의 분포와 관련된 지표. 

    - 목적: 실제 관측값과 예상 값의 차이를 비교하여 우연인지 확인.

- F분포
    - 형태: 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포. 자유도가 증가함에 따라 중심이 오른쪽으로 이동하고 모양이 정규분포에 가까워짐.
    - 매개변수: 자유도(df1,df2)

    - 특징: 두 독립적인 카이 제곱 분포의 비율.

    - 목적: 두 집단의 분산 차이를 비교하는데 사용됨.

    - **\*왜 카이제곱 분포의 비율인가?**<br>
        ->F 분포의 목적: 두 집단의 분산 차이를 비교.->이를 위해 정규 분포에서 표본을 뽑아 분산 계산.
        ->표본 분산값을 구한 후 분산의 확률분포를 알기 위해 표준화 진행->이를 통해 나온 값이 카이제곱분포.

# 4. 추론 통계

- 개념: 전체 집단(모집단)에서 추출한 일부분(표본)의 데이터를 분석하여, 모집단의 특성(모수)을 추정하거나 가설을 검정하는 통계 방법. ex) 날씨 예측

- 핵심 방법
    - 추정: 표본의 평균 등으로 모집단의 평균(모수)을 예측. 단일 값으로 예측하는 '점추정', 범위를 지정하는 '구간 추정'이 있음.
    - 가설 검정: 특정 가설이 표본 데이터와 비교하여 통계적으로 유의미한지 검증.

- 가설 검정 절차: 
    1. 가설 설정-귀무 가설과 대립가설 설정. 
        - 귀무 가설: 기각하고자 하는 명제
        - 대립 가설: 주장하고자 하는 명제

    2. 유의 수준 설정
        - 연구자가 결과를 판단할 때 허용할 수 있는 오류의 최대치로, 주로 0.05로 설정.
        - 귀무가설이 참이었는데 기각할 확률이 유의 수준

    3. 검정통계량 산출
        - 가설 검정의 결과를 판단하는데 사용되는 값. 
        - 데이터 분포나 특징에 따라 Z,T,F 값 등을 선택
    4. 가설 기각/채택 판단
        - 유의 확률(p-value)이 유의 수준보다 작을 시에 귀무가설 기각.
