# 1. 머신러닝이란?
- 모델이라는 소프트웨어를 학습하여 새로운 데이터를 예측하거나 결정을 내릴 수 있도록 하는 기술

- 종류
    - 지도학습
        - 정답 여부: 정답(label)이 있는 데이터 사용. 
        - 방식: 기계의 예측값이 데이터의 정답과 같아지도록 학습시킴. 
        - 예시) 회귀, 분류
    - 비지도학습
        - 정답 여부: 정답이 없는 데이터 사용.
        - 방식: 데이터의 패턴 혹은 유사성을 학습시키는 것. 
        - 예시)군집화
    - 강화학습
        - 정답 여부: 보상. 
        - 방식: 시행착오를 반복하여 보상을 최대화 하는 행동 학습. 
        - 예시) 알파고, 자율주행

## 1.1 모델링이란?
- 예측기(공식)을 만들어가는 전체적인 과정
- 기계가 무지 상태에서 주어진 데이터로 스스로 패턴을 찾아내 오차가 작은 최적의 공식을 완성할 때까지 학습하는 과정.
    - 과정: 설계->훈련->검증

# 2. 지도학습이란?
- 정답 레이블을 함께 주고 학습을 시키는 방법. 예측값(prediction)과 정답(label)이 최대한 같아지는 방향으로 학습함.

- 과정: 
    - 1. labeled data(독립변수(예측에 사용되는 입력 변수)와 종속변수(예측하고자 하는 대상 변수)가 모두 존재하는 데이터) 준비
    - 2. labeled data를 training set과 test set으로 나눔. 
    - 3. training set을 통해 모델을 학습.
        ->독립변수를 통해 추정한 예측값(prediction)이 주어진 종속변수(target/label)와 가까워지도록 훈련하는 과정

## 2.2 대표적인 지도학습 예시
- 회귀: 주어진 데이터를 기반으로 정답을 잘 맞추는 함수를 찾는 문제. 
    - 예측하고자 하는 값(=종속변수)가 연속형이면 사용
- 분류: 기존 데이터가 **어떤 레이블에 속하는지** 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별하는 것. 
    - 종속변수가 범주형이면 사용.

# 3. 선형회귀
- 선형 회귀란 데이터를 가장 잘 설명하는 회귀선을 찾는 과정
- 현재 데이터 요약 및 미래의 값을 예측하는데 활용
- 단순선형회귀(x가 한 개)와 다중선형회귀(x가 여러 개)로 나누어짐.

## 3.1 가중치와 편향
- 선형 회귀 기본 식인 y=wx+b를 기준으로 w는 가중치, b는 편향임.

1) 가중치- 독립변수 x가 종속변수 y에주는 영향력의 크기. 그래프에서는 기울기. 회귀계수라고도 불림

2) 편향- 독립변수 x가 0일때 나오는 기본값으로, 그래프에서는 y절편. 모델이 데이터에 더 유연하게 최적화 하도록 위아래로 움직이는 역할.


# 4. 단순 선형회귀
- 데이터를 직선형태(하나의 독립변수)로 표현하는 경우
- 잔차를 줄여나가며 직선을 결정짓는 적절한 회귀계수(w)와 편향(b)를 찾아나감

최적의 w와 b를 찾는 방법에는 최소 제곱법, 경사하강법이 있다. 

## 4.1 최소제곱법 (Least Squares Method)
- 잔차(실제값과 예측값의 차이)의 제곱의 합(RSS)이 **최소**가 되는 지점으로 최적의 회귀선을 구하는 방식
- 회귀선과 실제 데이터와의 차이를 최소화 하기 위한 시도

## 4.2 경사하강법(Gradient Descent)
- 최소제곱법을 통해 구한 함수가 복잡하여 **최솟값을 구하기 어려울 때** 사용.
- 잔차제곱을 줄이려는 목적은 그대로

    <span style="background-color: black">목적함수, 비용함수, 손실함수</span>

- 목적함수: 최적화하려는 전체 대상. 최대화/최소화가 목표
- 손실함수: 데이터 한 개에 대한 오차. 최소화가 목표
- 비용함수: 전체 데이터에 대한 평균 오차. 최소화가 목표

### 4.2.1 손실함수와 경사 하강법

- 오차를 구하는 식으로 **평균 제곱 오차(MSE)**를 사용함.

$$E = \frac{1}{n} \sum_{i=1}^{n} [t_i - y_i]^2$$

- t_i: i 번째 데이터포인트 실제값(정답)
- y_i: i 번째 예측값
- E: 오차
- n: 데이터 샘플의 총 개수
 
이 식에 $$y=Wx+n$$를 대입해 최종적으로 아래와 같은 손실함수 얻을 수 있다.

$$E(W, b) = \frac{1}{n} \sum_{i=1}^{n} [t_i - y_i]^2 = \frac{1}{n} \sum_{i=1}^{n} [t_i - (Wx_i + b)]^2$$

여기서 목표는 E를 최소화하는 W,b를 찾는 것. 

**W와 b**에 대해 **E가 변화하는 양상**을 보기 위해서 W로 x축을 구성하면, W에 대한 2차식이 됨. 

오차를 최소화 하기 위해서는 W가 최소 지점, <u>즉 기울기가 0이 되는 지점</u>으로 이동해야 한다.

따라서 미분계수가 **양수**일때 W는 **왼쪽**으로 가도록 업데이트
미분계수가 **음수**일때 W는 **오른쪽**으로 가도록 업데이트 한다. 

### 4.2.2 학습률
- 이동 보폭 크기라고 생각하기
- 만약 학습률이 너무 작다면->모델이 수렴할 때까지 너무 <u>오랜 시간</u>이 걸림
- 만약 학습률이 너무 크다면->모델이 수렴하지 못하고 <u>발산</u>함.

### 4.2.3 Local Minima 문제
-  전체에서 가장 낮은 점인 **Global Minima** 와 달리, 특정 구간에서만 최소인 지점. <u>국지적인 골짜기 지점임.</u> 
- 해결법: 모멘텀
    - 원리: 이전의 기울기와 방향을 고려하도록 관성을 부여. <br>
    ->기울기에 관성을 부여하여 작은 기울기는 쉽게 넘어갈 수 있도록 만듦. 

# 5. 다중 선형 회귀
- 독립 변수가 두 개 이상인 경우. 
- 독립 변수 x_i 가 2개인 경우 3차원 공간에 표현되고, 회귀선은 평면이 됨. 변수가 많아질 수록 비슷한 느낌으로 차원이 확장됨.
- 최소제곱법/경사하강법 사용 가능. 이때 **잔차**는 관측치와 평면 간의 차이가 됨.

# 6. 다중공선성
- 독립 변수들 간의 *강한 상관관계*로 인해 회귀 모델이 불안정해지는 문제. 
- 다중공선성이 높은 경우 어떤 독립 변수가 종족 변수에 얼마나 영향을 미치는지 정확한 구분이 어려움. 

## 6.1 다중 공선성 확인 방법
- 상관 계수(r) 
- VIF 지수 (분산 팽창 인수)

## 6.2 다중공선성 대처 방법
1. 변수 제거
2. 변수 변환
3. 규제선형모델 활용
4. PCA(주성분 분석)

# 7. 규제선형모델
- 모델이 학습 데이터에 **과적합(overfitting)** 되지 않도록 규제를 가하고자 등장한 모델
    <br>-> 선형회귀 모델에서는 특성에 곱해지는 **계수(기울기)의 크기를 조정하는 것

- 최소제곱법과 같이 RSS(잔차 제곱합)을 최소화하는 방식은 <u>회귀 계수가 쉽게 커지게 되어</u> 과적합이 발생할 가능성이 커짐
    <br>-> 따라서 RSS 최소화 방법과 회귀 계수 값이 커지지 않도록 하는 방법이 균형을 이뤄야 함.

## 7.1 규제선형모델의 종류
1. L2규제: W(회귀계수)의 제곱에 대해 페널티를 부여하는 방식, *릿지(Ridge)회귀*
2. L1규제:  W(회귀계수)의 **절댓값**에 대해 페널티를 부여하는 방식, *라쏘(Lasso) 회귀*
3. L2규제 + L1규제: *엘라스틱넷(Elastic Net) 회귀*

### 7.1.1 릿지 회귀(Ridge, L2규제)
- L2규제는 <u>기둥에 따라 다른 힘으로 누르는 망치</u>

- 회귀계수의 제곱을 패널티로 주기 때문에, **큰 회귀계수일 수록 훨씬 강한 패널티**를 받음.

- 모든 계수의 크기가 작아지지만, 규제를 통해 변수가 제거되지는 않음.

- 모든 변수를 유지하면서, 크기를 적절히 줄여 과적합 방지

### 7.1.2 라쏘 회귀(Lasso, L1 규제)
- L1규제는 <u>모든 기둥을 같은 힘으로 누르는 망치</u>

- 작은 기둥들은 힘을 버티지 못하고 완전히 사라질 수도 있음

- 회귀계수의 절댓값을 패널티로 주기 때문에, 회귀계수의 크기와 상관없이 같은 패널티를 받음

- 비중이 낮은 변수(작은 계수)는 아예 사라지고, 중요한 변수만 남음


### 7.1.3 엘라스틱넷 (Elastic Net, L1+L2)
- 불필요한 변수는 제고하고, 남은 변수들은 적절한 크기로 유지

- 안정적인 예측(모든 변수가 균형 있게 줄어듦)과 희소성(변수 제거)을 동시에 달성할 수 있음!

# 8. 모델 평가 방법
1. 평균 제곱 오차(MSE)
2. 평균 절대 오차(MAE)

        차이점: MSE는 제곱이기에 이상치에 민감하게 반응.
        MAE는 모든 오차를 동일하게 취급하여 이상치에 덜 민감.
