# 1. 다중선형회귀의 기본 가정 여섯가지

선형 회귀를 사용하고 분석 결과의 신뢰성을 확보하기 위해서는 다음 기본 가정 6가지가 필요하다.

1. 선형성
- 종속변수와 설명변수 간의 관계가 선형적이어야 한다.<br>
$E(y|X) = \beta_1+\beta_2X_2+...+\beta_nX_n$
- 위의 형태를 따라야함.

2. 독립성
- 각각의 설명변수가 서로 선형독립적이어야 한다.
- 선형독립적이지 않은 경우 **다중공선성**이 있기때문!

3. 오차항의 평균은 0이다(0에 가까워야 한다)
- 오차항($\epsilon$): 실제 값과 예측 값 사이의 차이
- 오차항은 다른 말로 **White noise**라고도 하는데, 백색 잡음의 성질을 만족하기 위해서 필요한 가정 중 하나임
- 백색 잡음이란: 특정 패턴 없이 랜덤하게 변하는 신호. **편향되면 안됨**
<br>ex) **오차항의 평균이 3이라면(=0이 아니라면)**, 회귀 모델이 **항상(평균적으로) 실제 값보다 3만큼 부족/넘치게** 예측

4. 등분산성
- 오차항의 분산이 일정해야 한다는 가정
- 오차항의 분산이 일정하지 않은 경우, 이분산성이 있음. 그 결과 모델의 예측이 **일부 구간**에서는 **매우 정확**하지만, **일부 구간**에서는 **오차가 매우 커질 수** 있다.
사진 삽입

5. 오차항은 자기상관되어 있지 않다.
- 오차항의 공분산은 항상 0이어야 된다는 가정
- 공분산이란?: 두 변수 간의 **상관된 정보**를 나타내는 지표. 예) 양의 상관관계, 음의 상관관계
- 오차항의 공분산이 0이 아닌 경우, 자기상관이 있음. 자기상관: 한 변수의 **현재 값이 과거 값과 상관관계**를 가지는 현상.
- 모델이 이전 데이터의 패턴을 학습하여 반복적인 오류를 만들 가능성 높음. 

6. 정규성(선택)
- 오차항이 정규분폴르 따른다는 가정
- 위배되어도 다중선형회귀분석의 결과에 큰 영향을 주지 않음

        선형성, 독립성(다중공선성) 가정은 회귀 분석 전, 후 모두 검증 가능.
        오차항 평균이 0이다, 등분산성, 정규성은 회귀 분석 후에 검증 가능
        오차항 자기상관 없다는 회귀 분석 전에는 간접적으로 검증 가능, 분석 후에는 최종 검증

# 2. 회귀 분석의 평가 방법

- 시각화
: 회귀선을 데이터와 함께 시각화할 때, 데이터 전반을 요약할 수 있음.
: 두 회귀선 중 어느 것이 데이터를 더 잘 요약했는지 객관적으로 비교하기 어려움.

- 통계지표: 유의성 검정, 성능 평가
    - 유의성 검정: 
        - F-statistic, Prob(F-statistic)
        - F-statistic이 클수록 통계적으로 유의미.
        - Prob(F-statistic)이 0.05보다 작다면 회귀 모델이 유의

        - T-statistic, Prob(T-statistic)
        - P > |t| (t통계량의 유의확률)가 0.05보다 작으면 해당 독립변수는 종속변수에 유의미한 영향을 미친다.
    - 성능평가
        - 결정 계수 (R-squared, $R^2$)
        - 조정된 결정 계수(Adjusted R-squared, Adjusted $R^2$)
	        - 변수 개수 증가에 덜 민감하도록 조정한 지표
        - AIC, BIC(SC)

# 3. 비선형회귀모델

## 3.1 다항식 회귀 모델
- 기존의 선형 회귀 $Y=a+bX$ 에서 독립 변수에 거듭제곱 함을 추가->$Y=a+bX+cX^2+dX^3+...$<br>
    ->데이터의 **곡선적(비선형적) 특성을 모델링**함.

- 장점: 선형 모델에 비해 회귀선을 데이터에 fit 하게 그릴 수 있음
- 단점: 너무 많은 독립변수를 이용하면 오버피팅 발생.
- 적합한 상황: 곡선을 그리는 데이터 패턴, 성장률, 감소율 등 시간에 따라 변화하는 속도가 다른 경우.

## 3.2 지수 회귀 모델
- 종속 변수가 지수적으로 변화하는 관계 모델링 할 때 사용.
- 사용 방법
- 지수함수 형태의 데이터는 $Y = ae^{bX}$ 수식 따름
    ->양 식에 로그를 취하면 lnY=lna + bX=> 선형회귀식과 비슷<br>
    ->모델이 ln Y와 X의 선형 관계를 학습. 이때 Y값을 예측 시, 모델이 예측한 Y값에  지수함수$(e^x)$를 적용하여 로그 변환하기 전 Y로 변형<br>
    ->$e^(lnY)=e^(lna+bX)$ ->Y=a+e^bx
- 적합한 상황: 값이 시간에 따라 지수적으로 증가하거나 감소하는 경우. 예) 기술 발전, 투자 수익률 증가

## 3.3 로그 회귀모델
- 종속변수와 독립변수의 관계가 로그 함수를 통해 더 잘 표현될 때 사용.
- 사용 방법: Y=a+bX 선형식에서 독립변수 X에 로그 연산을 취한 후, 선형회귀 적용.
- 적합한 상황: 인구 성장, 감영병 확산과 같이 **초기 빠른 증가 후 점진적 감소**를 보이는 데이터.

## 3.4 스플라인 회귀
- 데이터를 구간별로 나누고 각 구간에서 다른 선형 또는 비선형 함수를 적용하여 예측하는 방식
- 사용 방법: - 전체 데이터 범위를 여러 구간으로 나누고, 각 구간에 대해 별도의 회귀모델(선형, 다항 등) 적용
- 구간 경계에서의 **연속성을 유지**하는 것이 중요.
- 적합한 상황: 계절과 같이 여러 구간에서 데이터 패턴이 서로 다르게 나타나는 상황.
