# 1. 회귀와 분류의 차이

- 회귀: 연속적인 숫자값을 예측. 주어진 데이터(X)를 기반으로 정답(Y)를 잘 맞추는(fit)하는 함수를 찾는 문제
- 분류: 입력된 데이터를 주어진 항목으로 나눔. <br>
    -> 기존 데이터가 **어떤 레이블에 속하는지** 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 **레이블을 판별**하는 것


#### 분류의 종류
- 이진 분류: 예측하고자 하는 변수가 어떤 기준에 대해 참(True) 또는 거짓(False)의 값만을 가질 때
- 다중 분류: 예측하고자 하는 변수가 가질 수 있는 값이 3개 이상일 때

# 2. 분류 모델의 네 가지 종류
- 로지스틱 회귀: 독립변수와 종속변수의 *선형 관계성*에 기반해 분류
- 결정 트리: 데이터 *균일도*에 따른 규칙 기반의 분류
- 서포트 벡터 머신: 개별 클래스 간의 *최대 마진*을 효과적으로 찾아 분류
- 최소 근접 알고리즘: *근접 거리* 기준으로 분류

## 2.1 로지스틱 회귀
- 개념: **이진 분류 문제**를 푸는 대표적인 알고리즘으로, 샘플이 <u>특정 클래스에 속할 확률</u>을 추정하여 기준치에 따라 분류하는 것이 목표<br>
->독립 변수의 선형 조합 **(선형 회귀)**에 **로지스틱 함수를 적용**하여 출력값을 **0에서 1 사이로 변환**해주는 것을 의미

- 필요성: 직선으로 이진 분류 문제를 푸는데 한계가 있음. 
- 선형 회귀는 결과값이 무한대로 커질 수 있어 0과 1 사이의 확률을 표현할 수 없으나, 로지스틱 회귀는 시그모이드 함수를 통해 확률을 0~1 사이로 제한할 수 있음. 

### 2.1.1 시그모이드 함수
- 개념: 출력이 0과 1 사이의 값을 가지면서 S자 형태로 그려지는 함수

$S(x)=\frac{1}{(1+e^{-x})}$

- 특징: 
    - 입력값이 커지면 1에 수렴하고, 입력값이 작아지면 0에 수렴
    - 입력값(x)이 (-무한대, 무한대)일 때, 출력값(y)은 0~1의 값을 가짐

사용법: 시그모이드 함수가 $H(x)=\frac{1}{(1+e^{-{wx+b}})}=sigmoid(wx+b)$일 때, 인공지능은 주어진 데이터에 **적합한 가중치 w와 b를 계산해서 찾음**

## 2.2 의사결정나무
- 개념: *조건에 따라 데이터를 분류*하며, 최종적으로 데이터가 순수한 label의 집합으로 구성될 때까지 *분류를 반복*하는 분석 방법

        의사결정나무의 구성 요소

- **Root Node**: 가장 상위에 위치한 노드
- **Parent Node**: 주어진 노드의 상위에 있는 노드
- **Child Node**: 하나의 노드로부터 분리된 2개 이상의 노드들
- **Binary Tree**(이진 트리): Tree 중에서 children이 최대 2개인 tree
- **Leaf Nodes**: 자식마디가 없는 노드, 모델에서 label에 해당
- **Edge**: Parent node와 child node를 연결하는 연결선
- **Height**: 특정 노드에서 가장 멀리 있는 leaf node까지의 경로에 있는 edge 개수
- **Depth:** Root node에서 특정 노드에 도달하기 위한 edge의 수

## 2.3 CART(이진분할) 알고리즘
- 가장 널리 사용되는 의사결정나무 알고리즘. 데이터셋을 **임계값**을 기준으로 *두 child*로 나누는 알고리즘
- 임계값 나누는 기준: **불순도(지니 계수)가 낮아지는** 방향으로
	- 불순도란? : 분류하려는 데이터 집합에서 다른 클래스가 섞여 있는 정도를 의미

- CART 알고리즘의 주요 단계
    1. 임계값 설정
    2. 불순도 감소 알고리즘을 활용하여 불순도가 낮은 쪽으로 가지 형성

**실제 학습시 고려해야 할 것들**
1. Hyperparameter 설정
2. 시각화
3. Prunning(가지치기)- 불필요한 노드 지우기

## 2.4 SVM(서포트 벡터 머신)
- 개념: 클래스를 분류할 수 있는 **다양한 경계선 중 최적의 라인을 찾아내는** 알고리즘

        [SVM의 구성]
- Support vector (서포트 벡터): 데이터 중에서 결정 경계와 가장 가까이에 있는 데이터들의 집합
- Devision Boundary (결정 경계): 데이터 분류의 기준이 되는 경계
- Margin (마진): 결정 경계에서 서포트 벡터까지의 거리 (여유 공간)
- Hyperplane (초평면): n차원 공간의 (n-1)차원 평면
- Slack Variables (슬랙 변수/여유 변수): 완벽한 분리가 불가능할 때 분류를 위해 허용된 오차에 대한 변수

**가장 좋은 선: 데이터로부터 가장 멀리 떨어져 있는게 좋음**

## 2.5 KNN(최소 근접 알고리즘)

개념: **비슷한 특성을 가진 데이터끼리 서로 가까이 있는다는 점**을 이용하여<br>
데이터로부터 **거리가 가까운 k개의 다른 데이터 레이브를 참조하여 분류**하는 알고리즘. 

**특징: 어떠한 학습이 필요하지 않다.**

        KNN 알고리즘 계산 순서
1. 데이터 준비
2. K 값 설정 (원하는 K값을 설정. K는 가장 가까운 이웃의 개수를 나타내며, 보통 홀수개)
3. 새로운 데이터와 기존 모든 데이터 간의 거리 계산
4. 가장 가까운 K개의 이웃 선택
5. 분류 (k개의 이웃 중 가장 많이 등장하는 클래스가 예측 결과)

        최적의 k 선택하는 법
- 일반적으로 학습용 데이터 개수의 제곱근으로 설정. 
- 너무 크면 근접성이 떨어져 과소 적합.
- 너무 작으면 이상값, 잡음 데이터와 이웃 될 가능성 있음

# 3. 분류 평가 지표

## 3.1 혼동 행렬
- 개념: 분류 모델의 예측 결과를 **정확한 예측과 잘못된 예측으로 구분**하여 나타낸 표

- 혼동 행렬 종류

|구분|분류값|설명|
|---|---|---|
|예측이 정확한 경우|True Positive(TP)|정답이 참, 참으로 예측|
|   | True Negative(TN) | 정답이 거짓, 거짓으로 예측|
| 예측이 틀린 경우 |False Positive(FP) |정답이 거짓, 참으로 예측 |
|   | False Negative(FN)| 정답이 참, 거짓으로 예측 |

#### 3.1.1 혼동 행렬을 이용한 분류 모델 평가 지표

- 정확도: <u>모든 가능한 예측 중</u> 참인 비율
	- (T로 시작): (TP+TN)/(TP+TN+FP+FN)
- 정밀도: <u>참이라고 예측한 경우</u> 실제 참의 비율
	-TP/TP+FP
- 재현도: <u>실제로 참인 경우 중</u> 참으로 예측하는 비율
	- TP/TP+FN<br>
cf) 정확도, 정밀도, 재현도는 모두 0과 1사이의 값


**Threshold와 정확도, 재현도의 관계**


- Threshold: 모델의 참 거짓을 가르는 분류 기준 <br>예) Threshold가 0.5라면, 모델은 예측값이 0.5 이상일 때 참으로 예측

- 만약 Threshold가 0.5에서 0.3으로 줄어든다면, 0.3~0.5 값들은 거짓->참으로 바뀜.  <br>
->실제 참인 것 중 참이라고 예측한 것의 비율인 Recall 값 증가<br>
->반대로, 스팸의 경우 스팸이 아닌 메일을 스팸으로 분류(FP) 하는 현상을 막기 위해서는 Precision (정밀도)을 높여야 함.<br>
->Threshold를 높여야 함

- Threshold에 관해서 Recall과 Precision은 **반비례** 관계!

#### 3.1.2 F1-Score
- 개념: 정밀도(precision)와 재현율(recall)의 조화 평균

#### 3.1.3 ROC/AUC curve
- ROC:  **얼마나 분류가 잘 되었는가**를 보여주는 그래프

**좌상단으로 붙어있는 ROC 커브일 수록 성능이 좋다**

- AUC curve: ROC와 x축 사이의 면적(적분값)
    - 0.5에서 1사이의 값으로 모델 성능을 표현. **1에 가까울수록 분류 성능이 좋음**


# 4. 하이퍼파라미터 최적화 개념

- 하이퍼파라미터: 학습 시작 전 **사용자가 직접 설정하는 변수**

- 하이퍼파라미터 최적화: **적절한 하이퍼파라미터를 찾아** 모델 성능을 향상 시키는 것

- 최적화 과정

	- 1. 튜닝할 하이퍼파라미터(예: 학습률, 은닉층 수, 배치 크기)와 그 후보 값의 범위 설정
	- 2. 모델의 성능지표를 극대화하거나 손실함수를 최소화하는 함수를 정의
	- 3. 최적화 알고리즘 선택 (그리드 서치, 랜덤 서치, 베이지안 최적화 등)
	- 4. 학습 및 평가 반복(검증 데이터셋으로 성능 평가)
	- 5. 최적의 하이퍼파라미터 조합을 최종 선정
