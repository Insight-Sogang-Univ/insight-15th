# 1. 비지도 학습

- 개념: 모델에게 답을 가르쳐주지 않고 학습시키는 방법
- 특징: 
    - 데이터 자체에 내제된 구조를 파악하여 학습
    - 정답이 없으므로 모델의 성능을 명확히 측정 불가.

- 적합한 케이스: 패턴이 알려지지 않은 경우, 패턴이 계속해서 변하는 경우, 열린 문제를 해결하고 지식을 일반화해야 하는 경우

# 2. 군집화의 개념과 기본 과정
- 개념: 데이터를 비슷한 특성을 가진 그룹(군집)으로 나누는 비지도 학습 기법 .<br>
종속변수를 설정하지 않고, 데이터 내부의 고유한 패턴을 발견하고 추출하기 때문에 **비지도 학습**으로 분류

- 목표: 응집화 최대화(같은 군집에 속하는 데이터는 최대한 비슷하게), 분리도 최대화(서로 다른 군집은 최대한 분리)

- 군집화 과정
    1. 피처(특징) 선택 또는 추출
    2. 군집화 알고리즘 선택
    3. 군집 유효성 검증
    4. 결과 해석

### 2.1 군집화 고려 사항

1. 변수 유형 이해
- 연속형: 
	- 거리측정: 유클리디안 거리, 멘하탄 거리 등 사용 가능
	- 적합한 알고리즘: k-means, 계층적 군집화 등
	- 주의사항: 스케일링 필수
- 명목형
	- 거리측정: 해밍거리, 자카드 거리 등
	- 적합한 알고리즘: k-modes, 계층적 군집화(특별한 거리 함수 사용)
	-  처리방법: 원-핫 인코딩 또는 더미 변수 변환
- 혼합형
	- 적합한 알고리즘: K-prototypes, Gower distance 기반 계층적 군집화
	- 처리방법: 각 변수 유형에 맞는 거리함수를 조합하여 사용

2. 거리/유사도 정의와 측정
- 군집화는 데이터 간의 거리 또는 유사도를 기반으로 그룹을 형성

3. 차원 축소
- 모델 성능을 향상시키기 위해 유사한 변수들을 묶어 처리하는 차원 축소 고려

# 3. 거리 / 유사도 정의와 측정 방법

- 데이터 거리 측정 방법: 거리가 가까울수록 유사, 멀수록 다르다고 판단.
- 종류: 
    - 유클리디안 거리: 가장 일반적인 거리 측정 방법
    - 맨하탄 거리: 이상치에 덜 민감
    - 코사인 유사도: 벡터 간 각도 측정. 텍스트 데이터나 고차원 데이터에 효과적

### 3.1 6가지 연결 방법 (거리 계산 기준)

- 개념: 계층적 군집화에서 두 군집 간의 거리를 계산하는 방식. 각 방법은 군집 간 거리 측정 기준이 다르다. 

1. Single linkage(단일 연결법/최단 연결법)
- 두 군집 간 **가장 가까운 두 점 간 거리**로 군집 간 거리 정의
- 고립된 군집을 찾는것이 중점
- 이상치에 취약

2. Complete linkage(완전 연결법/최장 연결법)
- 두 군집 간 **가장 먼 점들 사이의 거리**로 군집 간 거리 정의
- 군집들의 내부 응집성에 중점을 둠.
- 이상치에 민감

3. Average linkage(평균 연결법)
- 두 군집의 **모든 항목에 대한 평균 거리**를 구함
- 최단 연결법, 최장 연결법보다 이상치에 덜 민감

4. Centroid Method(중심 연결법)
- 두 군집의 **중심 간의 거리를 측정**
- 시간 오래 걸리고, 덴드로그램에서 거리 축이 위에서 아래로 줄어드는 inversion이 발생할 수 있음

5. Median(중앙 연결법)
- 두 군집의 **중앙값** 사이의 거리
- 두 군집의 **중심점 사이의 산술적 중간 지점을 새로운 대표값**으로 설정하여 거리 측정
- 극단값에 덜 민감

6. Ward's Procedure(Ward 연결법)
- 군집의 병합 후 **군집 내 SSE(오차제곱합)의 증가분이 최소**인 것을 선택
- **병합된 군집의 SSE의 증가량이 가장 작아지는** 방향으로 군집 형성

# 4. 주성분 분석(PCA)의 목적

        차원 축소가 필요한 이유

- 차원이 크면 알고리즘이 데이터를 효과적/효율적으로 훈련할 수 없다.
- 차원 축소란 변수의 개수를 줄이는 것
- 종류:  
	- **선형 투영**: **주성분 분석(PCA)**, 특잇값 분해(SVD), 랜덤 투영 등
	- **매니폴드 학습(비선형 차원 축소)**: t-SNE, UMAP, Isomap

#### 주성분 분석(PCA)

- 목적: 다차원 데이터에서 가장 <u>정보량(분산)이 큰 방향</u>을 찾고, 그 방향을 기준으로 데이터를 투영하여 차원을 줄이는 것.
    - 이때 해당 차원의 가장 많은 분산을 담아내는 축이 '주성분'이 된다.
- 특징: 데이터의 분산을 최대한 보존하면서 차원을 줄여 시각화와 분석이 용이

# 5. 4가지 군집화 알고리즘의 특징과 과정

#### 5.1 계층적 군집화

- 개념: 데이터 간의 유사성을 기반으로 트리 구조를 형성하며, 상향식 또는 하향식 방식으로 군집을 형성해 나가는 방법
- 특징: **군집의 개수를 사전에 설정하지 않고**, 클러스트링이 종류된 후에 원하는 군집의 개수를 선택할 수 있음
- 트리 구조가 상향식이면 응집형 계층적 군집화, 하향식이면 분리형 계층적 군집화. 

응집형 계층적 군집화

- 각 데이터를 독립적인 군집으로 보고, 가장 유사도가 높은(거리가 가까운) 군집 쌍을 차례대로 병합하여 최종적으로 하나의 큰 군집으로 묶어 나가는 방식.

- 가장 가까운 클러스터를 정의하는 방식-단일 연결, 완전 연결

- 완전 연결을 사용한 응집형 군집화 과정

    1. 거리 행렬 계산
    2. 단일 클러스터 시작
    3. 가장 가까운 클러스터 병합
    4. 거리 행렬 업데이트
    5. 반복(모든 데이터 포인트가 하나의 클러스터에 속할 때 까지)

#### 5.2 k-means

- 개념: 데이터를 정해진 개수의 그룹으로 나누되, <u>각 그룹의 중심적과의 거리가 가장 가까운 데이터끼리 묶는 군집화 알고리즘.</u>

+ 단계: 
    1. 데이터 표본들 중에서 랜덤하게 k개의 중심점을 초기 클러스터 중심으로 선택
    2. 각 표본들을 가장 가까운 중심점(보통 유클리드 거리 사용)을 기준으로 해당 군집에 할당
    3. 각 클러스터에 할당된 표본들의 데이터 평균을 계산하여 중심점을 이동
    4. 중심점이 더 이상 변하지 않거나 사용자가 지정한 허용오차 또는 최대 반복 횟수에 도달할 때까지 반복

- 장점: 직관적이고 구현이 쉬움. 대용량 데이터에도 적용이 가능
- 단점: 초기 중심점값에 민감함. 군집 수 결정이 어려움. 아웃라이어에 민감. 기하학적인 모양의 군집은 파악하기 어려움.

[K-means++알고리즘]

- k-means에서 **초기 중심점들을 서로 멀리 떨어진 곳에 위치**시켜 더 일관되고 좋은 결과를 도출하는 개선된 초기화 방법

[엘보우 방법(Elbow Method)]

- 클래스 내 SSE를 바탕으로 그래프를 활용해 **최적의 클러스터 개수 k를 추정**하는 방법

- 왜곡이 빠르게 감소하는 지점의 k값을 찾음.

- 장점: 직관적, 구현이 간단

- 단점: 주관적 판단이 개입, 데이터에 따라 엘보우 지점이 명확하지 않을 수 있음.

### 5.3 DBSCAN

- 개념: **밀도가 높은 지역의 데이터를 하나의 군집**으로 묶고, 밀도 기준을 만족하지 못하는 점은 군집에 포함시키지 않는 군집화 알고리즘. 
- 특징: 이상치로 부터 자유로움, 특정 데이터가 여러 군집의 특정 거리 내에 있으면 가장 밀도가 높은 군집과 그룹화

- 단계: 

    1. 핵심/경계/잡음 샘플로 분류
    2. 클러스터 생성
    3. 경계 샘플 할당

- 장점
    - k-means처럼 클러스터 모양을 원형으로 가정하지 않아 임의의 기하학적 분포를 갖는 데이터도 잘 처리함.
    - 이상치에 민감하지 않음. 

- 단점 
    - epilson(두 포인트가 이웃이 되기 위한 포인트 사이의 최대 거리)과 min_samples(한 포인트가 군집이 되기 위한 eps 거리 내 최소 포인트 개수) 설정에 많은 영향을 받음
    - 연산량이 많아 속도가 느림
    - 다른 밀도 분포를 가진 데이터의 군집 분석을 잘 하지 못함
    - 변수가 늘어날수록 정확도가 떨어짐

#### 5.4 GMM

- 개념: 데이터가 여러 다른 모양의 가우시안 분포로 구성되었다고 가정하고, 각 분포를 클러스터로 인식하는 군집화 방법
    ->단일 분포로 표현하기 어려운 복잡한 형태의 확률 분포를 여러 개의 가우시안 분포를 합쳐서 표현하자는 아이디어

- 특징
    - 모델 기반 군집화 방법(데이터가 해당 모델로부터 생성되었다는 전제 하에 군집화 수행)
    - 각 군집을 **확률 분포**로 간주
    - 전체 데이터 분폴르 **여러 확률 분포의 혼합**으로 모델링
    - 데이터의 **생성 메커니즘**까지 모델링 가능

- 기본 가정
    - **관측된 데이터**: 특정 가우시안 확률 분포에 의해 생성됨
    - **전체 데이터셋**: 여러 개의 다변량 가우시안 분포가 섞여있음
    - **개별 데이터**: 우도에 따라 K개의 가우시안 분포 중 하나에 속함

- GMM 진행 과정
    1. 주어진 전체 데이터셋의 분포를 확인한다
    2. 전체 데이터 셋은 **서로 다른 정규 분포 형태의 확률 분포 곡선**으로 구성되어 있다고 가정한다.
    3. 전체 데이터셋을 구성하는 **여러 개의 정규분포 곡선 추출**, 개별 데이터가 이 중 어떤 정규분포에 속하는지 결정한다. 이때 **각각의 분포**가 **하나의 군집**이 된다.

- 장점: 
    - K-means보다 유연하게 다양한 데이터 세트에 잘 적용됨
    - K-means가 잘 작동하지 않는 타원형 분포나, 중첩된 군집 구조에서도 좋은 성능을 낼 수 있다.

- 단점
    - 군집화를 위한 수행 시간이 오래 걸림.
    - 가우시안 분포에 맞지 않는 데이터일 경우 계산 복잡도가 높아지고 성능이 저하될 수 있다.

- 군집화 평가 방법 (실루엣 계수, 던 인덱스)
1. 외부 평가
- 데이터에 **정답 레이블**이 존재하는 경우 사용
-군집화 결과와 실제 레이블(정답)을 비교하여 얼마나 유사한지 평가
- 대표적인 지표로 ARI(Adjusted Rand Index)가 있음
	- 조정된 랜드 지수(ARI)는 군집화 결과와 실제 레이블 간의 **일치도**를 측정하는 대표적인 지표
	- ARI는 -1부터 1 사이의 값을 가지며, 1에 가까울수록 일치, 0에 가까울수록 무작위에 유사함을 의미한다.
2. 내부 평가
- **정답 레이블이 없는 경우**에 사용
- 군집 내부 응집도와 군집 간 분리도를 기반으로 군집의 품질을 정량적으로 측정
	<br>->데이터 자체의 구조를 기반으로 군집 품질을 판단
- 대표적인 범용적 지표: **실루엣 계수(Silhouette Coefficient)**, **던 지수(Dunn Index)**

#### 실루엣 계수

- 개념: 클러스터 내 샘플들이 얼마나 조밀하게 모여 있는지를 측정하는 군집 품질 평가 방법. 

- 계산: 클러스터 응집력 $a^{(i)}$ 와 클러스터 분리도 $b^{(i)}$를 계산하고, 두 값의 차이를 둘 중 큰 값으로 나누어 실루엣 $s^{(i)}$을 계산한다.

- 해석: 
    - 실루엣 계수는 -1과 1 사이 값을 가지며, 1에 가까울 수록 분리도가 응집도 보다 큰 것을 의미
        <br>-> 샘플이 자신의 클러스터에 잘 속해있고, 다른 클러스터와는 멀리 떨어져 있으므로 **군집화가 잘되었다고 평가**

    - 0인 경우 분리도와 응집도가 같으므로 샘플이 클러스터의 경계에 위치하여 군집 간 거리가 모호하다고 평가

    - -1에 가까울 경우 분리도가 응집도보다 작다는 것을 의미<br>
        -> 군집이 섞여 있는 상태로, **군집화가 잘 되지 않았다고 평가**

##### Dunn Index

- 개념: 클러스터 간 **최소 거리(분리도)**와 클러스터 내 **최대 거리(응집도)**의 **비율**을 계산해 클러스터링의 품질을 평가하는 지표
- 평가: **값이 클수록 좋은 군집화 결과**를 의미
- 한계: 군집 수가 많아질수록 계산 비용이 증가
    - 극단적인 값(가장 멀리 떨어진 포인트, 가장 가까운 군집 등)에 큰 영향을 받음
    - 이상치에 민감하게 반응해 수치가 급격하게 감소할 수 있다
