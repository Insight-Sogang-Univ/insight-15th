# 1. 연관 분석

- 연관 분석의 정의
- 개념: 상품이나 서비스를 구매한느 등 일련의 거래나 사건 안에 존재하는 **항목 간의 일정한 <u>연관 규칙</u>을 발견**하는 분석
- 특징: 정답값이 별도로 존재하지 않기에 **비지도 학습**에 해당함.

### 1.1. 연관 규칙

- 형태: if-then의 형식을 띄고, A->B으로 표현됨
	- 조건절: '만일~라면'에 해당하는 부분
	- 결과절: 그 뒷부분에 해당하는 내용
	- 아이템 집합: 조건절 또는 결과절을 구성하는 아이템들의 집합

- 주의: 연관 규칙으로 상관관계를 파악할 뿐이지 **인과관계**는 알 수 없다.

### 1.2. 연관 규칙의 조건

- 일정 이상의 지지도 : 두 품목(품목 A와 B)이 함께 구매한 경우의 수가 일정 수준 이상이어야 한다. 
- 일정 이상의 신뢰도: 품목 A를 포함하는 거래 중 품목 B를 구입하는 경우의 수가 일정 수준 이상이어야 한다.

- 지지도: 전체 거래항목 중 **품목 A와 품목 B가 동시에 포함**하는 거래의 비율. 
    - 지지도 = P(A \cap B)
    <br> 지지도가 낮다면?-> 우연일 확률 존재, 실행 가능한 인사이트 도출 어려움.

- 신뢰도: **품목 A를 포함**하는 거래 수 중 **품목 A와 품목 B가 동시에 포함**하는 거래의 비율. 
    - 신뢰도 = \frac{P(A \cap B)}{P(A)} = \frac{지지도}{P(A)}
    - 신뢰도가 낮다면?-> 의미없는 연관 규칙. 추천 실패로 이어짐

# 2. 연관 분석의 평가 측도

- 향상도: 품목 A가 주어지지 않았을 때의 품목 B의 확률 대비, 품목 A가 주어졌을 때의 품목 B의 확률의 증가 비율
<br>->A와 B가 우연히 함께 등장할 확률과 실제 함께 등장할 확률을 비교! 
- 향상도 = $$\frac{P(A \cap B)}{P(A)P(B)} = \frac{신뢰도}{P(B)}$$

##### 2.1 독립

- 동시 확률=(A가 일어날 확률) * (A가 일어났을 때 B가 일어날 확률) <br>
=> $$P(A \cap B) = P(A) \times P(B|A)$$<br>
인데 이때 사건 A,B가 독립이라면 $$P(A \cap B) = P(A) \times P(B)$$가 됨
<br>따라서 향상도는 $$\dfrac{P(A∩B)}{P(A)⋅P(B)} = \dfrac{P(A)⋅P(B)}{P(A)⋅P(B)}=1$$

- 독립이 아니라면?
    - 실제로 관측한 $P(A∩B)$에 따라 향상도는 1보다 크거나, 1보다 작음

##### 2.2 실제 동시 발생 확률
- 우리가 실제로 관측한 - **$P(A∩B)$**가,
    
    **A와 B가 함께 등장할 확률의 기댓값** $P(A)\cdot P(B)$**보다** 
    
    - **크면 → 양의 연관성** (A 사면 B도 잘 삼)
    - **작으면 → 음의 연관성** (A 산 사람은 B 잘 안 삼)

##### 2.3 비율로 비교

- 향상도>1 : 우연적 기회보다 높은 확률 (두 품목이 서로 **양의 상관관계**)
- 향상도=1 : 독립
- 향상도<1 : 우연적 기회보다 낮은 확률 (두 품목이 서로 **음의 상관관계**)

		규칙의 효용성은 지지도, 신뢰도, 향상도 세 가지를 모두 반영해 평가

# 3. 연관 분석의 알고리즘

### 3.1 Apriori 알고리즘

- 등장배경: 연관 규칙을 찾기 위해 가능한 모든 조합을 시도하는 것은 매우 비효율적이다.

-> 상위 조합에서부터 차례로 스캔하면서 **특정 조합이 자주 발생하지 않는다면** 이의 결과물로 탄생한 후속 조합들까지 모두 후보에서 **배제**하는 방식의 알고리즘

- 원리: 집합 {A,B}의 지지도가 사용자가 정한 **최소 지지도를 넘지 못하면,** {A,B}를 포함한 {A,B,C}, {A,B,D} 등의 규칙들도 **모두 한번에 제거**된다

- 장점: 하나의 조합만 검사하고도 판단이 가능하므로 시간과 연산량을 효과적으로 절약. 
- 프로세스: 빈발 항목 집합(<u>일정한 **최소 지지도를 넘는** 항목들의 조합</u>)을 찾고, 단계적으로 조합을 생성하고 최소 지지도를 바탕으로 필터링.
<br> ex. 맥주, 치킨, 햄버거, 라면의 각 지지도를 계산하고 최소 지지도 (3/7)를 넘는 아이템만 다음 단계로 이동. 

<br>(맥주, 치킨), (맥주, 햄버거) 등 조합을 묶은 뒤 지지도 계산, 최소 지지도를 넘는 조합만 다음 단계로 진출

<br>최소 지지도를 넘는 집합이 없을때까지 반복

<br>최종 빈발 항목 집합을 선정하고 이를 바탕으로 **연관 규칙** 생성

- 단점: 여전히 데이터가 커질수록 속도가 느려짐

### 3.2 FP Growth 알고리즘

- 등장배경: Apriori 알고리즘은 여전히 데이터가 커질수록 속도가 느려진다는 단점을 보완하기 위해 등장.

- FP-Growth 알고리즘 소개: 
	- 빈발 품목 집합을 효율적으로 찾아낸다. 
	- FP-Tree 라는 구조를 이용하여 Apriori를 효과적으로 구현한다.
	- Tree, Array, Linked-List를 합쳐놓은 구조

- 특징: 모든 거래를 확인해 각 아이템마다 지지도를 계산하고, 최소 지지도 이상의 아이템만 선택
- 프로세스: 
	1. 모든 아이템을 빈도순으로 정렬 ex. 데이터 100건 중 우유가 80번, 빵이 50번 나왔다면 빈도는 우유>빵
	2. 맨 처음에 0 node를 만든 후, itemset을 tree에 하나씩 더한다. 이때 빈도가 높은 item이 0에 가까운 node(**=트리의 상단에 위치**)가 된다.
		ex. {우유, 빵) 이면 우유 node에 0+1->빵 node에 0+1. 그 다음 {우유, 빵, 시리얼}이면 0+1+1->0+1+1->0+1
	3. 빈발 항목 집합 추출
	4. 연관 규칙 도출

- 장점: 데이터를 두 번만 스캔하는 Tree 구조이기 때문에 Apriori보다 훨씬 빠르다. 
        <br>후보 Itemset을 생성할 필요 없이, Tree만 구성하면 끝이다.

- 단점: 대용량 데이터셋에서 메모리를 효율적으로 사용하지 않는다.
        <br>Aprior보다 설계하기 어렵고, 지지도 계산은 FP-Tree가 만들어져야 가능하다.

# 1. 경영 기초

## 1. 문제 정의와 문제 구조화

#### 1.1 문제 

- 현재의 상태와 '바라는 상태' 사이의 격차
	- 핵심: 경영자가 '의사결정'을 내려야 하는 구체적인 질문으로 치환되었을 때 비로소 '문제'가 됨.

- SWART 원칙: 
	- **S**pecific: 무엇을 어디까지 풀지 명확하게 정의
	- **M**easurable: 성공 여부를 숫자로 판단 가능
	- **A**ction-oriented: 행동을 이끌어내고 성과에 집중
	- **R**elevant: 조직의 목표와 관련이 있어야 함
	- **T**ime-bound: 마감 기한과 의사결정 시점이 존재함

#### 1.2 문제 구조화
- MECE 기법: 중복없이(Mutually Exclusive), 누락 없이(Collectively Exhaustive) 대상을 분류하거나 문제를 구조화하는 사고방식.
- 좋은 MECE를 하기 위한 규정: 

	1. 논리적 일관성: 한 층위에서는 **하나의 기준**으로만 나눠야 함

	2. 통찰력:

	3. 상호 배타성

## 2. 지표

- 북극성 지표
	- 기업이 가장 중요하게 생각하는 지표

	- 기업이 **고객에게 제공하는 핵심 가치**를 반영하여, 장기적인 성장과 성공을 측정하는 지표

- L1/L2 Metrics
	- L1: 비즈니스의 최종 목적이나 가장 핵심이 되는 성과 지표(전략 관점)

	- L2: L1 핵심 지표의 성장에 기여하는 구체적이고 실행 가능한 지표(전술 관점)

- AARRR(해적 지표)
	- 기업이 성장을 위해 사용자 행동을 5단계(획득, 활성화, 리텐션, 수익, 추천)로 분석하여 수치화하는 해적지표 프레임워크
    
	- 프로세스
    
		- Acquisition (획득/유입): 사용자가 어떤 경로로 우리 서비스를 방문하는가? (예: 앱 설치 수, CPC 광고 효율)
        
		- Activation (활성화/사용): 사용자가 처음 방문하여 긍정적인 첫 경험을 했는가? (예: 가입 전환율, 앱 내 주요 기능 사용)
        
		- Retention (리텐션/재방문): 사용자가 서비스를 다시 방문하는가? (예: 잔존율, 재방문 주기)
        
		- Revenue (수익화/매출): 사용자가 최종적으로 결제를 하는가? (예: 구매 전환율, 평균 주문 가치(AOV))
        
		- Referral (추천/전파): 사용자가 자발적으로 서비스를 추천하는가? (예: 추천인 코드 사용, 공유율) 

## 3. 지표 분석 방법

- 코호트 분석

	- 코호트(Cohort): 동질 집단. **정의된 시간 범위** 내에서 **공통된 특성이나 경험을 공유**하는 사용자 집단을 의미.

	- <u>'특정 기간에 특정 특성을 공유하는 사용자 집단'</u>이 코호트이며, 이를 분석하는 것이 코호트 분석

	- 분석 전에 데이터 세트의 데이터를 관련 그룹으로 나누는 일종의 행동 분석

	- **시간(날짜)**에 기반하는 사용자 세그먼트를 분류한다고 이해하기.

	- 주로 **사용자 유지**와 **이탈**을 체크하는 데 많이 사용
    
- 코호트 해석 방법

	- 세로축 방향: 코호트별 비교. ex. 신규 유저 가입자는 점점 증가하는 추세

	- 대각선 방향: 시점별 비교. ex. 3월 부터 1월까지 보았을 때, 전반적인 고객들의 구매율이 점점 떨어짐

	- 가로축 방향: 시간의 경과에 따른 변화 관찰. ex. 첫 방문(가입) 이후 시간이 경과할 수록 앱 활성률은 떨어짐

- 퍼널 분석
    - 개념: 사용자가 서비스에 유입되어 최종 목표(구매, 가입 등)에 **도달하기까지의 과정을 단계별로 시각화**하여, **어느 단계에서 가장 많이 이탈하는지 파악**하고 개선하는 데이터 분석 방법
