> ## 자료의 척도
> ### 자료: 수집, 분석, 요약되는 사실과 수치
> ### 양적자료와 질적자료가 있고, 양적자료는 구간과 비율척도로 나눠지고, 질적자료는 명목과 서열척도로 나눠짐

> - #### 양적 자료
>> ##### 1. 구간척도 (Ex. 온도, IQ, 시간)
>> : 값들 간 순서 존재, 간격 일정, 덧셈이나 평균 연산 가능\
>> 단, 곱셈과 나눗셈은 의미X
>>##### 2. 비율척도 (Ex. 거리, 높이, 무게, 시간)
>> : 절대영점과 임의의 측정단위로 구성, 동간성이 유지.\
 **0값은 그 변수에 대해 아무것도 존재하지 않는 상태**\
(Ex. 가격이 0이면, 비용이 전혀 안 들거나 공짜라는 말)
> - #### 질적자료
>> ##### 1. 명목척도 (Ex. 펀드 유형, 성별)
>>: 레이블이나 이름. 요소의 속성을 식별하는 척도
>> ##### 2. 서열척도 (Ex. 만족도, 직급체계)
>>: 분류 속성 + 항목 간 순서나 순위를 나타냄

> ## 기술통계
> #### 데이터를 있는 그대로 설명하고 이해하는 데 중점을 둔 통계 방법
- 
  ### 기술통계의 3요소
  > 1. ### 중심화 경향
  - 평균(mean): 총합을 사용하기에, 이상치(outlier)에 민감할 수 있음\
    `series.mean()`
  - 중앙값(median): 데이터의 한 가운데에 위치하는 값. 평균에 비해 이상치에 덜 민감.\
    `series.median()`
  - 최빈값(mode): 데이터에서 가장 자주 나타나는 값. 이산형 데이터에서 주로 쓰임\
    `series.mode()`
  > 2. ### 자료의 산포
  - 범위(range)\
  : 데이터 최대값과 최소값의 차이. 이상치에 매우 민감\
    `series.max() - series.min()`
  - 분산(variance)\
  : 분산값이 클수록 많이 퍼져있다는 것\
    `series.var()`
  - 표준편차(standard deviation)\
  : 분산에 제곱근을 취한 값(단위일치)\
    `series.std()`
  - 사분위수(quartile)\
  : 데이터의 표본을 네 개의 동일한 값으로 나눈 값
  (Q1: 하위 25퍼센트, Q2: 하위 50퍼센트, Q3: 하위 75퍼센트)\
    `series.quantile(0.25)`  # 25th percentile\
    `series.quantile(0.5)`   # 50th percentile (중앙값)\
    `series.quantile(0.75)`  # 75th percentile
    - numpy로 실행하는법\
    `np.percentile(series, 75)`
  - 사분위간 범위(IQR)\
  : Q3에서 Q1 값을 제외한 범위, 즉 데이터의 중앙 50%가 퍼져있는 범위 .이상치 탐지에 유용
    - 아래쪽 이상치: Q1 − 1.5×IQR 보다 작은 값
    - 위쪽 이상치: Q3 + 1.5×IQR 보다 큰 값
  - 변동계수 (coefficient of variation): 상대적인 변동성을 나타내는 통계지표
    - 변동계수 = (표준편차/평균) * 100
  > 3. ### 자료의 정리 및 시각화
  - 히스토그램, 박스플롯, 파이차트 등

### 확률분포
> - #### 확률변수
random한 세계에서 어떤 특정 사건이 발생할 때 나타날 수 있는 모든 경우의 수를 특정 값으로 대응시키는 함수
> - #### 확률분포(probability distribution)
확률변수가 특정한 값을 가질 확률을 나타내는 함수
> - #### 이산확률분포
확률 변수가 **이산적인(discrete)** 값을 가지는 경우에 해당\
이산적이라는 것은 유한하거나 셀 수 있다는 뜻이며, 특정 값을 선택할 수 있음
> - #### 연속확률분포
확률 변수가 **연속적인(continuous)** 값을 가지는 경우에 해당\
연속적인 값은 무한한 범위를 가지며, 특정 값을 선택하는 것이 불가능


### 대표적인 분포 모형
> #### 정규분포 (Normal Distribution)
> - 모집단의 분포가 정규분포를 따를 때 표본평균의 분포도 정규분포를 따름
> - 중심극한정리에 따르면, 표본의 크기가 커질수록 표본평균의 분포는 모집단의 분포와 상관없이 정규분포에 가까워짐
>>  **형태**\
종(bell) 모양\
평균을 기준으로 대칭\
표준편차가 클수록 곡선이 평평해짐 (데이터가 넓게 퍼짐)\
\
**매개변수**\
평균(μ), 표준편차(σ)\
\
**특징**\
가장 잘 알려진 (연속)확률분포\
많은 자연 현상과 사회 현상이 정규분포를 따르는 것으로 알려져 있음

>> #### Z분포 (표준정규분포)
>>>  평균이 0이고 표준편차가 1이 되도록 맞춘 표준적인 정규분포

# 표본분포


>#### T분포 (T-Distribution)
>>  **형태**\
종(bell) 모양\
평균을 기준으로 대칭\
자유도가 커짐에 따라 표준정규분포 N(0,1)에 수렴\
\
**매개변수**\
자유도(df)\
\
**특징**\
표본으로 모평균(혹은 모평균 간의 차이)을 추정할 때 사용되는 분포\
표본 크기가 작을 때 정규분포보다 더 정확한 결과를 제공함
>>> ##### 표준정규분포 vs T분포
>>> - 표준정규분포는 평균이 0이고 표준편차가 1인 정규분포로, 모집단의 평균과 분산을 정확히 알고 있을 때 사용
>>> - 이는 주어진 모집단에서 직접 계산되므로, 불확실성이 적음
>>> - t분포는 모집단의 분산을 모를 때, 표본의 분산을 사용해 모분산을 추정하여 만들어지기 때문에 표본의 크기가 작을수록 이 추정은 불확실성이 커지고, 이 불확실성 때문에 t분포는 표준정규분포보다 중심으로부터 더 퍼져 있음




> #### 카이제곱분포 (Chi-Square Distribution)
>>  **형태**\
0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포\
자유도가 증가함에 따라 중심이 오른쪽으로 이동하고 모양이 더 대칭적으로 변화하며, 정규분포에 가까워짐\
\
**매개변수**\
자유도(df)\
\
**특징**\
독립적인 표준 정규 분포를 따르는 변수들의 제곱 합\
즉, 표본분산의 분포와 관련된 지표\
주로 독립성 검정, 적합도 검정 등에 사용됨





> #### F분포 (F-Distribution)
>>  **형태**\
0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포\
자유도가 증가함에 따라 중심이 오른쪽으로 이동하고 모양이 더 대칭적으로 변화하며, 정규분포에 가까워짐\
\
**매개변수**\
자유도(df1, df2)\
\
**특징**\
두 독립적인 카이 제곱 분포의 비율\
두 집단 간의 분산 차이를 비교하는 데 사용됨

> #### 분포모형 활용
- 가설검정을 위한 검정통계량 산출에 사용
- 표본추출의 과정에서 인위적인 조작이 있었는지 확인 가능

## 추론통계
: 수집한 데이터를 기반으로 미래의, 혹은 조사하지 못한 전체 정보를 추론하고, 결론을 도출하는 방법을 다루는 분야
### 추정
: 모집단을 대표할 수 있는 일부를 표본으로 추출하여 분석한 통계량을 이용해서 모집단의 모수를 예측하는 과정
> #####모수
- Parameter. 모집단의 특성에 관계되는 평균, 비율 또는 표준편차 등 관심의 대상이 되는 미지의 참값
> #### 점추정
- 추정하고자 하는 모수를 하나의 수치로 추정
- 어떤 모수를 표본에 기초하여 계산된 하나의 값으로 추정

> #### 구간추정
- 추정하고자 하는 모수가 존재하리라 예상하는 구간을 정하여 추정
- 모수의 참 값이 포함되리라고 기대하는 추정치를 일정한 범위와 신뢰도를 가지고 나타내는 방법

### 가설검정 과정
> #### 가설검정
- 모집단의 실제 값이 얼마가 된다는 주장과 관련해, 표본의 정보를 사용해서 가설의 합당성 여부를 판정하는 과정
- 순서: **가설 설정 - 유의 수준 설정 - 검정통계량 산출 - 가설 기각/채택 판단**
> #### 가설 설정
- 귀무가설 H0와 대립가설 H1를 설정
- 귀무가설: 주로 기각하고자 하는 명제
- 대립가설: 내가 주장하고자 하는 명제
> #### 유의 수준 설정 
- 유의수준: 연구자가 결과를 판단할 때 허용할 수 있는 오류의 최대치로, 주로 0.05, 즉 5%로 설정
- 유의 수준을 **0.05**로 설정한다는 것은, "귀무가설이 실제로 참인데도 불구하고, **5%의 확률로 귀무가설을 잘못 기각할 수 있다**"는 것을 의미
- 1종 오류(𝛼) : 귀무가설이 실제로 참이었는데 기각할 확률\
    2종 오류(𝛽) : 귀무가설이 실제로 거짓인데 기각하지 못할 확률
- 가설검정은 1종 오류를 더 치명적인 것으로 파악
- 따라서 유의수준은 1종 오류의 확률을 의미
> #### 검정통계량 산출
- 검정통계량: 가설 검정의 결과를 판단하는 데 사용되는 값
- 데이터의 분포나 특징에 따라 **T값, F값** 등 어떤 통계량을 사용할지 선택
>> Z값 (Z-Score)
>>> - **정의**
>>>     - Z값 또는 Z-점수는 표준 정규 분포(평균이 0이고 표준편차가 1인 분포)를 사용하여 얻은 값
>>>     - 특정 데이터가 평균으로부터 얼마나 떨어져 있는지를 표준편차 단위로 나타냄
>>> - **계산**
>>>     - Z=(X−μ)/σ (X는 관측값, μ는 평균, σ는 표준편차)
>>> - **사용**
>>>     - 큰 표본 크기(일반적으로 30 이상)에 대한 가설 검정에 주로 사용
>>>     - 모집단의 평균에 대한 가설을 검정할 때 사용

>> T값 (T-Score)
>>> - **정의**
>>>     - T값은 T분포를 사용하여 얻은 값으로, 작은 표본 크기에 대한 가설 검정(T검정)에 주로 사용
>>>     - t분포는 정규분포와 유사하지만, 꼬리 부분이 더 두꺼워 작은 표본에서 발생할 수 있는 불확실성을 보다 잘 반영
>>> - **사용**
>>>     - 주로 모집단의 평균에 대한 가설을 검정하거나, 두 표본 평균 간의 차이를 검정할 때 사용


>> F값 (F-Score)
>>> - **정의**
>>>     - F값은 F분포를 사용하여 얻은 값으로, 주로 두 표본 분산 사이의 차이를 비교하는 데 사용
>>>     - F분포는 두 개의 독립적인 카이제곱 분포된 변수들의 비율로 정의
>>> - **사용**
>>>     - 주로 분산 분석(ANOVA)에서 두 개 이상의 그룹 간 평균의 차이를 검정하는 데 사용
