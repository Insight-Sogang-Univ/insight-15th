# 머신러닝 (Machine Learning)
## 머신러닝의 세 종류
> 지도학습
- 정답(label)이 있는 데이터 사용
- 기계의 예측값이 데이터의 정답과 같아지도록 학습
- ex. 회귀, 분류

> 비지도 학습 
- 정답(label)이 있는 데이터 사용
- 데이터의 패턴 혹은 유사성을 학습시키는 것
- ex. 군집화

> 강화 학습
- 보상
- 시행착오를 반복하여 보상을 최대화 하는 행동 학습
- ex. 알파고, 자율 주행

## 모델링
> 모델이란
- 학습의 결과물
- 학습을 마친 인공지능 그 자체
- 패턴이 저장된 상태
- 새로운 데이터를 넣으면 답을 주는 예측기 역할

> 모델링이란
- 예측기를 만들어가는 전체적인 과정
- 주어진 데이터 속에서 스스로 패턴을 찾아 최적의 공식을 완성할 때 까지 학습하는 과정

> 모델링의 과정
1. 설계 : 데이터 특성에 맞춰 어떤 알고리즘으로 학습할지 결정
2. 훈련 : 데이터를 계속 보여주며 정답을 맞히도록 수정
3. 검증 : 훈련시킨 모델이 새로운 데이터에서도 정답을 잘 맞히는지 테스트

## 지도 학습 
> 지도 학습이란
- 정답 레이블을 함께 주고 학습
- 예측값(prediction)과 정답(label)이 최대한 같아지는 방향으로 학습

> 지도 학습 과정
- 지도 학습을 위해서는 반드시 데이터가 labeled data의 형태여야 함
>> 1. labeled data ⇒ 독립변수(feature)와 종속변수
>> - 독립변수: 머신러닝에서 예측에 사용되는 입력 변수
>> - 종속변수: 머신러닝에서 예측하고자 하는 대상 변수
>> 2. labeled data 를 training set과 test set으로 나누기
>> - train set: 모델을 학습(훈련)시킬 때 쓰이는 데이터
>> - test set: 학습된 모델을 검증하기 위해 사용하는 데이터
>> 3. training set을 통해 모델 학습
>> - 독립변수(feature)를 통해 추정한 예측값(prediction)이 주어진 종속변수(target/label)와 같아지도록(가까워지도록) 모델을 훈련

> 회귀와 분류
- 회귀(Regression): 주어진 데이터를 기반으로 정답를 잘 맞히는 함수
- 분류(Classification) : 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별하는 것



# 회귀
- 데이터에서 패턴을 찾아 미래 값을 예측하는 것
- 종속변수가 연속적일 때 사용(*불연속적일 땐 분류)

## 선형회귀
> 선형회귀란


- 데이터를 가장 잘 설명하는 회귀선(ex. 직선 **y=wx+b**)을 찾는 과정
- 회귀선은 현재 데이터를 요약하는 동시에 미래 값을 예측
- 단순선형회귀 : x가 한 개
- 다중선형회귀 : x가 여러 개

> 가중치와 편향의 이해
- 선형 회귀의 기본 식인 **y= wx+b** 를 기준으로 했을 때
1. 가중치 w (Weight)
    - 독립변수 $x$가 종속변수 $y$에 주는 영향력의 크기
    - 기울기
    - 회귀계수
2. 편향 b (Bias)
    - 독립변수 $x$가 0일 때 나오는 기본값
    - y절편
    - 모델이 데이터에 더 유연하게 최적화될 수 있도록 위아래로 움직여주는 역할
> 단순선형회귀
- **하나의 독립변수($x$)** 만 존재하는 선형 회귀
- 데이터를 **직선형태**로 표현
- 데이터를 가장 잘 설명하는 **y= wx+b**을 찾는 과정
- 잔차를 줄여나가며 직선을 결정짓는 적절한 회귀계수(w)와 편향(b)을 찾음
> 잔차(Residual)
- 실제값과 예측값 간 차이
- 회귀선을 잘 찾는 과정 = 잔차를 줄여나가는 과정

## 최적의 w와 b를 찾는 방법
> 1. 최소제곱법(Least Squares Method)
- 잔차의 제곱의 합이 최소가 되는 지점으로 최적의 회귀선을 구하는 방식
- 
- 잔차를 곱하는 이유
    1. 양수-음수의 상쇄 방지: 제곱하지 않으면 더하는 과정에서 상쇄되기 때문
    2. 큰 잔차에서 더 큰 벌을 줘서 예측값을 강하게 조정 할 수 있기 때문
    3. 경사하강법 등의 알고리즘에서 미분 가능한 형태가 되기 위해서
- **"회귀선과 실제 데이터와의 차이를 최소화하기 위한 시도"**

> 2. 경사하강법(Gradient Descent)
- 최소제곱법을 통해 구한 함수가 복잡하여 최솟값을 구하기 어려울 때 사용
>> 목적함수, 비용함수, 손실함수

>> 1. 목적함수 (Objective Function)
>> - 우리가 달성하고 싶은 목표
>> - 최적화하려는 대상을 수학적으로 표현한 함수
>> - 문제 성격의 따라 최대화, 최소화 가능
>> 2. 손실함수 (Loss Function)
>> - 개별데이터의 대한 오차를 측정하는 함수
>> - 예측값과 실제값 y의 차이를 계산
>> 3. 비용함수 (Codt Function)
>> - 전체 데이터 셋에 대한 손실(Loss)함수를 합산하거나 평균을 내어 모델의 전체적인 성적을 평가
>> - 모델을 학습할 때 대부분은 비용함수를 최소화하는 것이 목표
>>     - 이때는 목적함수 = 비용함수로 보아도 무방
>>     - 목적함수가 원하는 최적화의 방향 = 비용함수

> 손실함수와 경사하강법


- 위 그림처럼 잔차를 제곱한 뒤 합하는 형태로 손실함수를 사용  

    - $t_i$ : $i$번째 데이터포인트 실제값(정답)
    - $y_i$ : $i$번째 예측값
    - $E$: Error
    - $n$: 데이터 샘플의 총 개수
- 이 식에 $y=Wx+b$를 대입해 최종적으로 아래와 같은 손실함수를 획득

- 우리의 목표는 $E$를 최소화하는 $W$, $b$를 찾는 것
---
W와 b에 대해 E가 변화하는 양상을 봐야 하니 x와 y축을 아래와 같이 구성하고, 최소 지점을 갖는 손실함수 그래프를 이런식으로 그릴 수 있음
1. 먼저 W로 x축을 구성한 경우
    
2. 임의의 W값에서 E값을 살펴본다면 지금 경우보다는 W값이 왼쪽으로, 즉 E값이 그래프에서 최소 지점으로 이동하도록 해야함
    
3. 미분 계수가 양수일 때, W는 왼쪽으로 (감소) 가야 하는 핵심 아이디어를 담아 W 값을 반복적으로 업데이트
4. 이렇게 W값을 업데이트 해나가고, 그러다 최솟값에 도달하면 미분값이 0이 되어서 “현재 값 - 0”으로 더이상 변화하지 않고 멈추게 됨

> 학습률(α: learning rate)
- 얼마큼 이동하는지_ 등산에서 보폭에 크기
- ‘얼만큼’ = “Learning Rate”
- 적당한 보폭이 가장 좋음 (어느 한 쪽으로 치우치지 않는것)
- Learning Rate가 너무 **작다면**, 모델이 수렴할 때 까지 너무 오랜 시간이 걸림
- Learning Rate가 너무 **크다면**, 모델은 수렴하지 못하고 발산 (=손실 값이 계속해서 커지는 것) 하게 됨

> Local Minima 문제
- 실제 이동량에서 작은 기울기에서는 조금씩, 큰 기울기에서는 많이 이동함
- 모든 비용함수가 하나의 최솟값을 가지는 매끈한 곡선 형태는 아니기에 Local Minima, Global Minima가 존재할 수 있음
- 작은 기울기에서는 업데이트가 조금씩 이루어지므로 Local Minima에서 나오기 어려움
    
    - 다행히 MSE, RMSE와 같은 비용함수에서는 함수가 볼록함수라는 것이 증명되었지만, 아닌 경우에는 주의!

> 해결법 - 모멘텀
- 기존 경사하강법은 이전 기울기는 다 잊어버리는 특성을 가짐
    - 현재 기울기만 기준으로 움직임
- 이전의 기울기와 방향을 고려하도록 관련성 부여
>> 모멘텀의 장점
- 기울기에 관성을 부여하여 작은 기울기를 쉽게 넘어가도록
- 지역 최소값을 탈출
- 모멘텀을 사용하지 않으면 
    - 아주 작은 언덕도 빠져나오지 못할 수 있으며 기울기가 매우 작은 구간을 빠져나오는데 아주 오랜 시간이 걸림

## 다중선형회귀

- $y$ = $W_1$ $x_1$ + $W_2$ $x_2$ + ... + $W_n$ $x_n$ + $b$ 
- 독립 변수가 **두 개 이상**인 경우 '다중선형회귀'
- 독립 변수 $x_i$가 2개인 경우 위와 같이 3차원 공간에 표현
- 회귀선은 평면
- 변수가 2개보다 많은 경우에도 유사하게 차원 확장
- 단순선형회귀와 마찬가지로 **최소제곱법/경사하강법**을 사용 가능
    - **잔차**는 관측치와 평면간의 차이

>> 다중공선성 (Multicollinearity)
- 독립 변수들 간의 **강한 상관관계**로 인해 회귀 모델이 **불안정**해지는 문제
- 심각하면 각 변수의 회귀 계수 추정이 부정확해지고 해석력도 떨어질 수 있으므로 중요한 체크 사항
- 다중공선성이 높은 경우 어떤 독립 변수가 종속 변수에 얼마나 영향을 미치는지 정확한 구분이 어려움

>> 다중공선성 확인 방법

>>> 상관계수(correlation coefficient, $r$)
- 변수 간의 통계적인 관계인 상관관계의 정도를 수치로 나타낸 것
- 상관관계
    - $-1 ≤ r ≤ 1$  사이의 값을 가진다.
    - $r>0$: 양의 상관관계, 한 변수가 증가하면 다른 변수도 증가하는 경향.
    - $r<0$: 음의 상관관계, 한 변수가 증가하면 다른 변수는 감소하는 경향.
    - $r=0$: 두 변수 사이에 직접적인 연관성이 없다.
- 히트맵 혹은 **corr()** 함수를 통해 상관계수 확인 가능
- `corr()` 함수  
    `import seaborn as sns`    
    `import matplotlib.pyplot as plt`  
    `sns.heatmap( df.corr(), annot = True)`  
    `plt.show()`  
    
- pairplot과 같은 시각화로 산점도를 찍어서도 확인
    

>>> VIF지수 (분산 팽창 인수)
- 회귀 모델의 결정계수 R-Square를 사용하여 계산
- VIF가 높으면 다중공선성이 존재한다고 판단해요.
    - **VIF = 1**: 해당 독립 변수는 다른 변수와 상관관계가 전혀 없음을 의미
    - **VIF < 5**: 일반적으로 다중공선성 문제가 없다고 간주
    - **VIF > 5**: 다중공선성의 징후가 있을 수 있으며, 주의를 요함
    - **VIF > 10**: 다중공선성이 심각하다고 간주되며, 이 변수를 포함한 회귀 모델은 불안정할 가능성이 큼. 이 경우 해당 변수를 제거하거나, 변수 간의 상관관계를 줄이기 위한 조치가 필요할 수 있음

>>> 다중공선성 대처 방법
1. 변수 제거
2. 변수 변환
    - 변수들을 더하거나 빼서 새로운 변수 생성
    - 독립변수를 더하거나 빼더라도 문제가 없는 경우
3. 규제선형모델 활용
- 릿지, 라쏘, 엘라스틱넷 등의 방법을 통해 모델의 복잡도를 줄일 수 있음
4. PCA(주성분 분석)
- 데이터의 차원을 축소하는 데에 사용되는 통계적 기법
- 고차원 데이터에서 중요한 정보를 최대한 보존하면서 데이터의 복잡성을 줄이는 것이 목표
    - 데이터의 차원이 높으면 분석과 시각화 모두 어려움
- **상관 있는 변수들을 묶어서 더 적은 수의 대표 축으로 바꾸는 것이 PCA의 핵심 아이디어**

## 규제선형모델
>> 규제선형모델이란
- 모델이 학습데이터에 과적합(overfitting) 되지 않도록 규제를 가하고자 등장한 모델
    - 선형회귀 모델에서는 특성에 곱해지는 계수(기울기)의 크기를 조정하는 것
- 선형모델에서는 회귀계수가 커지는 문제로 과적합 발생
- 다중회귀에서도 모델이 복잡해져 과적합 발생
- **비용 함수는 학습 데이터의 잔차 오류 값을 최소화하는 RSS 최소화 방법(최소제곱법)과, 과적합을 방지하기 위해 회귀 계수 값이 커지지 않도록 하는 방법이 서로 균형을 유지해야 함**
    - 오류는 최대한 줄이면서도, 특정 변수에 집착하지 않도록 균형을 맞춰 과적합을 방지

>> 규제선형모델의 종류
1. L2규제: W(회귀계수)의 제곱에 대해 페널티를 부여하는 방식, 릿지(Ridge)회귀
> - like 기둥에 따라 다른 힘으로 누르는 망치
> - 비율이 균등하지 않을 수는 있지만 전체적으로 작아짐  
  
> - 회귀계수($w_j$)의 제곱을 패널티로 주기 때문에, 큰 회귀계수일 수록 훨씬 강한 패널티
> - 모든 계수의 크기가 작아지지만, 규제를 통해 변수가 제거되지는 않음
> - **모든 변수를 유지하면서, 크기를 적절히 줄여 과적합 방지**

2. L1규제: W(회귀계수)의 절댓값에 대해 페널티를 부여하는 방식, 라쏘(Lasso) 회귀
> - like 모든 기둥을 같은 힘으로 누르는 망치
> - 작은 기둥들은 힘을 버티지 못하고 **완전히 사라질 수도 있음**
> 
> - 회귀계수($w_j$)의 절댓값을 패널티로 주기 때문에, 회귀계수의 크기와 상관없이 같은 패널티
> - 작은 계수가 0이 되기도 하기 때문에 변수가 완전히 사라질 수 있음
> - **비중이 낮은 변수(작은 계수)는 아예 사라지고, 중요한 변수만 남음**

3. L2규제 + L1규제: 엘라스틱넷(Elastic Net)회귀
> - 엘라스틱넷(Elastic Net)은 L1, L2 두 개의 망치를 동시에 사용
> - **L1 망치:** 작은 기둥을 아예 없애 불필요한 변수 제거
> - **L2 망치:** 남은 기둥을 균형 잡히게 눌러 크기 조절
> - 결과: 불필요한 변수는 제거하고, 남은 변수들은 적절한 크기로 유지
> 
> - L1의 변수 선택 기능(일부 계수 삭제)과 L2의 규제 기능(큰 계수를 줄임)을 동시에 활용하는 모델
> - L1은 상관관계가 높은 변수 중 일부만 선택하는 경향
> - L2를 함께 사용하면 상관관계가 높은 변수를 함께 고려하면서 규제 가능
> - **안정적인 예측(모든 변수가 균형 있게 줄어듦)과 희소성(변수 제거)을 동시에 달성**

## 모델평가방법
### 성능평가 지표
> 평균제곱오차(MSE)
- 예측값과 실제값 사이의 오차 제곱의 평균을 의미  

> 평균절대오차(MAE)
- 모델의 예측값과 실제값 사이의 절대값 오차의 평균을 의미  

> MSE와 MAE의 차이
- MSE는 큰 오차에 대해 민감하게 반응(제곱)_**이상치에 민감**
- MAE는 모든 오차를 동일하게 취급_**이상치 존재 경우에도 예측 성능 유지**

### 변수 유의성 평가
> T-검정
- **귀무가설 (**$H_0$**)**: 회귀계수(a)가 0이다 → 독립변수가 종속변수에 영향을 주지 않는다.
- **대립가설 (**$H_1$**)**: 회귀계수(a)가 0이 아니다 → 독립변수가 종속변수에 영향을 준다.
    1. t-값 계산 
    - 
    - $\hat{\beta}$ : 추정된 회귀계수
    - $SE(\hat{\beta})$ : 표준오차
    - t-분포를 따름.

    2. p-value 확인 및 판단
    - **p-value < 0.05** → 유의 수준 5%에서 $H_0$ 기각 → **해당 변수는 유의미하다.**
    - **p-value ≥ 0.05** → $H_0$ 기각 불가 → **해당 변수는 유의미하지 않다.**
- **회귀계수가 0이면 해당 독립변수는 의미 없음.**
- **p-value가 작을수록** 해당 변수는 종속 변수에 **더 큰 영향을 줄 확률이 높아짐.**
- **t-값이 크면** 귀무가설을 기각할 가능성이 높아짐.
