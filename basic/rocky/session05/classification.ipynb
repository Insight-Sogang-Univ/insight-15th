# 회귀와 분류의 차이

#### 회귀(Regression)
- 연속적인 숫자값을 예측
- 주어진 데이터(X)를 기반으로 정답(Y)를 잘 맞추는(fit)하는 함수(function)

- 위 그림에서 파란선을 찾는 문제 

#### 분류(Classification)
- 입력된 데이터를 주어진 항목들로 나눔
- 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별


### 회귀와 분류의 차이
- 회귀: 데이터가 **연속형 변수를 예측**하기 위해 사용될 때
- 분류: 데이터가 **범주형 변수를 예측**하기 위해 사용될 때
- -> 예측하고자 하는 값(=종속 변수)가 <span style="color:blue">연속형</span>이면 <span style="color:blue">회귀</span>, <span style="color:red">범주형</span>이면 <span style="color:red">분류</span>

#### 이진 분류와 다중 분류
> 이진 분류(Binary Classification)
- 예측하고자 하는 변수 어떤 기준에 대하여 참 또는 거짓의 값만을 가질 때
> 다중 분류(Multiclass Classification)
- 예측하고자 하는 변수가 가질 수 있는 값이 3개 이상일 때
> - 이진 분류와 다중 분류

## 1. 로지스틱 회귀(Logistic Regression)
- 이진 분류 문제를 푸는 대표적인 알고리즘
- 샘플이 특정 클래스에 속할 확률을 추정하여 기준치에 따라 분류하는 것이 목표
- 독립 변수의 선형 조합 (선형 회귀)에 로지스틱 함수를 적용하여 출력값을 0에서 1 사이로 변환해주는 것
- Ex) 이메일이 스팸인지 아닌지, 환자가 특정 질병을 가지고 있는지에 대한 여부
> #### 로지스틱 회귀의 필요성
- 이진 분류를 단순 선형 회귀로 풀려고 하면 문제 발생
> $y=wx+b$의 직선이 분류 문제에 적합하지 않은 이유
1. $x$와 $y$의 관계를 표현하기 위해서는 직선을 표현하는 함수가 아니라 **S자 형태**로 표현할 수 있는 함수가 필요
    - 시그모이드 함수는 입력값이 작거나 클 때는 천천히 변화하고, 중간값 근처에서는 급격히 변화
    - 이는 모델이 예측할 때 급격한 변화를 방지하고, 현실적인 확률 값을 제공
    - 특정 독립 변수의 변화가 크지 않은 경우, 시그모이드 함수는 확률을 급격히 바꾸지 않으며, 이는 노이즈에 민감하지 않은 안정적인 예측이 가능
2. 예측값은 0과 1사이의 값을 가져야 하는데, 위와 같이 직선으로 분류 작업을 수행하면 값이 음의 무한대부터 양의 무한대와 같이 범위를 벗어나는 값들도 가질 수 있음
##### 결론: 출력이 0과 1사이의 값을 가지면서 S자 형태로 그려지는 함수를 이용해야 함!
> #### 시그모이드 함수 (Sigmoid function)
- 시그모이드 함수 ≈ 로지스틱 함수:  출력이 0과 1사이의 값을 가지면서 S자 형태로 그려지는 함수

- 0과 1사이의 값으로 조정된 시그모이드 함수
- 출력이 0과 1 사이의 값을 가지면서 S자 형태로 그려지기 때문에, 이진 분류 작업에 적용 가능
- **시그모이드 함수 특징**
    - 입력값이 커지면 1에 수렴하고, 입력값이 작아지면 0에 수렴
    - 입력값($x$)이 $(-\infin, \infin)$일 때, 출력값($y$)은 0~1의 값을 가짐
- -> 출력값이 특정값 이상이면 1(True), 특정값 이하면 0(False)로 정해서 이진 분류 문제를 풀 때 사용!
> 시그모이드 함수의 가중치
- 그모이드 함수가 $H(x)=\frac{1}{1+e^{-(wx+b)}}=sigmoid(wx+b)=\sigma(wx+b)$일 때, -> 인공지능은 주어진 데이터에 적합한 가중치 $w$와 $b$를 계산
1. 가중치 $w$의 변화에 따른 그래프 변화(그래프의 기울기 변화)

- $w=0.5$ 
- w=1.0$ 
- $w=2.0$
2. 가중치 $b$의 변화에 따른 그래프 변화(그래프의 위치 변화)

- $b=0.5$ 
- $b=1.0$
- $b=1.5$

## 2. 의사결정나무(Decision Tree0)
- **조건에 따라 데이터를 분류**하며, 최종적으로 데이터가 순수한 label의 집합으로 구성될 때까지 **분류를 반복**

- 마치 **스무고개**처럼 반복
> #### 의사결정나무의 구성요소

- **Root Node**: 가장 상위에 위치한 노드
- **Parent Node**: 주어진 노드의 상위에 있는 노드
- **Child Node**: 하나의 노드로부터 분리된 2개 이상의 노드들
- **Binary Tree**(이진 트리): Tree 중에서 children이 최대 2개인 tree
- **Leaf Nodes**: 자식마디가 없는 노드, 모델에서 label에 해당
- **Edge**: Parent node와 child node를 연결하는 연결선
- **Height**: 특정 노드에서 가장 멀리 있는 leaf node까지의 경로에 있는 edge 개수
- **Depth:** Root node에서 특정 노드에 도달하기 위한 edge의 수
> #### CART(이진분할; Classification And Regression Tree) 알고리즘
- 가장 널리 사용되는 의사결정나무 알고리즘으로, 데이터셋을 **임계값**을 기준으로 **두 child**로 나누는 알고리즘
- 불순도(지니계수)가 낮아지는 방향으로 임계값을 나눠야 함
> ##### 불순도란?
- 불순도란, 분류하려는 데이터 집합에서 서로 다른 클래스(범주)가 섞여 있는 정도를 의미
- **CART 알고리즘**에서는 불순도를 확인하기 위해 **지니 지수**를 사용

> #### CART 알고리즘의 주요 단계
1. 임계값 설정

- 왼쪽과 오른쪽 그림은 기본적으로 같은 모델
- t1 .. t4의 임계값을 기준으로 하여 데이터를 그룹으로 나눔
2. 불순도 감소 알고리즘

- 의사결정나무를 만들어나갈 때 클래스를 정확하게 구분해줄 수 있는 분류기준을 찾는 것이 중요
- 즉,**이 데이터를 어떤 기준으로 분류했을때 동일한 객체들로만 잘 모아지게 할 수 있을까?**를 고려해서 분류 기준을 찾는것이 중요
- 이때, 불순도 알고리즘을 사용하여 현재 집단에 어느 정도 다른 객체들이 섞여있는지 확인하고 불순도가 낮은 쪽으로 가지를 형성

> #### 실제 학습 시 고려해야 할 것들: 모수 설정, 차이점 시각화, Prunning
1. Hyperparameter 설정
- `min_samples_split`: 분할되기 위해 노드가 가져야 하는 최소 샘플(데이터) 수
- `min_samples_leaf`: 리프 노드가 가지고 있어야 하는 최소 샘플 수
- `min_weight_fraction_leaf`: `min_samples_leaf`와 같지만 가중치가 부여된 전체 샘플 수에서의 비율
- `max_leaf_nodes`: 리프 노드의 최대 개수
- `max_features`: 각 노드에서 분할에 사용할 특성의 최대 수
2. 시각화

- 왼쪽 그림의 경우 규제가 없어서 과적합이 이루어진 반면, 오른쪽 그림은 min_samples_leaf 개수를 제어하여 적절히 분류되도록 함
3. Prunning(가지치기)
- 불필요한 노드 지우기

- 노드가 너무 많아지면 2번 왼쪽 그림처럼 과적합이 될 확률이 높음
- 이럴 때 Prunning을 통해 하부 트리를 제거하여 일반화 성능을 높일 수 있음
- pruning을 하게 되면 깊이가 줄어들고 결과의 개수가 줄어듦

## 3. SVM (Support Vector Machine)
- 클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘
- 데이터를 분리하는 초평면(Hyperplane) 중에서 데이터들과 가장 거리가 먼 초평면을 선택하여 분리하는 지도 학습 기반의 이진 선형 분류 모델


- 명확하게 분류할 수 있는 데이터 집단에서 뛰어난 성능을 보이며, 고차원 공간(다수의 feature)에서도 효과적으로 사용 가능
- 이러한 특성 때문에 다양한 머신러닝 응용분야에서 활용(ex. 패턴 인식, 이미지 분석, 바이오인포매틱스 등)
> #### SVM의 구성

- Support vector (서포트 벡터): 데이터 중에서 결정 경계와 가장 가까이에 있는 데이터들의 집합
- Decision Boundary (결정 경계): 데이터 분류의 기준이 되는 경계
- Margin (마진): 결정 경계에서 서포트 벡터까지의 거리(여유 공간)
- Hyperplane (초평면): $n$차원 공간의 ($n-1$)차원 평면
- Slack Variables (슬랙 변수/여유변수): 완벽한 분리가 불가능할 때 분류를 위해 허용된 오차에 대한 변수
> #### 최적의 선을 찾는 방법

- 이 중 가장 좋은 선(결정 경계)는 **F**
- C는 파란색에, E는 빨간색에 너무나 가까이 하고 있음
- 결정 경계는 데이터로부터 가장 멀리 떨어져 있는게 좋음
-  따라서 SVM은 **Margin (거리)가 가장 큰 경우를 선택**함으로써 최적의 선을 탐색
> #### SVM의 장단점
- 장점
    - 서포트 벡터만을 이용해서 결정 경계를 생성하므로 데이터가 적을 때 효과적
    - 새로운 데이터가 입력되면 전체 데이터 포인트와의 거리를 계산하지 않고 서포트 벡터와의 거리만 계산하면 되기 때문에 연산량 최소화
    - 정확성이 뛰어나며, 비선형 모델 분류 가능
    - 다른 모형보다 과대 적합의 가능성이 낮고, 노이즈의 영향이 적음
- 단점
    - 데이터 전처리 과정이 중요
    - 데이터가 많아질수록 최적화를 위한 평가 과정이 많아져서 속도가 느림

## 4. KNN (K-Nearest Neighbor)
- 데이터로부터 **거리가 가까운 k개의 다른 데이터 레이블을 참조하여 분류**하는 알고리즘
    - K개의 이웃 설정 -> 가까운 K개의 데이터를 기반으로 분류
- **비슷한 특성을 가진 데이터끼리 서로 가까이 있는다는 점**을 이용한 분류 알고리즘
> #### KNN 알고리즘 계산 순서
1. 데이터 준비
    - KNN은 미리 학습하는 과정이 없기 때문에, 데이터를 준비
    - 각 데이터는 특징 벡터와 레이블(어떤 class인지)로 구성
2. K 값 설정
    - 원하는 K 값을 설정합니다. K는 가장 가까운 이웃의 개수를 나타내며, 보통 홀수개로 설정
    - k를 짝수로 설정했을 때, 동점이 발생하는 문제를 방지하기 위해서 보통 홀수개로 설정
3. 거리 계산
    - 새로운 데이터(예측하려는 데이터)가 주어지면, 이 값과 **기존 모든 데이터 간의 거리를 계산**
        - 보통 유클리드 거리, 맨해튼 거리 등을 사용
4. 가장 가까운 K개의 이웃 선택
    - 계산된 거리 중에서 **가장 작은 거리 값을 가진 K개의 데이터를 선택**
    - “가장 가까운 이웃”인 데이터
5. 분류
    - K개의 이웃 중 가장 많이 등장하는 클래스가 예측 결과임
> #### 최적의 K 선택하는 법
- 일반적으로는 학습용 데이터 개수의 제곱근으로 설정
- K를 너무 크게 설정하면 주변에 있는 점과의 근접성이 떨어져서 분류가 잘 이뤄지지 않고 과소 적합
- K를 너무 작게 설정하면 이상값, 잡음 데이터와 이웃이 될 가능성이 있음
> #### KNN 특징 및 장단점
- **특징**: **어떠한 학습이 필요하지 않음**
    - 별도의 모델없이 데이터만을 이용해서,  새로운 데이터가 왔을 때 즉석에서 주변데이터들을 이용하여 classify
- 장점
    - 범주를 나눈 기준을 알지 못해도 데이터를 분류할 수 있음
    - 입력 데이터만 주어지면 바로 예측값을 구할 수 있어서 다른 분류 모형과는 달리 학습 과정이 필요 없음
    - 이해가 쉽고, 구현이 간단
    - 추가된 데이터의 처리가 용이
    - 훈련 데이터를 요약하거나 일반화된 모델로 만들지 않고, 원본 데이터 전체를 그대로 저장
- 단점
    - 다른 분류 모형과 달리 테스트 데이터의 개수에 따라 시간이 오래 걸림
    - 학습 데이터 모두를 거리 계산에 사용하기 때문에 학습 데이터의 양도 계산 시간에 영향을 미침
    - K의 값 결정이 어려움
    - 수치형 데이터가 아니면 유사도를 정의하기 어려움
    - 데이터 내 이상치가 존재하면 분류 성능에 큰 영향을 받음







# 분류 평가 지표

## 1. 혼동 행렬
- 분류 모델의 예측 결과를 **정확한 예측과 잘못된 예측으로 구분**하여 나타낸 표(행렬)

- **파란색 상자** : 예측을 올바르게 한 경우 (**예측과 실제 결과가 모두 일치**)
- **빨간색 상자** : 예측을 잘못한 경우 (**예측과 실제 결과가 불일치**)
- -> 예측 성공 여부(T/F) + 예측값(P/N)으로 구분하여 총 4종류의 경우의 수가 있음

> #### 혼동 행렬 구분
**TP, FP, FN, TN에서 T와 F(앞에 있는 값), P와 N(뒤에 있는 값)을 나눠서 생각**
- T_: 모델이 정답을 맞췄다.
- F_: 모델이 정답을 맞추지 못했다.
- _P: 모델이 참(양성)으로 예측했다
- _N: 모델이 거짓(음성)으로 예측했다.
---
- TP: 모델이 참이라고 예측해서 정답을 맞췄다.
- FP: 모델이 참이라고 예측해서 정답을 맞추지 못했다.
- FN: 모델이 거짓이라고 예측해서 정답을 맞추지 못했다. 
- TN: 모델이 거짓이라고 예측해서 정답을 맞췄다.
> #### 혼동 행렬을 이용한 분류 모델 평가 지표
- 정확도 (Accuracy): 모든 가능한 예측 중 참인 비율
- 정밀도 (precision): 참이라고 예측한 경우 실제 참의 비율
- 재현도 (recall): 실제로 참인 경우 중 참으로 예측하는 비율
- cf) 정확도, 정밀도, 재현도는 모두 0과 1사이의 값
##### 정확도(Accuaracy)
- 모델이 입력된 데이터에 대해 얼마나 정확하게 예측하는지

- 모든 가능한 예측 중 참인 비율(T로 시작) : **(TP+TN) / (TP+TN+FP+FN)**
- 단점 : 정답 레이블의 비율이 불균형하면 모델의 정확도를 신뢰할 수 없음
---
##### 정밀도(precision)

- 0과 1 사이의 값
- **TP/TP + FP**
- 관건은 **FP**
- 예측을 참이라고 한 경우 중 실제 참의 비율
- **거짓을 참으로 판단한 정도**를 알 수 있음
- 정밀도가 높을수록, **거짓을 참으로 예측한 정도가 낮음**
---
##### 재현도(recall)

- 0과 1 사이의 값
- **TP/TP + FN**
- 관건은 **FN**
- 실제 참이라고 한 경우 중 예측 참의 비율
- **참을 거짓으로 판단한 정도**를 알 수 있음
- 재현도가 높을수록, **참을 거짓으로 예측한 정도가 낮음**

> Precision과 Recall은 Trade-off 관계에 있음
- 분류를 할 때, 확률에 기반해서 Threshold가 넘으면 참, Threshold 미만이면 거짓으로 판단
- 따라서 이 경곗값(Threshold)를 조정하면 Precision과 Recall을 조정할 수 있음
    - 경곗값(Threshold)를 낮춤 → Positive 예측이 늘어남 → recall ⬆️, precision ⬇️
    - 경곗값(Threshold)를 높임 → Positive 예측이 감소함 → recall ⬇️, precision ⬆️
> #### 정밀도(recall) & 재현도(precision) 그래프
- 정밀도 & 재현도 그래프에서 두 value 값이 만나는 지점을 Threshold로 정하면, 예측 오류를 최소화 할 수 있음

- 각 점은 분류 모델이 해당 임계값에서 precision과 recall value를 어떻게 예측했는지 표시
- 모든 임계값에 대해 점을 그려서 모델의 예측 성능을 시각적으로 파악할 수 있음
- **두 value 값이 만나는 지점을 임계값으로 정할 때, 예측 오류를 최소화**할 수 있음
