# **머신러닝과 모델링**
## 머신러닝이란?
- 개념: 모델을 학습해 데이터에서 패턴을 발견하거나 새로운 데이터에 대해 예측하게 하는 기술
- 종류
    - 지도학습: 정답 라벨을 사용해 기계의 예측값이 데이터의 정답과 같아지도록 학습시키는 것 (회귀, 분류) -> **특정 카테고리로 나눔**
    - 비지도학습: 정답 라벨을 사용하지 않고 데이터의 패턴 혹은 유사성읋 학습시키는 것 (군집화) -> **카테고리는 없지만 유사한 것끼리 묶음**
    - 강화학습: 보상과 벌을 통해 상을 최대화하는 행동을 학습시키는 것
    > 군집화란?  
    데이터 속의 패턴 또는 데이터 간 유사도를 학습함으로써 비슷한 데이터를 큰 단위로 묶는 것  
---
## 모델링
- 모델 vs 모델링
    - 모델 = 학습의 결과물. 데이터를 학습하며 찾아낸 패턴이 저장된 인공지능 그 자체
    - 모델링 = 인공지능을 만들어가는 전체적인 과정. 기계가 오차를 최소화하는 공식을 완성할 때까지 학습하는 과정
- 모델링의 단계 $$ 설계 -> 훈련 -> 검증 $$  
    - 설계: 데이터를 어떤 알고리즘으로 학습시킬 것인가?
    - 훈련: 기계에게 데이터를 계속 제공하며 정답을 맞히도록 수정
    - 검증: 훈련시킨 모델을 새로운 데이터에 적용해 성능을 평가
---
## 지도학습
- 개념: 정답 레이블을 제공해서 학습을 시키는 방식. 예측값이 정답과 최대한 일치하도록 학습함
- 단계
    - labeled data 준비: **feature**(->x. input)와 **target**(->y. output)이 모두 존재하는 데이터 준비
    - training/test set 준비: labeled data를 훈련 데이터와 검증 데이터로 나눔. (test set은 학습시키지 않은 새로운 데이터)
    - 모델 학습: 훈련 데이터를 통해 모델 학습. 특성을 통해 추정한 예측값이 정답과 최대한 가까워지도록(**오차 최소화**)

**학습이 끝나면 모델은 데이터 사이의 통계적인 상관관계들을 **하나의 함수 형태**로 가지게 됨* 
- 회귀: 주어진 데이터를 기반으로 연속적인 숫자값을 갖는 정답(**연속형**)을 예측하는 것. regression
- 분류: 기존 데이터가 어떤 레이블에 속하는지 패턴을 찾아내어 새로운 데이터의 레이블(**범주형**)을 에측하는 것. classification
---


# **회귀**
- 개념: 데이터에서 패턴을 찾아내 미래 값을 예측하는 것. 종속변수(정답)가 연속형일 때 사용
- 종류: 선형회귀, 비선형회귀, 규제회귀, 다항회귀 등
---
## 선형회귀
- 개념: 데이터를 가장 잘 설명하는 직선 y = wx + b를 찾는 과정
> 회귀선의 역할: 현재 데이터 **요약** + 미래 값 **예측** 도구
- 회귀선 자세히 알아보기 $$ y = wx + b $$
    - 가중치 w: 회귀계수. 독립변수 x (->특성)가 종속변수 y (->정답)에 주는 **영향력의 크기**(=기울기)
    - 편향 b: 독립변수 x가 0일 때 나오는 기본값(=y절편). **모델이 데이터에 유연하게 최적화될 수 있도록 위아래로 움직여주는 역할**
- 단순선형회귀: 독립변수 x가 하나만 존재하는 선형회귀 (-> 회귀선이 직선). 잔차를 줄여나가며 직선을 결정짓는 회귀계수와 편향을 찾는 과정
> 만약 독립변수 x가 여러 개라면?
=> 다중선형회귀
- **잔차**: 실제값과 예측값 간의 차이
    > 오차랑 뭐가 다른가? 

    오차 = |실제값 - 실제 평균값| / 잔차 = |실제값 - 예측값|. 관용적으로 잔차를 오차라고 부르는 것뿐, 엄밀히 말하면 다르다
---
## 최적의 w와 b를 찾는 방법
- **최소제곱법**: 최소자승법. **${잔차}^2$의 합이 최소가 되는 지점**으로 최적의 회귀선을 구하는 방식 => 회귀선과 실제 데이터의 차이를 최소화하는 것이 목적
    > 잔차를 왜 제곱할까?

    - 양수-음수의 상쇄 방지
    - 큰 잔차에 더 큰 벌을 줘서 예측값을 강하게 조절
    - 경사하강법 등에서 미분가능한 상태가 되기 위해
- **경사하강법**: ${잔차}^2$의 합이 최소가 되게 하는 목적은 동일, but 최소제곱법을 통해 구한 함수가 복잡할 때 사용하는 방법
    - 목적함수, 비용함수, 손실함수  

    | **구분** | **정의** | **방향** |
    | --- | --- | --- |
    | 목적함수 | 최적화하려는 전체 대상(달성하고자 하는 목표) | 최대화/최소화 |
    | 손실함수 | **데이터 한 개**에 대한 오차 | 최소화 |
    | 비용함수 | **전체 데이터**에 대한 손실 합산 or 평균 오차 | 최소화 | 

    - 손실함수와 **경사하강법**: $$ E = \frac{1}{n} \sum_{i=1}^{n} [t_i - y_i]^2 $$ (E = error, n = 데이터 개수, $t_i$ = $i$번째 데이터포인트 정답, $y_i$ = $i$번째 예측값)  
    여기에 회귀식을 대입하면 $$ E(W, b) = \frac{1}{n} \sum_{i=1}^{n} [t_i - (Wx_i + b)]^2 $$ 이므로 E를 최소화하는 W,b를 찾는 게 목표  

        <div style="border:1px solid white; padding:10px; border-radius:5px;">

        1. 손실함수 그래프 그리기 (x축 = W, y축 = E): 아래로 볼록한 이차함수 형태

        2. 점을 움직이며 꼭짓점에 도달: 최소값은 이차함수의 꼭짓점에 위치하므로 현재 지점에서 미분계수가 양수이면 왼쪽으로, 미분계수가 음수이면 오른쪽으로 이동해야함 => 즉 미분계수와 이동방향이 반대가 될 떄 현재값(x좌표)이 작아지거나 커지면서 꼭짓점에 도달 가능  
        
        </div>
    - 학습률 α: 경사하강법에서 얼만큼씩 이동해서 최솟값을 찾아나갈 것인지를 나타내는 수치. 너무 작으면 모델이 수렴할 때까지의 시간이 매우 오래 걸리고, 너무 크면 모델이 발산(=손실값이 계속 커짐)하므로 하이퍼파라미터 실험을 통해 적당한 값으로 선택
    - **local minima 문제**: 모든 비용함수가 이차함수처럼 하나의 최소값을 갖는 것이 아니고, 특히 극소점에서는 이동량이 작기 때문에 여기서 빠져나오기 어려움
        - 해결법-**모멘텀**: 기존 경사하강법이 현재 기울기만 기준으로 움직인다는 한계를 극복하기 위해, **관성을 부여해 이전의 기울기와 방향을 고려하게 함**

        <div style="border:1px solid white; padding:10px; border-radius:5px;">

        

        - **빨간색 화살표** : 경사하강법이 이동하려고 하는 방향
        - **모멘텀 화살표** : 이전 단계의 이동을 반영한 모멘텀 방향(관성을 부여한 화살표)
        - **파란색 화살표** : 실제 이동 방향

        - 장점: 작은 기울기는 쉽게 넘어가고 극소점을 탈출할 수 있게 함
        
        </div>

## 다중선형회귀
- 개념: 회귀 과정에서 **독립변수가 두 개 이상**인 경우. 최소제곱법과 경사하강법 사용 가능$$ y = W_1 x_1 + W_2 x_2 + ... + W_n x_n + b  $$
> 독립변수가 두 개인 경우: 3차원 공간에 표현. 회귀선 = 평면 -> 잔차 = |관측치 - 평면|
- **다중공선성**(multicollinearity): 독립 변수들 간의 강한 상관관계로 인해 회귀 모델이 불안정해지는 문제 -> 각 변수의 회귀계수 추정이 부정확해지고, 해석력 저하 가능(어떤 독립변수가 종속변수에 얼마나 영향을 미치는지 정확히 구분 어려움)
    - 확인 방법
        - **상관계수**: 변수 간의 통계적 관계인 상관관계의 정도를 수치로 나타낸 것. -1~1 값을 가지며, 부호에 따라 양의 상관관계와 음의 상관관계로 나뉜다 (0이면 상관관계 X)
        - **VIF 지수**(분산 팽창 인수): 회귀 모델의 결정계수 R-Square를 사용하여 계산
            - VIF = 1: 해당 독립 변수는 다른 변수와 상관관계가 전혀 X
            - VIF < 5: 다중공선성 문제 X
            - VIF > 5: 다중공선성의 징후. 주의 요구
            - VIF > 10: 다중공선성 심각. 해당 변수 제거 등 변수 간의 상관관계를 줄이기 위한 조치 필요
    - 대처 방법: 변수 제거/변환, 규제선형모델 활용, **PCA**
        >PCA(주성분 분석)이란?  
        
        고차원 데이터의 차원을 축소함으로써 복잡성은 줄이고, 정보는 최대한 보존하는 것. 상관있는 변수들을 묶어 더 적은 수의 대표 축으로 바꾸는 것이 핵심

## 규제선형모델
- 개념: 과적합(<-잔차 제곱 최소화 과정에서 회귀 계수가 커짐) 방지를 위해 규제를 가하는 모델. 선형 회귀에서 계수의 크기를 조정하는 역할 => 특정 변수에 과의존하지 않게
- 종류
    - **L2규제**: W(회귀계수)의 제곱에 페널티 부여  ex) 릿지(Ridge)회귀
    - **L1규제**:  W(회귀계수)의 절댓값에 페널티 부여   ex) 라쏘(Lasso) 회귀
    - **L2규제 + L1규제**: ex) 엘라스틱넷(Elastic Net) 회귀
- **릿지** 회귀: 큰 회귀계수일수록 강한 패널티(=$ {회귀계수}^2 $)를 부여함으로써 계수 크기 조절 $$ \text{Cost Function} = \text{RSS} + \alpha \sum_{j=1}^{p} w_j^2 $$
- **라쏘** 회귀: 회귀계수의 크기와 상관없이 같은 패널티(=|회귀계수|)를 부여함으로써 비중이 큰 중요한 변수만 남게함 $$ \text{Cost Function} = \text{RSS} + \alpha \sum_{j=1}^{p} |w_j| $$
- **엘라스틱넷**: 릿지+랏쏘 => 불필요한 변수는 제거하고, 남은 변수들은 적절한 크기로 조절. 안정적 예측과 희소성 동시 달성, but 속도 느림 $$ \text{Cost Function} = \text{RSS} + \alpha_1 \sum_{j=1}^{p} |w_j| + \alpha_2 \sum_{j=1}^{p} w_j^2 $$

# **모델 평가 방법**
- 성능 평가 지표
    - **평균 제곱 오차**(Mean Squared Error): 예측값과 실제값 사이의 오차 제곱의 평균 $$ \text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 $$
    - **평균 절대 오차**(Mean Absolute Error): 예측값과 실제값 사이의 절대값 오차의 평균 $$ \text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i| $$
    > MSE와 MAE의 차이  

    MSE-큰 오차에 민감하게 반응(제곱하기 때문) -> 이상치 민감  
    MAE-모든 오차 동일 취급 -> 이상치 존재해도 전반적 예측 성능 유지  
    뭐가 더 좋고 나쁜 것은 X. 모델과 데이터의 특성, 평가 목적에 맞는 걸 선택하면 됨  
- 변수 유의성 평가
    - T-검정: 독립 변수의 회귀계수가 유의미한지 검정. H0 = 독립변수가 종속변수에 영향 X (회귀계수 = 0), H1 = 독립변수가 종속변수에 영향 O (회귀계수 ≠ 0)
        1. t-값 계산 $$ t = \frac{\hat{\beta} - 0}{SE(\hat{\beta})} $$ (- $\hat{\beta}$ : 추정된 회귀계수, $SE(\hat{\beta})$ : 표준오차)
        2. p-value 확인 및 판단: 0.05 미만이면 H0 기각, 변수 유의미로 해석. 값이 작을수록 해당 변수는 종속변수에 영향력 ↑
