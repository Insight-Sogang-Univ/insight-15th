# 인공 신경망
- 우리 뇌의 신경망을 구성하는 신경세포끼리 서로 연결되어 정보를 처리하는 구조를 모방한 모델
1. 입력을 받아 각각 가중치와 편향을 적용
2. 종합한 값을 활성화 함수에 넘김, 활성화 함수는 임계값을 기준 삼음
    기준 미달? 0에 가깝게 죽임
    기준 통과? 유의미한 값으로 살려서 내보냄
3. output이 출력

# 퍼센트론
- 초기 형태의 인공신경망으로 신경망(딥러닝)의 기원이 되는 선형 이진 분류기
1. 입력값과 가중치를 곱함
2. 곱한 값들의 총합을 구함 (편향을 더할지는 옵션)
3. 총합이 임계값을 넘으면 1, 넘지 않으면 0을 출력
-> 학습: 원하는 출력밧을 내보내도록 가중치를 조정

## 단층 퍼셉트론
- 단순히 입력층과 출력층으로만 이루어진 퍼셉트론
-> 값을 보내는 단계와 값을 받아서 출력하는 단계로만 이루어짐
-> AND, NAND, OR 논리 게이트를 구현할 수 있음
- 디지털 회로에서 기본적인 논리 연산을 수행하는데 사용되는 구성요소
- 게이트 연산에 쓰이는 것은 n개의 입력값과 하나의 출력값

## 대표적인 논리게이트
1. AND 
- 입력값이 모두 1인 경우에만 출력값이 1이 나오는 구조
2. OR
- 입력값이 모두 0인 경우에 출력값이 0이고 나머지에는 모두 1
3. NAND
- 입력값이 모두 1인 경우에만 출력값이 0, 나머지 경우에는 모두 출력값이 1인 구조
4. XOR
- 입력값 두 개가 서로 다른 값을 갖고 있을 때에만 출력값이 1인 구조

## 단층 퍼셉트론의 한계
-> XOR 게이트 구현 불가
퍼셉트론은 입력값과 가중치의 선형 결합으로 이루어진 구조이기에 단층 퍼셉트론은 직선 하나로 두 영역을 나눌 수 있는 선형 분류 문제에만 구현 가능

# 다층 퍼셉트론
중간에 층을 추가하여 입력층 + 하나 이상의 은닉층 + 출력층으로 구성
- 두개 이상의 은닉층을 갖는 MLP가 딥러닝의 시초 모델로 발전
- 심층 신경망 = 은닉층이 2개 이상인 신경망
- 딥러닝 = 심층 신경망을 학습(가중치를 스스로 찾도록) 시킴

## 학습 알고리즘의 기본 구조
1. 순전파를 통해 입력값으로부터 예측 수행
2. 오차 계산을 통해 예측 정확도 측정
3. 역전파를 통해 출력층에서 다시 입력층 방향으로 오차 전달
4. 가중치 업데이트
순전파와 역전파를 계속 반복 (불쌍해 ㅜㅜ)

## 연쇄 법칙
입력값이 은닉층으로 거친 후 최종 출력값이 나오는데 입력값이 최종 출력에 얼마나 영향을 미쳤는지 계산하려면 각 층이 다음 층에 미친 영향을 순서대로 곱해 나가야 함.
- 각 단계가 연쇄적으로 결과에 얼만큼 영향을 줬는지 알 수 있는데 이게 연쇄 법칙임
- 딥러닝 모델의 성능을 높이려면 모델의 오차를 계싼해 순전파와 역전파를 활용해서 오차를 줄이는 것이 필요함
- 연쇄법칙을 통해 구하는 기울기 = 손실함수를 줄이기 위해 가중치를 어떻게 바꿀까를 알려주는 신호

## 기울기 소실 문제
- 은닉층도 많고 각 단계마다 미분값이 1보다 작으면 0에 가까워짐
-> 앞쪽 층의 가중치가 거의 업데이트되지 않아 학습이 제대로 이뤄지지 안흔ㄴ 문제 발생

## 손실함수
모델의 예측이 얼마나 틀렸는지를 수치로 표현하는 방법

1. MSE
평균 제곱 오차로 회귀에서 사용
- 값이 연속적이고 예측값과 실제값 간의 차이를 직접적으로 측정하는 회귀문제에 적합

2. 이진 크로스 엔트로피
$L= -(y \log \hat{y} + (1-y) \log(1-\hat{y}))$
이진 분류 문제(로지스틱 회귀)에 사용되는 손실함수로
(1) 실제 정답이 1일때
- 예측 확률이 1에 가까울수록 loss 값은 0에 가까워짐 = 잘 맞힘
  예측 확률이 0에 가까울수록 loss가 급격히 증가
(2) 실제 정답이 0일때
- 예측 확률이 0에 가까울수록 loss값이 0에 가까워져 잘맞힘
- 예측 확률이 1에 가까울수록 loss가 급격히 증가

3. 크로스엔트로피 오차
$-\sum y \log \hat{y}$
- 다중 클래스 분류 문제에서 예측 확률이 실제 클래스와 얼마나 일치하는지를 평가하는데 적합
- 두 확률 분포 (진짜 분포/예측 분포)의 유사도를 측정하는 데에 유용

## 활성화 함수
각 층의 출력값에 비선형성을 추가해서 신경망이 복잡한 패턴도 학습할 수 있도록 해주는 함수

### 계단 함수
- 입력값과 가중치의 곱을 모두 더한 뒤, 그 합이 특정 기준을 넘으면 1, 아니면 0을 출력하는 함수로 주로 단층 퍼셉트론에서 사용
-- 입력 신호: 입력값과 가중치의 곱
-- 출력 신호: 0 or 1

### 시그모이드 함수
- 입력 신호를 0과 1 사이의 연속적인 값으로 변환하는 비선형 함수
계단 함수는 0 또는 1만 출력하는 반면, 시그모이드 함수는 x값에 따라 출력값이 연속적으로 변화 -> 확실히 맞다/틀리다가 아닌 얼마나 확실한지를 표현

### ReLU 함수
- 입력이 0보다 크면 그대로, 0이하면 0을 출력하는 비선형 함수
1. 계산이 간단하고 효율적
2. 비선형성 제공
- 은닉층에 비선형성을 주지 않으면 층을 아무리 많이 쌓아도 선형 모델과 같아지는 문제 발생
- ReLU 함수는 음수를 0으로, 양수는 그대로 처리하며 은닉층에 비선형성 추가
3. 기웃기 소실 문제 완화
- Sigmoid나 Tanh는 입력값이 크거나 작을 때 미분값이 0에 가까워져 역전파가 거의 전달되지 않아 기웃리 소실문제가 발생 but 이 함수는 양수 구간에서 미분값이 항상 1이므로 역전파가 안정적으로 전달

### 쌍곡탄젠트 함수
$f(x)=tanh(x)=\frac{e^x-e^{−x}}{e^x+e^{−x}}$

- 시그모이드 함수의 변형으로 출력값의 범위가 -1~1임
- 입력값의 중심을 0으로 맞추고 싶을 때 사용, 중심이 0이면 학습 과정에서 가중치 업데이트가 더 안정적으로 이루어지는 장점 존재
- 입력값이 매우 크거나 작으면 기울기 소실 문제 발생

### Softmax 함수
$f(x_i)=\frac{e^{x_i}}{∑_je^{x_j}}$
- 각 클래스에 대한 출력밧을 확률로 변환하는 함수
- 다중 클래스 분류 문제의 출력층에서 사용할 수 있음
-> 출력밧이 모두 0~1 사이이고 전체 합이 1이 되기에 각 클래스에 속할 확률을 직접 읽을 수 잇음
- Softmax와 크로스 엔트로피 손실함수를 조합하면 손실함수가 예측의 확률과 정답 확률과의 차이를 바로 계산할 수 있어 학습 효율이 높음
-> 가장 표준적인 다중 클래스의 분류 조합

## 경사 하강법
- 오차를 가장 작게 만드는 가중치와 편향을 찾는 방법
1. 배치
- 학습 과정에서 가중치와 같은 매개변수를 조정하기 위해 사용하는 데이터의 묶음으로 사용자가 임의로 정할 수 있는 하이퍼파라미터임

2. 에포크
- 모든 데이터셋을 학습하는 횟수

3. 이터레이션 또는 스텝
- 한 번의 에포크를 끝내기 위해서 필요한 배치의 수
1에포크에서 다루는 데이터의 수 = 배치의 크기 * 이터레이션

### 배치 경사 하강법
- 전체 데이터를 통해 학습시키기에 업데이트 횟수가 적음
- 전체 데이터를 모두 한 번에 처리하므로 메모리가 많이 필요
- 항상 같은 데이터(전체)에서 경사를 구하기에 수렴이 안정적

#### 배치 크기가 1인 확률적 경사하강법
- 전체 데이터에 대해서 계산을 하기에 시간이 너무 오래 걸림
but 배치 크기가 1인 확률적 경사하강법은 매개변수 값을 조정할 때 전체 데이터가 아니라 랜덤으로 선택된 하나의 데이터에 대해서만 계산
- 하나의 데이터에 대해서만 메모리를 저장하면 되므로 자원이 적은 컴퓨터에 대해서도 쉽게 사용이 가능
- 매개 변수의 변경폭이 불안정하고 정확도 낮음

#### 미니 배치 경사하강법
- 배치 크기를 적절이 지정하여 해당 데이터 개수만큼에 대해 매개 변수의 값을 조정
- 전제 데이터를 계산하는 것보다는 빠르고, 배치 크기가 1인 경사하강법보다 안정적


## 옵티마이저
### 모멘텀
- 경사하강법 + 관성
- 경사하강법에서 계산된 접선의 기울기에 바로 한 시점 전의 접선의 기울기값을 일정 비율만큼 반영

### 알엠에스프롭
- 기존에는 기울기가 큰 방향에서 너무 많이, 작은 방향에서는 너무 작게 이동
- 각 학습률을 조정해서 안정적이고 빠르게 학습

### 아담
- 방향과 스텝 사이즈를 적절하게 합친 방법으로 방향과 학습률 두 가지를 모두 잡기 위한 방법

#### 데이터 증강
- 현실의 데이터는 다양하게 변형되고 조금씩 바뀌는 경우 多
- 따라서 학습 데이터 셋에 일부러 다양ㅇ한 변화를 가함
- 목적: 데이터 부족 문제해결, 모델의 견고함 향상, 과적합 감소
- 적용 분야: 비전(이미지), 자연어 처리 등
- 주의점: 변형된 데이터가 원본의 본질적인 특성을 잃지 않아야 하며, 테스크에 적절한 증강 방식을 선택

장점
1. 모델의 성능 향상: 다양한 변형 데이터를 학습함으로써 더욱 일반화된 패턴을 익히게 됨
2. 데이터 의존성 감소: 다양한 패턴을 학습하도록 함
3. 과적합 완화: 작은 데이터셋을 사용한다면 모델이 쉽게 과적합될 수 있기에 약간의 변형을 거치 데이터들을 추가적으로 학습하게 해 다양한 입력값을 갖게 함
4. 데이터 프라이버시 보호: 원본 데이터의 노출을 최소화하며 성능은 향상
