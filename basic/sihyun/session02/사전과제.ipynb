# 자료의 첫도
## 질적 자료
### 명목 척도
레이블이나 이름으로 요소의 속성을 식별하려고 하는 척도 (성별 등)
### 서열척도
명목형의 분류 속성과 더불어, 항목 간의 순서 또는 순위를 나타내는 특성 (만족도, 직급체계)

## 양적 자료
### 구간(등간) 척도
값들 사이에 순서가 있고, 그 간격이 일정하여 덧셈이나 평균과 같은 연산이 가능한 척도 -> 곱셈이나 나눗셈은 의미를 가지지 못함 (온도, IQ, 시간)
### 비율 척도
절대 영점과 임의의 측정 단위로 구성하고, 등간성이 유지되는 척도 -> 0값이 그 변수에 대해 아무것도 존재하지 않는 상태를 나타냄. 가격이 0이면 비용이 안들거나 무료임 (거리, 높이, 무게, 시간)

### 자료 vs 척도
자료의 종류는 데이터의 대략적인 성격을 파악하는 큰 그림, 척도는 그 데이터를 다루기 위한 더욱 구체적인 규칙과 기준

# 기술 통계
데이터를 있는 그대로 설명하고 이해하는데 중점을 둔 통계 방법
답해주는 질문들
1. 한 회사의 평균 매출
2. 특정 제품의 불량률
3. 고객들의 연령 분포?
## 중심화 경향
1. 평균(mean): 데이터 값의 총합을 데이터의 개수로 나눈 값 -> 이상치에 민감
-> 가장 좋아하는 예시로는 (연도는 틀릴 수 있음) 1990년에 노스캐롤라이나 대학교 졸업생들의 학과별로 평균 연봉을 계산해본 결과 가장 높은 과는 어디었을까? 답은 지리학과였다. (마이클 조던이 지리학과 출신이며 혼자서 평균을 말도 안되게 올려놓았기에)

--> 절사 평균: 최대, 최소값을 빼고 평균을 계산하는 방식

2. 중앙값: 데이터를 크기 순서대로 일렬로 정렬했을 때 한가운데에 위치하는 값 -> 이상치에 덜 민감함

3. 최빈값: 데이터에서 가장 자주 나타나는 값
-> 데이터 분포에 이해를 줌

## 자료의 산포
데이터가 퍼져있는 정도를 설명하는 기술 통계량

1. 범위
-> 데이터의 최대, 최소값의 차이를 나타내는 지표로, 데이터의 전체 변동 범위를 나타냄 -> 이상치에 매우 민감

2. 분산
-> 데이터 값들이 평균으로부터 얼마나 멀리 퍼져있는지를 나타내는 중요한 통계적 지표
-> 크면 클수록 더 퍼져있다
--> 분산에 루트 씌우면 표준편차 나옴

3. 사분위수
-> 데이터의 표본을 네 개의 동일한 크기로 나눈 값
Q1=하위 25%, Q2= 중앙값, Q3= 상위 25%

4. 사분위간 범위
Q3에서 Q1값을 제외한 범위, 즉 데이터의 중앙 50%가 퍼져있는 범위
아래쪽 이상치: Q1 - 1.5*IQR
위쪽 이상치: Q3 + 1.5*IQR

5. 변동 계수
-> 데이터의 절대적인 변동성이 아닌 상대적인 변동성을 나타내는 통계적 지표
변동 계수 = (표준편차/평균)*100

(참고)
첨도: 확률 분포의 꼬리가 두꺼운 정도
왜도: 비대칭도, 실수 값 확률 변수의 확률 분포 비대칭성을 나타냄

# 확률분포
확률 변수가 특정한 값을 가질 확률을 나타내는 함수

## 확률의 다양한 반론들
1. 표본의 불확실성: 이상한 표본 아냐?? -> 표본 분포
2. 모수 추정: 그 표본이 모수 대표해?? -> 점추정
3. 추정의 현실성: 점 추정값이 모수와 일치하는게 비현실적이지 않아? -> 구간 추정
4. 구간 추정의 오류 관리: 이 구간이 모수를 포함 못하면 어케 할거야?? -> 신뢰수준
5. 적극적 주장 검증: 특정 주장이 맞는지 아닌지 어케 알아? -> 가설 검정
6. 가설검정의 오류 관리: 이 주장이 틀렸다는 결론이 나왔는데 내 결론이 틀렸을 가능성은? -> 유의 수준

## 다양한 확률 분포
1. 이산확률 분포: 확률변수가 discrete한 값을 가지는 경우로 유한하거나 셀 수 있으며 특정한 값을 선택할 수 있음
(베르누이 분포, 이항분포, 기하분포, 음이항 분포, 초기하 분포, 포아송 분포 등)
2. 연속확률 분포: 확률 변수가 continuous한 값을 가지는 경우로 무한한 범위를 가지며 특정 값을 선택하는 것은 불가능
(정규, T, F, 감마, 지수, 카이제곱, 균등 분포 등)

## 정규 분포
종 모양으로 평균을 기준으로 대칭, 표준편차가 클수록 넓게 퍼짐 -> 가장 잘 알려진 연속 분포
모집단의 분포가 정규 분포를 따를 때 표본 평균의 분포도 정규분포를 따르나 n이 30 이상이면 표본 평균의 분포는 근사적으로 정규 분포를 따른다고 가정할 수 있음

(가정)
T, 카이제곱, F 분포는 정규 분포를 따르는 모집단에서 추출한 표본 통계량들이 따르는 표본 평균이므로 정규 분포를 따르는 모집단에서 왔다는 가정임\

## T분포
종 모양으로 정규분포보다 꼬리가 두꺼우며 평균을 기준으로 대칭 -> 자유도가 커짐에 따라 표준 정규 분포에 수렴, 표본으로 모평균(혹은 모평균 간의 차이)를 추정할 때 사용하는 분포
What is 자유도?: 몇가지의 변수가 자유롭게 변할 수 있는가임 자유도가 높을수록 정규분포에 가까워지나 너무 많은 자유도를 가지면 오버피팅의 문제가 발생함
통상적으로 모집단의 분산이 알려지지 않은 경우 표본 분포를 결정할때 n이 30 미만이면 t, 이상이면 표준 정규 분포 사용

## 카이제곱
0에서 시작해 오른쪽으로 긴 꼬리를 가진 비대칭 분포
자유도가 증가함에 따라 중심이 오른쪽으로 이동하고 모양이 더 대칭적으로 변화
독립적인 표준 정규 분포를 따르는 변수들의 제곱 합

## F분포
0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포
자유도가 증가함에 따라 중심이 오른쪽으로 이동하고 모양이 더 대칭적으로 변화하고 정규 분포에 가까워짐
두 독립적인 카이 제곱 분포의 비율
두 집단 간의 분산 차이를 비교하는데 사용됨






# 추론 통계
: 수집한 데이터를 기반으로 미래의, 혹은 조사하지 못한 전체 정보를 추론하고 결론을 도출하는 방법을 다루는 분야

## 추정
모집단을 대표할 수 있는 일부를 표본으로 추출하여 분석한 통계량을 이용해 모집단의 모수를 예측
1. 점 추정: 추정하고자 하는 모수를 하나의 수치로 추정
2. 구간 추정: 추정하고자 하는 모수로 존재하리라 예상하는 구간을 정하여 추정

## 검정
표본을 추출하여 얻은 표본 통계량으로 모집단 모수에 대한 새로운 예상, 가설이 옳다고 판단할 수 있느지를 평가하는 과정

### 추정량 VS 추정치
추정량이란 표본 정보에 의존하는 확률 변수로서 모수를 추천하는데 사용되는 표본 통계량
추정치란 표본이 추출된 후 표본 자룔를 이용하여 계산한 통계량의 특정한 값

표본 평균은 확률 변수로서 표본에 포함되는 요소에 의존하기 때문에 표본마다 그의 값이 다르게 되어 모평균과 같지 않을 가능성이 매우 높다. 그래서 점 추정의 경우 오차에 대한 정보가 전혀 고려되지 않아 신뢰구간 추정 방법이 일반적으로 널리 사용

## 가설 검정의 절차
가설 검정은 모집단의 실제값이 얼마가 된다는 주장과 관련해 표본의 정보를 사용해서 가설의 합당성 여루블ㄹ 판정하느 ㄴ과정
가설 설정 -> 유의 수준 설정 -> 검정 총계량 산출 -> 가설 기각/채택 판단

1. 가설 설정
귀무 가설 H0와 H1를 설정 
-> 귀무 가설은 주로 기각하고자 하는 명제로 보통 ~는 차이가 없다
대립 가설은 내가 하고 싶은 말로 ~는 차이가 존재한다

2. 유의 수준 설정
연구자가 결과를 판단할 때 최대로 허용할 수 있는 오류의 최대치로 5%를 주로 설정
1종 오류: 귀무 가설이 실제로 참이었는데 기각할 확률
2종 오류: 귀무가설이 실제로 거짓인데 기각하지 못할 확률
 -> 1종 오류가 더 치명적인 것이기에 유의 수준은 1종 오류의 확률을 의미

3. 검정 통계량 산출
가설 검정의 결과를 판단하는데 사용되는 값
 -> Z, T, F 중 어떤거를 사용할지는 판단해야 함

4. 가설 기각/ 채택
유의 확률: p-value: 가설 검정에 쓰일 수 있는 또 다른 척도로 검정 통계량에서 유도할 수 있고 유의 확률이 유의 수준보다 작을 시에 귀무가설을 기각, 클 시에는 귀무 가설을 기각 X
