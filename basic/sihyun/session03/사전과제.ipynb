# 머신러닝
인공지능의 한 분야로, 모뎅이라는 소프트웨어를 학습하여 새로운 데이터를 예측하거나 계정을 내릴 수 있도록 하는 기술 -> 패턴 발견 & 패턴을 바탕으로 데이터에 대해 예측하는 것

## 종류
1. 지도학습: 정답이 있는 데이터를 사용해 기계의 예측값과 데이터의 정답이 같아지도록 학습시키는것(회귀, 분류)
2. 비지도 학습: 정답이 없는 데이터 사용으로 데이터의 패턴 혹은 유사성을 학습시키는 것(군집화)
3. 강화학습: 시행착오를 반복하여 보상을 최대화하는 행동 학습(알파고, 자율 주행 등)

## 모델링
1. 모델이란 학습의 결과물로서 학습을 마친 인공지능 그 자체(예측기)를 의미하기도 함
-> 데이터를 학습하먀 찾아낸 패턴이 저장된 상태이며, 새로운 데이터를 넣었을 때 답을 내놓는 예측기 역할

2. 모델링이란 예측기(공식)을 만들어가는 전체적인 과정
-> 아무것도 모르는 기계가 주어진 데이터 속에서 스스로 패턴을 찾아내 오차가 작은 최적의 공식을 완성할때까지 학습하는 과정

3. 모델링의 과정
3-1 설계: 데이터의 특성에 맞추어 어떤 알고리즘으로 학습할지 결정하는 단계
3-2 훈련: 데이터를 기계에게 계속 보여주며 정답을 맞히도록 수정하는 단계
3-3 검증: 훈련시킨 모뎅이 새로운 데이터에서도 정답을 맞히는지 테스트하는 단계

# 지도학습
기본적으로 데이터가 labeled data 형태여야함

1. labeled data란 독립변수와 종속변수가 모두 존재하는 데이터임 -> 입력 데이터와 그에 대응하는 출력값이 있는 데이터
-> 독립변수: 머신러닝에서 예측에 사용되는 입력 변수
-> 종속변수: 머신러닝에서 예측하고자 하는 대상 변수

2. training set, test set으로 나뉨
-> traning set: 모델을 학습(훈련)시킬 때 쓰이는 데이터
-> test set 학습된 모델을 검증하기 위해 사용하는 데이터
--> 학습시킨 데이터로 새로운 데이터를 얼마나 잘 예측하는지 공평하게 평가하기 위해서 나눔


3. training set을 통해 모델을 학습
-> 독립 변수를 통해 추정한 예측값이 주어진 종속변수와 같아지도록 모델을 훈련시키는 것

--> 예측값과 실제 정답 사이의 차이 즉 오차를 줄어들게 하도록 가중치를 스스로 수정함
-> 학습이 끝나면 데이터 사이의 통계적인 상관관계들을 하나의 함수 형태로 가지게 됨

# 회귀 & 분류
## 회귀 
주어진 데이터(X)를 기반으로 정답 (y)를 잘 맞히는 함수를 찾아내는것
ex. 주택 가격 예측에서 종속변수는 주택 가격, 독립변수는 주택 평 수, 여까지의 거리, 지역의 평균 소득 등

## 분류
기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별하는 것
ex. 대출 상환 여부 예측에서 종속 변수는 대출 상환 여부(0/1), 독립변수: 고객의 특성, 고객이 받은 대출의 특성

### 회귀 VS 분류
예측하고자 하는 값이 연속형이면 회귀, 범주형이면 분류를 사용

# 회귀
## 개요 
데이터에서 패턴을 찾아내어 미래값을 예측하는 것으로 통계나 머신러닝의 주요 개념
종속 변수가 연속적일 때 사용하며 선영, 비선영, 규제, 다항회귀 등 다양한 기법이 존재함

## 선형 회귀
데이터를 가장 잘 설명하는 회귀선을 찾는 과정임
현대 데이터를 요약하는 동시에 아직 관측되지 않은 미래 값을 예측하는 도구
단순(x가 한개), 다중(x가 여러개)로 나눌 수 있음

## 가중치와 편향의 이해
선형 회귀의 기본식
$y = wx + b$

1. 가중치
독립변수 x가 종속변수 y에 주는 영향력의 크기 -> 그래프에서는 기울기, 회귀계수라고도 부름

2. 편향 b
독립변수 x가 0일 때 나오는 기본 값으로 그래프에서는 y절편으로 나타남

## 단순 선형회귀
하나의 독립 변수만 존재하는 선형 회귀
데이터를 직선 형태(하나의 독립 변수)로 표현하는 경우
-> 잔차를 줄여가며 직선을 결정 짓는 적절한 회귀계수와 편향을 찾음

## 잔차
실제값과 예측값간의 차이를 잔차라고 하며 이를 줄여가는 과정이 필요

## 최소 제곱법
최적의 회귀선을 찾는 방식 중 하나로 최소 자승법이라고도 부름
-> 잔차의 제곱 합이 최소가 되는 지점으로 최적의 회귀선을 구하는 방식

## 경사하강법
최소 제곱법을 통해 구한 함수가 복잡하여 최솟값을 구하기 어려울 때는 보통 경사하강법을 사용

1. 목적 함수: 우리가 달성하고 싶은 목표
-> 문제의 성격별로 최대화할 수도 최소화할 수도 있음
2. 손실함수: 개별 데이터에 대한 오차를 측정하는 함수
-> 예측값과 실제값의 차이를 계산해서 한 개의 데이터 포인트에 대해 모델이 얼마나 틀렸는지 평가
3. 비용함수: 전체 데이터 셋에 대한 손실 함수를 합산하거나 평균을 내어 모델의 전체적인 성적을 평가하는 함수

### 손실함수
$E = \frac{1}{n} \sum_{i=1}^{n} [t_i - y_i]^2$


- $t_i$ : $i$번째 데이터포인트 실제값(정답)
- $y_i$ : $i$번째 예측값
- $E$: Error
- $n$: 데이터 샘플의 총 개수

$E(W, b) = \frac{1}{n} \sum_{i=1}^{n} [t_i - y_i]^2 = \frac{1}{n} \sum_{i=1}^{n} [t_i - (Wx_i + b)]^2$

1. 현재값과 최솟값을 비교(미분계수 이용)
2. 미분계수가 음수이면 오른쪽, 양수이면 왼쪽으로 이동
3. 최적값을 찾아가는 과정

#### 학습률
그 다음으로는 얼만큼 이동할 것인가가 문제임. -> learning Rate라고 부름
-> 작은 기울기에서는 조금, 큰 기울기에서는 많이 이동함
--> 모든 비용함수가 하나의 최솟값을 가지는 매끈한 곡선 형태는 아니기에 극소점, 극대점이 존재할 수 잇음

#### 해결법 - 모멘텀
기존 경사하강법은 이전 기울기는 다 잊어버리는 경향 -> 현재 기울기 기준으로 움직임 -> 이전 기울기와 방향을 고려하도록 관성을 부여 필요


# 다중선형회귀
$y = W_1 x_1 + W_2 x_2 + ... + W_n x_n + b$
독립 견수가 두 개 이상인 경우를 다중선형회귀라고 함
x가 2개인 경우에는 3차원 공간에 표현되고 회귀선은 평면이 됨 -> 이보다 변수가 많은 경우에도 비슷한 느낌으로 차원이 화장됨
마찬가지로 최소제곱법/경사하강법을 사용할 수 있으며 잔차는 관측치와 평면간의 차이가 됨

## 다중 공선성
독립변수들 간의 강한 상관관계로 인해 회귀 모델이 불안정해지는 문제
-> 다중 공선성이 문제가 심각하면 각 변수의 회귀 계수 추정이 부정확해지고 해석력이 떨어지기에 선형회귀에서 중요한 체크 사항
ex. 독립변수간에 강한 상관관계를 가진다면 이 두 변수 간에 다중 공선성이 발생함
-> 다중 공선성이 높으면 어떤 독립변수가 종속 변수에 얼마나 영향을 끼쳤는지 명확하게 구분하기 어렵기 때문에

### 확인 방법
1. 상관계수
변수간의 통계적 관계인 상관관계의 정도를 수치로 나타낸 것
2. VIF 지수
VIF 지수가 높으면 다중공선성이 존재한다고 판단, VIF-1 -> 해당 독립변수는 다른 변수와 상관관계가 전혀 없음 <5이면 묹가 없으며 5보다 크면 징후 존재 10보다 크면 심각하다고 생각하면 됨

### 대처 방법
1. 변수 제거
2. 변수 변환 -> 
변수들을 더하거나 빼서 새로운 변수 생성, 독립변수를 더하거나 빼더라도 문제가 없는 경우
3. 규제 선형 모델 활용
릿지, 라쏘, 엘라스틱넷 등의 방법을 통해 모델의 복잡도를 줄일 수 있다
4. PCA(주성분 분석)
데이터의 차원을 축소하는 데에 사용되는 통계적 기법으로 고차원 데이터에서 중요한 정보를 최대한 보존하면서 데이터의 복잡성을 줄이고자 하는 것
-? 데이터의 차원이 높으면 분석과 시각화 모두 어렵기 때문

# 규제선형모델
모델이 학습데이터에 과적합되지 않도록 규제를 가하고자 등장한 모델
비용함수는 학습 데이터의 잔차 오류 값을 최소화하는 RSS 최소화 방법(최소 제곱법)과 과적합을 방지하기 위해 회귀 계수 값이 커지지 않도록 하는 방법이 서로 균형을 유지하는 것
-> 오류를 최소화하면서 특정 변수에 집착하지 않도록 균형을 맞춰 과적합을 방지하는 것

## 종류
1. L2 규제: 회귀계수의 제곱에 패널티를 부여하는 방식
2. L1 규제: 회귀계수에 절댓값에 대해 패널티를 부여하는 방식
3. L2 + L1 방식

### L2 규제
회귀 계수의 제곱을 패널티로 주기에 큰 회귀계수일수록 훨씬 강한 패널티를 받음
-> 모든 계수가 작아지지만 규제를 통해 변수가 제거되지는 않음

### L1 규제
모든 기둥을 같은 힘으로 누르는 망치
회귀계수의 절댓값을 태널티로 주기에 회귀 계수 크기와 상관없이 같은 패널티
-> 작은 계수가 0이 되기도, 변수가 완전히 사라질 수도

### L1+L2 규제
불필요한 변수는 제거하고 남은 변수는 적절한 크기로 유지
-> 안정적인 예측(모든 변수가 균형있게 줄어듦)과 희소성(변수 제거)를 동시에 달성할 수 있으나 계산량이 많아져서 속도가 느림

# 모델 평가법
## 성능 평가 지표
1. 제곱 오차 (MSE)
예측값과 실제값 사이의 오차 제곱의 평균
2. 평균 절대 오차
모델의 예측값과 실제값 사이의 절대값 오차의 평균
-> MSE는 큰 오차에 대해 민감하게 반응 & MAE는 모든 오차를 동일하게 취급

## T 검정
t값을 계산하고 p-value를 확인
회귀계수가 0이면 해당 독립 변수는 의미 없으며 p-value가 작을 수록 해당 변수는 종속 변수에 더 큰 영향을 줄 확률이 높아짐
