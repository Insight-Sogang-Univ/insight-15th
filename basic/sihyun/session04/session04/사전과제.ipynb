# 선형회귀
## 다중 선형회귀의 수식적 이해
$y_i = \beta_0+\beta_1X_{1i}+\beta_2X_{2i}+...+\beta_kX_{ki}+\varepsilon_i$
-> 관심있는 y에 대해 영향을 줄 것 같은 여러 설명 변수(독립변수)를 채택하여 선형적으로 식을 작성한 모습

만약 상수항과 오차항을 제외하면
$y=w_0x_0+w_1x_1+\cdot\cdot\cdot+w_mx_m=\sum_{i=0}^mw_ix_i=w^Tx$
선형 회귀식은 결국 종속 변수 y에 대한 독립 변수들의 가중 평균이라고도 표현
설명 변수에 가중치를 곱해 더함으로써 y를 얼마나 잘 설명하는지를 나타냄

## 다중 선형회귀의 기본 가정 6가지
1. 선형성
$E(y|X) = \beta_1+\beta_2X_2+...+\beta_nX_n$
종속변수와 설명변수 간의 관계가 선형적이어야 한다는 가정 -> 당연함 

2. 독립성
각각의 설명변수가 서로 선형 독립적이어야 한다
선형독립적이지 않은 경우 다중 공선성이 존재함
    다중공선성: 독립 변수들이 서로 강한 상관관계를 가져서, 어떤 변수가 종속 변수에 가장 큰 영향을 주는지를 알 수 없는 상태
        있는지 알아보는 방법: 상관계수(r)이나 VIF를 구해보는 방법
        대처: 변수 제거, 규제 선형 모델, PCA

3. 오차항의 평균은 0 or 0에 근접해야 함
오차항 - 실제값과 예측 값 사이의 차이
오차항은 백색 잡음이라고도 함 (백색잡은: 특정 패텅 없이 랜덤하게 변하는 신호를 의미)

4. 등분산성
$Var(\varepsilon_i|X_i) = \sigma^2$
오차항의 분산이 일정해야 한다는 가정
-> 오차항의 분산이 일정하지 않은 경우, 이분산성이 있다고 함 
    횡단면 자료에서 많이 나타남 (횡단면 자료란 특정 시점에서 여러 개체를 관찰한 데이터 ex. 2025년 서울시 100개 기업의 매출액 등)
        회귀 분석의 결과가 정확하지 않을 수 있음

5. 오차항은 자기 상관되어 있지 않다.
$Cov(\varepsilon_i, \varepsilon_j|X_i) =0$ 
오차항의 공분산은 항상 0이어야 한다는 가정
    공분산이랑 두 변수 간의 상관된 정보를 나타내는 지표로 두 변수가 어떻게 함께 변하는지를 측정하는 값
        >0, x가 증가하면 y도 증가
        <0, x가 증가하면 y는 감소
        =0, x와 y는 독립
    오차항의 공분산이 0이 아닌 경우 자기 상관이 있다고 함
        자기 상관: 한 변수의 현재 값이 과거 값과 상관 관계를 가지는 현상
        시계열 자료에서 많이 발생함
        모델이 데이터를 랜덤하게 예측하는 것이 아니라, 이전 데이터의 패턴을 학습하여 반복적인 오류를 만들어 낼 수 있음

6. (선택) 정규성
오차항이 정규 분포를 따른다는 가정
위배되어도 다중 선형회귀분석의 결과에 큰 영향 X

## 회귀 분석 전 vs 후 검증해야 할 가정 정리
                전  후  방법
선형성           O   O   scatter plot
독립성           O   O   VIF 지수, 상관계수
오차항 평균=0,   X   O    np.mean
등분산성         X   O    잔차의 도표화
오차항 자기상관X, ?  O     Durbin-Watson 검정
정규성           X   O     Shapiro-Wilk 검증, Q-Q plot

-> 오차항에 대한 가정인가? 를 기준으로 나눔
1. 회귀식은 보통  $y = \beta_0 + \beta_1X + \epsilon$ 인데 오차항 엡실론은 모델 학습 후에 정해지는 정답과 예측 간의 차이
2. 모델 학습 전에는 X와 y만 가지고 있기 때문에 오차항을 계산할 수 있는 근거가 없기에 모델 학습 전에는 확실하게 알 수 없음
3. 모델 학습 후에 검증을 진행했는데 가정에 위배되면 다시해야함 ㅠㅠㅠㅠㅠ
4. 데이터 분석은 한번에 끝나는 것이 아니라 수정과 검증을 거듭 진행해야 함

여섯개의 가정을 완벽하게 만족시키기 어렵기에 다음과 같은 기준으로 타협하기도 함
1. 데이터 행 수가 충분히 많다면(30개 이상), 중심극한 정리에  의해 정규성 위배는 눈감아줌
2. 회귀 모델의 목적이 설명인가 예측인가
-> 설명인 경우 가정을 지키는 것이 중요, 가정이 깨지면 계산된 p-value를 믿을 수 없음
-> 예측인경우 가정이 조금깨지더라도 실제 예측 오차(MAE, RMSE)가 낮으면 그대로 사용

# 회귀 분석 평가 방법
회귀선을 데이터와 함께 시각화하거나 통계지표를 이용

## 시각화
회귀선이 데이터 전반을 잘 요약하는 모습
그러나 회귀선 간에 어느정도로 잘 요약하는지 객관적으로 비교하기 어렵기에 통계지표를 함께 써야 함

## 통계지표
### 모델 유의성 검정
1. F-statistic, Prob(F-statistic)
F-통계량은 회귀 모델이 유의미한지 검정하는 지표
귀무 가설 H0: 모든 회귀 계수가 0이다 (독립변수들이 종속변수들과 관계가 없다)
대립 가설 H1: 적어도 하나의 회귀 계수는 0이 아니다(독립변수 중 하나 이상이 종속 변수에 영향을 준다)

F-통계량 값이 클수록 모델이 통계적으로 유의미하다는것
Prob(F-통계량)은 p-value로 0.05보다 작다면 귀무가설을 기각할 수 있음 (적어도 하나의 독립 변수가 종속변수와 유의미한 관계를 갖고 있다)

F-검정은 a=모델이 설명할 수 있는 변동, b= 모델이 설명하지 못하는 변동으로 구분
F-통계량 -> a/b
F값이 1에 가까울 때는 독립 변수들이 종속변수를 설명하는 데에 아무런 도움 X
F값이 매우 클때 오차의 양에 비해 모델이 설명할 수 있는 변동의 양이 압도적으로 많아 통계적으로 유의미

--> F-통계량이 클수록 Prob(F-통계량)이 작아짐 -> 모델이 설명할 수 잇는 변동의 양이 오차의 양보다 크다 -> 모델이 유의미하다

2. T-통계, Prob(T-통계)
귀무 가설 H0: 해당 변수의 회귀 계수가 0이다
대립 가설 H1: 해당 변수의 회귀계수는 0이 아니다
P>|T-statistic|를 기준으로 가설 검정 수행

1. x1 > P>|T| 값이 0.000이므로 0.05보다 작다 -> x1이 종속 변수에 유의미한 영향을 미친다
2. x2 > P>|T| 값이 0.031이므로 0.05보다 작다 -> x2가 종속 변수에 유의미한 영향을 미친다
3. x3 > P>|T| 값이 0.567이므로 0.05보다 크다 -> x3가 종속 변수에 유의미한 영향을 미치지 않는다

3. F vS T

    검정 대상   목적                            해석 기준
F   전체 모델   회귀 모델이 유의미한지 검정        p<0.05 이면 회귀 모델이 유의미
T   개별 변수   특정 독립변수가 유의미한지 검정     p<0.05 이면 해당 변수가 유의미

## 모델의 성능 평가
1. 결정 계수 R^2은 회귀 모델이 종속 변수의 전체 변동성을 얼마나 잘 설명하는지 나타내는 지표로 모델이 데이터를 얼마나 잘 설명하는지 측정
단순히 평균값으로 데이터를 예측했을 때보다, 모델을 활용했을 때 오차가 얼마나 줄어드는지 수치화한것
    SST: 편균으로 예측했을 때의 총 오차
    SSE: 모델이 예측하고도 여전히 남은 오차(모델의 한계)
    SSR: 편균에 비해 회귀 모델이 얼마나 잘 맞혔는지
R^2=SSR/SST=1-SSE/SST

2. R^2 값이 높으면 다 좋은건가?
-> 독립 변수의 개수가 증가하면 그대로 증가하기에 데이터와 큰 관련성이 없는 변수를 추가해도 R^2값은 높아짐 -> Adjusted R^2를 사용

3. AIC
BIC에 비해 복잡성에 대한 패널티가 비교적 작다

4. BIC
AIC보다 엄격한 기준으로 데이터의 양에 따라 더 강한 패널티를 부과
-> 3,4 모두 값이 낮을수록 좋은 것

# 비선형 회귀
1. 다항식 회귀 모델
$Y=a+bX+cX^2+dX^3+\cdot\cdot\cdot$ 
형태로 독립 변수에 거듭제곱항을 추가하여 데이터의 비선형적 특성을 모델링

-> 선형 모델에 비해 회귀선을 데이터에 fit하게 그릴 수 있음
--> 너무 많은 독립변수를 이용하면 결과가 overfitting될 수 있음
        릿지, 라쏘 규제를 이용
적합한 상황
    관계가 직선적이지 않고 곡선을 그리는 데이터 패턴에서 유용
    성장률, 감소율 등이 시간에 따라 변화하는 속도가 다른 경우에 좋음

2. 지수 회귀 모델
 $Y = ae^{bX}$
종속변수가 지수적으로 변화하는 관계를 모델링할 때 사용
    지수 모델은 종속 변수 Y에 로그를 취한 후 선형 회귀를 적용하면 됨
        원래의 데이터 형태로 복구하기 위해 예측 값에 지수함수를 적용

-> 값이 시간에 따라 지수적으로 증가하거나 감소하는 경우 적합
ex. 기술 발전, 투자 수익률 증가 등

3. 로그 회귀 모델
$Y=a+b\log(X)$
예측된 값이 Y이므로 예측값에 별도의 처리를 하지 않고 그대로 사용

-> 데이터가 초기에 빠르게 증가하고 점차 증가율이 줄어드는 패턴을 보일 때 효과적
ex. 인구 성장, 감염병의 확산과 같이 초기 빠른 증가 후 점진적 감소를 보이는 데이터에 적합

4. 스플라인 회귀
데이터를 구간별로 나누고 각 구간에서 다른 선형 또는 비선형 함수를 적용하여 예측하는 방식
구간 경계에 대해서 연속성을 유지하는 것이 중요
- $Y = \begin{cases} a_1 + b_1X & \text{if } X < c_1 \\a_2 + b_2X & \text{if } c_1 \leq X < c_2 \\\cdots \\a_n + b_nX & \text{if } X \geq c_{n-1}\end{cases}$

-> 데이터 패턴이 여러 구간에서 서로 다르게 나타날 때 유용
ex. 계절에 따라 판매량이 변화하는 소매 데이터 분석 등에서 적합
