## 지도학습: 회귀 vs 분류
회귀는 연속적인 숫자값을 예측, 분류는 입력된 데이터를 주어진 항목들로 나누는 것

-회귀: 주어진 데이터(X)를 기반으로 정답 (Y)를 잘 맞추는(fit)하는 함수임 -> 연속형 변수를 예측하기 위해 사용
-분류: 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블 판별하는 것 -> 범주형 변수를 예측하기 위해 사용
--> 로지스틱 회귀(독립변수와 종속변수의 선형 관계성에 기반해 분류), 결정 트리(데이터 균일도에 따른 규칙 기반의 분류), 서포트 벡터 머신(개별 클래스 간의 최대 마진을 효과적으로 찾아 분류), 최소 근접(근점 거리 기준으로 분류)

(개인적인 궁금증)
로지스틱은 회귀인데 왜 분류에서 사용?
-GPT 피셜 수학적으로는 확률에 대한 회귀 문제인데 실제 사용은 확률을 임계값으로 잘라 클래스 결정하기 때문 


## 이진 분류 vs 다중 분류
이진 분류: 예측하고자 하는 변수가 어떤 기준에 대해 T,F 값만을 가질 때
다중 분류: 예측하고자 하는 변수가 가질 수 잇는 값이 3개 이상

### 로지스틱 회귀
이진 분류 문제를 푸는 대표적인 알고리즘으로 샘플이 특정 클래스에 속할 확률을 추정하여 기준치에 따라 분류하는 것

- 일반적인 선형회귀는 연속적인 값을 예측하는데 사용되나 로지스틱은 이진 분류 문제를 해결하는데 적합 

(필요성)
직선으로 이진 분류 문제를 풀 수 없는 경우
ex. 특정 임계값을 기준으로 합/불을 결정할 때 이를 선형으로 구분할 수 없음
-> 출력이 0과 1 사이의 값을 가지면서 S자 형태로 그려지는 함수를 이용해야 함 = 시그모이드함수

시그모이드 함수 (유사) 로지스틱 함수
(특징)
- 입력값이 커지면 1에 수렴, 작아지면 0에 수렴
- 입력값이 $(-\infin, \infin)$일때 출력값은 0~1의 값을 가짐
(가중치)
- 주어진 데이터에 적합한 가중치 w와 b를 계산해서 찾는다
- 가중치 w의 변화에 따른 그래프 변화(그래프의 기울기가 커짐에 따라 위로 세워짐)
- b가 커짐에 따라 위로 올라감 (당연)

### 의사결정 나무
조건에 따라 데이터를 분류하며, 최종적으로 데이터가 순수한 label의 집합으로 구설될 때까지 분류를 반복하는 분석 방법

- ROOT Node: 가장 상위에 있는 노드
- Parent Node: 주어진 노드의 상위에 있는 노드
- Child Node: 하나의 노드로부터 분리된 2개 이상의 노드들
- Binary Node: 트리 중에서 children이 최대 2개인 노드
- Leaf Nodes: 자식 마디가 없는 노드로 모델에서 label에 해당
- Edge: Parent와 child 노드를 연결하는 연결선
- Height: 특정 노드에서 가장 멀리 있는 leaf node까지의 경로에 있는 edge 수
- Depth: Root node에서 특정 노드에 도달하기 위한 edge 수

#### SVM
클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘
- 명확하게 분류할 수 잇는 데이터 집단에서 뛰어난 성능을 보이고, 고차원 공간에서 효과적으로 사용 가능
-> 패턴 인식, 이지 분석, 바이오인포매틱스 등 다양한 머신 러닝 응용분야에서 활용

(구성)
- support vector: 데이터 중에서 결정 경계와 가장 가까이에 있는 데이터들의 집합
- Decision Boundary: 데이터의 분류의 기준이 되는 경계
- Margin: 결정 경계에서 서포트 벡터까지의 거리
- Hyperplane: n차원 공간의 n-1 차원 평면
- Slack Variables: 완벽한 분리가 불가능할 때 분류를 위해 허용된 오차에 대한 변수

(가장 좋은 결정 경계)
- 결정 경계는 데이터로부터 가장 멀리 떨어져 있는게 좋음. 따라서 SVM은 margin이 가장 큰 경우를 선택함으로써 최적의 선을 찾음

(장점)
1. 서포트 벡터만을 이용해서 결정 경계를 생성하므로 데이터가 적을 때 효과적
2. 새로운 데이터가 입력되면 전체 데이터 포인트와의 거리를 계산하지 않고 서포트 벡터와의 거리만 계산하면 되기 때문에 연산량 최소화
3. 정확성이 뛰어나며, 비선형 모델 분류 가능
4. 다른 모형보다 과대 적합의 가능성이 낮고 노이즈의 영향이 적음

(단점)
1. 데이터 전처리 과정이 중요
2. 데이터가 많아질수록 최적화를 위한 평가 과정이 많아져 속도 느림

#### KNN
비슷한 특성을 가진 데이터끼리 서로 가까이 있다는 점을 이용한 분류 알고리즘으로 거리가 가까운 K개의 다른 데이터 레이블을 참조하여 분류

(순서)
1. 데이터 준비
미리 학습하는 과정 X, 각 데이터는 특징 벡터와 레이블(어떤 class인지)로 이루어짐
2. K 값 설정
가장 가까운 이웃의 개수를 나타내며 보통 홀수개로 설정 -> 짝수개로 하면 동점이 존재하기에
3. 거리 계산
새로운 데이터(예측하려는 데이터)가 주어지면, 이 값과 기존 모든 데이터 간의 거리 계산
- 유클리드 or 맨허튼 거리 사용
4. 가장 가까운 K개의 이웃 선택
계산된 거리 중에서 가장 작은 거리 값을 가진 K개의 데이터를 선택 -> 가장 가까운 이웃으로 선택
5. 분류하기
K개의 이웃 중 가장 많이 등장하는 클래스가 예측 결과

(How to decide K)
- 일반적으로 학습용 데이터 개수의 제곱근으로 설정
- K를 너무 크게 설정하면 주변에 있는 점과의 근접성이 떨어져서 분류가 잘 이뤄지지 않고 과소 적합
- K를 너무 작게 설정하면 이상값, 잡음 데이터와 이웃이 될 가능성 존재

(홀수로 설정해도 동점이 나오면?)
- 동점이 발생한 클래스들 중, 더 가까운 이웃에 해당하는 클래스로 분류
- 동점이 발생한 클래스들 중, 무작위로 하나 선택

(장점)
1. 범주를 나눈 기준을 알지 못해도 데이터를 분류할 수 잇음
2. 입력 데이터만 주어지면 바로 예측값을 구할 수 있어서 다른 분류 모형과는달리 학습과정이 필요없음
3. 이해가 쉽고 구현이 간단
4. 추가된 데이터의 처리가 용이
5. 훈련데이터를 요약하거나 일반화된 모델로 만들지 않고, 원본 데이터 전체를 그대로 저장

(단점)
1. 테스트 데이터 개수에 따라 시간이 오래 걸림
2. 학습데이터 모두를 거리 계산에 사용하기 때문에 학습 데이터의 양도 계산 시간에 영향을 미침
3. K값 결정 어려움
4. 수치형 데이터 아니면 유사도 정의 힘듦
5. 데이터 내 이상치 존재시 분류 성능에 영향

## 분류 평가 지표
### 혼동 행렬
분류 모델의 예측 결과를 정확한 예측과 잘못된 예측으로 구분하여 나타낸 표

(평가지표)
- 정확도: 모든 가능한 예측 중 참인 비율
(TP+TN)/(TP+TN+FP+FN)
단점: 정압 레이블비율이 불균형하면 모델의 정확도 신뢰 X

- 정밀도: 참이라고 예측한 경우 실제 참의 비율
(TP)/(TP+FP)
거짓을 참으로 판단한 정도를 알 수 있음 -> 정밀도가 높을수록 거짓을 참으로 예측한 정도가 낮음
- 재현도: 실제로 참인 경우 중 참으로 예측하는 비율
(TP)/(TP+FN)
참을 거짓으로 판단한 정도를 알 수 있음 -> 재현도가 높을수록 참을 거짓으로 예측한 정도가 낮음

(정밀도와 재현도는 trade off 관계에 있기에 상황에 따라 무엇을 높여야 하는지 다름)
- 암환자의 경우 FN을 낮춰야 하므로 재현율을 높여야 함
- 스팸 메일의 경우 스팸이 아닌 메일을 스팸으로 분류하는 현상이 치명적이므로 정밀도를 높여야 함

- 분류할 때 확률에 기반해서 Threshold가 넘으면 참, 미만이면 거짓으로 판단하므로 이 경계값을 조정하면 정밀도와 재현도를 조정할 수 잇음

-> 정밀도와 재현도가 만나는 지점을 임계값으로 정할 떄 예측 오류를 최소화할 수 잇음

### F1-score
정밀도와 재현도의 조화평균
-> 머신러닝 모델의 성능을 평가하는 주요 지표

### ROC/AUC Curve
ROC는 얼마나 분류가 잘 되었는가를 보여주는 그래프
-> 어떻게 분류를 하여도, 만족할 수 없는 부분이 생기기 때문에 완벽한 분류기는 나오기 힘듦

AUC Curve
ROC와 X 축 사이의 면적
모델의 성능을 숫자로 나타낼 수 있으며, 0.5~1 사이의 값을 가지며 1에 가까울수록 분류 성능이 좋은 것

## 하이퍼 파라미터
학습 시작 전에 사용자가 직접 설정하는 변수로 모델 학습 과정에 반영됨

- 하이퍼 파라미터 최적화란 Tuning을 거쳐 적절한 하이퍼 파라미터를 찾음으로써 모델 성능을 향상 시키는 것

(과정)
1. 하이퍼파라미터 탐색 범위 설정(최적값을 찾고 싶은 하이퍼파라미터의 범위를 설정)
2. 평가지표 계산함수 정의 (탐색하려는 하이퍼파라미터를 인수로 받아 평가지표값을 계산해주는 함수를 정의)
3. 1단계에서 샘플링한 하이퍼파라미터 값을 사용하여 검증 데이터로 정확도 평가
4. 위 내용을 특정횟수 반복하며, 정확도 결과를 보고 하이퍼파라미터의 범위 좁힘

(방법)
- Grid Search: 정해진 범위에서 하이퍼파라미터를 모두 순회
(장점): 범위가 넓고 스텝이 작을수록 꼼꼼하게 전 범위를 탐색하니 최적해를 정확이 찾음
(단점): 시간이 너무 오래 걸린다
(적용): 넓은 범위, 큰 step을 활용해 범위를 좁힘

- Random Search: 정해진 범위에서 하이퍼파라미터를 무작위로 탐색
(장점): 속도가 Grid보다 빠름
(단점): 무작위라는 한계 때문에 정확도가 떨어짐

- Bayesian Optimization: 사전 정보를 바탕으로 하이퍼 파라미터 값을 확률적으로 추정하며 탐색
(특징): 통계한 기반으로 가장 큰값이 나올 확률이 높은 지점을 찾아냄
