# 비지도학습
정답이 없는 데이터를 사용하여 데이터 속의 패턴 또는 각 데이터 간의 유사도를 기계가 학습하도록 하는 것 (군집화, 연관 분석 등에 사용)

## 특징
- 비지도 학습은 데이터 자체에 내제된 구조를 파악해서 학습함
- 정답 레이블이 없으므로 에이전트가 해야 하는 작업이 명확히 정의 X -> 모델의 성능을 명확히 측정할 수 없음
- 표현 학습(or 피쳐학습)을 통해 데어터 셋의 고유 패턴을 식별 O
--> 목표가 명확하게 정의된 구체적이고 좁은 범위의 문제는 지도학습보다 불리하나 목표 자체가 명확X, 데이터에서 새로운 특성을 찾는 등의 목적을 수행하기에는 적합

## 비지도학습이 적합한 케이스
1. 패턴이 아직 알려지지 않은 경우 -> 정답 모름
2. 패턴이 계속해서 변하는 경우 -> 정답을 익혀도 의미 X
3. 열린 문제를 해결하고 지식을 일반화해야 하는 경우

ex. 
1. 고객 세그먼트 분석 (성향 데이터를 바탕으로 비슷한 성향의 고객을 그룹 단위로 묶어 분석)
2. 이상 거래 탐지 (정상적인 거래 패턴을 군집으로 정의, 이외의 데이터는 이상치로 간주)
3. 자연어 처리 (텍스트 데이터를 군집화해 분석하여, 인사이트 도출)

## 군집화
데이터를 비슷한 특성을 가진 그룹으로 나누는 비지도 학습 기법
(참고) 군집: 유사한 데이터들의 집합으로 이때, 서로 다른 군집에 속한 개체들은 비슷하지 X

### 목표
유사성을 기반으로 비슷한 특징을 지닌 개체를 함께 그룹화하며 두 개의 목표를 가짐
1. 응집도 최대화: 같은 군집에 속하는 데이터끼리는 최대한 비슷하도록 함
2. 분리도 최대화: 서로 다른 군집은 최대한 분리되도록 함
--> 데이터 간의 유사성을 최대한 잘 유지하며, 서로 다른 그룹은 구분될 수 있도록 
--> 유유상종을 잘 묶어줘야 함

### 과정
1. 피처 선택 또는 추출: 군집화에 사용할 데이터의 피처를 선택
2. 군집화 알고리즘 선택: 데이터의 특성과 목표에 따라 적합한 군집화
3. 군집 유효성 검증: 군집화가 얼마나 잘 되었는지 평가
4. 결과 해석: 각 군집이 지닌 특성 분석 및 해석

### 고려사항
1. 변수 유형 이해
- 군집화에 사용될 피처의 종류와 특성을 명확하게 이해
- 변수 종류가 연속형, 명목형인지, 변수 개수에 따라 적합한 군집화 알고리즘 달라짐

2. 거리/ 유사도 정의와 특정
- 군집화는 데이터간의 거리 또는 유사도를 기반으로 그룹 형성
- 거리 측정 방법에 대한 정의가 중요(변수 유형에 따라 적합한 거리 측정 방식의 차이가 존재)

3. 차원 축소
- 모델의 성능을 향상시키기 위해 유사한 변수들을 묶어 처리하는 차원 축소
- 변수가 많아질수록 모델이 복잡해지고 효율성과 성능이 떨어짐
- 차원 축소를 통한 유사한 변수를 통합 or 필요없는 변수를 제거하여 변수 개수 줄임

### 변수 유형 이해
1. 연속형 변수
- 수치로 측정 가능한 변수
- 거리 측정 (유클리디안 거리, 맨하탄 거리 등)
- 적합한 알고리즘: K-means, 계층적 군집화 등
- 스케일링이 필수로 단위가 다르면 큰 값의 변수가 지배적 영향

2. 명목형 변수
- 특징: 범주로 구분되는 변수
- 거리 측정: 해밍 거리, 자카드 거리 등
- 적합한 알고리즘: L-modes, 계층적 군집화(특변한 거리 함수)
- 처리방법: 원-핫 인코딩 또는 더미 변수 변환

3. 혼합형 변수
- 특징: 연속형과 명목형이 섞인 데이터
- 적합한 알고리즘: K-prototypes, Gower distance 기반 계층적 군집화
- 처리 방법: 각 변수 유형에 맞는 거리 함수를 조합하여 사용


### 거리/유사도 정의와 측정
데이터의 거리 측정 방법: 군집화에서 데이터 포인트나 군집간의 유사성/비유사성을 수치로 계산하는 방법

1. 유클리디안 거리
$d(\mathbf{x}, \mathbf{y}) = \sqrt{\sum_{i=1}^n (x_i - y_i)^2}$
- 가장 일ㅇ반적인 거리 측정 방법
- 직선거리를 의미하며, K-means에서 기본적으로 사용

2. 맨하탄 거리
$d(\mathbf{x}, \mathbf{y}) = \sum_{i=1}^n |x_i - y_i|$$
- 각 차원의 차이가 절댓값의 함
- 이상치에 덜 민감한 특징

3. 코사인 유사도
$\text{similarity} = \frac{\mathbf{x} \cdot \mathbf{y}}{||\mathbf{x}|| \times ||\mathbf{y}||}$
- 벡터간의 각도를 측정'
- 텍스트 데이터나 고차원 데이터에 효과적

### 6가지 연결 방법
계층적 군집화에서 두 군집 간의 거리를 계산하는 다양한 방식, 각 방법은 군집 간 거리 측정 기준이 달라 서로 다른 군집화 결과를 만듦

1. Single linkage (단일 연결법, 최단 연결법)
군집 A와 B 사이의 가장 가까운 두 점 간 거리로 군집간 거리를 정의
- 서로 다른 군집에서 가장 가까운 두 점 사이의 거리를 군집간의 거리로 측정하는 방법
- 고립된 군집을 찾는데 중점
- 이상치에 취약

2. Complete linkage (완전 연결법, 최장 연결법)
군집 A, B 사이의 가장 먼 두 점 간 거리로 군집 간 거리 정의
- 군집에서 하나씩 관측값을 뽑았을 때 나타날 수 있는 거리의 최댓값으로 측정하여 가장 유사성이 큰 군집으로 병행
- 군집들의 내부 응집성에 중점을 둔 방법
- 최단 연결법과 같이 이상치에 민감

3. Average linkage (평균 연결법)
군집 A, B 모든 점 쌍 사이 거리의 평균
- 모든 항목에 대한 거리 평균을 구함
- 최단 연결법과 최장 연결법보다 이상치에 덜 민감
- 계산량이 불필요하게 많아짐

4. Centroid method (중심 연결법)
각 군집의 중심점 간의 거리
- 두 군집의 중심간의 거리를 측정
- 두 군집이 결합될때 새로운 군집의 평균은 가중 평균을 통해 구함
- 군집들의 중심을 계산해야 하기에 시간이 오래 걸림

5. 중앙 연결법
두 군집의 중앙값 사이의 거리
- 합쳐지는 두 군집의 중심점 사이의 산술적 중간 지점을 새로운 대표값으로 설정하여 거리 측정
- 데이터의 중앙값에 기반해 군집 간 거리를 정의 -> 극단값에 영향 덜 받음
- 기하학적 구조를 파악하기는 어려운 단점

6. Ward's Procedure (Ward 연결법)
군집 내 제곱합이 증가하지 않도록 병합
- 군집의 병합 후 군집 내 SSE의 증가분이 최소인 것을 선택
- 보통 두 군집이 합해지면 병합된 군집의 SSE는 병합 이전 각 군집의 SSE의 합보다 커지게 되는데 그 증가량이 가장 작아지는 방향으로 군집을 형성해 나가는 방법

### 차원 축소
군집화 알고리즘은 거리 기반 알고리즘인 경우 多
거리 기반 알고리즘의 경우, 데이터의 차원이 높아질수록 차원의 저주 발생
- 차원의 저주: 피처 공간이 너무 커서 알고리즘이 데이터를 효과적/효율적으로 훈련할 수 없는 현상
-> 고차원일수록 모든 점이 비슷하게 멀어져, 거리 기반 알고리즘의 분별력 감소

- 차원 축소: 차원이란 변수의 개수이고 차원 축소란 단순히 말해 변수의 개수를 줄이는 것
-- 목표: 변수를 제거하여 차원을 낮춰, 고차원 데이터를 저차원 공간에 투영하면서도 핵심 정보는 가능한 지켜내도록 하는 것이 목표

- 선형 투영: 고차원 공간에서 저차원 공간으로 선형적으로 테이터를 투영
-- 주성분 분석 (PCA, 특잇값 분해, 랜덤 투영 등)

- 매니폴드 학습(비선형 차원 축소): 유클리드 거리가 아닌 데이터 포인트들 사이의 곡선 거리를 고려하여 학습
-- t-SNE, UMAP, lsomap 등

(그 중에서 우리는 PCA 사용)
dbPCA 핵심 목적: 다차원 데이터에서 가장 정보량(분산)이 큰 방향을 찾고, 그 방향을 기준으로 데이터를 투영하여 차원을 줄이는 것
- 데이터의 분산을 최대한 보존하면서 차원을 줄여 시각화와 분석을 용이하게
- 상관관계가 높은 피처들을 결합해 선형적인 상관관계가 없는, 더 적은 수의 피처로 데이터를 표현
- 원본의 고차원 데이터에서 최대 분산 방향을 찾아 상관관계를 감소시키고 저차원 공간에 이를 투영

## PCA 사용
1. 분산 비율 기준으로 선택
2. 고정 개수 선택

(주의 사항)
1. 표준화 필수: PCA 사용 전 반드시 데이터 표준화 수행
2. 결측값 처리: 결측값 있으면 사용 X
3. 해석의 어려움: 주성분은 원본 특성들의 조합이라 직관적 해석 어려움
4. 정보 손실: 차원 축소 과정에서 일부 정보 손실

# 군집화 알고리즘
## 계층적 군집화
데이터간의 유사성을 기반으로 트리 구조를 형성하며 상향, 하향식으로 군집을 형성해나가는 방법

- 계층적 군집화 알고리즘은 데이터셋의 관측치를 사용해 덴드로그램 만듦 
(덴드로그램이란 가까운 두 개체 또는 군집을 점차적으로 병합해가는 과정을 시각적으로 표현한 트리 구조)
- 군집의 개수를 사전에 설정하지 않고, 클러스터링이 종료된 후에 원하는 군집의 개수를 선택
- 트리 구조를 상향식으로 하면 응집형, 하향식으로 하면 분리형임
- 응집형 클러스터링: 먼저 각 샘플이 독립적인 클러스터가 됨 -> 각 샘플이 독립적인 클러스터가 되어 이후 하나의 클러스터만 남을 때까지 가장 가까운 클러스터를 병합
- 분리형 클러스터링: 전체 샘플을 포함하는 하나의 클러스터에서 시작해, 유사성이 낮은 데이터들을 더 작은 클러스터로 나눔

### 응집형 계층적 군집화
모든 데이터 포인트를 각각 하나의 클러스터로 간주하고 시작 -> 점차 가까운 클러스터를 병합해 나가는 방식
- 단일 연결(가장 가까운 두 샘플의 거리)
- 완전 연결(가장 먼 두 샘플의 거리)
1. 거리 행렬 계산: 모든 데이터 포인트 간의 거리를 계산하여 거리 행렬을 만듦
2. 단일 클러스터 시작: 각 데이터 포인트를 개별적인 클러스터로 초기화
3. 가장 가까운 클러스터 병합: 정의된 기준(가장 멀리 있는 샘플간의 거리)에 따라 가장 가까운 두 클러스터를 병합해 새로운 클러스터 생성
4. 거리 행렬 업데이트: 새로 생성된 클러스터와 나머지 클러스터들 간의 거리를 다시 계산해서 거리 행렬을 업데이트
5. 반복: 모든 데이터 포인트가 하나의 클러스터에 속할 때까지 3,4번 반복

### K-means
데이터를 정해진 개수(K)의 그룹으로 나누되, 각 그룹의 중심점과의 거리가 가장 가까운 데이터끼리 묶는 군집화 알고리즘
- 군집화에서 가장 일반적으로 사용되는 알고리즘
- 연속적인 특성을 갖는 데이터 샘플들 중에서 군집 중심점을 선택해 중심에 가장 가까운 데이터들을 묶어나가는 군집화 기법
1. 데이터 표본들 중에서 랜덤하게 K개의 중심점을 초기 클러스터 중심으로 선택
2. 각 표본들은 가장 가까운 중심점에 할당
3. 각 클러스터에 할당된 표본들의 데이터 평균을 계산하여 중심점을 이동
4. 클러스터 할당이 변하지 않거나 사용자가 지정한 허용 오차 또는 최대 반복 횟수에 도달할때까지 2,3번 반복

장점: 직관적, 구현이 쉬움, 대용량 데이터에도 적용 O
단점: 초기 centroid 값에 민감, 군집 수 결정이 어려움, 아웃라이어에 민감, 기하학적인 모양의 군집은 파악 어려움

k-means ++ : 초기 중심점들을 서로 멀리 떨어진 곳에서 위치시켜 더 일관되고 좋은 결과를 도출하는 개선된 초기화 방법
- k-means에서는 초기 중심점을 랜덤하게 할당
- k-means++: 초기 중심점들을 서로 멀리 떨어진 곳에 위치시켜 더 일관되고 좋은 결과 도출

### 엘보우 방법
클래스 내 SSE를 바탕으로 그래프를 활용해 최적 클러스터 개수 k를 추정하는 방법
- 왜곡이 빠르게 감소하는 지점의 k값을 찾는 것 -> 최적 클러스터의 개수로 추정
- 기울기가 급격하게 변하는 지점이 최적의 k값으로 이 지점 이후로는 k를 증가시켜도 SSE 감소 효과가 미미함
장점: 시각적으로 최적의 k값을 쉽게 파악, 구현이 간단
단점: 주관적인 판단, 모호한 경우 존재

### DBSCAN
밀도가 높은 지역의 데이터를 하나의 군집으로 묶고, 밀도 기준으로 만족하지 못하는 점은 군집에 포함시키지 않는 알고리즘
- 밀집도: 특정 거리 내에 존재해야 하는 데이터 포인트의 최소 개수로 정의

- 특징
1. 명시적으로 이상치를 지정하여 클러스터링에서 제외
2. 다른 알고리즘 대비 이상치로부터 자유로움
3. 어떤 데이터가 여러 군집의 특정 거리 내에 있으면 가장 밀도가 높은 군집과 그룹화

- 주요 단계
1. 각 샘플을 다음 조건에 따라 분류
- 핵심 샘플: 특정 반경  ε 안에 있는 이웃 샘플이 지정된 개수 이상인 경우
- 경계 샘플:  ε 이내에 Minpts 보다 이웃이 적지만 다른 핵심 샘플의 반경  ε안에 있는 경우
- 잡음 샘플: 떨거지
2. 클러스터 생성: 개별 핵심 샘플 또는  ε  이내에 있는 핵심 샘플을 연결한 그룹을 클러스터로 생성
3. 경계 샘플 할당: 경계 샘플을 해당 핵심 샘플의 클러스터에 할당

장점: 
- k-means처럼 클러스터 모양을 원형으로 가정하지 않고, 밀도에 따라 클러스터 할당 -> 임의의 기하학적 분표를 갖는 데이터도 처리
- 이상치에 민감하지 않음: 노이즈를 통해 아웃라이어 검출 가능
- k- means나 계층 군집과 달리 모든 샘플을 클러스터에 할당하지 않음

단점:
-  ε과 최소 샘플 설정에 많은 영향
-  ε을 조절하기 쉽지 않으며, 적절한 조절을 위해 데이터셋에 대한 도메인 지식 필요
- 연산량이 많아 속도 느림
- 다른 밀도 분포를 가진 데이터의 군집 분석을 잘 하지 못함
- 차원의 저주 영향

### HDBSCAN
계층적 DBSCAN으로 계층적 군집화처럼 변화한 버전

작동 방식
1. 밀도 기반으로 1차 군집화
2. 거리 기준으로 밀도 기반 군집들을 반복적으로 연결

장점
- 여러 밀도 기준으로 반복 수행
- DBSCAN의 단일 밀도 기준 한계 보완


### GMM
데이터가 여러 다른 모양의 가우시안 분포로 구성되었다고 가정하고 각 분포를 클러스터로 인식하는 군집화 방법
-> 단일 분포로 표현하기 어려운 복잡한 형태의 확률 분포를 여러개의 가우시안 분포를 합쳐서 표현하자는 것

특징
- GMM은 모델 기반 군집화 방법
- 각 군집을 확률 분포로 간주
- 전체 데이터 분포를 여러 확률 분포의 혼합으로 모델링
- 데이터의 생성 메커니즘까지 모델링 가능


가정
- 관측된 데이터: 특정 가우시안 확률 분포에 의해 생성
- 전체 데이터셋: 여러개의 다변량 가우시안 분포가 섞여 있음
- 개별 데이터: 우도에 따라 K개의 가우시안 분포 중 하나에 속함
--> 섞인 데이터 분포에서 각각의 가우시안 분포를 추출하여 각각의 분포 기반으로 군집화 수행

진행과정
1. 주어진 전체 데이터셋의 분포 확인
2. 저너체 데이터 셋은 서로 다른 정규 분포 형태의 확률 분포 곡선으로 구성되어 있다고 가정
3. 전체 데이터 셋을 구성하는 여러 개의 정규 분포 곡선 추출, 개별 데이터가 이 주 어떤 정규 분포에 속하는지 결정

파라미터 추정
기댓값-최대화 알고리즘을 사용하여 모델 파라미터를 추정
- 각 군집의 중심, 모양, 크기, 방향

알고리즘 단계
- Initialization: 필요한 파라미터에 대해 초기 값을 선정
- E step: 현재 시그마를 통해 x가 특정 분포에 속할 사후 확률 계산
- M step: 계산된 사후 확률을 통해 다시 추정
- 수렴 조건이 만족할 때까지 반복

장점
- K-means보다 유연하게 다양한 데이터 셋에 잘 적용될 수 있음
- k-means가 잘 작동하지 않는 타원형이나 중첩된 군집 구조에서 좋은 성능

단점
- 군집화를 위한 수행시간 오래 걸리
- 가정한 분포(가우시안 분포)에 맞지 않는 데이터일 경우 계산 복잡도가 높아지고 성능이 저하


## 평가
### 외부 평가
- 데이터에 정답 레이블이 존재하는 경우 사용
- 군집화 결과와 실제 레이블(정답)을 비교하여 얼마나 유사한지 평가

### 내부 평가
- 정답 레이블이 없는 경우에 사용
- 군집 내부 응집도와 군집간 분리도를 기반으로 정량적으로 측정
- 각 알고리즘 자체의 내부평가 지표와 범용적으로 사용 가능한 내부 평가 지표 존재
- 알고리즘 별 내부 평가 지표와 목표
-- k-means: SSE를 최소화하기
-- GMM: 로드 우도(모델이 주어진 데이터를 얼마나 잘 설명하는지 나타내는 지표) 최대화

### 실루엣 지수
클러스터 내 샘플들이 얼마나 조밀하게 모여있는지를 측정하는 도구
$s^{(i)} = \frac{b^{(i)} - a^{(i)}}{\max \left\{ b^{(i)}, a^{(i)} \right\}}$

- 해석
1. 실루엣 계수는 -1~1 사이의 값 가짐
2. 분리도가 응집도보다 큰 경우 이상적인 실루엣 계수인 1에 가까워짐
- 샘플이 자신의 클러스터에 잘 속해있고, 다른 클러스터와는 멀리 떨어져 있음
- 군집화가 잘 되었다고 평가

3. 분리도와 응집도가 같을 경우 실루엣 계수는 0
- 샘플이 클러스터의 경계에 위치하여 군집 간 거리가 모호

4. 분리도가 응집도보다 작은 경우 실루엣 계수는 -1에 가까워짐
- 샘플이 잘못된 클러스터에 할당
--> 군집화가 잘 되지 않았다고 평가

### Dunn index
클러스터 간 최소 거리와 클러스터 내 최대 거리의 비율을 계산해 클러스터링의 품질을 평가하는 지표
$\text{Dunn Index} = \frac{
\min\limits_{i \ne j} \; d(C_i, C_j)}{\max\limits_{1 \le k \le K} \; \delta(C_k)}$

- 해석
1. 값이 클수록 좋은 군집화 결과를 의미
=> 군집간 거리가 크로(분리도가 높음), 군집 내부 거리는 작을수록(응집도가 높음) 좋은 군집화
2. 일반적으로 0 이상의 값을 가지며, 무한대까지 가능
3. 실루엣 계수와 종합적 평가 권장

한계
- 군집 수가 많을수록 계싼 비용이 증가
- 극단적인 값(가장 멀리 떨어진 포인트, 가장 가까운 군집 등)에 큰 영향
- 이상치에 민감하게 반응
