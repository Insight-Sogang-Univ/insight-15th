# 연관 분석
## 정의
- 상품이나 서비스를 구매하는 등 일련의 거래나 사건 안에 존재하는 항목 간의 일정한 연관 규칙을 발견하는 분석
ex. OTT 서비스, 금융 기관, 의료 분야, 요식 업계 등

## 분석
'A를 구매했을 때 B도 구매할 것이다'와 같이 if-then의 형식을 띰
- 조건절: 만일 ~~라면
- 결과절: 그 뒷부분에 해당하는 내용
- 아이템 집합: 조건절 또는 결과절을 구성하는 아이템들의 집합
--> 아이템의 집합은 여러개가 들어가도 되나 상호 배반이어야 함

(주의)
연관 규칙으로 상관관계는 파악할 수 있으나 인과관계는 파악할 수 없음
ex. A를 산 사람중 80%가 B도 샀다
1. 실험이 아닌 관찰 데이터이기에 
- 'A를 사게 한 원인'이 통제되지 않음
- A,B 사이에 교란 변수가 존재할 수 있음

2. 방향성은 그냥 수학적 비율임
- A가 원인이다가 아니라 A를 샀을 떄 B를 산 비율이 높은 것

### 협업 필터링 VS 연관 분석
- 협업 필터링은 상품마다 타 상품과 1:1 계산하지만 연관 분석은 상품마다 주문 내역을 1:n 계산
- 협업 필터링은 개인화된 상품 추천에 자주 활용
- 연관 분석은 상품 노출 또는 세트 상품 구성에 많이 활용

## 조건
1. 두 품목이 함께 구매한 경우의 수가 일정 수준 이상이어야 (지지도)
- 엄밀하게는 조건절이 발생하는 확률임 (전체 거래 항목 중 A가 구매되는 거래의 비율)
- 다만 알고리즘에서는 교집합을 사용
if. 지지도가 낮다면
- 우연일 수 잇는 규칙 포함 -> 몇 번밖에 등장하지 않는 규칙인데 우연히 신뢰도가 높을 수 있음
- 실행가능한 인사이트 도출 어려움

2. A를 포함하는 거래 중 B를 구입하는 경우의 수가 일정 수준 이상이어야 한다(신뢰도)
if. 신뢰도가 낮다면
- 의미 없는 연관 규칙 -> A를 샀다고 해도 B는 잘 안사는 경우
- 추천 실패로 이어짐 -> 신뢰도가 낮은 규칙을 기반으로 추천하거나 마케팅 타게팅하면 전환율 낮음

## 평가 측도
1. 향상도: 품목 A가 주어지지 않았을 때의 품목 B의 확률 대비 품목 A가 주어졌을 때 품목 B의 확률의 증가 비율
$향상도 = \frac{P(A \cap B)}{P(A)P(B)} = \frac{신뢰도}{P(B)}$
- 향상도 >1 = 우연적 기회보다 높은 확률
- 향상도 =1 = 독립
- 향상도 <1 = 우연적 기회보다 낮은 확률


## 연관 분석 알고리즘
### Apriori
- 개념: 상위 조합에서부터 차례로 스캔하면서 특정 조합이 자주 발생하지 않는다면 이의 결과물로 탄생한 후속 조합들까지 모두 후보에서 배제하는 방식
- 장점: 하나의 조합만 검사하고도 이에서 파생된 다른 조합들까지 후보에서 배제할 수 잇음
-> 집합 (A,B)의 지지도가 사용자가 정한 최소 지지도를 넘지 못하면 (A,B,C) 등 하위 규칙들을 전부 제거

- 프로세스
1. 단일 항목 집합 생성
- 모든 개별 상품에 대해 지지도를 계산
- 최소 지지도를 넘는 항목들만 다음 단계로

2. 2개 항목 집합 생성
- 1단계에서 선별된 상품들로 2개 항목의 조합을 만듦
- 이 조합에 대해 지지도를 계산하고, 최소 지지도를 넘지 않는 조합을 제거

3. 더 큰 항목 집합 생성
- 이전 단계에서 남은 조합들을 바탕으로 3개 이상의 항목으로 구성된 조합 생성
- 지지도를 계산하고 최소 지지도를 넘지 않는 조합 제거
- 더 이상 조합을 생성할 수 없을 때까지 반복

4. 최종 빈발 항목 선정
- 마지막으로 남은 빈발 항목 집합들을 바탕으로 연관 규칙 생성
- 각 규칙에 대해 지지도, 신뢰도, 향상도를 계산하여 가장 유용한 규칙 선택

- 장점
1. 수많은 상품 연관 구매 패턴 발견
2. 원리가 간단하고 이해 분석이 용이

- 단점
1. 여전히 데이터가 커질수록 속도가 느려짐

### FP Growth 알고리즘
- 배경: 여전히 느리다
- 개괄
1. FP-Growth 알고리즘은 빈발 품목 집합을 효율적으로 찾아냄
2. FP-Tree라는 구조를 이용하여 Apiori를 효과적으로 구현
3. 모든 거래를 확인해 각 아이템마다 지지도를 계산하고, 최소 지지도 이상의 아이템만 선택
3.1 모든 거래에서 빈도가 높은 아이템 순서대로 순서를 정렬
3.2 부모 노드를 중심으로 거래를 자식 노드로 추가하며 tree 생성
3.3 새로운 아이템이 나올 경우 부모 노드부터 시작하고 그렇지 않으면 기존의 노드에서 확장
3.4 위의 과정을 모든 거래에 대해 반복하며 FP Tree를 만들고 최소 지지도 이상의 패턴만 추출

- 프로세스
1. 데이터 셋 확인
2. 빈도수 확인: 남은 항목들을 각 거래별로 빈도순으로 재정렬하고 거래 데이터에서 가장 자주 구매된 항목이 먼저 오도록 재정렬
3. FP-트리 생성
- 이 트리는 각 거래를 노드로 표현하며 동일한 항목은 기존 노드에 연결해 빈도를 증가 시키고 새로운 항목은 새로운 노드로 추가
4. 빈반항목 집합 추출
- 특정 항목을 기준으로 부분 트리를 만들어 빈발 항목 집합을 추출 -> 기준이 된 항목과 연관된 빈발 항목 집합을 빠르게 도출
5. 연관 규칙 도출
- 각 규칙에 대해 지지도, 신뢰도, 향상도 등의 지표를 계산해 가장 유용한 규칙을 선택

- 장점
1. 데이터를 두 번만 스캔하는 트리 구조이기에 Apriori보다 훨씬 빠름
2. 후보 아이템 셋을 생성할 필요 없이 트리만 구성하면 끝

- 단점
1. 대용량 데이터 셋에서 메로리를 효율적으로 사용하지 않는다
2. Apriori에 비해 설계가 어렵고 지지도의 계산은 무조건 FP-Tree가 만들어져야 가능 


# 문제 정의와 구조화
## 문제 정의
문제란 현재와 바라는 상태의 격차
- 단, 현재의 안좋은 상황을 나열하는 것은 문제가 아니라 경영자가 '의사 결정'을 내려야 하는 구체적이 질문으로 치환되었을 때 문제가 됨

### framework
문제 정의: SMART
1. Specific: 구체적이고 범위가 명확함, 무엇을 어디까지 풀지 명확하게 정의
2. Measurable: 수치로 측정 가능함, 성공 여부를 숫자로 판단 가능
3. Action-oriented: 실행 지향적/ 결과 중심적, 행동을 이끌어내고 성과에 집중
4. Relevant: 관련성, 조직의 목표와 관련이 있어야 함
5. Time-bound: 시한성, 마감 기한과 의사결정 시점이 존재

문제 쪼개기: MECE
1. 상호 배타적: 서로 겹치는 부분이 없어야 함
2. 전체 포괄적: 누락된 부분이 없어야 함

--> MECE는 정말 좋은 말이다. 그러나 도달점이라기 보다는 지향점이라고 생각한다. 내가 아무리 열심히 짜도 빠진 부분이 있을 수도, 과대-과소 평가하는 부분이 있을 수도 있다. 그렇기에 LISS 방법을 쓰는 것이 현실적이라 생각한다

LISS: 겹치는 부분은 없고 전체를 다 보지는 않지만 중요한 부분은 다 봤다라는 의미

## 지표
목적이나 기준을 나타내는 표지로 추상적인 개념과 관련되는 지수, 척도 등을 말하며 측정하고 싶은 것을 숫자로 표현하는 것
(아래와 같은 지표들이 존재할 수 있음)
1. 효율성
2. 효과성
3. 수익성
4. 성장성
5. 안정성

### 지표 정의
분모, 분자에 절대적인 수치가 들어갈 수 없다면 어떻게?
- Proxy Metric 사용
-> 추상적이어서 프로덕트마다 나름의 가정과 정의, 논의가 필요한 지표를 말함
--> 궁금하지만 직접 측정할 수 없으니 대체물을 정하여 측정하는 의미로 정답은 존재 X

### 북극성 지표 L1/L2 Metric
- 북극성 지표: 기업이 가장 중요하게 생각하는 지표
- L1/L2 지표
1. L1: 전사적인, 고수준의 성과 지표 (전략 관점)
2. L2: L1 지표의 세부 사항을 측정하는 지표 (전술 관점)

### AARRR
프로덕트가 가치를 제공하는 방식이 다르면, 측정하는 지표도 달라져야 함
1. Acquisition
사용자 획득, 즉 사용자가 여러 채널을 통해 앱 혹은 웹사이트에 방문하는 것을 의미
-> 신규 고객을 어떻게 획득할 것인지, 어디서 획득할지, 주로 어떤 채널을 통해 획득할 것인지 등

2. Activation
사용자가 앱 또는 웹사이트에 첫 방문하여 주요 기능을 직접 사용하는 것을 의미
- sign-up: 유저가 제품에 들어온 시점
- setup moment: 유저가 제품의 핵심가치를 경험할 준비 완료
- A-HA: moment: 유저가 핵심 가치를 처음 경험한 순간
- habit moment: 유저가 제품 사용에 대한 습관 형성

3. Retention
신규 사용자를 장기 사용자로 전환 시키는 것: 장기 고객 양성 방법, 고객의 지속적인 구매 행위를 증가시킬 수 있는 방법

4. Revenue
제품 혹은 서비스를 유료 구매하기로 결정한 단계 -> 사용자의 소비를 유도해 매출을 높이는데 총력을 기울여야 하고, 최대한 사용자를 소비하게 만드는 방법

-> 측정 지표도 제품 성격에 따라 달라짐
ex. 게임이면 유료화 상품 판매 중심의 수익구조로 사용자가 구매한 게임 아이템, 스킨의 소비액 등이 주요 측정 지표가 됨

5. Referral
제품에 만족한 고객이 타인에게 추천하는 것. 충성도 고객을 기준으로 타 고객에게 퍼뜨리도록 하는 것

### 지표 분석 방법
#### 코호트 분석
1. 정의 
정의된 시간 범위 내에서 공통된 특성이나 경험을 공유하는 것을 코호트라고 하며 특정 기간에 특정 특성을 공유하는 사용자 집단을 의미
- 분석 전에 데이터 세트의 데이터를 관련 그룹으로 나누는 일종의 행동 분석
- 시간 (날짜)에 기반하는 사용자 세그먼트를 분류
- 주로 사용자 유지와 이탈을 체크하는데 사용

2. 해석
- 세로축: 같은 기간이 지난 후 코호트 간 리텐션을 비교
- 대각선: 시점별 비교 (전체적인 특성을 비교)
- 가로축: 시가느이 경과에 따른 변화 관찰

#### 퍼널 분석
1. 정의
유저가 접속해서 떠날 때까지 유저의 행동을 추적: 아래로 갈수록 점점 좁아짐
2. 개선 방법 예시
- 유입률 높이기
- 전환율 높이기
- 퍼널 단계 없애기
- 퍼널 순서 바꾸기
- 퍼널 단계 추가하기
- 코호트로 나누어 보기
