# 통계

> 들어가며...  
- **기술통계**: 표본 자체에 대한 관심/ 표본을 요약하여 표현  
- **추론통계**: 표본의 모집단에 대한 관심/ 표본으로부터 모집단을 추론

> ## 자료의 척도
**질적척도**  
- 명목척도: *레이블*이나 *이름*으로 요소의 속성을 식별  
- 서열척도: *순서* 또는 *순위*를 나타내는 특성

**양적 자료**
- 구간척도: 값들 사이에 *순서* 존재, 간격이 일정하여 덧셈이나 평균 등의 연산 가능/ *곱셈과 나눗셈은 의미 없음*<br>
- 비율척도: 0값이 아무것도 존재하지 않는다는 *의미존재*/ 절대영점과 임의의 측정단위로 구성, 동간성 유지


> ## 기술통계<br>
: 데이터를 있는 그대로 설명/이해하는데 중점  

### 3가지 주요한 도구 집합  
1. **중심화 경향**(평균, 중앙값, 최빈값...)  
  - 데이터의 중심적인 경향 나타냄
    - **평균**
      - (데이터 값의 총합/데이터의 개수)  
      - 중심 경향성의 대표 지표  
      - 총합 사용-> *이상치*에 민감  
       cf) 절사평균
         - 이상치에 대한 민감성 보완
         - 데이터의 양끝을 일정 비율로 절삭 후 평균 계산  
    <br>
    - **중앙값** 
      - 정렬 시 한가운데에 위치하는 값  
      - 이상치에 대한 민감성 적음  
      - 중심 경향성 나타내는 *대안적인* 지표  
     <br>
    - **최빈값**  
      - 데이터에서 가장 자주 나타나는 값  
      - 이산형 데이터 분포의 이해에 도움
     <br>
2. **자료의 산포**(분산, 표편, 범위...)  
 - 데이터가 퍼져있는 정도 설명  
   - **범위**  
    - (최대값-최소값): 데이터의 전체 변동 범위 나타냄  
    - 이상치에 매우 민감(양극단값 이용하므로)  
    <br>
   - **분산**  
      - 데이터 값들이 평균으로부터 얼마나 멀리 퍼져있는지 나타냄  
      - 분산 클수록 데이터가 퍼진 정도 큼  
    <br>
    - **표준편차**  
       - 분산에 *제곱근*을 취한 값  
       - 제곱근을 통한 *단위 일치*  
        - **변동계수**: 상대적인 변동성 나타내는 통계적 지표   
    <br>
    cf) **첨도와 왜도**  
    - **첨도**: 확률분포의 *꼬리*가 두꺼운 정도  
    - **왜도**: *비대칭도*, 실수 값 확률 변수의 확률 분포 비대칭성을 나타내는 지표  
    <br>
3. **자료의 정리 및 시각화**(히스토그램, 박스플롯, 파이차트...)

> ## 확률과 분포  
### 추론통계학의 의문점  
- 표본의 불확실성 -> 표본분포  
- 모수 추정 -> 점추정  
- 추정의 현실성 -> 구간추정  
- 구간추정의 오류 관리 -> 신뢰수준  
- 적극적 주장 검증 -> 가설검정  
- 가설검정의 오류 관리 -> 유의수준   
<br>

### 확률분포  
* **확률변수**: 무작위 세계에서 어떤 특정 사건이 발생할 때 나타날 수 있는 모든 경우의 수를 특정 값으로 대응시키는 함수  
<br>

- **확률분포**: 확률변수가 특정 값을 가질 확률을 나타내는 함수  
  (1) **이산확률분포**: 확률변수가 <u>이산적인</u>값을 가지는 경우  
    - 분포의 종류: 베르누이 분포, 이항분포, 기하분포, 음이항분포, 초기하분포, 포아송분포  
  (2) **연속확률분포**: 확률변수가 <u>연속적인</u>값을 가지는 경우  
    - 분포의 종류: 정규분포, T 분포, F분포, 감마분포, 지수분포, 카이제곱분포, 균등분포  

- **대표적인 분포 모형**  
  -> 정규분포, Z 분포, T분포, 카이제곱분포, F분포


> ## 추론통계- 추정  
- **추론통계란?**  
: 수집한 데이터를 기반으로 전체 정보를 추론하고 결론 도출하는 방법(ex. 통계적 가설검정, 신뢰구간)  
  - **추정**  
    -> 모집단의 일부를 표본으로 추출하여 분석한 통계량을 이용하여 집단의 *모수*를 예측하는 과정  
     (1) **점추정**: 추정하고자 하는 모수를 <u>하나의 수치</u>로 추정  
     (2) **구간추정**: 추정하고자 하는 모가 존재하리라 <u>예상하는 구간</u>을 정하여 추정  
       -> *일정한 범위*와 *신뢰도*를 가지고 표현
    <br>

  - **검정**  
  : 표본을 추출하여 얻은 표본통계량으로 모딥단의 모수에 댛ㄴ 새로운 예상/ 가설 등이 *옳다고 판단할 수 있는지*를 평가하는 과정  


> ## 추론통계- 검정  
### 가설 검정의 절차  
-> 가설 설정 - 유의 수준 설정 - 검정통계량 산출 - 가설 기각/채택 판단  
(1) **가설 설정**  
  - H0(귀무가설): 기각하고자 하는 명제  
  - H1(대립가설): 주장하고자 하는 명제 
  <br> 

(2) **유의 수준 설정**  
: 결과 판단에서 허용할 수 있는 <u>오류의 **최대치**</u>  
- 주로 0.05로 설정  
  - **1종 오류**: 참인 귀무가설을 기각할 확률  
  - **2종 오류**: 거짓을 귀무가설을 기각하지 못할 확률  
  -> 1종 오류가 더 치명적
  -> 유의수준: **1종 오류**의 확률 의미  
<br>

(3) **검정통계량 산출**  
**검정통계량**: 가설 검정의 결과를 판단하는 데 사용되는 값(Z값, T값, F값)  
<br>

(4) **가설 기각/채택 판단**  
  - **유의확률**  
    - 유의확률이 유의수준보다 **작을 시** 귀무가설 기각  
    -> 클 경우에는 귀무가설을 채택하는 것이 아닌, 기각 실패  
    <br>


### 데이터의 자료형에 따른 검정 방법
(1)**[수치형-수치형](피어슨) 상관분석**  
- **피어슨 상관 분석**: 두 변수 간의 피어슨 상관계수가 얼마나 *유의미한지* 검정하는 방법  
  -> H0: 두 변수 간 선형 상관관계 존재하지 않음  
  -> H1: 두 변수 간 선형 상관관계 존재  
  <br>

(2)**[범주형-수치형]F-test(ANOVA), T-test**  
- **T-test**: 범주가 *2개*일 때  
  -> H0: 두 그룹의 평균에 차이 없음  
  -> Ha: 두 그룹의 평균에 차이 있음  
  <br>

- **ANOVA test**: 범주가 *3개* 이상일 때  
  - F값: (집단 간 분산/집단 내 분산)  
<br>

(3) **[범주형-범주형]CHi2-test**  
- **자유도의 약 2배**가 기준량  
- 범주 개수에 제한이 없음  
- 범주형 변수의 자유도는 <u>범주의 수 -1</u>
