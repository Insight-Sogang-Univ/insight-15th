> # 회귀 기초

## **머신러닝과 모델링**  
<br>

### **머신러닝**  
: 모델이라는 소프트웨어를 학습하여 *새로운 데이터*를 예측 or 결정 내릴 수 있도록 하는 기술  
* *패턴* 발견
* 새로운 데이터에 대해 *예측*  
<br>

> **머신러닝의 종류**  

**1. 지도학습**  
  - 정답이 <u>있는</u> 데이터 사용  
  - (기계의 예측값 = 데이터의 정답) 되도록 학습  
  - 데이터가 반드시 **labeled data**(독립변수, 종속변수 모두 존재) 형태여야 함  
  - 예시) 회귀, 분류  
<br>

    - **회귀**: x,y에 대한 일차함수를 찾는 문제  
    - **분류**: 기존 데이터가 어떤 레이블에 속하는지에 대한 패턴 알고리즘으로 인지 -> 새로운 데이터의 *레이블* 판별 
    cf) 종속변수가 <u>연속형</u>이면 회귀, <u>범주형</u>이면 분류 사용
  <br>

**2. 비지도 학습**  
  - 정답이 <u>없는</u> 데이터 사용  
  - 데이터의 **패턴** 혹은 **유사성** 학습  
  - 예시) 군집화, 밀도 추정, 차원 축소
  <br>

**3. 강화 학습**  
  - 정답 여부: 보상  
  - 시행착오 **반복**, <u>보상을 최대화</u> 하는 행동 학습  
  - 예시) 알파고, 자율 주행  
<br>

### **모델링**  
* **모델** 이란?  
: 학습의 결과물(또는 학습을 마친 인공지능 그 자체)  
  - 패턴이 저장된 상태/ 입력된 새로운 데이터에 대한 정답 출력(예측기 역할)  
  <br>

* **모델링이란?**  
: *예측기*를 만들어가는 전체적인 과정  
<br>

> **모델링의 과정**  
   1. **설계**: 데이터의 특성에 맟춰 어떤 알고리즘으로 학습할지 결정  
   2. **훈련**: 데이터 제시 -> 정답에 가까워지도록 수정 
   3. **검증**: 훈련된 모델이 새로운 데이터에서도 정답을 잘 맞히는지 확인 



## **선형회귀**  
: 데이터를 가장 잘 설명하는 **회귀선**을 찾는 과정  
- 현재의 데이터를 <u>요약</u> + 미래의 값을 <u>예측</u>  
<br>

* **가중치**: 종속변수에 대한 독립변수의 *영향력의 크기*  
  - 그래프의 <u>기울기</u>로 나타남 
  - (=회귀계수)  
  <br>

* **편향**: 독립변수가 0일 때 나오는 기본값(y절편)  
  - 모델의 최적화를 위해 위아래로 움직여주는 역할  
  <br>

> **회귀의 종류**  
* **단순선형회귀**  
: <u>하나</u>의 독립변수만 존재  
  - 데이터를 *직선* 형태로 표현  
  - 최적의 일차함수 찾는 과정  
  - 잔차를 줄여가며 적절한 *회귀계수*와 *편향* 찾음  
  <br>
  cf) **잔차**: |실제값 - 예측값|  
  <br> 

    * **1) 최소제곱법**  
    : *잔차의 제곱의 합*이 <u>최소</u> 되는 지점으로 최적의 회귀선 계산  
    <br>

    * **2) 경사하강법**  
      - <u>최솟값을 구하기 어려울 때</u> 사용  
      - 손실함수 상에서 현재 값의 접선의 기울기 확인 -> 음수인 경우 **"현재 값 - 미분 값(음수)** -> 오른쪽으로 점이 이동  
      - 양수인 경우 **"현재 값 - 미분 값(양수)** -> 왼쪽으로 점이 이동  
      - 최솟값에 도달한다면 변화 중지  
<br>

* **다중선형회귀**  
: 독립 변수가 <u>두 개 이상</u>인 경우  
  - 잔차: 관측치와 *평면* 간 차이   
  <br>

  * **다중공선성**  
  : <u>독립 변수들 간의 </u> 강한 상관관계로 인해 회귀 모델이 불안정해지는 문제  
    - 심해지면 각 변수의 회귀계수 추정이 부정확해지고, 해석력 감소  
    - 높은 경우 *어떤 독립 변수*가 종속 변수에 *얼마나* 영향을 미치는지 정확하게 구분 어려움  
    <br>

    * **확인 방법**  
       (1) 상관계수  
       (2) VIF 지수  
       <br>

    * **대처 방법**  
      1. 변수 제거  
      2. 변수 변환  
      3. 규제선형모델 활용  
      4. 주성분 분석  




## **규제선형모델**  
: 모델에 **과적합**되지 않도록 규제를 하라고자 등장 -> <u>계수의 크기</u> 조정  
  - 오류는 최대한 줄이면서도, <u>특정한 변수에 집착하지 않도록</u> 균형 -> 과적합 방지  
  <br>

> **규제선형모델의 종류**  
1. **L2 규제**: 회귀계수의 **제곱**에 패널티 부여/ *릿지회귀*  
2. **L1 규제**: 회귀계수의 **절댓값**에 패널티 부여/ *라쏘 회귀*  
3. **L1+L2 규제**: *엘라스틱넷* 회귀  

## **모델 평가방법**   
### **1. 성능 평가 지표**  
**1) 평균 제곱 오차**  
: 예측값과 실제값 사이의 *오차 제곱의 평균*  
<br>

**2) 평균 절대 오차**  
: 모델의 예측값과 실제값 사이의 *절대값 오차의 평균*  
<br>

**3) MSE와 MAE의 차이**  
- **MSE**: 큰 오차에 민감하게 반응(이상치에 민감)  
- **MAE**: 모든 오차 동일하게 취급(이창치 존재하더라도 전반적인 예측 성능 유지 가능) 

<br>

### **2. 변수 유의성 평가**  
**T 검정**  
: <u>독립변수의 회귀계수가 유의미한지</u> 검정
