> # **다중선형회귀의 기본 가정 6가지**  
## **가정 1- 선형성**  
 - 종속변수와 설명 변수 간의 관계가 <u>선형적</u>이어야 한다는 가정
 <br>

## **가정 2-독립성**  
: 각각의 설명변수가 서로 <u>**선형독립적**</u>이어야한다는 가정  
  - 선형독립적이지 않은 경우: **다중공선성**  
     - 다중공선성: 독립 변수들이 <u>서로 강한 상관관계</u>를 가져서, 어떤 변수가 종속 변수에 가장 큰 영향을 주는지 알 수 없는 상태(상관계수 or VIF 계산으로 알 수 있음) 
     -> 변수 제거, 규제 선형 모델, PCA 방법으로 대처 가능  
<br>

## **가정 3- 오차항의 평균은 0이다(0에 근사)**  
: 오차항의 평균은 0이거나 0에 가까워야 한다는 가정  
  - **오차항**: 실제 값과 예측 값 사이의 차이  
    cf) *백색잡음*: <u>특정 패턴 없이 랜덤하게</u> 변하는 신호
      - 랜덤신호인 백색 잡음은 평균적으로 어느 방향에 치우치지 X  
      - 오차항이 0에 근사하지 않은 경우 <u>데이터의 패턴을 올바르게 반영하지 못하고 있음</u>을 의미
<br>

## **가정 4- 등분산성**  
: 오차항의 분산이 *일정*해야한다는 가정  
  - 오차항의 분산이 일정하지 않은 경우 -> **이분산성** 존재  
     - 특히 *횡단면 자료*에서 많이 나타남  
     - 등분산성ㄴ 가정을 만족하지 않으면, 모델의 예측이 <u>일부 구간</u>에서는 매우 *정확*하지만, 일부 구간에서는 <u>오차가 매우 커질 수 있음</u>  
  - **등분산성을 알아보는 방법**  
     방법 1. **잔차의 도표화**  
     방법 2. **검정**  
     <br>

## **가정 5- 오차항은 자기상관되어 있지 않다**
: 오차항의 공분산은 항상 0이어야 된다는 가정  
  - 오차항의 공분산이 0이 아닌 경우: **자기상관** 존재  
     - **자기상관**: 한 변수의 <u>현재 값이 과거 값과 상관관계</u>를 가지는 현상  
     - 시계열 자료에서 빈번  
     - 모델이 *이전 데이터의 패턴을 학습*하여 <u>반복적인 오류</u> 만들어낼 가능성 높음  
     <br>

## **가정 6(선택)- 정규성**  
: 오차항이 **정규분포** 따른다는 가정  
  - 위배되어도 다중선형회귀분석 결과에 큰 영향 주지 않음  
  - if)  오차항이 정규성을 만족한다는 가정을 따르지 않으면, p-value가 잘못 계산 가능-> 결과 해석의 신뢰성 하락  
<br>

### 가정들의 검증 단계  
**회귀 분석 전 검증 불가능**  
- 가정3, 가정4(등분산성), (가정 5), 가정 6(정규성)  
- 기준: '**오차항에 대한 가정인가?**  
  - 모델 학습 전에는 오차항을 계산할 수 있는 근거 없음  
<br>
* 모든 가정의 충족  
  - 데이터가 30개  이상인 경우 -> *중심극한정리*에 의해 <u>정규성 위배</u> 용인  
  - 회귀 모델의 **목적**에 따라  
    - 목적이 *설명*인 경우: 가정 충족 중요  
    - 목적이 *예측*인 경우: 예측률을 높이는게 중요   
    



> # **회귀분석 평가 방법**  
## 1) 시각화  
- 객관적인 비교가 어려움  
- 객관적 비교를 위해서는 통계지표를 활용하는 것이 좋음  
<br>

## 2) 통계지표  
**(1) 모델 유의성 검정**  
* F-통계량(*전체모델* 대상 검정)  
: <u>회귀모델</u>이 유의미한지 검정하는 지표  
  - 값이 클수록 모델이 통계적으로 유의미 
  - Prob(F-통계량의 유의확률)과 0.05 비교  
<br>

*T-통계량 (*개별변수* 대상 검정)  
: <u>각 독립변수가 종속변수에 유의미한 영향을 미치는지</u> 판단하는 지표  
  - 유의수준과 비교  
<br>

**(2)모델의 성능 평가**  
* **결정계수**: 회귀 모델이 <u>종속 변수의 전체 변동성</u>을 얼마나 잘 설명하는지 나타내는 지표  
  - 0과 1 사이의 값 가짐  
  - 1에 가까울수록 SSR과 SST의 값이 비슷해져 모델이 데이터를 *잘 설명* 한다고 할 수 있음  
  - 결정계수가 0이면: <u>모델이 데이터를 전혀 설명하지 못한다</u>  
  <br>

* **조정된 결정계수**  
 - 독립변수의 개수가 증가하면 결정계수가 데이터와의 관련성과 무관하게 *증가*  
 - 변수 개수에 따른 설명력 추가 부분을 상쇄하는 효과 내도록 설정  
 <br>

* **AIC, BIC** 
- 정보 기준이라고 불림  
- 값이 <u>낮을수록</u> 좋음  


> # **비선형회귀**  
## **1. 다항식 회귀 모델**  
- 독립 변수에 *거듭제곱 항*을 추가하여 데이터의 <u>곡선적(비선형적) 특성</u> 모델링  
- 장점: 선형 모델에 비해 회귀선을 데이터에 fit하게 그릴 수 있음  
- 단점: 너무 많은 독립변수 -> overfitting 유발 가능  
-> *곡선을 그리는 데이터 패턴*에 유용/ **성장률, 감소율** 등이 시간에 따라 변화하는 속도가 다른 경우 유용  
<br>

## **2. 지수 회귀 모델**  
: 종속변수가 지수적으로 변화하는 관계 모델링할 때 사용 가능  
* **사용방법**  
  - 종속 변수 Y에 <u>로그</u>를 취한 후 선형회귀 적용  
  - 원래 데이터 형태로 *복구* 하기 위해 <u>예측 값에 지수함수 적용</u>  
  <br>

* **적합한 상황** 
  - 값이 시간에 따라 **지수적으로 증가하거나 감소**하는 경우에 적합  
  - 기술 발전, 투자 수익률 증가 등  
  <br>

## **3. 로그 회귀모델** 
: 종속 변수와 독립 변수 간의 관계가 로그 함수를 통해 더 잘 표현될 때 사용 가능  
* **사용방법**  
 1. 독립변수 X에 로그함수 적용 
 2. 선형 회귀를 적용  
 -> 예측값에 <u>별도의 처리를 하지 않고</u> 그래도 사용 
 <br>

* **적합한 상황** 
 - 데이터가 초기에 빠르게 증가하고 점차 증가율이 줄어드는 패턴 보일 때  
 - <u>초기 빠른 증가 후 점진적 감소</u> 보이는 데이터에 적합  
 <br>

## **4. 스플라인 회귀**  
: 데이터 *구간별*로 나누고 각 구간에서 다른 선형 또는 비선형 함수를 적용하여 예측하는 방법  
* **사용방법**  
 - 전체 데이터 범위를 여러 구간으로 분류  
 - 각 구간에 대해 별도의 회귀모델 적용  
 - 구간 경계에서의 **연속성** 유지  
 <br>

* **적합한 상황**  
 - 데이터 패턴이 <u>여러 구간에서 서로 다르게</u> 나타날 때 유용  
 - 게절에 따라 판매량이 변화하는 소매 데이터 분석 등에 적합
