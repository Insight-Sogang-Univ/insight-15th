> # **분류란?**  
## **(1) 지도 학습: 회귀 vs 분류** 
- **회귀**: 주어진 데이터를 기반으로 정답을 잘 맞추는 <u>함수</u>를 찾는 문제(연속적인 숫자값 예측)  
- **분류**: 기존 데이터의 레이블 소속 패턴을 인지한 뒤 새로운 데이터에 대한 <u>레이블</u> 판별(입력된 데이터를 주어진 항목들로 나눔)  
cf) 가장 큰 차이는 변수의 종류에 따라!! (연속-회귀, 범주-분류)  
<br>

## **(2) 이진 분류와 다중 분류**  
- **이진분류**: 예측하고자 하는 변수가 어떤 기준에 대해 <u>참 또는 거짓</u>의 값만 가짐  
- **다중분류**: 예측하고자 하는 변수가 가질 수 있는 값이 <u>3개 이상</u>일 때  
<br>

> # **분류 모델**  
## **(1) 로지스틱 회귀**  
: '*이진분류*'에 사용되는 대표적인 알고리즘  
- 샘플이 <u>특정 클래스에 속할 확률을 추정</u> -> 기준치에 따라 분류  
- 선형 회귀에 로지스틱 함수를 적용 -> 출력값 0~1 사이로 변환  
<br>

### <로지스틱 회귀의 필요성>  
* **분류에 직선 사용 시의 문제점**  
   - 입력값에 따라 변화율이 다른 것을 고려하는 시그모이드 함수에서 더 현실적인 확률 값 제공 가능  
   - 예측값의 범위가 <u>음의 무한대부터 양의 무한대</u>와 같이 범위를 벗어날 수 있음  
   -> 출력이 0~1 사이의 값 + **S**자 형태로 그려지는 함수 이용!  
   <br>

### <시그모이드 함수>  
: 출력이 0~1의 값이며 S자 형태로 그려지는 함수 
* **시그모이드 함수의 특징**  
  - 입력값이 커지면 1에 수렴, 작아지면 0에 수렴  
  - 입력값이 $(-\infin, \infin)$일 때, 출력값은 0~1의 값을 가짐
  <br>

* **시그모이드 함수의 가중치**  
1. 가중치 w  
  - 커질수록 가운데 부분의 기울기가 가파름  
2. 가중치 b  
  - 커질수록 같은 x 값에서 가지는 y 값이 증가  
  <br>

## **(2) 의사결정나무**  
: <u>조건에 따라 데이터 분류</u>하며, 최종적으로 데이터가 순수한 label의 집합으로 구성될 때까지 분류를 **반복**하는 분석 방법  
### **CART(이진분할)**  
  : 가장 널리 이용/ 데이터셋을 *임계값*을 기준으로 <u>두 child</u>로 나누는 알고리즘  
  -> 임계값은 **불순도**(지니 계수)가 <u>낮아지는</u> 방향으로 설정  
  cf) **불순도**: 분류하려는 데이터 집합에서 서로 다른 범주가 섞여있는 정도  
  <br>

**<CART 알고리즘의 주요 단계>**  
1. **임계값 설정**  
  - 임계값을 기준으로 데이터 분류  
2. 불순도 감소 알고리즘  
  - 어떤 기준의 분류가 가장 적절한지에 대한 고민  
  <br>

## **(3)SVM**  
: 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘  
(<u>데이터를 분리하는 초평면</u> 중에서 데이터들과 <u>가장 거리가 먼</u> 초평면을 선택하여 분리하는 지도 학습 기반의 *이진 선형 분류 모델*)  
  - 명확하게 분류할 수 있는 데이터 집단에서 뛰어난 성능  
  - 고차원 공간에서 효과적으로 사용 가능 
  <br>

### **장점** 
  - 서포트 벡터만을 이용한 결정 경계 생성 -> 데이터 적을 때 효과적  
  - 새로운 데이터가 입력 시, 전체 데이터 포인트와의 거리 아닌 <u>서포트 벡터</u>와의 거리만 계산 -> *연산량 최소화*  
  - 정확성 높음, 비선형 모델 분류 가능  
  - 다른 모형보다 과대 적합의 가능성 낮음, 노이즈의 영향이 적음  
  <br>

### **단점**  
  - 데이터 전처리 과정 중요  
  - 데이터 증가 시 속도 감소  
<br>






## **(4)KNN**  
: <u>비슷한 특성</u> 가진 데이터끼리 서로 가까이 있다는 점 이용  
(K개의 이웃 설정 -> 가까운 K개의 데이터를 기반으로 분류)  
<br>

### **KNN 알고리즘 계산순서**  
데이터 준비 -> K값 설정 -> 거리 계산 -> 가장 가까운 K개의 이웃 선택 -> 분류하기  
<br>

**<최적의 K 선택하는 법>**  
  - 일반적으로 학습용 데이터 개수의 제곱근으로 설정  
  - K를 너무 크게 설정하면 과소 적합  
  - K를 너무 작게 설정하면 이상값, 잡음 데이터와 이웃이 될 가능성 있음  
  <br>

### **장점**  
- 범주를 나눈 기준을 알지 못해도 데이터 분류 가능  
- 입력 데이터만 주어지면 바로 예측값 구할 수 있우 다른 분류 모형과는 달리 <u>학습과정 불필요</u>  
- 이해 쉽고, 구현 간단  
- 추가된 데이터의 처리 용이  
- 훈련 데이터 요악하거나 일반화된 모델로 만들지 않고, 원본 데이터 전체 그대로 저장  
<br>

### **단점**  
- 테스트 데이터의 개수에 따라 시간 오래 걸림  
- 학습 데이터 모두를 거리 계산에 사용 -> 학습 데이터의 양도 계산 시간에 영향  
- K값의 결정 어려움  
- 수치형 데이터 아니면 유사도 정의 어려움  
- 데이터 내 *이상치* 존재하면 분류 성능에 큰 영향 



cf) **앙상블**  
- 여러 개의 개별 분류 모델들을 **결합**해 하나의 분류 모델보다  더 좋은 성능을 내는 머신러닝 기법  
- 여러 약 분류기를 병렬 또는 직렬로 결합하여 <u>강 분류기</u>로 만드는 것  
- 여러 개의 모델을 학습시키고 각 모델의 예측 결과를 종합해서 최종 예측  
- 종류: 보팅, 배깅, 부스팅  

> # **분류 평가 지표**  
## **(1) 혼동 행렬**  
: 분류 모델의 예측 결과를 <u>정확한</u> 예측과 <u>잘못된</u> 예측으로 구분하여 나타낸 행렬  
-> 예측 성공 여부 + 예측값으로 구분하여 총 4종류의 경우의 수로 나눔  
cf) **혼동 행렬을 이용한 분류 모델 평가 지표**  
- *정확도*: <u>모든 가능한 예측 중</u> 참인 비율  
- *정밀도*: <u>참이라고 예측한 경우</u> 실제 참의 비율  
- *재현도*: <u>실제로 참인 경우 중</u> 참으로 예측하는 비율  
-> 모두 0~1 사이의 값  
-> 정밀도^재현도 그래프에서 <u>두 값이 만나는 지점</u>을 역치로 정하면 예측 오류 *최소화* 가능  
<br>

## **(2) F1-Score**  
: 정밀도와 재현율의 조화 평균  
- 머신러닝 모델의 성능을 평가하는 주요 지표  
- 정밀도와 재현율 간의 균형을 효과적으로 평가하기 위함  
<br>

## **(3) ROC/AUC Curve**  
### **[ROC Curve]**  
- 얼마나 분류가 잘 되었는가를 보여주는 그래프  
- 완벽한 분류기는 불가능하므로, 현실적인 한계(비용, 시간 등)를 고려하여 적절히 타협한 최선의 ROC Curve를 도출해야함  
<br>

### **[AUC Curve]**  
: ROC와 x축 사이의 면적(적분값)  
- 모델의 성능을 *숫자*로 표현 가능  
- 0.5에서 1 사이의 값을 가짐  
- <u>1에 가까울수록</u> 분류 성능이 좋은 것에 해당  



> # **하이퍼파라미터 최적화**  
## **(1) 하이퍼파라미터 최적화**  
- **하이퍼파라미터**: 학습 시작 전에 <u>사용자가 직접 설정하는 변수</u> 
- **최적화**: <u>적절한 하이퍼파라미터</u>를 찾아 모델 성능을 향상시키는 것  
<br>

### **하이퍼파라미터 최적화 과정**  
1. 탐색 범위 설정(최적 값을 찾고 싶은 하이퍼파라미터의 범위 설정)  
2. 평가 지표 계산 함수(성능 평가 함수) 정의  
3. 1단계에서 샘플링한 하이퍼파라미터값을 사용하여 검증 데이터로 정확도 평가  
4. 위 단계를 특정 횟수 반복 -> 정확도 결과를 보고 하이퍼파라미어틔 범위 좁힘  
<br>

## **(2) 하이퍼파라미터 최적화 방법**  
- **Grid Search**: 정해진 범위에서 하이퍼파라미터 <u>모두 순회</u>  
  - P: 범위가 넓고 step이 작을수록 꼼꼼하게 전 범위를 탐색라니 최적 해를 *정확히* 찾기 가능  
  - N: 시간이 오래 걸림  
  - 적용: 넓은 범위, 큰 step 활용해 범위 감소  
  <br>

- **Random Search**: 정해진 범위에서 <u>무작위로 탐색</u>  
  - P: 속도가 Grid Search보다 빠름  
  - N: 무작위라는 한계 때문에 *정확도 감소* -> 다른 모델에 비해 사용 빈도 적음  
  <br>

- **Bayesian Optimization**: 사전 정보를 바탕으로 값을 <u>확률적으로 추정하며 탐색</u>  
  - '가우시안 과정'이라는 통계학을 기반으로 만들어지 모델로, 여러 개의 하이퍼파라미터들에 대해 '습득 기능'을 적용했을 때, <u>가장 큰 값</u> 이 나올 확률이 높은 지점 탐색
