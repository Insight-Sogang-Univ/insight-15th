# **군집화**  
> ## **1. 군집화**  
## 1-1 머신러닝: **비지도학습**  
**비지도 학습의 특징**  
- <u>데이터 자체에 내제</u>된 구조를 파악해서 학습  
- 정답 레이블 x -> 모델의 성능을 명확히 측정 불가  
- by 표현학습 -> 데이터셋의 *고유 패턴* 식별 가능  
-> *목표 명확히 정의*: (구체적&좁은 범위의 문제) 지도학습보다 불리  
-> *목표 불명확*: 데이터에서 <u>새로운 특성을 찾는</u> 등의 목적 수행에 적합  
<br>

**비지도학습이 적합한 케이스**  
1. 패턴 미정 -> 정답을 아직 알 수 없음  
2. 패턴이 계속해서 *변동* -> 정답을 익혀도 큰 의미 없음(답 변동)  
3. *열린 문제*를 해결 + 지식을 *일반화*  


## 1-2. 군집화란?  
**군집화**: 데이터를 <u>비슷한 특성을 가진 그룹</u>으로 분류  
-> 각 그룹이 어떻게 구성되어있는지 파악하면서 전체 데이터 구조 이해 가능  
cf) 군집: 유사한 데이터들의 집합/ 서로 다른 군집에 속한 개체들은 *비슷하지 않아야 함*  
<br>
- 지도학습: 독립변수와 종속 변수 사이의 관계에서 패턴을 찾아 종속 변수를 예측하고자 하는 구조  
- 군집화: 종속변수 <u>설정하지 않고</u>, 데이터 내부의 패턴 인ㅅ기 -> 유사성을 가진 그룹으로 분류  
<br>

**[군집화의 목표]**  
- **응집도 최대화**: 같은 군집에 속하는 데이터끼리는 최대한 유사하도록  
- **분리도 최대화**: 서로 다른 군집은 최대한 분리되도록  
<br>

## 1-3. 군집화 과정  
**[군집화의 기본 과정]**  
1. *피처 선택 또는 추출*: 군집화에 사용할 데이터의 피처(특징)을 선택  
2. *군집화 알고리즘 선택*: 데이터의 특성과 목표에 따라 적합한 군집화 알고리즘 선택  
3. *군집 유효성 검증*: 군집화가 얼마나 잘 되었는지 평가  
4. *결과 해석*: 각 군집이 지닌 특성 분석 및 해석  

> ## **2. 군집화를 위한 데이터 준비**  
## 2-1. 군집화의 주요 고려사항  
-> 군집화는 한 번의 시도로 끝나는 것이 아닌, <u>여러 번의 반복 시도</u> 통해 최적의 결과를 찾아내는 과정  
1. *변수 유형 이해*  
  - 군집화에 사용될 피처의 <u>종류와 특성</u>을 명확하게 이해해야함  
  - 변수 종류가 연속형 또는 명목형인지, 그리고 변수 개수에 따라 적합한 군집화 알고리즘 변화   
<br>

2. *거리/유사도 정의와 측정*  
 - 군집화는 데이터 간의 <u>거리 또는 유사도</u>를 기반으로 그룹 형성  
 - 거리 측정 방법에 대한 정의가 중요 (변수 유형에 따라 적합한 거리 측정 방식의 차이 존재)  
 <br>

3. *차원 축소*  
  - 모델의 성능 향상시키기 위해 유사한 변수들을 묶어 처리  
  - 변수가 많아질수록 모델이 복잡해지고, 효율성과 성능 저하 가능  
  - 차원 축소를 통해 <u>유사한 변수 통합</u> or <u>필요없는 변수 제거</u> -> 변수 개수 감소 가능
  <br>

## 2-2. 변수 유형 이해  
**[변수 유형별 군집화 고려사항]**  
-> 유형에 따라 적합한 <u>거리 측정 방법</u>과 <u>군집화 알고리즘</u> 변화  
1. *연속형 변수*  
  - 특징: <u>수치</u>로 측정 가능한 변수  
  - 거리 측정: 유클리디안 거리, 맨하탄 거리 등 사용 가능  
  - 적합한 알고리즘: K-means, 계층적 군집화 등  
  - 주의사항: <u>스케일링</u> 필수! (단위 때문)  
  <br>

2. *명목형 변수*  
  - 특징: <u>범주</u>로 구분되는 변수  
  - 거리 측정: 해밍 거리, 자카드 거리 등  
  - 적합한 알고리즘: K-modes, 계층적 군집화  
  - 처리 방법- 원-핫 인토딩 또는 더미 변수 변환  
  <br>

3. *혼합형 변수*  
  - 특징: 연속형 + 명목형  
  - 적합한 알고리즘: K-prototypes, Gower distance 기반 계층적 군집화  
  - 처리 방법: 각 변수 유형에 맞는 거리 함수 조합  
  

## 2-3. 거리/유사도 정의와 측정  
**[데이터의 거리 측정 방법]**  
: 거리가 가까울수록 유사, 멀수록 상이  
1. *유클리디안 거리*  
  - 가장 <u>일반적인</u> 거리 측정 방법  
  - <u>직선거리</u>의미, K-means에서 기본적으로 사용  
  <br>

2. *맨하탄 거리*  
  - 각 차원의 차이의 절대값의 합  
  - <u>이상치에 덜 민감</u>  
<br>

3. *코사인 유사도* 
  - 벡터 간의 <u>각도</u> 측정  
  - 텍스트 데이터나 고차원 데이터에 효과적  
  <br>

**[6가지 연결 방법]**   
1. *단일 연결법/최단 연결법*  
  - 군집 A와 B 사이의 <u>가장 가까운 두 점 간 거리</u>로 군집 간 거리 정의  
     - <u>고립된 군집</u> 찾는데 중점  
     - 이상치에 취약  
     <br>

2. *완전 연결법/최장 연결법*  
  - 군집 A와 B 사이의 <u>가장 먼 두 점 간 거리</u>로 군집 간 거리 정의  
     - 가장 유사성이 큰 군집으로 병합해 나가는 방법  
     - <u>군집들의 내부 응집성</u>에 중점  
     - 이상치에 민감 
     <br>

3. *평균 연결법*  
 - 군집 A와 B의 <u>모든 점 쌍 사이 거리의 평균</u>  
     - **모든 항목에 대한 거리 평균**을 구함  
     - 위의 두 방법보다 이상치에 덜 민감  
     - 계산량이 불필요하게 증가 가능(n)
     <br>

4. *중심 연결법*  
- 각 군집의 <u>중심점</u> 간의 거리  
     - 두 군집이 결합될 때 새로운 군집의 평균은 <u>가중 평균</u>을 통해 구함  
     - 시간 오래 걸림  
     - 덴드로그램에서 거리 축이 위에서 아래로 줄어드는 inveresion 발생 가능성 존재  
     <br>

5. *중앙 연결법*  
 - 두 군집의 <u>중앙값</u> 사이의 거리  
     - 합쳐지는 <u>두 군집의 중심점 사이의 산술적 중간 지점</u>을 새로운 대표값으로 설정 -> 거리 측정  
     - 데이터의 중앙값에 기반해 군집 간 거리 정의 -> <u>극단값</u>에 영향 덜 받음  
     - 기하학적 구조 파악 어려움  
     <br>

6. *Ward 연결법*  
 - 군집 내 제곱합이 증가하지 않도록 병합  
     - 군집의 <u>병합 후</u> 군집 내 <u>오차 제곱합의 증가분이 최소</u>인 것을 선택 
     - 보통 병합된 군집의 SSZ는 이전에 비해 증가 -> 그 증가량이 가장 작아지는 방향으로 군집 형성 

## 2-4. 차원 축소  
**[차원의 저주]**  
- 거리 기반 알고리즘 -> 데이터의 차워닝 높아질 수록 차원의 저주 발생  
<br>

**차원의 저주**  
: 피처 공간이 너무 커서 알고리즘이 데이터를 효과적/효율적으로 훈련할 수 없는 현상  
-> 고차원일수록 거리 기반 알고리즘의 분별력 감소  
<br>

**차원 축소**  
- 차원: 변수의 개수  
- 차원 축소 = 변수의 개수 감소  
- **목표**: 변수를 제거하여 차원 축소 -> 고차원의 데이터를 <u>저차원에 투영</u> -> 핵심 정보는 유지  => 노이즈 감소하여 패턴 학습에 유리 
<br>

1. **선형 투영**  
  - 고차원에서 저차원 공간으로 <u>선형적으로</u> 데이터 투영  
2. **매니폴드 학습(비선형 차원 축소)** : 유클리드 거리가 아닌 <u>ㄷ이터 포인트들</u> 사이의 곡선 거리 고려하며 학습  
<br>

**[PCA]**  
- 목적: 다차원 데이터에서 <u>가장 정보량(분산)이 큰 방향</u> 탐색 -> 그 방향 기준으로 데이터 투영 => 차원 축소  
- 특징  
   - <u>분산 최대한 보존</u>하면서 차원 축소 -> 시각화와 분석 용이  
   - 상관관계 높은 피처들 결합 -> 적은 수의 피처로 데이터 표현 지향  
   - 원본에서 <u>최대 분산 방향</u> 찾아 상관관계 감소 -> 저차원 공간에 투영  
     -> 새로 파생된 성분: **주성분** 
     <br>

- 주의사항  
  1. 표준화 필수  
  2. 결측값 처리  
  3. 해석의 어려움  
  4. 정보 손실  
  


> ## **3. 군집화 알고리즘**  
## 3-1. 계층적 군집화  
: 데이터 간의 유사성을 기반으로 트리 구조를 형성, 상향식 or 하향식 방식으로 군집 형성  
-**특징**  
  - 덴드로그램 형성  
  - 군집의 개수 사전 설정 x -> 클러스터링 종료 이후 개수 선택 가능  
  - 상향식, 하향식 구성에 따라 응집형과 분리형으로 구분(상향식-응집형/ 하향식-분리형)  
  - 소규모 데이터에 대한 군집 수 탐색 or 클러스터링의 전반적인 구조를 시각적으로 분석할 때 유용  
  <br>

**응집형/분리형 계층적 군집화**  
- *응집형*: 각 샘플이 독립적인 클러스터 -> 하나의 클러스터만 남을 때까지 가장 가까운 클러스터 병합  
- *분할형*: 전체 샘플 포함하는 하나의 클러스터에서 시작 -> 유사성 낮은 데이터들을 더 작게 분할  
- 두개 모두 거리 행렬에 기반한 <u>계층적 구조</u>로 조직화  
- 결과: 일반적으로 이진 트리 또는 덴드로그램  
<br>

## 3-2. K-means  
: 데이터를 정해진 개수(K)의 그룹으로 나누되, <u>각 그룹의 중심점과의 거리가 가장 가까운 데이터</u>끼리 묶는 군집화 알고리즘  
- 군집화에서 가장 일반적으로 사용  
- 연속적인 특정 갖는 데이터 샘플 중 군집 중심점 선택 -> 해당 *중심에 가장 가까운 데이터*들을 병합  
- 프로토타입 기반 군집화에 속함  
- 유클리디안 거리 지표 사용  
<br>

**Pros**  
- 직관적이고 구현 쉬움  
- 대용량 데이터에도 적용 가능  

**cons** 
- *초기값*에 민감  
- 군집 수 결정이 어려움 
- 이상치에 민감  
- 기하학적 모양의 군집 파악 어려움  
<br>

**K-means++ 알고리즘**  
: 초기 중심점들을 서로 멀리 떨어진 곳에 위치 -> 더 일관되고 좋은 결과 도출  
- k-means: 초기 중심점 랜덤 할당  
- k-means ++: 초기 중심점들을 서로 멀리 멀리 떨어진 곳에 위치  
<br>

### 엘보우 방법
: 클래스 내 SSE를 바탕으로 그래프를 활용해 <u>최적 클러스터 개수 K개를 추정</u>하는 방법  
- 왜곡이 빠르게 감소하는 지점의 k값을 찾는 것  
<br>

**pros**  
- 직관적  
- 구현이 간단
<br>

**cons**   
- 주관적 판단  
- 모호한 경우  


## 3-2. DBSCAN  
: 밀도가 높은 지역의 데이터를 하나의 군집으로 묶고, 밀도 기준을 만족하지 못하는 점은 군집에 포함시키지 않는 군집화 알고리즘(밀도에 따른 군집화)  
<br>

**특징**  
- 명시적 이상치 지정 -> 클러스터링에서 제외  
- 다른 알고리즘 대비 이상치로부터 자유로움  
- 어떤 데이터가 여러 군집의 특정 거리 내 존재 -> 가장 밀도가 높은 군집과 그룹화 
<br>

**단계**  
샘플 분류-클러스터 생성-경계 샘플 할당  
<br>

**pros**  
- *밀도*에 따른 클러스터 할당 
- 이상치에 둔감  
- 모든 샘플을 클러스터에 할당하지 X(잡음 샘플 구분)  
<br>

**cons**  
- epsilon과 min_samples 설정에 많은 영향 받음  
- epsilon 조절 어려움, 적절한 조절을 위해서는 데이터셋에 대한 도메인 지식 필요  
- 연샨량이 많아 속도 느림  
- 다른 밀도 분포 가진 데이터의 군집 분석 잘 하지 못함 
- 차원의 저주 영향을 받음  
<br>

### [HDBSCAN]  
- 계층적 DBSCAN  
**작동 방식**  
1. 밀도 기반으로 1차 군집화  
2. 거리 기준으로 밀도 기반 군집들을 반복적으로 연결  
<br>

**pros**  
- 여러 밀도 기준으로 반복 수행  
- DBSCAN의 단일 밀도 기준 한계 보완  





## 3-4. GMM  
:데이터가 여러 다른 모향의 가우시안 분포로 구성되었다고 가정 -> 각 분포를 클러스터로 인식하는 군집화 방법
<br>

**특징** 
- <u>모델 기반 군집화</u> 방법  
- 각 군집을 <u>확률 분포</u>로 간주  
- 전체 데이터 분포를 <u>야러 확률 분포의 혼합</u>으로 모델링  
- 데이터의 <u>생성 메커니즘</u>까지 모델링 가능  
<br>

**pros**  
- k-means보다 유연하게 다양한 데이터 세트에 잘 적용 가능  
- k-means가 잘 작동하지 않는 타원형 분포나, 중첩된 군집 구조에서 좋은 성능 가능
<br>

**cons**  
- 수행 시간이 오래 걸림 
- 가정한 분포에 맞지 않은 데이터 -> 계산 복잡도 증가, 성능 저하  



> ## **4. 군집화 평가 방법**  
## 4-1. 외부 평가와 내부 평가  
### 1.외부평가  
- <u>정답 레이블이 존재하는 경우</u>에 사용  
- *군집화 결과*와 *실제 레이블*을 비교하여 얼마나 유사한지 평가  
<br>

### 2.내부 평가  
- <u>정답 레이블이 없는 경우</u>에 사용  
- 군집 내부 *응집도*와 군집 간 *분리도*를 기반으로 군집의 품질을 정량적으로 측정  
- 각 알고리즘 자체의 내부 평가 지표와 범용적으로 사용 가능한 내부 평가 지표 존재  
<br>

## 4-2. 실루엣 계수  
: 클러스터 내 샘플들이 얼마나 조밀하게 모여 있는지 측정> 군집의 품질 확인 가능  
<br>

**실루엣 계수 해석**  
- -1과 1 사이값  
- <u>분리도가 응집도보다 큰 경우</u>: 이상적인 실루엣 계수인 1에 근사  
- 분리도와 응집도가 <u>같은 경우</u>: 실루엣 계수=0  
- 분리도가 응집도보다 <u>작은 경우</u>: 실루엣 계수는 -1에 근사  
<br>

## 4-3. Dunn Index  
: 클러스터 간 *최소거리*와 클러스터 내 *최대 거리*의 비율을 계산 -> 클러스터링 품질 평가하는 지표  
<br>

**한계**  
- 군집 수 증가할수록 계산 비용 증가  
- 극단적인 값에 큰 영향 받음  
- 이상치에 민감하게 반응 -> 수치 급감 가능
