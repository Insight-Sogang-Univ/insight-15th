# **1. 연관분석**  
> ## **1. 연관 분석이란?**  
: 일련의 거래나 사건 안에 존재하는 항목 간의 일정한 <u>연관규칙</u>을 발견하는 분석  
-> *비지도 학습*에 해당  
## 1-2. 연관규칙 분석  
### **연관규칙의 형태**  
- 형식: if-then(A->B)  
  - 조건절: '만일 ~라면'에 해당하는 부분  
  - 결과절: 그 뒷부분에 해당하는 내용  
  - 아이템 집합: 조건절 또는 결과절을 구성하는 아이템들의 집합(요소들의 개수에는 제한이 없으나, <u>상호 배반</u>이어야 함)  
<br>

### **연관규칙 주의** : 인과관계 알 수 없음  
: only 상관관계 파악 가능  
**why?**  
1. 실험이 아닌 <u>관찰 데이터</u>이기 때문  
2. 방향성은 그냥 수학적 비율일 뿐  
3. 상관관계 ≠ 인과관계 
<br>

### 연관 분석 vs 협업 필터링  
* **협업필터링(상관관계)**  
  - 상관관계 기반  
  - 상품마다 타 상품과 <u>1:1</u> 계산  
  - 사용자의 개인 선호도에 따라 *선호 상품* 추출 가능  
  - 타 상품의 유사도 *순위대로* 노출  
  - 복잡한 수치 계산 가능 -> 개인화된 상품 추천에 활용 
  <br>

* **연관분석**  
- 교집합 기반  
- 상품마다 주문내역을 <u>1:n</u> 계산  
- 상품과 연관성이 높은 1개 이상의 상품을 세트로 추출 가능  
- 연관지표가 높은 *세트상품(1개이상)* 노출  
- 다양한 세트 상품 추출 가능-> 상품 노출 또는 세트상품 구성에 활용
<br>



> ## **2. 주요개념**  
## 2-1. (적절한) 연관규칙의 조건  
1. 두 품목 함께 구매한 경우의 수가 일정 수준 이상이어야 한다.(**일정 이상의 지지도**)  
2. 품목 A를 포함하는 거래 중 품목 B를 구입하는 경우의 수가 일정 수준 이상이어야(**일정 이상의 신뢰도**)  
**지지도**  
: 전체 거래항목 중 <u>동시에 품목 A,B 포함</u>하는 거래의 비율(교집합)  
- 조건절이 발생하는 확률  
*낮은 경우*  
  - 우연일 수 있는 규칙 포함-> 통계적 신뢰도 없음  
  - 실행 가능한 인사이트 도출 어려움
  <br>

**신뢰도** 
: *품목 A를 포함하는 거래* 수 중 <u>품목 A와 B가 동시에 포함</u>하는 거래의 비율  
*낮은 경우*  
- 의미없는 연관 규칙 -> 추천/진열 무의미  
- 추천 실패로 귀결 -> 클릭률, 전환율 모두 낮아짐
<br>

## 2-2. 연관 분석의 평가 척도 
**향상도**  
: 품목 B의 확률이 품목 A의 제공 여부에 따라 얼마나 증가하는지의 비율  
-> A와 B가 우연히 등장할 확률과 <u>실제 함께 등장할 확률</u> 비교  
- 독립인 경우: 향상도 **1**  
- 독립이 아닌 경우: 1보다 크거나 작음  
cf) 규칙의 효용성: <u>지지도, 신뢰도, 향상도</u> 세가지 모두 반영해 평가  



> ## **4. 연관 분석 알고리즘**  
## 4-1. Apriori 알고리즘  
### **선험적 경험 이용** 
- 배경: 가능한 모든 조합을 시도하는 것의 비효율성  
- 개념: 상위 조합에서부터 차례로 스캔 -> 특정 조합이 자주 발생하지 않는다면 이의 후속 조합들까지도 모두 후보에서 <u>배제</u>하는 방식  
- 장점: <u>하나의 조합만 검사</u>하고도 배제 가능성 발생 -> 시간과 연산령을 효과적으로 감소 가능  
<br>

**Apriori 알고리즘 프로세스**  
1. 단일 항목 집합 생성  
2. 2개 항목 집합 생성  
3. 더 큰 항목 집합 생성  
4. 최종 빈발 항목 집합 선정  
  -> 남은 항목 집합 바탕으로의 *연관 규칙* 생성
  <br>

## 4-2. FP Growth 알고리즘  
- Apriori의 연산 속도 문제를 해결할 수 있게 <u>자료구조</u>를 잘 이용한 알고리즘  
- 배경: Apriori 알고리즘에서 데이터가 커질수록 속도가 느려진다는 단점 존재(각 단계에서 데이터 스캔 재진행하기 때문)  
**How?**  
- FP-Tree 구조 이용 -> 효과적으로 구현  
- 모든 거래를 확인해 각 아이템마다 지지도 계산 -> <u>최소 지지도 이상의 아이템만 선택</u>  
1. 데이터셋 확인  
2. 빈도순 재정렬 
3. FP-트리 생성  
4. 빈발 항목 집합 추출  
5. 연관 규칙 도출  


# **2. 경영기초**   
> ## **1. 문제 정의와 구조화** 
## 1-1. 문제 정의   
문제: '현재 상태'와 '바라는 상태' 사이의 격차  
-> 단순히 나쁜 상황을 <u>나열</u>하는 것은 문제가 아님  
-> 경영자가 '의사결정'을 내려야 하는 <u>구체적인 질문으로 치환</u>되었을 때 비로소 문제가 됨  
<br>

## 1-2. 문제 구조화  
**MECE**  
: 문제 분해시 반드시 지켜야하는 철칙  
1. 상호 배타적: 겹치는 부분 없음  
2. 전체 포괄적: 누락된 부분 없음  
cf) *좋은 MECE를 하기 위한 규정*  
  1. 논리적 일관성  
  2. 통찰력  
  3. 상호 배타성
  <br>


> ## **2. 지표**  
## 2.1 지표 
: 방향이나 목적, 기준 따위를 나타내는 표지  
(분모 대비 분자로 표현)
<br>

## 2.2 지표 정의  
**Proxy Metric**  
: 추상적이여서 프로덕트마다 나름의 *가정과 정의,논의*가 필요한 지표 의미  
<br>

## 2.3 북극성 지표  
: 기업이 *가장 중요하게* 생각하는 지표  
**L1/L2 지표**  
 - L1: 전사적인, 고수준의 성과 지표(전략 관점)  
 - L2: L1 지표의 세부사항을 측정하는 지표(전술 관점) 
<br>

## 2.4 AARRR(해석지표)  
1. Acquisition  
: 사용자가 앱 혹은 웹사이트에 방문하는 것  
  - 신규 고객의 획득/ 어디로부터 획득/ 어떤 채널을 통한 획득 고민  
  - 대체로 조회수, 가입 사용자 수 등 지표 통헤 측정  
2. Activation  
: 사용자가 앱 또는 웹사이트에 첫 방문하여 주요 기능을 직접 사용하는 것  
  - 제품의 성격에 따라 다양한 기준으로 측정 
  - 제품 핵심 기능의 활용도, 사용자의 온라인 접속 시간 등 존재  
3. Retention  
  - 유지 단계에서 <u>신규 사용자를 장기 사용자로 전환</u>하는 것에 초점  
  - 기존 사용자의 유지보다 더 많은 비용 투입  
  - 측정 지표: 익일 유지율, 1주차 유지율, 2주차 유지율, 월 유지율, EDM 메일 개설율 등  
4. Revenue  
  - 제품 혹은 서비스를 유료 구매하기로 결정한 단계  
  - 사용자의 소비 유도 -> 매출 상승에 총력 투여 -> 최대한 사용자를 소비하게끔 만들거나 매출 늘릴 수 있는 행위 하도록 유도  
  - 고객의 평생 가치 지속적으로 상승, 고객 유치 비용 게속 감소 시키기!  
5. Referral  
  -제품의 지속적인 성장을 위햏 매우 중요한 단계  
  - '어떻게 충성도 높은 사용자를 브랜드 홍보자로 만들 것인지', '어떻게 제품을 잠재 고객들에게 추천하도록 만들 것인지' 고민  
  - 측정 지표: 순 추천고객 지수, 바이럴 지수 등  
<br>

> ## **3. 지표 분석 방법**  
## 3-1. 코호트 분석  
**코호트**  
 - 동질 집단  
 - 정의된 시간 범위 내에서 공통된 특성이나 경험을 공융하는 사용자 집단 의미  
 - 특정 기간에 특정 특성을 <u>공유</u>하는 사용자 집단  
 - 일종의 행동 분석  
<br>

## 3-2. 퍼널 분석  
: 유저가 접속해서 떠날 때까지 유저의 행동 추적  
**개선방법**  
1. 유입률 높이기 
2. 전환율 높이기  
3. 퍼널 단계 없애기  
4. 퍼널 순서 바꾸기  
5. 퍼널 단계 추가하기  
6. 코호트로 나눠보기
