# <span style="color:skyblue; background-color:white">통계</span>

- 통계는 우리가 다양한 비즈니스와 경제 상황을 이해하는데 도움을 주는 숫자로 표현된 사실을 나타냄

- 더 나은 의사결정을 내리는 데 도움을 줌



<span style="color:skyblue; background-color:white">모집단</span>: 전체 집단 <BR>
표본 집단: 모집단에 대해 일부 샘플

**기술통계**: 수치와 시각화를 통해 *샘플을 요약하여 표현*하는 방법론
**추론통계**: 추정과 검정을 통해 *샘플로부터 모집단을 추론*하는 방법론

### <span style="color:skyblue; background-color:white">1-2. 자료의 척도</span>

- **질적 자료**
    - **명목척도**
        
        : **이름으로 요소의 속성을 식별**하려고 하는 척도
        
        - ex. 펀드 유형(국내주식: 1, 국제주식: 2, 채권: 3), 성별(남: 0, 여: 1)
    - **서열척도**
        
        : 항목 간의 **순서 또는 순위를 나타내는 특성**을 지니는 척도
        
        - ex. 만족도(`좋음` `보통` `나쁨`), 직급 체계(사원 < 대리 < 과장 < 차장 < 부장)

- **양적 자료**
    - **구간(등간)척도**
        
        : 값들 사이에 순서가 있고, 그 간격이 일정하여 덧셈이나 평균과 같은 연산이 가능한 척도
        
        - **단, 곱셈과 나눗셈은 의미를 지니지 못한다.**
        - ex. 온도(섭씨, 화씨), IQ, 시간
    - **비율척도**
        
        : 절대영점과 임의의 측정단위로 구성하고, 동간성이 유지되는 척도
        
        - 이 척도에서는 `0` 값이 그 변수에 대해 아무것도 존재하지 않는 상태를 나타냄. 가격이 0이면, 비용이 전혀 안 들거나 공짜라는 사실을 나타냄.
        - ex. 거리, 높이, 무게, 시간




# <span style="color:skyblue; background-color:white">기술통계</span>


- 수집한 데이터의 특성과 패턴을 있는 그대로 보여주는 역할
- ex) 회사의 올해 평균 매출은 얼마인가? <BR>   제품 A의 불량률은? <BR>   고객들의 연령 분포는?

- 1️⃣ 중심화 경향 (평균, 중앙값, 최빈값)
평균: 데이터 값의 총합을 데이터 개수로 나눈 값 <BR> 총합을 사용하기에, *이상치에 민감*할 수 있다. <BR>
중앙값: 데이터를 크기 순서대로 일렬로 정렬했을 때, **한가운데에 위치하는 값** <BR>
최빈값: 데이터에서 가장 자주 나타나는 값
- 2️⃣ 자료의 산포(퍼진정도) (분산, 표준편차, 범위)
범위: 데이터의 최대값과 최소값의 차이를 나타내는 지표 <BR>
분산(variance): 데이터 값들이 평균으로부터 얼마나 멀리 퍼져있는지를 나타내는 지표 <BR>
분산값이 클수록 데이터가 많이 퍼져있다는 뜻 <BR>
표준편차: *분산에 제곱근을 취한 값*
분산의 단위는 관측값 단위의 제곱이기에 대상 데이터와 일치하지 않음. 그렇기에 제곱근을 씌워 단위를 일치시켜주는 것 <BR>
사분위수: <BR>
 <BR>
Q1 = 하위 25퍼센트에 위치한 값, `하위 사분위수` <BR>
Q2 = 하위 50퍼센트에 위치한 값, `중앙값` <BR>
Q3 = 하위 75퍼센트(상위 25%)에 위치한 값, `상위 사분위수` <BR>
사분위간 범위(IQR) Q3-Q1 <BR>
 <BR>
변동계수: 상대적인 변동성을 나타내는 지표 <BR>
변동계수 = (포준편차/평균)*100

<span style="color:skyblue; background-color:white">첨도와 왜도</span> <BR>
첨도: 확률분포의 꼬리가 두꺼운 정도를 나타내는 척도 <BR>
왜도: 비대칭도, skewness 
- 3️⃣ 자료의 정리 및 시각화 (히스토그램, 박스플롯, 파이차트)


# <span style="color:skyblue; background-color:white">확률과 분포</span>

- **표본의 불확실성**: 혹시 운 없이 이상한 표본만 뽑힌 건 아닐까?
→ **표본분포**(Sampling Distribution)
- **모수 추정**: 그 표본의 통계량이 정말로 모수를 잘 반영한다고 할 수 있나?
→ **점추정**(Point Estimation)
- **추정의 현실성**: 점추정 값이 모수와 정확히 일치하는 건 비현실적이지 않나?
→ **구간추정**(Interval Estimation)
- **구간추정의 오류 관리**: 이 구간이 틀릴(모수를 포함하지 못할) 가능성은 어떻게 관리하지?
→ **신뢰수준**(Confidence Level)
- **적극적 주장 검증**: 특정 주장이 맞는지 틀리는지 직접 판단을 내릴 수는 없나?
→ **가설검정**(Hypothesis Testing)
- **가설검정의 오류 관리**: '이 주장은 틀렸다'는 결론을 내렸는데, 사실은 내 결론이 틀렸을 가능성은 어떻게 관리하지?
→ **유의수준**(Significance Level, 제1종 오류)

# <span style="color:skyblue; background-color:white">확률분포</span>

(probability distribution)는 확률변수가 특정한 값을 가질 확률을 나타내는 함수를 의미 <BR>

이산확률분포: 확률 변수가 이산적인(discrete) 값을 가지는 경우 <BR>
연속확률분포: 확률 변수가 연속적인(continuous) 값을 가지는 경우 <BR>



 <BR>
- 모집단의 분포가 정규분포를 따를 때 표본평균의 분포도 정규분포를 따르게 됨
- 중심극한정리에 따르면, 표본의 크기가 커질수록 표본평균의 분포는 모집단의 분포와 상관없이 정규분포에 가까워짐
- 즉, 모집단의 분포를 몰라도 표본의 크기가 충분히 크다면(n>=30) 포본평균의 분포는 근사적으로 정규분포를 따른다고 가정할 수 있음


<span style="color:skyblue; background-color:white">Z분포(표준정규분포)</span> <BR>


표본분포: 정규분포를 따르는 모집단에서 추출한 표본(sample) 통계량들이 따르는 분포 <BR>
종류: T분포, 카이제곱분포, F분포 등



자유도: 통계에서 *"얼마나 많은 정보가 독립적으로 변할 수 있는가"*를 나타내는 숫자

# <span style="color:skyblue; background-color:white">추론 통계</span>

추론 통계: 수집한 데이터를 기반으로 미래의 / 혹은 조사하지 못한 전체 정보를 추론하고 결론을 도출하는 방법을 다루는 분야


추정: 모집단을 대표할 수 있는 일부를 표본으로 추출하여 분석한 통계량을 이용해서 모집단의 모수를 예측하는 과정 <BR>
    - 점추정 : 추정하고자 하는 모수를 하나의 수치로 추정 <BR>
    - 구간추정 : 추정하고자 하는 모수가 존재하리라 예상하는 구간을 정하여 추정 <BR>

<span style="color:skyblue; background-color:white">모수</span>: 모집단의 특성에 관계되는 평균, 비율 또는 표준편차 등 관심의 대상이 되는 미지의 참값

<span style="color:skyblue; background-color:white">검정</span>: 표본을 추출하여 얻은 표본통계량으로 모집단 모수에 대한 새로운 예상·가설 등이 옳다고 판단할 수 있는지를 평가하는 과정

- 점추정: 어떤 모수를 표본에 기초하여 계산된 하나의 값으로 추정하는 것

- 구간추정: 모수의 참 값이 포함되리라고 기대하는 추정치를 일정한 범위와 신뢰도를 가지고 나타내는 방법

가설 검정의 절차 <BR>
**가설 설정 - 유의 수준 설정 - 검정통계량 산출 - 가설 기각/채택 판단**


1️⃣ 귀무가설 H0와 대립가설 H1를 설정하기

귀무가설은 주로 **기각하고자 하는 명제**로 설정함. 보통 “~는 차이가 없다”라는 명제.

같은 말로, 대립가설은 **내가 주장하고자 하는 명제** 로 설정함. 보통 “~는 차이가 존재한다”라는 명제.

2️⃣ 유의 수준 설정

유의수준이란 연구자가 결과를 판단할 때 허용할 수 있는 오류의 최대치로, 주로 0.05, 즉 5%로 설정

**1종, 2종 오류**

- **1종 오류(𝛼)** : 귀무가설이 실제로 참이었는데 기각할 확률
- **2종 오류(𝛽)** : 귀무가설이 실제로 거짓인데 기각하지 못할 확률

가설검정은 1종 오류를 더 치명적인 것으로 파악함

따라서 유의수준은 1종 오류의 확률을 의미하게 됨



3️⃣ 검정통계량 산출

검정통계량은 가설 검정의 결과를 판단하는 데 사용되는 값

데이터의 분포나 특징에 따라 T값, F값 등 어떤 통계량을 사용할지 정해야 함.

Z값: Z값 또는 Z-점수는 표준 정규 분포(평균이 0이고 표준편차가 1인 분포)를 사용하여 얻은 값

T값: T값은 T분포를 사용하여 얻은 값으로, 작은 표본 크기에 대한 가설 검정(T검정)에 주로 사용됨

F값: F값은 F분포를 사용하여 얻은 값으로, 주로 두 표본 분산 사이의 차이를 비교하는 데 사용됨

유의확률(p-value)

- 가설 검정에 쓰일 수 있는 또 다른 척도
- 유의확률은 검정통계량에서 유도할 수 있으며 유의 확률(p)가 유의 수준(α)보다 작을 시에 귀무가설을 기각
- 클 시에는 귀무 가설을 기각하지 못함

데이터의 자료형마다 검정 방법이 다름.
- 수치형~수치형: 피어슨 상관분석
- 범주형~수치형: F-테스트, T-테스트
- 범주형~범주형: Chi2테스트
- 수치형~범주형: 없다...
