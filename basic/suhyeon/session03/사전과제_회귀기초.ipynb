# <span style="color:white; background-color:pink">회귀 기초</span>

### <span style="color:white; background-color:pink">머신 러닝</span>

- 인공지능의 한 분야로, **모델**이라는 소프트웨어를 학습하여 새로운 데이터를 예측하거나 결정을 내릴 수 있도록 하는 기술 <BR>
- 데이터에서 패턴을 발견하는 것과 이 패턴을 바탕으로 새로운 데이터에 대해 예측하는 것을 가능하게 함 <BR>


|  | 정답 여부 | 방식 | 예시 |
|------|------|------|------|
| 지도 학습 | 정답이 있는 데이터 사용 | 기계의 예측값이 데이터의 정답과 같아지도록 학습시키는 것 | 회귀, 분류 |
| 비지도 학습 | 정답이 없는 데이터 사용 | 데이터의 패턴 혹은 유사성을 학습시키는 것 | 군집화 |
| 강화 학습 | 보상 | 시행착오를 반복하여 보상을 최대화하는 행동 학습 | 알파고, 자율주행 |

<span style="color:white; background-color:pink">지도학습</span> <BR>

- 사진 하나하나 기린인지 고양이인지 알려준다. 다음에 답을 맞힐 수 있도록
- 예측값과 정답이 최대한 같아지는 방향으로 학습 <BR>

- 지도 학습을 위해서는 반드시 데이터가 labaeled data의 형태여야 함
1. **labeled data** ⇒ **독립변수(feature)**와 **종속변수(target/label)**가 모두 존재하는 데이터. 입력 데이터와 그에 대응하는 출력값이 있는 데이터를 의미 
    - **독립변수**: 머신러닝에서 **예측에 사용되는 입력 변수**
    - **종속변수**: 머신러닝에서 **예측하고자 하는 대상 변수**
2. **labeled data** 를 **training set**과 **test set**으로 나눔
    - **train set**: 모델을 학습(훈련)시킬 때 쓰이는 데이터
    - **test set**: 학습된 모델을 검증하기 위해 사용하는 데이터 → unseen data
3. training set을 통해 모델을 학습

<span style="color:white; background-color:pink">비지도학습</span> <BR>

- 사진을 여러 개 주고 답을 알려주지 않는다. 컴퓨터가 데이터의 특성을 스스로 파악하도록 한다. 데이터의 특성을 파악하여 비슷한 사진끼리 분류하는 것을 **군집화**라고 함

## <span style="color:white; background-color:pink">모델링</span> <BR>



모델은 *학습의 결과물*
- 데이터를 학습하며 찾아낸 패턴이 저장된 상태, 새로운 데이터를 넣었을 때 답을 내놓는 예측기 역할
- ex) 공부 시간 x를 넣으면 시험 점수 y가 나오는 공식 y = ax + b 도 모델이라고 볼 수 있음!

## <span style="color:white; background-color:pink">선형 회귀</span> <BR>

**데이터를 가장 잘 설명하는 회귀선**을 찾는 과정
-  현재 데이터를 요약
- 아직 관측되지 않은 미래 값을 예측하는 도구

1) 가중치 $w$ (Weight)

- 독립변수 $x$가 종속변수 $y$에 주는 **영향력의 크기**
- 그래프에서는 **기울기**
- **회귀계수**라고도 불림

2) 편향 $b$ (Bias)

- 독립변수 $x$가 0일 때 나오는 **기본값**
- 그래프에서는 **y절편**으로 나타남
- 모델이 데이터에 더 유연하게 최적화될 수 있도록 위아래로 움직여주는 역할

## <span style="color:white; background-color:pink">다중공선성</span> <BR>

- 문제가 심각하면 회귀계수 추정 불가

- 예를 들어: 주택 가격을 예측하는 모델에서, '주택의 크기'와 '방의 개수'라는 두 독립변수가 강한 상관관계를 가진다면(상관계수 1에 가까운), 이 두 변수 간에 다중공선성 발생.

## <span style="color:white; background-color:pink">규제선형모델</span> <BR>
 <BR>
두 번째이다.
underfitting, good fitting, overfiting.

그래서 규제선형모델이란?
- 모델이 학습데이터에 과적합되지 않도록 규제를 가하고자 등장한 모델

회귀계수가 커진다는 것은
- 특정 독립 변수에 과하게 의존한다는 뜻

1. L2규제: W(회귀계수)의 **제곱**에 대해 페널티를 부여하는 방식, **릿지(Ridge)회귀**
2. L1규제:  W(회귀계수)의 **절댓값**에 대해 페널티를 부여하는 방식, **라쏘(Lasso) 회귀**
3. L2규제 + L1규제**: 엘라스틱넷(Elastic Net) 회귀**

## <span style="color:white; background-color:pink">모델평가방법</span> <BR>

성능평가지표

1. 평균제곱오차: 예측값과 실제값 사이의 오차 제곱의 평균
2. 평균절대오차: 모델의 예측값과 실제값 사이의 절대값 오차의 평균
3. MSE와 MAE의 차이
- MSE는 큰 오차에 대해 민감하게 반응, 이상치에 민감
- MAE는 모든 오차를 동일하게 취급, 이상치에 민감X

변수유의성 평가

t검정은 독립변수의 회귀계수가 유의미한지 검정하는 데 사용됨.
