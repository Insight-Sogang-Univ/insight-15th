## <span style="color:green; background-color:white">선형 회귀</span>

다중선형회귀의 식은 이렇게 생겼음. <BR>


여기서 상수항과 오차항을 제외하면 이렇게 됨

 <BR>
해석(제미나이): 선형 결합식을 세가지 다른 표기법으로 나타낸 것. <BR>
여러 개의 입력값에 각각의 가중치(w)를 곱해서 더하면 예측값(y)가 된다는 뜻. <BR>
<span style="color:pink">x</span>=우리가 가진 데이터의 특성(피쳐)들이다. (예시: 집 값 예측 시 방 개수, 면적 등) <BR>
<span style="color:pink">w</span>=각 특성이 결과에 얼마나 영향을 주는지 나타내는 가중치(weight) <BR>
<span style="color:pink">w0x0</span>=편향: 보통 x0는 항상 1로 간주. 걍 절편이랍니다

그러나 선형회귀를 모든 상황에 쓸 수 있는 건 아니다!!!


## <span style="color:green; background-color:white">다중선형회귀의 기본 가정 6가지</span>

가정1 - <span style="color:red">선형성</span>

종속변수와 설명변수 간의 관계가 선형적이어야 한다.

가정2 - <span style="color:red">독립성</span>

각각의 설명변수가 서로 선형독립적이어야 한다.
- 그렇지 않은 경우 <span style="color:red">다중공선성</span> 이 있다고 함 <BR>
-> VIF 구하기 <BR>
-> 대처법: 변수 제거, 규제 선형 모델, PCA 

가정3 - 오차항의 평균은 <span style="color:red">0</span> 이다(0에 가까워야 한다)

오차항: 실제 값과 예측값 사이의 차이 (White noise)라고 부르기도 함. <BR>
왜 그렇게 부르냐면 특정 패턴 없이 랜덤하게 변하는 신호라서

가정4 - <span style="color:red">등분산성</span>

오차항의 분산이 일정해야 한다.
오차항이ㅡ 분산이 일정하지 않은 경우, <span style="color:red">이분산성</span>이 있다고 한다.
횡단면 자료에서 나타난다.
횡단면 자료(Cross-sectional Data)란 특정 시점에 여러 대상으로부터 수집한 데이터를 말함
(시계열 자료, 패널 자료와 비교)

| 종류 | 특징 | 예시 |
|------|------|------|
| 횡단면 자료 | 같은 시간에 여러 대상을 관찰 | 2026년 3월, 전국 가구의 소득과 소비 지출액 조사 |
| 시계열 자료 | 한 대상을 여러 시간에 걸쳐 관찰 | 삼성전자 주가의 10년간 가격 변동 |
| 패널 자료 | 여러 대상을 여러 시간에 걸쳐 관찰 | A, B, C 세 가구를 5년간 추적하며 조사한 데이터 |

왜 횡단면 자료에서 이분산성이 많이 나타날까? <BR>
"고소득으로 갈수록 오차가 커진다" <BR>
**현상**: 저소득층은 소득이 적어 선택의 폭이 좁기 때문에 소비 패턴이 비슷비슷(오차가 작음). 하지만 고소득층은 누군가는 검소하게 살고, 누군가는 호화롭게 살기 때문에 소비의 변동 폭이 매우 큼(오차가 큼).

**결과**: 이렇게 서로 다른 특성을 가진 개체들을 한꺼번에 모아놓고 분석하다 보니, 특정 구간에서 오차의 크기가 확 벌어지는 이분산성이 자주 발생하는 것.


가정5 - 오차항은 <span style="color:red">자기상관</span>되어 있지 않다. (non-autocorrelation)

오차항의 공분산은 항상 0이어야 한다. <BR>
공분산이란? 두 변수 간의 상관된 정보를 나타내는 지표. 즉, 두 변수가 어떻게 함께 변하는지를 측정하는 값 <BR>
자기상관: 한 변수의 현재 값이 과거 값과 상관관계를 가지는 현상. <BR>
**시계열 자료**에서 많이 나타남.

가정6 - <span style="color:red">정규성</span>

오차항은 정규분포를 따른다.
* 가정 6은 위배되어도 다중선형회귀분석의 결과에 큰 영향을 주지 않는다.
* 오차항이 정규분포를 따라야 하는 이유 <BR>
회귀계수들의 신뢰구간과 p-value를 계산할 때 사용되는 공식은 사실 오차항이 정규분포를 따른다고 가정하고 만들어진 것임. 따라서 이런 공식들을 사용해서 p-value에 대해 유의미한 해석을 하기 위해서는 가정이 만족해야 함!



이 가정을 위배한다면 **다시 해야 함**


### <span style="color:green; background-color:white">회귀 분석의 평가 방법</span>

<span style="color:white; background-color:lightgreen">시각화</span> <BR>


시각화만으로는 어느 회귀선이 나은지 객관적으로 보기 어렵

<span style="color:white; background-color:lightgreen">통계 지표</span> <BR>


- F-statistic은 모델 전체에 대한 평가이다.
    - F-statistic가 클수록 통계적으로 유의미하다.
    - Prob(F-statistic)이 0.05보다 작다면 회귀 모델이 유의미하다.
- T-statistic은 개별 변수의 유의성에 대한 평가이다.
    - P > |t| (t통계량의 유의확률)가 0.05보다 작으면 해당 독립변수는 종속변수에 유의미한 영향을 미친다.

<span style="color:red">R스퀘어</span> <BR>
결정계수 $R^2$는 회귀 모델이 종속 변수의 전체 변동성을 얼마나 잘 설명하는지 나타내는 지표 <BR>
- **SST :** 평균($\bar{Y}$)으로 예측했을 때의 총 오차 (우리가 극복해야 할 전체 변동성)
- **SSE:** 모델($\hat{Y}$)이 예측하고도 여전히 남은 오차 (모델의 한계)
- **SSR:** 평균에 비해 **Regression** 모델이 얼마나 더 잘 맞혔는지 (모델의 성과) <BR>
 <BR>

- R스퀘어의 값은 0과 1 사이
- 1에 가까울수록, SSR과 SST의 값이 비슷해 지고, SSE는 0에 가까워지므로 모델이 데이터를 잘 설명한다고 할 수 있음
- R-squared 값이 0이면, 데이터들의 평균을 구한 것과 모델로 예측을 한 것이 동일한 것으로, 모델이 데이터를 전혀 설명하지 못한다고 봄

<span style="color:red">조정된 R스퀘어</span> <BR>

$R^2$는 독립 변수의 개수가 증가하면 증가할수록 자연스레 증가
→ 데이터와 큰 관련성이 없는 변수를 추가해도 $R^2$값은 높아지는 현상이 생김 <BR>



<span style="color:red">AIC, BIC</span> <BR>

두 지표 모두 정보 기준(information criteria)이라고도 불리며, 값이 낮을수록 좋다고 평가함


## <span style="color:green; background-color:white">비선형회귀</span>

<span style="color:white; background-color:lightgreen">1. 다항식 회귀 모델</span>



1. 기존의 선형 모델 $Y = a + bX$에 대해 $cX^2, dX^3$ 등을 포함하여 모델을 확장
2. 확장한 상태에 선형 회귀를 적용

- 곡선을 그리는 데이터 패턴에서 유용
- 성장률, 감소율 등이 시간에 따라 변화하는 속도가 다른 경우에 유용

<span style="color:white; background-color:lightgreen">2. 지수 회귀 모델</span>



- 종속 변수가 지수적으로 변화하는 관계를 모델링할 때 사용

1. 종속 변수 Y에 로그함수를 적용
2. 로그 변환된 Y와 기존의 독립변수 X에 대해 선형 회귀 적용
→ 모델이 ln Y(로그 변환된 Y)와 X의 선형 관계를 학습
3. Y값을 예측 시, 모델이 예측한 Y값에 지수함수$(e^x)$를 적용하여 로그 변환하기 전 Y로 변형

- 값이 시간에 따라 **지수적으로 증가하거나 감소**하는 경우 적합
- 기술 발전, 투자 수익률 증가 등

<span style="color:white; background-color:lightgreen">1. 로그 회귀 모델</span>



- 종속 변수와 독립 변수 간의 관계가 로그 함수를 통해 더 잘 표현될 때 사용

1. 독립 변수 X에 로그함수를 적용
2. 선형 회귀를 적용

- 데이터가 초기에 빠르게 증가하고 점차 증가율이 줄어드는 패턴을 보일 때 효과적
- 인구 성장, 감염병의 확산과 같이 초기 빠른 증가 후 점진적 감소를 보이는 데이터에 적합

<span style="color:white; background-color:lightgreen">1. 스플라인 회귀 모델</span>



- 데이터를 구간별로 나누고 각 구간에서 다른 선형 또는 비선형 함수를 적용하여 예측하는 방식

- 전체 데이터 범위를 여러 구간으로 나누고, 각 구간에 대해 별도의 회귀모델(선형, 다항 등) 적용
- 구간 경계에서의 **연속성을 유지**하는 것이 중요합니다.



- 데이터 패턴이 여러 구간에서 서로 다르게 나타날 때 유용
- 계절에 따라 판매량이 변화하는 소매 데이터 분석 등에서 적합ㄴ
