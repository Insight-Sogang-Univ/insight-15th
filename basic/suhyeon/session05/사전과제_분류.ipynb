## 3. <span style="color:white; background-color:hotpink">SVM</span>

SVM: 클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘

데이터를 분리하는 초평면(Hyperplane) 중에서 데이터들과 가장 거리가 먼 초평면을 선택하여 분리하는 지도 학습 기반의 이진 선형 분류 모델







- Support vector (서포트 벡터): 데이터 중에서 결정 경계와 가장 가까이에 있는 데이터들의 집합
- Decision Boundary (결정 경계): 데이터 분류의 기준이 되는 경계
- Margin (마진): 결정 경계에서 서포트 벡터까지의 거리(여유 공간)
- Hyperplane (초평면): $n$차원 공간의 ($n-1$)차원 평면
- Slack Variables (슬랙 변수/여유변수): 완벽한 분리가 불가능할 때 분류를 위해 허용된 오차에 대한 변수





정답은 F <BR>
이유: 결정 경계는 데이터로부터 가장 멀리 떨어진 것이 좋다. 즉, <span style="color:hotpink; background-color:grey">**Margin이 가장 큰 경우를 선택**</span>

## 3. <span style="color:white; background-color:hotpink">KNN 케이엔엔</span>

케이엔엔의 가정: 유유상종

<span style="color:hotpink; background-color:grey">**비슷한 특성을 가진 데이터끼리 서로 가까이 있는다는 점**</span>을 이용한 분류 알고리즘

"데이터로부터 <span style="color:yellow; background-color:grey">**거리**</span>가 <span style="color:yellow; background-color:grey">**r가까운**</span> <span style="color:yellow; background-color:grey">**k개**</span>의 다른 데이터 레이블을 참조하여 분류하는 알고리즘"

### k개의 이웃 설정 --> 가까운 k개의 데이터를 기반으로!

### 순서

1. 데이터 준비 (KNN은 미리 학습하는 과정 **없음**)

2. K값 설정
원하는 K값을 설정. K는 가장 가까운 이웃의 개수를 나타냄. 보통 <span style="color:yellow; background-color:grey">**홀수**</span>개로 설정

3. 거리 계산
새로운 데이터(예측하려는)가 주어지면, 이 값과 기존 모든 데이터 간의 거리를 계산
<span style="color:cyan; background-color:grey">유클리드 거리</span>나 <span style="color:cyan; background-color:grey">맨헤튼 거리</span> 등을 사용

<span style="color:cyan; background-color:grey">거리 메트릭 종류</span>임.

4. 가장 가까운 K개의 이웃 선택
계산된 거리 중에서 가장 작은 거리 값을 가진 K개의 데이터를 선택, 
이 데이터 포인트들이 "가장 가까운 이웃"

5. 분류하기
K개의 이웃 중 가장 많이 등장하는 클래스가 예측 결과가 됨

# <span style="color:crimson; background-color:white">분류 평가 지표</span>

(1) <span style="color:white; background-color:hotpink">분류평가지표 첫번째-혼동 행렬</span>



- **파란색 상자** : 예측을 올바르게 한 경우 (**예측과 실제 결과가 모두 일치**)
- **빨간색 상자** : 예측을 잘못한 경우 (**예측과 실제 결과가 불일치**)



<span style="color:white; background-color:hotpink">혼동 행렬</span>을 이용한 분류 모델 평가 지표: 정확도, 정밀도, 재현도
- 정확도, 정밀도, 재현도는 모두 0과 1 사이의 값





<span style="color:cyan; background-color:grey">**프리시전(정밀도)와 리콜(재현도) 의 트레이드 오프 관계**</span>

분류할 때, 확률에 기반해서 Threshold가 넘으면 참, 미만이면 거짓으로 판단
- 따라서 경곗값(threshold)을 조정하면 프리시전과 리콜을 조정할 수 있다.
- THRES 낮춤 -> Positive 예측 늘어남 -> 리콜 ⬆️, 프리시전 ⬇️
- THRES 높임 -> Positive 예측 감소함 -> 리콜 ⬇️, 프리시전 ⬆️



이런 관계~
-  두 value 값이 만나는 지점을 임계값으로 정할 때, 예측 오류를 최소화할 수 있음

(2) <span style="color:white; background-color:hotpink">분류평가지표 두번째-F1 스코어</span>
F1 스코어: **프리시전과 재현율의 조화 평균**



음~ 내일 정신 바짝 차려야겠다

(3) <span style="color:white; background-color:hotpink">분류평가지표 세번째-ROC / AUC Curve</span>

- ROC 커브: 얼마나 분류가 잘 되었는가를 보여주는 그래프



**True Positive Rate** (TPR) : 참인 것들 중에 참이라고 예측한 비율 (=Recall)
**False Positive Rate** (FPR) : 거짓인 것들 중에 참이라고 잘못 예측한 비율

- 어떻게 분류를 해도, 만족할 수 없는 부분이 생기기 때문에 완벽한 분류기는 나올 수 없음(ㅠㅠ)

- AUC Curve: ROC와 x축 사이의 면적(적분값)



모델의 성능을 숫자로 나타낼 수 있음. 0.5에서 1사이의 값을 가지며, <span style="color:hotpink; background-color:grey">**1에 가까울수록 분류 성능이 좋은 것**</span>

# <span style="color:crimson; background-color:white">하이퍼파라미터 최적화</span>

- <span style="color:white; background-color:hotpink">하이퍼파라미터</span>: <span style="color:hotpink; background-color:grey">학습 시작 전에 사용자가 직접 설정하는 변수</span>
- 하이퍼파라미터 최적화: <span style="color:cyan; background-color:grey">Tuning</span> 을 거쳐 <span style="color:hotpink; background-color:grey">적절한 하이퍼파라미터를 찾아</span> 모델 성능을 향상시키는 것



하/파 예시: decision tree의 max_depth,  random forest의 n_estimators  

### 과정

1. Hyperparameter 탐색 범위 설정 (최적 값을 찾고 싶은 하이퍼파라미터의 범위를 설정)
2. 평가 지표 계산 함수(성능 평가 함수) 정의(탐색하려는 Hyperparameter를 인수로 받아 평가지표 값을 계산해주는 함수를 정의)
3. 1단계에서 샘플링한 Hyperparameter 값을 사용하여 검증 데이터로 정확도 평가
4. 위 단계를 특정 횟수 반복하며, 정확도 결과를 보고 하이퍼파라미터의 범위 좁힘

### 방법(3가지-그리드, 랜덤, 베이지안)

1. 그리드 서치



정해진 범위에서 Hyperparameter를 모두 순회하며 가장 좋은 성능을 내는 값을 찾는 기법
- 단점: 시간이 너무 오래 걸린다

2. 랜덤 서치



정해진 범위에서 Hyperparameter를 무작위로 탐색해 가장 좋은 성능을 내는 값을 찾는 기법
- 속도는 빠른데 무작위라는 한계 때문에 **정확도가 떨어짐.**

3. 베이지안 최적화



사전 정보를 바탕으로 최적 Hyperparameter 값을 확률적으로 추정하며 탐색하는 기법

- "Gausain Process"라는 통계학을 기반으로 만들어진 모델로, 여러 개의 하이퍼 파라미터들에 대해서, "Aquisition Fucntion"을 적용했을 때, "가장 큰 값"이 나올 확률이 높은 지점을 찾아냄

- 과정(시행착오 반복인듯?): 초기값 몇 개 무작위 실험 -> 결과를 바탕으로 어떤 파라미터 조합을 쓸지 확률 모델로 예측-> 다음 실험할 조합을 가장 좋을 것 같은 위치로 선택 -> 결과 반영 후 다시 반복
