# **통계**
___

### 들어가기 전 알아보는 통계와 통계학의 의미

마지막 시간에 데이터 전처리를 배우면서 아웃라이어와 같은 이상치를 제거하는 것을 배우기도 했습니다. 이를 보면서 한가지 예시가 생각이 났었는데, 위 사진에 보이는 마이클 조던에 대한 예시입니다. 미국 UNC에서 전공별 연봉을 조사한 결과 지리학과가 제일 높았고, 그 이유는 마이클 조던의 연봉이 앞도적으로 높았기 때문에 그런 결과가 나왔습니다. 그래서 이때 조사결과를 봤을때에는 평균보다 중앙값이 더 신뢰할 수 있었습니다. 이 사례에서 연봉의 평균, 중앙값과 같이 숫자로 표현된 사실들을 통계라 부릅니다. 그리고 이 전에 평균대신 중앙값이 더 좋다고 생각하는 과정을 하기까지의 의시결정 과정처럼 자료를 수집하는 것 뿐 아니라 분석,표현하는하는 기술을 배우는 것을 통계학이라고 합니다.

# 1. 자료의 척도

### 자료란?
설명과 해석을 위해 수집된고 분석되는 사실을 말합니다.

### 자료의 척도
자료의 척도는 저희가 저번 시간에 배웠던 데이터의 종류와 같이 크게는 2가지 세부적으로 총 4가지로 나눌수 있습니다.
자료의 척도는 크게 연산의 의미가 없는 질적 자료와 수치로 표현되는 양적 자료로 나뉩니다. 그리고 질적 자료에서는 명목척도와 서열 척도, 그리고 양적 자료는 구간척도와 비율 척도로 나뉩니다. (아래 표와 같습니다.)
| 질적 자료 | 양적 자료 |
|-----|-----|
| 명목 척도 | 구간 척도 |
| 서열 척도 | 비율척도 |
- 명목 척도:
    이름이나 요소의 속성을 식별 하려는 요소입니다.
- 서열 척도:
    순서나 순위를 나타내는 속성입니다.
- 구간 척도:
    값들 사이에 순서가 있고, 간격이 일정해 연산이 가능합니다.
- 비율 척도:
    영점(0)과 임의의 측정단위로 구성합니다.

# 2. 기술통계

### 기술통계란?
데이터의 특성이나 패턴을 있는 그대로 보여주기 위해 아래 표의 있는 방법을 사용해 객관적 숫자로 간략히 보여주는 통계 방법입니다.


### 중심화 경향
중심화 경향이란 말 그대로 데이터의 중심적인 경향을 나타냅니다.

- 평균 

평균은 절대적인 수치를 자료의 수로 나눈것입니다. 
하지만 처음에 말씀드린 조던의 예시처럼 이상치에 민감하여 절사평균을 쓰기도 합니디.
    
    - 절사평균이란??
        데이터의 양끝을 잘라낸후 평균을 내는것으로 이번에 최가온님이 금메달을 딴 하프파이프 종목에서 점수를 낼때 사용하는 방식입니다.

- 중앙값

중앙값은 데이터를 크기로 정렬했을때 가운데에 있는 값입니다.(짝수라면 가운데 평균값)

- 최빈값

최빈값은 데이터중 제일 많이 등장한 값입니다.

### 자료의 산포

- 범위

범위는 최댓값과 최솟값의 차이로, 직관적으로 보기 쉬우나 이상치에 민감하다는 단점이 있습니다. 

- 분산

$$
\frac{\sum_{i=1}^{N}(x_i - \mu)^2}{N}
$$

편차를 제곱해서 더한 다음 자료의 수많큼 나눠준 것으로 위식과 같이 구합니다. 크면 클수록 데이터가 많이 퍼져있음을 의미하며 반대로 작을수록 데이터가 많이 모여있음을 뜻합니다.

- 표준편차
$$
\sqrt{\frac{\sum_{i=1}^{N}(x_i - \mu)^2}{N}}
$$
위와 같은 식으로 구하며 분산에 루트를 붙인 값입니다. 의미는 분산과 비슷하지만 루트를 붙이는 이유는 편차의 제곱을 더했기 때문에 직관적으로 알기 쉽기 위해서 루트를 붙였습니다.

- 사분위수

데이터의 표본의 
Q1 = 하위 25%
Q2 = 하위 50%
Q3 = 하위 75%
에 위치한 값들을 말합니다.

- 사분위간 범위

위에 사분위수에서 Q1에서 Q3까지의 값을 제외한 범위로 

    - 아래쪽 이상치: Q1 − 1.5×IQR 보다 작은 값
    - 위쪽 이상치: Q3 + 1.5×IQR 보다 큰 값
    로 설정해 이상치를 탐지합니다.

- 변동계수

상대적인 변동성을 알아보는 것으로 위에 살펴본 값중 표준편차와 평균을 사용해 구하며, 
공식은 (표준편차/평균)*100 입니다.

### 자료의 정리 및 시각화

- 첨도

첨도는 꼬리의 두께를 정규분포와 비교하는것이며 3을 기준으로 이보다 크면 양의 첨도,작으면 음의 첨도로 불리며,크면 클수록 이상치가 존재할 확률이 높습니다.

- 왜도

왜도는 자료의 비대칭성을 나타낸 값으로 중앙값과 평균이 같으면 0이며, 음수면 왼쪽에 양수면 오른쪽에 데이터가 쏠려있다는 뜻입니다.

# 3. 확률 검정

### 확률분포

먼저 확률변수와 확률분포란?

확률변수: 나타날 수 있는 모든 경우의 수를 숫자에 대응시키는 것으로 주사위를 던져을때 나오는 눈의 개수와 같이 생각할 수 있습니다.
확률분포: 확률변수가 특정한 값을 가질 확률로 위의 예시에서는 각 주사위의 숫자가 나올 확률인 1/6을 말합니다.

    - 이산확률분포 : 시험점수와 같이 유한한, 즉 셀 수 있는 값들을 가질때 사용합니다.
    - 연속확률분표 : 키나 온도와 같이 갯수가 무한한 값들을 가질떄 사용합니다.

### 대표적인 분포 모형

- 정규분포



정규분포란 대표적인 분포로 표본의 크기가 커질수록 정규분포와 가까운 분포를 띄게 됩니다. 공식은 아래와 같습니다.
$$
f(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \, e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

- Z분포



Z분포, 즉 표준정규분포는 많은 데이터들이 따르는 표본이며, 정규분포에서 평균이 0 분산이 1일때를 말하며 다른 분포들을 사용하기 위해 사용되는 가정으로 많이 쓰입니다.

- T분포




T분포는 모집단이 적을때 사용하는 분포로 자유도가 커질수록 정규분포로 나오며 공식은 아래와 같고, 표본이 커질수록 정규분포에 가까워 지는것을 알 수 있습니다.
$$
f(t) =
\frac{\Gamma\left(\frac{\nu+1}{2}\right)}
{\sqrt{\nu\pi}\,\Gamma\left(\frac{\nu}{2}\right)}
\left(1+\frac{t^2}{\nu}\right)^{-\frac{\nu+1}{2}}
$$

- 카이제곱분포



카이제곱분포는 이름에서 보이듯 확률변수들을 제곱해서 값을 구할때 자주 사용됩니다.전에 구했던 분산 역시도 제곱과 관련되있기 떄문에 이와 연관이 깊습니다.

- F분포





F분포는 위에 나온 두 카이제곱 분포를 비교할때 사용합니다.

# 4. 추론 통계

### 추론 통계란?
추론 통계는 표본이 신뢰할만 한지 아니면 이상한 표본만 뽑힌건 아닐지와 같은 질문에 답하며 점차 논리적으로 발전해 나가는 것이라고 볼 수 있습니다.

### 검정

먼저 가설 검정의 단계는 다음과 같습니다
 
 가설 설정 &Rightarrow; 유의 수준 설정 &Rightarrow; 검정통계량 산출 &Rightarrow; 가설 기각 혹은 채택 판단

- 가설 설정
가설 설정에 종류로는 귀무가설과 대립가설이 있습니다
    - 귀무가설: 기각하고자 하는 명제
    - 대립가설: 내가 주장하고자 하는 명제

- 유의 수준 설정
    유의수준이란 연구자가 결과를 판단할 떄 허용할 수 있는 오류의 최대로 0.05로 설정하였을때 5%확률로 오류가 발생할 수 있음을 뜻합니다.
        - 오류의 종류
            1종 오류: 귀무가설이 참인데 기각할 확률
            2종 오류: 귀무가설이 거짓인데 기각하지 못할 확률

- 검정통계량 산출
검정통계량이란 가설 검정의 결과를 판단할 때 쓰며 종류는 다음과 같습니다.
    - Z값: 절댓값이 클수록 이상치의 가능성이 높아짐을 뜻합니다
    - T값: 절댓값이 클수록 모집단과 표본 평균의 차이가 있을 가능성이 높아집니다.
    - F값: 두 표본의 분산비로서 1에 가까울수록 두 그룹의 퍼진정도가 비슷합니다.

- 가설 기각 혹은 채택 판단
    - P-값 계산 및 해석: 먼저 설정한 유의수준과 P값을 비교해 결론을 내립니다. 

### 데이터의 자료형에 따른 검정 방법

- 수치형 - 수치형
 
 공분산과 상관계수를 통해 알아봅니다.

    - 공분산이 1에 가까우면 서로 양의 관계 -1에 가까우면 음의 관계, 0이면 관계없다는 뜻입니다.
 
    - 상관계수는 절댓값이 크면 클수록 상관관계가 강하다는 뜻입니다.


- 범주형 - 수치형

T-test 나 ANOVA 검정을 통해 알아 봅니다
    
    - T-test: 범주가 두개일때 사용하며 t값은 2를 커지면 커질수록 두 범주가 차이가 크다고 생각합니다
    
    - ANOVA: 범주가 3개 이상일때 사용하며,평균대비 각 그룹간 차이를 알아낼 수 있습니다.

- 범주형 -범주형
    
    관측빈도나 기대빈도를 통해 카이제곱 통계량을을 구하고 자유도의 두배를 기준으로 판단합니다.
