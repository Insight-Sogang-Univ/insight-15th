# 1. 머신러닝
---

### 먼저 머신러닝이란?
- 단어의 뜻: 소프투웨어를 학습해 새 데이터를 예측하거나 결정을 내리게 해줍니다.
- 활용: 데이터에서 패턴을 발견하는 것과 이 패턴을 바탕으로 새로운 데이터에 대해 예측하게 해줍니다.

### 머신러닝의 종류



위 그림과 같이 머신러닝은 크게 3가지로 나뉩니다
- 지도 학습
    - 지도학습은 기계의 예측값이 데이터의 정답과 같아지도록 학습합니다. 예를 들어 이메일을 받앗을 때 스팸을 분류한다던가 집의 크기와 위치를 보고 집값을 예측한다던지와 같이 정답이 있는 데이터를 통해 학습합니다.
- 비지도 학습
    - 비지도 학습은 지도학습과 반대로 정답이 없는 데이터에서 패턴이나 구조를 찾는 방법입니다. 예시로는 유튜브 알고리즘이 있는데, 이용자들의 시청 패턴을 보고 비슷한 취향의 사람들 끼리 묶어 그 그룹이 좋아한 영상을 추천해 주는것이 비지도 학습의 예시입니다.
- 강화 학습
    - 강화학습은 보상을 받으면서 더 좋은 행동을 배우는 것으로 강아지를 훈련시키는 것과 유사합니다. 예시로는 자율주행이 있는데, 사고없이 달리면 보상을 사고가 나면 패널티를 부여함으로써 더 안전한 주행을 하게하는 것이 강화 학습의 예시입니다. 


### 모델링
- 모델
    - 모델은 학습의 결과물로 학습을 마친 인공지능입니다. 
- 모델링
    - 모델링은 모델은 만들어가는 과정을 말합니다. 오차를 줄여가는 과정이라고 볼 수 있습니다.
- 모델링의 과정
    - 과정은 크게 세단계로
    설계 → 훈련 → 검증 순으로 진행됩니다.
        - 설계: 데이터의 특성을 파악하고 이에 맞추어 어떤 알고리즘으로 학습할지 결정합니다
        - 훈련: 데이터를 기계에 보여줘 정답을 맞히게 수정하는 단계입니다.
        - 검증: 훈련이 끝난 모델이 새로운 데이터에서도 잘 맞히는지 테스트합니다.

# 2. 회귀
---

### 회귀란?
- 위에서 다루었던 지도 학습중 하나로 데이터에서 패턴을 찾아서 미래의 값을 예측합니다.

### 선형회귀
선형 회귀란 가중치(W) 와 편향(B)을 이용해 데이터를 잘 표현하는* 하나의 직선을 찾는 과정입니다.
여기서 회귀선의 공식은 다음과 같습니다.

$
y =  Wx+B
$

**여기서 회귀선을 통해 데이터 요약 및 예측을 진행할거기 떄문에 매우 중요합니다.*

여기서 두 변수인 가중치와 편향에 대해 먼저 알아 보겠습니다.
    
    - 가중치(W) : 독립변수인 x가 종속변수인 y에 주는 영향력의 크기로 회귀계수로 라고도 불립니다.
    
    - 편향(B) : 모델이 더 유연하게 최적화할 수 있도록 위아래로 움직여 줍니다.

이제 선형회귀의 종류를 알아보겠습니다. 크게 두가지가 있는데
    
    - 단순선형회귀: 하나의 변수로 하나의 결과를 예측하는 것입니다.
        - Ex) 우유섭취량 -> 키
    
    - 다중선형회귀: 여러 변수로 하나의 결과 예측하는 것입니다.
        - Ex) 아빠의 키, 엄마의 키 -> 아들의 키 

### 단순선형회귀에서 최소제곱법과 경사하강법
이제 가중치와 편향을 잘 조정해서 오차를 최대한 줄이는 방법을 알아볼 것입니다. 제목을 보면 알 수 있듯 두가지 방법이 대표적입니다.
- 최소제곱법



위와 같은 그림에서 오차의 제곱의 합들이 최소가 되는 W와 B를 찾는 것입니다.

- 경사하강법

경사하강법역시 오차를 최소화하는 가중치와 편향을 찾기 위한 최적화 방법입니다.

이때 새로운 가중치 $W'$ 은 기존 가중치 $W$에 학습률  
$\eta$ 를 오차함수 기울기와 곱하여 뺴준 값을 사용하며 식은 아래와 같습니다.

$
W' := W - \eta \frac{\partial J(W)}{\partial W}
$

임의로 학습률 지정하고 공식을 사용하면 됩니다.

이 공식을 사용하는 이유는



이차함수의 특성상 위 사진과 같이 최소가 되는 점을 기준으로 왼쪽에 있으면 그래프의 기울기가 음수이고, 오른쪽에 있으면 그래프의 기울기가 양수가 됩니다. 

따라서 기존 가중치가 왼쪽에 있어서 그래프의 기울기가 음수였다면 상수(>0)×오차함수 기울기를 빼주는, 즉 하나의 양수를 더해줘 오른쪽으로 이동시킵니다.

반대로 그래프의 기울기가 양수였을때는 상수(>0)×오차함수 기울기를 빼서 왼쪽으로 옮기는 과정을 수행하여 점점 오차 최소지점에 도달하는 것이 경사하강법입니다.

추가로 여기서 학습률을 잘 조정해야하는데, 
먼저 학습률을 너무 작은 경우, 아래 그림과 같은 상황이 발생합니다.



그러면 학습은 확실히 하지만, 시간이 오래 소요되는 단점이 있습니다.

반대로 학습률이 너무 큰 경우에는 아래와 같은 상황이 발생합니다.



오차 최소지점에 오히려 멀어지는 상황이 발생합니다. 따라서 학습률을 임의로 몇개 지정해보고 그 중 택하는 방법으로 학습률을 정합니다.

### 다중회귀속 다중공산성

- 다중공산성이란?
    - 다중공산성이란 두 독립변수가 서로 강한 영향을 미치는 상황을 말하는 것으로 예를 들어 시험점수를 예측할대 공부시간과 앉는시간을 둘 다 반영 했을떄 공부시간과 앉아있는 시간끼리도 강한 상관관계가 있으므로 어떤 독립변수가 강한영향을 끼치는지 알기 어렵습니다.
- 확인법
    - 상관계수 활용: 전에 배웠던 상관계수를 이용해 서로 강한 영향을 끼칠거 같은 두 독립변수의 상관계수를 구하여 확인합니다
    - VIF지수 활용: 회귀 모델의 결정계수 R-Square를 사용하여 VIF지수를 구하고 이 값들을 통해 확인합니다.
- 대처법
    - 변수 제거
    - 변수 변환
        - 변수들을 더하거나 빼서 새로운 변수 생성
        - 독립변수를 더하거나 빼더라도 문제가 없는 경우
    - 규제선형모델 활용
        - 나중에 다룰 릿지, 라쏘, 엘라스틱넷 등의 규제선형모델을 활용할 수 있습니다. 
    - PCA(주성분 분석)
        - 상관 있는 변수들을 묶어서 더 적은 수의 대표 축으로 바꾸는 것으로, 고차원 데이터에서 중요한 정보를 최대한 보존하면서 데이터의 복잡성을 줄이는 것입니다.



### 규제선형모델

- 규제선형모델이란 선형회귀 모델에 패널티(규제)를 추가해서 과적합을 막는 방법입니다.
    - 과적합: 너무 많은 데이터를 학습해 편향이 생기는 것으로 예를 들어 강아지를 인식하는데 검은 강아지를 너무 많이 인식해 다른 검정 동물을 보고도 강아지라고 인식하는 것입니다.
- 규제선형모델의 종류
    위에서 잠깐 다뤘듯이 릿지,라쏘 엘라스틱넷의 세가지 종류가 있습니다.
    - 릿지:  회귀계수의 제곱 합에 페널티를 부여해 계수를 적게 만듦니다.
    - 라쏘: 필요 없는 변수의 계수를 0으로 만들어 변수를 선택할 수 있게 해줍니다.
    - 엘라스틱넷: 릿지와 라쏘를 섞은것입니다.

# 3. 모델 평가 방법
---

### 성능 평가 지표

성능평가지표는 크게 두가지가 있습니다.
    
    - 평균 제곱 오차
        - 예측값과 실제값 사이의 오차 제곱의 평균을 의미합니다.

    - 평균 절대 오차
        - 모델의 예측값과 실제값 사이의 절대값 오차의 평균을 의미합니다.

두 지표중 평균제곱오차는 이상치에 민감하게 반응하지만 평균 절대 오차는 이상치가 있어도 성능을 유지하는 만큼 이상치를 어떻게 다룰거냐에 따라 다르게 사용합니다.

### 변수 유의성 평가

마지막시간에 배운 T검정을 활용해 t값과 pvalue를 활용해 평가합니다.
