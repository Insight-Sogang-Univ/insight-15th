# 1. 다중선형회귀
---

### REVIEW 다중선형회귀란?
- 수식: $y_i = \beta_0+\beta_1X_{1i}+\beta_2X_{2i}+...+\beta_kX_{ki}+\varepsilon_i$
- 의미: 여러 독립변수를 채택하여 선형적으로 식을 작성한 것입니다. 각 변수의 가중치를 곱해 값을 비교해가면서 어떤 변수가 종속변수를 잘 성명하는지도 알 수 있습니다. 하지만 모든 상황에서 선형회귀를 쓸 수 있는것은 아니며 선제되어야 하는 가정에 대해서 알아보겠습니다. 

### 다중선형회귀의 기본가정 6가지



기본가정은 6가지가 필요합니다.
- 선형성
    - 종속변수와 설명변수 간의 관계가 선형적이여야 한다는 가정입니다.
- 독립성
    - 저번시간에 배운 다중공산성을 따져보고 결과적으로 서로 선형독립적이여야 한다는 가정입니다.
- 오차항의 평균은 0이다
    - 실제 값과 예측 값 사이의 차이가 0에 가까워야 한다는 가정입니다.
- 등분산성
    - 오차항의 분산이 일정해야 한다는 가정으로 회귀 분석의 정확성을 위해 필요합니다.
- 오차항은 자기상관되어 있지 않다.
    - 오차항의 공분산은 0이 되어야 한다는 가정으로 과거의 값이 현재의 값과 상관이 없어야 한다는 과정입니다.
- 정규성
    - 오차항이 정규분포를 따른다는 가정으로 위배되더라도 분석의 결과에 영향을 크게 주지 않지만, p-value 값이 잘못 계산될수 있기에 지켜줘야 한다.

### 분석 전에 검증해야 할 가정과 분석 후에 검증해야 할 가정
- 선형성
    - 전,후 둘다 가능
- 독립성
    - 전,후 둘다 가능 
- 오차항의 평균은 0이다
    - 후에만 가능
- 등분산성
    - 후에만 가능
- 오차항은 자기상관되어 있지 않다.
    - 전에는 간접적으로 가능 후에는 직접적으로 가능
- 정규성
    - 후에만 가능



### 회귀분석 평가 방법
- 시각화

    - 데이터를 한눈에 보기 쉽게 표현하는것으로,직관적으로 보기 쉽지만, 객관적으로 비교하기는 어렵습니다.
- 통계지표
    -  모델의 성능 평가
        - F검정:
            회귀도델이 유의미한지 검정하는것으로,  p-value가 0.05보다 작으면 유의미하다.
        - T검정:
            특정독립변수가 유의미한지 검정하는 것으로, F검정때와 같이 p-value가 0.05보다 작으면 유의미합니다.
    -  모델 유의성 검정
        - 결정계수:
            종속 변수의 전체 변동성을 설명하는 지표로, 1에 가까워질수록 모델이 데이터를 잘 설명하며, 0에 가까워질수록 모델이 데이터를 설명하지 못합니다.

# 2. 비선형회귀
---

### 비선형회귀란?
- 비선형회귀란 독립변수와 종석변수의 직선이 아닌 곡선 형태를 쓰는 것으로 $ y = ax + b $ 형태가 아니라 $ y = ae^{bx} $, $ y = a + b\ln(x) $, 혹은 $ y = ax^2 + bx + c $ 와 같이 곡선의 형태로 나타나며 종류는 아래의 4종류로 나뉩니다.
    - 다항식
    - 지수
    - 로그
    - 스플라인



### 다항식 회귀 모델


- 사용: 주로 곡선을 그리는 데이터에서 자주 쓰이며, 시간에 따른 추이를 볼때 자주 사용됩니다. 
- 장점과 단점: 독립변수의 거듭제곱항을 사용하며 회귀선보다 좀 더 정확하게 예측 할 수 있지만, 과적합이 될수 있어 규제가 필요합니다.
- 공식
$$
y = a_0 + a_1 x + a_2 x^2 + \cdots + a_n x^n
$$

### 지수회귀모델



- 사용: 기술발전이나 투자 수익률 같이 지수적으로 증가하거나 감소하는 경우에 자주 사용됩니다.
- 장점과 단점: 해석이 쉽고, 초기 데이터를 잘 설명하나, 후반부로 갈수록 급격히 기울기의 절댓값이 커져 장기적으로는 과장될수 있습니다.
- 공식
$$
y = a e^{bx}
$$

### 로그회귀모델



 - 사용: 공부시간에 따른 시험점수 같이 초기 빠른증가후 점진적으로 증가하는 데이터에 사용됩니다. 
 - 장점과 단점: 한계효과를 설명 할 수 있다는 장점이 있지만, 초기 데이터에 민감해 초기의 이상치가 있을시에 부정확할 수 있습니다.
 - 공식
 $$
y = a + b \ln(x)
$$


### 스플라인 회귀



- 사용: 계쩔별 상품 판매량 패턴이 여러가지인 데이터를 분석할때 유용합니다.
- 장점과 단점: 데이터 패턴이 여러가지인 데이터를 분석할 수 있을만큼 많은 데이터를 설명할 수 있다는 장점이 있지만, 그만큼 모델이 복잡하다는 단점이 있습니다.
- 공식 
$$
S(x)=
\begin{cases}
f_1(x), & x_0 \le x < x_1 \\
f_2(x), & x_1 \le x < x_2 \\
\cdots \\
f_n(x), & x_{n-1} \le x \le x_n
\end{cases}
$$
