# 1. 분류의 정의
---

### REVIEW 회귀와 분류
- 회귀: 데이터가 주어졌을때 이를 기반으로 정답을 잘 맞추는 함수
- 분류: 기존 데이터가 어떤 레이블에 속하는지 패턴을 파악하고 이에 맞게 새로운 데이터의 레이블을 판별하는 것

- 차이점

    회귀는 연속형 변수를 예측하지만, 분류는 범주형 변수를 예측한다.

---
# 2. 분류에 쓰이는 모델
---

### 로지스틱 회귀


- 정의: 독립 변수의 선형 회귀에 로지스틱 함수를 적용히여 출력값을 0에서 1사이로 변환해주는 것을 의미합니다.

- 사용이유 직선회귀 출력시 음의 무한대부터 양의 무한대까지 값을 갖는다는 점과 모델의 급격한 변화를 방지하기 위해 이를 보완한 시그모이드 함수를 적용해 값을 0~1사이로 압축시키고 현실적인 확률 값을 구한다.

- 수식
$$
\sigma(x) = \frac{1}{1 + e^{-x}}
$$

- 장점: 해석이 쉽고 계산이 빠르다

- 단점: 복잡한 데이터에서는 성능이 낮다

### 결정 트리 



- 정의: 조건에 따라 데이터를 조건에 따라 나눠 트리구조로 분류한다.

- 사용이유: 한눈에 보기 쉬우며, 전처리 과정이 거의 필요없다. 

- 불순도: 같은 클래스끼리 모일시 순도가 높아지는데, 그 반대개념이다.
    - 지니 불순도: 통계적 분산 정도를 정량화해서 나타낸 것입니다.
    - 정보이득: 불한 전,후의 불순도 감소량

- 장점: 데이터 전처리 과정이 거의 필요없고 비선형 관계도 잘 나타낸다

- 단점: 과적합이 잘 발생하며, 작은 데이터 변화에도 구조가 크게 변한다.

### 서포트 백터 머신


- 정의: 데이터를 가장 잘 구분하는 초평면 경계를 찾는 모델

- 사용이유: 고차원 데이터를 다루기 위해서 사용된다.

- 구성 
    - Support vector (서포트 벡터):데이터중 초평면과 가장가까이 있는 데이터들의 집합
    - Decision Boundary (결정 경계): 데이터 분류의 기준
    - Margin (마진): 결정경계와 서포트 백터 사이 거리
    - Hyperplane (초평면): n차원 공간의 (n-1)차원 평면
    - Slack Variables (슬랙 변수/여유변수): 오차에 대한 변수

- 장점: 과적합에 비교적 강하며, 소량 데이터에서도 성능을 유지한다.

- 단점: 데이터 셋이 클수록 느려지며 결과해석이 어렵다.

### 최소 근접

- 정의: 새로운 데이터를 받았을때 근처의 데이터를 통해 예측하는 모델

- 사용이유: 구현이 쉽고 직관적이다

- 순서 
    - 데이터 준비
    - K값 설정: 몇개의 데이터를 비교할기 미리 정한다.
    - 거리계산: 유클리드 거리, 멘해튼 거리, 혹은 민코프스키 거리같이 하나의 거리를 정하고 데이터간의 거리를 계산한다.
    - K개의 데이터 선택
    - 분류하기

- 장점: 학습과정이 필요없고,추가된 데이터 처리가 쉽다. 

- 단점: 계산량이 많고,이상치에 민감하다.

# 3. 분류평가지표
---

### 혼동행렬


- 정의 모델의 예측과 결과를 정확한 예측과 잘못된 예측으로 구분

| 구분 | 의미 | 설명 |
|------|------|------|
| TP (True Positive) | 정답 맞춤 | 모델이 참(P)으로 예측했고 실제도 참 |
| FP (False Positive) | 오답 | 모델이 참(P)으로 예측했지만 실제는 거짓 |
| FN (False Negative) | 오답 | 모델이 거짓(N)으로 예측했지만 실제는 참 |
| TN (True Negative) | 정답 맞춤 | 모델이 거짓(N)으로 예측했고 실제도 거짓 |

- 정확도: 모든 예측중 참인 비율 (TP+TN) / (TP+TN+FP+FN)

- 정밀도: 예측한 것중 참인 비율 (TP) / (TP+FP)

- 재현도: 실제 참인 것중 예측도 참인 비율 (TP) / (TP+FN)


### F-1 score

머신러닝 모델의 성능을 평가하는 주요지표입니다. 조화평균을 사용하는 이유는 정밀도와 재현율 간의 균형을 효과적으로 평가하기 위해서입니다.

### ROC curve,AUC curve


- ROC curve: 얼마나 분류가 잘 되었는가를 나타내며 직각으로 꺾일수록 분류가 잘된것입니다.

- AUC curve: ROC와 x축 사이의 면적으로 모델의 성능을 수치로 알 수 있으며, 1에 가까울수록 좋습니다.

# 4. 하이퍼파라미터 최적화
---

### 하이퍼파라미터 정의


하이퍼파라미터란 사용자가 직접 설정하는 변수로 경사하강법의 학습률이 대표적인 예시입니다.

### 하이퍼파라미터 최적화 과정



탐색범위설정 -> 평가지표계산함수 정의 -> 검증데이터를 통한 정확도 평가 

위 단계들을 반복하여 정확도 결과를 보고 범위를 좁힙니다.
(경사하강법에서와 같습니다.)

### 하이퍼파라미터 최적화 방법

최적화 방법에는 총 3가지가 있는데 다음과 같습니다.

- Grid Search: 정해진 범위에서 모두 순회
    - 장점: 정확하다
    - 단점: 시간이 오래걸린다
    - 예시:
    
    학습률: [0.01, 0.1, 1]
    편차: [0.01, 0.1, 1]

    이라고 가정하면, 가능한 조합은 다음과 총 9개로
    (0.01, 0.01), (0.01, 0.1), (0.01, 1),
    (0.1, 0.01), (0.1, 0.1), (0.1, 1),
    (1, 0.01), (1, 0.1), (1, 1)

    이 모든 조합에 대해 모델을 학습하고, 그중 가장 성능이 좋은 값을 선택한다.
- Random Search: 정해진 범위에서 무작위로 탐색
    - 장점: 시간이 빠르다
    - 단점: 정확도가 떨어진다.
    - 예시: Grid Search와 같은 예시에서 범위를 정해두고, 그중에서 랜덤하게 4개만 뽑아서 실험한다고 하면
(0.01, 1), (0.1, 0.01), (1, 0.1), (0.1, 0.1)
이런 식으로 일부만 선택해서 학습한다.

- Bayesian Optimization: 사전 정보를 바탕으로하여 하이퍼파라미터값을 확률적으로 추정하며 탐색
    - 방법: 초기에 랜덤으로 실행한후 패턴을 보고 좋을거 같은 조합을 예측해 값을 찾는다.
    - 예시: 
    처음에 아래와 같이 4개를 랜덤으로 시도했다고 하자:

    (0.01, 0.01), (0.1, 0.1), (1, 1)

    이 결과를 보고
    “학습률은 0.1 근처가 좋고, 편차는 0.01~0.1 사이가 좋아 보인다”
    라고 판단하면, 다음에는
    (0.1, 0.01), (0.05, 0.1) 같은 좋을 것 같은 영역을 집중적으로 탐색한다
