# 1. 군집화의 정의
---

### REVIEW
- 지도학습 
    훈련 데이터를 사용해 입력값과 출력값사이의 

- 비지도학습
    정답이 없는 데이터에서 기계가 스스로 구조, 패턴, 숨겨진 특징을 찾아내는 머신러닝 방식

### 지도학습 vs 비지도학습

| 구분 | 지도학습 | 비지도학습 |
|------|--------|-----------|
| 정답(label) | 있음 | 없음 |
| 목적 | 예측 | 패턴 발견 |
| 대표 문제 | 회귀, 분류 | 군집화, 차원축소 |
| 예시 | 스팸메일 분류 | 고객 그룹 나누기 |

### 군집화
- 정의: 데이터를 비슷한 특성을 가진 그룹(군집)으로 나누는 비지도 학습 기법. 

### 군집

- 정의: 유사한 데이터들의 조합

### 군집화 과정
- 피처 선택,추출 -> 군집화 알고리즘 선택 -> 군집 유효성 검증 -> 결과 해석
    -  피처 선택 / 추출  
    군집화를 수행하기 전에 어떤 데이터를 사용할지 결정하는 단계
    - 군집화 알고리즘 선택  
    데이터의 특성에 맞는 군집화 알고리즘을 선택하는 단계
    - 군집 유효성 검증  
    생성된 군집이 얼마나 잘 나뉘었는지 평가하는 단계 
    - 결과 해석  
    군집화 결과를 실제 의미 있는 정보로 해석하는 단계.

# 2. 거리 / 유사도 정의와 측정
---

## 데이터의 거리 측정
    - 군집화에서 데이터 포인트나 군집간의 유사성을 계산하는 방법입니다. 거리가 가까울수록 비슷하고 멀수록 다릅니다.

### 유클리디안 거리

    
    - 두점사이의 직선거리 
    
    - 수식
$$
d(x, y) = \sqrt{\sum_{i=1}^{n}(x_i - y_i)^2}
$$ 거리
### 맨하탄 거리

    
    - 각 차원사이의 절댓값
    
    - 수식
$$
d(\mathbf{x}, \mathbf{y}) = \sum_{i=1}^{n} |x_i - y_i|
$$
### 코사인 유사도

    
    - 백터간의 각도
    
    - 수식
$$
\text{similarity} = \frac{\mathbf{x} \cdot \mathbf{y}}{||\mathbf{x}|| \times ||\mathbf{y}||}
$$

### 6가지 연결 방법
- 단일 연결법:
    
    군집 사이의 가장 가까운 두점 간 거리로 정의 -> 고립된 군집 찾기
- 완전 연결법:
    
    군집 사이의 가장 먼 두점 간 거리로 정의 -> 내부 응집성에 중심 
- 평균 연결법:
    
    모든 항목에 대한 거리 평균 -> 이상치에 민감하지 않다.
- 중심 연결법:
    
    가중평균을 통해 구한 중심점 사이의 거리 -> 계산 오래걸림
- 중앙 연결법:
    
    두 군집의 중심점 사이의 산술적 중간 지점 -> 이상치에 민감하지 않다.
- Ward 연결법:
    
    제곱합이 증가하지 않도록 병합 -> 분산 최소화


### 차원축소


- 목적

    차원이 높아짐 -> 거리가 길어짐 -> 거리의 분별력 감소 -> 차원축소를 통해 거리기반 알고리즘 분별력 확보 밒 데이터 패턴 인식

차원 축소 유형 
    
    - 선형 투영
    고차원 공간에서 선형적으로 저차원 공간에 데이터를 투영
    
    - 비선형 차원 축소
    데이터 포인트들 사이의 곡선 거리를 고려해 학습

# 3. 주성분 분석(PCA)
---

### PCA란?


위에 나온 차원축소 기법으로 분산을 보존하며 상관관계가 없는 적은 수의 피처(쥬성분)로 표현하여 시각화와 분석에 용이하게 만드는 것
    
- 수행과정 
    
    
1. 데이터 표준화  
    → 변수 간 스케일 차이 제거 (필수)

2. 공분산 행렬 계산  
    → 변수 간 관계 파악

3. 고유값 & 고유벡터 계산  
    → 데이터의 주요 방향(축) 찾기

4. 주성분 선택  
    → 설명 분산 비율 기준으로 선택

5. 데이터 투영 
    → 선택된 주성분으로 차원 축소

- 주의사항 

    표준화가 필수적이고 결측값이 있을시 사용할 수 없으며, 해석이 어렵고 차원축소 특성상 정보가 손실될 수 있다.

# 4. 군집화 알고리즘 특징
---

### 군집화 알고리즘

데이터 간의 유사도를 기준으로 비슷한 데이터끼리 군집으로 묶는 비지도 학습 알고리즘

### 


### 계층적 군집화



데이터 간의 유사성 기반으로 트리 구조 형성을 하며, 상향 또는 하향식으로 군집을 형성
- 특징

    원하는 군집의 개수 선택 가능
    상향식이냐 하향식이냐에 따라 응집형과 분리형으로 구분

- 응집형  
    각 데이터 → 하나의 군집으로 시작 → 가장 가까운 군집끼리 점차 병합  
    
    위 과정을 하나의 군집이 될때까지 반복합니다. 

- 분할형  
    전체 데이터를 하나의 군집으로 시작 → 점차 작은 군집으로 분할  



### k-means



데이터를 정해진 개수의 그룹으로 나누고 중심점 거리가 가까운 데이터 끼리 묶는 알고리즘

직관적이고 구현이 쉽지만, 이상치에 민감하다.

- k-means++

    k-means의 단점 중 하나인 초기 중심점 선택 문제를 개선하기 위해 제안된 방법이다.
    초기 중심점을 무작위로 고르는 대신, 서로 멀리 떨어진 점들이 중심점으로 선택될 가능성을 높이는 방식을 사용해 더 안정적인 군집화를 유도한다.
    이를 통해 k-means보다 수렴 속도와 결과 품질이 좋아지는 경우가 많다.

### DBSCAN



밀도가 높은 지역의 데이터를 하나의 군집으로 묶고, 다른 점들은 군집에 포함시키지 않는 알고리즘

- 특징 
    
    이상치에 민감하지 않고, 모든 샘플에 클러스터를 할당하지 않지만, 속도가 느리고, 변수가 늘어날수록 정확도가 떨어짐
- HDBSCAN
HDBSCAN은 DBSCAN을 계층적 군집화 방식으로 확장한 알고리즘  

-  작동 방식
1. 다양한 밀도 기준으로 DBSCAN 수행  
2. 군집 간 연결 관계로 계층 구조(dendrogram) 생성  
3. 가장 안정적인 군집 선택  



### Gaussian Mixture Model (GMM)



여러 개의 가우시안 분포를 합쳐 표현하며, 데이터의 생성 메커니즘까지 모델링 가능하다.

- 기본가정
    관측된 데이터-> 특정 가우시안 확률분포에 의해 생성
    전체 데이터셋-> 여러개의 다변량 가우시안 분포가 섞임
    개별 데이터-> 우도에 따른 K개의 가우시안 분포 중 하나

K-means가 잘 작동하지 않는 타원분포나, 중첩된 군집 구조에서 좋은 성능을 내지만, 군집화하는 시간 오래걸린다.

# 5. 군집화 평가 방법
---

### 외부평가와 내부평가
- 외부평가
    정답레이블이 존재하는 경우 사용하며 군집화 결과와 정답을 비교해 평가
- 내부평가
    정답레이블이 존재하지 않을 경우 사용 응집도와 분리도를 기반으로 판단하며
    알고리즘별로 사용하는 내부평가지표와 범용적으로 사용하는 내부지표가 있다.

### 실루엣 계수
클러스터내 샘플들이 얼마나 조밀한지 측정합니다.
- 구성 요소

- $a^{(i)}$ : 같은 클러스터 내 다른 점들과의 평균 거리  
  → 작을수록 좋음 (응집도 ↑)

- $b^{(i)}$ : 가장 가까운 다른 클러스터와의 평균 거리  
  → 클수록 좋음 (분리도 ↑)

$$
s^{(i)} = \frac{b^{(i)} - a^{(i)}}{\max \{ b^{(i)}, a^{(i)} \}}
$$
- 값의 범위와 해석
  
  값이 1에 가까울수록 좋고 -1에 가까울수록 좋지 않다.



### Dunn Index

클러스터간 최소 거리와 최대거리의 비율을 계산해 클러스터링의 품질을 평가

- 구성 요소

- $\delta(C_i, C_j)$ :  
  서로 다른 두 클러스터 $C_i, C_j$ 간의 거리 (보통 최소 거리 사용)  
  → 클수록 좋음 (클러스터 간 분리도 ↑)

- $\Delta(C_l)$ :  
  클러스터 $C_l$ 내부의 최대 거리 (지름, diameter)  
  → 작을수록 좋음 (클러스터 내부 응집도 ↑)

- 값 해석
  
  값이 클수록 좋습니다.
