# 1. 연관 분석
---

### REVIEW
- 비지도학습
    정답이 없는 데이터에서 기계가 스스로 구조, 패턴, 숨겨진 특징을 찾아내는 머신러닝 방식
---
### 연관분석이란?
    비지도 학습의 일종으로,데이터 안에서 같이 나타나는 패턴을 찾는 기법으로 상품이나 서비스를 구매할때 사건 안에 존재하는 항목간의 일정한 연관 규칙을 발견하는 분석이다.
---
### 연관규칙
    만일 ~ 한다면과 같은 조건절과 ~한다의 결과절로 아루어져 있으며, 아이템집하은 두 절을 구성하는 아이템의 집합이다.
---
### 연관규칙 vs 인과관계
| 구분 | 연관규칙 (Association Rule) | 인과관계 (Causality)       |
| -- | ----------------------- | ---------------------- |
| 정의 | 함께 나타나는 패턴을 찾는 것        | 한 변수가 다른 변수에 영향을 주는 관계 |
| 의미 | A가 있으면 B도 자주 나타남        | A가 B를 발생시킴             |
| 해석 | 단순 동시 발생 (상관 수준)        | 원인 → 결과 관계             |
| 예시 | 맥주를 사면 기저귀도 산다          | 광고를 하면 매출이 증가한다        |

---
### 지지도, 신뢰도, 향상도



연관규칙의 조건으로는 일정이상의 지지도와 신뢰도가 있어야 한다.

- 지지도란?
    전체 데이터 중 두 A와 B가 함께 등장하는 비율 
$$
support(A \rightarrow B) = P(A \cap B)
$$

- 신뢰도란?
    A가 있을때 B가 같이 나올 확률 (=조건부확률)
$$
confidence(A \rightarrow B) = P(B \mid A) = \frac{P(A \cap B)}{P(A)}
$$

이를 통해 연관분석을 진행했다면, 평가를 해야하는데 향상도가 연관 분석의 평가 측도가 된다.
- 향상도
    A와 B가 얼마나 관련 있는지 보는 지표로 우연인지 진짜 관계가 있는지 판단하기 위해 사용한다.
$$
lift(A \rightarrow B) = \frac{P(A \cap B)}{P(A)P(B)}
$$

lift 가 1이 넘으면 같이 나올 가능성이 높으면, 1이면 서로 독립 1보다 작으며뉴 같이 안나온다.

---
### 연관 분석 알고리즘이란?
    데이터에서 자주 같이 등장하는 항목 조합을 찾고, 규칙을 만드는 알고리즘
---
### Apriori 알고리즘


- 소개
    효율적으로 팀색하기 위해 자주 발생하지 않는 특정조합을 베제하는 알고리즘
- 과정
    단일 항목 집합 생성 -> 2개 항목 집합 생성 -> 최소 지지도 못넘는 조합 제거 -> 큰 항목 집합 생성 -> 최소 지지도 못넘는 조합 제거 -> 큰 항목 집합생성 밒 최소 지지도 못넘는 조합 제거 반복 -> 최종 반발 항목 집합 선정
- 장점
    이해분석에 용이하다
- 단점
    데이터가 커질수록 속도가 느려진다.
---
### FP-Growth


- 소개
    데이터가 커져도 속도를 유지하기 위해 FP tree 구조를 이용해 Apriori 효율적으로 이용한다
- 과정
데이터셋을 확인 -> 빈도순으로 재정렬 -> 거래를 노드로 표현 -> 동일 항목 등자시 빈도 증가, 새로운 항목 등장시 새로운 노드로 추가 -> FP-트리를 바탕으로 특정 항목 기준으로 부분 트리를 만든다 -> 이를 바탕으로 연관 규칙을 만든다. -> 지표들을 통해 가장 유용한 규칙을 선택한다.
- 장점
    Apriori 보다 훨씬 빠르다.
- 단점
    설계가 어렵고, 대용량 데이터셋에서 메모리가 비효율적이다.

# 2. 경영 기초
---

### 문제 정의와 문제 구조화


- 문제 정의
    현재 상태와 바라는 상태의 격차를 나타내며 이를 정교하게 정의하기 위해 SMART 원칙을 사용한다.
- SMART 원칙
| 요소                 | 의미              | 
| ------------------ | --------------- | 
| Specific (구체성)     | 목표가 명확한가        | 
| Measurable (측정가능성) | 성과를 수치로 측정 가능한가 | 
| Achievable (달성가능성) | 현실적으로 가능한가      | 
| Relevant (관련성)     | 목표가 중요한가        | 
| Time-bound (기한)    | 기한이 있는가         | 

- 문제 구조화


위의 표처럼 하나의 문제를 하위의 여러문제로 정의하는 것이다. 이때 MECE를 규칙으로 문제를 쪼개는데 서로 겹치는 부분이 없어야 하며, 전체 포괄적이여야 한다.

### 지표
- 지표란 무엇일까?
    지표는 방향이나 목적,기준을 나타내는 표지로 측정하고 싶은것을 숫자로 표현한것입니다. 
- 북극성 지표와 L1,L@ 지표
    북극성 지표는 가장 중하게 생각하는 지표이며 L1은 가장 전사적인 성과 지표, L2는 지표의 세부 사항을 측정하는 지표입니다.
- AARRR (해적 지표)

사진과 같이 획득(Acquisition), 활성화(Activation), 리텐션(Retention), 수익(Revenue), 추천(Referral)의 5단계 퍼널을 통해 사용자 데이터를 측정하고 병목 구간을 개선하여 성장을 도모합니다
### 지표분석방법 
- 코호트 분석


    정의된시간 내에서 공통된 특성이나 경험을 공유하는 사용자 집단을 분석
- 퍼널 분석


    유저가 접속해서 떠날 때까지 유저의 행동을 추적함
