### 1) 자료의 척도  
- 양적 자료(수치로 표현)  
    - 구간척도: 값들 사이에 순서가 있고, 그 간격이 일정하여 덧셈이나 평균과 같은 연산이 가능한 척도 (ex: 온도, IQ, 시간)  
    - 비율척도: 절대영점과 임의의 측정단위로 구성하고, 동간성이 유지되는 척도 (ex: 거리, 높이, 무게, 시간)  
<br>
- 질적 자료(수치로 표현되더라도 연산의 의미 X)  
    - 명목척도: 레이블이나 이름으로 요소의 속성을 식별하려고 하는 척도 (ex: 펀드 유형, 성별)  
    - 서열척도: 명목형의 분류 속성과 더불어, 항목 간의 순서 또는 순위를 나타내는 특성을 지니는 척도 (ex: 만족도, 직급 체계)  
<br>
- 덧셈이 가능한가? -> 불가능하면 명목척도
- 0값이 의미를 가지는가? -> 가지지 않으면 구간척도, 가지면 비율척도  
- 자료의 종류: 데이터의 대략적인 성격을 파악하는 큰 그림  
- 척도: 그 데이터를 다루기 위한 더욱 구체적인 규칙과 기준  

### 2) 기술통계  
- 개념: 데이터를 있는 그대로 설명하고 이해하는 데 중점을 둔 통계 방법  
<br>
1. 중심화 경향: 데이터의 중심적인 경향을 나타내는 기술 통계량
    - 평균(mean): 데이터 값의 총합을 데이터의 개수로 나눈 값, 가장 대표적인 지표 -> 이상치에 민감할 수 있음. (절사 평균으로 보완)  
    - 중앙값(median): 데이터를 크기 순서대로 일렬로 정렬했을 때, 한가운데에 위치하는 값(데이터가 짝수개라면 가운데 있는 두 값의 평균)  
    - 최빈값(mode): 데이터에서 가장 자주 나타나는 값  
<br>
2. 자료의 산포: 데이터가 퍼져있는 정도를 설명하는 기술 통계량  
    - 범위(range): 데이터의 최대값과 최소값의 차이를 나타내는 지표로, 데이터의 전체 변동 범위를 나타낸다. -> 이상치에 매우 민감함.  
    - 분산(variance): 데이터 값들이 평균으로부터 얼마나 멀리 퍼져있는지를 나타내는 중요한 통계적 지표  
    - 표준편차(standard deviation): 분산에 제곱근을 취한 값  
    - 사분위수(quartile): 데이터의 표본을 네 개의 동일한 크기로 나눈 값  
    - 사분위간 범위(IQR): Q3에서 Q1값을 제외한 범위, 즉 데이터의 중앙 50%가 퍼져 있는 범위 -> 이상치 탐지에 유용함. 
        - 아래쪽 이상치: Q1 - 1.5*IQR  
        - 위쪽 이상치: Q3 + 1.5*IQR  
    - 변동계수(coefficient of variation): 데이터의 절대적인 변동성이 아닌 상대적인 변동성을 나타내는 통계적 지표  
    - 첨도: 확률분포의 꼬리가 두꺼운 정도를 나타내는 척도
    - 왜도: 비대칭도, 실수 값 확률 변수의 확률 분포 비대칭성을 나타내는 지표  
<br>
3. 자료의 정리 및 시각화: 히스토그램, 박스플롯, 파이차트  

### 3) 확률과 분포  
- 확률변수: random한 세계에서 어떤 특정 사건이 발생할 때 나타날 수 있는 모든 경우의 수를 특정 값으로 대응시키는 함수  
<br>
- 확률분포(probability distribution): 확률변수가 특정한 값을 가질 확률을 나타내는 함수  
    - 이산확률분포: 확률 변수가 유한하거나 셀 수 있는 값을 가지는 경우  
        - 베르누이분포 (Bernoulli Distribution)
        - 이항분포 (Binomial Distribution)
        - 기하분포 (Geometric Distribution)
        - 음이항분포 (Negative Binomial Distribution)
        - 초기하분포 (Hypergeometric Distribution)
        - 포아송분포 (Poisson Distribution)
    - 연속확률분포: 확률 변수가 연속적인 값을 가지는 경우(특정 값 선택 불가능)  
        - 정규분포 (Normal Distribution)
        - T분포 (T-Distribution)
        - F분포 (F-Distribution)
        - 감마분포 (Gamma Distribution)
        - 지수분포 (Exponential Distribution)
        - 카이제곱분포 (Chi-squared Distribution)
        - 균등분포 (Uniform Distribution)

#### 1. 정규분포 (Normal Distribution)  
- 형태: 종 모양, 평균을 기준으로 대칭, 표준편차가 클수록 곡선이 평평해짐.  
- 특징: 가장 잘 알려진 연속확률분포, 많은 자연 현상과 사회 현상이 정규분포를 따르는 것으로 알려져 있음.  

- 중심극한정리: 모집단의 분포를 몰라도 표본의 크기가 충분히 크다면(n>=30) 표본평균의 분포는 근사적으로 정규분포를 따른다고 가정  

#### 2. Z분포 (표준정규분포)  
- 평균이 0이고 표준편차가 1이 되도록 맞춘 표준적인 정규분포  
<br>


#### 3. T분포 (T-Distribution)  
- 형태: 종 모양, 정규분포보다 꼬리가 두꺼움, 평균을 기준으로 대칭, 자유도가 커짐에 따라 표준정규분포에 수렴  
- 특징: 표본으로 모평균을 추정할 때 사용되는 분포  


#### 4. 카이제곱분포 (Chi-Square Distribution)  
- 형태: 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포 (자유도가 증가할수록 중심이 오른쪽으로 이동하고 정규분포에 가까워짐)  
- 특징: 독립적인 표준 정규분포를 따르는 변수들의 제곱합 (주로 독립성 검정, 적합도 검정 등에 사용)  
<br>


#### 5. F분포 (F-Distribution)  
- 형태: 0에서 시작하여 오른쪽으로 긴 꼬리를 가진 비대칭 분포 (자유도가 증가할수록 중심이 오른쪽으로 이동하고 정규분포에 가까워짐)  
- 특징: 두 독립적인 카이제곱분포의 비율, 두 집단 간의 분산 차이를 비교하는 데 사용됨.
<br>  


### 4) 추론통계  
- 개념: 수집한 데이터를 기반으로 미래의, 혹은 조사하지 못한 전체 정보를 추론하고 결론을 도출하는 방법을 다루는 분야  
<br>
- 추정: 모집단을 대표할 수 있는 일부를 표본으로 추출하여 분석한 통계량을 이용해서 모집단의 모수를 예측하는 과정  
    - 점추정: 추정하고자 하는 모수를 하나의 수치로 추정 -> 오차에 대한 정보가 고려되지 않기 때문에 어느 정도 오차를 포함하는 신뢰구간 추정 방법이 일반적으로 사용됨.  
    - 구간추정: 추정하고자 하는 모수가 존재하리라 예상하는 구간을 정하여 추정  
- 검정: 표본을 추출하여 얻은 표본통계량으로 모집단 모수에 대한 새로운 예상·가설 등이 옳다고 판단할 수 있는지를 평가하는 과정  

- 가설 검정의 절차
    1. 가설 설정
        - H0 귀무가설: 주로 기각하고자 하는 명제로 설정  
        - H1 대립가설: 내가 주장하고자 하는 명제  
    2. 유의수준 설정  
        - 유의수준: 연구자가 결과를 판단할 때 허용할 수 있는 오류의 최대치 (주로 0.05)  
        - 1종 오류: 귀무가설이 실제로 참이었는데 기각할 확률 -> 더 치명적  
        - 2종 오류: 귀무가설이 실제로 거짓인데 기각하지 못할 확률  
    3. 검정통계량 산출  
        - 검정통계량: 가설 검정의 결과를 판단하는 데 사용되는 값 -> 데이터의 분포나 특징에 따라 T값, F값, Z값 등 어떤 통계량을 사용할지 정해야 함.  
    4. 가설 기각/채택 판단  
        - 유의확률(p-value): 유의 확률(p)가 유의 수준(α)보다 작을 시에 귀무가설을 기각, 클 시에는 귀무 가설을 기각하지 못함.
