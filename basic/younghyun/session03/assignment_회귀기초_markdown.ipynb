## **머신러닝과 모델링**

### 1) 머신러닝(Machine Learning)  
- 개념: 인공지능의 한 분야로, 모델이라는 소프트웨어를 학습하여 새로운 데이터를 예측하거나 결정을 내릴 수 있도록 하는 기술  
- 종류
    - 지도 학습: 정답이 있는 데이터 사용, 기계의 예측값이 데이터의 정답과 같아지도록 학습시키는 것 (ex. 회귀, 분류)  
    - 비지도 학습: 정답이 없는 데이터 사용, 데이터의 패턴 혹은 유사성을 학습시키는 것 (ex. 군집화)  
    - 강화 학습: 시행착오를 반복하여 보상을 최대화 하는 행동 학습 (ex. 알파고, 자율 주행)  


### 2) 모델링(Modeling)  
- 모델: 학습의 결과물, 데이터를 학습하며 찾아낸 패턴이 저장된 상태이자 새로운 데이터를 넣었을 때 답을 내놓는 예측기  
- 모델링: 예측기를 만들어가는 전체적인 과정  
    1. 설계: 데이터의 특성에 맞추어 어떤 알고리즘으로 학습할지 결정하는 단계
    2. 훈련: 데이터를 기계에게 계속 보여주며 정답을 맞히도록 수정하는 단계
    3. 검증: 훈련시킨 모델이 새로운 데이터에서도 정답을 잘 맞히는지 테스트하는 단계  

### 3) 지도 학습
- 개념: 정답 레이블을 함께 주고 학습을 시키는 방법  
- 과정
    1. labeled data: 독립변수와 종속변수가 모두 존재하는 데이터  
    2. labeled data를 training set/test set으로 나누어주기
    3. training set을 통해 모델 학습  

종속 변수가 연속형이면 회귀, 범주형이면 분류 사용

## **회귀(Regression)**

### 1) 선형회귀  
- 개념: 데이터를 가장 잘 설명하는 회귀선을 찾는 과정, 회귀선은 현재 데이터를 요약하는 동시에 관측되지 않은 미래 값을 예측하는 도구가 됨.

#### 가중치와 편향의 이해  
기본식 : $y=wx+b$  
(1) 가중치 $w$
- 독립변수 x가 종속변수 y에 주는 영향력의 크기 (그래프에서는 기울기로 나타남.)  
(2) 편향 $b$  
- 독립변수 x가 0일 때 나오는 기본값 (그래프에서는 y절편)  
- 모델이 데이터에 더 유연하게 최적화될 수 있도록 위아래로 움직여주는 역할  

#### 단순선형회귀  
- 하나의 독립변수만 존재하는 선형회귀  
- 데이터를 직선 형태로 표현하는 경우 -> 잔차를 줄여나가며 직선을 결정짓는 적절한 회귀계수와 편향을 찾아감.  

#### 잔차(Residual): 실제값과 예측값 간의 차이

### 2) 최적의 $w$와 $b$를 찾는 방법

- 최소제곱법(Least Squares Method): 잔차의 제곱의 합이 최소가 되는 지점으로 최적의 회귀선을 구하는 방식 -> 회귀선과 실제 데이터와의 차이를 최소화하기 위한 시도  

- 경사하강법(Gradient Descent)

    - 목적함수: 우리가 달성하고 싶은 목표  
    - 손실함수: 개별 데이터에 대한 오차를 측정하는 함수
    - 비용함수: 전체 데이터셋에 대한 손실함수를 합산하거나 평균을 내어 모델의 전체적인 성적을 평가하는 함수  

(1) 손실함수와 경사하강법  
<br>
$E(W, b) = \frac{1}{n} \sum_{i=1}^{n} [t_i - y_i]^2 = \frac{1}{n} \sum_{i=1}^{n} [t_i - (Wx_i + b)]^2$
<br>  
우리의 목표는 $E$를 최소화하는 $W$, $b$를 찾는 것
원리: 최솟값 왼쪽의 값은 점점 오른쪽으로 이동하고, 최솟값 오른쪽의 값은 왼쪽으로 이동하며 최종적으로 최솟값에 도달
<br>  
(2) 학습률(learning rate)  
너무 작으면 모델이 수렴할 때까지 오랜 시간이 걸림, 너무 크면 모델이 수렴하지 못하고 발산  
자동으로 학습률 찾는 방법? Learning Rate Scheduler, Grid Search, Random Search

(3) Local Minima 문제  
실제 이동량을 보면 작은 기울기에서 조금씩, 큰 기울기에서 많이 이동  
<br>
(4) 해결법 - 모멘텀  
이전의 기울기와 방향을 고려하도록 관성 부여  
장점: 작은 기울기를 쉽게 넘어 지역 최소값을 탈출할 수 있게 함.  

### 3) 다중선형회귀  
$y = W_1 x_1 + W_2 x_2 + ... + W_n x_n + b$
- 개념: 독립변수가 두 개 이상인 경우  
- 3차원 공간에 표현되고 회귀선은 평면이 됨. 

#### 다중공선성(Multicollinearity)  
- 개념: 독립변수들 간의 강한 상관관계로 인해 회귀 모델이 불안정해지는 문제  
- 확인 방법
    - 상관계수  
    - VIF지수(분산 팽창 인수)  
- 대처 방법
    - 변수 제거  
    - 변수 변환  
    - 규제선형모델 활용  
    - PCA(주성분 분석)

### 4) 규제선형모델  
- 개념: 모델이 학습 데이터에 과적합되지 않도록 규제를 가하고자 등장한 모델 -> 선형회귀 모델에서는 특성에 곱해지는 계수의 크기를 조정하는 것  
- 종류  
    - L2규제: 회귀계수의 제곱에 대해 페널티를 부여 -> 릿지 회귀  
    - L1규제: 회귀계수의 절댓값에 대해 페널티를 부여 -> 라쏘 회귀  
    - L2+L1규제: 엘라스틱넷 회귀  

(1) 릿지회귀
- 제약식: L2 norm  
- 변수 선택: 불가능  
- solution: closed form  
- 장점: 변수간 상관관계가 높아도 좋은 성능  
- 특징: 크기가 큰 변수를 우선적으로 줄임  

(2) 라쏘회귀
- 제약식: L1 norm  
- 변수 선택: 가능  
- solution: 명시해 없음  
- 장점: 변수간 상관관계가 높으면 성능 저하  
- 특징: 비중요 변수를 우선적으로 줄임  

(3) 엘라스틱넷
- 제약식: L1 + L2 norm  
- 변수 선택: 가능  
- solution: 명시해 없음  
- 장점: 변수간 상관관계를 반영한 정규화  
- 특징: 상관관계가 큰 변수를 동시에 선택/배제

## **모델 평가 방법**

### 1) 성능 평가 지표  
- 평균 제곱 오차(Mean Squared Error, MSE): 예측값과 실제값 사이의 오차 제곱의 평균
- 평균 절대 오차 (Mean Absolute Error, MAE): 모델의 예측값과 실제값 사이의 절대값 오차의 평균  
- MSE와 MAE의 차이
    - MSE는 큰 오차에 대해 민감하게 반응하기 때문에 이상치에 민감함.  
    - MAE는 모든 오차를 동일하게 취급해 이상치가 존재하는 경우에도 전반적인 예측 성능 유지 가능  

### 2) 변수 유의성 평가  
- T-검정
    - t-값 계산  
    $t = \frac{\hat{\beta} - 0}{SE(\hat{\beta})}$
    - p-value 확인 및 판단
    - p-value가 작을수록 해당 변수는 종속변수에 더 큰 영향을 줄 확률이 높아짐. 
    - t-값이 크면 귀무가설을 기각할 가능성이 높아짐.
