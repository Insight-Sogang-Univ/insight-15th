## **선형회귀**

### 1) 다중선형회귀의 수식적 이해  
$y_i = \beta_0+\beta_1X_{1i}+\beta_2X_{2i}+...+\beta_kX_{ki}+\varepsilon_i$

-> 상수항, 오차항 제외: $y=w_0x_0+w_1x_1+\cdot\cdot\cdot+w_mx_m=\sum_{i=0}^mw_ix_i=w^Tx$

### 2) 다중선형회귀의 기본 가정 6가지  

**1. 선형성 (Linearity)**  
<br>
$E(y|X) = \beta_1+\beta_2X_2+...+\beta_nX_n$
- 종속변수와 설명변수 간의 관계가 선형적이어야 한다. 

**2. 독립성 (Independence)**  
<br>
$X_i, X_j \ is \ linearly\ independent \ for\ all \ i,j$  
- 각각의 설명변수가 서로 선형독립적이어야 한다.  
- 선형독립적이지 않은 경우 -> 다중공선성 有 (독립변수들이 서로 강한 상관관계를 가져서, 어떤 변수가 종속변수에 가장 큰 영향을 주는지를 알 수 없는 상태)

**3. 오차항의 평균은 0이다(0에 가까워야 한다)**
<br>
$E(\varepsilon_i|X_i) = 0$  
- 오차항: 실제값과 예측값 사이의 차이

**4. 등분산성 (Homoscedasticity)**  
<br>
$Var(\varepsilon_i|X_i) = \sigma^2$  
- 오차항의 분산이 일정해야 한다.  
- 오차항의 분산이 일정하지 않은 경우 -> 이분산성이 있다. (횡단면 자료에서 많이 나타남.)  
- 등분산성을 알아보는 방법
    - 잔차의 도표화  
    - 검정 (Brown-Forsythe, Breusch-Pagan)

**5. 오차항은 자기상관되어 있지 않다. (Non-autocorrelation)**  
<br>
$Cov(\varepsilon_i, \varepsilon_j|X_i) =0$  
- 오차항의 공분산은 항상 0이어야 된다.  
- 오차항의 공분산이 0이 아닌 경우 -> 자기상관(autocorrelation)이 있다. (한 변수의 현재 값이 과거 값과 상관관계를 가지는 현상, 시계열 자료에서 많이 나타남.)  
- 자기상관을 확인하는 방법  
    - 더빈-왓슨 검정  
    - ACF, PACF 함수 찍어보기

**6. 정규성 (Normality)**  
<br> 
$\varepsilon_i \sim N(0, \sigma^2)$  
- 오차항이 정규분포를 따른다. (위배되어도 다중선형회귀분석의 결과에 큰 영향을 주지 않음.)  
- 정규성을 알아보는 법  
    - 샤피로-월크 검정(Shapiro-Wilk Test)    
    - 자퀴-베라 검정(Jarque-Bera Test, JB)    
    - Q-Q plot 찍어보기

- 가정 1,2를 제외하고는 회귀 분석 후에 검증 가능하다. (오차항에 대한 가정인지가 기준)  
- 모델 학습 전에는 오차항을 계산할 수 있는 근거가 없기 때문이다.  

### 3) 회귀분석 평가 방법  

**1. 시각화**  
<br>



- 회귀선이 데이터 전반을 잘 요약하지만 둘 중 어느 회귀선이 해당 데이터를 더 잘 요약하는지 객관적으로 비교하기는 어렵다.  
- 두 집단의 차이의 정도를 객관적으로 비교하기 위해서 통계지표를 활용하는 것이 좋다. 

**2. 통계지표**  
<br>
(1) 모델 유의성 검정  
- F-statistic: 회귀 모델이 유의미한지 검정하는 지표 (모델 전체)  
    - 귀무가설 $H_0$: 모든 회귀 계수가 0이다(독립 변수들이 종속 변수들과 관계가 없다).  
    - 대립가설 $H_1$: 적어도 하나의 회귀 계수는 0이 아니다.
                        (독립 변수 중 하나 이상이 종속 변수에 영향을 준다)
    - F-statistic의 값이 클수록 모델이 통계적으로 유의미하다는 것을 의미한다.  
    - Prob가 0.05보다 작다면, 귀무가설을 기각할 수 있다.
<br>
- T-statistic: 각 독립변수가 종속변수 y에 유의미한 영향을 미치는지 판단하는 지표 (개별 변수의 유의성)    
    - 귀무가설 $H_0$: 해당 변수의 회귀 계수가 0이다.
                        (이 변수는 종속 변수에 영향을 미치지 않는다)
    - 대립가설 $H_1$: 해당 변수의 회귀 계수는 0이 아니다.
                        (이 변수는 종속 변수에 유의미한 영향을 준다)
    - P > |t| (t통계량의 유의확률)가 0.05보다 작으면 해당 독립변수는 종속변수에 유의미한 영향을 미친다.

(2) 모델의 성능 평가
- 결정계수 (R-squared, $R^2$): 회귀 모델이 종속 변수의 전체 변동성을 얼마나 잘 설명하는지 나타내는 지표  
    - $0 ≤ R^2 ≤ 1$
    - 1에 가까울수록, SSR과 SST의 값이 비슷해지고, SSE는 0에 가까워지므로 모델이 데이터를 잘 설명한다고 할 수 있다.  
- 조정된 결정계수 (Adjusted R-squared, Adjusted $R^2$)  
- AIC, BIC(SC): 두 지표 모두 정보 기준이라고도 불리며, 값이 낮을수록 좋다고 평가한다.  
    - AIC = Akaike information criterion (BIC에 비해 복잡성에 대한 패널티가 비교적 작다.)  
    - BIC(SC): SC = Schwarz Criterion, BIC = Bayesian Information Criterion (AIC보다 더 엄격한 기준으로, 데이터의 양에 따라 더 강한 패널티를 부과한다.)

## **비선형회귀**  

### 1) 다항식 회귀 모델 (Polynomial Regression Model) 

- 기존의 선형 회귀: $Y=a+bX$
- 다항식 회귀 모델: $Y=a+bX+cX^2+dX^3+\cdot\cdot\cdot$의 형태로, 독립 변수에 거듭제곱 항을 추가하여 데이터의 곡선적(비선형적) 특성을 모델링한다.
- 사용 방법
    1. 기존의 선형 모델 $Y = a + bX$에 대해 $cX^2, dX^3$ 등을 포함하여 모델을 확장
    2. 확장한 상태에 선형 회귀를 적용

- 장점: 선형 모델에 비해 회귀선을 데이터에 fit하게 그릴 수 있다.  
- 단점: 너무 많은 feature(독립변수)를 이용하면 결과가 overfitting 될 수 있다.  
- 관계가 직선적이지 않고 곡선을 그리는 데이터 패턴에서 유용하다. 성장률, 감소율 등이 시간에 따라 변화하는 속도가 다른 경우에 좋다.  

### 2) 지수 회귀 모델 (Exponential model)

- 종속변수가 지수적으로 변화하는 관계를 모델링할 때 사용할 수 있다.  
- 사용 방법  
    - 종속변수 Y에 로그함수 적용
    - 로그 변환된 Y와 기존의 독립변수 X에 대해 선형 회귀 적용 → 모델이 ln Y(로그 변환된 Y)와 X의 선형 관계를 학습  
    - Y값을 예측 시, 모델이 예측한 Y값에 지수함수$(e^x)$를 적용하여 로그 변환하기 전 Y로 변형
- 값이 시간에 따라 지수적으로 증가하거나 감소하는 경우(기술 발전, 투자 수익률 증가 등)에 적합하다. 

### 3) 로그 회귀모델 (log regression model)

- 종속변수와 독립변수 간의 관계가 로그함수를 통해 더 잘 표현될 때 사용할 수 있다.  
- 사용 방법  
    - 독립변수 X에 로그함수 적용
    - 선형회귀를 적용  
-> 이렇게 예측된 값은 Y이므로 예측값에 별도의 처리를 하지 않고 그대로 사용할 수 있다.  
- 데이터가 초기에 빠르게 증가하고 점차 증가율이 줄어드는 패턴(인구 성장, 감염병의 확산)을 보일 때 적합하다. 

### 4) 스플라인 회귀

- 데이터를 구간별로 나누고 각 구간에서 다른 선형 또는 비선형 함수를 적용하여 예측하는 방식이다.  
- 사용 방법  
    - 전체 데이터 범위를 여러 구간으로 나누고, 각 구간에 대해 별도의 회귀모델 적용  
    - 구간 경계에서의 연속성을 유지하는 것이 중요하다.  
- 데이터 패턴이 여러 구간에서 서로 다르게 나타날 때(계절에 따라 판매량이 변화하는 소매 데이터 분석 등) 적합하다.
