## **1. 분류 (Classification)**

### 1) 지도학습  
- 회귀 (Regression): 주어진 데이터 x를 기반으로 정답 y를 잘 맞추는 함수 (연속형 변수 예측)  
- 분류 (Classification): 기존 데이터가 어떤 레이블에 속하는지 패턴을 알고리즘으로 인지한 뒤에 새롭게 관측된 데이터에 대한 레이블을 판별하는 것 (범주형 변수 예측)  
<br>
### 2) 이진 분류와 다중 분류  
- 이진 분류(Binary Classification): 예측하고자 하는 변수기 어떤 기준에 대하여 True 또는 False의 값만을 가질 때  
- 다중 분류(Multiclass Classification): 예측하고자 하는 변수가 가질 수 있는 값이 3개 이상일 때

## **2. 분류 모델**  

### 1) 로지스틱 회귀(Logistic Regression)  
- 개념: 이진 분류 문제를 푸는 대표적인 알고리즘으로, 샘플이 특정 클래스에 속할 확률을 추정하여 기준치에 따라 분류하는 것을 목표로 함.  
- 시그모이드 함수 (Sigmoid function): 출력이 0~1 값을 가지면서 S자 형태로 그려지는 함수


- 시그모이드 함수 특징  
    - 입력값이 커지면 1에 수렴하고, 작아지면 0에 수렴  
    - 입력값이 $(-\infin, \infin)$일 때, 출력값은 0~1의 값을 가짐.  

-> 승산의 개념을 적용하여 로지스틱 회귀모델의 결과를 선형적으로 해석할 수 있음.

### 2) 의사결정나무 (Decision Tree)  
 

- 개념: 조건에 따라 데이터를 분류하며, 최종적으로 데이터가 순수한 label의 집합으로 구성될 때까지 분류를 반복하는 분석 방법  
- 구성 요소  
    - Root Node: 가장 상위에 위치한 노드  
    - Parent Node: 주어진 노드의 상위에 있는 노드  
    - Child Node: 하나의 노드로부터 분리된 2개 이상의 노드들  
    - Binary Tree: Tree 중에서 children이 최대 2개인 tree  
    - Leaf Nodes: 자식마디가 없는 노드, 모델에서 label에 해당  
    - Edge: Parent Node와 Child Node를 연결하는 연결선  
    - Height: 특정 노드에서 가장 멀리 있는 leaf node까지의 경로에 있는 edge 개수  
    - Depth: Root Node에서 특정 노드에 도달하기 위한 edge의 수

#### CART (이진분할) 알고리즘
- CART: 데이터셋을 임계값을 기준으로 두 child로 나누는 알고리즘(가장 널리 사용되는 의사결정나무 알고리즘)  
- 임계값은 불순도(지니계수)가 낮아지는 방향으로 나눔.  
- 불순도: 분류하려는 데이터 집합에서 서로 다른 클래스가 섞여있는 정도  
- 주요 단계: 임계값 설정 -> 불순도 감소 알고리즘  
- 한계: Greedy Algorithm - 당장의 지니계수를 낮추는 판단만 하여 가장 효율적인 대안을 제시할 수 없음. 

#### 실제 학습 시 고려해야 할 것  
- Hyperparameter 설정  
- 시각화  
- Prunning(가지치기)

### 3) SVM (Support Vector Machine)  

- SVM: 클래스를 분류할 수 있는 다양한 경계선 중 최적의 라인을 찾아내는 알고리즘  
-> 데이터를 분리하는 초평면 중에서 데이터들과 가장 거리가 먼 초평면을 선택하여 분리하는 지도 학습 기반의 이진 선형 분류 모델  
- 고차원 공간 및 명확하게 분류할 수 있는 데이터 집단에서 뛰어난 성능을 보임. (ex. 패턴 인식, 이미지 분석, 바이오인포매틱스)

- 구성  
    - Support vector: 데이터 중에서 결정 경계와 가장 가까이에 있는 데이터들의 집합  
    - Decision Boundary: 데이터 분류의 기준이 되는 경계  
    - Margin: 결정 경계에서 서포트 벡터까지의 거리  
    - Hyperplane: n차원 공간의 (n-1)차원 평면 - 초평면  
    - Slack Variables: 완벽한 분리가 불가능할 때 분류를 위해 허용된 오차에 대한 변수  

- 장점  
    - 서포트 벡터만을 이용해서 결정 경계를 생성하므로 데이터가 적을 때 효과적임.  
    - 새로운 데이터가 입력되면 전체 데이터 포인트와의 거리를 계산하지 않고 서포트 벡터와의 거리만 계산하면 되기 때문에 연산량이 최소화됨.  
    - 뛰어난 정확성 및 비선형 모델 분류 가능  
    - 과대 적합 가능성이 낮고, 노이즈의 영향이 적음.

- 단점  
    - 데이터 전처리 과정이 중요  
    - 데이터가 많아질수록 최적화를 위한 평가 과정이 많아져서 속도가 느림.

### 4) KNN (K-Nearest Neighbor)  
- KNN: 데이터로부터 거리가 가까운 k개의 다른 데이터 레이블을 참조하여 분류하는 알고리즘  
- 계산 순서  
    1. 데이터 준비  
    2. K값 설정 (보통 홀수개)  
    3. 거리 계산 (유클리드/맨해튼 등)  
    4. 가장 가까운 K개의 이웃 선택  
    5. 분류하기  

- 최적의 K 선택 방법  
    - 일반적으로 학습용 데이터 개수의 제곱근으로 설정  
    - 너무 크게 설정하면 과소 적합  
    - 너무 작게 설정하면 이상값, 잡음 데이터와 이웃이 될 가능성이 있음.

- 장점  
    - 범주를 나눈 기준을 알지 못해도 데이터 분류 가능  
    - 입력 데이터만 주어지면 바로 예측값을 구할 수 있어서 학습 과정이 필요없음.  
    - 이해와 추가된 데이터의 처리가 용이, 구현이 간단함.  
    - 훈련 데이터를 요약하거나 일반화된 모델로 만들지 않고, 원본 데이터 전체를 그대로 저장함.

- 단점  
    - 테스트 데이터 개수에 따라 시간이 오래 걸림.  
    - K 값 결정이 어려움.  
    - 수치형 데이터가 아니라면 유사도를 정의하기 어려움.  
    - 데이터 내 이상치가 존재하면 분류 성능에 큰 영향을 받음.

#### 앙상블 (Ensemble)  
- 여러 개의 개별 분류 모델들을 결합해 하나의 분류 모델보다 더 좋은 성능을 내는 머신러닝 기법  
    - 보팅(Voting): 다른 알고리즘의 모델을 병렬로 사용  
    - 배깅(Bagging): 동일 알고리즘의 모델을 병렬로 사용  
    - 부스팅(Boosting): 동일 알고리즘의 모델을 직렬로 사용

## **3. 분류 평가 지표**

### 1) 혼동 행렬  
- 개념: 분류 모델의 예측 결과를 정확한 예측과 잘못된 예측으로 구분하여 나타낸 표

#### Accuracy (정확도) 
$(TP+TN) / (TP+TN+FP+FN)$
- 모델이 입력된 데이터에 대해 얼마나 정확하게 예측하는지를 나타냄.  
- 단점: 정답 레이블의 비율이 불균형하면 모델의 정확도를 신뢰할 수 없음.

#### Precision (정밀도)  
$TP/TP+FP$
- 0과 1사이의 값  
- 예측을 참이라고 한 경우 중 실제 참의 비율 (거짓을 참으로 판단한 정도를 알 수 있음.)  
- 정밀도가 높을수록, 거짓을 참으로 예측한 정도가 낮음.

#### 재현도 (Recall)  
$TP/TP+FN$
- 0과 1사이의 값  
- 실제 참이라고 한 경우 중 예측 참의 비율 (참을 거직으로 판단한 정도를 알 수 있음.)  
- 재현도가 높을수록, 참을 거짓으로 예측한 정도가 낮음.
- 정밀도와 재현도는 trade-off 관계

### 2) F1-Score  
- 정밀도(precision)와 재현율(recall)의 조화 평균

$F1\ Score = \frac{2}{\frac{1}{Precision} + \frac{1}{Recall}} = \frac{2 \times Precision \times Recall}{Precision + Recall}$

### 3) ROC / AUC Curve  
- ROC Curve: 얼마나 분류가 잘 되었는가를 보여주는 그래프
<br>  


- AUC Curve: ROC와 x축 사이의 면적
<br>  
  
-> 0.5~1의 값을 가지며, 1에 가까울수록 분류 성능이 좋음.

## **4. 하이퍼파라미터 최적화**

### 1) 하이퍼파라미터 최적화  

#### 하이퍼파라미터  
- 하이퍼파라미터: 모델 학습 과정에 반영되는 값으로, 학습 시작 전에 사용자가 직접 설정하는 변수  
- 하이퍼파라미터 최적화: 적절한 하이퍼파라미터를 찾아 모델 성능을 향상시키는 것

- 과정  
    1. Hyperparameter 탐색 범위 설정 (최적 값을 찾고 싶은 하이퍼파라미터의 범위를 설정)
    2. 평가 지표 계산 함수(성능 평가 함수) 정의(탐색하려는 Hyperparameter를 인수로 받아 평가지표 값을 계산해주는 함수를 정의)
    3. 1단계에서 샘플링한 Hyperparameter 값을 사용하여 검증 데이터로 정확도 평가
    4. 위 단계를 특정 횟수 반복하며, 정확도 결과를 보고 하이퍼파라미터의 범위 좁힘.

### 2) 하이퍼파라미터 최적화 방법  
- Grid Search: 정해진 범위에서 Hyperparameter를 모두 순회하며 가장 좋은 성능을 내는 값을 찾는 기법  
    - 장점: 범위가 넓고 step이 작을수록 꼼꼼하게 전범위를 탐색해 최적해를 정확히 찾을 수 있음. 
    - 단점: 시간이 오래 걸림.  
    - 적용: 넓은 범위, 큰 step을 활용해 범위를 좁힘.

- Random Search: 정해진 범위에서 Hyperparameter를 무작위로 탐색해 가장 좋은 성능을 내는 값을 찾는 기법  
    - 장점: 속도가 Grid Search보다 빠름.  
    - 단점: 무작위라는 한계 때문에 정확도가 떨어짐. (사용빈도가 비교적 적음.)

- Bayesian Optimization: 사전 정보를 바탕으로 최적 Hyperparameter 값을 확률적으로 추정하며 탐색하는 기법  
    - 특징: "Gausain Process"라는 통계학을 기반으로 만들어진 모델로, 여러 개의 하이퍼 파라미터들에 대해서 "Aquisition Fucntion"을 적용했을 때 "가장 큰 값"이 나올 확률이 높은 지점을 찾아냄.  
    - 방법
        - 초기 값 몇 개를 무작위로 실험
        - 그 결과를 바탕으로 어떤 파라미터의 조합이 좋을 지를 확률 모델로 예측
        - 다음 실험할 조합을 가장 좋을 것 같은 위치로 선택
        - 결과를 반영하고 다시 반복
