# 2. 테이블을 합치기

위 head를 보면, info 테이블은 특수문자 -> 숫자 -> abc 순으로 정렬이 되어 있고, sale 테이블은 글로벌 판매량을 기준으로 정렬되어 있습니다.

```python
# 실습 1: 우선 GAME_ID를 기준으로 병합해봅시다.
# 참고로, DB에서는 병합의 기준이 되는 키를 Primary Key라고 부른다네요

df = pd.merge(sale, info, on='GAME_ID', how='inner')

print(df.shape)
df.head(10)
```

```python
# Name_x, Name_y 이런 중복열이 생겨요...
# -> 기준(on)에 리스트 형식으로, 공통된 열을 다 넣어주세요.
df = pd.merge(sale, info, on=['GAME_ID', 'Name', 'Platform'], how='inner')

print(df.shape)
df.head(10)
```

```python
# (여기는 그냥 실행만 해주세요)
sale_drop = sale.drop(columns=['GAME_ID'])
info_drop = info.drop(columns=['GAME_ID'])
```

```python
# 이름만을 Key로 하여 병합한다면 이런 일이 생깁니다.

df_merged_by_name = pd.merge(sale_drop, info_drop, on=['Name'])

df_merged_by_name.shape
# 행 수가 이상하게 늘어났죠?
```

# 4. 값 매핑과 대체

```python
# 열 이름 바꾸기

# 'Year'를 'Release_Year'로, 'Name'를 'Title'로 바꾸기
df = df.rename(columns={'Year':'Release_Year', 'Name':'Title'})

print(df.columns)
```

```python
# astype(바꿀 자료형) - 자료형 변경
# 원본 데이터셋은 Year가 float입니다. 이를 정수형으로 바꿔 봅시다

# 그 전에 NaN을 일단 0으로 바꿔줍시다. (NaN이 있으면 변환이 안되므로)
df['Release_Year'] = df['Release_Year'].fillna(0)
df['Release_Year'] = df['Release_Year'].astype(int)

print(df.dtypes)
```

```python
# replace()

# 'X360'을 알아보기 쉽게 'Xbox_360'으로 바꾸기
df['Platform'] = df['Platform'].replace('X360', 'Xbox_360')

# 여러 개를 동시에 바꾸고 싶다면 리스트로 묶어주면 됩니다.
# df['Platform'] = df['Platform'].replace(['PS3', 'PS4'], ['PlayStation3', 'PlayStation4'])
```

```python
# 판다스로 원핫인코딩 - pd.get_dummies()

df_genre = df[['Title','Genre']]

df_encoded = pd.get_dummies(df_genre, columns=['Genre'], drop_first=True)

df_encoded.head(3)
```

```python
# apply(함수)

# 글로벌 판매량이 백만장 이상이면 밀리언셀러로 표기해 봅시다
df['Hit'] = df['Global_Sales'].apply(lambda x: '밀리언셀러' if x >= 1.0 else '일반')

print(df['Hit'].value_counts(normalize=True))
```

# 5. 그룹화, 집계함수

```python
# groupby()

# 각 플랫폼별로 글로벌 판매량의 합계는 얼마일까?

df.groupby('Platform')['Global_Sales'].sum() # -> 시리즈

df.groupby('Platform')[['Global_Sales']].sum() # -> 데이터프레임 (fancy 방식)
```

```python
# 판매량 합계가 많은 순서대로 보기

# Series이니까.. 정렬을 할 수 있겠죠?
print(type(df.groupby('Platform')['Global_Sales'].sum()))

df.groupby('Platform')['Global_Sales'].sum().sort_values(ascending=False).head()
```

```python
# agg()

# 리스트 [ ] 안에 원하는 집계함수 이름들을 문자열로 넣어줍니다.
# 저는 판매량 평균이 높은 순으로 정렬도 해봤어요

df.groupby('Platform')['Global_Sales'].agg(['sum', 'mean', 'count']).sort_values(by='mean', ascending=False).head(10)
```

```python
# 참고 - 사용자 정의 함수를 agg에 쓰기

def get_range(x):
    return x.max() - x.min()

df.groupby('Publisher')['Global_Sales'].agg(
    ['max', 'min', get_range, lambda x: x.max() / x.mean()]
).sort_values(by='max', ascending=False).head(10)
```

```python
# 실습 4는 여기에!

df.groupby('Genre')['Global_Sales'].agg(
    최대판매량=('max'),
    최소판매량=('min'),
    판매량격차=(get_range),
    최대평균비율=(lambda x: x.max() / x.mean())
).sort_values(by='최대판매량', ascending=False).head(10)
```
