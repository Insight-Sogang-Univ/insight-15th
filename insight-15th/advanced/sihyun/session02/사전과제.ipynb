# CNN
## MLP: 위치/크기/외전 등의 변화에 취약
1. convolution으로 특징 추출: 모델이 특징을 학습함
2. Pooling으로 차원 축소: 위치/크기 변화에 덜 민감하게 만든다
3. 어떤 영역에 특정한 특징을 존재하는가를 보기 때문에 위치에 무관하다

## MLP: 파라미터 多多多
1. 부분 연결 구조 & Convolution 연산으로 이미지의 특정 피쳐에서만 집중
2. 효율적인 파라미터 사용: 위 예시에서 파라미터 수가 차이 많이 남

## MLP: 공간 정보 손실 (2차원 이상의 이미지 데이터를 1차원으로 옮기며 공간 정보 손실)
1. 다차원 이미지를 그대로 입력받고, 공간 정보를 보존하면서 학습
2. 특징을 어느정도 추출한 후 flatten 하기에 그 손실의 저옫가 훨씬 낮다

## 합성곱과 풀링
: 이미지의 특징을 추출하는 방법

-> 이미지는 어떻게 분석해야 하나?
- 공간적인 관계와 지역적 패턴이 중요함 따라서 이를 위해 CNN을 활용

CNN은 위치 별 특징을 추출하는 부분과 클래스를 분류하는 부분으로 나눌 수 있음
1. 특징 추출: Convolution layer와 Pooling later를 여러겹 쌓는 형태로 구성
- Convolution: 입력데이터에 필터를 적용 후 활성화 함수를 반영하는 필수 요소
- Pooling: 합성곱 다음에 위치하며, 이미지의 공간 크기를 줄여 중요한 특징을 강조하는 역할. 모델의 계산 효율성을 높이고 과적합을 방지하는데 도움 줌

2. 이미지 분류: CNN 모델의 마지막에 위치하고 fully connected layer로 구성

즉, 이미지를 인식하기 위한 모델은 이미지의 특징을 학습하는 부분과 어떤 이미지인지 분류하는 부분으로 구성

학습: CNN 모델에서 새로 추가된 부분, 합성곱과 풀링 사용
분류: 기존에 배운 회귀, 분류와 유사한 방식


## 특징 추출
### Convolution layer
1. 필터 또는커널이 이미지 위를 슬라이딩
2. 각 위치에서 요소별 곱셈 후 합계 계산
3. 결과를 Feature Map에 저장

### Padding
1. filter를 거치며 정보를 잃어버리는 것을 막기 위해 이미지 바깥을 0으로 둘러싸는 기법

### Stride
1. 이미지를 몇 칸씩 점프하면서 훑을 것인지 결정하는 값
=1 -> 한칸 씩 이동하며 세밀한 특징 추출
=2 -> 두칸 씩 이동하며 빠른 처리, 크기 축소
즉, stride가 커질수록 feature map의 크기가 작아짐

### Pooling
1. 피처맵에서 특징을 추출하기 위해 이미지를 격자 형태로 분할한 뒤 해당 영역 안에서 가장 큰 값 또는 평균 값을 취하는 것
-> 왜씀? 창문이 있다보다 정확히 어디에 있다가 중요할 수 있기에

## RNN(순차적 데이터)
- 순차 데이터란 순서에 의미가 있으며 순서가 달라질 경우 의미가 손상되는 데이터
1. 음성 및 오디오 데이터
2. 자연어
3. 생물학적 서열 데이터
4. 비디오 데이터

- 기존 정형 데이터와의 차이점
1. 샘플의 독립성: 각 행은 독립적인 정보를 나타내므로, 해으이 순서가 바뀌어도 데이터 셋의 의미가 변하지 않음
2. 특성 순서의 독립성: 나이,키 순서로 분석하든 순서를 바꿔서 하든 각 특성이 담고 있는 정보는 동일

- 순차 데이터의 특징
1. 요소의 순서 의존성: 데이터를 구성하는 요소드르이 순서 자체가 핵심 정보
2. 자기 상관성: 이전 시점의 데이터가 다음 시점의 데이터에 영향을 미침

- 정의: 순환하는 구조를 가진, 순차 데이터 처리에 특화된 인공 신경망
(나도 RNN 마스터!!)
1. 시퀀스 데이터를 입력 받아 순서 정보를 유지하며 처리
2. 순환하는 은닉층이 매 시점의 은닉 상태를 업데이트
3. 이전 시점의 값을 현재 시점으로 넘겨준다

- 핵심 원리: 가중치 공유
-> 동일한 가중치를 시퀀스의 다양한 시점에서 반복적으로 사용
RNN이 시퀀스 데이터의 길이와 시점 위치에 상관없이 효과적으로 작동하게 하는 핵심 원리
- 이점: 학습 파라미터의 수 감소, 일반화 능력 향상

- 한계
1. 기울기 소실: 시쿼스 뒤쪽의 오차가 앞쪽까지 제대로 전달되지 않아 먼 과거의 정보를 학습하지 못하는 현상
2. 기울기 폭주: 시퀀스 뒤쪽의 오차가 역전파 과정에서 비정상적으로 커지는 문제
3. 느린 훈련 시간: 계산 과정이 순차적으로 이루어져야만 하는 구조적 한계
4. 장기 의존성 문제 -> 시퀀스 앞부분의 중요한 정보를 잊어버려 맥락 파악 능력이 급격히 저하되는 현상

## LSTM
- 셀 상태(장기 기억): 이 기억은 절대 변하지 않는 핵심 기억
- 엘리트: 현재 상황에서 바로 사용하는 기억으로 셀 상태에서 필요한 정보만 꺼내서 , 지금 시점에서의 판단이나 출력으로 사용
-> 진짜 장기 기억과 단기 기억 및 출력을 따로 분리해서 관리 -> 매우 정교하지만 그만큼 복잡하다는 특징

(나도 LSTM 마스터!!)
1. 핵심 아이디어: 기억할 내용과 잊어버릴 내용을 선택해 중요한 정보들을 오래 가져감
2. RNN과의 차이는 2개의 순환되는 층을 사용함(RNN은 한개)
3. Gate를 통해 필요한 정보들만 통과
- Forget: 정보를 얼마나 잊어버릴지
- Input: 정보를 얼마나 사용할지
- Output: 다음 층으로 어떤 정보를 전달할지
4. Final memoery cell: 1~3을 결합해 현재 정보를 얼마나 기억할지 계산

장점: Vanishinf gradient problem을 효과적으로 완화
단점: 많이 복잡해 RNN보다 학습 파라미터가 많음

## GRU
- 모든 기억과 행동이 하나의 상태로 통합
기억과 행동이 하나로 통합되어 있어 LSTM보다 훨씬 단순하고 즉각적인 반응 -> 정교함은 떨어짐
(나도 GRU 마스터!!)
1. forget과 input을 update 하나로 합침
2. Reset gate를 활용
3. Gate가 3->2개로 줄음

장점: 성능과 효율성 사이의 균형이 좋다
단점: 긴 시퀀스 처리에서의 한계는 여전히 존재

## Seq2Seq
1. 인코더: 의미를 정리하는 사람
- 핵심 의도: 컨텍스트 벡터
2. 디코더: 그걸 받아서 결과를 만드는 사람
- 컨텍스트 벡터를 이용해 새로운 결과를 만들어냄

- 특징
1. 입력과 출력의 길이가 달라도 됨
2. 인코더와 디코더가 분리되어 있다
3. 문장 벡터라는 다리를 건넌다

한계: 고정된 컨텍스트벡터에 소스 문장의 정보 압축 -> 병목 현상 발생
- 고정된 길이에 입력 시퀀스의 모든 정보를 담지 못하고 손실
- 긴 입력 시퀀스의 경우 정보 손실이 더 심각하고 모델 성능 하락으로 이어짐
- 기울기 소실 문제
- 병렬화 불가능
