## Attention 등장배경
- 기존 Seq2Seq 모델은 입력 문장의 모든 정보를 하나의 고정된 Context Vector에 압축
- Decoder는 이 Context Vector만을 이용해 출력 문장을 생성
- 입력 문장이 길어질수록 하나의 벡터에 모든 정보를 담기 어렵고 중요한 정보가 손실되는 문제가 발생
- Decoder는 매번 같은 Context Vector만 참고하기에 출력 단어마다 필요한 정보를 다르게 반영하기 어려움

But Attention은 말 그대로 중요한 정보에 더 집중하는 메커니즘
-> 결과를 만들 때 현재 시점에서 가장 중요한 부분에 더 높은 가중치를 부여하여 활용 (출력 단어를 생성할 때 입력 문장 속 단어들 중 어떤 단어에 얼마나 집중할지 결정하는 기술)
-> 출력 시점마다 새로운 context vector를 동적으로 계산함 -> 병렬

### 장점
1. 시간 경과에 대한 유연성
- RNN은 장기 의존성 문제를 가짐
-> attention은 모든 단어에서 다른 모든 단어로 직접 연결될 수 있기에 멀리 떨어진 단어 사이의 관계라도 정보 손실 없이 명확하게 파악

2. 공간에 대한 유연성
- 고정된 필터 (CNN)을 사용하기에 이미지의 지역적인 부분을 중심으로 정보를 처리
- Attention은 이미지의 모든 칙셀이 다른 모든 픽셀과 직접 연결되어, 픽셀 간의 거리에 상관없이 한 번에 전체 이미지의 전역적인 관계를 학습할 수 있음

3. 병렬화 
- RNN은 구조적으로 순차 계산이 필수 적이기에 병렬 처리가 어렵고 학습 속도가 매우 느림
- Attention은 self-attention에서 문장 내 모든 단어의 관계를 한번에, 독립적으로 계산 가능함 -> 병렬 처리가 가능하고 긴 시퀀스도 더 효율적으로 학습할 수 있음

### Attention is All you need
1. 배경: RNN기반의 모델들은 느린 학습 속도, 장기 의존성 문제가 존재
2. Transformer 등장: RNN의 순환 구조를 완전히 제거하고 self-attention에 기반한 Transformer 모델을 제안
3. 발견: Attention만으로도 더 좋은 성능을 낼 수 있음을 보여주었고 모든 단어들을 동시에 계산할 수 잇는 병렬화가 가능해짐
4. 효과:
4.1 LLM의 탄생: BERT와 GPT 라는 두 거대 모델의 직접적인 기반
4.2 NLP 평정: 번역, 요약, 챗봇 등 거의 모든 자연어 처리 분야에서 트랜스포머 기반 모델들의 표준으로 자리잡음
4.3 타 분야로 확장: 트랜스포머의 성공은 AI를 NLP를 넘어 컴퓨터 비전, 음성처리, 신약 개발 등 모든 영역으로 확장

## 트랜스포머
- RNN의 순차적인 계산 방식을 버리고 attention만으로 문장의 의미와 구조를 파악하는 모델

### 핵심 기술
- self-attention: 문장 안에서 어떤 단어가 다른 단어들과 얼마나 중요한 관계를 맺고 있는지 한 번에 파악하는 기술
- Multi-Head Attention: 셀프 어텐션을 여러 개의 Head로 동시에, 여러 관점에서 실행
- Positional Encoding: 단어의 위치 정보를 벡터에 추가하여 각 단어가 문장의 몇 번째 위치에 존재하는지 알 수 있음

##### self-attention
1. Query: 질문, 요청으로 지금 내가 알고 싶거나 초점을 맞추고 있는 대상
2. Key: 모든 정보들이 달고 있는 '이름표', '색인'
3. Value: 키와 한 쌍으로 묶여 있는 '실제 내용물', 쿼리와 키의 관련 계산도 끝나고 가장 관련성이 높다고 판단된 키가 선택되면, 모델은 그 키에 해당하는 밸류를 가져와 사용

#### cross-attention, self-attention
1. cross-attention: 하나의 시퀀스가 완전히 다른 시퀀스를 참고하여 정보를 만들어내는 매커니즘
- 오픈북 시험마냥 디코더가 출력을 작성하면서, 계속해서 입력의 어떤 부분을 참고할지 결정하는 과정과 똑같음

2. self-attention: 문장자체를 뜯어보며 내부의 관계를 파악
- 단어의 프로필 만들기 (벡터 임베딩)
- 단어간 관계 점수 계산 (내정 및 정렬 점수)
- 중요도 계산 (소프트맥스 & 어탠션 가중치)
- 새로운 벡터 생성

#### Scaled Dot-Product Attention
##### Dot-Product Attention
- 쿼리와 키 벡터를 내적하여 유사도를 구함
-> 문제점: Attention Score의 극단화
- 벡터 차원이 커질수록 내적 값이 지나치게 커지거나 작아질 수 있음. 이런 극단적인 값이 softmax에 입력되면 특정 단어에 확률이 과도하게 집중. gradient가 0에 가까워지는 Vanishing Gradient 문제가 발생할 수 있음

-> 내적을 진행한 후에 값의 크기를 맞추는 스케일링 과정을 추가함
계산된 내적 값이 과도하게 증가하는 것을 방지하여 안정적인 분포를 형성하고 Softmax 함수가 안정적으로 학습을 진행할 수 잇음
- 이를 통해 ContextVector 와 Self attention: 다른 단어들과의 관계를 반영한 문맥적 표현이 반영



## Multi-Head Attention
1. single-head는 문장 내 단어들 간의 관계를 파악하기 위해 가중치 행렬만을 학습하고 번역에 사용
-> 하나의 어텐션 분포 안에서 각 단어 간 문법적인 관계, 의미적인 관계, 위치 관계 등 모든 종류의 정보를 한 번에 담아 가중 평균을 구해버리기에 결국 각 단어 간의 중요한 관계 놓침

2. Multi-head는 한 단어와 다른 단어 간의 관계를 여러 차원으로 나누어 병렬로 학습

2. 병렬 어텐션 계산 - 서로에게 전혀 간섭하지 않고 병렬로 Scaled Dot-priduct Attention을 계산

## 트랜스포머 전처리 단계
1. 작동 방식: 텍스트를 입력 받고 그 텍스트를 토큰화하고 임베딩하여 Positional Encodinf으로 순서 정보를 추가하여 인코더 레이어에서 실질적인 연산을 수행

2. 토큰화: 입력 텍스트를 모델이 처리할 수 있는 단위로 나눔
3. 임베딩: 숫자의 벡터로 변환
4. 포지셔널 인코딩

### 인코더, 디코더
1. 인코더 = 입력 문장을 이해하고 요약된 의미 벡터로 변환하는 과정
2. 디코더 = 분석한 입력 문장의 의미 벡터를 받아서 출력 문장을 순차적으로 생성하는 역할

### 전체 흐름
1. 입력 준비: 문장을 의미 + 위치 벡터로 바꾸기
2. 인코더: 입력 문장의 문맥적 의미 깊이 파악
3. 디코더: 번역 문장을 한단어씩 생성
