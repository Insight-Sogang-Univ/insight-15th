```python
"""
서강대학교 - 게시물 허가 페이지 크롤러

설치:
    pip install beautifulsoup4

실행:
    python sogang_permisstion_post_crawler.py
"""

import json
import sys
from datetime import datetime
from bs4 import BeautifulSoup

# 출력 생략 방지
sys.stdout.reconfigure(encoding="utf-8")

OUTPUT_PATH = "sogang_permisstion_post.json"

HTML = """
<div data-v-72bb78ca="" class="container mx-auto"><div data-v-72bb78ca=""><h2 class="text-center zoom-text text-[#1E1E1E] md:my-[60px] my-[40px] text-[24px] md:text-[32px] font-[700] line-height-140">게시물 허가</h2></div><div class="flex flex-col gap-[2.5rem] md:gap-[3.75rem]" data-v-72bb78ca=""><div class="" data-v-72bb78ca=""><div class="flex justify-between items-center" data-v-72bb78ca=""><div class="flex items-center gap-2 pb-3" data-v-72bb78ca=""><div class="bg-[#A72018] w-[2px] h-[24px]" data-v-72bb78ca=""></div><div class="text-[#1E1E1E] text-[1rem] md:text-[1.25rem] font-[700] line-height-140 zoom-text" data-v-72bb78ca=""> 학교 게시물 및 현수막 게시 관련 </div></div></div><div class="text-[#454545] text-[0.875rem] md:text-[1rem] font-[400] line-height-180 md:pl-[10px] zoom-text" data-v-72bb78ca=""> 학생지원팀에서는 깨끗한 우리 교정 관리와 게시물의 효과적인 홍보를 위하여 학교 게시물 및 현수막 게시 관리를 하고 있습니다.<br data-v-72bb78ca=""> 이와 관련 학교 모든 구성원의 규정 준수 및 협조와 자율적인 관리를 부탁드립니다. </div></div><div class="flex flex-col gap-[1rem] md:gap-[3rem] overflow-auto" data-v-72bb78ca=""><div class="flex justify-between items-center" data-v-72bb78ca=""><div class="flex items-center gap-2 pb-3" data-v-72bb78ca=""><div class="bg-[#A72018] w-[2px] h-[24px]" data-v-72bb78ca=""></div><div class="text-[#1E1E1E] text-[1rem] md:text-[1.25rem] font-[700] line-height-140 zoom-text" data-v-72bb78ca=""> 교내 게시물 부착 절차 </div></div></div><div class="overflow-x-auto" data-v-72bb78ca=""><div class="w-[764px] md:w-full" data-v-72bb78ca=""><picture data-v-72bb78ca=""><source media="(min-width:1920px)" srcset="https://www.sogang.ac.kr/_nuxt/img1_1920.12c75809.svg" data-v-72bb78ca=""><source media="(min-width:1024px)" srcset="https://www.sogang.ac.kr/_nuxt/img1_1024.7c313222.svg" data-v-72bb78ca=""><source media="(min-width:768px)" srcset="https://www.sogang.ac.kr/_nuxt/img1_768.b51772de.svg" data-v-72bb78ca=""><img class="w-full h-full" alt="img1_768" src="https://www.sogang.ac.kr/_nuxt/img1_768.b51772de.svg" data-v-72bb78ca=""></picture></div></div></div><div class="" data-v-72bb78ca=""><div class="flex justify-between items-center" data-v-72bb78ca=""><div class="flex items-center gap-2 pb-3" data-v-72bb78ca=""><div class="bg-[#A72018] w-[2px] h-[24px]" data-v-72bb78ca=""></div><div class="text-[#1E1E1E] text-[1rem] md:text-[1.25rem] font-[700] line-height-140 zoom-text" data-v-72bb78ca=""> 게시물 </div></div></div><div class="text-[#454545] text-[0.875rem] md:text-[1rem] font-[400] line-height-180 md:ml-2 zoom-text" data-v-72bb78ca=""> 1. 자보, 포스터 및 기타 게시 문건이 이에 포함됩니다.<br data-v-72bb78ca=""> 2. 학내 게시물은 10장, 학외 게시물은 5장에 대하여 게시스티커를 발급하며, 게시기간은 7일(휴일포함)입니다. <br data-v-72bb78ca=""> 3. 게시물의 성격에 부합하는 게시판에 자율부착 및 자율철거를 원칙으로 합니다. <br data-v-72bb78ca=""> 4. 상기 사항을 준수하지 않은 게시물이나 게시판 이외의 장소에 은 학생지원팀에서 임의로 철거 가능하며, 철거된 게시물은 보관하지 않습니다. </div></div><div class="" data-v-72bb78ca=""><div class="flex justify-between items-center" data-v-72bb78ca=""><div class="flex items-center gap-2 pb-3" data-v-72bb78ca=""><div class="bg-[#A72018] w-[2px] h-[24px]" data-v-72bb78ca=""></div><div class="text-[#1E1E1E] text-[1rem] md:text-[1.25rem] font-[700] line-height-140 zoom-text" data-v-72bb78ca=""> 현수막 </div></div></div><div class="text-[#454545] text-[0.875rem] md:text-[1rem] font-[400] line-height-180 md:ml-2 zoom-text" data-v-72bb78ca=""> 1. 현수막 및 각종 플래카드가 이에 포함됩니다.<br data-v-72bb78ca=""> 2. 정문밖 게시대 사용은 발전홍보팀에 미리 연락후 승인을 받아야만 사용가능합니다. (신R관앞 게시대는 공학부 행정팀에서 관리합니다.) <br data-v-72bb78ca=""> 3. 학내 현수막 게시가 가능한 장소가 매우 제한되어 있는 관계로 학생지원팀을 방문하여 허가를 받은후, 관련 스티커를 부착하여 게시하여야 합니다. <br data-v-72bb78ca=""> 4. 게시기간은 14일(휴일포함)입니다. <br data-v-72bb78ca=""> 5. 현수막 및 각종 플래카드는 자율부착 및 자율철거를 원칙으로 합니다. <br data-v-72bb78ca=""> 6. 상기 사항을 준수하지 않은 현수막 및 플래카드는 학생지원팀에서 임의로 철거 가능하며, 철거된 현수막은 보관하지 않습니다. </div></div><div class="" data-v-72bb78ca=""><div class="flex justify-between items-center" data-v-72bb78ca=""><div class="flex items-center gap-2 pb-3" data-v-72bb78ca=""><div class="bg-[#A72018] w-[2px] h-[24px]" data-v-72bb78ca=""></div><div class="text-[#1E1E1E] text-[1rem] md:text-[1.25rem] font-[700] line-height-140 zoom-text" data-v-72bb78ca=""> 기타 </div></div></div><div class="text-[#454545] text-[0.875rem] md:text-[1rem] font-[400] md:line-height-180 ml-2 zoom-text" data-v-72bb78ca=""> 1. 학교 시설물 및 자연 보호를 위하여 정해진 게시대에만 게시가능합니다.<br data-v-72bb78ca=""> 2. 나무, 가로등, 운동장 펜스, 곤자가 플라자 전지역 및 기타 보호가 필요한 학교 시설물 등에 불법 부착할 경우, 학생지원팀 허가 여부에 관계없이 철거하며, 철거된 현수막은 보관하지 않습니다. </div></div></div><div data-v-72bb78ca="" class="mt-[120px] my-[102px] xl:mt-[162px]"><div class="w-full border border-t-[#EFEFEF]"></div><div class="flex justify-center break-words max-w-full md:justify-end md:items-center items-start gap-[8px] mt-[28px]"><div class="flex justify-end items-center h-fit gap-[8px] pl-[20px]"><img src="https://www.sogang.ac.kr/_nuxt/icon_phone.d36efbb6.svg" alt="icon_phone"><p class="text-[#861F1C] text-[14px] font-[700] line-height-180 zoom-text">담당부서</p><div class="h-[12px] border border-r border-[#EFEFEF]"></div></div><p class="text-[#727272] text-[14px] md:text-[16px] font-[400] break-words break-normal line-height-180 zoom-text">학생지원팀(02-705-8740)</p></div></div></div>
"""


def parse(html: str) -> dict:
    soup = BeautifulSoup(html, "html.parser")

    result = {
        "page_url": "https://www.sogang.ac.kr/ko/campus/student-support/permisstion-post",
        "crawled_at": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "page_title": "",
        "sections": [],
        "department": {"label": "", "contact": ""},
    }

    # 페이지 제목
    h2 = soup.find("h2")
    if h2:
        result["page_title"] = h2.get_text(strip=True)

    # 본문 섹션
    content_div = soup.find("div", class_=lambda c: c and "flex-col" in c)
    if content_div:
        for section in content_div.find_all("div", recursive=False):
            heading_el = section.find(
                "div", class_=lambda c: c and "font-[700]" in c and "zoom-text" in c
            )
            heading = heading_el.get_text(strip=True) if heading_el else ""
            if not heading:
                continue

            text_el = section.find(
                "div", class_=lambda c: c and "454545" in c and "zoom-text" in c
            )
            text_lines = []
            if text_el:
                for br in text_el.find_all("br"):
                    br.replace_with("\n")
                for line in text_el.get_text().split("\n"):
                    line = line.strip()
                    if line:
                        text_lines.append(line)

            images = []
            picture = section.find("picture")
            if picture:
                sources = [
                    {"media": s.get("media", ""), "srcset": s.get("srcset", "")}
                    for s in picture.find_all("source")
                ]
                img = picture.find("img")
                images.append({
                    "sources": sources,
                    "default_src": img.get("src", "") if img else "",
                    "alt": img.get("alt", "") if img else "",
                })

            result["sections"].append({
                "heading": heading,
                "text_lines": text_lines,
                "images": images,
            })

    # 담당부서
    dept_div = soup.find("div", class_=lambda c: c and "mt-[120px]" in c)
    if dept_div:
        label_el = dept_div.find("p", class_=lambda c: c and "861F1C" in c)
        contact_el = dept_div.find("p", class_=lambda c: c and "727272" in c)
        result["department"] = {
            "label": label_el.get_text(strip=True) if label_el else "담당부서",
            "contact": contact_el.get_text(strip=True) if contact_el else "",
        }

    return result


def main():
    print("=" * 55)
    print("  서강대학교 게시물 허가 페이지 파싱")
    print("=" * 55)

    data = parse(HTML)

    print(f"\n  페이지 제목 : {data['page_title']}")
    print(f"  섹션 수     : {len(data['sections'])}개\n")

    for s in data["sections"]:
        print(f"  ▶ {s['heading']}")
        for line in s["text_lines"]:
            print(f"      {line}")
        for img in s["images"]:
            print(f"      📷 [{img['alt']}] {img['default_src']}")
        print()

    dept = data["department"]
    print(f"  {dept['label']}: {dept['contact']}")

    # JSON 저장
    with open(OUTPUT_PATH, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)

    print(f"\n[완료] '{OUTPUT_PATH}' 저장 완료!")


if __name__ == "__main__":
    main()
```
