# 시계열 데이터
- 순차적인 시간의 흐름에 따라 수집된 데이터로 연속한 관측치는 서로 상관관계가 존재하고 순서가 중요한 '시퀀스 데이터'로 간주
-> 미래 데이터 예측을 목적으로 분석하고 모델링하는 통계적인 방법과 기법


1. 단변량: 하나의 숫자가 시간에 따라 변하는 것
- 시간에 따라 연속적으로 측정된 단일 변수로 이루어진 시계열 데이터
- 주식, 온도. 매출 등
2. 다변량: 여러 숫자가 동시에 시간에 따라 변하는 것
- 같은 시간에 기록된 둘 이상의 변수로 이루어진 시계열 데이터
- 경제 지표, 기후 모델

### 특징
1. 시간 의존성
- 시간 순서에 따라 관측되므로 데이터의 순서를 바꾸면 의미가 달라짐
- 과거의 값이 현재의 값에 영향

2. 계절성 및 추세
- 시간에 따라 장기적으로 증가/감소하는 추세 + 일정 주기로 반복되는 패턴
- 계절성: 일정 주기에 따라 반복적으로 나타나는 데이터의 변동
- 추세: 시간의 흐름에 따라 나타나는 장기적인 증가/감소 경향

3. 자기 상관성: 자기 자신끼리 상관관계를 갖는 특성
-> 회귀 분석이 불가능함 (오차항 독립성 가정 위반, 허위 상관관계 문제, 예측의 어려움과 모델의 복잡성 증가)

### 분석 요인
1. 추세 요인
- 데이터 전반에 걸쳐 시간에 따라 지속적으로 증가/감소하는 경향

2. 계절 요인
- 일정한 주기(계절 , 월, 요일 등 정기적인 주기)로 반복되는 패턴

3. 순환 요인
- 계절성보다 더 길고 불규칙한 주기로 반복되는 변동

4. 불규칙 요인
- 추세, 계절성, 순환성을 모두 제거하고 남은 예측 불가능한 변동

## 전처리
- 시계열 데이터의 특성으로 인해 왜곡되지 않도록 시계열 특성을 반영한 전처리 필요

1. 결측치: 특정 시점에서 값이 측정되지 않았거나 누락된 상태
- 삭제: 비어있는 시점의 데이터를 제거 - 시간의 흐름의 연속석이 깨지고 분석구간이 짧아짐
- 대치: 평균, 중앙값, 최빈값. 0으로 결측치를 채움 - 패턴 왜곡 가능
- 보간법: 누락된 값을 추정하는 수학적 방법

2. 이상치: 일반적이지 않은 값: IQR, Z-score, STL 분해, 분류 및 회귀 트리, 클러스터링 기반 이상 탐지, 오토 인코더 방법 사용
- 결측치로 간주 : 같은 방법으로 처리

3. 평활화: 노이즈를 줄이고 전체적인 흐름이나 추세를 더 명확하게 파악하기 위해 데이터를 부드럽게 만드는 기술
- 노이즈: 원래 데이터 안에 자연스럽게 섞여 있는 설명 불가능한 랜덤 변동
- 이동 평균: MA
: 노이즈 제저, 추세 파악, 예측의 기초 자료로 사용

4. 요소 분해: 주어진 시계열 데이터를 개별 성분으로 분해하는 과정
- 추세, 계절, 순환, 불규칙 성분으로 분해

- 가법 모형
$y_t = T_t + S_t + R_t$
- 시계열 = 추세 + 계절성 + 불규칙성
- 각 변동 요인들이 독립적으로 더해지며, 모든 요소의 옇양이 일정함

- 승법 모형
$y_t = T_t * S_t * R_t$
- 시게열 = 추세*계절성*불규칙성
- 추세가 커질수록 계절성이나 불규칙성의 효과도 비례해서 커짐

- seasonal_decompose:: 전통적인 시계열 분해 기법으로 이동 평균을 통해 추세를 추출한 뒤 잔여 성분에서 계절성 분리
1. 추세 추출: 중심 이동 평균을 적용해서 추세 성분 추정
2. Detrending: 원 시계열에서 추세를 제거해서 시계열 생성
3. 계절성 추정: 디트렌드 시계열에서 동일한 계정 인덱스에 해당하는 평균을 취해 계절패턴 도출, 연간 평균이 0이되도록 정규화
4. 잔차 산출

- STL: 추세와 계절성이 시간에 따라 변화하는 경우에도 유연하게 포착, 이상치에 강건한 분해 가능
- Inner Loop
1. 추세 제거
2. Cycle-subseries 추출 후 각 subseries에 LOESS 적용
3. Low-pass filter로 저주파 성분 제거
4. 계절 성분 확정
5. 계절성 제거:
6. LOESS로 추세 갱신
→ 통상 2회 반복으로 수렴
- Outer Loop
잔차 크기 기준으로 강건성 가중치를 부여해 이상치의 영향력을 줄인 후 Inner Loop 재실행

### 정상성 판단: 시간에 상관없이 일정한 성질을 띠고 있다
- 평균 분산 등 통계적 특성이 시간에 따라 일정한 수준으로 유지되는 것

- 강한 정상성: 완벽하나 비현실적
1. 개념: 시간을 언제로 잡든 데이터의 확률 분포 자체가 똑같은 상태
2. 특징: 평균, 분산 뿐만 아니라 데이터의 왜도, 첨도 등 모든 통계적 특성이 일정
3. 한계: 현실 데이터는 잡음과 외부 요인이 항상 존재하기에 이런 상태는 거의 없음

- 약한 정상성: 실무에서 쓰는 현실적 기준
1. 평균이 일정: 시간이 지나도 일정한 수준 유지
2. 분산 일정: 데이터의 오르내림 변동 폭이 일정한 수준
3. 자기공분산이 시차에만 의존: 데이터 간의 관계가 특정 시점이 아니라 시간의 간격에 의해서만 결정
-> 약 정상성을 띠는 시계열은 시간이 흘러도 평균, 분산이 변하지 않으며, 자기 공분산이시차에만 의존

-판단
1. 시각적 판단: 일정한 기준선을 중심으로 일정한 폭으로 진동하는 형태
- 비정상: 추세가 있는 경우 (시간에 따라 데이터가 지속적으로 상승 or 하락)
- 계절성: 1년 등의 특정 주기로 패턴이 반복
-> 추세가 뚜렷한 것은 차분한 결과로 해석

2. 통계적 검정: 단위근이 있는지 확인
- 과거 충격이 영구적으로 누적되는 특성을 의미 (대표적인 비정상성의 원인으로 시계열이 random하게 움직이게 만들어 평균으로 회귀하지 않고 분산이 시간에 따라 증가하며 예측이 어려움)

- 정상성 확보를 위한 방법
1. 추세 제거: 연이은 관측값들의 차이를 계산하여 차분하게 만듦
2. 계절성 제거: 시계열이 주기적인 패턴을 가질 때 적용
3. 분산 안정화: 시계열의 변동 폭이 점점 커지거나 작아질 때 로그 변환하여 사용

#### 정리
1. 정상 비정상 여부 판단: 단위근 검정
2. 비정상 시계열이면 추세가 있으면 차분과 추세 제거, 계절성이 있다면 계절 차분 or deseasoning, 분산 변화가 있으면 로그 변환, box-cox 변환
3. 정상성이 확보되면 시계열 구조에 맞게 모델 적용

### 단변량 시계열 분석
1. 자기 회귀 모델: 과거의 자기 자신의 값을 독립 변수로 삼아 현재 값을 예측하는 회귀 -> 시계열 데이터의 자기 상관성을 활용

$AR(p):  y_{t}=c +\phi _1y_{t-1}+\phi_2y_{t-2}+...+\phi _py_{t-p}+\varepsilon _t$
- 과거 p개 시점의 데이터의 선형 조합을 이용해 예측하는 모델

$AR(1):  y_{t}=c +\phi_1y_{t-1}+\varepsilon _t$
- y의 현재값은 1시점 이전의 y값의 영향을 받는다

2. 이동 평균 모델: 과거 시점의 잔차를 독립변수로 삼아 현재 값을 예측하는 모델
- 잔차: 실제 값과 예측 값의 차이

$MA(q): y_t = \mu +\varepsilon _t+ \theta _1\varepsilon_{t-1}+\theta_2\varepsilon_{t-2}+...+\theta_q\varepsilon_{t-q}$
- q차 이동 평균 모형으로 과거 q개의 예측오차의 선형 결합으로 예측하는 모델

3. AR, MA 모델의 전제 조건
- 정상성: AR과 MA 모델은안정적 시계열 분석으로 확룰적 안정성을 전제로 추정
조건: 특성 방적식의 모든해의 절댓값이 1보다 커야 정상성 확보
- MA 모델은 잔차들은 이미 정상성을 만족하기에 잔차들의 조합으로 이뤄진 MA 모형도 항상 정상성 만족

- 가역성: MA 모형을 AR 형태로 재표현할 수 있는지 여부
-- MA 모형에서 현재 시점의 값들을 과거 시점의 오차항으로 예측

#### ARMA 모형
- 과거 p개의 관측값과 과거 q개의 오차를 활용해서 예측하는 방법
- AR 모형의 정상성과 MA 모형의 가역성 모두 만족해야 함

#### 자기 상관함수
- 두 그래프를 통해 현재 값이 과거 값과 어떤 관계가 있는지 파악하고 이를 통해 적절한 차수를 결정
- 자기 상관계수
-- ARMA 모형에서 사용할 q개의 오차 결정 방법
-- 두 시점 사이의 상관관계 파악(간접 영향까지 모두 파악)
-- ARMA 모형에서 사용할 p개의 시점 결정 방법

#### 불안정 시계열 분석
1. 비정상 시계열 데이터 (추세, 계쩔성 존재)
2. 정상성 확보(차분 or 계절성 차분)
3. 정상 시계열
4. ARMA를 사용해 안정적 시계열 분석

##### ARIMA
$ARMA(p, q) + \text{차분} \rightarrow ARIMA(p,d,q)$

##### SARIMA
$ARIMA(p,d,q) + \text{계절성(Seasonal) 차분} \\ \rightarrow SARIMA(p,d,q)(P, D, Q)_m$
- 시계열 데이터의 계절성을 처리할 수 있는 모델
- ARIMA 모델에 계절성 변동을 반영한 모델

### 다변량 시계열 분석
- 자신의 과거 값과 다른 변수들의 과거 값을 함께 사용하여 미래를 예측하는 모델
1. 적합한 경우
- 변수들이 서로 영향을 주고 받는 경우
- 경제, 금융 지표처럼 상호 의존성이 중요한 경우

2. 부적합한 경우
- 변수가 너무 많으면 파라미터 폭발
- 변수 간 관계가 명확하지 않은 경우
- 장기 예측이 필요한 경우

이를 활용한 분석 방법
1. 충격 반응 함수: 한 변수의 충격이 다른 변수에 미치는 영향을 시간 흐름에 따라 분석

2. 예측오차 분산 분해: 예측 오차에 각 변수가 얼마나 기여하는지 분석

### 딥러닝을 활용한 시계열 분석
1. 배경 (전통 통계 모델의 한계)
- 데이터가 많으면 처리 속도가 느려짐
- 크기가 고정된 데이터만 사용 가능
- 비선형 패턴 포착이 어려움

2. RNN 계열
적합한 경우
- 순차적 패턴이 중요한 시계열 데이터
- 과거 정보가 미래 예측에 영향을 주는 문제
-시계열 길이와 목적에 따라 모델 선택 가능

부적합한 경우
- 매우 긴 시계열에서는 학습이 어려움
- 병렬 처리 효울이 낮음
- 트랜스포머 계열 대비 학습 속도가 느림

3. 트램스포머 계열
- 인포머(장기 예측 특화)
--- 특징: 불필요한 연산을 줄여 계산 복잡도를 대폭 낮품
- 오토 포머(주기성 포착 특화)
--- 특징: 시계열의 전통적 기법인 요소 분해 개념을 딥러닝 구조에 직접 결합
- TFT
--- 특징: 과거 데이터 뿐만 아니라 다양한 외부 변수를 섞어서 예측할 수 있는 다목적 아키텍쳐

4. 파운데이션 모델: 거대한 데이터로 사전 학습하여 다양한 문제를 해결할 수 있는 범용 모델
- 특징: 틀정 문제 전용이 아니며, 범용적이고, 사전 학습 후 파인튜닝으로 특화
- 유형: LLM 기반 모델과 시게열 전용 모델이 존재
- LLM 기반 모델 활용 방식프롬포트 엔지니어링 이나 시계열 데이터로 파인 튜닝, 시계열을 텍스트로 변환 후 LLM 활용
